## (old)Propagating synthetic big wave

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, FFMpegWriter
from pathlib import Path


# ==================================================
# Settings
# ==================================================
OUT_DIR = Path("propagating_wave_outputs")
OUT_DIR.mkdir(parents=True, exist_ok=True)

NX = 250
NY = 250

DX_KM = 1.0
DY_KM = 1.0

DT_SECONDS = 30
TOTAL_TIME_SECONDS = 3600*4

SPEED_MS = 150.0
SPEED_KMS = SPEED_MS / 1000.0

WAVE_LAMBDA_KM = 800.0
WAVE_ANGLE_DEG = 15.0

WAVE_AMP = 1.0

# Large envelope so the wave fills most/all of the FOV
WAVE_SIGMA_X_KM = 1200.0
WAVE_SIGMA_Y_KM = 1200.0

NOISE_STD = 0.02
RANDOM_SEED = 1

FPS = 20
DPI = 200

CMAP = "RdBu_r"


# ==================================================
# Coordinate grid
# ==================================================
x_km = np.arange(NX) * DX_KM
y_km = np.arange(NY) * DY_KM
x2d, y2d = np.meshgrid(x_km, y_km)

x0 = 0.5 * (NX - 1) * DX_KM
y0 = 0.5 * (NY - 1) * DY_KM

times_s = np.arange(0, TOTAL_TIME_SECONDS + DT_SECONDS, DT_SECONDS)
times_min = times_s / 60.0
nt = len(times_s)

theta = np.deg2rad(WAVE_ANGLE_DEG)

# Propagation direction coordinate
s_km = (x2d - x0) * np.cos(theta) + (y2d - y0) * np.sin(theta)

# Envelope
envelope = np.exp(
    -0.5 * (
        ((x2d - x0) / WAVE_SIGMA_X_KM) ** 2
        + ((y2d - y0) / WAVE_SIGMA_Y_KM) ** 2
    )
)

k = 2.0 * np.pi / WAVE_LAMBDA_KM


# ==================================================
# Create time series
# ==================================================
rng = np.random.default_rng(RANDOM_SEED)

wave_cube = np.zeros((nt, NY, NX), dtype=np.float32)

for it, t in enumerate(times_s):
    displacement_km = SPEED_KMS * t

    phase = k * (s_km - displacement_km)

    field = WAVE_AMP * envelope * np.cos(phase)
    field += rng.normal(0.0, NOISE_STD, size=field.shape)

    wave_cube[it] = field.astype(np.float32)


# Consistent color limits
vmax = np.nanpercentile(np.abs(wave_cube), 99)
vmin = -vmax


# ==================================================
# Animation
# ==================================================
fig, ax = plt.subplots(figsize=(7, 6))

im = ax.imshow(
    wave_cube[0],
    origin="upper",
    cmap=CMAP,
    vmin=vmin,
    vmax=vmax,
    extent=[x_km.min(), x_km.max(), y_km.max(), y_km.min()],
)

cbar = fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
cbar.set_label("Synthetic wave amplitude")

title = ax.set_title("")
ax.set_xlabel("E-W distance (km)")
ax.set_ylabel("N-S distance (km)")
ax.set_aspect("equal")


def update(frame):
    im.set_data(wave_cube[frame])
    title.set_text(
        f"Propagating Synthetic Wave | "
        f"t = {times_min[frame]:.1f} min | "
        f"speed = {SPEED_MS:.0f} m/s"
    )
    return im, title


ani = FuncAnimation(
    fig,
    update,
    frames=nt,
    interval=1000 / FPS,
    blit=False,
)

out_mp4 = OUT_DIR / "propagating_wave_animation.mp4"

writer = FFMpegWriter(fps=FPS, bitrate=3000)
ani.save(out_mp4, writer=writer, dpi=DPI)

plt.close(fig)


# ==================================================
# Keograms
# ==================================================
center_x_idx = NX // 2
center_y_idx = NY // 2

# N-S keogram: vertical slice through center x
keogram_ns = wave_cube[:, :, center_x_idx]

# E-W keogram: horizontal slice through center y
keogram_ew = wave_cube[:, center_y_idx, :]


# --------------------------
# N-S keogram
# --------------------------
fig_ns, ax_ns = plt.subplots(figsize=(9, 5))

im_ns = ax_ns.imshow(
    keogram_ns.T,
    origin="lower",
    aspect="auto",
    cmap=CMAP,
    vmin=vmin,
    vmax=vmax,
    extent=[
        times_min.min(),
        times_min.max(),
        y_km.min(),
        y_km.max(),
    ],
)

cbar_ns = fig_ns.colorbar(im_ns, ax=ax_ns, fraction=0.046, pad=0.04)
cbar_ns.set_label("Synthetic wave amplitude")

ax_ns.set_title("N-S Keogram")
ax_ns.set_xlabel("Time (min)")
ax_ns.set_ylabel("N-S distance (km)")

plt.tight_layout()

out_ns = OUT_DIR / "keogram_NS.png"
fig_ns.savefig(out_ns, dpi=DPI, bbox_inches="tight")
plt.close(fig_ns)


# --------------------------
# E-W keogram
# --------------------------
fig_ew, ax_ew = plt.subplots(figsize=(9, 5))

im_ew = ax_ew.imshow(
    keogram_ew.T,
    origin="lower",
    aspect="auto",
    cmap=CMAP,
    vmin=vmin,
    vmax=vmax,
    extent=[
        times_min.min(),
        times_min.max(),
        x_km.min(),
        x_km.max(),
    ],
)

cbar_ew = fig_ew.colorbar(im_ew, ax=ax_ew, fraction=0.046, pad=0.04)
cbar_ew.set_label("Synthetic wave amplitude")

ax_ew.set_title("E-W Keogram")
ax_ew.set_xlabel("Time (min)")
ax_ew.set_ylabel("E-W distance (km)")

plt.tight_layout()

out_ew = OUT_DIR / "keogram_EW.png"
fig_ew.savefig(out_ew, dpi=DPI, bbox_inches="tight")
plt.close(fig_ew)

np.save(OUT_DIR / "keogram_NS.npy", keogram_ns)
np.save(OUT_DIR / "keogram_EW.npy", keogram_ew)
np.save(OUT_DIR / "times_s.npy", times_s)
np.save(OUT_DIR / "x_km.npy", x_km)
np.save(OUT_DIR / "y_km.npy", y_km)

# ==================================================
# Save one final frame too
# ==================================================
fig_last, ax_last = plt.subplots(figsize=(7, 6))

im_last = ax_last.imshow(
    wave_cube[-1],
    origin="upper",
    cmap=CMAP,
    vmin=vmin,
    vmax=vmax,
    extent=[x_km.min(), x_km.max(), y_km.max(), y_km.min()],
)

cbar_last = fig_last.colorbar(im_last, ax=ax_last, fraction=0.046, pad=0.04)
cbar_last.set_label("Synthetic wave amplitude")

ax_last.set_title("Final Frame: Propagating Synthetic Wave")
ax_last.set_xlabel("E-W distance (km)")
ax_last.set_ylabel("N-S distance (km)")
ax_last.set_aspect("equal")

plt.tight_layout()

out_last = OUT_DIR / "final_frame.png"
fig_last.savefig(out_last, dpi=DPI, bbox_inches="tight")
plt.close(fig_last)


print("Saved:")
print(out_mp4)
print(out_ns)
print(out_ew)
print(out_last)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, FFMpegWriter
from pathlib import Path
from scipy.ndimage import map_coordinates


# ==================================================
# Settings
# ==================================================
OUT_DIR = Path("propagating_wave_outputs")
OUT_DIR.mkdir(parents=True, exist_ok=True)

NX = 250
NY = 250

DX_KM = 1.0
DY_KM = 1.0

DT_SECONDS = 30
TOTAL_TIME_SECONDS = 3600 * 4

SPEED_MS = 150.0
SPEED_KMS = SPEED_MS / 1000.0

WAVE_LAMBDA_KM = 400.0
WAVE_ANGLE_DEG = 15.0

WAVE_AMP = 1.0

# Keep large envelope
WAVE_SIGMA_X_KM = 800.0
WAVE_SIGMA_Y_KM = 800.0

NOISE_STD = 0.02
RANDOM_SEED = 1

FPS = 20
DPI = 200

CMAP = "RdBu_r"

# Directional keograms
KEOGRAM_ANGLES_DEG = np.arange(0, 166, 15)
KEOGRAM_WIDTH_PIXELS = 3


# ==================================================
# Coordinate grid
# ==================================================
x_km = np.arange(NX) * DX_KM
y_km = np.arange(NY) * DY_KM
x2d, y2d = np.meshgrid(x_km, y_km)

x0 = 0.5 * (NX - 1) * DX_KM
y0 = 0.5 * (NY - 1) * DY_KM

times_s = np.arange(0, TOTAL_TIME_SECONDS + DT_SECONDS, DT_SECONDS)
times_min = times_s / 60.0
nt = len(times_s)

theta = np.deg2rad(WAVE_ANGLE_DEG)

k = 2.0 * np.pi / WAVE_LAMBDA_KM


# ==================================================
# Create time series
# ==================================================
rng = np.random.default_rng(RANDOM_SEED)

wave_cube = np.zeros((nt, NY, NX), dtype=np.float32)

for it, t in enumerate(times_s):
    displacement_km = SPEED_KMS * t

    # Moving wave packet center
    x_center_t = x0 + displacement_km * np.cos(theta)
    y_center_t = y0 + displacement_km * np.sin(theta)

    # Moving envelope
    envelope_t = np.exp(
        -0.5 * (
            ((x2d - x_center_t) / WAVE_SIGMA_X_KM) ** 2
            + ((y2d - y_center_t) / WAVE_SIGMA_Y_KM) ** 2
        )
    )

    # Directional coordinate measured from the original center
    s_km = (x2d - x0) * np.cos(theta) + (y2d - y0) * np.sin(theta)

    # Propagating phase
    phase = k * (s_km - displacement_km)

    field = WAVE_AMP * envelope_t * np.cos(phase)
    field += rng.normal(0.0, NOISE_STD, size=field.shape)

    wave_cube[it] = field.astype(np.float32)


vmax = np.nanpercentile(np.abs(wave_cube), 99)
vmin = -vmax


# ==================================================
# Animation
# ==================================================
fig, ax = plt.subplots(figsize=(7, 6))

im = ax.imshow(
    wave_cube[0],
    origin="upper",
    cmap=CMAP,
    vmin=vmin,
    vmax=vmax,
    extent=[x_km.min(), x_km.max(), y_km.max(), y_km.min()],
)

cbar = fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
cbar.set_label("Synthetic wave amplitude")

title = ax.set_title("")
ax.set_xlabel("E-W distance (km)")
ax.set_ylabel("N-S distance (km)")
ax.set_aspect("equal")


def update(frame):
    im.set_data(wave_cube[frame])
    title.set_text(
        f"Moving Synthetic Wave Packet | "
        f"t = {times_min[frame]:.1f} min | "
        f"speed = {SPEED_MS:.0f} m/s"
    )
    return im, title


ani = FuncAnimation(
    fig,
    update,
    frames=nt,
    interval=1000 / FPS,
    blit=False,
)

out_mp4 = OUT_DIR / "moving_wave_packet_animation.mp4"

writer = FFMpegWriter(fps=FPS, bitrate=3000)
ani.save(out_mp4, writer=writer, dpi=DPI)

plt.close(fig)


# ==================================================
# Save first frame with directional keogram cuts
# ==================================================
fig_first, ax_first = plt.subplots(figsize=(7, 6))

im_first = ax_first.imshow(
    wave_cube[0],
    origin="upper",
    cmap=CMAP,
    vmin=vmin,
    vmax=vmax,
    extent=[x_km.min(), x_km.max(), y_km.max(), y_km.min()],
)

cbar_first = fig_first.colorbar(im_first, ax=ax_first, fraction=0.046, pad=0.04)
cbar_first.set_label("Synthetic wave amplitude")

cx_km = x0
cy_km = y0
half_length_km = 0.5 * min((NX - 1) * DX_KM, (NY - 1) * DY_KM)

for angle_deg in KEOGRAM_ANGLES_DEG:
    theta_cut = np.deg2rad(angle_deg)

    dx_line = half_length_km * np.cos(theta_cut)
    dy_line = half_length_km * np.sin(theta_cut)

    x_line = [cx_km - dx_line, cx_km + dx_line]
    y_line = [cy_km - dy_line, cy_km + dy_line]

    ax_first.plot(
        x_line,
        y_line,
        linewidth=1.2,
        alpha=0.8,
        label=f"{angle_deg:.0f}°",
    )

ax_first.set_title("First Frame with Directional Keogram Cuts")
ax_first.set_xlabel("E-W distance (km)")
ax_first.set_ylabel("N-S distance (km)")
ax_first.set_aspect("equal")

ax_first.legend(
    loc="upper right",
    fontsize=7,
    ncol=2,
    framealpha=0.8,
)

plt.tight_layout()

out_first = OUT_DIR / "first_frame_with_keogram_cuts.png"
fig_first.savefig(out_first, dpi=DPI, bbox_inches="tight")
plt.close(fig_first)


# ==================================================
# Directional keogram helper
# ==================================================
def make_directional_keogram(wave_cube, angle_deg, width_pixels=3):
    """
    angle_deg:
        0 deg  = E-W slice
        90 deg = N-S slice
    """

    theta = np.deg2rad(angle_deg)

    ux = np.cos(theta)
    uy = np.sin(theta)

    px = -np.sin(theta)
    py = np.cos(theta)

    cx = (NX - 1) / 2
    cy = (NY - 1) / 2

    half_length_pix = min(NX, NY) // 2
    distances_pix = np.arange(-half_length_pix, half_length_pix + 1)

    offsets = np.arange(width_pixels) - width_pixels // 2

    keogram = np.full((nt, len(distances_pix)), np.nan, dtype=float)

    for it in range(nt):
        sampled_slices = []

        for off in offsets:
            x_sample = cx + distances_pix * ux + off * px
            y_sample = cy + distances_pix * uy + off * py

            valid = (
                (x_sample >= 0) & (x_sample <= NX - 1) &
                (y_sample >= 0) & (y_sample <= NY - 1)
            )

            values = np.full(len(distances_pix), np.nan)

            values[valid] = map_coordinates(
                wave_cube[it],
                [y_sample[valid], x_sample[valid]],
                order=1,
                mode="nearest",
            )

            sampled_slices.append(values)

        keogram[it] = np.nanmean(sampled_slices, axis=0)

    distance_km = distances_pix * DX_KM

    return keogram, distance_km


# ==================================================
# Create and save directional keograms
# ==================================================
saved_keograms = []

for angle_deg in KEOGRAM_ANGLES_DEG:
    keogram, distance_km = make_directional_keogram(
        wave_cube,
        angle_deg,
        width_pixels=KEOGRAM_WIDTH_PIXELS,
    )

    fig_k, ax_k = plt.subplots(figsize=(9, 5))

    im_k = ax_k.imshow(
        keogram.T,
        origin="lower",
        aspect="auto",
        cmap=CMAP,
        vmin=vmin,
        vmax=vmax,
        extent=[
            times_min.min(),
            times_min.max(),
            distance_km.min(),
            distance_km.max(),
        ],
    )

    cbar_k = fig_k.colorbar(im_k, ax=ax_k, fraction=0.046, pad=0.04)
    cbar_k.set_label("Synthetic wave amplitude")

    ax_k.set_title(f"Directional Keogram: {angle_deg:.0f}°")
    ax_k.set_xlabel("Time (min)")
    ax_k.set_ylabel("Distance along slice (km)")

    plt.tight_layout()

    out_png = OUT_DIR / f"keogram_angle_{angle_deg:03.0f}.png"
    out_npy = OUT_DIR / f"keogram_angle_{angle_deg:03.0f}.npy"

    fig_k.savefig(out_png, dpi=DPI, bbox_inches="tight")
    plt.close(fig_k)

    np.save(out_npy, keogram)

    saved_keograms.append(out_png)


np.save(OUT_DIR / "times_s.npy", times_s)
np.save(OUT_DIR / "x_km.npy", x_km)
np.save(OUT_DIR / "y_km.npy", y_km)
np.save(OUT_DIR / "keogram_angles_deg.npy", KEOGRAM_ANGLES_DEG)


# ==================================================
# Save final frame
# ==================================================
fig_last, ax_last = plt.subplots(figsize=(7, 6))

im_last = ax_last.imshow(
    wave_cube[-1],
    origin="upper",
    cmap=CMAP,
    vmin=vmin,
    vmax=vmax,
    extent=[x_km.min(), x_km.max(), y_km.max(), y_km.min()],
)

cbar_last = fig_last.colorbar(im_last, ax=ax_last, fraction=0.046, pad=0.04)
cbar_last.set_label("Synthetic wave amplitude")

ax_last.set_title("Final Frame: Moving Synthetic Wave Packet")
ax_last.set_xlabel("E-W distance (km)")
ax_last.set_ylabel("N-S distance (km)")
ax_last.set_aspect("equal")

plt.tight_layout()

out_last = OUT_DIR / "final_frame.png"
fig_last.savefig(out_last, dpi=DPI, bbox_inches="tight")
plt.close(fig_last)


# ==================================================
# Print outputs
# ==================================================
print("Saved:")
print(out_mp4)
print(out_first)
print(out_last)

print("\nDirectional keograms:")
for f in saved_keograms:
    print(f)

print(f"\nNumber of directional keograms: {len(KEOGRAM_ANGLES_DEG)}")
```

```python
import pickle
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, FFMpegWriter
from pathlib import Path
from scipy.ndimage import map_coordinates


# ==================================================
# Settings
# ==================================================
PKL_FILE = r"C:\Users\domin\Downloads\processed_data_full_03_25_2025.pkl"

OUT_DIR = Path("BLO_perturbation_keogram_outputs_new")
OUT_DIR.mkdir(parents=True, exist_ok=True)

DX_KM = 1.0
DY_KM = 1.0

DT_SECONDS = 120

FPS = 20
DPI = 200

CMAP = "RdBu_r"

# Make keograms every 5 degrees
KEOGRAM_ANGLES_DEG = np.arange(0, 176, 5)

# Only draw every 15 degrees on the first frame
PLOT_CUT_ANGLES_DEG = np.arange(0, 176, 15)

KEOGRAM_WIDTH_PIXELS = 7 # Average a wider strip, less noisy. 

# Keogram display limits
KEOGRAM_TIME_MIN_MAX = 300
KEOGRAM_DISTANCE_MIN = -250
KEOGRAM_DISTANCE_MAX = 250


# ==================================================
# Load PKL perturbation array
# ==================================================
with open(PKL_FILE, "rb") as file:
    data = pickle.load(file)

# PKL format:
# data[5] = ["Perturbation Images", perturbation_array]
# data[7] = ["Datetime", [date, start_time]]
wave_cube = np.asarray(data[5][1], dtype=np.float32)
date_start = data[7][1]

# Expected shape: time, y, x
if wave_cube.ndim != 3:
    raise ValueError(
        f"Expected perturbation array with shape (time, y, x), got {wave_cube.shape}"
    )

nt, NY, NX = wave_cube.shape

print("Perturbation array shape:", wave_cube.shape)
print("Date/start time:", date_start)


# ==================================================
# Coordinate grid
# ==================================================
x_km = np.arange(NX) * DX_KM
y_km = np.arange(NY) * DY_KM

times_s = np.arange(nt) * DT_SECONDS
times_min = times_s / 60.0


# ==================================================
# Color limits
# ==================================================
vmax = np.nanpercentile(np.abs(wave_cube), 99)
if not np.isfinite(vmax) or vmax == 0:
    vmax = 1.0

vmin = -vmax


# ==================================================
# Animation
# ==================================================
fig, ax = plt.subplots(figsize=(7, 6))

im = ax.imshow(
    wave_cube[0],
    origin="upper",
    cmap=CMAP,
    vmin=vmin,
    vmax=vmax,
    extent=[x_km.min(), x_km.max(), y_km.max(), y_km.min()],
)

cbar = fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
cbar.set_label("Perturbation amplitude")

title = ax.set_title("")
ax.set_xlabel("E-W distance (km)")
ax.set_ylabel("N-S distance (km)")
ax.set_aspect("equal")


def update(frame):
    im.set_data(wave_cube[frame])
    title.set_text(
        f"BLO Perturbation Image | "
        f"Frame {frame}/{nt - 1} | "
        f"t = {times_min[frame]:.1f} min"
    )
    return im, title


ani = FuncAnimation(
    fig,
    update,
    frames=nt,
    interval=1000 / FPS,
    blit=False,
)

out_mp4 = OUT_DIR / "BLO_perturbation_animation.mp4"

writer = FFMpegWriter(fps=FPS, bitrate=3000)
ani.save(out_mp4, writer=writer, dpi=DPI)

plt.close(fig)


# ==================================================
# Save first frame with directional keogram cuts
# Only show every 15 degrees, but still create keograms every 5 degrees later
# ==================================================
fig_first, ax_first = plt.subplots(figsize=(7, 6))

im_first = ax_first.imshow(
    wave_cube[0],
    origin="upper",
    cmap=CMAP,
    vmin=vmin,
    vmax=vmax,
    extent=[x_km.min(), x_km.max(), y_km.max(), y_km.min()],
)

cbar_first = fig_first.colorbar(im_first, ax=ax_first, fraction=0.046, pad=0.04)
cbar_first.set_label("Perturbation amplitude")

cx_km = 0.5 * (NX - 1) * DX_KM
cy_km = 0.5 * (NY - 1) * DY_KM
half_length_km = 0.5 * min((NX - 1) * DX_KM, (NY - 1) * DY_KM)

for angle_deg in PLOT_CUT_ANGLES_DEG:
    theta_cut = np.deg2rad(angle_deg)

    dx_line = half_length_km * np.cos(theta_cut)
    dy_line = half_length_km * np.sin(theta_cut)

    x_line = [cx_km - dx_line, cx_km + dx_line]
    y_line = [cy_km - dy_line, cy_km + dy_line]

    ax_first.plot(
        x_line,
        y_line,
        linewidth=1.4,
        alpha=0.85,
        label=f"{angle_deg:.0f}°",
    )

ax_first.set_title("First Perturbation Frame with Directional Keogram Cuts")
ax_first.set_xlabel("E-W distance (km)")
ax_first.set_ylabel("N-S distance (km)")
ax_first.set_aspect("equal")

ax_first.legend(
    loc="upper right",
    fontsize=7,
    ncol=2,
    framealpha=0.8,
)

plt.tight_layout()

out_first = OUT_DIR / "first_perturbation_frame_with_keogram_cuts_every_15deg.png"
fig_first.savefig(out_first, dpi=DPI, bbox_inches="tight")
plt.close(fig_first)


# ==================================================
# Directional keogram helper
# ==================================================
def make_directional_keogram(wave_cube, angle_deg, width_pixels=3):
    """
    angle_deg:
        0 deg  = E-W slice
        90 deg = N-S slice
    """

    nt, NY, NX = wave_cube.shape

    theta = np.deg2rad(angle_deg)

    ux = np.cos(theta)
    uy = np.sin(theta)

    px = -np.sin(theta)
    py = np.cos(theta)

    cx = (NX - 1) / 2
    cy = (NY - 1) / 2

    half_length_pix = min(NX, NY) // 2
    distances_pix = np.arange(-half_length_pix, half_length_pix + 1)

    offsets = np.arange(width_pixels) - width_pixels // 2

    keogram = np.full((nt, len(distances_pix)), np.nan, dtype=float)

    for it in range(nt):
        sampled_slices = []

        for off in offsets:
            x_sample = cx + distances_pix * ux + off * px
            y_sample = cy + distances_pix * uy + off * py

            valid = (
                (x_sample >= 0) & (x_sample <= NX - 1) &
                (y_sample >= 0) & (y_sample <= NY - 1)
            )

            values = np.full(len(distances_pix), np.nan)

            values[valid] = map_coordinates(
                wave_cube[it],
                [y_sample[valid], x_sample[valid]],
                order=1,
                mode="nearest",
            )

            sampled_slices.append(values)

        keogram[it] = np.nanmean(sampled_slices, axis=0)

    distance_km = distances_pix * DX_KM

    return keogram, distance_km


# ==================================================
# Create and save directional keograms
# Make every 5 degrees
# Display only 0 to 300 min and -250 to 250 km
# Save full keogram array as .npy
# ==================================================
saved_keograms = []

for angle_deg in KEOGRAM_ANGLES_DEG:
    keogram, distance_km = make_directional_keogram(
        wave_cube,
        angle_deg,
        width_pixels=KEOGRAM_WIDTH_PIXELS,
    )

    fig_k, ax_k = plt.subplots(figsize=(9, 5))

    im_k = ax_k.imshow(
        keogram.T,
        origin="lower",
        aspect="auto",
        cmap=CMAP,
        vmin=vmin,
        vmax=vmax,
        extent=[
            times_min.min(),
            times_min.max(),
            distance_km.min(),
            distance_km.max(),
        ],
    )

    cbar_k = fig_k.colorbar(im_k, ax=ax_k, fraction=0.046, pad=0.04)
    cbar_k.set_label("Perturbation amplitude")

    ax_k.set_title(f"BLO Directional Keogram: {angle_deg:.0f}°")
    ax_k.set_xlabel("Time (min)")
    ax_k.set_ylabel("Distance along slice (km)")

    ax_k.set_xlim(0, KEOGRAM_TIME_MIN_MAX)
    ax_k.set_ylim(KEOGRAM_DISTANCE_MIN, KEOGRAM_DISTANCE_MAX)

    plt.tight_layout()

    out_png = OUT_DIR / f"BLO_keogram_angle_{angle_deg:03.0f}.png"
    out_npy = OUT_DIR / f"BLO_keogram_angle_{angle_deg:03.0f}.npy"

    fig_k.savefig(out_png, dpi=DPI, bbox_inches="tight")
    plt.close(fig_k)

    np.save(out_npy, keogram)

    saved_keograms.append(out_png)


# ==================================================
# Save arrays and metadata
# ==================================================
np.save(OUT_DIR / "times_s.npy", times_s)
np.save(OUT_DIR / "x_km.npy", x_km)
np.save(OUT_DIR / "y_km.npy", y_km)
np.save(OUT_DIR / "keogram_angles_deg.npy", KEOGRAM_ANGLES_DEG)
np.save(OUT_DIR / "first_frame_cut_angles_deg.npy", PLOT_CUT_ANGLES_DEG)

with open(OUT_DIR / "Date_and_Start_Time.txt", "w") as f:
    f.write(f"Date: {date_start[0]}\n")
    f.write(f"Start time: {date_start[1]}\n")


# ==================================================
# Save final frame
# ==================================================
fig_last, ax_last = plt.subplots(figsize=(7, 6))

im_last = ax_last.imshow(
    wave_cube[-1],
    origin="upper",
    cmap=CMAP,
    vmin=vmin,
    vmax=vmax,
    extent=[x_km.min(), x_km.max(), y_km.max(), y_km.min()],
)

cbar_last = fig_last.colorbar(im_last, ax=ax_last, fraction=0.046, pad=0.04)
cbar_last.set_label("Perturbation amplitude")

ax_last.set_title("Final BLO Perturbation Frame")
ax_last.set_xlabel("E-W distance (km)")
ax_last.set_ylabel("N-S distance (km)")
ax_last.set_aspect("equal")

plt.tight_layout()

out_last = OUT_DIR / "final_perturbation_frame.png"
fig_last.savefig(out_last, dpi=DPI, bbox_inches="tight")
plt.close(fig_last)


# ==================================================
# Print outputs
# ==================================================
print("Saved:")
print(out_mp4)
print(out_first)
print(out_last)

print("\nDirectional keograms:")
for f in saved_keograms:
    print(f)

print(f"\nNumber of directional keograms made: {len(KEOGRAM_ANGLES_DEG)}")
print(f"First-frame cut lines shown: every 15 degrees, {len(PLOT_CUT_ANGLES_DEG)} lines")
print("Keogram plot x-limit: 0 to 300 min")
print("Keogram plot y-limit: -250 to 250 km")
print("Output folder:", OUT_DIR)
```

## Read pkl, calculate slope

```python
import pickle
import csv
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, FFMpegWriter
from pathlib import Path
from scipy.ndimage import map_coordinates


# ==================================================
# Settings
# ==================================================
PKL_FILE = r"C:\Users\domin\Downloads\processed_data_full_03_25_2025.pkl"

OUT_DIR = Path("BLO_perturbation_keogram_outputs_new")
OUT_DIR.mkdir(parents=True, exist_ok=True)

DX_KM = 1.0
DY_KM = 1.0

DT_SECONDS = 120

FPS = 20
DPI = 200

CMAP = "RdBu_r"

# Make keograms every 5 degrees
KEOGRAM_ANGLES_DEG = np.arange(0, 176, 5)

# Only draw every 15 degrees on the first frame
PLOT_CUT_ANGLES_DEG = np.arange(0, 176, 15)

# Average a wider strip, less noisy
KEOGRAM_WIDTH_PIXELS = 7

# Keogram display limits
KEOGRAM_TIME_MIN_MAX = 300
KEOGRAM_DISTANCE_MIN = -250
KEOGRAM_DISTANCE_MAX = 250

# 2D FFT settings
SAVE_FFT_IMAGES = True
FFT_EXCLUDE_RADIUS_PIXELS = 5


# ==================================================
# Load PKL perturbation array
# ==================================================
with open(PKL_FILE, "rb") as file:
    data = pickle.load(file)

# PKL format:
# data[5] = ["Perturbation Images", perturbation_array]
# data[7] = ["Datetime", [date, start_time]]
wave_cube = np.asarray(data[5][1], dtype=np.float32)
date_start = data[7][1]

# Expected shape: time, y, x
if wave_cube.ndim != 3:
    raise ValueError(
        f"Expected perturbation array with shape (time, y, x), got {wave_cube.shape}"
    )

nt, NY, NX = wave_cube.shape

print("Perturbation array shape:", wave_cube.shape)
print("Date/start time:", date_start)


# ==================================================
# Coordinate grid
# ==================================================
x_km = np.arange(NX) * DX_KM
y_km = np.arange(NY) * DY_KM

times_s = np.arange(nt) * DT_SECONDS
times_min = times_s / 60.0


# ==================================================
# Color limits
# ==================================================
vmax = np.nanpercentile(np.abs(wave_cube), 99)
if not np.isfinite(vmax) or vmax == 0:
    vmax = 1.0

vmin = -vmax


# ==================================================
# Animation
# ==================================================
fig, ax = plt.subplots(figsize=(7, 6))

im = ax.imshow(
    wave_cube[0],
    origin="upper",
    cmap=CMAP,
    vmin=vmin,
    vmax=vmax,
    extent=[x_km.min(), x_km.max(), y_km.max(), y_km.min()],
)

cbar = fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
cbar.set_label("Perturbation amplitude")

title = ax.set_title("")
ax.set_xlabel("E-W distance (km)")
ax.set_ylabel("N-S distance (km)")
ax.set_aspect("equal")


def update(frame):
    im.set_data(wave_cube[frame])
    title.set_text(
        f"BLO Perturbation Image | "
        f"Frame {frame}/{nt - 1} | "
        f"t = {times_min[frame]:.1f} min"
    )
    return im, title


ani = FuncAnimation(
    fig,
    update,
    frames=nt,
    interval=1000 / FPS,
    blit=False,
)

out_mp4 = OUT_DIR / "BLO_perturbation_animation.mp4"

writer = FFMpegWriter(fps=FPS, bitrate=3000)
ani.save(out_mp4, writer=writer, dpi=DPI)

plt.close(fig)


# ==================================================
# Save first frame with directional keogram cuts
# Only show every 15 degrees, but still create keograms every 5 degrees later
# ==================================================
fig_first, ax_first = plt.subplots(figsize=(7, 6))

im_first = ax_first.imshow(
    wave_cube[0],
    origin="upper",
    cmap=CMAP,
    vmin=vmin,
    vmax=vmax,
    extent=[x_km.min(), x_km.max(), y_km.max(), y_km.min()],
)

cbar_first = fig_first.colorbar(im_first, ax=ax_first, fraction=0.046, pad=0.04)
cbar_first.set_label("Perturbation amplitude")

cx_km = 0.5 * (NX - 1) * DX_KM
cy_km = 0.5 * (NY - 1) * DY_KM
half_length_km = 0.5 * min((NX - 1) * DX_KM, (NY - 1) * DY_KM)

for angle_deg in PLOT_CUT_ANGLES_DEG:
    theta_cut = np.deg2rad(angle_deg)

    dx_line = half_length_km * np.cos(theta_cut)
    dy_line = half_length_km * np.sin(theta_cut)

    x_line = [cx_km - dx_line, cx_km + dx_line]
    y_line = [cy_km - dy_line, cy_km + dy_line]

    ax_first.plot(
        x_line,
        y_line,
        linewidth=1.4,
        alpha=0.85,
        label=f"{angle_deg:.0f}°",
    )

ax_first.set_title("First Perturbation Frame with Directional Keogram Cuts")
ax_first.set_xlabel("E-W distance (km)")
ax_first.set_ylabel("N-S distance (km)")
ax_first.set_aspect("equal")

ax_first.legend(
    loc="upper right",
    fontsize=7,
    ncol=2,
    framealpha=0.8,
)

plt.tight_layout()

out_first = OUT_DIR / "first_perturbation_frame_with_keogram_cuts_every_15deg.png"
fig_first.savefig(out_first, dpi=DPI, bbox_inches="tight")
plt.close(fig_first)


# ==================================================
# Directional keogram helper
# ==================================================
def make_directional_keogram(wave_cube, angle_deg, width_pixels=3):
    """
    angle_deg:
        0 deg  = E-W slice
        90 deg = N-S slice
    """

    nt, NY, NX = wave_cube.shape

    theta = np.deg2rad(angle_deg)

    ux = np.cos(theta)
    uy = np.sin(theta)

    px = -np.sin(theta)
    py = np.cos(theta)

    cx = (NX - 1) / 2
    cy = (NY - 1) / 2

    half_length_pix = min(NX, NY) // 2
    distances_pix = np.arange(-half_length_pix, half_length_pix + 1)

    offsets = np.arange(width_pixels) - width_pixels // 2

    keogram = np.full((nt, len(distances_pix)), np.nan, dtype=float)

    for it in range(nt):
        sampled_slices = []

        for off in offsets:
            x_sample = cx + distances_pix * ux + off * px
            y_sample = cy + distances_pix * uy + off * py

            valid = (
                (x_sample >= 0) & (x_sample <= NX - 1) &
                (y_sample >= 0) & (y_sample <= NY - 1)
            )

            values = np.full(len(distances_pix), np.nan)

            values[valid] = map_coordinates(
                wave_cube[it],
                [y_sample[valid], x_sample[valid]],
                order=1,
                mode="nearest",
            )

            sampled_slices.append(values)

        keogram[it] = np.nanmean(sampled_slices, axis=0)

    distance_km = distances_pix * DX_KM

    return keogram, distance_km


# ==================================================
# 2D FFT helper for keogram slope
# ==================================================
def estimate_keogram_fft_slope(keogram, exclude_radius_pixels=5):
    """
    Estimate dominant keogram orientation using 2D FFT.

    keogram shape:
        axis 0 = time pixels
        axis 1 = distance pixels

    Returns:
        slope_distance_pixel_per_time_pixel

    Interpretation:
        slope > 0 means feature tilts upward to the right
        slope < 0 means feature tilts downward to the right
    """

    Z = np.asarray(keogram, dtype=float)

    mean_val = np.nanmean(Z)
    Z = np.where(np.isfinite(Z), Z, mean_val)

    # Remove mean
    Z = Z - np.mean(Z)

    nt, nd = Z.shape

    # Apply 2D Hann window to reduce edge artifacts
    win_t = np.hanning(nt)
    win_d = np.hanning(nd)
    window = win_t[:, None] * win_d[None, :]
    Z_win = Z * window

    # 2D FFT
    F = np.fft.fftshift(np.fft.fft2(Z_win))
    power = np.abs(F) ** 2

    # Frequency axes in cycles/pixel
    freq_t = np.fft.fftshift(np.fft.fftfreq(nt))
    freq_d = np.fft.fftshift(np.fft.fftfreq(nd))

    center_t = nt // 2
    center_d = nd // 2

    # Exclude DC and very low frequencies
    yy, xx = np.indices(power.shape)
    rr = np.sqrt((yy - center_t) ** 2 + (xx - center_d) ** 2)

    power_masked = power.copy()
    power_masked[rr <= exclude_radius_pixels] = 0

    # Find strongest FFT peak
    peak_t_idx, peak_d_idx = np.unravel_index(
        np.argmax(power_masked),
        power_masked.shape,
    )

    ft = freq_t[peak_t_idx]
    fd = freq_d[peak_d_idx]

    peak_power = power_masked[peak_t_idx, peak_d_idx]

    # Constant phase lines:
    # ft*time_pixel + fd*distance_pixel = constant
    #
    # distance_pixel / time_pixel = -ft / fd
    if np.isclose(fd, 0):
        slope_distance_per_time_pixel = np.inf
        orientation_angle_deg = 90.0
    else:
        slope_distance_per_time_pixel = -ft / fd
        orientation_angle_deg = np.rad2deg(
            np.arctan2(slope_distance_per_time_pixel, 1.0)
        )

    return {
        "peak_t_idx": peak_t_idx,
        "peak_d_idx": peak_d_idx,
        "freq_time_cycles_per_pixel": ft,
        "freq_distance_cycles_per_pixel": fd,
        "slope_distance_pixel_per_time_pixel": slope_distance_per_time_pixel,
        "orientation_angle_deg": orientation_angle_deg,
        "peak_power": peak_power,
        "power_spectrum": power,
        "power_masked": power_masked,
        "freq_t": freq_t,
        "freq_d": freq_d,
    }


# ==================================================
# Create and save directional keograms
# Also estimate dominant keogram slope using 2D FFT
# ==================================================
saved_keograms = []
saved_fft_images = []
fft_summary_rows = []

for angle_deg in KEOGRAM_ANGLES_DEG:
    keogram, distance_km = make_directional_keogram(
        wave_cube,
        angle_deg,
        width_pixels=KEOGRAM_WIDTH_PIXELS,
    )

    # ------------------------------
    # Save keogram plot
    # ------------------------------
    fig_k, ax_k = plt.subplots(figsize=(9, 5))

    im_k = ax_k.imshow(
        keogram.T,
        origin="lower",
        aspect="auto",
        cmap=CMAP,
        vmin=vmin,
        vmax=vmax,
        extent=[
            times_min.min(),
            times_min.max(),
            distance_km.min(),
            distance_km.max(),
        ],
    )

    cbar_k = fig_k.colorbar(im_k, ax=ax_k, fraction=0.046, pad=0.04)
    cbar_k.set_label("Perturbation amplitude")

    ax_k.set_title(f"BLO Directional Keogram: {angle_deg:.0f}°")
    ax_k.set_xlabel("Time (min)")
    ax_k.set_ylabel("Distance along slice (km)")

    ax_k.set_xlim(0, KEOGRAM_TIME_MIN_MAX)
    ax_k.set_ylim(KEOGRAM_DISTANCE_MIN, KEOGRAM_DISTANCE_MAX)

    plt.tight_layout()

    out_png = OUT_DIR / f"BLO_keogram_angle_{angle_deg:03.0f}.png"
    out_npy = OUT_DIR / f"BLO_keogram_angle_{angle_deg:03.0f}.npy"

    fig_k.savefig(out_png, dpi=DPI, bbox_inches="tight")
    plt.close(fig_k)

    np.save(out_npy, keogram)
    saved_keograms.append(out_png)

    # ------------------------------
    # 2D FFT slope estimate
    # ------------------------------
    fft_result = estimate_keogram_fft_slope(
        keogram,
        exclude_radius_pixels=FFT_EXCLUDE_RADIUS_PIXELS,
    )

    fft_summary_rows.append({
        "angle_deg": float(angle_deg),
        "peak_t_idx": int(fft_result["peak_t_idx"]),
        "peak_d_idx": int(fft_result["peak_d_idx"]),
        "freq_time_cycles_per_pixel": float(fft_result["freq_time_cycles_per_pixel"]),
        "freq_distance_cycles_per_pixel": float(fft_result["freq_distance_cycles_per_pixel"]),
        "slope_distance_pixel_per_time_pixel": float(
            fft_result["slope_distance_pixel_per_time_pixel"]
        ),
        "orientation_angle_deg": float(fft_result["orientation_angle_deg"]),
        "peak_power": float(fft_result["peak_power"]),
    })

    # ------------------------------
    # Save FFT image
    # ------------------------------
    if SAVE_FFT_IMAGES:
        power_plot = np.log10(fft_result["power_spectrum"] + 1)

        fig_fft, ax_fft = plt.subplots(figsize=(6, 5))

        im_fft = ax_fft.imshow(
            power_plot,
            origin="lower",
            aspect="auto",
            cmap="gray",
        )

        ax_fft.plot(
            fft_result["peak_d_idx"],
            fft_result["peak_t_idx"],
            "ro",
            markersize=5,
        )

        ax_fft.set_title(
            f"2D FFT Power: {angle_deg:.0f}°\n"
            f"slope = {fft_result['slope_distance_pixel_per_time_pixel']:.3f} pix/pix"
        )
        ax_fft.set_xlabel("Distance-frequency index")
        ax_fft.set_ylabel("Time-frequency index")

        cbar_fft = fig_fft.colorbar(im_fft, ax=ax_fft, fraction=0.046, pad=0.04)
        cbar_fft.set_label("log10(power + 1)")

        plt.tight_layout()

        out_fft_png = OUT_DIR / f"BLO_keogram_angle_{angle_deg:03.0f}_fft.png"
        fig_fft.savefig(out_fft_png, dpi=DPI, bbox_inches="tight")
        plt.close(fig_fft)

        saved_fft_images.append(out_fft_png)


# ==================================================
# Save FFT slope summary CSV
# ==================================================
fft_csv = OUT_DIR / "keogram_fft_slope_summary.csv"

with open(fft_csv, "w", newline="") as f:
    writer = csv.DictWriter(
        f,
        fieldnames=[
            "angle_deg",
            "peak_t_idx",
            "peak_d_idx",
            "freq_time_cycles_per_pixel",
            "freq_distance_cycles_per_pixel",
            "slope_distance_pixel_per_time_pixel",
            "orientation_angle_deg",
            "peak_power",
        ],
    )
    writer.writeheader()
    writer.writerows(fft_summary_rows)


# ==================================================
# Save arrays and metadata
# ==================================================
np.save(OUT_DIR / "times_s.npy", times_s)
np.save(OUT_DIR / "x_km.npy", x_km)
np.save(OUT_DIR / "y_km.npy", y_km)
np.save(OUT_DIR / "keogram_angles_deg.npy", KEOGRAM_ANGLES_DEG)
np.save(OUT_DIR / "first_frame_cut_angles_deg.npy", PLOT_CUT_ANGLES_DEG)

with open(OUT_DIR / "Date_and_Start_Time.txt", "w") as f:
    f.write(f"Date: {date_start[0]}\n")
    f.write(f"Start time: {date_start[1]}\n")


# ==================================================
# Save final frame
# ==================================================
fig_last, ax_last = plt.subplots(figsize=(7, 6))

im_last = ax_last.imshow(
    wave_cube[-1],
    origin="upper",
    cmap=CMAP,
    vmin=vmin,
    vmax=vmax,
    extent=[x_km.min(), x_km.max(), y_km.max(), y_km.min()],
)

cbar_last = fig_last.colorbar(im_last, ax=ax_last, fraction=0.046, pad=0.04)
cbar_last.set_label("Perturbation amplitude")

ax_last.set_title("Final BLO Perturbation Frame")
ax_last.set_xlabel("E-W distance (km)")
ax_last.set_ylabel("N-S distance (km)")
ax_last.set_aspect("equal")

plt.tight_layout()

out_last = OUT_DIR / "final_perturbation_frame.png"
fig_last.savefig(out_last, dpi=DPI, bbox_inches="tight")
plt.close(fig_last)


# ==================================================
# Print outputs
# ==================================================
print("Saved:")
print(out_mp4)
print(out_first)
print(out_last)

print("\nDirectional keograms:")
for f in saved_keograms:
    print(f)

print("\nFFT slope summary saved:")
print(fft_csv)

if SAVE_FFT_IMAGES:
    print("\nFFT images:")
    for f in saved_fft_images:
        print(f)

print(f"\nNumber of directional keograms made: {len(KEOGRAM_ANGLES_DEG)}")
print(f"First-frame cut lines shown: every 15 degrees, {len(PLOT_CUT_ANGLES_DEG)} lines")
print("Keogram plot x-limit: 0 to 300 min")
print("Keogram plot y-limit: -250 to 250 km")
print("Output folder:", OUT_DIR)
```

## Overlay
Add another first row figure where each selected vertical profile is converted into an equivalent time span of 25 minutes, then overlaid at its correct center time: 0, 100, 150, and 200 min.

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from scipy.ndimage import gaussian_filter


# ==================================================
# Settings
# ==================================================
BASE_OUT_DIR = Path("BLO_perturbation_keogram_outputs_new")

ANGLE_DEG = 90

KEOGRAM_TIME_MIN_MAX = 300

# Dynamic distance range
KEOGRAM_DISTANCE_MIN = 0
KEOGRAM_DISTANCE_MAX = 250

# Additional row to plot
MIDDLE_DISTANCE_KM = 125

PROFILE_TOTAL_TIME_MIN = 15.0

TIMES_TO_PLOT = [
    0,
    50,
    100,
    125,
    150,
    175,
    200,
    250,
]

DPI = 200
CMAP = "RdBu_r"


# ==================================================
# Optional 2D keogram smoothing
# ==================================================
APPLY_2D_SMOOTHING = True

# Gaussian smoothing widths in pixels.
#
# The keogram shape is:
#     axis 0 = time
#     axis 1 = distance
#
# Larger values produce stronger smoothing.
SMOOTH_SIGMA_TIME_PIXELS = 1.5
SMOOTH_SIGMA_DISTANCE_PIXELS = 2.0

# Gaussian boundary handling:
# "nearest", "reflect", "mirror", "wrap", or "constant"
SMOOTH_MODE = "nearest"

# Prevent division by values that are effectively zero
SMOOTH_WEIGHT_TOLERANCE = 1e-12


# ==================================================
# Helper function for filename-safe distance labels
# ==================================================
def distance_filename_label(distance):
    """
    Convert a distance into a filename-safe label.

    Examples:
        -100 -> minus100km
           0 -> 0km
         125 -> plus125km
         250 -> plus250km
    """
    distance_int = int(round(distance))

    if distance_int < 0:
        return f"minus{abs(distance_int)}km"
    elif distance_int > 0:
        return f"plus{distance_int}km"
    else:
        return "0km"


# ==================================================
# NaN-aware 2D Gaussian smoothing
# ==================================================
def smooth_keogram_2d(
    data,
    sigma_time,
    sigma_distance,
    mode="nearest",
    weight_tolerance=1e-12,
):
    """
    Apply NaN-aware 2D Gaussian smoothing to a keogram.

    Parameters
    ----------
    data : ndarray
        Two-dimensional keogram with shape:
            time x distance

    sigma_time : float
        Gaussian smoothing width along the time axis,
        measured in pixels.

    sigma_distance : float
        Gaussian smoothing width along the distance axis,
        measured in pixels.

    mode : str
        Boundary handling mode passed to scipy.ndimage.gaussian_filter.

    weight_tolerance : float
        Minimum smoothed weight required for a valid output value.

    Returns
    -------
    smoothed : ndarray
        Smoothed keogram with the same shape as the input.
    """
    data = np.asarray(
        data,
        dtype=float,
    )

    if data.ndim != 2:
        raise ValueError(
            "The keogram must be a two-dimensional array."
        )

    if sigma_time < 0:
        raise ValueError(
            "sigma_time must be greater than or equal to zero."
        )

    if sigma_distance < 0:
        raise ValueError(
            "sigma_distance must be greater than or equal to zero."
        )

    # If both sigma values are zero, return an unchanged copy.
    if sigma_time == 0 and sigma_distance == 0:
        return data.copy()

    valid_mask = np.isfinite(
        data
    )

    # Replace NaNs and infinities with zero before filtering.
    data_filled = np.where(
        valid_mask,
        data,
        0.0,
    )

    # The weights identify locations containing valid input data.
    weights = valid_mask.astype(
        float
    )

    sigma = (
        sigma_time,
        sigma_distance,
    )

    smoothed_data = gaussian_filter(
        data_filled,
        sigma=sigma,
        mode=mode,
    )

    smoothed_weights = gaussian_filter(
        weights,
        sigma=sigma,
        mode=mode,
    )

    smoothed = np.full(
        data.shape,
        np.nan,
        dtype=float,
    )

    valid_output = (
        smoothed_weights
        > weight_tolerance
    )

    smoothed[valid_output] = (
        smoothed_data[valid_output]
        / smoothed_weights[valid_output]
    )

    return smoothed


# ==================================================
# Output folder for selected range
# ==================================================
min_distance_label = distance_filename_label(
    KEOGRAM_DISTANCE_MIN
)

middle_distance_label = distance_filename_label(
    MIDDLE_DISTANCE_KM
)

max_distance_label = distance_filename_label(
    KEOGRAM_DISTANCE_MAX
)

range_folder = (
    f"range_"
    f"{min_distance_label}_"
    f"to_"
    f"{max_distance_label}"
)

if APPLY_2D_SMOOTHING:
    smoothing_folder = (
        f"smoothed_"
        f"time_sigma_{SMOOTH_SIGMA_TIME_PIXELS:g}_"
        f"distance_sigma_{SMOOTH_SIGMA_DISTANCE_PIXELS:g}"
    )
else:
    smoothing_folder = "unsmoothed"

OUT_DIR = (
    BASE_OUT_DIR
    / range_folder
    / smoothing_folder
)

OUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


# ==================================================
# Input files
# ==================================================
KEOGRAM_FILE = (
    BASE_OUT_DIR
    / f"BLO_keogram_angle_{ANGLE_DEG:03d}.npy"
)

TIMES_FILE = (
    BASE_OUT_DIR
    / "times_s.npy"
)


# ==================================================
# Read keogram
# ==================================================
keogram_original = np.load(
    KEOGRAM_FILE
).astype(float)

times_s = np.load(
    TIMES_FILE
)

times_min = times_s / 60.0


# ==================================================
# Optionally smooth the complete 2D keogram
# ==================================================
if APPLY_2D_SMOOTHING:
    print(
        "Applying NaN-aware 2D Gaussian smoothing..."
    )

    print(
        f"Time-axis sigma: "
        f"{SMOOTH_SIGMA_TIME_PIXELS:g} pixels"
    )

    print(
        f"Distance-axis sigma: "
        f"{SMOOTH_SIGMA_DISTANCE_PIXELS:g} pixels"
    )

    keogram = smooth_keogram_2d(
        data=keogram_original,
        sigma_time=SMOOTH_SIGMA_TIME_PIXELS,
        sigma_distance=SMOOTH_SIGMA_DISTANCE_PIXELS,
        mode=SMOOTH_MODE,
        weight_tolerance=SMOOTH_WEIGHT_TOLERANCE,
    )

    processing_label = (
        f"2D Gaussian Smoothed, "
        f"σt={SMOOTH_SIGMA_TIME_PIXELS:g} px, "
        f"σd={SMOOTH_SIGMA_DISTANCE_PIXELS:g} px"
    )

    filename_processing_label = (
        f"smoothed_"
        f"st{SMOOTH_SIGMA_TIME_PIXELS:g}_"
        f"sd{SMOOTH_SIGMA_DISTANCE_PIXELS:g}"
    )

else:
    print(
        "2D smoothing is disabled. "
        "Using the original keogram."
    )

    keogram = keogram_original.copy()

    processing_label = "Unsmoothed"

    filename_processing_label = "unsmoothed"


# ==================================================
# Validate keogram and time dimensions
# ==================================================
if keogram.ndim != 2:
    raise ValueError(
        f"Expected a 2D keogram, but the loaded array "
        f"has shape {keogram.shape}."
    )

nt, nspace = keogram.shape

if len(times_min) != nt:
    raise ValueError(
        f"The keogram contains {nt} time samples, "
        f"but times_s.npy contains {len(times_min)} samples."
    )


# ==================================================
# Construct distance axis
# ==================================================
distance_km = np.arange(
    -(nspace // 2),
    nspace // 2 + 1,
)

if len(distance_km) != nspace:
    distance_km = (
        np.arange(nspace)
        - nspace // 2
    )


# ==================================================
# Validate requested distance range
# ==================================================
available_distance_min = np.nanmin(
    distance_km
)

available_distance_max = np.nanmax(
    distance_km
)

if KEOGRAM_DISTANCE_MIN >= KEOGRAM_DISTANCE_MAX:
    raise ValueError(
        "KEOGRAM_DISTANCE_MIN must be smaller than "
        "KEOGRAM_DISTANCE_MAX."
    )

if KEOGRAM_DISTANCE_MIN < available_distance_min:
    raise ValueError(
        f"Requested minimum distance is "
        f"{KEOGRAM_DISTANCE_MIN} km, but the keogram "
        f"only extends to {available_distance_min} km."
    )

if KEOGRAM_DISTANCE_MAX > available_distance_max:
    raise ValueError(
        f"Requested maximum distance is "
        f"{KEOGRAM_DISTANCE_MAX} km, but the keogram "
        f"only extends to {available_distance_max} km."
    )

if not (
    KEOGRAM_DISTANCE_MIN
    <= MIDDLE_DISTANCE_KM
    <= KEOGRAM_DISTANCE_MAX
):
    raise ValueError(
        f"MIDDLE_DISTANCE_KM = {MIDDLE_DISTANCE_KM} km "
        f"must be inside the selected distance range "
        f"{KEOGRAM_DISTANCE_MIN} to "
        f"{KEOGRAM_DISTANCE_MAX} km."
    )


# ==================================================
# Select rows at minimum, 0 km, middle,
# and maximum distances
# ==================================================
idx_min = np.argmin(
    np.abs(
        distance_km
        - KEOGRAM_DISTANCE_MIN
    )
)

idx_0km = np.argmin(
    np.abs(
        distance_km
        - 0
    )
)

idx_middle = np.argmin(
    np.abs(
        distance_km
        - MIDDLE_DISTANCE_KM
    )
)

idx_max = np.argmin(
    np.abs(
        distance_km
        - KEOGRAM_DISTANCE_MAX
    )
)

actual_distance_min = distance_km[
    idx_min
]

actual_distance_0 = distance_km[
    idx_0km
]

actual_distance_middle = distance_km[
    idx_middle
]

actual_distance_max = distance_km[
    idx_max
]

row_min = keogram[
    :,
    idx_min,
]

row_0km = keogram[
    :,
    idx_0km,
]

row_middle = keogram[
    :,
    idx_middle,
]

row_max = keogram[
    :,
    idx_max,
]


# ==================================================
# Select time columns
# ==================================================
column_indices = []
columns = []

for requested_time in TIMES_TO_PLOT:
    idx = np.argmin(
        np.abs(
            times_min
            - requested_time
        )
    )

    column_indices.append(
        idx
    )

    columns.append(
        keogram[idx, :]
    )


# ==================================================
# Print selected rows and columns
# ==================================================
print("\nKeogram processing:")
print(processing_label)

print("\nSelected rows:")

print(
    f"{KEOGRAM_DISTANCE_MIN:7.1f} km requested -> "
    f"row index {idx_min}, "
    f"actual distance = {actual_distance_min:.1f} km"
)

print(
    f"{0:7.1f} km requested -> "
    f"row index {idx_0km}, "
    f"actual distance = {actual_distance_0:.1f} km"
)

print(
    f"{MIDDLE_DISTANCE_KM:7.1f} km requested -> "
    f"row index {idx_middle}, "
    f"actual distance = {actual_distance_middle:.1f} km"
)

print(
    f"{KEOGRAM_DISTANCE_MAX:7.1f} km requested -> "
    f"row index {idx_max}, "
    f"actual distance = {actual_distance_max:.1f} km"
)

print("\nSelected columns:")

for requested_time, idx in zip(
    TIMES_TO_PLOT,
    column_indices,
):
    print(
        f"{requested_time:6.1f} min -> "
        f"frame {idx}, "
        f"actual time = {times_min[idx]:.2f} min"
    )


# ==================================================
# Color limits
# ==================================================
finite_keogram_values = np.abs(
    keogram[np.isfinite(keogram)]
)

if finite_keogram_values.size > 0:
    vmax = np.nanpercentile(
        finite_keogram_values,
        99,
    )
else:
    vmax = 1.0

if not np.isfinite(vmax) or vmax == 0:
    vmax = 1.0

vmin = -vmax

time_colors = [
    "black",
    "blue",
    "green",
    "purple",
    "orange",
    "pink",
    "brown",
    "magenta",
]

if len(TIMES_TO_PLOT) > len(time_colors):
    raise ValueError(
        "There are more TIMES_TO_PLOT entries than colors "
        "in time_colors."
    )


# ==================================================
# Plot keogram with selected rows and time columns
# ==================================================
fig, ax = plt.subplots(
    figsize=(9, 5)
)

im = ax.imshow(
    keogram.T,
    origin="lower",
    aspect="auto",
    cmap=CMAP,
    vmin=vmin,
    vmax=vmax,
    extent=[
        times_min.min(),
        times_min.max(),
        distance_km.min(),
        distance_km.max(),
    ],
)

cbar = fig.colorbar(
    im,
    ax=ax,
    fraction=0.046,
    pad=0.04,
)

cbar.set_label(
    "Perturbation amplitude"
)

ax.axhline(
    actual_distance_0,
    color="black",
    linewidth=2.0,
    label=f"{actual_distance_0:.0f} km",
)

ax.axhline(
    actual_distance_middle,
    color="blue",
    linewidth=2.0,
    label=f"{actual_distance_middle:.0f} km",
)

ax.axhline(
    actual_distance_max,
    color="red",
    linewidth=2.0,
    label=f"{actual_distance_max:.0f} km",
)

for requested_time, idx, color in zip(
    TIMES_TO_PLOT,
    column_indices,
    time_colors,
):
    ax.axvline(
        times_min[idx],
        color=color,
        linewidth=1.8,
        linestyle=":",
        label=f"{requested_time:.0f} min",
    )

ax.set_xlim(
    0,
    KEOGRAM_TIME_MIN_MAX,
)

ax.set_ylim(
    KEOGRAM_DISTANCE_MIN,
    KEOGRAM_DISTANCE_MAX,
)

ax.set_title(
    f"{ANGLE_DEG}° BLO Directional Keogram\n"
    f"{processing_label}"
)

ax.set_xlabel(
    "Time (min)"
)

ax.set_ylabel(
    "Distance along slice (km)"
)

ax.legend(
    loc="upper right",
    fontsize=8,
)

plt.tight_layout()

out_keogram_plot = (
    OUT_DIR
    / (
        f"BLO_keogram_{ANGLE_DEG:03d}_"
        f"range_{min_distance_label}_to_{max_distance_label}_"
        f"{filename_processing_label}_"
        f"with_selected_rows_and_columns.png"
    )
)

fig.savefig(
    out_keogram_plot,
    dpi=DPI,
    bbox_inches="tight",
)

plt.close(fig)


# ==================================================
# Plot minimum-distance row
# ==================================================
fig_min, ax_min = plt.subplots(
    figsize=(8, 4)
)

ax_min.plot(
    times_min,
    row_min,
    linewidth=1.5,
)

ax_min.set_xlim(
    0,
    KEOGRAM_TIME_MIN_MAX,
)

ax_min.set_title(
    f"{actual_distance_min:.0f} km Row of "
    f"{ANGLE_DEG}° BLO Keogram\n"
    f"{processing_label}"
)

ax_min.set_xlabel(
    "Time (min)"
)

ax_min.set_ylabel(
    "Perturbation amplitude"
)

ax_min.grid(
    True,
    alpha=0.3,
)

plt.tight_layout()

out_min = (
    OUT_DIR
    / (
        f"BLO_keogram_{ANGLE_DEG:03d}_"
        f"{min_distance_label}_row_"
        f"{filename_processing_label}.png"
    )
)

fig_min.savefig(
    out_min,
    dpi=DPI,
    bbox_inches="tight",
)

plt.close(fig_min)


# ==================================================
# Overlay selected profiles onto minimum-distance row
# ==================================================
distance_mask = (
    (distance_km >= KEOGRAM_DISTANCE_MIN)
    & (distance_km <= KEOGRAM_DISTANCE_MAX)
)

distance_for_overlay = distance_km[
    distance_mask
]

# Normalize selected distance range:
#
# KEOGRAM_DISTANCE_MIN -> 0
# KEOGRAM_DISTANCE_MAX -> 1
distance_norm = (
    distance_for_overlay
    - KEOGRAM_DISTANCE_MIN
) / (
    KEOGRAM_DISTANCE_MAX
    - KEOGRAM_DISTANCE_MIN
)

zero_distance_fraction = (
    0.0
    - KEOGRAM_DISTANCE_MIN
) / (
    KEOGRAM_DISTANCE_MAX
    - KEOGRAM_DISTANCE_MIN
)

zero_distance_time_offset = (
    zero_distance_fraction
    * PROFILE_TOTAL_TIME_MIN
)

middle_distance_fraction = (
    MIDDLE_DISTANCE_KM
    - KEOGRAM_DISTANCE_MIN
) / (
    KEOGRAM_DISTANCE_MAX
    - KEOGRAM_DISTANCE_MIN
)

middle_distance_time_offset = (
    middle_distance_fraction
    * PROFILE_TOTAL_TIME_MIN
)

fig_min_overlay, ax_min_overlay = plt.subplots(
    figsize=(9, 4)
)

ax_min_overlay.plot(
    times_min,
    row_min,
    color="black",
    linewidth=1.5,
    label=f"{actual_distance_min:.0f} km row",
)

for requested_time, idx, col, color in zip(
    TIMES_TO_PLOT,
    column_indices,
    columns,
    time_colors,
):
    profile_time = (
        times_min[idx]
        + distance_norm
        * PROFILE_TOTAL_TIME_MIN
    )

    profile_amp = col[
        distance_mask
    ]

    ax_min_overlay.plot(
        profile_time,
        profile_amp,
        color=color,
        linewidth=1.5,
        alpha=0.8,
        label=(
            f"profile starting at "
            f"{requested_time:.0f} min"
        ),
    )

    ax_min_overlay.axvline(
        times_min[idx],
        color=color,
        linestyle=":",
        linewidth=1.2,
        alpha=0.8,
    )

ax_min_overlay.set_xlim(
    0,
    KEOGRAM_TIME_MIN_MAX,
)

ax_min_overlay.set_title(
    f"{actual_distance_min:.0f} km Row of "
    f"{ANGLE_DEG}° Keogram with Selected Profiles Overlaid\n"
    f"{processing_label}; "
    f"{KEOGRAM_DISTANCE_MIN:.0f} to "
    f"{KEOGRAM_DISTANCE_MAX:.0f} km mapped to "
    f"{PROFILE_TOTAL_TIME_MIN:.0f} min"
)

ax_min_overlay.set_xlabel(
    "Time (min)"
)

ax_min_overlay.set_ylabel(
    "Perturbation amplitude"
)

ax_min_overlay.grid(
    True,
    alpha=0.3,
)

ax_min_overlay.legend(
    loc="best",
    fontsize=8,
)

plt.tight_layout()

out_min_overlay = (
    OUT_DIR
    / (
        f"BLO_keogram_{ANGLE_DEG:03d}_"
        f"{min_distance_label}_row_with_profiles_"
        f"range_{min_distance_label}_to_{max_distance_label}_"
        f"{PROFILE_TOTAL_TIME_MIN:.0f}min_mapping_"
        f"{filename_processing_label}.png"
    )
)

fig_min_overlay.savefig(
    out_min_overlay,
    dpi=DPI,
    bbox_inches="tight",
)

plt.close(fig_min_overlay)


# ==================================================
# Plot 0-km, 125-km, and 250-km rows together
# ==================================================
fig_rows, ax_rows = plt.subplots(
    figsize=(9, 4)
)

ax_rows.plot(
    times_min,
    row_0km,
    color="black",
    linewidth=1.5,
    label=f"{actual_distance_0:.0f} km",
)

ax_rows.plot(
    times_min,
    row_middle,
    color="blue",
    linewidth=1.5,
    label=f"{actual_distance_middle:.0f} km",
)

ax_rows.plot(
    times_min,
    row_max,
    color="red",
    linewidth=1.5,
    label=f"{actual_distance_max:.0f} km",
)

ax_rows.set_xlim(
    0,
    KEOGRAM_TIME_MIN_MAX,
)

ax_rows.set_title(
    f"{actual_distance_0:.0f} km, "
    f"{actual_distance_middle:.0f} km, and "
    f"{actual_distance_max:.0f} km Rows of "
    f"{ANGLE_DEG}° BLO Keogram\n"
    f"{processing_label}"
)

ax_rows.set_xlabel(
    "Time (min)"
)

ax_rows.set_ylabel(
    "Perturbation amplitude"
)

ax_rows.grid(
    True,
    alpha=0.3,
)

ax_rows.legend(
    loc="best",
    fontsize=9,
)

plt.tight_layout()

out_rows_combined = (
    OUT_DIR
    / (
        f"BLO_keogram_{ANGLE_DEG:03d}_"
        f"{distance_filename_label(actual_distance_0)}_"
        f"{distance_filename_label(actual_distance_middle)}_and_"
        f"{distance_filename_label(actual_distance_max)}_rows_"
        f"{filename_processing_label}.png"
    )
)

fig_rows.savefig(
    out_rows_combined,
    dpi=DPI,
    bbox_inches="tight",
)

plt.close(fig_rows)


# ==================================================
# Plot selected time columns/profiles
# ==================================================
fig_c, ax_c = plt.subplots(
    figsize=(6, 7)
)

for requested_time, idx, col, color in zip(
    TIMES_TO_PLOT,
    column_indices,
    columns,
    time_colors,
):
    ax_c.plot(
        col,
        distance_km,
        linewidth=1.5,
        color=color,
        label=f"{requested_time:.0f} min",
    )

ax_c.axhline(
    actual_distance_0,
    color="black",
    linestyle="--",
    linewidth=1.2,
    label=f"{actual_distance_0:.0f} km",
)

ax_c.axhline(
    actual_distance_middle,
    color="blue",
    linestyle="--",
    linewidth=1.2,
    label=f"{actual_distance_middle:.0f} km",
)

ax_c.axhline(
    actual_distance_max,
    color="red",
    linestyle="--",
    linewidth=1.2,
    label=f"{actual_distance_max:.0f} km",
)

ax_c.set_ylim(
    KEOGRAM_DISTANCE_MIN,
    KEOGRAM_DISTANCE_MAX,
)

ax_c.set_title(
    f"Profiles Along {ANGLE_DEG}° Slice "
    f"at Selected Times\n"
    f"{processing_label}"
)

ax_c.set_xlabel(
    "Perturbation amplitude"
)

ax_c.set_ylabel(
    "Distance along slice (km)"
)

ax_c.grid(
    True,
    alpha=0.3,
)

ax_c.legend(
    loc="best",
    fontsize=8,
)

plt.tight_layout()

time_label = "_".join(
    [
        f"{int(t)}"
        for t in TIMES_TO_PLOT
    ]
)

out_columns = (
    OUT_DIR
    / (
        f"BLO_keogram_{ANGLE_DEG:03d}_"
        f"profiles_{time_label}min_"
        f"range_{min_distance_label}_to_{max_distance_label}_"
        f"{filename_processing_label}.png"
    )
)

fig_c.savefig(
    out_columns,
    dpi=DPI,
    bbox_inches="tight",
)

plt.close(fig_c)


# ==================================================
# Optionally save the processed keogram array
# ==================================================
if APPLY_2D_SMOOTHING:
    out_smoothed_keogram = (
        OUT_DIR
        / (
            f"BLO_keogram_angle_{ANGLE_DEG:03d}_"
            f"{filename_processing_label}.npy"
        )
    )

    np.save(
        out_smoothed_keogram,
        keogram,
    )
else:
    out_smoothed_keogram = None


# ==================================================
# Print outputs
# ==================================================
print("\nSaved:")
print(out_keogram_plot)
print(out_min)
print(out_min_overlay)
print(out_rows_combined)
print(out_columns)

if out_smoothed_keogram is not None:
    print(out_smoothed_keogram)

print("\nKeogram shape:")
print(
    f"time x distance = {keogram.shape}"
)

print("\nProcessing:")
print(processing_label)

print("\nExtracted row arrays:")

print(
    f"row_min shape = {row_min.shape}"
)

print(
    f"row_0km shape = {row_0km.shape}"
)

print(
    f"row_middle shape = {row_middle.shape}"
)

print(
    f"row_max shape = {row_max.shape}"
)

print("\nExtracted column arrays:")

for requested_time, col in zip(
    TIMES_TO_PLOT,
    columns,
):
    print(
        f"{requested_time:.0f} min "
        f"column shape = {col.shape}"
    )

print("\nPlot limits:")

print(
    f"Time range: 0 to "
    f"{KEOGRAM_TIME_MIN_MAX} min"
)

print(
    f"Distance range: "
    f"{KEOGRAM_DISTANCE_MIN} to "
    f"{KEOGRAM_DISTANCE_MAX} km"
)

print("\nOverlay mapping:")

print(
    f"{KEOGRAM_DISTANCE_MIN:.0f} km "
    f"-> selected time"
)

print(
    f"0 km -> selected time + "
    f"{zero_distance_time_offset:.2f} min"
)

print(
    f"{MIDDLE_DISTANCE_KM:.0f} km "
    f"-> selected time + "
    f"{middle_distance_time_offset:.2f} min"
)

print(
    f"{KEOGRAM_DISTANCE_MAX:.0f} km "
    f"-> selected time + "
    f"{PROFILE_TOTAL_TIME_MIN:.2f} min"
)
```

## Bias Correction

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

from scipy.ndimage import gaussian_filter, gaussian_filter1d
from scipy.interpolate import PchipInterpolator


# ==================================================
# Settings
# ==================================================
BASE_OUT_DIR = Path(
    "BLO_perturbation_keogram_outputs_new"
)

ANGLE_DEG = 45

KEOGRAM_TIME_MIN_MAX = 300

# Displayed distance range
KEOGRAM_DISTANCE_MIN = 0
KEOGRAM_DISTANCE_MAX = 250

# Additional row to plot
MIDDLE_DISTANCE_KM = 125

PROFILE_TOTAL_TIME_MIN = 10

TIMES_TO_PLOT = [
    0,
    50,
    100,
    125,
    150,
    175,
    200,
    250,
]

DPI = 200
CMAP = "RdBu_r"


# ==================================================
# Optional 2D smoothing
# ==================================================
APPLY_2D_SMOOTHING = True

# Keogram dimensions:
# axis 0 = time
# axis 1 = distance
SMOOTH_SIGMA_TIME_PIXELS = 1.5
SMOOTH_SIGMA_DISTANCE_PIXELS = 2.0

SMOOTH_MODE = "nearest"
SMOOTH_WEIGHT_TOLERANCE = 1e-12


# ==================================================
# Overlay-profile smoothing
# ==================================================
# Smooth each selected distance profile again before it is
# compressed into the overlay plot. This applies to both the
# uncorrected and corrected plot sets; therefore the corrected
# overlay uses smoothed, bias-corrected profiles.
APPLY_OVERLAY_PROFILE_SMOOTHING = True
OVERLAY_PROFILE_SMOOTH_SIGMA_PIXELS = 2.0
OVERLAY_PROFILE_SMOOTH_MODE = "nearest"


# ==================================================
# Bias-correction settings
# ==================================================
APPLY_BIAS_CORRECTION = True

# Correction values at 250 km.
#
# A shape-preserving PCHIP interpolation is used
# through these points.
## For 90 degrees:
#BIAS_TIME_POINTS_MIN = np.array(
#    [
#        0.0,
#       150.0,
#        300.0,
#    ],
#    dtype=float,
#)

#BIAS_AT_MAX_DISTANCE = np.array(
#    [
#        0.35,
#        0.05,
#        -0.02,
#    ],
#    dtype=float,
#)

## For 45 degrees:

BIAS_TIME_POINTS_MIN = np.array(
    [
        0.0,
        150.0,
        300.0,
    ],
    dtype=float,
)

BIAS_AT_MAX_DISTANCE = np.array(
    [
        .2,
        0.04,
        -0.01,
    ],
    dtype=float,
)

BIAS_DISTANCE_MIN_KM = 0.0
BIAS_DISTANCE_MAX_KM = 250.0

# Quadratic distance dependence:
#
# distance factor =
#     normalized_distance ** 2 => 1
BIAS_DISTANCE_POWER = 1.0 #2

# Set correction to zero outside the defined ranges
BIAS_ZERO_OUTSIDE_DISTANCE_RANGE = True
BIAS_ZERO_OUTSIDE_TIME_RANGE = True

# Prevent any tiny negative interpolation values
CLIP_NEGATIVE_TIME_CORRECTION = False


# ==================================================
# Helper: filename-safe numeric label
# ==================================================
def safe_number_label(value):
    """
    Convert a number into a short filename-safe label.

    Examples
    --------
    1.5 -> 1p5
    2.0 -> 2
    """
    value_float = float(value)

    if value_float.is_integer():
        return str(
            int(value_float)
        )

    return (
        f"{value_float:g}"
        .replace("-", "m")
        .replace(".", "p")
    )


# ==================================================
# Helper: filename-safe distance label
# ==================================================
def distance_filename_label(distance):
    """
    Convert distance into a filename-safe label.

    Examples
    --------
    -100 -> m100km
       0 -> 0km
     125 -> p125km
     250 -> p250km
    """
    distance_int = int(
        round(distance)
    )

    if distance_int < 0:
        return f"m{abs(distance_int)}km"

    if distance_int > 0:
        return f"p{distance_int}km"

    return "0km"


# ==================================================
# NaN-aware 2D Gaussian smoothing
# ==================================================
def smooth_keogram_2d(
    data,
    sigma_time,
    sigma_distance,
    mode="nearest",
    weight_tolerance=1e-12,
):
    """
    Apply NaN-aware Gaussian smoothing to a 2D keogram.

    Parameters
    ----------
    data : ndarray
        Keogram with shape:
            time x distance

    sigma_time : float
        Gaussian sigma along the time axis, in pixels.

    sigma_distance : float
        Gaussian sigma along the distance axis, in pixels.

    mode : str
        Boundary handling mode used by gaussian_filter.

    weight_tolerance : float
        Minimum valid-data weight for the output.

    Returns
    -------
    ndarray
        Smoothed keogram with the same shape.
    """
    data = np.asarray(
        data,
        dtype=float,
    )

    if data.ndim != 2:
        raise ValueError(
            "The keogram must be a two-dimensional array."
        )

    if sigma_time < 0:
        raise ValueError(
            "sigma_time must be nonnegative."
        )

    if sigma_distance < 0:
        raise ValueError(
            "sigma_distance must be nonnegative."
        )

    if (
        sigma_time == 0
        and sigma_distance == 0
    ):
        return data.copy()

    valid_mask = np.isfinite(
        data
    )

    data_filled = np.where(
        valid_mask,
        data,
        0.0,
    )

    weights = valid_mask.astype(
        float
    )

    sigma = (
        sigma_time,
        sigma_distance,
    )

    smoothed_values = gaussian_filter(
        data_filled,
        sigma=sigma,
        mode=mode,
    )

    smoothed_weights = gaussian_filter(
        weights,
        sigma=sigma,
        mode=mode,
    )

    smoothed = np.full(
        data.shape,
        np.nan,
        dtype=float,
    )

    valid_output = (
        smoothed_weights
        > weight_tolerance
    )

    smoothed[valid_output] = (
        smoothed_values[valid_output]
        / smoothed_weights[valid_output]
    )

    return smoothed


# ==================================================
# Create 2D bias-correction keogram
# ==================================================
def create_bias_correction_keogram(
    times_min,
    distance_km,
    bias_time_points_min,
    bias_at_max_distance,
    bias_distance_min_km,
    bias_distance_max_km,
    distance_power=2.0,
    zero_outside_distance_range=True,
    zero_outside_time_range=True,
    clip_negative_time_correction=True,
):
    """
    Create a two-dimensional additive correction field.

    Time dependence
    ---------------
    A shape-preserving PCHIP interpolation passes through:

        0 min   -> 0.40 at 250 km
        150 min -> 0.05 at 250 km
        300 min -> 0.00 at 250 km

    Distance dependence
    -------------------
    The spatial dependence is quadratic:

        distance_factor =
            normalized_distance ** distance_power

    For distance_power = 2:

        0 km   -> factor 0.00
        125 km -> factor 0.25
        250 km -> factor 1.00

    Returns
    -------
    correction_keogram : ndarray
        Correction field with shape:
            time x distance

    bias_vs_time : ndarray
        Time-dependent correction at maximum distance.

    distance_factor : ndarray
        Spatial correction multiplier.

    time_interpolator : PchipInterpolator
        Shape-preserving time interpolator.
    """
    times_min = np.asarray(
        times_min,
        dtype=float,
    )

    distance_km = np.asarray(
        distance_km,
        dtype=float,
    )

    bias_time_points_min = np.asarray(
        bias_time_points_min,
        dtype=float,
    )

    bias_at_max_distance = np.asarray(
        bias_at_max_distance,
        dtype=float,
    )

    if times_min.ndim != 1:
        raise ValueError(
            "times_min must be one-dimensional."
        )

    if distance_km.ndim != 1:
        raise ValueError(
            "distance_km must be one-dimensional."
        )

    if (
        len(bias_time_points_min)
        != len(bias_at_max_distance)
    ):
        raise ValueError(
            "BIAS_TIME_POINTS_MIN and "
            "BIAS_AT_MAX_DISTANCE must have "
            "the same length."
        )

    if len(bias_time_points_min) < 2:
        raise ValueError(
            "At least two time points are required."
        )

    if not np.all(
        np.diff(bias_time_points_min) > 0
    ):
        raise ValueError(
            "BIAS_TIME_POINTS_MIN must be "
            "strictly increasing."
        )

    if (
        bias_distance_max_km
        <= bias_distance_min_km
    ):
        raise ValueError(
            "BIAS_DISTANCE_MAX_KM must be greater "
            "than BIAS_DISTANCE_MIN_KM."
        )

    if distance_power <= 0:
        raise ValueError(
            "distance_power must be greater than zero."
        )

    # ==================================================
    # Shape-preserving time dependence
    # ==================================================
    time_interpolator = PchipInterpolator(
        bias_time_points_min,
        bias_at_max_distance,
        extrapolate=False,
    )

    bias_vs_time = time_interpolator(
        times_min
    )

    if zero_outside_time_range:
        inside_time_range = (
            (times_min >= bias_time_points_min[0])
            & (times_min <= bias_time_points_min[-1])
        )

        bias_vs_time = np.where(
            inside_time_range,
            bias_vs_time,
            0.0,
        )
    else:
        # If extrapolation is disabled, values outside the
        # range are NaN. Replace them with endpoint values.
        bias_vs_time = np.where(
            times_min < bias_time_points_min[0],
            bias_at_max_distance[0],
            bias_vs_time,
        )

        bias_vs_time = np.where(
            times_min > bias_time_points_min[-1],
            bias_at_max_distance[-1],
            bias_vs_time,
        )

    if clip_negative_time_correction:
        bias_vs_time = np.maximum(
            bias_vs_time,
            0.0,
        )

    # ==================================================
    # Quadratic distance dependence
    # ==================================================
    normalized_distance = (
        distance_km
        - bias_distance_min_km
    ) / (
        bias_distance_max_km
        - bias_distance_min_km
    )

    if zero_outside_distance_range:
        inside_distance_range = (
            (distance_km >= bias_distance_min_km)
            & (distance_km <= bias_distance_max_km)
        )

        normalized_distance = np.where(
            inside_distance_range,
            normalized_distance,
            0.0,
        )
    else:
        normalized_distance = np.clip(
            normalized_distance,
            0.0,
            1.0,
        )

    distance_factor = (
        normalized_distance ** distance_power
    )

    # ==================================================
    # Build 2D correction field
    # ==================================================
    correction_keogram = (
        bias_vs_time[:, np.newaxis]
        * distance_factor[np.newaxis, :]
    )

    return (
        correction_keogram,
        bias_vs_time,
        distance_factor,
        time_interpolator,
    )


# ==================================================
# NaN-aware 1D Gaussian smoothing
# ==================================================
def smooth_profile_1d(
    profile,
    sigma,
    mode="nearest",
    weight_tolerance=1e-12,
):
    """
    Apply NaN-aware Gaussian smoothing to one profile.
    """
    profile = np.asarray(
        profile,
        dtype=float,
    )

    if profile.ndim != 1:
        raise ValueError(
            "The profile must be one-dimensional."
        )

    if sigma < 0:
        raise ValueError(
            "sigma must be nonnegative."
        )

    if sigma == 0:
        return profile.copy()

    valid_mask = np.isfinite(
        profile
    )

    values_filled = np.where(
        valid_mask,
        profile,
        0.0,
    )

    weights = valid_mask.astype(
        float
    )

    smoothed_values = gaussian_filter1d(
        values_filled,
        sigma=sigma,
        mode=mode,
    )

    smoothed_weights = gaussian_filter1d(
        weights,
        sigma=sigma,
        mode=mode,
    )

    smoothed_profile = np.full(
        profile.shape,
        np.nan,
        dtype=float,
    )

    valid_output = (
        smoothed_weights
        > weight_tolerance
    )

    smoothed_profile[valid_output] = (
        smoothed_values[valid_output]
        / smoothed_weights[valid_output]
    )

    return smoothed_profile


# ==================================================
# Symmetric color limits
# ==================================================
def get_symmetric_color_limits(
    data,
    percentile=99.0,
):
    """
    Return symmetric color limits centered on zero.
    """
    data = np.asarray(
        data,
        dtype=float,
    )

    finite_values = np.abs(
        data[np.isfinite(data)]
    )

    if finite_values.size == 0:
        vmax = 1.0
    else:
        vmax = np.nanpercentile(
            finite_values,
            percentile,
        )

    if (
        not np.isfinite(vmax)
        or vmax == 0
    ):
        vmax = 1.0

    return -vmax, vmax


# ==================================================
# Standard plots for one keogram
# ==================================================
def create_keogram_plots(
    keogram,
    dataset_name,
    filename_label,
    out_dir,
    times_min,
    distance_km,
    row_indices,
    actual_distances,
    column_indices,
    time_colors,
):
    """
    Create the standard set of keogram plots.
    """
    out_dir = Path(
        out_dir
    )

    out_dir.mkdir(
        parents=True,
        exist_ok=True,
    )

    idx_min = row_indices["min"]
    idx_zero = row_indices["zero"]
    idx_middle = row_indices["middle"]
    idx_max = row_indices["max"]

    distance_min_actual = actual_distances["min"]
    distance_zero_actual = actual_distances["zero"]
    distance_middle_actual = actual_distances["middle"]
    distance_max_actual = actual_distances["max"]

    row_min = keogram[
        :,
        idx_min,
    ]

    row_zero = keogram[
        :,
        idx_zero,
    ]

    row_middle = keogram[
        :,
        idx_middle,
    ]

    row_max = keogram[
        :,
        idx_max,
    ]

    columns = [
        keogram[idx, :]
        for idx in column_indices
    ]

    vmin, vmax = get_symmetric_color_limits(
        keogram,
        percentile=99.0,
    )

    outputs = {}

    # ==================================================
    # 2D keogram
    # ==================================================
    fig, ax = plt.subplots(
        figsize=(9, 5)
    )

    im = ax.imshow(
        keogram.T,
        origin="lower",
        aspect="auto",
        cmap=CMAP,
        vmin=vmin,
        vmax=vmax,
        extent=[
            times_min.min(),
            times_min.max(),
            distance_km.min(),
            distance_km.max(),
        ],
    )

    cbar = fig.colorbar(
        im,
        ax=ax,
        fraction=0.046,
        pad=0.04,
    )

    cbar.set_label(
        "Perturbation amplitude"
    )

    ax.axhline(
        distance_zero_actual,
        color="black",
        linewidth=2.0,
        label=f"{distance_zero_actual:.0f} km",
    )

    ax.axhline(
        distance_middle_actual,
        color="blue",
        linewidth=2.0,
        label=f"{distance_middle_actual:.0f} km",
    )

    ax.axhline(
        distance_max_actual,
        color="red",
        linewidth=2.0,
        label=f"{distance_max_actual:.0f} km",
    )

    for requested_time, idx, color in zip(
        TIMES_TO_PLOT,
        column_indices,
        time_colors,
    ):
        ax.axvline(
            times_min[idx],
            color=color,
            linewidth=1.8,
            linestyle=":",
            label=f"{requested_time:.0f} min",
        )

    ax.set_xlim(
        0,
        KEOGRAM_TIME_MIN_MAX,
    )

    ax.set_ylim(
        KEOGRAM_DISTANCE_MIN,
        KEOGRAM_DISTANCE_MAX,
    )

    ax.set_title(
        f"{ANGLE_DEG}° BLO Directional Keogram\n"
        f"{dataset_name}"
    )

    ax.set_xlabel(
        "Time (min)"
    )

    ax.set_ylabel(
        "Distance along slice (km)"
    )

    ax.legend(
        loc="upper right",
        fontsize=8,
    )

    plt.tight_layout()

    outputs["keogram"] = (
        out_dir
        / f"keogram_{ANGLE_DEG:03d}_{filename_label}.png"
    )

    fig.savefig(
        outputs["keogram"],
        dpi=DPI,
        bbox_inches="tight",
    )

    plt.close(fig)

    # ==================================================
    # Minimum-distance row
    # ==================================================
    fig_min, ax_min = plt.subplots(
        figsize=(8, 4)
    )

    ax_min.plot(
        times_min,
        row_min,
        linewidth=1.5,
    )

    ax_min.set_xlim(
        0,
        KEOGRAM_TIME_MIN_MAX,
    )

    ax_min.set_title(
        f"{distance_min_actual:.0f} km Row\n"
        f"{dataset_name}"
    )

    ax_min.set_xlabel(
        "Time (min)"
    )

    ax_min.set_ylabel(
        "Perturbation amplitude"
    )

    ax_min.grid(
        True,
        alpha=0.3,
    )

    plt.tight_layout()

    outputs["minimum_row"] = (
        out_dir
        / f"row_min_{ANGLE_DEG:03d}_{filename_label}.png"
    )

    fig_min.savefig(
        outputs["minimum_row"],
        dpi=DPI,
        bbox_inches="tight",
    )

    plt.close(fig_min)

    # ==================================================
    # Profiles overlaid on minimum-distance row
    # ==================================================
    distance_mask = (
        (distance_km >= KEOGRAM_DISTANCE_MIN)
        & (distance_km <= KEOGRAM_DISTANCE_MAX)
    )

    distance_for_overlay = distance_km[
        distance_mask
    ]

    distance_norm = (
        distance_for_overlay
        - KEOGRAM_DISTANCE_MIN
    ) / (
        KEOGRAM_DISTANCE_MAX
        - KEOGRAM_DISTANCE_MIN
    )

    fig_overlay, ax_overlay = plt.subplots(
        figsize=(9, 4)
    )

    ax_overlay.plot(
        times_min,
        row_min,
        color="black",
        linewidth=1.5,
        label=f"{distance_min_actual:.0f} km row",
    )

    for requested_time, idx, col, color in zip(
        TIMES_TO_PLOT,
        column_indices,
        columns,
        time_colors,
    ):
        profile_time = (
            times_min[idx]
            + distance_norm
            * PROFILE_TOTAL_TIME_MIN
        )

        profile_amplitude = col[
            distance_mask
        ]

        if APPLY_OVERLAY_PROFILE_SMOOTHING:
            profile_amplitude = smooth_profile_1d(
                profile=profile_amplitude,
                sigma=(
                    OVERLAY_PROFILE_SMOOTH_SIGMA_PIXELS
                ),
                mode=OVERLAY_PROFILE_SMOOTH_MODE,
                weight_tolerance=(
                    SMOOTH_WEIGHT_TOLERANCE
                ),
            )

        ax_overlay.plot(
            profile_time,
            profile_amplitude,
            color=color,
            linewidth=1.5,
            alpha=0.8,
            label=(
                f"profile starting at "
                f"{requested_time:.0f} min"
            ),
        )

        ax_overlay.axvline(
            times_min[idx],
            color=color,
            linestyle=":",
            linewidth=1.2,
            alpha=0.8,
        )

    ax_overlay.set_xlim(
        0,
        KEOGRAM_TIME_MIN_MAX,
    )

    if APPLY_OVERLAY_PROFILE_SMOOTHING:
        overlay_smoothing_text = (
            f"1D profile smoothing "
            f"σ={OVERLAY_PROFILE_SMOOTH_SIGMA_PIXELS:g} pixels"
        )
    else:
        overlay_smoothing_text = (
            "No additional profile smoothing"
        )

    ax_overlay.set_title(
        f"Selected Profiles Overlaid\n"
        f"{dataset_name}; "
        f"{overlay_smoothing_text}"
    )

    ax_overlay.set_xlabel(
        "Time (min)"
    )

    ax_overlay.set_ylabel(
        "Perturbation amplitude"
    )

    ax_overlay.grid(
        True,
        alpha=0.3,
    )

    ax_overlay.legend(
        loc="best",
        fontsize=8,
    )

    plt.tight_layout()

    outputs["overlay"] = (
        out_dir
        / f"overlay_{ANGLE_DEG:03d}_{filename_label}.png"
    )

    fig_overlay.savefig(
        outputs["overlay"],
        dpi=DPI,
        bbox_inches="tight",
    )

    plt.close(fig_overlay)

    # ==================================================
    # 0, 125, and 250 km rows
    # ==================================================
    fig_rows, ax_rows = plt.subplots(
        figsize=(9, 4)
    )

    ax_rows.plot(
        times_min,
        row_zero,
        color="black",
        linewidth=1.5,
        label=f"{distance_zero_actual:.0f} km",
    )

    ax_rows.plot(
        times_min,
        row_middle,
        color="blue",
        linewidth=1.5,
        label=f"{distance_middle_actual:.0f} km",
    )

    ax_rows.plot(
        times_min,
        row_max,
        color="red",
        linewidth=1.5,
        label=f"{distance_max_actual:.0f} km",
    )

    ax_rows.set_xlim(
        0,
        KEOGRAM_TIME_MIN_MAX,
    )

    ax_rows.set_title(
        f"{distance_zero_actual:.0f}, "
        f"{distance_middle_actual:.0f}, and "
        f"{distance_max_actual:.0f} km Rows\n"
        f"{dataset_name}"
    )

    ax_rows.set_xlabel(
        "Time (min)"
    )

    ax_rows.set_ylabel(
        "Perturbation amplitude"
    )

    ax_rows.grid(
        True,
        alpha=0.3,
    )

    ax_rows.legend(
        loc="best",
        fontsize=9,
    )

    plt.tight_layout()

    outputs["combined_rows"] = (
        out_dir
        / f"rows_0_125_250_{ANGLE_DEG:03d}_{filename_label}.png"
    )

    fig_rows.savefig(
        outputs["combined_rows"],
        dpi=DPI,
        bbox_inches="tight",
    )

    plt.close(fig_rows)

    # ==================================================
    # Selected time profiles
    # ==================================================
    fig_profiles, ax_profiles = plt.subplots(
        figsize=(6, 7)
    )

    for requested_time, col, color in zip(
        TIMES_TO_PLOT,
        columns,
        time_colors,
    ):
        ax_profiles.plot(
            col,
            distance_km,
            linewidth=1.5,
            color=color,
            label=f"{requested_time:.0f} min",
        )

    ax_profiles.axhline(
        distance_zero_actual,
        color="black",
        linestyle="--",
        linewidth=1.2,
        label=f"{distance_zero_actual:.0f} km",
    )

    ax_profiles.axhline(
        distance_middle_actual,
        color="blue",
        linestyle="--",
        linewidth=1.2,
        label=f"{distance_middle_actual:.0f} km",
    )

    ax_profiles.axhline(
        distance_max_actual,
        color="red",
        linestyle="--",
        linewidth=1.2,
        label=f"{distance_max_actual:.0f} km",
    )

    ax_profiles.set_ylim(
        KEOGRAM_DISTANCE_MIN,
        KEOGRAM_DISTANCE_MAX,
    )

    ax_profiles.set_title(
        f"Selected Profiles\n"
        f"{dataset_name}"
    )

    ax_profiles.set_xlabel(
        "Perturbation amplitude"
    )

    ax_profiles.set_ylabel(
        "Distance along slice (km)"
    )

    ax_profiles.grid(
        True,
        alpha=0.3,
    )

    ax_profiles.legend(
        loc="best",
        fontsize=8,
    )

    plt.tight_layout()

    outputs["profiles"] = (
        out_dir
        / f"profiles_{ANGLE_DEG:03d}_{filename_label}.png"
    )

    fig_profiles.savefig(
        outputs["profiles"],
        dpi=DPI,
        bbox_inches="tight",
    )

    plt.close(fig_profiles)

    return outputs


# ==================================================
# Short output paths for Windows
# ==================================================
sigma_t_label = safe_number_label(
    SMOOTH_SIGMA_TIME_PIXELS
)

sigma_d_label = safe_number_label(
    SMOOTH_SIGMA_DISTANCE_PIXELS
)

range_folder = (
    f"r_"
    f"{int(KEOGRAM_DISTANCE_MIN)}_"
    f"{int(KEOGRAM_DISTANCE_MAX)}"
)

if APPLY_2D_SMOOTHING:
    smoothing_folder = (
        f"sig_"
        f"{sigma_t_label}_"
        f"{sigma_d_label}"
    )
else:
    smoothing_folder = "raw"

OUT_DIR = (
    BASE_OUT_DIR
    / range_folder
    / smoothing_folder
    / "pchip_bias"
)

UNCORRECTED_OUT_DIR = (
    OUT_DIR
    / "uncorr"
)

CORRECTION_OUT_DIR = (
    OUT_DIR
    / "field"
)

CORRECTED_OUT_DIR = (
    OUT_DIR
    / "corr"
)

COMPARISON_OUT_DIR = (
    OUT_DIR
    / "compare"
)

for directory in [
    UNCORRECTED_OUT_DIR,
    CORRECTION_OUT_DIR,
    CORRECTED_OUT_DIR,
    COMPARISON_OUT_DIR,
]:
    directory.mkdir(
        parents=True,
        exist_ok=True,
    )


# ==================================================
# Input files
# ==================================================
KEOGRAM_FILE = (
    BASE_OUT_DIR
    / f"BLO_keogram_angle_{ANGLE_DEG:03d}.npy"
)

TIMES_FILE = (
    BASE_OUT_DIR
    / "times_s.npy"
)


# ==================================================
# Read keogram
# ==================================================
keogram_original = np.load(
    KEOGRAM_FILE
).astype(float)

times_s = np.load(
    TIMES_FILE
)

times_min = times_s / 60.0

if keogram_original.ndim != 2:
    raise ValueError(
        f"Expected a two-dimensional keogram, "
        f"but found shape {keogram_original.shape}."
    )

nt, nspace = keogram_original.shape

if len(times_min) != nt:
    raise ValueError(
        f"The keogram contains {nt} time samples, "
        f"but times_s.npy contains "
        f"{len(times_min)} samples."
    )


# ==================================================
# Optional smoothing
# ==================================================
if APPLY_2D_SMOOTHING:
    print(
        "Applying NaN-aware 2D Gaussian smoothing."
    )

    keogram_uncorrected = smooth_keogram_2d(
        data=keogram_original,
        sigma_time=SMOOTH_SIGMA_TIME_PIXELS,
        sigma_distance=SMOOTH_SIGMA_DISTANCE_PIXELS,
        mode=SMOOTH_MODE,
        weight_tolerance=SMOOTH_WEIGHT_TOLERANCE,
    )

    uncorrected_dataset_name = (
        f"Uncorrected, smoothed "
        f"σt={SMOOTH_SIGMA_TIME_PIXELS:g}, "
        f"σd={SMOOTH_SIGMA_DISTANCE_PIXELS:g}"
    )
else:
    print(
        "Smoothing disabled."
    )

    keogram_uncorrected = (
        keogram_original.copy()
    )

    uncorrected_dataset_name = (
        "Uncorrected, unsmoothed"
    )


# ==================================================
# Construct distance axis
# ==================================================
distance_km = np.arange(
    -(nspace // 2),
    nspace // 2 + 1,
)

if len(distance_km) != nspace:
    distance_km = (
        np.arange(nspace)
        - nspace // 2
    )


# ==================================================
# Validate distance settings
# ==================================================
available_distance_min = np.nanmin(
    distance_km
)

available_distance_max = np.nanmax(
    distance_km
)

if (
    KEOGRAM_DISTANCE_MIN
    >= KEOGRAM_DISTANCE_MAX
):
    raise ValueError(
        "KEOGRAM_DISTANCE_MIN must be smaller than "
        "KEOGRAM_DISTANCE_MAX."
    )

if (
    KEOGRAM_DISTANCE_MIN
    < available_distance_min
):
    raise ValueError(
        f"Requested minimum distance is "
        f"{KEOGRAM_DISTANCE_MIN} km, but the keogram "
        f"extends only to {available_distance_min} km."
    )

if (
    KEOGRAM_DISTANCE_MAX
    > available_distance_max
):
    raise ValueError(
        f"Requested maximum distance is "
        f"{KEOGRAM_DISTANCE_MAX} km, but the keogram "
        f"extends only to {available_distance_max} km."
    )

if not (
    KEOGRAM_DISTANCE_MIN
    <= MIDDLE_DISTANCE_KM
    <= KEOGRAM_DISTANCE_MAX
):
    raise ValueError(
        f"MIDDLE_DISTANCE_KM = "
        f"{MIDDLE_DISTANCE_KM} km must lie inside "
        f"the displayed distance range."
    )


# ==================================================
# Create correction keogram
# ==================================================
(
    bias_correction_keogram,
    bias_vs_time,
    bias_distance_factor,
    bias_time_interpolator,
) = create_bias_correction_keogram(
    times_min=times_min,
    distance_km=distance_km,
    bias_time_points_min=BIAS_TIME_POINTS_MIN,
    bias_at_max_distance=BIAS_AT_MAX_DISTANCE,
    bias_distance_min_km=BIAS_DISTANCE_MIN_KM,
    bias_distance_max_km=BIAS_DISTANCE_MAX_KM,
    distance_power=BIAS_DISTANCE_POWER,
    zero_outside_distance_range=(
        BIAS_ZERO_OUTSIDE_DISTANCE_RANGE
    ),
    zero_outside_time_range=(
        BIAS_ZERO_OUTSIDE_TIME_RANGE
    ),
    clip_negative_time_correction=(
        CLIP_NEGATIVE_TIME_CORRECTION
    ),
)


# ==================================================
# Apply bias correction
# ==================================================
if APPLY_BIAS_CORRECTION:
    keogram_corrected = (
        keogram_uncorrected
        + bias_correction_keogram
    )

    corrected_dataset_name = (
        "Bias corrected: PCHIP time dependence, "
        f"distance power={BIAS_DISTANCE_POWER:g}"
    )
else:
    keogram_corrected = (
        keogram_uncorrected.copy()
    )

    corrected_dataset_name = (
        "Bias correction disabled"
    )

# Preserve invalid regions
keogram_corrected[
    ~np.isfinite(keogram_uncorrected)
] = np.nan


# ==================================================
# Select distance rows
# ==================================================
idx_min = np.argmin(
    np.abs(
        distance_km
        - KEOGRAM_DISTANCE_MIN
    )
)

idx_zero = np.argmin(
    np.abs(
        distance_km
        - 0.0
    )
)

idx_middle = np.argmin(
    np.abs(
        distance_km
        - MIDDLE_DISTANCE_KM
    )
)

idx_max = np.argmin(
    np.abs(
        distance_km
        - KEOGRAM_DISTANCE_MAX
    )
)

row_indices = {
    "min": idx_min,
    "zero": idx_zero,
    "middle": idx_middle,
    "max": idx_max,
}

actual_distances = {
    "min": distance_km[idx_min],
    "zero": distance_km[idx_zero],
    "middle": distance_km[idx_middle],
    "max": distance_km[idx_max],
}


# ==================================================
# Select time columns
# ==================================================
column_indices = []

for requested_time in TIMES_TO_PLOT:
    idx = np.argmin(
        np.abs(
            times_min
            - requested_time
        )
    )

    column_indices.append(
        idx
    )


# ==================================================
# Plot colors
# ==================================================
time_colors = [
    "black",
    "blue",
    "green",
    "purple",
    "orange",
    "pink",
    "brown",
    "magenta",
]

if len(TIMES_TO_PLOT) > len(time_colors):
    raise ValueError(
        "There are more requested times than "
        "available colors."
    )


# ==================================================
# Create uncorrected plots
# ==================================================
uncorrected_outputs = create_keogram_plots(
    keogram=keogram_uncorrected,
    dataset_name=uncorrected_dataset_name,
    filename_label="uncorrected",
    out_dir=UNCORRECTED_OUT_DIR,
    times_min=times_min,
    distance_km=distance_km,
    row_indices=row_indices,
    actual_distances=actual_distances,
    column_indices=column_indices,
    time_colors=time_colors,
)


# ==================================================
# Create corrected plots
# ==================================================
corrected_outputs = create_keogram_plots(
    keogram=keogram_corrected,
    dataset_name=corrected_dataset_name,
    filename_label="corrected",
    out_dir=CORRECTED_OUT_DIR,
    times_min=times_min,
    distance_km=distance_km,
    row_indices=row_indices,
    actual_distances=actual_distances,
    column_indices=column_indices,
    time_colors=time_colors,
)


# ==================================================
# Plot 2D correction keogram
# ==================================================
fig_corr, ax_corr = plt.subplots(
    figsize=(9, 5)
)

correction_max = np.nanmax(
    np.abs(bias_correction_keogram)
)

if (
    not np.isfinite(correction_max)
    or correction_max == 0
):
    correction_max = 1.0

im_corr = ax_corr.imshow(
    bias_correction_keogram.T,
    origin="lower",
    aspect="auto",
    cmap=CMAP,
    vmin=-correction_max,
    vmax=correction_max,
    extent=[
        times_min.min(),
        times_min.max(),
        distance_km.min(),
        distance_km.max(),
    ],
)

cbar_corr = fig_corr.colorbar(
    im_corr,
    ax=ax_corr,
    fraction=0.046,
    pad=0.04,
)

cbar_corr.set_label(
    "Additive perturbation correction"
)

for time_point in BIAS_TIME_POINTS_MIN:
    ax_corr.axvline(
        time_point,
        color="black",
        linestyle=":",
        linewidth=1.2,
    )

ax_corr.axhline(
    BIAS_DISTANCE_MIN_KM,
    color="black",
    linestyle="--",
    linewidth=1.2,
)

ax_corr.axhline(
    MIDDLE_DISTANCE_KM,
    color="blue",
    linestyle="--",
    linewidth=1.2,
)

ax_corr.axhline(
    BIAS_DISTANCE_MAX_KM,
    color="red",
    linestyle="--",
    linewidth=1.2,
)

ax_corr.set_xlim(
    0,
    KEOGRAM_TIME_MIN_MAX,
)

ax_corr.set_ylim(
    KEOGRAM_DISTANCE_MIN,
    KEOGRAM_DISTANCE_MAX,
)

ax_corr.set_title(
    "2D Bias-Correction Keogram\n"
    f"PCHIP time dependence and distance power "
    f"{BIAS_DISTANCE_POWER:g}"
)

ax_corr.set_xlabel(
    "Time (min)"
)

ax_corr.set_ylabel(
    "Distance along slice (km)"
)

plt.tight_layout()

out_correction_keogram_plot = (
    CORRECTION_OUT_DIR
    / f"bias2d_{ANGLE_DEG:03d}.png"
)

fig_corr.savefig(
    out_correction_keogram_plot,
    dpi=DPI,
    bbox_inches="tight",
)

plt.close(fig_corr)


# ==================================================
# Plot correction versus time at selected distances
# ==================================================
fig_corr_rows, ax_corr_rows = plt.subplots(
    figsize=(9, 4)
)

ax_corr_rows.plot(
    times_min,
    bias_correction_keogram[:, idx_zero],
    color="black",
    linewidth=1.5,
    label=(
        f"{actual_distances['zero']:.0f} km"
    ),
)

ax_corr_rows.plot(
    times_min,
    bias_correction_keogram[:, idx_middle],
    color="blue",
    linewidth=1.5,
    label=(
        f"{actual_distances['middle']:.0f} km"
    ),
)

ax_corr_rows.plot(
    times_min,
    bias_correction_keogram[:, idx_max],
    color="red",
    linewidth=1.5,
    label=(
        f"{actual_distances['max']:.0f} km"
    ),
)

ax_corr_rows.scatter(
    BIAS_TIME_POINTS_MIN,
    BIAS_AT_MAX_DISTANCE,
    color="red",
    s=40,
    zorder=5,
    label="Specified values at 250 km",
)

ax_corr_rows.set_xlim(
    0,
    KEOGRAM_TIME_MIN_MAX,
)

ax_corr_rows.set_title(
    "Bias Correction at Selected Distances"
)

ax_corr_rows.set_xlabel(
    "Time (min)"
)

ax_corr_rows.set_ylabel(
    "Additive correction"
)

ax_corr_rows.grid(
    True,
    alpha=0.3,
)

ax_corr_rows.legend(
    loc="best",
)

plt.tight_layout()

out_correction_rows_plot = (
    CORRECTION_OUT_DIR
    / f"bias_rows_{ANGLE_DEG:03d}.png"
)

fig_corr_rows.savefig(
    out_correction_rows_plot,
    dpi=DPI,
    bbox_inches="tight",
)

plt.close(fig_corr_rows)


# ==================================================
# Plot time correction at 250 km
# ==================================================
fig_time, ax_time = plt.subplots(
    figsize=(8, 4)
)

ax_time.plot(
    times_min,
    bias_vs_time,
    linewidth=1.8,
    label="PCHIP interpolation",
)

ax_time.scatter(
    BIAS_TIME_POINTS_MIN,
    BIAS_AT_MAX_DISTANCE,
    s=45,
    zorder=5,
    label="Specified points",
)

ax_time.set_xlim(
    0,
    KEOGRAM_TIME_MIN_MAX,
)

ax_time.set_title(
    "Time-Dependent Correction at 250 km"
)

ax_time.set_xlabel(
    "Time (min)"
)

ax_time.set_ylabel(
    "Additive correction"
)

ax_time.grid(
    True,
    alpha=0.3,
)

ax_time.legend(
    loc="best",
)

plt.tight_layout()

out_time_correction_plot = (
    CORRECTION_OUT_DIR
    / f"bias_time_{ANGLE_DEG:03d}.png"
)

fig_time.savefig(
    out_time_correction_plot,
    dpi=DPI,
    bbox_inches="tight",
)

plt.close(fig_time)


# ==================================================
# Plot distance correction factor
# ==================================================
fig_distance, ax_distance = plt.subplots(
    figsize=(8, 4)
)

ax_distance.plot(
    distance_km,
    bias_distance_factor,
    linewidth=1.8,
)

selected_distance_points = np.array(
    [
        0.0,
        125.0,
        250.0,
    ]
)

selected_distance_factors = (
    (
        selected_distance_points
        - BIAS_DISTANCE_MIN_KM
    )
    / (
        BIAS_DISTANCE_MAX_KM
        - BIAS_DISTANCE_MIN_KM
    )
) ** BIAS_DISTANCE_POWER

ax_distance.scatter(
    selected_distance_points,
    selected_distance_factors,
    s=45,
    zorder=5,
)

ax_distance.set_xlim(
    KEOGRAM_DISTANCE_MIN,
    KEOGRAM_DISTANCE_MAX,
)

ax_distance.set_title(
    f"Distance Correction Factor "
    f"(power={BIAS_DISTANCE_POWER:g})"
)

ax_distance.set_xlabel(
    "Distance along slice (km)"
)

ax_distance.set_ylabel(
    "Correction fraction"
)

ax_distance.grid(
    True,
    alpha=0.3,
)

plt.tight_layout()

out_distance_correction_plot = (
    CORRECTION_OUT_DIR
    / f"bias_distance_{ANGLE_DEG:03d}.png"
)

fig_distance.savefig(
    out_distance_correction_plot,
    dpi=DPI,
    bbox_inches="tight",
)

plt.close(fig_distance)


# ==================================================
# Corrected versus uncorrected row comparison
# ==================================================
fig_compare, axes_compare = plt.subplots(
    nrows=3,
    ncols=1,
    figsize=(10, 9),
    sharex=True,
)

comparison_rows = [
    (
        idx_zero,
        actual_distances["zero"],
    ),
    (
        idx_middle,
        actual_distances["middle"],
    ),
    (
        idx_max,
        actual_distances["max"],
    ),
]

for ax_compare, (
    row_index,
    actual_distance,
) in zip(
    axes_compare,
    comparison_rows,
):
    ax_compare.plot(
        times_min,
        keogram_uncorrected[:, row_index],
        linewidth=1.4,
        label="Uncorrected",
    )

    ax_compare.plot(
        times_min,
        keogram_corrected[:, row_index],
        linewidth=1.4,
        label="Corrected",
    )

    ax_compare.set_xlim(
        0,
        KEOGRAM_TIME_MIN_MAX,
    )

    ax_compare.set_ylabel(
        "Perturbation"
    )

    ax_compare.set_title(
        f"{actual_distance:.0f} km"
    )

    ax_compare.grid(
        True,
        alpha=0.3,
    )

    ax_compare.legend(
        loc="best",
    )

axes_compare[-1].set_xlabel(
    "Time (min)"
)

fig_compare.suptitle(
    f"{ANGLE_DEG}° Corrected versus Uncorrected",
    y=0.995,
)

plt.tight_layout()

out_comparison_rows_plot = (
    COMPARISON_OUT_DIR
    / f"compare_rows_{ANGLE_DEG:03d}.png"
)

fig_compare.savefig(
    out_comparison_rows_plot,
    dpi=DPI,
    bbox_inches="tight",
)

plt.close(fig_compare)


# ==================================================
# Side-by-side 2D comparison
# ==================================================
combined_data = np.concatenate(
    [
        keogram_uncorrected.ravel(),
        keogram_corrected.ravel(),
    ]
)

comparison_vmin, comparison_vmax = (
    get_symmetric_color_limits(
        combined_data,
        percentile=99.0,
    )
)

fig_2d, axes_2d = plt.subplots(
    nrows=1,
    ncols=2,
    figsize=(14, 5),
    sharex=True,
    sharey=True,
)

im_uncorrected = axes_2d[0].imshow(
    keogram_uncorrected.T,
    origin="lower",
    aspect="auto",
    cmap=CMAP,
    vmin=comparison_vmin,
    vmax=comparison_vmax,
    extent=[
        times_min.min(),
        times_min.max(),
        distance_km.min(),
        distance_km.max(),
    ],
)

axes_2d[0].set_title(
    "Uncorrected"
)

axes_2d[0].set_xlabel(
    "Time (min)"
)

axes_2d[0].set_ylabel(
    "Distance along slice (km)"
)

axes_2d[0].set_xlim(
    0,
    KEOGRAM_TIME_MIN_MAX,
)

axes_2d[0].set_ylim(
    KEOGRAM_DISTANCE_MIN,
    KEOGRAM_DISTANCE_MAX,
)

axes_2d[1].imshow(
    keogram_corrected.T,
    origin="lower",
    aspect="auto",
    cmap=CMAP,
    vmin=comparison_vmin,
    vmax=comparison_vmax,
    extent=[
        times_min.min(),
        times_min.max(),
        distance_km.min(),
        distance_km.max(),
    ],
)

axes_2d[1].set_title(
    "Bias Corrected"
)

axes_2d[1].set_xlabel(
    "Time (min)"
)

axes_2d[1].set_xlim(
    0,
    KEOGRAM_TIME_MIN_MAX,
)

axes_2d[1].set_ylim(
    KEOGRAM_DISTANCE_MIN,
    KEOGRAM_DISTANCE_MAX,
)

cbar_2d = fig_2d.colorbar(
    im_uncorrected,
    ax=axes_2d,
    fraction=0.025,
    pad=0.03,
)

cbar_2d.set_label(
    "Perturbation amplitude"
)

fig_2d.suptitle(
    f"{ANGLE_DEG}° BLO Keogram Comparison"
)

fig_2d.subplots_adjust(
    left=0.07,
    right=0.90,
    bottom=0.12,
    top=0.87,
    wspace=0.12,
)

out_2d_comparison_plot = (
    COMPARISON_OUT_DIR
    / f"compare_2d_{ANGLE_DEG:03d}.png"
)

fig_2d.savefig(
    out_2d_comparison_plot,
    dpi=DPI,
    bbox_inches="tight",
)

plt.close(fig_2d)


# ==================================================
# Save arrays using short filenames
# ==================================================
out_uncorrected_npy = (
    UNCORRECTED_OUT_DIR
    / f"uncorr_{ANGLE_DEG:03d}.npy"
)

out_correction_npy = (
    CORRECTION_OUT_DIR
    / f"bias2d_{ANGLE_DEG:03d}.npy"
)

out_corrected_npy = (
    CORRECTED_OUT_DIR
    / f"corr_{ANGLE_DEG:03d}.npy"
)

out_bias_time_npy = (
    CORRECTION_OUT_DIR
    / f"bias_time_{ANGLE_DEG:03d}.npy"
)

out_bias_distance_npy = (
    CORRECTION_OUT_DIR
    / f"bias_dist_{ANGLE_DEG:03d}.npy"
)

np.save(
    out_uncorrected_npy,
    keogram_uncorrected,
)

np.save(
    out_correction_npy,
    bias_correction_keogram,
)

np.save(
    out_corrected_npy,
    keogram_corrected,
)

np.save(
    out_bias_time_npy,
    bias_vs_time,
)

np.save(
    out_bias_distance_npy,
    bias_distance_factor,
)


# ==================================================
# Print correction checks
# ==================================================
print(
    "\nBias-correction checks:"
)

check_times = [
    0.0,
    150.0,
    300.0,
]

check_distances = [
    0.0,
    125.0,
    250.0,
]

for check_time in check_times:
    time_idx = np.argmin(
        np.abs(
            times_min
            - check_time
        )
    )

    print(
        f"\nRequested time: "
        f"{check_time:.1f} min"
    )

    print(
        f"Actual time: "
        f"{times_min[time_idx]:.2f} min"
    )

    for check_distance in check_distances:
        distance_idx = np.argmin(
            np.abs(
                distance_km
                - check_distance
            )
        )

        correction_value = (
            bias_correction_keogram[
                time_idx,
                distance_idx,
            ]
        )

        print(
            f"  Distance "
            f"{distance_km[distance_idx]:7.1f} km: "
            f"correction = "
            f"{correction_value:.6f}"
        )


# ==================================================
# Print saved outputs
# ==================================================
print(
    "\nOutput root:"
)

print(
    OUT_DIR.resolve()
)

print(
    "\nSaved uncorrected plots:"
)

for output_path in uncorrected_outputs.values():
    print(
        output_path
    )

print(
    "\nSaved corrected plots:"
)

for output_path in corrected_outputs.values():
    print(
        output_path
    )

print(
    "\nSaved correction plots:"
)

print(
    out_correction_keogram_plot
)

print(
    out_correction_rows_plot
)

print(
    out_time_correction_plot
)

print(
    out_distance_correction_plot
)

print(
    "\nSaved comparison plots:"
)

print(
    out_comparison_rows_plot
)

print(
    out_2d_comparison_plot
)

print(
    "\nSaved arrays:"
)

print(
    out_uncorrected_npy
)

print(
    out_correction_npy
)

print(
    out_corrected_npy
)

print(
    out_bias_time_npy
)

print(
    out_bias_distance_npy
)

print(
    "\nKeogram shapes:"
)

print(
    f"Uncorrected: "
    f"{keogram_uncorrected.shape}"
)

print(
    f"Correction:  "
    f"{bias_correction_keogram.shape}"
)

print(
    f"Corrected:   "
    f"{keogram_corrected.shape}"
)
```

## Correlation instead of overlaying

Create a new script. Instead of plotting selected profiles and overlaying: For each row, shift the row by -20 to 20 mins and calculate the correlation between that row and the 0 km row. Plot the 2d map of correlation (x axis is the shift, y axis is the distance, color is the correlation).

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

from scipy.ndimage import gaussian_filter
from scipy.interpolate import PchipInterpolator


# ==================================================
# Settings
# ==================================================
BASE_OUT_DIR = Path(
    "BLO_perturbation_keogram_outputs_new"
)

ANGLE_DEG = 45

# Time range used for correlation
CORRELATION_TIME_MIN = 0.0
CORRELATION_TIME_MAX = 300.0

# Distance range included in the map
CORRELATION_DISTANCE_MIN_KM = 0.0
CORRELATION_DISTANCE_MAX_KM = 250.0

# Shift range is now specified in MINUTES
SHIFT_MIN_MINUTES = -140.0
SHIFT_MAX_MINUTES = 140.0

# Shifts are still applied at integer-pixel resolution.
# The actual shift increment will equal one time pixel.
MIN_VALID_CORRELATION_POINTS = 20

# Correlation values below this are not used in the slope fit
MIN_CORRELATION_FOR_SLOPE = 0.50

# Optional distance limits for fitting the ridge
SLOPE_FIT_DISTANCE_MIN_KM = 0.0
SLOPE_FIT_DISTANCE_MAX_KM = 250.0

DPI = 200

CORRELATION_CMAP = "RdBu_r"

# Display only correlations from 0.5 to 1.0
CORRELATION_VMIN = 0.50
CORRELATION_VMAX = 1.00


# ==================================================
# Optional 2D smoothing
# ==================================================
APPLY_2D_SMOOTHING = True

# Keogram dimensions:
# axis 0 = time
# axis 1 = distance
SMOOTH_SIGMA_TIME_PIXELS = 1.5
SMOOTH_SIGMA_DISTANCE_PIXELS = 2.0

SMOOTH_MODE = "nearest"
SMOOTH_WEIGHT_TOLERANCE = 1e-12


# ==================================================
# Bias-correction settings
# ==================================================
APPLY_BIAS_CORRECTION = True

BIAS_TIME_POINTS_MIN = np.array(
    [
        0.0,
        150.0,
        300.0,
    ],
    dtype=float,
)

BIAS_AT_MAX_DISTANCE = np.array(
    [
        0.20,
        0.04,
        -0.01,
    ],
    dtype=float,
)

BIAS_DISTANCE_MIN_KM = 0.0
BIAS_DISTANCE_MAX_KM = 250.0

# 1.0 = linear distance dependence
# 2.0 = quadratic distance dependence
BIAS_DISTANCE_POWER = 1.0

BIAS_ZERO_OUTSIDE_DISTANCE_RANGE = True
BIAS_ZERO_OUTSIDE_TIME_RANGE = True
CLIP_NEGATIVE_TIME_CORRECTION = False


# ==================================================
# Filename-safe numeric label
# ==================================================
def safe_number_label(value):
    value_float = float(value)

    if value_float.is_integer():
        return str(int(value_float))

    return (
        f"{value_float:g}"
        .replace("-", "m")
        .replace(".", "p")
    )


# ==================================================
# NaN-aware 2D Gaussian smoothing
# ==================================================
def smooth_keogram_2d(
    data,
    sigma_time,
    sigma_distance,
    mode="nearest",
    weight_tolerance=1e-12,
):
    """
    Apply NaN-aware Gaussian smoothing to a 2D keogram.
    """
    data = np.asarray(
        data,
        dtype=float,
    )

    if data.ndim != 2:
        raise ValueError(
            "The keogram must be two-dimensional."
        )

    if sigma_time < 0 or sigma_distance < 0:
        raise ValueError(
            "Smoothing sigma values must be nonnegative."
        )

    if sigma_time == 0 and sigma_distance == 0:
        return data.copy()

    valid_mask = np.isfinite(data)

    data_filled = np.where(
        valid_mask,
        data,
        0.0,
    )

    weights = valid_mask.astype(float)

    sigma = (
        sigma_time,
        sigma_distance,
    )

    smoothed_values = gaussian_filter(
        data_filled,
        sigma=sigma,
        mode=mode,
    )

    smoothed_weights = gaussian_filter(
        weights,
        sigma=sigma,
        mode=mode,
    )

    smoothed = np.full(
        data.shape,
        np.nan,
        dtype=float,
    )

    valid_output = (
        smoothed_weights
        > weight_tolerance
    )

    smoothed[valid_output] = (
        smoothed_values[valid_output]
        / smoothed_weights[valid_output]
    )

    return smoothed


# ==================================================
# Create 2D bias-correction keogram
# ==================================================
def create_bias_correction_keogram(
    times_min,
    distance_km,
    bias_time_points_min,
    bias_at_max_distance,
    bias_distance_min_km,
    bias_distance_max_km,
    distance_power=1.0,
    zero_outside_distance_range=True,
    zero_outside_time_range=True,
    clip_negative_time_correction=False,
):
    """
    Create the time x distance additive bias field.
    """
    times_min = np.asarray(
        times_min,
        dtype=float,
    )

    distance_km = np.asarray(
        distance_km,
        dtype=float,
    )

    bias_time_points_min = np.asarray(
        bias_time_points_min,
        dtype=float,
    )

    bias_at_max_distance = np.asarray(
        bias_at_max_distance,
        dtype=float,
    )

    if (
        len(bias_time_points_min)
        != len(bias_at_max_distance)
    ):
        raise ValueError(
            "Bias time points and bias values must have "
            "the same length."
        )

    if not np.all(
        np.diff(bias_time_points_min) > 0
    ):
        raise ValueError(
            "BIAS_TIME_POINTS_MIN must be strictly increasing."
        )

    if bias_distance_max_km <= bias_distance_min_km:
        raise ValueError(
            "The maximum bias distance must exceed "
            "the minimum bias distance."
        )

    if distance_power <= 0:
        raise ValueError(
            "BIAS_DISTANCE_POWER must be positive."
        )

    # ==================================================
    # PCHIP time dependence
    # ==================================================
    interpolator = PchipInterpolator(
        bias_time_points_min,
        bias_at_max_distance,
        extrapolate=False,
    )

    bias_vs_time = interpolator(
        times_min
    )

    if zero_outside_time_range:
        inside_time = (
            (times_min >= bias_time_points_min[0])
            & (times_min <= bias_time_points_min[-1])
        )

        bias_vs_time = np.where(
            inside_time,
            bias_vs_time,
            0.0,
        )
    else:
        bias_vs_time = np.where(
            times_min < bias_time_points_min[0],
            bias_at_max_distance[0],
            bias_vs_time,
        )

        bias_vs_time = np.where(
            times_min > bias_time_points_min[-1],
            bias_at_max_distance[-1],
            bias_vs_time,
        )

    if clip_negative_time_correction:
        bias_vs_time = np.maximum(
            bias_vs_time,
            0.0,
        )

    # ==================================================
    # Distance dependence
    # ==================================================
    normalized_distance = (
        distance_km
        - bias_distance_min_km
    ) / (
        bias_distance_max_km
        - bias_distance_min_km
    )

    if zero_outside_distance_range:
        inside_distance = (
            (distance_km >= bias_distance_min_km)
            & (distance_km <= bias_distance_max_km)
        )

        normalized_distance = np.where(
            inside_distance,
            normalized_distance,
            0.0,
        )
    else:
        normalized_distance = np.clip(
            normalized_distance,
            0.0,
            1.0,
        )

    distance_factor = (
        normalized_distance ** distance_power
    )

    correction_keogram = (
        bias_vs_time[:, np.newaxis]
        * distance_factor[np.newaxis, :]
    )

    return (
        correction_keogram,
        bias_vs_time,
        distance_factor,
    )


# ==================================================
# NaN-aware Pearson correlation
# ==================================================
def nan_pearson_correlation(
    reference,
    comparison,
    minimum_points=20,
):
    """
    Calculate Pearson correlation using finite overlapping values.
    """
    reference = np.asarray(
        reference,
        dtype=float,
    )

    comparison = np.asarray(
        comparison,
        dtype=float,
    )

    if reference.shape != comparison.shape:
        raise ValueError(
            "The two arrays must have the same shape."
        )

    valid = (
        np.isfinite(reference)
        & np.isfinite(comparison)
    )

    if np.count_nonzero(valid) < minimum_points:
        return np.nan

    reference_valid = reference[valid]
    comparison_valid = comparison[valid]

    reference_centered = (
        reference_valid
        - np.mean(reference_valid)
    )

    comparison_centered = (
        comparison_valid
        - np.mean(comparison_valid)
    )

    denominator = np.sqrt(
        np.sum(reference_centered ** 2)
        * np.sum(comparison_centered ** 2)
    )

    if (
        not np.isfinite(denominator)
        or denominator <= 0
    ):
        return np.nan

    correlation = (
        np.sum(
            reference_centered
            * comparison_centered
        )
        / denominator
    )

    return np.clip(
        correlation,
        -1.0,
        1.0,
    )


# ==================================================
# Correlation for one integer-pixel shift
# ==================================================
def calculate_shifted_correlation(
    reference_row,
    comparison_row,
    shift_pixels,
    minimum_points=20,
):
    """
    Shift comparison_row and correlate it with reference_row.

    Positive shift:
        comparison row moves toward later times.

    Negative shift:
        comparison row moves toward earlier times.

    The data are not wrapped around.
    """
    reference_row = np.asarray(
        reference_row,
        dtype=float,
    )

    comparison_row = np.asarray(
        comparison_row,
        dtype=float,
    )

    if reference_row.shape != comparison_row.shape:
        raise ValueError(
            "Reference and comparison rows must have "
            "the same shape."
        )

    ntime = len(reference_row)

    if abs(shift_pixels) >= ntime:
        return np.nan

    if shift_pixels > 0:
        reference_overlap = reference_row[
            shift_pixels:
        ]

        comparison_overlap = comparison_row[
            :-shift_pixels
        ]

    elif shift_pixels < 0:
        shift_abs = abs(
            shift_pixels
        )

        reference_overlap = reference_row[
            :-shift_abs
        ]

        comparison_overlap = comparison_row[
            shift_abs:
        ]

    else:
        reference_overlap = reference_row
        comparison_overlap = comparison_row

    return nan_pearson_correlation(
        reference=reference_overlap,
        comparison=comparison_overlap,
        minimum_points=minimum_points,
    )


# ==================================================
# Calculate complete correlation map
# ==================================================
def calculate_correlation_map(
    keogram,
    distance_km,
    reference_distance_km,
    selected_distance_mask,
    shifts_pixels,
    minimum_points=20,
):
    """
    Correlate every selected distance row with the 0 km row
    for every requested time shift.
    """
    keogram = np.asarray(
        keogram,
        dtype=float,
    )

    distance_km = np.asarray(
        distance_km,
        dtype=float,
    )

    shifts_pixels = np.asarray(
        shifts_pixels,
        dtype=int,
    )

    reference_index = np.argmin(
        np.abs(
            distance_km
            - reference_distance_km
        )
    )

    selected_indices = np.where(
        selected_distance_mask
    )[0]

    selected_distances = distance_km[
        selected_indices
    ]

    reference_row = keogram[
        :,
        reference_index,
    ]

    correlation_map = np.full(
        (
            len(selected_indices),
            len(shifts_pixels),
        ),
        np.nan,
        dtype=float,
    )

    for output_row_index, keogram_row_index in enumerate(
        selected_indices
    ):
        comparison_row = keogram[
            :,
            keogram_row_index,
        ]

        for shift_index, shift_pixels in enumerate(
            shifts_pixels
        ):
            correlation_map[
                output_row_index,
                shift_index,
            ] = calculate_shifted_correlation(
                reference_row=reference_row,
                comparison_row=comparison_row,
                shift_pixels=int(shift_pixels),
                minimum_points=minimum_points,
            )

    return (
        correlation_map,
        selected_distances,
        reference_index,
    )


# ==================================================
# Maximum-correlation shift for each row
# ==================================================
def find_best_shifts(
    correlation_map,
    shifts_minutes,
):
    """
    Find the shift in minutes having the largest correlation
    for every distance row.
    """
    correlation_map = np.asarray(
        correlation_map,
        dtype=float,
    )

    shifts_minutes = np.asarray(
        shifts_minutes,
        dtype=float,
    )

    best_shift_minutes = np.full(
        correlation_map.shape[0],
        np.nan,
        dtype=float,
    )

    maximum_correlation = np.full(
        correlation_map.shape[0],
        np.nan,
        dtype=float,
    )

    for row_index in range(
        correlation_map.shape[0]
    ):
        correlation_row = correlation_map[
            row_index,
            :
        ]

        finite = np.isfinite(
            correlation_row
        )

        if not np.any(finite):
            continue

        finite_indices = np.where(
            finite
        )[0]

        local_max_index = np.argmax(
            correlation_row[finite]
        )

        full_max_index = finite_indices[
            local_max_index
        ]

        best_shift_minutes[row_index] = (
            shifts_minutes[full_max_index]
        )

        maximum_correlation[row_index] = (
            correlation_row[full_max_index]
        )

    return (
        best_shift_minutes,
        maximum_correlation,
    )


# ==================================================
# Fit maximum-correlation ridge
# ==================================================
def fit_maximum_correlation_slope(
    distances_km,
    best_shift_minutes,
    maximum_correlation,
    minimum_correlation=0.5,
    distance_min_km=None,
    distance_max_km=None,
):
    """
    Fit:

        shift_minutes = slope * distance_km + intercept

    Returns
    -------
    slope_min_per_km : float
        Time-delay slope.

    intercept_minutes : float
        Fitted shift at zero distance.

    equivalent_speed_km_per_min : float
        Reciprocal of the time-delay slope.

    equivalent_speed_m_per_s : float
        Equivalent signed speed.

    r_squared : float
        Coefficient of determination.

    fit_mask : ndarray
        Points used in the fit.
    """
    distances_km = np.asarray(
        distances_km,
        dtype=float,
    )

    best_shift_minutes = np.asarray(
        best_shift_minutes,
        dtype=float,
    )

    maximum_correlation = np.asarray(
        maximum_correlation,
        dtype=float,
    )

    fit_mask = (
        np.isfinite(distances_km)
        & np.isfinite(best_shift_minutes)
        & np.isfinite(maximum_correlation)
        & (
            maximum_correlation
            >= minimum_correlation
        )
    )

    if distance_min_km is not None:
        fit_mask &= (
            distances_km
            >= distance_min_km
        )

    if distance_max_km is not None:
        fit_mask &= (
            distances_km
            <= distance_max_km
        )

    if np.count_nonzero(fit_mask) < 2:
        return {
            "slope_min_per_km": np.nan,
            "intercept_minutes": np.nan,
            "speed_km_per_min": np.nan,
            "speed_m_per_s": np.nan,
            "r_squared": np.nan,
            "fit_mask": fit_mask,
        }

    fit_distances = distances_km[
        fit_mask
    ]

    fit_shifts = best_shift_minutes[
        fit_mask
    ]

    slope_min_per_km, intercept_minutes = np.polyfit(
        fit_distances,
        fit_shifts,
        deg=1,
    )

    predicted_shifts = (
        slope_min_per_km
        * fit_distances
        + intercept_minutes
    )

    residual_sum_squares = np.sum(
        (
            fit_shifts
            - predicted_shifts
        ) ** 2
    )

    total_sum_squares = np.sum(
        (
            fit_shifts
            - np.mean(fit_shifts)
        ) ** 2
    )

    if total_sum_squares > 0:
        r_squared = (
            1.0
            - residual_sum_squares
            / total_sum_squares
        )
    else:
        r_squared = np.nan

    if np.isclose(
        slope_min_per_km,
        0.0,
    ):
        speed_km_per_min = np.nan
        speed_m_per_s = np.nan
    else:
        speed_km_per_min = (
            1.0
            / slope_min_per_km
        )

        speed_m_per_s = (
            speed_km_per_min
            * 1000.0
            / 60.0
        )

    return {
        "slope_min_per_km": slope_min_per_km,
        "intercept_minutes": intercept_minutes,
        "speed_km_per_min": speed_km_per_min,
        "speed_m_per_s": speed_m_per_s,
        "r_squared": r_squared,
        "fit_mask": fit_mask,
    }


# ==================================================
# Plot correlation map and fitted ridge
# ==================================================
def plot_correlation_map(
    correlation_map,
    shifts_minutes,
    selected_distances,
    best_shift_minutes,
    maximum_correlation,
    slope_results,
    dataset_name,
    output_file,
):
    """
    Plot correlation versus time shift and distance.
    """
    fig, ax = plt.subplots(
        figsize=(10, 6)
    )

    if len(shifts_minutes) > 1:
        shift_spacing = np.median(
            np.diff(shifts_minutes)
        )
    else:
        shift_spacing = 1.0

    if len(selected_distances) > 1:
        distance_spacing = np.median(
            np.diff(selected_distances)
        )
    else:
        distance_spacing = 1.0

    im = ax.imshow(
        correlation_map,
        origin="lower",
        aspect="auto",
        interpolation="nearest",
        cmap=CORRELATION_CMAP,
        vmin=CORRELATION_VMIN,
        vmax=CORRELATION_VMAX,
        extent=[
            shifts_minutes[0]
            - shift_spacing / 2.0,
            shifts_minutes[-1]
            + shift_spacing / 2.0,
            selected_distances[0]
            - distance_spacing / 2.0,
            selected_distances[-1]
            + distance_spacing / 2.0,
        ],
    )

    cbar = fig.colorbar(
        im,
        ax=ax,
        fraction=0.046,
        pad=0.04,
    )

    cbar.set_label(
        "Pearson correlation with 0 km row"
    )

    ax.axvline(
        0.0,
        color="black",
        linestyle="--",
        linewidth=1.2,
        label="Zero shift",
    )

    # Only show maximum-correlation points at or above
    # the correlation threshold.
    ridge_mask = (
        np.isfinite(best_shift_minutes)
        & np.isfinite(maximum_correlation)
        & (
            maximum_correlation
            >= MIN_CORRELATION_FOR_SLOPE
        )
    )

    if np.any(ridge_mask):
        ax.plot(
            best_shift_minutes[ridge_mask],
            selected_distances[ridge_mask],
            color="black",
            linewidth=1.2,
            label="Maximum correlation",
        )

        ax.scatter(
            best_shift_minutes[ridge_mask],
            selected_distances[ridge_mask],
            color="black",
            s=8,
        )

    slope = slope_results[
        "slope_min_per_km"
    ]

    intercept = slope_results[
        "intercept_minutes"
    ]

    r_squared = slope_results[
        "r_squared"
    ]

    speed_m_per_s = slope_results[
        "speed_m_per_s"
    ]

    if np.isfinite(slope):
        fit_distances = np.linspace(
            SLOPE_FIT_DISTANCE_MIN_KM,
            SLOPE_FIT_DISTANCE_MAX_KM,
            300,
        )

        fit_shifts = (
            slope
            * fit_distances
            + intercept
        )

        ax.plot(
            fit_shifts,
            fit_distances,
            color="lime",
            linestyle="--",
            linewidth=2.0,
            label="Linear ridge fit",
        )

        fit_text = (
            f"Shift slope = {slope:.5f} min/km\n"
            f"Equivalent speed = {speed_m_per_s:.2f} m/s\n"
            f"$R^2$ = {r_squared:.3f}"
        )

        ax.text(
            0.02,
            0.03,
            fit_text,
            transform=ax.transAxes,
            ha="left",
            va="bottom",
            fontsize=9,
            bbox={
                "facecolor": "white",
                "edgecolor": "black",
                "alpha": 0.85,
            },
        )

    ax.set_xlim(
        SHIFT_MIN_MINUTES,
        SHIFT_MAX_MINUTES,
    )

    ax.set_ylim(
        CORRELATION_DISTANCE_MIN_KM,
        CORRELATION_DISTANCE_MAX_KM,
    )

    ax.set_title(
        f"{ANGLE_DEG}° Shifted Row Correlation\n"
        f"{dataset_name}"
    )

    ax.set_xlabel(
        "Shift applied to row (min)"
    )

    ax.set_ylabel(
        "Distance along slice (km)"
    )

    ax.legend(
        loc="upper right",
        fontsize=9,
    )

    plt.tight_layout()

    fig.savefig(
        output_file,
        dpi=DPI,
        bbox_inches="tight",
    )

    plt.close(fig)


# ==================================================
# Output paths
# ==================================================
sigma_t_label = safe_number_label(
    SMOOTH_SIGMA_TIME_PIXELS
)

sigma_d_label = safe_number_label(
    SMOOTH_SIGMA_DISTANCE_PIXELS
)

distance_folder = (
    f"r_"
    f"{int(CORRELATION_DISTANCE_MIN_KM)}_"
    f"{int(CORRELATION_DISTANCE_MAX_KM)}"
)

shift_folder = (
    f"shift_min_"
    f"{safe_number_label(SHIFT_MIN_MINUTES)}_"
    f"{safe_number_label(SHIFT_MAX_MINUTES)}"
)

if APPLY_2D_SMOOTHING:
    smoothing_folder = (
        f"sig_"
        f"{sigma_t_label}_"
        f"{sigma_d_label}"
    )
else:
    smoothing_folder = "raw"

OUT_DIR = (
    BASE_OUT_DIR
    / distance_folder
    / smoothing_folder
    / "pchip_bias"
    / "row_correlation_minutes"
    / shift_folder
)

UNCORRECTED_OUT_DIR = (
    OUT_DIR
    / "uncorrected"
)

CORRECTED_OUT_DIR = (
    OUT_DIR
    / "corrected"
)

for directory in [
    UNCORRECTED_OUT_DIR,
    CORRECTED_OUT_DIR,
]:
    directory.mkdir(
        parents=True,
        exist_ok=True,
    )


# ==================================================
# Input files
# ==================================================
KEOGRAM_FILE = (
    BASE_OUT_DIR
    / f"BLO_keogram_angle_{ANGLE_DEG:03d}.npy"
)

TIMES_FILE = (
    BASE_OUT_DIR
    / "times_s.npy"
)


# ==================================================
# Read keogram and time
# ==================================================
keogram_original = np.load(
    KEOGRAM_FILE
).astype(float)

times_s = np.load(
    TIMES_FILE
).astype(float)

times_min = times_s / 60.0

if keogram_original.ndim != 2:
    raise ValueError(
        f"Expected a 2D keogram, but found "
        f"{keogram_original.shape}."
    )

nt, nspace = keogram_original.shape

if len(times_min) != nt:
    raise ValueError(
        f"The keogram has {nt} time samples, but "
        f"times_s.npy has {len(times_min)} samples."
    )

if len(times_min) < 2:
    raise ValueError(
        "At least two time samples are required."
    )


# ==================================================
# Determine minutes represented by one time pixel
# ==================================================
time_differences_min = np.diff(
    times_min
)

finite_time_differences = time_differences_min[
    np.isfinite(time_differences_min)
]

if len(finite_time_differences) == 0:
    raise ValueError(
        "Could not determine the time spacing."
    )

MINUTES_PER_PIXEL = np.median(
    finite_time_differences
)

if (
    not np.isfinite(MINUTES_PER_PIXEL)
    or MINUTES_PER_PIXEL <= 0
):
    raise ValueError(
        "The calculated minutes-per-pixel value is invalid."
    )

maximum_time_spacing_deviation = np.max(
    np.abs(
        finite_time_differences
        - MINUTES_PER_PIXEL
    )
)

print(
    "\nTime sampling:"
)

print(
    f"One time pixel = "
    f"{MINUTES_PER_PIXEL:.6f} minutes"
)

print(
    f"One time pixel = "
    f"{MINUTES_PER_PIXEL * 60.0:.3f} seconds"
)

print(
    f"Maximum deviation from median spacing = "
    f"{maximum_time_spacing_deviation:.6e} minutes"
)


# ==================================================
# Convert requested minute range to integer shifts
# ==================================================
shift_min_pixels = int(
    np.ceil(
        SHIFT_MIN_MINUTES
        / MINUTES_PER_PIXEL
    )
)

shift_max_pixels = int(
    np.floor(
        SHIFT_MAX_MINUTES
        / MINUTES_PER_PIXEL
    )
)

shifts_pixels = np.arange(
    shift_min_pixels,
    shift_max_pixels + 1,
    dtype=int,
)

shifts_minutes = (
    shifts_pixels
    * MINUTES_PER_PIXEL
)

print(
    "\nShift range:"
)

print(
    f"Requested: "
    f"{SHIFT_MIN_MINUTES:.2f} to "
    f"{SHIFT_MAX_MINUTES:.2f} minutes"
)

print(
    f"Actual: "
    f"{shifts_minutes[0]:.2f} to "
    f"{shifts_minutes[-1]:.2f} minutes"
)

print(
    f"Pixel shifts: "
    f"{shifts_pixels[0]} to "
    f"{shifts_pixels[-1]}"
)


# ==================================================
# Construct distance axis
# ==================================================
distance_km = np.arange(
    -(nspace // 2),
    nspace // 2 + 1,
    dtype=float,
)

if len(distance_km) != nspace:
    distance_km = (
        np.arange(
            nspace,
            dtype=float,
        )
        - nspace // 2
    )


# ==================================================
# Validate requested ranges
# ==================================================
available_distance_min = np.nanmin(
    distance_km
)

available_distance_max = np.nanmax(
    distance_km
)

if (
    CORRELATION_DISTANCE_MIN_KM
    >= CORRELATION_DISTANCE_MAX_KM
):
    raise ValueError(
        "The minimum correlation distance must be smaller "
        "than the maximum."
    )

if (
    CORRELATION_DISTANCE_MIN_KM
    < available_distance_min
    or CORRELATION_DISTANCE_MAX_KM
    > available_distance_max
):
    raise ValueError(
        "The requested distance range extends outside "
        "the available keogram."
    )

if CORRELATION_TIME_MIN >= CORRELATION_TIME_MAX:
    raise ValueError(
        "CORRELATION_TIME_MIN must be smaller than "
        "CORRELATION_TIME_MAX."
    )


# ==================================================
# Restrict to requested time range
# ==================================================
time_mask = (
    (times_min >= CORRELATION_TIME_MIN)
    & (times_min <= CORRELATION_TIME_MAX)
)

if np.count_nonzero(time_mask) < MIN_VALID_CORRELATION_POINTS:
    raise ValueError(
        "The selected time interval has too few samples."
    )

times_correlation_min = times_min[
    time_mask
]

keogram_time_selected = keogram_original[
    time_mask,
    :
]


# ==================================================
# Optional smoothing
# ==================================================
if APPLY_2D_SMOOTHING:
    print(
        "\nApplying NaN-aware 2D Gaussian smoothing."
    )

    keogram_uncorrected = smooth_keogram_2d(
        data=keogram_time_selected,
        sigma_time=SMOOTH_SIGMA_TIME_PIXELS,
        sigma_distance=SMOOTH_SIGMA_DISTANCE_PIXELS,
        mode=SMOOTH_MODE,
        weight_tolerance=SMOOTH_WEIGHT_TOLERANCE,
    )

    uncorrected_dataset_name = (
        f"Uncorrected, smoothed "
        f"σt={SMOOTH_SIGMA_TIME_PIXELS:g}, "
        f"σd={SMOOTH_SIGMA_DISTANCE_PIXELS:g}"
    )
else:
    keogram_uncorrected = (
        keogram_time_selected.copy()
    )

    uncorrected_dataset_name = (
        "Uncorrected, unsmoothed"
    )


# ==================================================
# Create and apply bias correction
# ==================================================
(
    bias_correction_keogram,
    bias_vs_time,
    bias_distance_factor,
) = create_bias_correction_keogram(
    times_min=times_correlation_min,
    distance_km=distance_km,
    bias_time_points_min=BIAS_TIME_POINTS_MIN,
    bias_at_max_distance=BIAS_AT_MAX_DISTANCE,
    bias_distance_min_km=BIAS_DISTANCE_MIN_KM,
    bias_distance_max_km=BIAS_DISTANCE_MAX_KM,
    distance_power=BIAS_DISTANCE_POWER,
    zero_outside_distance_range=(
        BIAS_ZERO_OUTSIDE_DISTANCE_RANGE
    ),
    zero_outside_time_range=(
        BIAS_ZERO_OUTSIDE_TIME_RANGE
    ),
    clip_negative_time_correction=(
        CLIP_NEGATIVE_TIME_CORRECTION
    ),
)

if APPLY_BIAS_CORRECTION:
    keogram_corrected = (
        keogram_uncorrected
        + bias_correction_keogram
    )

    corrected_dataset_name = (
        "Bias corrected: PCHIP time dependence, "
        f"distance power={BIAS_DISTANCE_POWER:g}"
    )
else:
    keogram_corrected = (
        keogram_uncorrected.copy()
    )

    corrected_dataset_name = (
        "Bias correction disabled"
    )

keogram_corrected[
    ~np.isfinite(keogram_uncorrected)
] = np.nan


# ==================================================
# Select distance rows
# ==================================================
selected_distance_mask = (
    (distance_km >= CORRELATION_DISTANCE_MIN_KM)
    & (distance_km <= CORRELATION_DISTANCE_MAX_KM)
)

selected_distances = distance_km[
    selected_distance_mask
]


# ==================================================
# Uncorrected correlation map
# ==================================================
print(
    "\nCalculating uncorrected correlation map."
)

(
    correlation_map_uncorrected,
    selected_distances_uncorrected,
    reference_index_uncorrected,
) = calculate_correlation_map(
    keogram=keogram_uncorrected,
    distance_km=distance_km,
    reference_distance_km=0.0,
    selected_distance_mask=selected_distance_mask,
    shifts_pixels=shifts_pixels,
    minimum_points=MIN_VALID_CORRELATION_POINTS,
)

(
    best_shift_uncorrected_minutes,
    maximum_correlation_uncorrected,
) = find_best_shifts(
    correlation_map=correlation_map_uncorrected,
    shifts_minutes=shifts_minutes,
)

uncorrected_slope_results = (
    fit_maximum_correlation_slope(
        distances_km=selected_distances_uncorrected,
        best_shift_minutes=(
            best_shift_uncorrected_minutes
        ),
        maximum_correlation=(
            maximum_correlation_uncorrected
        ),
        minimum_correlation=(
            MIN_CORRELATION_FOR_SLOPE
        ),
        distance_min_km=(
            SLOPE_FIT_DISTANCE_MIN_KM
        ),
        distance_max_km=(
            SLOPE_FIT_DISTANCE_MAX_KM
        ),
    )
)


# ==================================================
# Corrected correlation map
# ==================================================
print(
    "\nCalculating corrected correlation map."
)

(
    correlation_map_corrected,
    selected_distances_corrected,
    reference_index_corrected,
) = calculate_correlation_map(
    keogram=keogram_corrected,
    distance_km=distance_km,
    reference_distance_km=0.0,
    selected_distance_mask=selected_distance_mask,
    shifts_pixels=shifts_pixels,
    minimum_points=MIN_VALID_CORRELATION_POINTS,
)

(
    best_shift_corrected_minutes,
    maximum_correlation_corrected,
) = find_best_shifts(
    correlation_map=correlation_map_corrected,
    shifts_minutes=shifts_minutes,
)

corrected_slope_results = (
    fit_maximum_correlation_slope(
        distances_km=selected_distances_corrected,
        best_shift_minutes=(
            best_shift_corrected_minutes
        ),
        maximum_correlation=(
            maximum_correlation_corrected
        ),
        minimum_correlation=(
            MIN_CORRELATION_FOR_SLOPE
        ),
        distance_min_km=(
            SLOPE_FIT_DISTANCE_MIN_KM
        ),
        distance_max_km=(
            SLOPE_FIT_DISTANCE_MAX_KM
        ),
    )
)


# ==================================================
# Plot maps
# ==================================================
uncorrected_plot_file = (
    UNCORRECTED_OUT_DIR
    / f"correlation_minutes_{ANGLE_DEG:03d}_uncorrected.png"
)

corrected_plot_file = (
    CORRECTED_OUT_DIR
    / f"correlation_minutes_{ANGLE_DEG:03d}_corrected.png"
)

plot_correlation_map(
    correlation_map=correlation_map_uncorrected,
    shifts_minutes=shifts_minutes,
    selected_distances=selected_distances_uncorrected,
    best_shift_minutes=(
        best_shift_uncorrected_minutes
    ),
    maximum_correlation=(
        maximum_correlation_uncorrected
    ),
    slope_results=uncorrected_slope_results,
    dataset_name=uncorrected_dataset_name,
    output_file=uncorrected_plot_file,
)

plot_correlation_map(
    correlation_map=correlation_map_corrected,
    shifts_minutes=shifts_minutes,
    selected_distances=selected_distances_corrected,
    best_shift_minutes=(
        best_shift_corrected_minutes
    ),
    maximum_correlation=(
        maximum_correlation_corrected
    ),
    slope_results=corrected_slope_results,
    dataset_name=corrected_dataset_name,
    output_file=corrected_plot_file,
)


# ==================================================
# Save arrays
# ==================================================
np.save(
    OUT_DIR / "shifts_minutes.npy",
    shifts_minutes,
)

np.save(
    OUT_DIR / "shifts_pixels.npy",
    shifts_pixels,
)

np.save(
    OUT_DIR / "distances_km.npy",
    selected_distances,
)

np.save(
    UNCORRECTED_OUT_DIR
    / f"correlation_{ANGLE_DEG:03d}_uncorrected.npy",
    correlation_map_uncorrected,
)

np.save(
    UNCORRECTED_OUT_DIR
    / f"best_shift_minutes_{ANGLE_DEG:03d}_uncorrected.npy",
    best_shift_uncorrected_minutes,
)

np.save(
    CORRECTED_OUT_DIR
    / f"correlation_{ANGLE_DEG:03d}_corrected.npy",
    correlation_map_corrected,
)

np.save(
    CORRECTED_OUT_DIR
    / f"best_shift_minutes_{ANGLE_DEG:03d}_corrected.npy",
    best_shift_corrected_minutes,
)


# ==================================================
# Save CSV summaries
# ==================================================
uncorrected_summary = np.column_stack(
    [
        selected_distances_uncorrected,
        best_shift_uncorrected_minutes,
        maximum_correlation_uncorrected,
    ]
)

corrected_summary = np.column_stack(
    [
        selected_distances_corrected,
        best_shift_corrected_minutes,
        maximum_correlation_corrected,
    ]
)

uncorrected_csv = (
    UNCORRECTED_OUT_DIR
    / f"ridge_{ANGLE_DEG:03d}_uncorrected.csv"
)

corrected_csv = (
    CORRECTED_OUT_DIR
    / f"ridge_{ANGLE_DEG:03d}_corrected.csv"
)

np.savetxt(
    uncorrected_csv,
    uncorrected_summary,
    delimiter=",",
    header=(
        "distance_km,"
        "best_shift_minutes,"
        "maximum_correlation"
    ),
    comments="",
)

np.savetxt(
    corrected_csv,
    corrected_summary,
    delimiter=",",
    header=(
        "distance_km,"
        "best_shift_minutes,"
        "maximum_correlation"
    ),
    comments="",
)


# ==================================================
# Save slope results
# ==================================================
slope_output_file = (
    OUT_DIR
    / f"correlation_slope_{ANGLE_DEG:03d}.txt"
)

with open(
    slope_output_file,
    "w",
    encoding="utf-8",
) as file:
    file.write(
        f"Minutes per pixel: "
        f"{MINUTES_PER_PIXEL:.10f}\n\n"
    )

    file.write(
        "UNCORRECTED\n"
    )

    for key, value in uncorrected_slope_results.items():
        if key != "fit_mask":
            file.write(
                f"{key}: {value}\n"
            )

    file.write(
        "\nCORRECTED\n"
    )

    for key, value in corrected_slope_results.items():
        if key != "fit_mask":
            file.write(
                f"{key}: {value}\n"
            )


# ==================================================
# Print results
# ==================================================
def print_slope_results(
    label,
    results,
):
    print(
        f"\n{label}:"
    )

    print(
        f"  Shift slope: "
        f"{results['slope_min_per_km']:.8f} min/km"
    )

    print(
        f"  Intercept: "
        f"{results['intercept_minutes']:.6f} min"
    )

    print(
        f"  Equivalent map slope: "
        f"{results['speed_km_per_min']:.8f} km/min"
    )

    print(
        f"  Equivalent signed speed: "
        f"{results['speed_m_per_s']:.4f} m/s"
    )

    print(
        f"  R²: "
        f"{results['r_squared']:.6f}"
    )

    print(
        f"  Points used: "
        f"{np.count_nonzero(results['fit_mask'])}"
    )


print_slope_results(
    "Uncorrected maximum-correlation ridge",
    uncorrected_slope_results,
)

print_slope_results(
    "Corrected maximum-correlation ridge",
    corrected_slope_results,
)

print(
    "\nSaved plots:"
)

print(
    uncorrected_plot_file
)

print(
    corrected_plot_file
)

print(
    "\nSaved slope report:"
)

print(
    slope_output_file
)

print(
    "\nOutput directory:"
)

print(
    OUT_DIR.resolve()
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

from scipy.ndimage import gaussian_filter
from scipy.interpolate import PchipInterpolator


# ==================================================
# Settings
# ==================================================
BASE_OUT_DIR = Path(
    "BLO_perturbation_keogram_outputs_new"
)

# Process every directional keogram:
# 0, 5, 10, ..., 175 degrees
KEOGRAM_ANGLES_DEG = np.arange(
    0,
    176,
    5,
    dtype=int,
)

# Time range used for correlation
CORRELATION_TIME_MIN = 0.0
CORRELATION_TIME_MAX = 300.0

# Distance range included in the correlation map
CORRELATION_DISTANCE_MIN_KM = 0.0
CORRELATION_DISTANCE_MAX_KM = 250.0

# Shift range in minutes
SHIFT_MIN_MINUTES = -200.0
SHIFT_MAX_MINUTES = 200.0

# Minimum number of overlapping time samples
MIN_VALID_CORRELATION_POINTS = 20

# Minimum maximum-correlation value used in the slope fit
MIN_CORRELATION_FOR_SLOPE = 0.50

# Distance range used in the ridge fit
SLOPE_FIT_DISTANCE_MIN_KM = 0.0
SLOPE_FIT_DISTANCE_MAX_KM = 250.0

DPI = 200

CORRELATION_CMAP = "RdBu_r"

# Correlation color scale
CORRELATION_VMIN = 0.70
CORRELATION_VMAX = 1.00


# ==================================================
# Optional 2D smoothing
# ==================================================
APPLY_2D_SMOOTHING = True

# Keogram dimensions:
# axis 0 = time
# axis 1 = distance
SMOOTH_SIGMA_TIME_PIXELS = 1.5
SMOOTH_SIGMA_DISTANCE_PIXELS = 2.0

SMOOTH_MODE = "nearest"
SMOOTH_WEIGHT_TOLERANCE = 1e-12


# ==================================================
# Bias-correction settings
# ==================================================
APPLY_BIAS_CORRECTION = True

BIAS_TIME_POINTS_MIN = np.array(
    [
        0.0,
        150.0,
        300.0,
    ],
    dtype=float,
)

BIAS_AT_MAX_DISTANCE = np.array(
    [
        0.20,
        0.04,
        -0.01,
    ],
    dtype=float,
)

BIAS_DISTANCE_MIN_KM = 0.0
BIAS_DISTANCE_MAX_KM = 250.0

# 1.0 = linear distance dependence
# 2.0 = quadratic distance dependence
BIAS_DISTANCE_POWER = 1.0

BIAS_ZERO_OUTSIDE_DISTANCE_RANGE = True
BIAS_ZERO_OUTSIDE_TIME_RANGE = True
CLIP_NEGATIVE_TIME_CORRECTION = False


# ==================================================
# Filename-safe numeric label
# ==================================================
def safe_number_label(value):
    value_float = float(value)

    if value_float.is_integer():
        return str(
            int(value_float)
        )

    return (
        f"{value_float:g}"
        .replace("-", "m")
        .replace(".", "p")
    )


# ==================================================
# NaN-aware 2D Gaussian smoothing
# ==================================================
def smooth_keogram_2d(
    data,
    sigma_time,
    sigma_distance,
    mode="nearest",
    weight_tolerance=1e-12,
):
    """
    Apply NaN-aware Gaussian smoothing to a 2D keogram.
    """
    data = np.asarray(
        data,
        dtype=float,
    )

    if data.ndim != 2:
        raise ValueError(
            "The keogram must be two-dimensional."
        )

    if sigma_time < 0 or sigma_distance < 0:
        raise ValueError(
            "Smoothing sigma values must be nonnegative."
        )

    if sigma_time == 0 and sigma_distance == 0:
        return data.copy()

    valid_mask = np.isfinite(
        data
    )

    data_filled = np.where(
        valid_mask,
        data,
        0.0,
    )

    weights = valid_mask.astype(
        float
    )

    sigma = (
        sigma_time,
        sigma_distance,
    )

    smoothed_values = gaussian_filter(
        data_filled,
        sigma=sigma,
        mode=mode,
    )

    smoothed_weights = gaussian_filter(
        weights,
        sigma=sigma,
        mode=mode,
    )

    smoothed = np.full(
        data.shape,
        np.nan,
        dtype=float,
    )

    valid_output = (
        smoothed_weights
        > weight_tolerance
    )

    smoothed[valid_output] = (
        smoothed_values[valid_output]
        / smoothed_weights[valid_output]
    )

    return smoothed


# ==================================================
# Create 2D bias-correction keogram
# ==================================================
def create_bias_correction_keogram(
    times_min,
    distance_km,
    bias_time_points_min,
    bias_at_max_distance,
    bias_distance_min_km,
    bias_distance_max_km,
    distance_power=1.0,
    zero_outside_distance_range=True,
    zero_outside_time_range=True,
    clip_negative_time_correction=False,
):
    """
    Create the time x distance additive bias field.
    """
    times_min = np.asarray(
        times_min,
        dtype=float,
    )

    distance_km = np.asarray(
        distance_km,
        dtype=float,
    )

    bias_time_points_min = np.asarray(
        bias_time_points_min,
        dtype=float,
    )

    bias_at_max_distance = np.asarray(
        bias_at_max_distance,
        dtype=float,
    )

    if (
        len(bias_time_points_min)
        != len(bias_at_max_distance)
    ):
        raise ValueError(
            "Bias time points and bias values must have "
            "the same length."
        )

    if not np.all(
        np.diff(bias_time_points_min) > 0
    ):
        raise ValueError(
            "BIAS_TIME_POINTS_MIN must be strictly increasing."
        )

    if bias_distance_max_km <= bias_distance_min_km:
        raise ValueError(
            "The maximum bias distance must exceed "
            "the minimum bias distance."
        )

    if distance_power <= 0:
        raise ValueError(
            "BIAS_DISTANCE_POWER must be positive."
        )

    # ==================================================
    # PCHIP time dependence
    # ==================================================
    interpolator = PchipInterpolator(
        bias_time_points_min,
        bias_at_max_distance,
        extrapolate=False,
    )

    bias_vs_time = interpolator(
        times_min
    )

    if zero_outside_time_range:
        inside_time = (
            (times_min >= bias_time_points_min[0])
            & (times_min <= bias_time_points_min[-1])
        )

        bias_vs_time = np.where(
            inside_time,
            bias_vs_time,
            0.0,
        )
    else:
        bias_vs_time = np.where(
            times_min < bias_time_points_min[0],
            bias_at_max_distance[0],
            bias_vs_time,
        )

        bias_vs_time = np.where(
            times_min > bias_time_points_min[-1],
            bias_at_max_distance[-1],
            bias_vs_time,
        )

    if clip_negative_time_correction:
        bias_vs_time = np.maximum(
            bias_vs_time,
            0.0,
        )

    # ==================================================
    # Distance dependence
    # ==================================================
    normalized_distance = (
        distance_km
        - bias_distance_min_km
    ) / (
        bias_distance_max_km
        - bias_distance_min_km
    )

    if zero_outside_distance_range:
        inside_distance = (
            (distance_km >= bias_distance_min_km)
            & (distance_km <= bias_distance_max_km)
        )

        normalized_distance = np.where(
            inside_distance,
            normalized_distance,
            0.0,
        )
    else:
        normalized_distance = np.clip(
            normalized_distance,
            0.0,
            1.0,
        )

    distance_factor = (
        normalized_distance ** distance_power
    )

    correction_keogram = (
        bias_vs_time[:, np.newaxis]
        * distance_factor[np.newaxis, :]
    )

    return (
        correction_keogram,
        bias_vs_time,
        distance_factor,
    )


# ==================================================
# NaN-aware Pearson correlation
# ==================================================
def nan_pearson_correlation(
    reference,
    comparison,
    minimum_points=20,
):
    """
    Calculate Pearson correlation using finite overlapping values.
    """
    reference = np.asarray(
        reference,
        dtype=float,
    )

    comparison = np.asarray(
        comparison,
        dtype=float,
    )

    if reference.shape != comparison.shape:
        raise ValueError(
            "The two arrays must have the same shape."
        )

    valid = (
        np.isfinite(reference)
        & np.isfinite(comparison)
    )

    if np.count_nonzero(valid) < minimum_points:
        return np.nan

    reference_valid = reference[
        valid
    ]

    comparison_valid = comparison[
        valid
    ]

    reference_centered = (
        reference_valid
        - np.mean(reference_valid)
    )

    comparison_centered = (
        comparison_valid
        - np.mean(comparison_valid)
    )

    denominator = np.sqrt(
        np.sum(reference_centered ** 2)
        * np.sum(comparison_centered ** 2)
    )

    if (
        not np.isfinite(denominator)
        or denominator <= 0
    ):
        return np.nan

    correlation = (
        np.sum(
            reference_centered
            * comparison_centered
        )
        / denominator
    )

    return np.clip(
        correlation,
        -1.0,
        1.0,
    )


# ==================================================
# Correlation for one integer-pixel shift
# ==================================================
def calculate_shifted_correlation(
    reference_row,
    comparison_row,
    shift_pixels,
    minimum_points=20,
):
    """
    Shift comparison_row and correlate it with reference_row.

    Positive shift:
        comparison row moves toward later times.

    Negative shift:
        comparison row moves toward earlier times.

    No circular wrapping is used.
    """
    reference_row = np.asarray(
        reference_row,
        dtype=float,
    )

    comparison_row = np.asarray(
        comparison_row,
        dtype=float,
    )

    if reference_row.shape != comparison_row.shape:
        raise ValueError(
            "Reference and comparison rows must have "
            "the same shape."
        )

    ntime = len(
        reference_row
    )

    if abs(shift_pixels) >= ntime:
        return np.nan

    if shift_pixels > 0:
        reference_overlap = reference_row[
            shift_pixels:
        ]

        comparison_overlap = comparison_row[
            :-shift_pixels
        ]

    elif shift_pixels < 0:
        shift_abs = abs(
            shift_pixels
        )

        reference_overlap = reference_row[
            :-shift_abs
        ]

        comparison_overlap = comparison_row[
            shift_abs:
        ]

    else:
        reference_overlap = reference_row
        comparison_overlap = comparison_row

    return nan_pearson_correlation(
        reference=reference_overlap,
        comparison=comparison_overlap,
        minimum_points=minimum_points,
    )


# ==================================================
# Calculate complete correlation map
# ==================================================
def calculate_correlation_map(
    keogram,
    distance_km,
    reference_distance_km,
    selected_distance_mask,
    shifts_pixels,
    minimum_points=20,
):
    """
    Correlate every selected distance row with the reference row
    for every requested shift.
    """
    keogram = np.asarray(
        keogram,
        dtype=float,
    )

    distance_km = np.asarray(
        distance_km,
        dtype=float,
    )

    shifts_pixels = np.asarray(
        shifts_pixels,
        dtype=int,
    )

    reference_index = np.argmin(
        np.abs(
            distance_km
            - reference_distance_km
        )
    )

    selected_indices = np.where(
        selected_distance_mask
    )[0]

    selected_distances = distance_km[
        selected_indices
    ]

    reference_row = keogram[
        :,
        reference_index,
    ]

    correlation_map = np.full(
        (
            len(selected_indices),
            len(shifts_pixels),
        ),
        np.nan,
        dtype=float,
    )

    for output_row_index, keogram_row_index in enumerate(
        selected_indices
    ):
        comparison_row = keogram[
            :,
            keogram_row_index,
        ]

        for shift_index, shift_pixels in enumerate(
            shifts_pixels
        ):
            correlation_map[
                output_row_index,
                shift_index,
            ] = calculate_shifted_correlation(
                reference_row=reference_row,
                comparison_row=comparison_row,
                shift_pixels=int(shift_pixels),
                minimum_points=minimum_points,
            )

    return (
        correlation_map,
        selected_distances,
        reference_index,
    )


# ==================================================
# Maximum-correlation shift for each row
# ==================================================
def find_best_shifts(
    correlation_map,
    shifts_minutes,
):
    """
    Find the shift in minutes having the largest correlation
    for every distance row.
    """
    correlation_map = np.asarray(
        correlation_map,
        dtype=float,
    )

    shifts_minutes = np.asarray(
        shifts_minutes,
        dtype=float,
    )

    best_shift_minutes = np.full(
        correlation_map.shape[0],
        np.nan,
        dtype=float,
    )

    maximum_correlation = np.full(
        correlation_map.shape[0],
        np.nan,
        dtype=float,
    )

    for row_index in range(
        correlation_map.shape[0]
    ):
        correlation_row = correlation_map[
            row_index,
            :
        ]

        finite = np.isfinite(
            correlation_row
        )

        if not np.any(finite):
            continue

        finite_indices = np.where(
            finite
        )[0]

        local_max_index = np.argmax(
            correlation_row[finite]
        )

        full_max_index = finite_indices[
            local_max_index
        ]

        best_shift_minutes[row_index] = (
            shifts_minutes[full_max_index]
        )

        maximum_correlation[row_index] = (
            correlation_row[full_max_index]
        )

    return (
        best_shift_minutes,
        maximum_correlation,
    )


# ==================================================
# Fit maximum-correlation ridge
# ==================================================
def fit_maximum_correlation_slope(
    distances_km,
    best_shift_minutes,
    maximum_correlation,
    minimum_correlation=0.5,
    distance_min_km=None,
    distance_max_km=None,
):
    """
    Fit:

        shift_minutes = slope * distance_km + intercept
    """
    distances_km = np.asarray(
        distances_km,
        dtype=float,
    )

    best_shift_minutes = np.asarray(
        best_shift_minutes,
        dtype=float,
    )

    maximum_correlation = np.asarray(
        maximum_correlation,
        dtype=float,
    )

    fit_mask = (
        np.isfinite(distances_km)
        & np.isfinite(best_shift_minutes)
        & np.isfinite(maximum_correlation)
        & (
            maximum_correlation
            >= minimum_correlation
        )
    )

    if distance_min_km is not None:
        fit_mask &= (
            distances_km
            >= distance_min_km
        )

    if distance_max_km is not None:
        fit_mask &= (
            distances_km
            <= distance_max_km
        )

    number_of_points = np.count_nonzero(
        fit_mask
    )

    if number_of_points < 2:
        return {
            "slope_min_per_km": np.nan,
            "intercept_minutes": np.nan,
            "speed_km_per_min": np.nan,
            "speed_m_per_s": np.nan,
            "r_squared": np.nan,
            "number_of_fit_points": number_of_points,
            "fit_mask": fit_mask,
        }

    fit_distances = distances_km[
        fit_mask
    ]

    fit_shifts = best_shift_minutes[
        fit_mask
    ]

    slope_min_per_km, intercept_minutes = np.polyfit(
        fit_distances,
        fit_shifts,
        deg=1,
    )

    predicted_shifts = (
        slope_min_per_km
        * fit_distances
        + intercept_minutes
    )

    residual_sum_squares = np.sum(
        (
            fit_shifts
            - predicted_shifts
        ) ** 2
    )

    total_sum_squares = np.sum(
        (
            fit_shifts
            - np.mean(fit_shifts)
        ) ** 2
    )

    if total_sum_squares > 0:
        r_squared = (
            1.0
            - residual_sum_squares
            / total_sum_squares
        )
    else:
        r_squared = np.nan

    if np.isclose(
        slope_min_per_km,
        0.0,
    ):
        speed_km_per_min = np.nan
        speed_m_per_s = np.nan
    else:
        speed_km_per_min = (
            1.0
            / slope_min_per_km
        )

        speed_m_per_s = (
            speed_km_per_min
            * 1000.0
            / 60.0
        )

    return {
        "slope_min_per_km": slope_min_per_km,
        "intercept_minutes": intercept_minutes,
        "speed_km_per_min": speed_km_per_min,
        "speed_m_per_s": speed_m_per_s,
        "r_squared": r_squared,
        "number_of_fit_points": number_of_points,
        "fit_mask": fit_mask,
    }


# ==================================================
# Plot correlation map and fitted ridge
# ==================================================
def plot_correlation_map(
    angle_deg,
    correlation_map,
    shifts_minutes,
    selected_distances,
    best_shift_minutes,
    maximum_correlation,
    slope_results,
    dataset_name,
    output_file,
):
    """
    Plot correlation versus time shift and distance.
    """
    fig, ax = plt.subplots(
        figsize=(10, 6)
    )

    if len(shifts_minutes) > 1:
        shift_spacing = np.median(
            np.diff(shifts_minutes)
        )
    else:
        shift_spacing = 1.0

    if len(selected_distances) > 1:
        distance_spacing = np.median(
            np.diff(selected_distances)
        )
    else:
        distance_spacing = 1.0

    im = ax.imshow(
        correlation_map,
        origin="lower",
        aspect="auto",
        interpolation="nearest",
        cmap=CORRELATION_CMAP,
        vmin=CORRELATION_VMIN,
        vmax=CORRELATION_VMAX,
        extent=[
            shifts_minutes[0]
            - shift_spacing / 2.0,
            shifts_minutes[-1]
            + shift_spacing / 2.0,
            selected_distances[0]
            - distance_spacing / 2.0,
            selected_distances[-1]
            + distance_spacing / 2.0,
        ],
    )

    cbar = fig.colorbar(
        im,
        ax=ax,
        fraction=0.046,
        pad=0.04,
    )

    cbar.set_label(
        "Pearson correlation with 0 km row"
    )

    ax.axvline(
        0.0,
        color="black",
        linestyle="--",
        linewidth=1.2,
        label="Zero shift",
    )

    ridge_mask = (
        np.isfinite(best_shift_minutes)
        & np.isfinite(maximum_correlation)
        & (
            maximum_correlation
            >= MIN_CORRELATION_FOR_SLOPE
        )
    )

    if np.any(ridge_mask):
        ax.plot(
            best_shift_minutes[ridge_mask],
            selected_distances[ridge_mask],
            color="black",
            linewidth=1.2,
            label="Maximum correlation",
        )

        ax.scatter(
            best_shift_minutes[ridge_mask],
            selected_distances[ridge_mask],
            color="black",
            s=8,
        )

    slope = slope_results[
        "slope_min_per_km"
    ]

    intercept = slope_results[
        "intercept_minutes"
    ]

    r_squared = slope_results[
        "r_squared"
    ]

    speed_m_per_s = slope_results[
        "speed_m_per_s"
    ]

    if np.isfinite(slope):
        fit_distances = np.linspace(
            SLOPE_FIT_DISTANCE_MIN_KM,
            SLOPE_FIT_DISTANCE_MAX_KM,
            300,
        )

        fit_shifts = (
            slope
            * fit_distances
            + intercept
        )

        ax.plot(
            fit_shifts,
            fit_distances,
            color="lime",
            linestyle="--",
            linewidth=2.0,
            label="Linear ridge fit",
        )

        fit_text = (
            f"Shift slope = {slope:.5f} min/km\n"
            f"Equivalent speed = {speed_m_per_s:.2f} m/s\n"
            f"$R^2$ = {r_squared:.3f}"
        )

        ax.text(
            0.02,
            0.03,
            fit_text,
            transform=ax.transAxes,
            ha="left",
            va="bottom",
            fontsize=9,
            bbox={
                "facecolor": "white",
                "edgecolor": "black",
                "alpha": 0.85,
            },
        )

    ax.set_xlim(
        SHIFT_MIN_MINUTES,
        SHIFT_MAX_MINUTES,
    )

    ax.set_ylim(
        CORRELATION_DISTANCE_MIN_KM,
        CORRELATION_DISTANCE_MAX_KM,
    )

    ax.set_title(
        f"{angle_deg}° Shifted Row Correlation\n"
        f"{dataset_name}"
    )

    ax.set_xlabel(
        "Shift applied to row (min)"
    )

    ax.set_ylabel(
        "Distance along slice (km)"
    )

    ax.legend(
        loc="upper right",
        fontsize=9,
    )

    plt.tight_layout()

    fig.savefig(
        output_file,
        dpi=DPI,
        bbox_inches="tight",
    )

    plt.close(fig)


# ==================================================
# Print slope results
# ==================================================
def print_slope_results(
    angle_deg,
    label,
    results,
):
    print(
        f"\n{angle_deg}° — {label}:"
    )

    print(
        f"  Shift slope: "
        f"{results['slope_min_per_km']:.8f} min/km"
    )

    print(
        f"  Intercept: "
        f"{results['intercept_minutes']:.6f} min"
    )

    print(
        f"  Equivalent speed: "
        f"{results['speed_m_per_s']:.4f} m/s"
    )

    print(
        f"  R²: "
        f"{results['r_squared']:.6f}"
    )

    print(
        f"  Points used: "
        f"{results['number_of_fit_points']}"
    )


# ==================================================
# Output paths
# ==================================================
sigma_t_label = safe_number_label(
    SMOOTH_SIGMA_TIME_PIXELS
)

sigma_d_label = safe_number_label(
    SMOOTH_SIGMA_DISTANCE_PIXELS
)

distance_folder = (
    f"r_"
    f"{int(CORRELATION_DISTANCE_MIN_KM)}_"
    f"{int(CORRELATION_DISTANCE_MAX_KM)}"
)

shift_folder = (
    f"shift_min_"
    f"{safe_number_label(SHIFT_MIN_MINUTES)}_"
    f"{safe_number_label(SHIFT_MAX_MINUTES)}"
)

if APPLY_2D_SMOOTHING:
    smoothing_folder = (
        f"sig_"
        f"{sigma_t_label}_"
        f"{sigma_d_label}"
    )
else:
    smoothing_folder = "raw"

OUT_DIR = (
    BASE_OUT_DIR
    / distance_folder
    / smoothing_folder
    / "pchip_bias"
    / "row_correlation_minutes_all_angles"
    / shift_folder
)

UNCORRECTED_OUT_DIR = (
    OUT_DIR
    / "uncorrected"
)

CORRECTED_OUT_DIR = (
    OUT_DIR
    / "corrected"
)

SLOPE_REPORT_OUT_DIR = (
    OUT_DIR
    / "slope_reports"
)

for directory in [
    OUT_DIR,
    UNCORRECTED_OUT_DIR,
    CORRECTED_OUT_DIR,
    SLOPE_REPORT_OUT_DIR,
]:
    directory.mkdir(
        parents=True,
        exist_ok=True,
    )


# ==================================================
# Read common time array
# ==================================================
TIMES_FILE = (
    BASE_OUT_DIR
    / "times_s.npy"
)

times_s = np.load(
    TIMES_FILE
).astype(float)

times_min = times_s / 60.0

if len(times_min) < 2:
    raise ValueError(
        "At least two time samples are required."
    )


# ==================================================
# Determine minutes per time pixel
# ==================================================
time_differences_min = np.diff(
    times_min
)

finite_time_differences = time_differences_min[
    np.isfinite(time_differences_min)
]

if len(finite_time_differences) == 0:
    raise ValueError(
        "Could not determine the time spacing."
    )

MINUTES_PER_PIXEL = np.median(
    finite_time_differences
)

if (
    not np.isfinite(MINUTES_PER_PIXEL)
    or MINUTES_PER_PIXEL <= 0
):
    raise ValueError(
        "The calculated minutes-per-pixel value is invalid."
    )

maximum_time_spacing_deviation = np.max(
    np.abs(
        finite_time_differences
        - MINUTES_PER_PIXEL
    )
)

print(
    "Angles to process:"
)

print(
    KEOGRAM_ANGLES_DEG
)

print(
    f"Number of angles: "
    f"{len(KEOGRAM_ANGLES_DEG)}"
)

print(
    "\nTime sampling:"
)

print(
    f"One time pixel = "
    f"{MINUTES_PER_PIXEL:.6f} minutes"
)

print(
    f"One time pixel = "
    f"{MINUTES_PER_PIXEL * 60.0:.3f} seconds"
)

print(
    f"Maximum spacing deviation = "
    f"{maximum_time_spacing_deviation:.6e} minutes"
)


# ==================================================
# Convert minute shifts to integer pixel shifts
# ==================================================
shift_min_pixels = int(
    np.ceil(
        SHIFT_MIN_MINUTES
        / MINUTES_PER_PIXEL
    )
)

shift_max_pixels = int(
    np.floor(
        SHIFT_MAX_MINUTES
        / MINUTES_PER_PIXEL
    )
)

shifts_pixels = np.arange(
    shift_min_pixels,
    shift_max_pixels + 1,
    dtype=int,
)

shifts_minutes = (
    shifts_pixels
    * MINUTES_PER_PIXEL
)

print(
    "\nShift range:"
)

print(
    f"Requested: "
    f"{SHIFT_MIN_MINUTES:.2f} to "
    f"{SHIFT_MAX_MINUTES:.2f} min"
)

print(
    f"Actual: "
    f"{shifts_minutes[0]:.2f} to "
    f"{shifts_minutes[-1]:.2f} min"
)

print(
    f"Pixel shifts: "
    f"{shifts_pixels[0]} to "
    f"{shifts_pixels[-1]}"
)


# ==================================================
# Select common time range
# ==================================================
time_mask = (
    (times_min >= CORRELATION_TIME_MIN)
    & (times_min <= CORRELATION_TIME_MAX)
)

if np.count_nonzero(time_mask) < MIN_VALID_CORRELATION_POINTS:
    raise ValueError(
        "The selected time interval has too few samples."
    )

times_correlation_min = times_min[
    time_mask
]


# ==================================================
# Save common arrays
# ==================================================
np.save(
    OUT_DIR / "angles_deg.npy",
    KEOGRAM_ANGLES_DEG,
)

np.save(
    OUT_DIR / "shifts_minutes.npy",
    shifts_minutes,
)

np.save(
    OUT_DIR / "shifts_pixels.npy",
    shifts_pixels,
)

np.save(
    OUT_DIR / "correlation_times_min.npy",
    times_correlation_min,
)


# ==================================================
# Storage for combined angle results
# ==================================================
combined_results = []

successful_angles = []
failed_angles = []


# ==================================================
# Process every angle
# ==================================================
for angle_number, angle_deg in enumerate(
    KEOGRAM_ANGLES_DEG,
    start=1,
):
    print(
        "\n"
        + "=" * 70
    )

    print(
        f"Processing angle "
        f"{angle_deg}° "
        f"({angle_number}/{len(KEOGRAM_ANGLES_DEG)})"
    )

    print(
        "=" * 70
    )

    KEOGRAM_FILE = (
        BASE_OUT_DIR
        / f"BLO_keogram_angle_{angle_deg:03d}.npy"
    )

    if not KEOGRAM_FILE.exists():
        print(
            f"WARNING: File not found: "
            f"{KEOGRAM_FILE}"
        )

        failed_angles.append(
            (
                int(angle_deg),
                "file not found",
            )
        )

        continue

    try:
        # ==================================================
        # Read keogram
        # ==================================================
        keogram_original = np.load(
            KEOGRAM_FILE
        ).astype(float)

        if keogram_original.ndim != 2:
            raise ValueError(
                f"Expected a 2D keogram, found "
                f"{keogram_original.shape}."
            )

        nt, nspace = keogram_original.shape

        if len(times_min) != nt:
            raise ValueError(
                f"Keogram has {nt} time samples, but "
                f"times_s.npy has {len(times_min)}."
            )

        # ==================================================
        # Construct distance axis
        # ==================================================
        distance_km = np.arange(
            -(nspace // 2),
            nspace // 2 + 1,
            dtype=float,
        )

        if len(distance_km) != nspace:
            distance_km = (
                np.arange(
                    nspace,
                    dtype=float,
                )
                - nspace // 2
            )

        available_distance_min = np.nanmin(
            distance_km
        )

        available_distance_max = np.nanmax(
            distance_km
        )

        if (
            CORRELATION_DISTANCE_MIN_KM
            < available_distance_min
            or CORRELATION_DISTANCE_MAX_KM
            > available_distance_max
        ):
            raise ValueError(
                f"Requested distance range "
                f"{CORRELATION_DISTANCE_MIN_KM} to "
                f"{CORRELATION_DISTANCE_MAX_KM} km "
                f"is outside available range "
                f"{available_distance_min} to "
                f"{available_distance_max} km."
            )

        # ==================================================
        # Select time interval
        # ==================================================
        keogram_time_selected = keogram_original[
            time_mask,
            :
        ]

        # ==================================================
        # Optional smoothing
        # ==================================================
        if APPLY_2D_SMOOTHING:
            keogram_uncorrected = smooth_keogram_2d(
                data=keogram_time_selected,
                sigma_time=SMOOTH_SIGMA_TIME_PIXELS,
                sigma_distance=SMOOTH_SIGMA_DISTANCE_PIXELS,
                mode=SMOOTH_MODE,
                weight_tolerance=SMOOTH_WEIGHT_TOLERANCE,
            )

            uncorrected_dataset_name = (
                f"Uncorrected, smoothed "
                f"σt={SMOOTH_SIGMA_TIME_PIXELS:g}, "
                f"σd={SMOOTH_SIGMA_DISTANCE_PIXELS:g}"
            )
        else:
            keogram_uncorrected = (
                keogram_time_selected.copy()
            )

            uncorrected_dataset_name = (
                "Uncorrected, unsmoothed"
            )

        # ==================================================
        # Create bias correction
        # ==================================================
        (
            bias_correction_keogram,
            bias_vs_time,
            bias_distance_factor,
        ) = create_bias_correction_keogram(
            times_min=times_correlation_min,
            distance_km=distance_km,
            bias_time_points_min=BIAS_TIME_POINTS_MIN,
            bias_at_max_distance=BIAS_AT_MAX_DISTANCE,
            bias_distance_min_km=BIAS_DISTANCE_MIN_KM,
            bias_distance_max_km=BIAS_DISTANCE_MAX_KM,
            distance_power=BIAS_DISTANCE_POWER,
            zero_outside_distance_range=(
                BIAS_ZERO_OUTSIDE_DISTANCE_RANGE
            ),
            zero_outside_time_range=(
                BIAS_ZERO_OUTSIDE_TIME_RANGE
            ),
            clip_negative_time_correction=(
                CLIP_NEGATIVE_TIME_CORRECTION
            ),
        )

        # ==================================================
        # Apply bias correction
        # ==================================================
        if APPLY_BIAS_CORRECTION:
            keogram_corrected = (
                keogram_uncorrected
                + bias_correction_keogram
            )

            corrected_dataset_name = (
                "Bias corrected: PCHIP time dependence, "
                f"distance power={BIAS_DISTANCE_POWER:g}"
            )
        else:
            keogram_corrected = (
                keogram_uncorrected.copy()
            )

            corrected_dataset_name = (
                "Bias correction disabled"
            )

        keogram_corrected[
            ~np.isfinite(keogram_uncorrected)
        ] = np.nan

        # ==================================================
        # Select distance rows
        # ==================================================
        selected_distance_mask = (
            (
                distance_km
                >= CORRELATION_DISTANCE_MIN_KM
            )
            & (
                distance_km
                <= CORRELATION_DISTANCE_MAX_KM
            )
        )

        selected_distances = distance_km[
            selected_distance_mask
        ]

        if len(selected_distances) == 0:
            raise ValueError(
                "No rows lie inside the selected distance range."
            )

        # ==================================================
        # Per-angle folders
        # ==================================================
        angle_uncorrected_dir = (
            UNCORRECTED_OUT_DIR
            / f"angle_{angle_deg:03d}"
        )

        angle_corrected_dir = (
            CORRECTED_OUT_DIR
            / f"angle_{angle_deg:03d}"
        )

        angle_uncorrected_dir.mkdir(
            parents=True,
            exist_ok=True,
        )

        angle_corrected_dir.mkdir(
            parents=True,
            exist_ok=True,
        )

        # ==================================================
        # Uncorrected map
        # ==================================================
        (
            correlation_map_uncorrected,
            selected_distances_uncorrected,
            reference_index_uncorrected,
        ) = calculate_correlation_map(
            keogram=keogram_uncorrected,
            distance_km=distance_km,
            reference_distance_km=0.0,
            selected_distance_mask=selected_distance_mask,
            shifts_pixels=shifts_pixels,
            minimum_points=MIN_VALID_CORRELATION_POINTS,
        )

        (
            best_shift_uncorrected_minutes,
            maximum_correlation_uncorrected,
        ) = find_best_shifts(
            correlation_map=correlation_map_uncorrected,
            shifts_minutes=shifts_minutes,
        )

        uncorrected_slope_results = (
            fit_maximum_correlation_slope(
                distances_km=(
                    selected_distances_uncorrected
                ),
                best_shift_minutes=(
                    best_shift_uncorrected_minutes
                ),
                maximum_correlation=(
                    maximum_correlation_uncorrected
                ),
                minimum_correlation=(
                    MIN_CORRELATION_FOR_SLOPE
                ),
                distance_min_km=(
                    SLOPE_FIT_DISTANCE_MIN_KM
                ),
                distance_max_km=(
                    SLOPE_FIT_DISTANCE_MAX_KM
                ),
            )
        )

        # ==================================================
        # Corrected map
        # ==================================================
        (
            correlation_map_corrected,
            selected_distances_corrected,
            reference_index_corrected,
        ) = calculate_correlation_map(
            keogram=keogram_corrected,
            distance_km=distance_km,
            reference_distance_km=0.0,
            selected_distance_mask=selected_distance_mask,
            shifts_pixels=shifts_pixels,
            minimum_points=MIN_VALID_CORRELATION_POINTS,
        )

        (
            best_shift_corrected_minutes,
            maximum_correlation_corrected,
        ) = find_best_shifts(
            correlation_map=correlation_map_corrected,
            shifts_minutes=shifts_minutes,
        )

        corrected_slope_results = (
            fit_maximum_correlation_slope(
                distances_km=(
                    selected_distances_corrected
                ),
                best_shift_minutes=(
                    best_shift_corrected_minutes
                ),
                maximum_correlation=(
                    maximum_correlation_corrected
                ),
                minimum_correlation=(
                    MIN_CORRELATION_FOR_SLOPE
                ),
                distance_min_km=(
                    SLOPE_FIT_DISTANCE_MIN_KM
                ),
                distance_max_km=(
                    SLOPE_FIT_DISTANCE_MAX_KM
                ),
            )
        )

        # ==================================================
        # Plot maps
        # ==================================================
        uncorrected_plot_file = (
            angle_uncorrected_dir
            / (
                f"correlation_minutes_"
                f"{angle_deg:03d}_uncorrected.png"
            )
        )

        corrected_plot_file = (
            angle_corrected_dir
            / (
                f"correlation_minutes_"
                f"{angle_deg:03d}_corrected.png"
            )
        )

        plot_correlation_map(
            angle_deg=angle_deg,
            correlation_map=(
                correlation_map_uncorrected
            ),
            shifts_minutes=shifts_minutes,
            selected_distances=(
                selected_distances_uncorrected
            ),
            best_shift_minutes=(
                best_shift_uncorrected_minutes
            ),
            maximum_correlation=(
                maximum_correlation_uncorrected
            ),
            slope_results=(
                uncorrected_slope_results
            ),
            dataset_name=uncorrected_dataset_name,
            output_file=uncorrected_plot_file,
        )

        plot_correlation_map(
            angle_deg=angle_deg,
            correlation_map=(
                correlation_map_corrected
            ),
            shifts_minutes=shifts_minutes,
            selected_distances=(
                selected_distances_corrected
            ),
            best_shift_minutes=(
                best_shift_corrected_minutes
            ),
            maximum_correlation=(
                maximum_correlation_corrected
            ),
            slope_results=(
                corrected_slope_results
            ),
            dataset_name=corrected_dataset_name,
            output_file=corrected_plot_file,
        )

        # ==================================================
        # Save arrays
        # ==================================================
        np.save(
            angle_uncorrected_dir
            / (
                f"correlation_"
                f"{angle_deg:03d}_uncorrected.npy"
            ),
            correlation_map_uncorrected,
        )

        np.save(
            angle_uncorrected_dir
            / (
                f"best_shift_minutes_"
                f"{angle_deg:03d}_uncorrected.npy"
            ),
            best_shift_uncorrected_minutes,
        )

        np.save(
            angle_uncorrected_dir
            / (
                f"maximum_correlation_"
                f"{angle_deg:03d}_uncorrected.npy"
            ),
            maximum_correlation_uncorrected,
        )

        np.save(
            angle_corrected_dir
            / (
                f"correlation_"
                f"{angle_deg:03d}_corrected.npy"
            ),
            correlation_map_corrected,
        )

        np.save(
            angle_corrected_dir
            / (
                f"best_shift_minutes_"
                f"{angle_deg:03d}_corrected.npy"
            ),
            best_shift_corrected_minutes,
        )

        np.save(
            angle_corrected_dir
            / (
                f"maximum_correlation_"
                f"{angle_deg:03d}_corrected.npy"
            ),
            maximum_correlation_corrected,
        )

        np.save(
            angle_uncorrected_dir
            / "distances_km.npy",
            selected_distances_uncorrected,
        )

        np.save(
            angle_corrected_dir
            / "distances_km.npy",
            selected_distances_corrected,
        )

        # ==================================================
        # Save ridge CSV files
        # ==================================================
        uncorrected_summary = np.column_stack(
            [
                selected_distances_uncorrected,
                best_shift_uncorrected_minutes,
                maximum_correlation_uncorrected,
            ]
        )

        corrected_summary = np.column_stack(
            [
                selected_distances_corrected,
                best_shift_corrected_minutes,
                maximum_correlation_corrected,
            ]
        )

        uncorrected_csv = (
            angle_uncorrected_dir
            / (
                f"ridge_"
                f"{angle_deg:03d}_uncorrected.csv"
            )
        )

        corrected_csv = (
            angle_corrected_dir
            / (
                f"ridge_"
                f"{angle_deg:03d}_corrected.csv"
            )
        )

        np.savetxt(
            uncorrected_csv,
            uncorrected_summary,
            delimiter=",",
            header=(
                "distance_km,"
                "best_shift_minutes,"
                "maximum_correlation"
            ),
            comments="",
        )

        np.savetxt(
            corrected_csv,
            corrected_summary,
            delimiter=",",
            header=(
                "distance_km,"
                "best_shift_minutes,"
                "maximum_correlation"
            ),
            comments="",
        )

        # ==================================================
        # Save individual slope report
        # ==================================================
        slope_output_file = (
            SLOPE_REPORT_OUT_DIR
            / (
                f"correlation_slope_"
                f"{angle_deg:03d}.txt"
            )
        )

        with open(
            slope_output_file,
            "w",
            encoding="utf-8",
        ) as file:
            file.write(
                f"Angle: {angle_deg} degrees\n"
            )

            file.write(
                f"Minutes per pixel: "
                f"{MINUTES_PER_PIXEL:.10f}\n\n"
            )

            file.write(
                "UNCORRECTED\n"
            )

            for key, value in (
                uncorrected_slope_results.items()
            ):
                if key != "fit_mask":
                    file.write(
                        f"{key}: {value}\n"
                    )

            file.write(
                "\nCORRECTED\n"
            )

            for key, value in (
                corrected_slope_results.items()
            ):
                if key != "fit_mask":
                    file.write(
                        f"{key}: {value}\n"
                    )

        # ==================================================
        # Add to combined summary
        # ==================================================
        combined_results.append(
            [
                float(angle_deg),

                uncorrected_slope_results[
                    "slope_min_per_km"
                ],
                uncorrected_slope_results[
                    "intercept_minutes"
                ],
                uncorrected_slope_results[
                    "speed_km_per_min"
                ],
                uncorrected_slope_results[
                    "speed_m_per_s"
                ],
                uncorrected_slope_results[
                    "r_squared"
                ],
                uncorrected_slope_results[
                    "number_of_fit_points"
                ],

                corrected_slope_results[
                    "slope_min_per_km"
                ],
                corrected_slope_results[
                    "intercept_minutes"
                ],
                corrected_slope_results[
                    "speed_km_per_min"
                ],
                corrected_slope_results[
                    "speed_m_per_s"
                ],
                corrected_slope_results[
                    "r_squared"
                ],
                corrected_slope_results[
                    "number_of_fit_points"
                ],
            ]
        )

        successful_angles.append(
            int(angle_deg)
        )

        print_slope_results(
            angle_deg=angle_deg,
            label=(
                "Uncorrected maximum-correlation ridge"
            ),
            results=uncorrected_slope_results,
        )

        print_slope_results(
            angle_deg=angle_deg,
            label=(
                "Corrected maximum-correlation ridge"
            ),
            results=corrected_slope_results,
        )

    except Exception as error:
        print(
            f"ERROR processing {angle_deg}°:"
        )

        print(
            error
        )

        failed_angles.append(
            (
                int(angle_deg),
                str(error),
            )
        )


# ==================================================
# Save combined angle summary
# ==================================================
combined_results = np.asarray(
    combined_results,
    dtype=float,
)

combined_csv = (
    OUT_DIR
    / "all_angle_slope_summary.csv"
)

combined_npy = (
    OUT_DIR
    / "all_angle_slope_summary.npy"
)

combined_header = (
    "angle_deg,"
    "uncorrected_slope_min_per_km,"
    "uncorrected_intercept_minutes,"
    "uncorrected_speed_km_per_min,"
    "uncorrected_speed_m_per_s,"
    "uncorrected_r_squared,"
    "uncorrected_number_of_fit_points,"
    "corrected_slope_min_per_km,"
    "corrected_intercept_minutes,"
    "corrected_speed_km_per_min,"
    "corrected_speed_m_per_s,"
    "corrected_r_squared,"
    "corrected_number_of_fit_points"
)

if combined_results.size > 0:
    np.savetxt(
        combined_csv,
        combined_results,
        delimiter=",",
        header=combined_header,
        comments="",
    )

    np.save(
        combined_npy,
        combined_results,
    )


# ==================================================
# Plot speed versus angle
# ==================================================
if combined_results.size > 0:
    angles_processed = combined_results[
        :,
        0,
    ]

    uncorrected_speed_m_per_s = combined_results[
        :,
        4,
    ]

    corrected_speed_m_per_s = combined_results[
        :,
        10,
    ]

    fig_speed, ax_speed = plt.subplots(
        figsize=(10, 5)
    )

    ax_speed.plot(
        angles_processed,
        uncorrected_speed_m_per_s,
        marker="o",
        linewidth=1.5,
        label="Uncorrected",
    )

    ax_speed.plot(
        angles_processed,
        corrected_speed_m_per_s,
        marker="o",
        linewidth=1.5,
        label="Corrected",
    )

    ax_speed.axhline(
        0.0,
        color="black",
        linestyle="--",
        linewidth=1.0,
    )

    ax_speed.set_xlim(
        KEOGRAM_ANGLES_DEG.min(),
        KEOGRAM_ANGLES_DEG.max(),
    )

    ax_speed.set_xlabel(
        "Keogram angle (degrees)"
    )

    ax_speed.set_ylabel(
        "Equivalent signed speed (m/s)"
    )

    ax_speed.set_title(
        "Maximum-Correlation Ridge Speed versus Angle"
    )

    ax_speed.grid(
        True,
        alpha=0.3,
    )

    ax_speed.legend(
        loc="best",
    )

    plt.tight_layout()

    speed_angle_plot = (
        OUT_DIR
        / "speed_vs_angle.png"
    )

    fig_speed.savefig(
        speed_angle_plot,
        dpi=DPI,
        bbox_inches="tight",
    )

    plt.close(
        fig_speed
    )


# ==================================================
# Plot R-squared versus angle
# ==================================================
if combined_results.size > 0:
    uncorrected_r_squared = combined_results[
        :,
        5,
    ]

    corrected_r_squared = combined_results[
        :,
        11,
    ]

    fig_r2, ax_r2 = plt.subplots(
        figsize=(10, 5)
    )

    ax_r2.plot(
        angles_processed,
        uncorrected_r_squared,
        marker="o",
        linewidth=1.5,
        label="Uncorrected",
    )

    ax_r2.plot(
        angles_processed,
        corrected_r_squared,
        marker="o",
        linewidth=1.5,
        label="Corrected",
    )

    ax_r2.set_xlim(
        KEOGRAM_ANGLES_DEG.min(),
        KEOGRAM_ANGLES_DEG.max(),
    )

    ax_r2.set_ylim(
        0.0,
        1.0,
    )

    ax_r2.set_xlabel(
        "Keogram angle (degrees)"
    )

    ax_r2.set_ylabel(
        "R²"
    )

    ax_r2.set_title(
        "Maximum-Correlation Ridge Fit R² versus Angle"
    )

    ax_r2.grid(
        True,
        alpha=0.3,
    )

    ax_r2.legend(
        loc="best",
    )

    plt.tight_layout()

    r2_angle_plot = (
        OUT_DIR
        / "r_squared_vs_angle.png"
    )

    fig_r2.savefig(
        r2_angle_plot,
        dpi=DPI,
        bbox_inches="tight",
    )

    plt.close(
        fig_r2
    )


# ==================================================
# Save processing log
# ==================================================
processing_log_file = (
    OUT_DIR
    / "processing_log.txt"
)

with open(
    processing_log_file,
    "w",
    encoding="utf-8",
) as file:
    file.write(
        "Requested angles:\n"
    )

    file.write(
        ", ".join(
            str(angle)
            for angle in KEOGRAM_ANGLES_DEG
        )
        + "\n\n"
    )

    file.write(
        f"Number requested: "
        f"{len(KEOGRAM_ANGLES_DEG)}\n"
    )

    file.write(
        f"Number successful: "
        f"{len(successful_angles)}\n"
    )

    file.write(
        f"Number failed: "
        f"{len(failed_angles)}\n\n"
    )

    file.write(
        "Successful angles:\n"
    )

    file.write(
        ", ".join(
            str(angle)
            for angle in successful_angles
        )
        + "\n\n"
    )

    file.write(
        "Failed angles:\n"
    )

    if failed_angles:
        for angle_deg, reason in failed_angles:
            file.write(
                f"{angle_deg}: {reason}\n"
            )
    else:
        file.write(
            "None\n"
        )


# ==================================================
# Print final summary
# ==================================================
print(
    "\n"
    + "=" * 70
)

print(
    "ALL-ANGLE PROCESSING COMPLETE"
)

print(
    "=" * 70
)

print(
    "Angles requested:"
)

print(
    KEOGRAM_ANGLES_DEG
)

print(
    f"\nNumber requested: "
    f"{len(KEOGRAM_ANGLES_DEG)}"
)

print(
    f"Number successful: "
    f"{len(successful_angles)}"
)

print(
    f"Number failed: "
    f"{len(failed_angles)}"
)

print(
    "\nSuccessful angles:"
)

print(
    successful_angles
)

if failed_angles:
    print(
        "\nFailed angles:"
    )

    for angle_deg, reason in failed_angles:
        print(
            f"  {angle_deg}°: {reason}"
        )

print(
    "\nCombined CSV:"
)

print(
    combined_csv
)

print(
    "\nOutput folder:"
)

print(
    OUT_DIR.resolve()
)
```

## Additional local maxima

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

from scipy.ndimage import gaussian_filter
from scipy.interpolate import PchipInterpolator
from scipy.signal import find_peaks


# ==================================================
# Settings
# ==================================================
BASE_OUT_DIR = Path(
    "BLO_perturbation_keogram_outputs_new"
)

# Process every directional keogram:
# 0, 5, 10, ..., 175 degrees
KEOGRAM_ANGLES_DEG = np.arange(
    0,
    176,
    5,
    dtype=int,
)

# Time range used for correlation
CORRELATION_TIME_MIN = 0.0
CORRELATION_TIME_MAX = 300.0

# Distance range included in the correlation map
CORRELATION_DISTANCE_MIN_KM = 0.0
CORRELATION_DISTANCE_MAX_KM = 250.0

# Shift range in minutes
SHIFT_MIN_MINUTES = -200.0
SHIFT_MAX_MINUTES = 200.0

# Minimum number of overlapping time samples
MIN_VALID_CORRELATION_POINTS = 20

# Minimum maximum-correlation value used in the slope fit
MIN_CORRELATION_FOR_SLOPE = 0.50

# Distance range used in the ridge fit
SLOPE_FIT_DISTANCE_MIN_KM = 0.0
SLOPE_FIT_DISTANCE_MAX_KM = 250.0

DPI = 200

CORRELATION_CMAP = "RdBu_r"

# Correlation color scale
CORRELATION_VMIN = 0.70
CORRELATION_VMAX = 1.00

# Local maxima search on the 0 km correlation-shift profile.
# The left maximum is searched at negative shifts and the right
# maximum is searched at positive shifts. The zero-shift point is
# therefore excluded automatically.
LOCAL_MAXIMUM_MIN_PROMINENCE = 0.0
LOCAL_MAXIMUM_MIN_DISTANCE_PIXELS = 1


# ==================================================
# Optional 2D smoothing
# ==================================================
APPLY_2D_SMOOTHING = True

# Keogram dimensions:
# axis 0 = time
# axis 1 = distance
SMOOTH_SIGMA_TIME_PIXELS = 1.5
SMOOTH_SIGMA_DISTANCE_PIXELS = 2.0

SMOOTH_MODE = "nearest"
SMOOTH_WEIGHT_TOLERANCE = 1e-12


# ==================================================
# Bias-correction settings
# ==================================================
APPLY_BIAS_CORRECTION = False

BIAS_TIME_POINTS_MIN = np.array(
    [
        0.0,
        150.0,
        300.0,
    ],
    dtype=float,
)

BIAS_AT_MAX_DISTANCE = np.array(
    [
        0.20,
        0.04,
        -0.01,
    ],
    dtype=float,
)

BIAS_DISTANCE_MIN_KM = 0.0
BIAS_DISTANCE_MAX_KM = 250.0

# 1.0 = linear distance dependence
# 2.0 = quadratic distance dependence
BIAS_DISTANCE_POWER = 1.0

BIAS_ZERO_OUTSIDE_DISTANCE_RANGE = True
BIAS_ZERO_OUTSIDE_TIME_RANGE = True
CLIP_NEGATIVE_TIME_CORRECTION = False


# ==================================================
# Filename-safe numeric label
# ==================================================
def safe_number_label(value):
    value_float = float(value)

    if value_float.is_integer():
        return str(
            int(value_float)
        )

    return (
        f"{value_float:g}"
        .replace("-", "m")
        .replace(".", "p")
    )


# ==================================================
# NaN-aware 2D Gaussian smoothing
# ==================================================
def smooth_keogram_2d(
    data,
    sigma_time,
    sigma_distance,
    mode="nearest",
    weight_tolerance=1e-12,
):
    """
    Apply NaN-aware Gaussian smoothing to a 2D keogram.
    """
    data = np.asarray(
        data,
        dtype=float,
    )

    if data.ndim != 2:
        raise ValueError(
            "The keogram must be two-dimensional."
        )

    if sigma_time < 0 or sigma_distance < 0:
        raise ValueError(
            "Smoothing sigma values must be nonnegative."
        )

    if sigma_time == 0 and sigma_distance == 0:
        return data.copy()

    valid_mask = np.isfinite(
        data
    )

    data_filled = np.where(
        valid_mask,
        data,
        0.0,
    )

    weights = valid_mask.astype(
        float
    )

    sigma = (
        sigma_time,
        sigma_distance,
    )

    smoothed_values = gaussian_filter(
        data_filled,
        sigma=sigma,
        mode=mode,
    )

    smoothed_weights = gaussian_filter(
        weights,
        sigma=sigma,
        mode=mode,
    )

    smoothed = np.full(
        data.shape,
        np.nan,
        dtype=float,
    )

    valid_output = (
        smoothed_weights
        > weight_tolerance
    )

    smoothed[valid_output] = (
        smoothed_values[valid_output]
        / smoothed_weights[valid_output]
    )

    return smoothed


# ==================================================
# Create 2D bias-correction keogram
# ==================================================
def create_bias_correction_keogram(
    times_min,
    distance_km,
    bias_time_points_min,
    bias_at_max_distance,
    bias_distance_min_km,
    bias_distance_max_km,
    distance_power=1.0,
    zero_outside_distance_range=True,
    zero_outside_time_range=True,
    clip_negative_time_correction=False,
):
    """
    Create the time x distance additive bias field.
    """
    times_min = np.asarray(
        times_min,
        dtype=float,
    )

    distance_km = np.asarray(
        distance_km,
        dtype=float,
    )

    bias_time_points_min = np.asarray(
        bias_time_points_min,
        dtype=float,
    )

    bias_at_max_distance = np.asarray(
        bias_at_max_distance,
        dtype=float,
    )

    if (
        len(bias_time_points_min)
        != len(bias_at_max_distance)
    ):
        raise ValueError(
            "Bias time points and bias values must have "
            "the same length."
        )

    if not np.all(
        np.diff(bias_time_points_min) > 0
    ):
        raise ValueError(
            "BIAS_TIME_POINTS_MIN must be strictly increasing."
        )

    if bias_distance_max_km <= bias_distance_min_km:
        raise ValueError(
            "The maximum bias distance must exceed "
            "the minimum bias distance."
        )

    if distance_power <= 0:
        raise ValueError(
            "BIAS_DISTANCE_POWER must be positive."
        )

    # ==================================================
    # PCHIP time dependence
    # ==================================================
    interpolator = PchipInterpolator(
        bias_time_points_min,
        bias_at_max_distance,
        extrapolate=False,
    )

    bias_vs_time = interpolator(
        times_min
    )

    if zero_outside_time_range:
        inside_time = (
            (times_min >= bias_time_points_min[0])
            & (times_min <= bias_time_points_min[-1])
        )

        bias_vs_time = np.where(
            inside_time,
            bias_vs_time,
            0.0,
        )
    else:
        bias_vs_time = np.where(
            times_min < bias_time_points_min[0],
            bias_at_max_distance[0],
            bias_vs_time,
        )

        bias_vs_time = np.where(
            times_min > bias_time_points_min[-1],
            bias_at_max_distance[-1],
            bias_vs_time,
        )

    if clip_negative_time_correction:
        bias_vs_time = np.maximum(
            bias_vs_time,
            0.0,
        )

    # ==================================================
    # Distance dependence
    # ==================================================
    normalized_distance = (
        distance_km
        - bias_distance_min_km
    ) / (
        bias_distance_max_km
        - bias_distance_min_km
    )

    if zero_outside_distance_range:
        inside_distance = (
            (distance_km >= bias_distance_min_km)
            & (distance_km <= bias_distance_max_km)
        )

        normalized_distance = np.where(
            inside_distance,
            normalized_distance,
            0.0,
        )
    else:
        normalized_distance = np.clip(
            normalized_distance,
            0.0,
            1.0,
        )

    distance_factor = (
        normalized_distance ** distance_power
    )

    correction_keogram = (
        bias_vs_time[:, np.newaxis]
        * distance_factor[np.newaxis, :]
    )

    return (
        correction_keogram,
        bias_vs_time,
        distance_factor,
    )


# ==================================================
# NaN-aware Pearson correlation
# ==================================================
def nan_pearson_correlation(
    reference,
    comparison,
    minimum_points=20,
):
    """
    Calculate Pearson correlation using finite overlapping values.
    """
    reference = np.asarray(
        reference,
        dtype=float,
    )

    comparison = np.asarray(
        comparison,
        dtype=float,
    )

    if reference.shape != comparison.shape:
        raise ValueError(
            "The two arrays must have the same shape."
        )

    valid = (
        np.isfinite(reference)
        & np.isfinite(comparison)
    )

    if np.count_nonzero(valid) < minimum_points:
        return np.nan

    reference_valid = reference[
        valid
    ]

    comparison_valid = comparison[
        valid
    ]

    reference_centered = (
        reference_valid
        - np.mean(reference_valid)
    )

    comparison_centered = (
        comparison_valid
        - np.mean(comparison_valid)
    )

    denominator = np.sqrt(
        np.sum(reference_centered ** 2)
        * np.sum(comparison_centered ** 2)
    )

    if (
        not np.isfinite(denominator)
        or denominator <= 0
    ):
        return np.nan

    correlation = (
        np.sum(
            reference_centered
            * comparison_centered
        )
        / denominator
    )

    return np.clip(
        correlation,
        -1.0,
        1.0,
    )


# ==================================================
# Correlation for one integer-pixel shift
# ==================================================
def calculate_shifted_correlation(
    reference_row,
    comparison_row,
    shift_pixels,
    minimum_points=20,
):
    """
    Shift comparison_row and correlate it with reference_row.

    Positive shift:
        comparison row moves toward later times.

    Negative shift:
        comparison row moves toward earlier times.

    No circular wrapping is used.
    """
    reference_row = np.asarray(
        reference_row,
        dtype=float,
    )

    comparison_row = np.asarray(
        comparison_row,
        dtype=float,
    )

    if reference_row.shape != comparison_row.shape:
        raise ValueError(
            "Reference and comparison rows must have "
            "the same shape."
        )

    ntime = len(
        reference_row
    )

    if abs(shift_pixels) >= ntime:
        return np.nan

    if shift_pixels > 0:
        reference_overlap = reference_row[
            shift_pixels:
        ]

        comparison_overlap = comparison_row[
            :-shift_pixels
        ]

    elif shift_pixels < 0:
        shift_abs = abs(
            shift_pixels
        )

        reference_overlap = reference_row[
            :-shift_abs
        ]

        comparison_overlap = comparison_row[
            shift_abs:
        ]

    else:
        reference_overlap = reference_row
        comparison_overlap = comparison_row

    return nan_pearson_correlation(
        reference=reference_overlap,
        comparison=comparison_overlap,
        minimum_points=minimum_points,
    )


# ==================================================
# Calculate complete correlation map
# ==================================================
def calculate_correlation_map(
    keogram,
    distance_km,
    reference_distance_km,
    selected_distance_mask,
    shifts_pixels,
    minimum_points=20,
):
    """
    Correlate every selected distance row with the reference row
    for every requested shift.
    """
    keogram = np.asarray(
        keogram,
        dtype=float,
    )

    distance_km = np.asarray(
        distance_km,
        dtype=float,
    )

    shifts_pixels = np.asarray(
        shifts_pixels,
        dtype=int,
    )

    reference_index = np.argmin(
        np.abs(
            distance_km
            - reference_distance_km
        )
    )

    selected_indices = np.where(
        selected_distance_mask
    )[0]

    selected_distances = distance_km[
        selected_indices
    ]

    reference_row = keogram[
        :,
        reference_index,
    ]

    correlation_map = np.full(
        (
            len(selected_indices),
            len(shifts_pixels),
        ),
        np.nan,
        dtype=float,
    )

    for output_row_index, keogram_row_index in enumerate(
        selected_indices
    ):
        comparison_row = keogram[
            :,
            keogram_row_index,
        ]

        for shift_index, shift_pixels in enumerate(
            shifts_pixels
        ):
            correlation_map[
                output_row_index,
                shift_index,
            ] = calculate_shifted_correlation(
                reference_row=reference_row,
                comparison_row=comparison_row,
                shift_pixels=int(shift_pixels),
                minimum_points=minimum_points,
            )

    return (
        correlation_map,
        selected_distances,
        reference_index,
    )


# ==================================================
# Maximum-correlation shift for each row
# ==================================================
def find_best_shifts(
    correlation_map,
    shifts_minutes,
):
    """
    Find the shift in minutes having the largest correlation
    for every distance row.
    """
    correlation_map = np.asarray(
        correlation_map,
        dtype=float,
    )

    shifts_minutes = np.asarray(
        shifts_minutes,
        dtype=float,
    )

    best_shift_minutes = np.full(
        correlation_map.shape[0],
        np.nan,
        dtype=float,
    )

    maximum_correlation = np.full(
        correlation_map.shape[0],
        np.nan,
        dtype=float,
    )

    for row_index in range(
        correlation_map.shape[0]
    ):
        correlation_row = correlation_map[
            row_index,
            :
        ]

        finite = np.isfinite(
            correlation_row
        )

        if not np.any(finite):
            continue

        finite_indices = np.where(
            finite
        )[0]

        local_max_index = np.argmax(
            correlation_row[finite]
        )

        full_max_index = finite_indices[
            local_max_index
        ]

        best_shift_minutes[row_index] = (
            shifts_minutes[full_max_index]
        )

        maximum_correlation[row_index] = (
            correlation_row[full_max_index]
        )

    return (
        best_shift_minutes,
        maximum_correlation,
    )


# ==================================================
# Extract the correlation-shift profile at 0 km
# ==================================================
def extract_zero_km_correlation_profile(
    correlation_map,
    selected_distances,
):
    """
    Return the correlation-versus-shift profile for the selected
    distance row nearest 0 km.
    """
    correlation_map = np.asarray(
        correlation_map,
        dtype=float,
    )

    selected_distances = np.asarray(
        selected_distances,
        dtype=float,
    )

    if correlation_map.ndim != 2:
        raise ValueError(
            "correlation_map must be two-dimensional."
        )

    if correlation_map.shape[0] != len(selected_distances):
        raise ValueError(
            "The number of correlation-map rows must equal "
            "the number of selected distances."
        )

    zero_distance_row_index = int(
        np.argmin(
            np.abs(selected_distances)
        )
    )

    zero_distance_km = float(
        selected_distances[zero_distance_row_index]
    )

    zero_km_profile = correlation_map[
        zero_distance_row_index,
        :
    ].copy()

    return (
        zero_km_profile,
        zero_distance_row_index,
        zero_distance_km,
    )


# ==================================================
# Find strongest local maximum on each side of zero
# ==================================================
def find_left_and_right_local_maxima(
    correlation_profile,
    shifts_minutes,
    minimum_prominence=0.0,
    minimum_distance_pixels=1,
):
    """
    Find the strongest finite local maximum at negative shift
    and the strongest finite local maximum at positive shift.

    The point at zero shift is not included in either search.

    If scipy.signal.find_peaks finds no strict local maximum on one
    side, the largest finite value on that side is used as a fallback.
    """
    correlation_profile = np.asarray(
        correlation_profile,
        dtype=float,
    )

    shifts_minutes = np.asarray(
        shifts_minutes,
        dtype=float,
    )

    if correlation_profile.shape != shifts_minutes.shape:
        raise ValueError(
            "correlation_profile and shifts_minutes must "
            "have the same shape."
        )

    def find_one_side(side_mask):
        side_indices = np.where(
            side_mask
            & np.isfinite(correlation_profile)
        )[0]

        if len(side_indices) == 0:
            return {
                "index": -1,
                "shift_minutes": np.nan,
                "correlation": np.nan,
                "used_fallback": False,
            }

        side_values = correlation_profile[
            side_indices
        ]

        local_indices, properties = find_peaks(
            side_values,
            prominence=minimum_prominence,
            distance=minimum_distance_pixels,
        )

        used_fallback = False

        if len(local_indices) > 0:
            local_values = side_values[
                local_indices
            ]

            strongest_local_position = local_indices[
                int(np.nanargmax(local_values))
            ]

            full_index = int(
                side_indices[strongest_local_position]
            )
        else:
            used_fallback = True

            strongest_side_position = int(
                np.nanargmax(side_values)
            )

            full_index = int(
                side_indices[strongest_side_position]
            )

        return {
            "index": full_index,
            "shift_minutes": float(
                shifts_minutes[full_index]
            ),
            "correlation": float(
                correlation_profile[full_index]
            ),
            "used_fallback": used_fallback,
        }

    left_result = find_one_side(
        shifts_minutes < 0.0
    )

    right_result = find_one_side(
        shifts_minutes > 0.0
    )

    return {
        "left": left_result,
        "right": right_result,
    }


# ==================================================
# Fit maximum-correlation ridge
# ==================================================
def fit_maximum_correlation_slope(
    distances_km,
    best_shift_minutes,
    maximum_correlation,
    minimum_correlation=0.5,
    distance_min_km=None,
    distance_max_km=None,
):
    """
    Fit:

        shift_minutes = slope * distance_km + intercept
    """
    distances_km = np.asarray(
        distances_km,
        dtype=float,
    )

    best_shift_minutes = np.asarray(
        best_shift_minutes,
        dtype=float,
    )

    maximum_correlation = np.asarray(
        maximum_correlation,
        dtype=float,
    )

    fit_mask = (
        np.isfinite(distances_km)
        & np.isfinite(best_shift_minutes)
        & np.isfinite(maximum_correlation)
        & (
            maximum_correlation
            >= minimum_correlation
        )
    )

    if distance_min_km is not None:
        fit_mask &= (
            distances_km
            >= distance_min_km
        )

    if distance_max_km is not None:
        fit_mask &= (
            distances_km
            <= distance_max_km
        )

    number_of_points = np.count_nonzero(
        fit_mask
    )

    if number_of_points < 2:
        return {
            "slope_min_per_km": np.nan,
            "intercept_minutes": np.nan,
            "speed_km_per_min": np.nan,
            "speed_m_per_s": np.nan,
            "r_squared": np.nan,
            "number_of_fit_points": number_of_points,
            "fit_mask": fit_mask,
        }

    fit_distances = distances_km[
        fit_mask
    ]

    fit_shifts = best_shift_minutes[
        fit_mask
    ]

    slope_min_per_km, intercept_minutes = np.polyfit(
        fit_distances,
        fit_shifts,
        deg=1,
    )

    predicted_shifts = (
        slope_min_per_km
        * fit_distances
        + intercept_minutes
    )

    residual_sum_squares = np.sum(
        (
            fit_shifts
            - predicted_shifts
        ) ** 2
    )

    total_sum_squares = np.sum(
        (
            fit_shifts
            - np.mean(fit_shifts)
        ) ** 2
    )

    if total_sum_squares > 0:
        r_squared = (
            1.0
            - residual_sum_squares
            / total_sum_squares
        )
    else:
        r_squared = np.nan

    if np.isclose(
        slope_min_per_km,
        0.0,
    ):
        speed_km_per_min = np.nan
        speed_m_per_s = np.nan
    else:
        speed_km_per_min = (
            1.0
            / slope_min_per_km
        )

        speed_m_per_s = (
            speed_km_per_min
            * 1000.0
            / 60.0
        )

    return {
        "slope_min_per_km": slope_min_per_km,
        "intercept_minutes": intercept_minutes,
        "speed_km_per_min": speed_km_per_min,
        "speed_m_per_s": speed_m_per_s,
        "r_squared": r_squared,
        "number_of_fit_points": number_of_points,
        "fit_mask": fit_mask,
    }


# ==================================================
# Plot correlation map, fitted ridge, and 0 km profile
# ==================================================
def plot_correlation_map(
    angle_deg,
    correlation_map,
    shifts_minutes,
    selected_distances,
    best_shift_minutes,
    maximum_correlation,
    slope_results,
    zero_km_correlation_profile,
    zero_km_distance,
    zero_km_local_maxima,
    dataset_name,
    output_file,
):
    """
    Plot:
      1. correlation versus time shift and distance;
      2. correlation versus shift for the 0 km row.

    The strongest local maximum on the negative-shift side and the
    strongest local maximum on the positive-shift side are marked.
    """
    fig, (
        ax_map,
        ax_profile,
    ) = plt.subplots(
        nrows=2,
        ncols=1,
        figsize=(10, 9),
        gridspec_kw={
            "height_ratios": [3.0, 1.25],
        },
        sharex=True,
    )

    if len(shifts_minutes) > 1:
        shift_spacing = np.median(
            np.diff(shifts_minutes)
        )
    else:
        shift_spacing = 1.0

    if len(selected_distances) > 1:
        distance_spacing = np.median(
            np.diff(selected_distances)
        )
    else:
        distance_spacing = 1.0

    im = ax_map.imshow(
        correlation_map,
        origin="lower",
        aspect="auto",
        interpolation="nearest",
        cmap=CORRELATION_CMAP,
        vmin=CORRELATION_VMIN,
        vmax=CORRELATION_VMAX,
        extent=[
            shifts_minutes[0]
            - shift_spacing / 2.0,
            shifts_minutes[-1]
            + shift_spacing / 2.0,
            selected_distances[0]
            - distance_spacing / 2.0,
            selected_distances[-1]
            + distance_spacing / 2.0,
        ],
    )

    cbar = fig.colorbar(
        im,
        ax=ax_map,
        fraction=0.046,
        pad=0.04,
    )

    cbar.set_label(
        "Pearson correlation with 0 km row"
    )

    ax_map.axvline(
        0.0,
        color="black",
        linestyle="--",
        linewidth=1.2,
        label="Zero shift",
    )

    ax_map.axhline(
        zero_km_distance,
        color="white",
        linestyle=":",
        linewidth=1.5,
        label=f"Profile row ({zero_km_distance:g} km)",
    )

    ridge_mask = (
        np.isfinite(best_shift_minutes)
        & np.isfinite(maximum_correlation)
        & (
            maximum_correlation
            >= MIN_CORRELATION_FOR_SLOPE
        )
    )

    if np.any(ridge_mask):
        ax_map.plot(
            best_shift_minutes[ridge_mask],
            selected_distances[ridge_mask],
            color="black",
            linewidth=1.2,
            label="Maximum correlation",
        )

        ax_map.scatter(
            best_shift_minutes[ridge_mask],
            selected_distances[ridge_mask],
            color="black",
            s=8,
        )

    slope = slope_results[
        "slope_min_per_km"
    ]

    intercept = slope_results[
        "intercept_minutes"
    ]

    r_squared = slope_results[
        "r_squared"
    ]

    speed_m_per_s = slope_results[
        "speed_m_per_s"
    ]

    if np.isfinite(slope):
        fit_distances = np.linspace(
            SLOPE_FIT_DISTANCE_MIN_KM,
            SLOPE_FIT_DISTANCE_MAX_KM,
            300,
        )

        fit_shifts = (
            slope
            * fit_distances
            + intercept
        )

        ax_map.plot(
            fit_shifts,
            fit_distances,
            color="lime",
            linestyle="--",
            linewidth=2.0,
            label="Linear ridge fit",
        )

        fit_text = (
            f"Shift slope = {slope:.5f} min/km\n"
            f"Equivalent speed = {speed_m_per_s:.2f} m/s\n"
            f"$R^2$ = {r_squared:.3f}"
        )

        ax_map.text(
            0.02,
            0.03,
            fit_text,
            transform=ax_map.transAxes,
            ha="left",
            va="bottom",
            fontsize=9,
            bbox={
                "facecolor": "white",
                "edgecolor": "black",
                "alpha": 0.85,
            },
        )

    ax_map.set_ylim(
        CORRELATION_DISTANCE_MIN_KM,
        CORRELATION_DISTANCE_MAX_KM,
    )

    ax_map.set_title(
        f"{angle_deg}° Shifted Row Correlation\n"
        f"{dataset_name}"
    )

    ax_map.set_ylabel(
        "Distance along slice (km)"
    )

    ax_map.legend(
        loc="upper right",
        fontsize=8,
    )

    # ==================================================
    # 0 km correlation-shift line
    # ==================================================
    ax_profile.plot(
        shifts_minutes,
        zero_km_correlation_profile,
        color="black",
        linewidth=1.4,
        label=f"{zero_km_distance:g} km correlation profile",
    )

    ax_profile.axvline(
        0.0,
        color="gray",
        linestyle="--",
        linewidth=1.0,
    )

    left_peak = zero_km_local_maxima[
        "left"
    ]

    right_peak = zero_km_local_maxima[
        "right"
    ]

    if np.isfinite(
        left_peak["shift_minutes"]
    ):
        ax_profile.scatter(
            left_peak["shift_minutes"],
            left_peak["correlation"],
            s=55,
            marker="o",
            zorder=5,
            label=(
                "Left local max: "
                f"{left_peak['shift_minutes']:.2f} min"
            ),
        )

        ax_profile.axvline(
            left_peak["shift_minutes"],
            linestyle=":",
            linewidth=1.2,
        )

        ax_profile.annotate(
            (
                f"{left_peak['shift_minutes']:.2f} min\n"
                f"r={left_peak['correlation']:.3f}"
            ),
            xy=(
                left_peak["shift_minutes"],
                left_peak["correlation"],
            ),
            xytext=(8, -35),
            textcoords="offset points",
            ha="left",
            va="top",
            fontsize=8,
            arrowprops={
                "arrowstyle": "->",
                "linewidth": 0.8,
            },
        )

    if np.isfinite(
        right_peak["shift_minutes"]
    ):
        ax_profile.scatter(
            right_peak["shift_minutes"],
            right_peak["correlation"],
            s=55,
            marker="s",
            zorder=5,
            label=(
                "Right local max: "
                f"{right_peak['shift_minutes']:.2f} min"
            ),
        )

        ax_profile.axvline(
            right_peak["shift_minutes"],
            linestyle=":",
            linewidth=1.2,
        )

        ax_profile.annotate(
            (
                f"{right_peak['shift_minutes']:.2f} min\n"
                f"r={right_peak['correlation']:.3f}"
            ),
            xy=(
                right_peak["shift_minutes"],
                right_peak["correlation"],
            ),
            xytext=(-8, -35),
            textcoords="offset points",
            ha="right",
            va="top",
            fontsize=8,
            arrowprops={
                "arrowstyle": "->",
                "linewidth": 0.8,
            },
        )

    ax_profile.set_xlim(
        SHIFT_MIN_MINUTES,
        SHIFT_MAX_MINUTES,
    )

    ax_profile.set_xlabel(
        "Shift applied to row (min)"
    )

    ax_profile.set_ylabel(
        "Correlation"
    )

    ax_profile.set_title(
        f"Correlation-shift line at {zero_km_distance:g} km"
    )

    ax_profile.grid(
        True,
        alpha=0.3,
    )

    ax_profile.legend(
        loc="best",
        fontsize=8,
    )

    plt.tight_layout()

    fig.savefig(
        output_file,
        dpi=DPI,
        bbox_inches="tight",
    )

    plt.close(fig)


# ==================================================
# Print slope results
# ==================================================
def print_slope_results(
    angle_deg,
    label,
    results,
):
    print(
        f"\n{angle_deg}° — {label}:"
    )

    print(
        f"  Shift slope: "
        f"{results['slope_min_per_km']:.8f} min/km"
    )

    print(
        f"  Intercept: "
        f"{results['intercept_minutes']:.6f} min"
    )

    print(
        f"  Equivalent speed: "
        f"{results['speed_m_per_s']:.4f} m/s"
    )

    print(
        f"  R²: "
        f"{results['r_squared']:.6f}"
    )

    print(
        f"  Points used: "
        f"{results['number_of_fit_points']}"
    )


# ==================================================
# Output paths
# ==================================================
sigma_t_label = safe_number_label(
    SMOOTH_SIGMA_TIME_PIXELS
)

sigma_d_label = safe_number_label(
    SMOOTH_SIGMA_DISTANCE_PIXELS
)

distance_folder = (
    f"r_"
    f"{int(CORRELATION_DISTANCE_MIN_KM)}_"
    f"{int(CORRELATION_DISTANCE_MAX_KM)}"
)

shift_folder = (
    f"shift_min_"
    f"{safe_number_label(SHIFT_MIN_MINUTES)}_"
    f"{safe_number_label(SHIFT_MAX_MINUTES)}"
)

if APPLY_2D_SMOOTHING:
    smoothing_folder = (
        f"sig_"
        f"{sigma_t_label}_"
        f"{sigma_d_label}"
    )
else:
    smoothing_folder = "raw"

OUT_DIR = (
    BASE_OUT_DIR
    / distance_folder
    / smoothing_folder
    / "pchip_bias"
    / "row_correlation_minutes_all_angles"
    / shift_folder
)

UNCORRECTED_OUT_DIR = (
    OUT_DIR
    / "uncorrected"
)

CORRECTED_OUT_DIR = (
    OUT_DIR
    / "corrected"
)

SLOPE_REPORT_OUT_DIR = (
    OUT_DIR
    / "slope_reports"
)

for directory in [
    OUT_DIR,
    UNCORRECTED_OUT_DIR,
    CORRECTED_OUT_DIR,
    SLOPE_REPORT_OUT_DIR,
]:
    directory.mkdir(
        parents=True,
        exist_ok=True,
    )


# ==================================================
# Read common time array
# ==================================================
TIMES_FILE = (
    BASE_OUT_DIR
    / "times_s.npy"
)

times_s = np.load(
    TIMES_FILE
).astype(float)

times_min = times_s / 60.0

if len(times_min) < 2:
    raise ValueError(
        "At least two time samples are required."
    )


# ==================================================
# Determine minutes per time pixel
# ==================================================
time_differences_min = np.diff(
    times_min
)

finite_time_differences = time_differences_min[
    np.isfinite(time_differences_min)
]

if len(finite_time_differences) == 0:
    raise ValueError(
        "Could not determine the time spacing."
    )

MINUTES_PER_PIXEL = np.median(
    finite_time_differences
)

if (
    not np.isfinite(MINUTES_PER_PIXEL)
    or MINUTES_PER_PIXEL <= 0
):
    raise ValueError(
        "The calculated minutes-per-pixel value is invalid."
    )

maximum_time_spacing_deviation = np.max(
    np.abs(
        finite_time_differences
        - MINUTES_PER_PIXEL
    )
)

print(
    "Angles to process:"
)

print(
    KEOGRAM_ANGLES_DEG
)

print(
    f"Number of angles: "
    f"{len(KEOGRAM_ANGLES_DEG)}"
)

print(
    "\nTime sampling:"
)

print(
    f"One time pixel = "
    f"{MINUTES_PER_PIXEL:.6f} minutes"
)

print(
    f"One time pixel = "
    f"{MINUTES_PER_PIXEL * 60.0:.3f} seconds"
)

print(
    f"Maximum spacing deviation = "
    f"{maximum_time_spacing_deviation:.6e} minutes"
)


# ==================================================
# Convert minute shifts to integer pixel shifts
# ==================================================
shift_min_pixels = int(
    np.ceil(
        SHIFT_MIN_MINUTES
        / MINUTES_PER_PIXEL
    )
)

shift_max_pixels = int(
    np.floor(
        SHIFT_MAX_MINUTES
        / MINUTES_PER_PIXEL
    )
)

shifts_pixels = np.arange(
    shift_min_pixels,
    shift_max_pixels + 1,
    dtype=int,
)

shifts_minutes = (
    shifts_pixels
    * MINUTES_PER_PIXEL
)

print(
    "\nShift range:"
)

print(
    f"Requested: "
    f"{SHIFT_MIN_MINUTES:.2f} to "
    f"{SHIFT_MAX_MINUTES:.2f} min"
)

print(
    f"Actual: "
    f"{shifts_minutes[0]:.2f} to "
    f"{shifts_minutes[-1]:.2f} min"
)

print(
    f"Pixel shifts: "
    f"{shifts_pixels[0]} to "
    f"{shifts_pixels[-1]}"
)


# ==================================================
# Select common time range
# ==================================================
time_mask = (
    (times_min >= CORRELATION_TIME_MIN)
    & (times_min <= CORRELATION_TIME_MAX)
)

if np.count_nonzero(time_mask) < MIN_VALID_CORRELATION_POINTS:
    raise ValueError(
        "The selected time interval has too few samples."
    )

times_correlation_min = times_min[
    time_mask
]


# ==================================================
# Save common arrays
# ==================================================
np.save(
    OUT_DIR / "angles_deg.npy",
    KEOGRAM_ANGLES_DEG,
)

np.save(
    OUT_DIR / "shifts_minutes.npy",
    shifts_minutes,
)

np.save(
    OUT_DIR / "shifts_pixels.npy",
    shifts_pixels,
)

np.save(
    OUT_DIR / "correlation_times_min.npy",
    times_correlation_min,
)


# ==================================================
# Storage for combined angle results
# ==================================================
combined_results = []
combined_zero_km_peak_results = []

successful_angles = []
failed_angles = []


# ==================================================
# Process every angle
# ==================================================
for angle_number, angle_deg in enumerate(
    KEOGRAM_ANGLES_DEG,
    start=1,
):
    print(
        "\n"
        + "=" * 70
    )

    print(
        f"Processing angle "
        f"{angle_deg}° "
        f"({angle_number}/{len(KEOGRAM_ANGLES_DEG)})"
    )

    print(
        "=" * 70
    )

    KEOGRAM_FILE = (
        BASE_OUT_DIR
        / f"BLO_keogram_angle_{angle_deg:03d}.npy"
    )

    if not KEOGRAM_FILE.exists():
        print(
            f"WARNING: File not found: "
            f"{KEOGRAM_FILE}"
        )

        failed_angles.append(
            (
                int(angle_deg),
                "file not found",
            )
        )

        continue

    try:
        # ==================================================
        # Read keogram
        # ==================================================
        keogram_original = np.load(
            KEOGRAM_FILE
        ).astype(float)

        if keogram_original.ndim != 2:
            raise ValueError(
                f"Expected a 2D keogram, found "
                f"{keogram_original.shape}."
            )

        nt, nspace = keogram_original.shape

        if len(times_min) != nt:
            raise ValueError(
                f"Keogram has {nt} time samples, but "
                f"times_s.npy has {len(times_min)}."
            )

        # ==================================================
        # Construct distance axis
        # ==================================================
        distance_km = np.arange(
            -(nspace // 2),
            nspace // 2 + 1,
            dtype=float,
        )

        if len(distance_km) != nspace:
            distance_km = (
                np.arange(
                    nspace,
                    dtype=float,
                )
                - nspace // 2
            )

        available_distance_min = np.nanmin(
            distance_km
        )

        available_distance_max = np.nanmax(
            distance_km
        )

        if (
            CORRELATION_DISTANCE_MIN_KM
            < available_distance_min
            or CORRELATION_DISTANCE_MAX_KM
            > available_distance_max
        ):
            raise ValueError(
                f"Requested distance range "
                f"{CORRELATION_DISTANCE_MIN_KM} to "
                f"{CORRELATION_DISTANCE_MAX_KM} km "
                f"is outside available range "
                f"{available_distance_min} to "
                f"{available_distance_max} km."
            )

        # ==================================================
        # Select time interval
        # ==================================================
        keogram_time_selected = keogram_original[
            time_mask,
            :
        ]

        # ==================================================
        # Optional smoothing
        # ==================================================
        if APPLY_2D_SMOOTHING:
            keogram_uncorrected = smooth_keogram_2d(
                data=keogram_time_selected,
                sigma_time=SMOOTH_SIGMA_TIME_PIXELS,
                sigma_distance=SMOOTH_SIGMA_DISTANCE_PIXELS,
                mode=SMOOTH_MODE,
                weight_tolerance=SMOOTH_WEIGHT_TOLERANCE,
            )

            uncorrected_dataset_name = (
                f"Uncorrected, smoothed "
                f"σt={SMOOTH_SIGMA_TIME_PIXELS:g}, "
                f"σd={SMOOTH_SIGMA_DISTANCE_PIXELS:g}"
            )
        else:
            keogram_uncorrected = (
                keogram_time_selected.copy()
            )

            uncorrected_dataset_name = (
                "Uncorrected, unsmoothed"
            )

        # ==================================================
        # Create bias correction
        # ==================================================
        (
            bias_correction_keogram,
            bias_vs_time,
            bias_distance_factor,
        ) = create_bias_correction_keogram(
            times_min=times_correlation_min,
            distance_km=distance_km,
            bias_time_points_min=BIAS_TIME_POINTS_MIN,
            bias_at_max_distance=BIAS_AT_MAX_DISTANCE,
            bias_distance_min_km=BIAS_DISTANCE_MIN_KM,
            bias_distance_max_km=BIAS_DISTANCE_MAX_KM,
            distance_power=BIAS_DISTANCE_POWER,
            zero_outside_distance_range=(
                BIAS_ZERO_OUTSIDE_DISTANCE_RANGE
            ),
            zero_outside_time_range=(
                BIAS_ZERO_OUTSIDE_TIME_RANGE
            ),
            clip_negative_time_correction=(
                CLIP_NEGATIVE_TIME_CORRECTION
            ),
        )

        # ==================================================
        # Apply bias correction
        # ==================================================
        if APPLY_BIAS_CORRECTION:
            keogram_corrected = (
                keogram_uncorrected
                + bias_correction_keogram
            )

            corrected_dataset_name = (
                "Bias corrected: PCHIP time dependence, "
                f"distance power={BIAS_DISTANCE_POWER:g}"
            )
        else:
            keogram_corrected = (
                keogram_uncorrected.copy()
            )

            corrected_dataset_name = (
                "Bias correction disabled"
            )

        keogram_corrected[
            ~np.isfinite(keogram_uncorrected)
        ] = np.nan

        # ==================================================
        # Select distance rows
        # ==================================================
        selected_distance_mask = (
            (
                distance_km
                >= CORRELATION_DISTANCE_MIN_KM
            )
            & (
                distance_km
                <= CORRELATION_DISTANCE_MAX_KM
            )
        )

        selected_distances = distance_km[
            selected_distance_mask
        ]

        if len(selected_distances) == 0:
            raise ValueError(
                "No rows lie inside the selected distance range."
            )

        # ==================================================
        # Per-angle folders
        # ==================================================
        angle_uncorrected_dir = (
            UNCORRECTED_OUT_DIR
            / f"angle_{angle_deg:03d}"
        )

        angle_corrected_dir = (
            CORRECTED_OUT_DIR
            / f"angle_{angle_deg:03d}"
        )

        angle_uncorrected_dir.mkdir(
            parents=True,
            exist_ok=True,
        )

        angle_corrected_dir.mkdir(
            parents=True,
            exist_ok=True,
        )

        # ==================================================
        # Uncorrected map
        # ==================================================
        (
            correlation_map_uncorrected,
            selected_distances_uncorrected,
            reference_index_uncorrected,
        ) = calculate_correlation_map(
            keogram=keogram_uncorrected,
            distance_km=distance_km,
            reference_distance_km=0.0,
            selected_distance_mask=selected_distance_mask,
            shifts_pixels=shifts_pixels,
            minimum_points=MIN_VALID_CORRELATION_POINTS,
        )

        (
            best_shift_uncorrected_minutes,
            maximum_correlation_uncorrected,
        ) = find_best_shifts(
            correlation_map=correlation_map_uncorrected,
            shifts_minutes=shifts_minutes,
        )

        uncorrected_slope_results = (
            fit_maximum_correlation_slope(
                distances_km=(
                    selected_distances_uncorrected
                ),
                best_shift_minutes=(
                    best_shift_uncorrected_minutes
                ),
                maximum_correlation=(
                    maximum_correlation_uncorrected
                ),
                minimum_correlation=(
                    MIN_CORRELATION_FOR_SLOPE
                ),
                distance_min_km=(
                    SLOPE_FIT_DISTANCE_MIN_KM
                ),
                distance_max_km=(
                    SLOPE_FIT_DISTANCE_MAX_KM
                ),
            )
        )

        # ==================================================
        # Corrected map
        # ==================================================
        (
            correlation_map_corrected,
            selected_distances_corrected,
            reference_index_corrected,
        ) = calculate_correlation_map(
            keogram=keogram_corrected,
            distance_km=distance_km,
            reference_distance_km=0.0,
            selected_distance_mask=selected_distance_mask,
            shifts_pixels=shifts_pixels,
            minimum_points=MIN_VALID_CORRELATION_POINTS,
        )

        (
            best_shift_corrected_minutes,
            maximum_correlation_corrected,
        ) = find_best_shifts(
            correlation_map=correlation_map_corrected,
            shifts_minutes=shifts_minutes,
        )

        corrected_slope_results = (
            fit_maximum_correlation_slope(
                distances_km=(
                    selected_distances_corrected
                ),
                best_shift_minutes=(
                    best_shift_corrected_minutes
                ),
                maximum_correlation=(
                    maximum_correlation_corrected
                ),
                minimum_correlation=(
                    MIN_CORRELATION_FOR_SLOPE
                ),
                distance_min_km=(
                    SLOPE_FIT_DISTANCE_MIN_KM
                ),
                distance_max_km=(
                    SLOPE_FIT_DISTANCE_MAX_KM
                ),
            )
        )

        # ==================================================
        # Extract 0 km correlation-shift profiles
        # ==================================================
        (
            zero_km_profile_uncorrected,
            zero_km_row_uncorrected,
            zero_km_distance_uncorrected,
        ) = extract_zero_km_correlation_profile(
            correlation_map=correlation_map_uncorrected,
            selected_distances=selected_distances_uncorrected,
        )

        (
            zero_km_profile_corrected,
            zero_km_row_corrected,
            zero_km_distance_corrected,
        ) = extract_zero_km_correlation_profile(
            correlation_map=correlation_map_corrected,
            selected_distances=selected_distances_corrected,
        )

        zero_km_peaks_uncorrected = (
            find_left_and_right_local_maxima(
                correlation_profile=zero_km_profile_uncorrected,
                shifts_minutes=shifts_minutes,
                minimum_prominence=(
                    LOCAL_MAXIMUM_MIN_PROMINENCE
                ),
                minimum_distance_pixels=(
                    LOCAL_MAXIMUM_MIN_DISTANCE_PIXELS
                ),
            )
        )

        zero_km_peaks_corrected = (
            find_left_and_right_local_maxima(
                correlation_profile=zero_km_profile_corrected,
                shifts_minutes=shifts_minutes,
                minimum_prominence=(
                    LOCAL_MAXIMUM_MIN_PROMINENCE
                ),
                minimum_distance_pixels=(
                    LOCAL_MAXIMUM_MIN_DISTANCE_PIXELS
                ),
            )
        )

        # ==================================================
        # Plot maps
        # ==================================================
        uncorrected_plot_file = (
            angle_uncorrected_dir
            / (
                f"correlation_minutes_"
                f"{angle_deg:03d}_uncorrected.png"
            )
        )

        corrected_plot_file = (
            angle_corrected_dir
            / (
                f"correlation_minutes_"
                f"{angle_deg:03d}_corrected.png"
            )
        )

        plot_correlation_map(
            angle_deg=angle_deg,
            correlation_map=(
                correlation_map_uncorrected
            ),
            shifts_minutes=shifts_minutes,
            selected_distances=(
                selected_distances_uncorrected
            ),
            best_shift_minutes=(
                best_shift_uncorrected_minutes
            ),
            maximum_correlation=(
                maximum_correlation_uncorrected
            ),
            slope_results=(
                uncorrected_slope_results
            ),
            zero_km_correlation_profile=(
                zero_km_profile_uncorrected
            ),
            zero_km_distance=(
                zero_km_distance_uncorrected
            ),
            zero_km_local_maxima=(
                zero_km_peaks_uncorrected
            ),
            dataset_name=uncorrected_dataset_name,
            output_file=uncorrected_plot_file,
        )

        plot_correlation_map(
            angle_deg=angle_deg,
            correlation_map=(
                correlation_map_corrected
            ),
            shifts_minutes=shifts_minutes,
            selected_distances=(
                selected_distances_corrected
            ),
            best_shift_minutes=(
                best_shift_corrected_minutes
            ),
            maximum_correlation=(
                maximum_correlation_corrected
            ),
            slope_results=(
                corrected_slope_results
            ),
            zero_km_correlation_profile=(
                zero_km_profile_corrected
            ),
            zero_km_distance=(
                zero_km_distance_corrected
            ),
            zero_km_local_maxima=(
                zero_km_peaks_corrected
            ),
            dataset_name=corrected_dataset_name,
            output_file=corrected_plot_file,
        )

        # ==================================================
        # Save arrays
        # ==================================================
        np.save(
            angle_uncorrected_dir
            / (
                f"correlation_"
                f"{angle_deg:03d}_uncorrected.npy"
            ),
            correlation_map_uncorrected,
        )

        np.save(
            angle_uncorrected_dir
            / (
                f"best_shift_minutes_"
                f"{angle_deg:03d}_uncorrected.npy"
            ),
            best_shift_uncorrected_minutes,
        )

        np.save(
            angle_uncorrected_dir
            / (
                f"maximum_correlation_"
                f"{angle_deg:03d}_uncorrected.npy"
            ),
            maximum_correlation_uncorrected,
        )

        np.save(
            angle_corrected_dir
            / (
                f"correlation_"
                f"{angle_deg:03d}_corrected.npy"
            ),
            correlation_map_corrected,
        )

        np.save(
            angle_corrected_dir
            / (
                f"best_shift_minutes_"
                f"{angle_deg:03d}_corrected.npy"
            ),
            best_shift_corrected_minutes,
        )

        np.save(
            angle_corrected_dir
            / (
                f"maximum_correlation_"
                f"{angle_deg:03d}_corrected.npy"
            ),
            maximum_correlation_corrected,
        )

        np.save(
            angle_uncorrected_dir
            / "distances_km.npy",
            selected_distances_uncorrected,
        )

        np.save(
            angle_corrected_dir
            / "distances_km.npy",
            selected_distances_corrected,
        )

        np.save(
            angle_uncorrected_dir
            / "zero_km_correlation_profile.npy",
            zero_km_profile_uncorrected,
        )

        np.save(
            angle_corrected_dir
            / "zero_km_correlation_profile.npy",
            zero_km_profile_corrected,
        )

        zero_km_profile_uncorrected_csv = np.column_stack(
            [
                shifts_minutes,
                zero_km_profile_uncorrected,
            ]
        )

        zero_km_profile_corrected_csv = np.column_stack(
            [
                shifts_minutes,
                zero_km_profile_corrected,
            ]
        )

        np.savetxt(
            angle_uncorrected_dir
            / "zero_km_correlation_profile.csv",
            zero_km_profile_uncorrected_csv,
            delimiter=",",
            header="shift_minutes,correlation",
            comments="",
        )

        np.savetxt(
            angle_corrected_dir
            / "zero_km_correlation_profile.csv",
            zero_km_profile_corrected_csv,
            delimiter=",",
            header="shift_minutes,correlation",
            comments="",
        )

        # ==================================================
        # Save ridge CSV files
        # ==================================================
        uncorrected_summary = np.column_stack(
            [
                selected_distances_uncorrected,
                best_shift_uncorrected_minutes,
                maximum_correlation_uncorrected,
            ]
        )

        corrected_summary = np.column_stack(
            [
                selected_distances_corrected,
                best_shift_corrected_minutes,
                maximum_correlation_corrected,
            ]
        )

        uncorrected_csv = (
            angle_uncorrected_dir
            / (
                f"ridge_"
                f"{angle_deg:03d}_uncorrected.csv"
            )
        )

        corrected_csv = (
            angle_corrected_dir
            / (
                f"ridge_"
                f"{angle_deg:03d}_corrected.csv"
            )
        )

        np.savetxt(
            uncorrected_csv,
            uncorrected_summary,
            delimiter=",",
            header=(
                "distance_km,"
                "best_shift_minutes,"
                "maximum_correlation"
            ),
            comments="",
        )

        np.savetxt(
            corrected_csv,
            corrected_summary,
            delimiter=",",
            header=(
                "distance_km,"
                "best_shift_minutes,"
                "maximum_correlation"
            ),
            comments="",
        )

        # ==================================================
        # Save individual slope report
        # ==================================================
        slope_output_file = (
            SLOPE_REPORT_OUT_DIR
            / (
                f"correlation_slope_"
                f"{angle_deg:03d}.txt"
            )
        )

        with open(
            slope_output_file,
            "w",
            encoding="utf-8",
        ) as file:
            file.write(
                f"Angle: {angle_deg} degrees\n"
            )

            file.write(
                f"Minutes per pixel: "
                f"{MINUTES_PER_PIXEL:.10f}\n\n"
            )

            file.write(
                "UNCORRECTED\n"
            )

            for key, value in (
                uncorrected_slope_results.items()
            ):
                if key != "fit_mask":
                    file.write(
                        f"{key}: {value}\n"
                    )

            file.write(
                "\nCORRECTED\n"
            )

            for key, value in (
                corrected_slope_results.items()
            ):
                if key != "fit_mask":
                    file.write(
                        f"{key}: {value}\n"
                    )

            file.write(
                "\n0 KM CORRELATION-PROFILE LOCAL MAXIMA\n"
            )

            file.write(
                "Uncorrected left shift (min): "
                f"{zero_km_peaks_uncorrected['left']['shift_minutes']}\n"
            )
            file.write(
                "Uncorrected left correlation: "
                f"{zero_km_peaks_uncorrected['left']['correlation']}\n"
            )
            file.write(
                "Uncorrected right shift (min): "
                f"{zero_km_peaks_uncorrected['right']['shift_minutes']}\n"
            )
            file.write(
                "Uncorrected right correlation: "
                f"{zero_km_peaks_uncorrected['right']['correlation']}\n"
            )
            file.write(
                "Corrected left shift (min): "
                f"{zero_km_peaks_corrected['left']['shift_minutes']}\n"
            )
            file.write(
                "Corrected left correlation: "
                f"{zero_km_peaks_corrected['left']['correlation']}\n"
            )
            file.write(
                "Corrected right shift (min): "
                f"{zero_km_peaks_corrected['right']['shift_minutes']}\n"
            )
            file.write(
                "Corrected right correlation: "
                f"{zero_km_peaks_corrected['right']['correlation']}\n"
            )

        # ==================================================
        # Add to combined summary
        # ==================================================
        combined_results.append(
            [
                float(angle_deg),

                uncorrected_slope_results[
                    "slope_min_per_km"
                ],
                uncorrected_slope_results[
                    "intercept_minutes"
                ],
                uncorrected_slope_results[
                    "speed_km_per_min"
                ],
                uncorrected_slope_results[
                    "speed_m_per_s"
                ],
                uncorrected_slope_results[
                    "r_squared"
                ],
                uncorrected_slope_results[
                    "number_of_fit_points"
                ],

                corrected_slope_results[
                    "slope_min_per_km"
                ],
                corrected_slope_results[
                    "intercept_minutes"
                ],
                corrected_slope_results[
                    "speed_km_per_min"
                ],
                corrected_slope_results[
                    "speed_m_per_s"
                ],
                corrected_slope_results[
                    "r_squared"
                ],
                corrected_slope_results[
                    "number_of_fit_points"
                ],
            ]
        )

        combined_zero_km_peak_results.append(
            [
                float(angle_deg),
                zero_km_peaks_uncorrected["left"][
                    "shift_minutes"
                ],
                zero_km_peaks_uncorrected["left"][
                    "correlation"
                ],
                zero_km_peaks_uncorrected["right"][
                    "shift_minutes"
                ],
                zero_km_peaks_uncorrected["right"][
                    "correlation"
                ],
                zero_km_peaks_corrected["left"][
                    "shift_minutes"
                ],
                zero_km_peaks_corrected["left"][
                    "correlation"
                ],
                zero_km_peaks_corrected["right"][
                    "shift_minutes"
                ],
                zero_km_peaks_corrected["right"][
                    "correlation"
                ],
            ]
        )

        successful_angles.append(
            int(angle_deg)
        )

        print_slope_results(
            angle_deg=angle_deg,
            label=(
                "Uncorrected maximum-correlation ridge"
            ),
            results=uncorrected_slope_results,
        )

        print_slope_results(
            angle_deg=angle_deg,
            label=(
                "Corrected maximum-correlation ridge"
            ),
            results=corrected_slope_results,
        )

        print(
            "\n0 km correlation-profile local maxima:"
        )
        print(
            "  Uncorrected left: "
            f"{zero_km_peaks_uncorrected['left']['shift_minutes']:.3f} min, "
            f"r={zero_km_peaks_uncorrected['left']['correlation']:.6f}"
        )
        print(
            "  Uncorrected right: "
            f"{zero_km_peaks_uncorrected['right']['shift_minutes']:.3f} min, "
            f"r={zero_km_peaks_uncorrected['right']['correlation']:.6f}"
        )
        print(
            "  Corrected left: "
            f"{zero_km_peaks_corrected['left']['shift_minutes']:.3f} min, "
            f"r={zero_km_peaks_corrected['left']['correlation']:.6f}"
        )
        print(
            "  Corrected right: "
            f"{zero_km_peaks_corrected['right']['shift_minutes']:.3f} min, "
            f"r={zero_km_peaks_corrected['right']['correlation']:.6f}"
        )

    except Exception as error:
        print(
            f"ERROR processing {angle_deg}°:"
        )

        print(
            error
        )

        failed_angles.append(
            (
                int(angle_deg),
                str(error),
            )
        )


# ==================================================
# Save combined 0 km local-maximum summary
# ==================================================
combined_zero_km_peak_results = np.asarray(
    combined_zero_km_peak_results,
    dtype=float,
)

combined_zero_km_peak_csv = (
    OUT_DIR
    / "all_angle_zero_km_local_maxima.csv"
)

combined_zero_km_peak_npy = (
    OUT_DIR
    / "all_angle_zero_km_local_maxima.npy"
)

combined_zero_km_peak_header = (
    "angle_deg,"
    "uncorrected_left_shift_minutes,"
    "uncorrected_left_correlation,"
    "uncorrected_right_shift_minutes,"
    "uncorrected_right_correlation,"
    "corrected_left_shift_minutes,"
    "corrected_left_correlation,"
    "corrected_right_shift_minutes,"
    "corrected_right_correlation"
)

if combined_zero_km_peak_results.size > 0:
    np.savetxt(
        combined_zero_km_peak_csv,
        combined_zero_km_peak_results,
        delimiter=",",
        header=combined_zero_km_peak_header,
        comments="",
    )

    np.save(
        combined_zero_km_peak_npy,
        combined_zero_km_peak_results,
    )


# ==================================================
# Save combined angle summary
# ==================================================
combined_results = np.asarray(
    combined_results,
    dtype=float,
)

combined_csv = (
    OUT_DIR
    / "all_angle_slope_summary.csv"
)

combined_npy = (
    OUT_DIR
    / "all_angle_slope_summary.npy"
)

combined_header = (
    "angle_deg,"
    "uncorrected_slope_min_per_km,"
    "uncorrected_intercept_minutes,"
    "uncorrected_speed_km_per_min,"
    "uncorrected_speed_m_per_s,"
    "uncorrected_r_squared,"
    "uncorrected_number_of_fit_points,"
    "corrected_slope_min_per_km,"
    "corrected_intercept_minutes,"
    "corrected_speed_km_per_min,"
    "corrected_speed_m_per_s,"
    "corrected_r_squared,"
    "corrected_number_of_fit_points"
)

if combined_results.size > 0:
    np.savetxt(
        combined_csv,
        combined_results,
        delimiter=",",
        header=combined_header,
        comments="",
    )

    np.save(
        combined_npy,
        combined_results,
    )


# ==================================================
# Plot speed versus angle
# ==================================================
if combined_results.size > 0:
    angles_processed = combined_results[
        :,
        0,
    ]

    uncorrected_speed_m_per_s = combined_results[
        :,
        4,
    ]

    corrected_speed_m_per_s = combined_results[
        :,
        10,
    ]

    fig_speed, ax_speed = plt.subplots(
        figsize=(10, 5)
    )

    ax_speed.plot(
        angles_processed,
        uncorrected_speed_m_per_s,
        marker="o",
        linewidth=1.5,
        label="Uncorrected",
    )

    ax_speed.plot(
        angles_processed,
        corrected_speed_m_per_s,
        marker="o",
        linewidth=1.5,
        label="Corrected",
    )

    ax_speed.axhline(
        0.0,
        color="black",
        linestyle="--",
        linewidth=1.0,
    )

    ax_speed.set_xlim(
        KEOGRAM_ANGLES_DEG.min(),
        KEOGRAM_ANGLES_DEG.max(),
    )

    ax_speed.set_xlabel(
        "Keogram angle (degrees)"
    )

    ax_speed.set_ylabel(
        "Equivalent signed speed (m/s)"
    )

    ax_speed.set_title(
        "Maximum-Correlation Ridge Speed versus Angle"
    )

    ax_speed.grid(
        True,
        alpha=0.3,
    )

    ax_speed.legend(
        loc="best",
    )

    plt.tight_layout()

    speed_angle_plot = (
        OUT_DIR
        / "speed_vs_angle.png"
    )

    fig_speed.savefig(
        speed_angle_plot,
        dpi=DPI,
        bbox_inches="tight",
    )

    plt.close(
        fig_speed
    )


# ==================================================
# Plot R-squared versus angle
# ==================================================
if combined_results.size > 0:
    uncorrected_r_squared = combined_results[
        :,
        5,
    ]

    corrected_r_squared = combined_results[
        :,
        11,
    ]

    fig_r2, ax_r2 = plt.subplots(
        figsize=(10, 5)
    )

    ax_r2.plot(
        angles_processed,
        uncorrected_r_squared,
        marker="o",
        linewidth=1.5,
        label="Uncorrected",
    )

    ax_r2.plot(
        angles_processed,
        corrected_r_squared,
        marker="o",
        linewidth=1.5,
        label="Corrected",
    )

    ax_r2.set_xlim(
        KEOGRAM_ANGLES_DEG.min(),
        KEOGRAM_ANGLES_DEG.max(),
    )

    ax_r2.set_ylim(
        0.0,
        1.0,
    )

    ax_r2.set_xlabel(
        "Keogram angle (degrees)"
    )

    ax_r2.set_ylabel(
        "R²"
    )

    ax_r2.set_title(
        "Maximum-Correlation Ridge Fit R² versus Angle"
    )

    ax_r2.grid(
        True,
        alpha=0.3,
    )

    ax_r2.legend(
        loc="best",
    )

    plt.tight_layout()

    r2_angle_plot = (
        OUT_DIR
        / "r_squared_vs_angle.png"
    )

    fig_r2.savefig(
        r2_angle_plot,
        dpi=DPI,
        bbox_inches="tight",
    )

    plt.close(
        fig_r2
    )


# ==================================================
# Save processing log
# ==================================================
processing_log_file = (
    OUT_DIR
    / "processing_log.txt"
)

with open(
    processing_log_file,
    "w",
    encoding="utf-8",
) as file:
    file.write(
        "Requested angles:\n"
    )

    file.write(
        ", ".join(
            str(angle)
            for angle in KEOGRAM_ANGLES_DEG
        )
        + "\n\n"
    )

    file.write(
        f"Number requested: "
        f"{len(KEOGRAM_ANGLES_DEG)}\n"
    )

    file.write(
        f"Number successful: "
        f"{len(successful_angles)}\n"
    )

    file.write(
        f"Number failed: "
        f"{len(failed_angles)}\n\n"
    )

    file.write(
        "Successful angles:\n"
    )

    file.write(
        ", ".join(
            str(angle)
            for angle in successful_angles
        )
        + "\n\n"
    )

    file.write(
        "Failed angles:\n"
    )

    if failed_angles:
        for angle_deg, reason in failed_angles:
            file.write(
                f"{angle_deg}: {reason}\n"
            )
    else:
        file.write(
            "None\n"
        )


# ==================================================
# Print final summary
# ==================================================
print(
    "\n"
    + "=" * 70
)

print(
    "ALL-ANGLE PROCESSING COMPLETE"
)

print(
    "=" * 70
)

print(
    "Angles requested:"
)

print(
    KEOGRAM_ANGLES_DEG
)

print(
    f"\nNumber requested: "
    f"{len(KEOGRAM_ANGLES_DEG)}"
)

print(
    f"Number successful: "
    f"{len(successful_angles)}"
)

print(
    f"Number failed: "
    f"{len(failed_angles)}"
)

print(
    "\nSuccessful angles:"
)

print(
    successful_angles
)

if failed_angles:
    print(
        "\nFailed angles:"
    )

    for angle_deg, reason in failed_angles:
        print(
            f"  {angle_deg}°: {reason}"
        )

print(
    "\nCombined CSV:"
)

print(
    combined_csv
)

print(
    "\nOutput folder:"
)

print(
    OUT_DIR.resolve()
)
```

## Also plot original keogram and smoothed ranged keogram. 

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

from scipy.ndimage import gaussian_filter
from scipy.interpolate import PchipInterpolator
from scipy.signal import find_peaks


# ==================================================
# Settings
# ==================================================
BASE_OUT_DIR = Path(
    "BLO_perturbation_keogram_outputs_new"
)

# Process every directional keogram:
# 0, 5, 10, ..., 175 degrees
KEOGRAM_ANGLES_DEG = np.arange(
    0,
    176,
    5,
    dtype=int,
)

# Time range used for correlation
CORRELATION_TIME_MIN = 0.0
CORRELATION_TIME_MAX = 300.0

# Distance range included in the correlation map
CORRELATION_DISTANCE_MIN_KM = 0.0
CORRELATION_DISTANCE_MAX_KM = 250.0

# Shift range in minutes
SHIFT_MIN_MINUTES = -200.0
SHIFT_MAX_MINUTES = 200.0

# Minimum number of overlapping time samples
MIN_VALID_CORRELATION_POINTS = 20

# Minimum maximum-correlation value used in the slope fit
MIN_CORRELATION_FOR_SLOPE = 0.50

# Distance range used in the ridge fit
SLOPE_FIT_DISTANCE_MIN_KM = 0.0
SLOPE_FIT_DISTANCE_MAX_KM = 250.0

DPI = 200

CORRELATION_CMAP = "RdBu_r"
KEOGRAM_CMAP = "RdBu_r"

# Use symmetric automatic keogram limits based on this percentile.
# Set KEOGRAM_ABS_LIMIT to a number to force fixed limits.
KEOGRAM_ABS_LIMIT = None
KEOGRAM_AUTO_PERCENTILE = 99.0

# Correlation color scale
CORRELATION_VMIN = 0.70
CORRELATION_VMAX = 1.00

# Local maxima search on the 0 km correlation-shift profile.
# The left maximum is searched at negative shifts and the right
# maximum is searched at positive shifts. The zero-shift point is
# therefore excluded automatically.
LOCAL_MAXIMUM_MIN_PROMINENCE = 0.0
LOCAL_MAXIMUM_MIN_DISTANCE_PIXELS = 1


# ==================================================
# Optional 2D smoothing
# ==================================================
APPLY_2D_SMOOTHING = True

# Keogram dimensions:
# axis 0 = time
# axis 1 = distance
SMOOTH_SIGMA_TIME_PIXELS = 1.5
SMOOTH_SIGMA_DISTANCE_PIXELS = 2.0

SMOOTH_MODE = "nearest"
SMOOTH_WEIGHT_TOLERANCE = 1e-12


# ==================================================
# Bias-correction settings
# ==================================================
APPLY_BIAS_CORRECTION = False

BIAS_TIME_POINTS_MIN = np.array(
    [
        0.0,
        150.0,
        300.0,
    ],
    dtype=float,
)

BIAS_AT_MAX_DISTANCE = np.array(
    [
        0.20,
        0.04,
        -0.01,
    ],
    dtype=float,
)

BIAS_DISTANCE_MIN_KM = 0.0
BIAS_DISTANCE_MAX_KM = 250.0

# 1.0 = linear distance dependence
# 2.0 = quadratic distance dependence
BIAS_DISTANCE_POWER = 1.0

BIAS_ZERO_OUTSIDE_DISTANCE_RANGE = True
BIAS_ZERO_OUTSIDE_TIME_RANGE = True
CLIP_NEGATIVE_TIME_CORRECTION = False


# ==================================================
# Filename-safe numeric label
# ==================================================
def safe_number_label(value):
    value_float = float(value)

    if value_float.is_integer():
        return str(
            int(value_float)
        )

    return (
        f"{value_float:g}"
        .replace("-", "m")
        .replace(".", "p")
    )


# ==================================================
# NaN-aware 2D Gaussian smoothing
# ==================================================
def smooth_keogram_2d(
    data,
    sigma_time,
    sigma_distance,
    mode="nearest",
    weight_tolerance=1e-12,
):
    """
    Apply NaN-aware Gaussian smoothing to a 2D keogram.
    """
    data = np.asarray(
        data,
        dtype=float,
    )

    if data.ndim != 2:
        raise ValueError(
            "The keogram must be two-dimensional."
        )

    if sigma_time < 0 or sigma_distance < 0:
        raise ValueError(
            "Smoothing sigma values must be nonnegative."
        )

    if sigma_time == 0 and sigma_distance == 0:
        return data.copy()

    valid_mask = np.isfinite(
        data
    )

    data_filled = np.where(
        valid_mask,
        data,
        0.0,
    )

    weights = valid_mask.astype(
        float
    )

    sigma = (
        sigma_time,
        sigma_distance,
    )

    smoothed_values = gaussian_filter(
        data_filled,
        sigma=sigma,
        mode=mode,
    )

    smoothed_weights = gaussian_filter(
        weights,
        sigma=sigma,
        mode=mode,
    )

    smoothed = np.full(
        data.shape,
        np.nan,
        dtype=float,
    )

    valid_output = (
        smoothed_weights
        > weight_tolerance
    )

    smoothed[valid_output] = (
        smoothed_values[valid_output]
        / smoothed_weights[valid_output]
    )

    return smoothed


# ==================================================
# Create 2D bias-correction keogram
# ==================================================
def create_bias_correction_keogram(
    times_min,
    distance_km,
    bias_time_points_min,
    bias_at_max_distance,
    bias_distance_min_km,
    bias_distance_max_km,
    distance_power=1.0,
    zero_outside_distance_range=True,
    zero_outside_time_range=True,
    clip_negative_time_correction=False,
):
    """
    Create the time x distance additive bias field.
    """
    times_min = np.asarray(
        times_min,
        dtype=float,
    )

    distance_km = np.asarray(
        distance_km,
        dtype=float,
    )

    bias_time_points_min = np.asarray(
        bias_time_points_min,
        dtype=float,
    )

    bias_at_max_distance = np.asarray(
        bias_at_max_distance,
        dtype=float,
    )

    if (
        len(bias_time_points_min)
        != len(bias_at_max_distance)
    ):
        raise ValueError(
            "Bias time points and bias values must have "
            "the same length."
        )

    if not np.all(
        np.diff(bias_time_points_min) > 0
    ):
        raise ValueError(
            "BIAS_TIME_POINTS_MIN must be strictly increasing."
        )

    if bias_distance_max_km <= bias_distance_min_km:
        raise ValueError(
            "The maximum bias distance must exceed "
            "the minimum bias distance."
        )

    if distance_power <= 0:
        raise ValueError(
            "BIAS_DISTANCE_POWER must be positive."
        )

    # ==================================================
    # PCHIP time dependence
    # ==================================================
    interpolator = PchipInterpolator(
        bias_time_points_min,
        bias_at_max_distance,
        extrapolate=False,
    )

    bias_vs_time = interpolator(
        times_min
    )

    if zero_outside_time_range:
        inside_time = (
            (times_min >= bias_time_points_min[0])
            & (times_min <= bias_time_points_min[-1])
        )

        bias_vs_time = np.where(
            inside_time,
            bias_vs_time,
            0.0,
        )
    else:
        bias_vs_time = np.where(
            times_min < bias_time_points_min[0],
            bias_at_max_distance[0],
            bias_vs_time,
        )

        bias_vs_time = np.where(
            times_min > bias_time_points_min[-1],
            bias_at_max_distance[-1],
            bias_vs_time,
        )

    if clip_negative_time_correction:
        bias_vs_time = np.maximum(
            bias_vs_time,
            0.0,
        )

    # ==================================================
    # Distance dependence
    # ==================================================
    normalized_distance = (
        distance_km
        - bias_distance_min_km
    ) / (
        bias_distance_max_km
        - bias_distance_min_km
    )

    if zero_outside_distance_range:
        inside_distance = (
            (distance_km >= bias_distance_min_km)
            & (distance_km <= bias_distance_max_km)
        )

        normalized_distance = np.where(
            inside_distance,
            normalized_distance,
            0.0,
        )
    else:
        normalized_distance = np.clip(
            normalized_distance,
            0.0,
            1.0,
        )

    distance_factor = (
        normalized_distance ** distance_power
    )

    correction_keogram = (
        bias_vs_time[:, np.newaxis]
        * distance_factor[np.newaxis, :]
    )

    return (
        correction_keogram,
        bias_vs_time,
        distance_factor,
    )


# ==================================================
# NaN-aware Pearson correlation
# ==================================================
def nan_pearson_correlation(
    reference,
    comparison,
    minimum_points=20,
):
    """
    Calculate Pearson correlation using finite overlapping values.
    """
    reference = np.asarray(
        reference,
        dtype=float,
    )

    comparison = np.asarray(
        comparison,
        dtype=float,
    )

    if reference.shape != comparison.shape:
        raise ValueError(
            "The two arrays must have the same shape."
        )

    valid = (
        np.isfinite(reference)
        & np.isfinite(comparison)
    )

    if np.count_nonzero(valid) < minimum_points:
        return np.nan

    reference_valid = reference[
        valid
    ]

    comparison_valid = comparison[
        valid
    ]

    reference_centered = (
        reference_valid
        - np.mean(reference_valid)
    )

    comparison_centered = (
        comparison_valid
        - np.mean(comparison_valid)
    )

    denominator = np.sqrt(
        np.sum(reference_centered ** 2)
        * np.sum(comparison_centered ** 2)
    )

    if (
        not np.isfinite(denominator)
        or denominator <= 0
    ):
        return np.nan

    correlation = (
        np.sum(
            reference_centered
            * comparison_centered
        )
        / denominator
    )

    return np.clip(
        correlation,
        -1.0,
        1.0,
    )


# ==================================================
# Correlation for one integer-pixel shift
# ==================================================
def calculate_shifted_correlation(
    reference_row,
    comparison_row,
    shift_pixels,
    minimum_points=20,
):
    """
    Shift comparison_row and correlate it with reference_row.

    Positive shift:
        comparison row moves toward later times.

    Negative shift:
        comparison row moves toward earlier times.

    No circular wrapping is used.
    """
    reference_row = np.asarray(
        reference_row,
        dtype=float,
    )

    comparison_row = np.asarray(
        comparison_row,
        dtype=float,
    )

    if reference_row.shape != comparison_row.shape:
        raise ValueError(
            "Reference and comparison rows must have "
            "the same shape."
        )

    ntime = len(
        reference_row
    )

    if abs(shift_pixels) >= ntime:
        return np.nan

    if shift_pixels > 0:
        reference_overlap = reference_row[
            shift_pixels:
        ]

        comparison_overlap = comparison_row[
            :-shift_pixels
        ]

    elif shift_pixels < 0:
        shift_abs = abs(
            shift_pixels
        )

        reference_overlap = reference_row[
            :-shift_abs
        ]

        comparison_overlap = comparison_row[
            shift_abs:
        ]

    else:
        reference_overlap = reference_row
        comparison_overlap = comparison_row

    return nan_pearson_correlation(
        reference=reference_overlap,
        comparison=comparison_overlap,
        minimum_points=minimum_points,
    )


# ==================================================
# Calculate complete correlation map
# ==================================================
def calculate_correlation_map(
    keogram,
    distance_km,
    reference_distance_km,
    selected_distance_mask,
    shifts_pixels,
    minimum_points=20,
):
    """
    Correlate every selected distance row with the reference row
    for every requested shift.
    """
    keogram = np.asarray(
        keogram,
        dtype=float,
    )

    distance_km = np.asarray(
        distance_km,
        dtype=float,
    )

    shifts_pixels = np.asarray(
        shifts_pixels,
        dtype=int,
    )

    reference_index = np.argmin(
        np.abs(
            distance_km
            - reference_distance_km
        )
    )

    selected_indices = np.where(
        selected_distance_mask
    )[0]

    selected_distances = distance_km[
        selected_indices
    ]

    reference_row = keogram[
        :,
        reference_index,
    ]

    correlation_map = np.full(
        (
            len(selected_indices),
            len(shifts_pixels),
        ),
        np.nan,
        dtype=float,
    )

    for output_row_index, keogram_row_index in enumerate(
        selected_indices
    ):
        comparison_row = keogram[
            :,
            keogram_row_index,
        ]

        for shift_index, shift_pixels in enumerate(
            shifts_pixels
        ):
            correlation_map[
                output_row_index,
                shift_index,
            ] = calculate_shifted_correlation(
                reference_row=reference_row,
                comparison_row=comparison_row,
                shift_pixels=int(shift_pixels),
                minimum_points=minimum_points,
            )

    return (
        correlation_map,
        selected_distances,
        reference_index,
    )


# ==================================================
# Maximum-correlation shift for each row
# ==================================================
def find_best_shifts(
    correlation_map,
    shifts_minutes,
):
    """
    Find the shift in minutes having the largest correlation
    for every distance row.
    """
    correlation_map = np.asarray(
        correlation_map,
        dtype=float,
    )

    shifts_minutes = np.asarray(
        shifts_minutes,
        dtype=float,
    )

    best_shift_minutes = np.full(
        correlation_map.shape[0],
        np.nan,
        dtype=float,
    )

    maximum_correlation = np.full(
        correlation_map.shape[0],
        np.nan,
        dtype=float,
    )

    for row_index in range(
        correlation_map.shape[0]
    ):
        correlation_row = correlation_map[
            row_index,
            :
        ]

        finite = np.isfinite(
            correlation_row
        )

        if not np.any(finite):
            continue

        finite_indices = np.where(
            finite
        )[0]

        local_max_index = np.argmax(
            correlation_row[finite]
        )

        full_max_index = finite_indices[
            local_max_index
        ]

        best_shift_minutes[row_index] = (
            shifts_minutes[full_max_index]
        )

        maximum_correlation[row_index] = (
            correlation_row[full_max_index]
        )

    return (
        best_shift_minutes,
        maximum_correlation,
    )


# ==================================================
# Extract the correlation-shift profile at 0 km
# ==================================================
def extract_zero_km_correlation_profile(
    correlation_map,
    selected_distances,
):
    """
    Return the correlation-versus-shift profile for the selected
    distance row nearest 0 km.
    """
    correlation_map = np.asarray(
        correlation_map,
        dtype=float,
    )

    selected_distances = np.asarray(
        selected_distances,
        dtype=float,
    )

    if correlation_map.ndim != 2:
        raise ValueError(
            "correlation_map must be two-dimensional."
        )

    if correlation_map.shape[0] != len(selected_distances):
        raise ValueError(
            "The number of correlation-map rows must equal "
            "the number of selected distances."
        )

    zero_distance_row_index = int(
        np.argmin(
            np.abs(selected_distances)
        )
    )

    zero_distance_km = float(
        selected_distances[zero_distance_row_index]
    )

    zero_km_profile = correlation_map[
        zero_distance_row_index,
        :
    ].copy()

    return (
        zero_km_profile,
        zero_distance_row_index,
        zero_distance_km,
    )


# ==================================================
# Find strongest local maximum on each side of zero
# ==================================================
def find_left_and_right_local_maxima(
    correlation_profile,
    shifts_minutes,
    minimum_prominence=0.0,
    minimum_distance_pixels=1,
):
    """
    Find the strongest finite local maximum at negative shift
    and the strongest finite local maximum at positive shift.

    The point at zero shift is not included in either search.

    If scipy.signal.find_peaks finds no strict local maximum on one
    side, the largest finite value on that side is used as a fallback.
    """
    correlation_profile = np.asarray(
        correlation_profile,
        dtype=float,
    )

    shifts_minutes = np.asarray(
        shifts_minutes,
        dtype=float,
    )

    if correlation_profile.shape != shifts_minutes.shape:
        raise ValueError(
            "correlation_profile and shifts_minutes must "
            "have the same shape."
        )

    def find_one_side(side_mask):
        side_indices = np.where(
            side_mask
            & np.isfinite(correlation_profile)
        )[0]

        if len(side_indices) == 0:
            return {
                "index": -1,
                "shift_minutes": np.nan,
                "correlation": np.nan,
                "used_fallback": False,
            }

        side_values = correlation_profile[
            side_indices
        ]

        local_indices, properties = find_peaks(
            side_values,
            prominence=minimum_prominence,
            distance=minimum_distance_pixels,
        )

        used_fallback = False

        if len(local_indices) > 0:
            local_values = side_values[
                local_indices
            ]

            strongest_local_position = local_indices[
                int(np.nanargmax(local_values))
            ]

            full_index = int(
                side_indices[strongest_local_position]
            )
        else:
            used_fallback = True

            strongest_side_position = int(
                np.nanargmax(side_values)
            )

            full_index = int(
                side_indices[strongest_side_position]
            )

        return {
            "index": full_index,
            "shift_minutes": float(
                shifts_minutes[full_index]
            ),
            "correlation": float(
                correlation_profile[full_index]
            ),
            "used_fallback": used_fallback,
        }

    left_result = find_one_side(
        shifts_minutes < 0.0
    )

    right_result = find_one_side(
        shifts_minutes > 0.0
    )

    return {
        "left": left_result,
        "right": right_result,
    }


# ==================================================
# Fit maximum-correlation ridge
# ==================================================
def fit_maximum_correlation_slope(
    distances_km,
    best_shift_minutes,
    maximum_correlation,
    minimum_correlation=0.5,
    distance_min_km=None,
    distance_max_km=None,
):
    """
    Fit:

        shift_minutes = slope * distance_km + intercept
    """
    distances_km = np.asarray(
        distances_km,
        dtype=float,
    )

    best_shift_minutes = np.asarray(
        best_shift_minutes,
        dtype=float,
    )

    maximum_correlation = np.asarray(
        maximum_correlation,
        dtype=float,
    )

    fit_mask = (
        np.isfinite(distances_km)
        & np.isfinite(best_shift_minutes)
        & np.isfinite(maximum_correlation)
        & (
            maximum_correlation
            >= minimum_correlation
        )
    )

    if distance_min_km is not None:
        fit_mask &= (
            distances_km
            >= distance_min_km
        )

    if distance_max_km is not None:
        fit_mask &= (
            distances_km
            <= distance_max_km
        )

    number_of_points = np.count_nonzero(
        fit_mask
    )

    if number_of_points < 2:
        return {
            "slope_min_per_km": np.nan,
            "intercept_minutes": np.nan,
            "speed_km_per_min": np.nan,
            "speed_m_per_s": np.nan,
            "r_squared": np.nan,
            "number_of_fit_points": number_of_points,
            "fit_mask": fit_mask,
        }

    fit_distances = distances_km[
        fit_mask
    ]

    fit_shifts = best_shift_minutes[
        fit_mask
    ]

    slope_min_per_km, intercept_minutes = np.polyfit(
        fit_distances,
        fit_shifts,
        deg=1,
    )

    predicted_shifts = (
        slope_min_per_km
        * fit_distances
        + intercept_minutes
    )

    residual_sum_squares = np.sum(
        (
            fit_shifts
            - predicted_shifts
        ) ** 2
    )

    total_sum_squares = np.sum(
        (
            fit_shifts
            - np.mean(fit_shifts)
        ) ** 2
    )

    if total_sum_squares > 0:
        r_squared = (
            1.0
            - residual_sum_squares
            / total_sum_squares
        )
    else:
        r_squared = np.nan

    if np.isclose(
        slope_min_per_km,
        0.0,
    ):
        speed_km_per_min = np.nan
        speed_m_per_s = np.nan
    else:
        speed_km_per_min = (
            1.0
            / slope_min_per_km
        )

        speed_m_per_s = (
            speed_km_per_min
            * 1000.0
            / 60.0
        )

    return {
        "slope_min_per_km": slope_min_per_km,
        "intercept_minutes": intercept_minutes,
        "speed_km_per_min": speed_km_per_min,
        "speed_m_per_s": speed_m_per_s,
        "r_squared": r_squared,
        "number_of_fit_points": number_of_points,
        "fit_mask": fit_mask,
    }


# ==================================================
# Keogram color limit
# ==================================================
def determine_keogram_abs_limit(
    full_keogram,
    limited_keogram,
):
    """
    Return one symmetric absolute color limit shared by both
    keogram panels.
    """
    if KEOGRAM_ABS_LIMIT is not None:
        abs_limit = float(
            KEOGRAM_ABS_LIMIT
        )

        if (
            not np.isfinite(abs_limit)
            or abs_limit <= 0
        ):
            raise ValueError(
                "KEOGRAM_ABS_LIMIT must be positive and finite."
            )

        return abs_limit

    finite_values = []

    for data in [
        full_keogram,
        limited_keogram,
    ]:
        data = np.asarray(
            data,
            dtype=float,
        )

        finite = np.abs(
            data[np.isfinite(data)]
        )

        if finite.size > 0:
            finite_values.append(
                finite
            )

    if len(finite_values) == 0:
        return 1.0

    all_finite = np.concatenate(
        finite_values
    )

    abs_limit = np.nanpercentile(
        all_finite,
        KEOGRAM_AUTO_PERCENTILE,
    )

    if (
        not np.isfinite(abs_limit)
        or abs_limit <= 0
    ):
        abs_limit = np.nanmax(
            all_finite
        )

    if (
        not np.isfinite(abs_limit)
        or abs_limit <= 0
    ):
        abs_limit = 1.0

    return float(
        abs_limit
    )


# ==================================================
# Plot keograms, correlation map, ridge, and profile
# ==================================================
def plot_correlation_map(
    angle_deg,
    full_keogram,
    full_times_min,
    full_distance_km,
    limited_keogram,
    limited_times_min,
    limited_distance_km,
    correlation_map,
    shifts_minutes,
    selected_distances,
    best_shift_minutes,
    maximum_correlation,
    slope_results,
    zero_km_correlation_profile,
    zero_km_distance,
    zero_km_local_maxima,
    dataset_name,
    output_file,
):
    """
    Plot:
      1. full-range original keogram;
      2. limited-range smoothed/corrected keogram;
      3. correlation versus time shift and distance;
      4. correlation versus shift for the 0 km row.
    """
    fig, (
        ax_full,
        ax_limited,
        ax_map,
        ax_profile,
    ) = plt.subplots(
        nrows=4,
        ncols=1,
        figsize=(11, 16),
        gridspec_kw={
            "height_ratios": [
                1.55,
                1.55,
                2.5,
                1.15,
            ],
        },
    )

    keogram_abs_limit = determine_keogram_abs_limit(
        full_keogram=full_keogram,
        limited_keogram=limited_keogram,
    )

    # ==================================================
    # Full-range original keogram
    # ==================================================
    full_mesh = ax_full.pcolormesh(
        full_times_min,
        full_distance_km,
        full_keogram.T,
        shading="auto",
        cmap=KEOGRAM_CMAP,
        vmin=-keogram_abs_limit,
        vmax=keogram_abs_limit,
    )

    full_cbar = fig.colorbar(
        full_mesh,
        ax=ax_full,
        fraction=0.025,
        pad=0.02,
    )

    full_cbar.set_label(
        "Keogram value"
    )

    ax_full.axhline(
        0.0,
        color="black",
        linestyle="--",
        linewidth=1.0,
    )

    ax_full.set_xlim(
        np.nanmin(full_times_min),
        np.nanmax(full_times_min),
    )

    ax_full.set_ylim(
        np.nanmin(full_distance_km),
        np.nanmax(full_distance_km),
    )

    ax_full.set_title(
        f"{angle_deg}° Full-Range Original Keogram"
    )

    ax_full.set_xlabel(
        "Time (min)"
    )

    ax_full.set_ylabel(
        "Distance along slice (km)"
    )

    # ==================================================
    # Limited-range processed keogram
    # ==================================================
    limited_mesh = ax_limited.pcolormesh(
        limited_times_min,
        limited_distance_km,
        limited_keogram.T,
        shading="auto",
        cmap=KEOGRAM_CMAP,
        vmin=-keogram_abs_limit,
        vmax=keogram_abs_limit,
    )

    limited_cbar = fig.colorbar(
        limited_mesh,
        ax=ax_limited,
        fraction=0.025,
        pad=0.02,
    )

    limited_cbar.set_label(
        "Keogram value"
    )

    ax_limited.axhline(
        0.0,
        color="black",
        linestyle="--",
        linewidth=1.0,
    )

    ax_limited.set_xlim(
        CORRELATION_TIME_MIN,
        CORRELATION_TIME_MAX,
    )

    ax_limited.set_ylim(
        CORRELATION_DISTANCE_MIN_KM,
        CORRELATION_DISTANCE_MAX_KM,
    )

    ax_limited.set_title(
        "Limited-Range Processed Keogram\n"
        f"{dataset_name}"
    )

    ax_limited.set_xlabel(
        "Time (min)"
    )

    ax_limited.set_ylabel(
        "Distance along slice (km)"
    )

    # ==================================================
    # Correlation map
    # ==================================================
    if len(shifts_minutes) > 1:
        shift_spacing = np.median(
            np.diff(shifts_minutes)
        )
    else:
        shift_spacing = 1.0

    if len(selected_distances) > 1:
        distance_spacing = np.median(
            np.diff(selected_distances)
        )
    else:
        distance_spacing = 1.0

    im = ax_map.imshow(
        correlation_map,
        origin="lower",
        aspect="auto",
        interpolation="nearest",
        cmap=CORRELATION_CMAP,
        vmin=CORRELATION_VMIN,
        vmax=CORRELATION_VMAX,
        extent=[
            shifts_minutes[0]
            - shift_spacing / 2.0,
            shifts_minutes[-1]
            + shift_spacing / 2.0,
            selected_distances[0]
            - distance_spacing / 2.0,
            selected_distances[-1]
            + distance_spacing / 2.0,
        ],
    )

    cbar = fig.colorbar(
        im,
        ax=ax_map,
        fraction=0.025,
        pad=0.02,
    )

    cbar.set_label(
        "Pearson correlation with 0 km row"
    )

    ax_map.axvline(
        0.0,
        color="black",
        linestyle="--",
        linewidth=1.2,
        label="Zero shift",
    )

    ax_map.axhline(
        zero_km_distance,
        color="white",
        linestyle=":",
        linewidth=1.5,
        label=f"Profile row ({zero_km_distance:g} km)",
    )

    ridge_mask = (
        np.isfinite(best_shift_minutes)
        & np.isfinite(maximum_correlation)
        & (
            maximum_correlation
            >= MIN_CORRELATION_FOR_SLOPE
        )
    )

    if np.any(ridge_mask):
        ax_map.plot(
            best_shift_minutes[ridge_mask],
            selected_distances[ridge_mask],
            color="black",
            linewidth=1.2,
            label="Maximum correlation",
        )

        ax_map.scatter(
            best_shift_minutes[ridge_mask],
            selected_distances[ridge_mask],
            color="black",
            s=8,
        )

    slope = slope_results[
        "slope_min_per_km"
    ]

    intercept = slope_results[
        "intercept_minutes"
    ]

    r_squared = slope_results[
        "r_squared"
    ]

    speed_m_per_s = slope_results[
        "speed_m_per_s"
    ]

    if np.isfinite(slope):
        fit_distances = np.linspace(
            SLOPE_FIT_DISTANCE_MIN_KM,
            SLOPE_FIT_DISTANCE_MAX_KM,
            300,
        )

        fit_shifts = (
            slope
            * fit_distances
            + intercept
        )

        ax_map.plot(
            fit_shifts,
            fit_distances,
            color="lime",
            linestyle="--",
            linewidth=2.0,
            label="Linear ridge fit",
        )

        fit_text = (
            f"Shift slope = {slope:.5f} min/km\n"
            f"Equivalent speed = {speed_m_per_s:.2f} m/s\n"
            f"$R^2$ = {r_squared:.3f}"
        )

        ax_map.text(
            0.02,
            0.03,
            fit_text,
            transform=ax_map.transAxes,
            ha="left",
            va="bottom",
            fontsize=9,
            bbox={
                "facecolor": "white",
                "edgecolor": "black",
                "alpha": 0.85,
            },
        )

    ax_map.set_xlim(
        SHIFT_MIN_MINUTES,
        SHIFT_MAX_MINUTES,
    )

    ax_map.set_ylim(
        CORRELATION_DISTANCE_MIN_KM,
        CORRELATION_DISTANCE_MAX_KM,
    )

    ax_map.set_title(
        f"{angle_deg}° Shifted Row Correlation\n"
        f"{dataset_name}"
    )

    ax_map.set_xlabel(
        "Shift applied to row (min)"
    )

    ax_map.set_ylabel(
        "Distance along slice (km)"
    )

    ax_map.legend(
        loc="upper right",
        fontsize=8,
    )

    # ==================================================
    # 0 km correlation-shift line
    # ==================================================
    ax_profile.plot(
        shifts_minutes,
        zero_km_correlation_profile,
        color="black",
        linewidth=1.4,
        label=f"{zero_km_distance:g} km correlation profile",
    )

    ax_profile.axvline(
        0.0,
        color="gray",
        linestyle="--",
        linewidth=1.0,
    )

    left_peak = zero_km_local_maxima[
        "left"
    ]

    right_peak = zero_km_local_maxima[
        "right"
    ]

    if np.isfinite(
        left_peak["shift_minutes"]
    ):
        ax_profile.scatter(
            left_peak["shift_minutes"],
            left_peak["correlation"],
            s=55,
            marker="o",
            zorder=5,
            label=(
                "Left local max: "
                f"{left_peak['shift_minutes']:.2f} min"
            ),
        )

        ax_profile.axvline(
            left_peak["shift_minutes"],
            linestyle=":",
            linewidth=1.2,
        )

        ax_profile.annotate(
            (
                f"{left_peak['shift_minutes']:.2f} min\n"
                f"r={left_peak['correlation']:.3f}"
            ),
            xy=(
                left_peak["shift_minutes"],
                left_peak["correlation"],
            ),
            xytext=(8, 15),
            textcoords="offset points",
            ha="left",
            va="bottom",
            fontsize=8,
            arrowprops={
                "arrowstyle": "->",
                "linewidth": 0.8,
            },
        )

    if np.isfinite(
        right_peak["shift_minutes"]
    ):
        ax_profile.scatter(
            right_peak["shift_minutes"],
            right_peak["correlation"],
            s=55,
            marker="s",
            zorder=5,
            label=(
                "Right local max: "
                f"{right_peak['shift_minutes']:.2f} min"
            ),
        )

        ax_profile.axvline(
            right_peak["shift_minutes"],
            linestyle=":",
            linewidth=1.2,
        )

        ax_profile.annotate(
            (
                f"{right_peak['shift_minutes']:.2f} min\n"
                f"r={right_peak['correlation']:.3f}"
            ),
            xy=(
                right_peak["shift_minutes"],
                right_peak["correlation"],
            ),
            xytext=(-8, 15),
            textcoords="offset points",
            ha="right",
            va="bottom",
            fontsize=8,
            arrowprops={
                "arrowstyle": "->",
                "linewidth": 0.8,
            },
        )

    ax_profile.set_xlim(
        SHIFT_MIN_MINUTES,
        SHIFT_MAX_MINUTES,
    )

    ax_profile.set_xlabel(
        "Shift applied to row (min)"
    )

    ax_profile.set_ylabel(
        "Correlation"
    )

    ax_profile.set_title(
        f"Correlation-Shift Line at {zero_km_distance:g} km"
    )

    ax_profile.grid(
        True,
        alpha=0.3,
    )

    ax_profile.legend(
        loc="best",
        fontsize=8,
    )

    plt.tight_layout()

    fig.savefig(
        output_file,
        dpi=DPI,
        bbox_inches="tight",
    )

    plt.close(fig)


# ==================================================
# Print slope results
# ==================================================
def print_slope_results(
    angle_deg,
    label,
    results,
):
    print(
        f"\n{angle_deg}° — {label}:"
    )

    print(
        f"  Shift slope: "
        f"{results['slope_min_per_km']:.8f} min/km"
    )

    print(
        f"  Intercept: "
        f"{results['intercept_minutes']:.6f} min"
    )

    print(
        f"  Equivalent speed: "
        f"{results['speed_m_per_s']:.4f} m/s"
    )

    print(
        f"  R²: "
        f"{results['r_squared']:.6f}"
    )

    print(
        f"  Points used: "
        f"{results['number_of_fit_points']}"
    )


# ==================================================
# Output paths
# ==================================================
sigma_t_label = safe_number_label(
    SMOOTH_SIGMA_TIME_PIXELS
)

sigma_d_label = safe_number_label(
    SMOOTH_SIGMA_DISTANCE_PIXELS
)

distance_folder = (
    f"r_"
    f"{int(CORRELATION_DISTANCE_MIN_KM)}_"
    f"{int(CORRELATION_DISTANCE_MAX_KM)}"
)

shift_folder = (
    f"shift_min_"
    f"{safe_number_label(SHIFT_MIN_MINUTES)}_"
    f"{safe_number_label(SHIFT_MAX_MINUTES)}"
)

if APPLY_2D_SMOOTHING:
    smoothing_folder = (
        f"sig_"
        f"{sigma_t_label}_"
        f"{sigma_d_label}"
    )
else:
    smoothing_folder = "raw"

OUT_DIR = (
    BASE_OUT_DIR
    / distance_folder
    / smoothing_folder
    / "pchip_bias"
    / "row_correlation_minutes_all_angles"
    / shift_folder
)

UNCORRECTED_OUT_DIR = (
    OUT_DIR
    / "uncorrected"
)

CORRECTED_OUT_DIR = (
    OUT_DIR
    / "corrected"
)

SLOPE_REPORT_OUT_DIR = (
    OUT_DIR
    / "slope_reports"
)

for directory in [
    OUT_DIR,
    UNCORRECTED_OUT_DIR,
    CORRECTED_OUT_DIR,
    SLOPE_REPORT_OUT_DIR,
]:
    directory.mkdir(
        parents=True,
        exist_ok=True,
    )


# ==================================================
# Read common time array
# ==================================================
TIMES_FILE = (
    BASE_OUT_DIR
    / "times_s.npy"
)

times_s = np.load(
    TIMES_FILE
).astype(float)

times_min = times_s / 60.0

if len(times_min) < 2:
    raise ValueError(
        "At least two time samples are required."
    )


# ==================================================
# Determine minutes per time pixel
# ==================================================
time_differences_min = np.diff(
    times_min
)

finite_time_differences = time_differences_min[
    np.isfinite(time_differences_min)
]

if len(finite_time_differences) == 0:
    raise ValueError(
        "Could not determine the time spacing."
    )

MINUTES_PER_PIXEL = np.median(
    finite_time_differences
)

if (
    not np.isfinite(MINUTES_PER_PIXEL)
    or MINUTES_PER_PIXEL <= 0
):
    raise ValueError(
        "The calculated minutes-per-pixel value is invalid."
    )

maximum_time_spacing_deviation = np.max(
    np.abs(
        finite_time_differences
        - MINUTES_PER_PIXEL
    )
)

print(
    "Angles to process:"
)

print(
    KEOGRAM_ANGLES_DEG
)

print(
    f"Number of angles: "
    f"{len(KEOGRAM_ANGLES_DEG)}"
)

print(
    "\nTime sampling:"
)

print(
    f"One time pixel = "
    f"{MINUTES_PER_PIXEL:.6f} minutes"
)

print(
    f"One time pixel = "
    f"{MINUTES_PER_PIXEL * 60.0:.3f} seconds"
)

print(
    f"Maximum spacing deviation = "
    f"{maximum_time_spacing_deviation:.6e} minutes"
)


# ==================================================
# Convert minute shifts to integer pixel shifts
# ==================================================
shift_min_pixels = int(
    np.ceil(
        SHIFT_MIN_MINUTES
        / MINUTES_PER_PIXEL
    )
)

shift_max_pixels = int(
    np.floor(
        SHIFT_MAX_MINUTES
        / MINUTES_PER_PIXEL
    )
)

shifts_pixels = np.arange(
    shift_min_pixels,
    shift_max_pixels + 1,
    dtype=int,
)

shifts_minutes = (
    shifts_pixels
    * MINUTES_PER_PIXEL
)

print(
    "\nShift range:"
)

print(
    f"Requested: "
    f"{SHIFT_MIN_MINUTES:.2f} to "
    f"{SHIFT_MAX_MINUTES:.2f} min"
)

print(
    f"Actual: "
    f"{shifts_minutes[0]:.2f} to "
    f"{shifts_minutes[-1]:.2f} min"
)

print(
    f"Pixel shifts: "
    f"{shifts_pixels[0]} to "
    f"{shifts_pixels[-1]}"
)


# ==================================================
# Select common time range
# ==================================================
time_mask = (
    (times_min >= CORRELATION_TIME_MIN)
    & (times_min <= CORRELATION_TIME_MAX)
)

if np.count_nonzero(time_mask) < MIN_VALID_CORRELATION_POINTS:
    raise ValueError(
        "The selected time interval has too few samples."
    )

times_correlation_min = times_min[
    time_mask
]


# ==================================================
# Save common arrays
# ==================================================
np.save(
    OUT_DIR / "angles_deg.npy",
    KEOGRAM_ANGLES_DEG,
)

np.save(
    OUT_DIR / "shifts_minutes.npy",
    shifts_minutes,
)

np.save(
    OUT_DIR / "shifts_pixels.npy",
    shifts_pixels,
)

np.save(
    OUT_DIR / "correlation_times_min.npy",
    times_correlation_min,
)


# ==================================================
# Storage for combined angle results
# ==================================================
combined_results = []
combined_zero_km_peak_results = []

successful_angles = []
failed_angles = []


# ==================================================
# Process every angle
# ==================================================
for angle_number, angle_deg in enumerate(
    KEOGRAM_ANGLES_DEG,
    start=1,
):
    print(
        "\n"
        + "=" * 70
    )

    print(
        f"Processing angle "
        f"{angle_deg}° "
        f"({angle_number}/{len(KEOGRAM_ANGLES_DEG)})"
    )

    print(
        "=" * 70
    )

    KEOGRAM_FILE = (
        BASE_OUT_DIR
        / f"BLO_keogram_angle_{angle_deg:03d}.npy"
    )

    if not KEOGRAM_FILE.exists():
        print(
            f"WARNING: File not found: "
            f"{KEOGRAM_FILE}"
        )

        failed_angles.append(
            (
                int(angle_deg),
                "file not found",
            )
        )

        continue

    try:
        # ==================================================
        # Read keogram
        # ==================================================
        keogram_original = np.load(
            KEOGRAM_FILE
        ).astype(float)

        if keogram_original.ndim != 2:
            raise ValueError(
                f"Expected a 2D keogram, found "
                f"{keogram_original.shape}."
            )

        nt, nspace = keogram_original.shape

        if len(times_min) != nt:
            raise ValueError(
                f"Keogram has {nt} time samples, but "
                f"times_s.npy has {len(times_min)}."
            )

        # ==================================================
        # Construct distance axis
        # ==================================================
        distance_km = np.arange(
            -(nspace // 2),
            nspace // 2 + 1,
            dtype=float,
        )

        if len(distance_km) != nspace:
            distance_km = (
                np.arange(
                    nspace,
                    dtype=float,
                )
                - nspace // 2
            )

        available_distance_min = np.nanmin(
            distance_km
        )

        available_distance_max = np.nanmax(
            distance_km
        )

        if (
            CORRELATION_DISTANCE_MIN_KM
            < available_distance_min
            or CORRELATION_DISTANCE_MAX_KM
            > available_distance_max
        ):
            raise ValueError(
                f"Requested distance range "
                f"{CORRELATION_DISTANCE_MIN_KM} to "
                f"{CORRELATION_DISTANCE_MAX_KM} km "
                f"is outside available range "
                f"{available_distance_min} to "
                f"{available_distance_max} km."
            )

        # ==================================================
        # Select time interval
        # ==================================================
        keogram_time_selected = keogram_original[
            time_mask,
            :
        ]

        # ==================================================
        # Optional smoothing
        # ==================================================
        if APPLY_2D_SMOOTHING:
            keogram_uncorrected = smooth_keogram_2d(
                data=keogram_time_selected,
                sigma_time=SMOOTH_SIGMA_TIME_PIXELS,
                sigma_distance=SMOOTH_SIGMA_DISTANCE_PIXELS,
                mode=SMOOTH_MODE,
                weight_tolerance=SMOOTH_WEIGHT_TOLERANCE,
            )

            uncorrected_dataset_name = (
                f"Uncorrected, smoothed "
                f"σt={SMOOTH_SIGMA_TIME_PIXELS:g}, "
                f"σd={SMOOTH_SIGMA_DISTANCE_PIXELS:g}"
            )
        else:
            keogram_uncorrected = (
                keogram_time_selected.copy()
            )

            uncorrected_dataset_name = (
                "Uncorrected, unsmoothed"
            )

        # ==================================================
        # Create bias correction
        # ==================================================
        (
            bias_correction_keogram,
            bias_vs_time,
            bias_distance_factor,
        ) = create_bias_correction_keogram(
            times_min=times_correlation_min,
            distance_km=distance_km,
            bias_time_points_min=BIAS_TIME_POINTS_MIN,
            bias_at_max_distance=BIAS_AT_MAX_DISTANCE,
            bias_distance_min_km=BIAS_DISTANCE_MIN_KM,
            bias_distance_max_km=BIAS_DISTANCE_MAX_KM,
            distance_power=BIAS_DISTANCE_POWER,
            zero_outside_distance_range=(
                BIAS_ZERO_OUTSIDE_DISTANCE_RANGE
            ),
            zero_outside_time_range=(
                BIAS_ZERO_OUTSIDE_TIME_RANGE
            ),
            clip_negative_time_correction=(
                CLIP_NEGATIVE_TIME_CORRECTION
            ),
        )

        # ==================================================
        # Apply bias correction
        # ==================================================
        if APPLY_BIAS_CORRECTION:
            keogram_corrected = (
                keogram_uncorrected
                + bias_correction_keogram
            )

            corrected_dataset_name = (
                "Bias corrected: PCHIP time dependence, "
                f"distance power={BIAS_DISTANCE_POWER:g}"
            )
        else:
            keogram_corrected = (
                keogram_uncorrected.copy()
            )

            corrected_dataset_name = (
                "Bias correction disabled"
            )

        keogram_corrected[
            ~np.isfinite(keogram_uncorrected)
        ] = np.nan

        # ==================================================
        # Select distance rows
        # ==================================================
        selected_distance_mask = (
            (
                distance_km
                >= CORRELATION_DISTANCE_MIN_KM
            )
            & (
                distance_km
                <= CORRELATION_DISTANCE_MAX_KM
            )
        )

        selected_distances = distance_km[
            selected_distance_mask
        ]

        if len(selected_distances) == 0:
            raise ValueError(
                "No rows lie inside the selected distance range."
            )

        # ==================================================
        # Per-angle folders
        # ==================================================
        angle_uncorrected_dir = (
            UNCORRECTED_OUT_DIR
            / f"angle_{angle_deg:03d}"
        )

        angle_corrected_dir = (
            CORRECTED_OUT_DIR
            / f"angle_{angle_deg:03d}"
        )

        angle_uncorrected_dir.mkdir(
            parents=True,
            exist_ok=True,
        )

        angle_corrected_dir.mkdir(
            parents=True,
            exist_ok=True,
        )

        # ==================================================
        # Uncorrected map
        # ==================================================
        (
            correlation_map_uncorrected,
            selected_distances_uncorrected,
            reference_index_uncorrected,
        ) = calculate_correlation_map(
            keogram=keogram_uncorrected,
            distance_km=distance_km,
            reference_distance_km=0.0,
            selected_distance_mask=selected_distance_mask,
            shifts_pixels=shifts_pixels,
            minimum_points=MIN_VALID_CORRELATION_POINTS,
        )

        (
            best_shift_uncorrected_minutes,
            maximum_correlation_uncorrected,
        ) = find_best_shifts(
            correlation_map=correlation_map_uncorrected,
            shifts_minutes=shifts_minutes,
        )

        uncorrected_slope_results = (
            fit_maximum_correlation_slope(
                distances_km=(
                    selected_distances_uncorrected
                ),
                best_shift_minutes=(
                    best_shift_uncorrected_minutes
                ),
                maximum_correlation=(
                    maximum_correlation_uncorrected
                ),
                minimum_correlation=(
                    MIN_CORRELATION_FOR_SLOPE
                ),
                distance_min_km=(
                    SLOPE_FIT_DISTANCE_MIN_KM
                ),
                distance_max_km=(
                    SLOPE_FIT_DISTANCE_MAX_KM
                ),
            )
        )

        # ==================================================
        # Corrected map
        # ==================================================
        (
            correlation_map_corrected,
            selected_distances_corrected,
            reference_index_corrected,
        ) = calculate_correlation_map(
            keogram=keogram_corrected,
            distance_km=distance_km,
            reference_distance_km=0.0,
            selected_distance_mask=selected_distance_mask,
            shifts_pixels=shifts_pixels,
            minimum_points=MIN_VALID_CORRELATION_POINTS,
        )

        (
            best_shift_corrected_minutes,
            maximum_correlation_corrected,
        ) = find_best_shifts(
            correlation_map=correlation_map_corrected,
            shifts_minutes=shifts_minutes,
        )

        corrected_slope_results = (
            fit_maximum_correlation_slope(
                distances_km=(
                    selected_distances_corrected
                ),
                best_shift_minutes=(
                    best_shift_corrected_minutes
                ),
                maximum_correlation=(
                    maximum_correlation_corrected
                ),
                minimum_correlation=(
                    MIN_CORRELATION_FOR_SLOPE
                ),
                distance_min_km=(
                    SLOPE_FIT_DISTANCE_MIN_KM
                ),
                distance_max_km=(
                    SLOPE_FIT_DISTANCE_MAX_KM
                ),
            )
        )

        # ==================================================
        # Extract 0 km correlation-shift profiles
        # ==================================================
        (
            zero_km_profile_uncorrected,
            zero_km_row_uncorrected,
            zero_km_distance_uncorrected,
        ) = extract_zero_km_correlation_profile(
            correlation_map=correlation_map_uncorrected,
            selected_distances=selected_distances_uncorrected,
        )

        (
            zero_km_profile_corrected,
            zero_km_row_corrected,
            zero_km_distance_corrected,
        ) = extract_zero_km_correlation_profile(
            correlation_map=correlation_map_corrected,
            selected_distances=selected_distances_corrected,
        )

        zero_km_peaks_uncorrected = (
            find_left_and_right_local_maxima(
                correlation_profile=zero_km_profile_uncorrected,
                shifts_minutes=shifts_minutes,
                minimum_prominence=(
                    LOCAL_MAXIMUM_MIN_PROMINENCE
                ),
                minimum_distance_pixels=(
                    LOCAL_MAXIMUM_MIN_DISTANCE_PIXELS
                ),
            )
        )

        zero_km_peaks_corrected = (
            find_left_and_right_local_maxima(
                correlation_profile=zero_km_profile_corrected,
                shifts_minutes=shifts_minutes,
                minimum_prominence=(
                    LOCAL_MAXIMUM_MIN_PROMINENCE
                ),
                minimum_distance_pixels=(
                    LOCAL_MAXIMUM_MIN_DISTANCE_PIXELS
                ),
            )
        )

        # ==================================================
        # Plot maps
        # ==================================================
        uncorrected_plot_file = (
            angle_uncorrected_dir
            / (
                f"correlation_minutes_"
                f"{angle_deg:03d}_uncorrected.png"
            )
        )

        corrected_plot_file = (
            angle_corrected_dir
            / (
                f"correlation_minutes_"
                f"{angle_deg:03d}_corrected.png"
            )
        )

        plot_correlation_map(
            angle_deg=angle_deg,
            full_keogram=keogram_original,
            full_times_min=times_min,
            full_distance_km=distance_km,
            limited_keogram=keogram_uncorrected[
                :,
                selected_distance_mask,
            ],
            limited_times_min=times_correlation_min,
            limited_distance_km=selected_distances_uncorrected,
            correlation_map=(
                correlation_map_uncorrected
            ),
            shifts_minutes=shifts_minutes,
            selected_distances=(
                selected_distances_uncorrected
            ),
            best_shift_minutes=(
                best_shift_uncorrected_minutes
            ),
            maximum_correlation=(
                maximum_correlation_uncorrected
            ),
            slope_results=(
                uncorrected_slope_results
            ),
            zero_km_correlation_profile=(
                zero_km_profile_uncorrected
            ),
            zero_km_distance=(
                zero_km_distance_uncorrected
            ),
            zero_km_local_maxima=(
                zero_km_peaks_uncorrected
            ),
            dataset_name=uncorrected_dataset_name,
            output_file=uncorrected_plot_file,
        )

        plot_correlation_map(
            angle_deg=angle_deg,
            full_keogram=keogram_original,
            full_times_min=times_min,
            full_distance_km=distance_km,
            limited_keogram=keogram_corrected[
                :,
                selected_distance_mask,
            ],
            limited_times_min=times_correlation_min,
            limited_distance_km=selected_distances_corrected,
            correlation_map=(
                correlation_map_corrected
            ),
            shifts_minutes=shifts_minutes,
            selected_distances=(
                selected_distances_corrected
            ),
            best_shift_minutes=(
                best_shift_corrected_minutes
            ),
            maximum_correlation=(
                maximum_correlation_corrected
            ),
            slope_results=(
                corrected_slope_results
            ),
            zero_km_correlation_profile=(
                zero_km_profile_corrected
            ),
            zero_km_distance=(
                zero_km_distance_corrected
            ),
            zero_km_local_maxima=(
                zero_km_peaks_corrected
            ),
            dataset_name=corrected_dataset_name,
            output_file=corrected_plot_file,
        )

        # ==================================================
        # Save arrays
        # ==================================================
        np.save(
            angle_uncorrected_dir
            / (
                f"correlation_"
                f"{angle_deg:03d}_uncorrected.npy"
            ),
            correlation_map_uncorrected,
        )

        np.save(
            angle_uncorrected_dir
            / (
                f"best_shift_minutes_"
                f"{angle_deg:03d}_uncorrected.npy"
            ),
            best_shift_uncorrected_minutes,
        )

        np.save(
            angle_uncorrected_dir
            / (
                f"maximum_correlation_"
                f"{angle_deg:03d}_uncorrected.npy"
            ),
            maximum_correlation_uncorrected,
        )

        np.save(
            angle_corrected_dir
            / (
                f"correlation_"
                f"{angle_deg:03d}_corrected.npy"
            ),
            correlation_map_corrected,
        )

        np.save(
            angle_corrected_dir
            / (
                f"best_shift_minutes_"
                f"{angle_deg:03d}_corrected.npy"
            ),
            best_shift_corrected_minutes,
        )

        np.save(
            angle_corrected_dir
            / (
                f"maximum_correlation_"
                f"{angle_deg:03d}_corrected.npy"
            ),
            maximum_correlation_corrected,
        )

        np.save(
            angle_uncorrected_dir
            / "distances_km.npy",
            selected_distances_uncorrected,
        )

        np.save(
            angle_corrected_dir
            / "distances_km.npy",
            selected_distances_corrected,
        )

        np.save(
            angle_uncorrected_dir
            / "zero_km_correlation_profile.npy",
            zero_km_profile_uncorrected,
        )

        np.save(
            angle_corrected_dir
            / "zero_km_correlation_profile.npy",
            zero_km_profile_corrected,
        )

        zero_km_profile_uncorrected_csv = np.column_stack(
            [
                shifts_minutes,
                zero_km_profile_uncorrected,
            ]
        )

        zero_km_profile_corrected_csv = np.column_stack(
            [
                shifts_minutes,
                zero_km_profile_corrected,
            ]
        )

        np.savetxt(
            angle_uncorrected_dir
            / "zero_km_correlation_profile.csv",
            zero_km_profile_uncorrected_csv,
            delimiter=",",
            header="shift_minutes,correlation",
            comments="",
        )

        np.savetxt(
            angle_corrected_dir
            / "zero_km_correlation_profile.csv",
            zero_km_profile_corrected_csv,
            delimiter=",",
            header="shift_minutes,correlation",
            comments="",
        )

        # ==================================================
        # Save ridge CSV files
        # ==================================================
        uncorrected_summary = np.column_stack(
            [
                selected_distances_uncorrected,
                best_shift_uncorrected_minutes,
                maximum_correlation_uncorrected,
            ]
        )

        corrected_summary = np.column_stack(
            [
                selected_distances_corrected,
                best_shift_corrected_minutes,
                maximum_correlation_corrected,
            ]
        )

        uncorrected_csv = (
            angle_uncorrected_dir
            / (
                f"ridge_"
                f"{angle_deg:03d}_uncorrected.csv"
            )
        )

        corrected_csv = (
            angle_corrected_dir
            / (
                f"ridge_"
                f"{angle_deg:03d}_corrected.csv"
            )
        )

        np.savetxt(
            uncorrected_csv,
            uncorrected_summary,
            delimiter=",",
            header=(
                "distance_km,"
                "best_shift_minutes,"
                "maximum_correlation"
            ),
            comments="",
        )

        np.savetxt(
            corrected_csv,
            corrected_summary,
            delimiter=",",
            header=(
                "distance_km,"
                "best_shift_minutes,"
                "maximum_correlation"
            ),
            comments="",
        )

        # ==================================================
        # Save individual slope report
        # ==================================================
        slope_output_file = (
            SLOPE_REPORT_OUT_DIR
            / (
                f"correlation_slope_"
                f"{angle_deg:03d}.txt"
            )
        )

        with open(
            slope_output_file,
            "w",
            encoding="utf-8",
        ) as file:
            file.write(
                f"Angle: {angle_deg} degrees\n"
            )

            file.write(
                f"Minutes per pixel: "
                f"{MINUTES_PER_PIXEL:.10f}\n\n"
            )

            file.write(
                "UNCORRECTED\n"
            )

            for key, value in (
                uncorrected_slope_results.items()
            ):
                if key != "fit_mask":
                    file.write(
                        f"{key}: {value}\n"
                    )

            file.write(
                "\nCORRECTED\n"
            )

            for key, value in (
                corrected_slope_results.items()
            ):
                if key != "fit_mask":
                    file.write(
                        f"{key}: {value}\n"
                    )

            file.write(
                "\n0 KM CORRELATION-PROFILE LOCAL MAXIMA\n"
            )

            file.write(
                "Uncorrected left shift (min): "
                f"{zero_km_peaks_uncorrected['left']['shift_minutes']}\n"
            )
            file.write(
                "Uncorrected left correlation: "
                f"{zero_km_peaks_uncorrected['left']['correlation']}\n"
            )
            file.write(
                "Uncorrected right shift (min): "
                f"{zero_km_peaks_uncorrected['right']['shift_minutes']}\n"
            )
            file.write(
                "Uncorrected right correlation: "
                f"{zero_km_peaks_uncorrected['right']['correlation']}\n"
            )
            file.write(
                "Corrected left shift (min): "
                f"{zero_km_peaks_corrected['left']['shift_minutes']}\n"
            )
            file.write(
                "Corrected left correlation: "
                f"{zero_km_peaks_corrected['left']['correlation']}\n"
            )
            file.write(
                "Corrected right shift (min): "
                f"{zero_km_peaks_corrected['right']['shift_minutes']}\n"
            )
            file.write(
                "Corrected right correlation: "
                f"{zero_km_peaks_corrected['right']['correlation']}\n"
            )

        # ==================================================
        # Add to combined summary
        # ==================================================
        combined_results.append(
            [
                float(angle_deg),

                uncorrected_slope_results[
                    "slope_min_per_km"
                ],
                uncorrected_slope_results[
                    "intercept_minutes"
                ],
                uncorrected_slope_results[
                    "speed_km_per_min"
                ],
                uncorrected_slope_results[
                    "speed_m_per_s"
                ],
                uncorrected_slope_results[
                    "r_squared"
                ],
                uncorrected_slope_results[
                    "number_of_fit_points"
                ],

                corrected_slope_results[
                    "slope_min_per_km"
                ],
                corrected_slope_results[
                    "intercept_minutes"
                ],
                corrected_slope_results[
                    "speed_km_per_min"
                ],
                corrected_slope_results[
                    "speed_m_per_s"
                ],
                corrected_slope_results[
                    "r_squared"
                ],
                corrected_slope_results[
                    "number_of_fit_points"
                ],
            ]
        )

        combined_zero_km_peak_results.append(
            [
                float(angle_deg),
                zero_km_peaks_uncorrected["left"][
                    "shift_minutes"
                ],
                zero_km_peaks_uncorrected["left"][
                    "correlation"
                ],
                zero_km_peaks_uncorrected["right"][
                    "shift_minutes"
                ],
                zero_km_peaks_uncorrected["right"][
                    "correlation"
                ],
                zero_km_peaks_corrected["left"][
                    "shift_minutes"
                ],
                zero_km_peaks_corrected["left"][
                    "correlation"
                ],
                zero_km_peaks_corrected["right"][
                    "shift_minutes"
                ],
                zero_km_peaks_corrected["right"][
                    "correlation"
                ],
            ]
        )

        successful_angles.append(
            int(angle_deg)
        )

        print_slope_results(
            angle_deg=angle_deg,
            label=(
                "Uncorrected maximum-correlation ridge"
            ),
            results=uncorrected_slope_results,
        )

        print_slope_results(
            angle_deg=angle_deg,
            label=(
                "Corrected maximum-correlation ridge"
            ),
            results=corrected_slope_results,
        )

        print(
            "\n0 km correlation-profile local maxima:"
        )
        print(
            "  Uncorrected left: "
            f"{zero_km_peaks_uncorrected['left']['shift_minutes']:.3f} min, "
            f"r={zero_km_peaks_uncorrected['left']['correlation']:.6f}"
        )
        print(
            "  Uncorrected right: "
            f"{zero_km_peaks_uncorrected['right']['shift_minutes']:.3f} min, "
            f"r={zero_km_peaks_uncorrected['right']['correlation']:.6f}"
        )
        print(
            "  Corrected left: "
            f"{zero_km_peaks_corrected['left']['shift_minutes']:.3f} min, "
            f"r={zero_km_peaks_corrected['left']['correlation']:.6f}"
        )
        print(
            "  Corrected right: "
            f"{zero_km_peaks_corrected['right']['shift_minutes']:.3f} min, "
            f"r={zero_km_peaks_corrected['right']['correlation']:.6f}"
        )

    except Exception as error:
        print(
            f"ERROR processing {angle_deg}°:"
        )

        print(
            error
        )

        failed_angles.append(
            (
                int(angle_deg),
                str(error),
            )
        )


# ==================================================
# Save combined 0 km local-maximum summary
# ==================================================
combined_zero_km_peak_results = np.asarray(
    combined_zero_km_peak_results,
    dtype=float,
)

combined_zero_km_peak_csv = (
    OUT_DIR
    / "all_angle_zero_km_local_maxima.csv"
)

combined_zero_km_peak_npy = (
    OUT_DIR
    / "all_angle_zero_km_local_maxima.npy"
)

combined_zero_km_peak_header = (
    "angle_deg,"
    "uncorrected_left_shift_minutes,"
    "uncorrected_left_correlation,"
    "uncorrected_right_shift_minutes,"
    "uncorrected_right_correlation,"
    "corrected_left_shift_minutes,"
    "corrected_left_correlation,"
    "corrected_right_shift_minutes,"
    "corrected_right_correlation"
)

if combined_zero_km_peak_results.size > 0:
    np.savetxt(
        combined_zero_km_peak_csv,
        combined_zero_km_peak_results,
        delimiter=",",
        header=combined_zero_km_peak_header,
        comments="",
    )

    np.save(
        combined_zero_km_peak_npy,
        combined_zero_km_peak_results,
    )


# ==================================================
# Save combined angle summary
# ==================================================
combined_results = np.asarray(
    combined_results,
    dtype=float,
)

combined_csv = (
    OUT_DIR
    / "all_angle_slope_summary.csv"
)

combined_npy = (
    OUT_DIR
    / "all_angle_slope_summary.npy"
)

combined_header = (
    "angle_deg,"
    "uncorrected_slope_min_per_km,"
    "uncorrected_intercept_minutes,"
    "uncorrected_speed_km_per_min,"
    "uncorrected_speed_m_per_s,"
    "uncorrected_r_squared,"
    "uncorrected_number_of_fit_points,"
    "corrected_slope_min_per_km,"
    "corrected_intercept_minutes,"
    "corrected_speed_km_per_min,"
    "corrected_speed_m_per_s,"
    "corrected_r_squared,"
    "corrected_number_of_fit_points"
)

if combined_results.size > 0:
    np.savetxt(
        combined_csv,
        combined_results,
        delimiter=",",
        header=combined_header,
        comments="",
    )

    np.save(
        combined_npy,
        combined_results,
    )


# ==================================================
# Plot speed versus angle
# ==================================================
if combined_results.size > 0:
    angles_processed = combined_results[
        :,
        0,
    ]

    uncorrected_speed_m_per_s = combined_results[
        :,
        4,
    ]

    corrected_speed_m_per_s = combined_results[
        :,
        10,
    ]

    fig_speed, ax_speed = plt.subplots(
        figsize=(10, 5)
    )

    ax_speed.plot(
        angles_processed,
        uncorrected_speed_m_per_s,
        marker="o",
        linewidth=1.5,
        label="Uncorrected",
    )

    ax_speed.plot(
        angles_processed,
        corrected_speed_m_per_s,
        marker="o",
        linewidth=1.5,
        label="Corrected",
    )

    ax_speed.axhline(
        0.0,
        color="black",
        linestyle="--",
        linewidth=1.0,
    )

    ax_speed.set_xlim(
        KEOGRAM_ANGLES_DEG.min(),
        KEOGRAM_ANGLES_DEG.max(),
    )

    ax_speed.set_xlabel(
        "Keogram angle (degrees)"
    )

    ax_speed.set_ylabel(
        "Equivalent signed speed (m/s)"
    )

    ax_speed.set_title(
        "Maximum-Correlation Ridge Speed versus Angle"
    )

    ax_speed.grid(
        True,
        alpha=0.3,
    )

    ax_speed.legend(
        loc="best",
    )

    plt.tight_layout()

    speed_angle_plot = (
        OUT_DIR
        / "speed_vs_angle.png"
    )

    fig_speed.savefig(
        speed_angle_plot,
        dpi=DPI,
        bbox_inches="tight",
    )

    plt.close(
        fig_speed
    )


# ==================================================
# Plot R-squared versus angle
# ==================================================
if combined_results.size > 0:
    uncorrected_r_squared = combined_results[
        :,
        5,
    ]

    corrected_r_squared = combined_results[
        :,
        11,
    ]

    fig_r2, ax_r2 = plt.subplots(
        figsize=(10, 5)
    )

    ax_r2.plot(
        angles_processed,
        uncorrected_r_squared,
        marker="o",
        linewidth=1.5,
        label="Uncorrected",
    )

    ax_r2.plot(
        angles_processed,
        corrected_r_squared,
        marker="o",
        linewidth=1.5,
        label="Corrected",
    )

    ax_r2.set_xlim(
        KEOGRAM_ANGLES_DEG.min(),
        KEOGRAM_ANGLES_DEG.max(),
    )

    ax_r2.set_ylim(
        0.0,
        1.0,
    )

    ax_r2.set_xlabel(
        "Keogram angle (degrees)"
    )

    ax_r2.set_ylabel(
        "R²"
    )

    ax_r2.set_title(
        "Maximum-Correlation Ridge Fit R² versus Angle"
    )

    ax_r2.grid(
        True,
        alpha=0.3,
    )

    ax_r2.legend(
        loc="best",
    )

    plt.tight_layout()

    r2_angle_plot = (
        OUT_DIR
        / "r_squared_vs_angle.png"
    )

    fig_r2.savefig(
        r2_angle_plot,
        dpi=DPI,
        bbox_inches="tight",
    )

    plt.close(
        fig_r2
    )


# ==================================================
# Save processing log
# ==================================================
processing_log_file = (
    OUT_DIR
    / "processing_log.txt"
)

with open(
    processing_log_file,
    "w",
    encoding="utf-8",
) as file:
    file.write(
        "Requested angles:\n"
    )

    file.write(
        ", ".join(
            str(angle)
            for angle in KEOGRAM_ANGLES_DEG
        )
        + "\n\n"
    )

    file.write(
        f"Number requested: "
        f"{len(KEOGRAM_ANGLES_DEG)}\n"
    )

    file.write(
        f"Number successful: "
        f"{len(successful_angles)}\n"
    )

    file.write(
        f"Number failed: "
        f"{len(failed_angles)}\n\n"
    )

    file.write(
        "Successful angles:\n"
    )

    file.write(
        ", ".join(
            str(angle)
            for angle in successful_angles
        )
        + "\n\n"
    )

    file.write(
        "Failed angles:\n"
    )

    if failed_angles:
        for angle_deg, reason in failed_angles:
            file.write(
                f"{angle_deg}: {reason}\n"
            )
    else:
        file.write(
            "None\n"
        )


# ==================================================
# Print final summary
# ==================================================
print(
    "\n"
    + "=" * 70
)

print(
    "ALL-ANGLE PROCESSING COMPLETE"
)

print(
    "=" * 70
)

print(
    "Angles requested:"
)

print(
    KEOGRAM_ANGLES_DEG
)

print(
    f"\nNumber requested: "
    f"{len(KEOGRAM_ANGLES_DEG)}"
)

print(
    f"Number successful: "
    f"{len(successful_angles)}"
)

print(
    f"Number failed: "
    f"{len(failed_angles)}"
)

print(
    "\nSuccessful angles:"
)

print(
    successful_angles
)

if failed_angles:
    print(
        "\nFailed angles:"
    )

    for angle_deg, reason in failed_angles:
        print(
            f"  {angle_deg}°: {reason}"
        )

print(
    "\nCombined CSV:"
)

print(
    combined_csv
)

print(
    "\nOutput folder:"
)

print(
    OUT_DIR.resolve()
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

from scipy.ndimage import gaussian_filter
from scipy.interpolate import PchipInterpolator
from scipy.signal import find_peaks


# ==================================================
# Settings
# ==================================================
BASE_OUT_DIR = Path(
    "BLO_perturbation_keogram_outputs_new"
)

# Process every directional keogram:
# 0, 5, 10, ..., 175 degrees
KEOGRAM_ANGLES_DEG = np.arange(
    0,
    176,
    5,
    dtype=int,
)

# Time range used for correlation
CORRELATION_TIME_MIN = 0.0
CORRELATION_TIME_MAX = 70.0

# Distance range included in the correlation map
CORRELATION_DISTANCE_MIN_KM = 0.0
CORRELATION_DISTANCE_MAX_KM = 250.0

# Shift range in minutes
SHIFT_MIN_MINUTES = -60.0
SHIFT_MAX_MINUTES = 60.0

# Minimum number of overlapping time samples
MIN_VALID_CORRELATION_POINTS = 10

# Minimum maximum-correlation value used in the slope fit
MIN_CORRELATION_FOR_SLOPE = 0.50

# Distance range used in the ridge fit
SLOPE_FIT_DISTANCE_MIN_KM = 0.0
SLOPE_FIT_DISTANCE_MAX_KM = 250.0

DPI = 200

CORRELATION_CMAP = "RdBu_r"
KEOGRAM_CMAP = "RdBu_r"

# Use symmetric automatic keogram limits based on this percentile.
# Set KEOGRAM_ABS_LIMIT to a number to force fixed limits.
KEOGRAM_ABS_LIMIT = None
KEOGRAM_AUTO_PERCENTILE = 99.0

# Correlation color scale
CORRELATION_VMIN = 0.00
CORRELATION_VMAX = 1.00

# Local maxima search on the 0 km correlation-shift profile.
# The left maximum is searched at negative shifts and the right
# maximum is searched at positive shifts. The zero-shift point is
# therefore excluded automatically.
LOCAL_MAXIMUM_MIN_PROMINENCE = 0.0
LOCAL_MAXIMUM_MIN_DISTANCE_PIXELS = 1


# ==================================================
# Optional 2D smoothing
# ==================================================
APPLY_2D_SMOOTHING = True

# Keogram dimensions:
# axis 0 = time
# axis 1 = distance
SMOOTH_SIGMA_TIME_PIXELS = 1.5
SMOOTH_SIGMA_DISTANCE_PIXELS = 2.0

SMOOTH_MODE = "nearest"
SMOOTH_WEIGHT_TOLERANCE = 1e-12


# ==================================================
# Bias-correction settings
# ==================================================
APPLY_BIAS_CORRECTION = False

BIAS_TIME_POINTS_MIN = np.array(
    [
        0.0,
        150.0,
        300.0,
    ],
    dtype=float,
)

BIAS_AT_MAX_DISTANCE = np.array(
    [
        0.20,
        0.04,
        -0.01,
    ],
    dtype=float,
)

BIAS_DISTANCE_MIN_KM = 0.0
BIAS_DISTANCE_MAX_KM = 250.0

# 1.0 = linear distance dependence
# 2.0 = quadratic distance dependence
BIAS_DISTANCE_POWER = 1.0

BIAS_ZERO_OUTSIDE_DISTANCE_RANGE = True
BIAS_ZERO_OUTSIDE_TIME_RANGE = True
CLIP_NEGATIVE_TIME_CORRECTION = False


# ==================================================
# Filename-safe numeric label
# ==================================================
def safe_number_label(value):
    value_float = float(value)

    if value_float.is_integer():
        return str(
            int(value_float)
        )

    return (
        f"{value_float:g}"
        .replace("-", "m")
        .replace(".", "p")
    )


# ==================================================
# NaN-aware 2D Gaussian smoothing
# ==================================================
def smooth_keogram_2d(
    data,
    sigma_time,
    sigma_distance,
    mode="nearest",
    weight_tolerance=1e-12,
):
    """
    Apply NaN-aware Gaussian smoothing to a 2D keogram.
    """
    data = np.asarray(
        data,
        dtype=float,
    )

    if data.ndim != 2:
        raise ValueError(
            "The keogram must be two-dimensional."
        )

    if sigma_time < 0 or sigma_distance < 0:
        raise ValueError(
            "Smoothing sigma values must be nonnegative."
        )

    if sigma_time == 0 and sigma_distance == 0:
        return data.copy()

    valid_mask = np.isfinite(
        data
    )

    data_filled = np.where(
        valid_mask,
        data,
        0.0,
    )

    weights = valid_mask.astype(
        float
    )

    sigma = (
        sigma_time,
        sigma_distance,
    )

    smoothed_values = gaussian_filter(
        data_filled,
        sigma=sigma,
        mode=mode,
    )

    smoothed_weights = gaussian_filter(
        weights,
        sigma=sigma,
        mode=mode,
    )

    smoothed = np.full(
        data.shape,
        np.nan,
        dtype=float,
    )

    valid_output = (
        smoothed_weights
        > weight_tolerance
    )

    smoothed[valid_output] = (
        smoothed_values[valid_output]
        / smoothed_weights[valid_output]
    )

    return smoothed


# ==================================================
# Create 2D bias-correction keogram
# ==================================================
def create_bias_correction_keogram(
    times_min,
    distance_km,
    bias_time_points_min,
    bias_at_max_distance,
    bias_distance_min_km,
    bias_distance_max_km,
    distance_power=1.0,
    zero_outside_distance_range=True,
    zero_outside_time_range=True,
    clip_negative_time_correction=False,
):
    """
    Create the time x distance additive bias field.
    """
    times_min = np.asarray(
        times_min,
        dtype=float,
    )

    distance_km = np.asarray(
        distance_km,
        dtype=float,
    )

    bias_time_points_min = np.asarray(
        bias_time_points_min,
        dtype=float,
    )

    bias_at_max_distance = np.asarray(
        bias_at_max_distance,
        dtype=float,
    )

    if (
        len(bias_time_points_min)
        != len(bias_at_max_distance)
    ):
        raise ValueError(
            "Bias time points and bias values must have "
            "the same length."
        )

    if not np.all(
        np.diff(bias_time_points_min) > 0
    ):
        raise ValueError(
            "BIAS_TIME_POINTS_MIN must be strictly increasing."
        )

    if bias_distance_max_km <= bias_distance_min_km:
        raise ValueError(
            "The maximum bias distance must exceed "
            "the minimum bias distance."
        )

    if distance_power <= 0:
        raise ValueError(
            "BIAS_DISTANCE_POWER must be positive."
        )

    # ==================================================
    # PCHIP time dependence
    # ==================================================
    interpolator = PchipInterpolator(
        bias_time_points_min,
        bias_at_max_distance,
        extrapolate=False,
    )

    bias_vs_time = interpolator(
        times_min
    )

    if zero_outside_time_range:
        inside_time = (
            (times_min >= bias_time_points_min[0])
            & (times_min <= bias_time_points_min[-1])
        )

        bias_vs_time = np.where(
            inside_time,
            bias_vs_time,
            0.0,
        )
    else:
        bias_vs_time = np.where(
            times_min < bias_time_points_min[0],
            bias_at_max_distance[0],
            bias_vs_time,
        )

        bias_vs_time = np.where(
            times_min > bias_time_points_min[-1],
            bias_at_max_distance[-1],
            bias_vs_time,
        )

    if clip_negative_time_correction:
        bias_vs_time = np.maximum(
            bias_vs_time,
            0.0,
        )

    # ==================================================
    # Distance dependence
    # ==================================================
    normalized_distance = (
        distance_km
        - bias_distance_min_km
    ) / (
        bias_distance_max_km
        - bias_distance_min_km
    )

    if zero_outside_distance_range:
        inside_distance = (
            (distance_km >= bias_distance_min_km)
            & (distance_km <= bias_distance_max_km)
        )

        normalized_distance = np.where(
            inside_distance,
            normalized_distance,
            0.0,
        )
    else:
        normalized_distance = np.clip(
            normalized_distance,
            0.0,
            1.0,
        )

    distance_factor = (
        normalized_distance ** distance_power
    )

    correction_keogram = (
        bias_vs_time[:, np.newaxis]
        * distance_factor[np.newaxis, :]
    )

    return (
        correction_keogram,
        bias_vs_time,
        distance_factor,
    )


# ==================================================
# NaN-aware Pearson correlation
# ==================================================
def nan_pearson_correlation(
    reference,
    comparison,
    minimum_points=20,
):
    """
    Calculate Pearson correlation using finite overlapping values.
    """
    reference = np.asarray(
        reference,
        dtype=float,
    )

    comparison = np.asarray(
        comparison,
        dtype=float,
    )

    if reference.shape != comparison.shape:
        raise ValueError(
            "The two arrays must have the same shape."
        )

    valid = (
        np.isfinite(reference)
        & np.isfinite(comparison)
    )

    if np.count_nonzero(valid) < minimum_points:
        return np.nan

    reference_valid = reference[
        valid
    ]

    comparison_valid = comparison[
        valid
    ]

    reference_centered = (
        reference_valid
        - np.mean(reference_valid)
    )

    comparison_centered = (
        comparison_valid
        - np.mean(comparison_valid)
    )

    denominator = np.sqrt(
        np.sum(reference_centered ** 2)
        * np.sum(comparison_centered ** 2)
    )

    if (
        not np.isfinite(denominator)
        or denominator <= 0
    ):
        return np.nan

    correlation = (
        np.sum(
            reference_centered
            * comparison_centered
        )
        / denominator
    )

    return np.clip(
        correlation,
        -1.0,
        1.0,
    )


# ==================================================
# Correlation for one integer-pixel shift
# ==================================================
def calculate_shifted_correlation(
    reference_row,
    comparison_row,
    shift_pixels,
    minimum_points=20,
):
    """
    Shift comparison_row and correlate it with reference_row.

    Positive shift:
        comparison row moves toward later times.

    Negative shift:
        comparison row moves toward earlier times.

    No circular wrapping is used.
    """
    reference_row = np.asarray(
        reference_row,
        dtype=float,
    )

    comparison_row = np.asarray(
        comparison_row,
        dtype=float,
    )

    if reference_row.shape != comparison_row.shape:
        raise ValueError(
            "Reference and comparison rows must have "
            "the same shape."
        )

    ntime = len(
        reference_row
    )

    if abs(shift_pixels) >= ntime:
        return np.nan

    if shift_pixels > 0:
        reference_overlap = reference_row[
            shift_pixels:
        ]

        comparison_overlap = comparison_row[
            :-shift_pixels
        ]

    elif shift_pixels < 0:
        shift_abs = abs(
            shift_pixels
        )

        reference_overlap = reference_row[
            :-shift_abs
        ]

        comparison_overlap = comparison_row[
            shift_abs:
        ]

    else:
        reference_overlap = reference_row
        comparison_overlap = comparison_row

    return nan_pearson_correlation(
        reference=reference_overlap,
        comparison=comparison_overlap,
        minimum_points=minimum_points,
    )


# ==================================================
# Calculate complete correlation map
# ==================================================
def calculate_correlation_map(
    keogram,
    distance_km,
    reference_distance_km,
    selected_distance_mask,
    shifts_pixels,
    minimum_points=20,
):
    """
    Correlate every selected distance row with the reference row
    for every requested shift.
    """
    keogram = np.asarray(
        keogram,
        dtype=float,
    )

    distance_km = np.asarray(
        distance_km,
        dtype=float,
    )

    shifts_pixels = np.asarray(
        shifts_pixels,
        dtype=int,
    )

    reference_index = np.argmin(
        np.abs(
            distance_km
            - reference_distance_km
        )
    )

    selected_indices = np.where(
        selected_distance_mask
    )[0]

    selected_distances = distance_km[
        selected_indices
    ]

    reference_row = keogram[
        :,
        reference_index,
    ]

    correlation_map = np.full(
        (
            len(selected_indices),
            len(shifts_pixels),
        ),
        np.nan,
        dtype=float,
    )

    for output_row_index, keogram_row_index in enumerate(
        selected_indices
    ):
        comparison_row = keogram[
            :,
            keogram_row_index,
        ]

        for shift_index, shift_pixels in enumerate(
            shifts_pixels
        ):
            correlation_map[
                output_row_index,
                shift_index,
            ] = calculate_shifted_correlation(
                reference_row=reference_row,
                comparison_row=comparison_row,
                shift_pixels=int(shift_pixels),
                minimum_points=minimum_points,
            )

    return (
        correlation_map,
        selected_distances,
        reference_index,
    )


# ==================================================
# Maximum-correlation shift for each row
# ==================================================
def find_best_shifts(
    correlation_map,
    shifts_minutes,
):
    """
    Find the shift in minutes having the largest correlation
    for every distance row.
    """
    correlation_map = np.asarray(
        correlation_map,
        dtype=float,
    )

    shifts_minutes = np.asarray(
        shifts_minutes,
        dtype=float,
    )

    best_shift_minutes = np.full(
        correlation_map.shape[0],
        np.nan,
        dtype=float,
    )

    maximum_correlation = np.full(
        correlation_map.shape[0],
        np.nan,
        dtype=float,
    )

    for row_index in range(
        correlation_map.shape[0]
    ):
        correlation_row = correlation_map[
            row_index,
            :
        ]

        finite = np.isfinite(
            correlation_row
        )

        if not np.any(finite):
            continue

        finite_indices = np.where(
            finite
        )[0]

        local_max_index = np.argmax(
            correlation_row[finite]
        )

        full_max_index = finite_indices[
            local_max_index
        ]

        best_shift_minutes[row_index] = (
            shifts_minutes[full_max_index]
        )

        maximum_correlation[row_index] = (
            correlation_row[full_max_index]
        )

    return (
        best_shift_minutes,
        maximum_correlation,
    )


# ==================================================
# Extract the correlation-shift profile at 0 km
# ==================================================
def extract_zero_km_correlation_profile(
    correlation_map,
    selected_distances,
):
    """
    Return the correlation-versus-shift profile for the selected
    distance row nearest 0 km.
    """
    correlation_map = np.asarray(
        correlation_map,
        dtype=float,
    )

    selected_distances = np.asarray(
        selected_distances,
        dtype=float,
    )

    if correlation_map.ndim != 2:
        raise ValueError(
            "correlation_map must be two-dimensional."
        )

    if correlation_map.shape[0] != len(selected_distances):
        raise ValueError(
            "The number of correlation-map rows must equal "
            "the number of selected distances."
        )

    zero_distance_row_index = int(
        np.argmin(
            np.abs(selected_distances)
        )
    )

    zero_distance_km = float(
        selected_distances[zero_distance_row_index]
    )

    zero_km_profile = correlation_map[
        zero_distance_row_index,
        :
    ].copy()

    return (
        zero_km_profile,
        zero_distance_row_index,
        zero_distance_km,
    )


# ==================================================
# Find strongest local maximum on each side of zero
# ==================================================
def find_left_and_right_local_maxima(
    correlation_profile,
    shifts_minutes,
    minimum_prominence=0.0,
    minimum_distance_pixels=1,
):
    """
    Find the strongest finite local maximum at negative shift
    and the strongest finite local maximum at positive shift.

    The point at zero shift is not included in either search.

    If scipy.signal.find_peaks finds no strict local maximum on one
    side, the largest finite value on that side is used as a fallback.
    """
    correlation_profile = np.asarray(
        correlation_profile,
        dtype=float,
    )

    shifts_minutes = np.asarray(
        shifts_minutes,
        dtype=float,
    )

    if correlation_profile.shape != shifts_minutes.shape:
        raise ValueError(
            "correlation_profile and shifts_minutes must "
            "have the same shape."
        )

    def find_one_side(side_mask):
        side_indices = np.where(
            side_mask
            & np.isfinite(correlation_profile)
        )[0]

        if len(side_indices) == 0:
            return {
                "index": -1,
                "shift_minutes": np.nan,
                "correlation": np.nan,
                "used_fallback": False,
            }

        side_values = correlation_profile[
            side_indices
        ]

        local_indices, properties = find_peaks(
            side_values,
            prominence=minimum_prominence,
            distance=minimum_distance_pixels,
        )

        used_fallback = False

        if len(local_indices) > 0:
            local_values = side_values[
                local_indices
            ]

            strongest_local_position = local_indices[
                int(np.nanargmax(local_values))
            ]

            full_index = int(
                side_indices[strongest_local_position]
            )
        else:
            used_fallback = True

            strongest_side_position = int(
                np.nanargmax(side_values)
            )

            full_index = int(
                side_indices[strongest_side_position]
            )

        return {
            "index": full_index,
            "shift_minutes": float(
                shifts_minutes[full_index]
            ),
            "correlation": float(
                correlation_profile[full_index]
            ),
            "used_fallback": used_fallback,
        }

    left_result = find_one_side(
        shifts_minutes < 0.0
    )

    right_result = find_one_side(
        shifts_minutes > 0.0
    )

    return {
        "left": left_result,
        "right": right_result,
    }


# ==================================================
# Fit maximum-correlation ridge
# ==================================================
def fit_maximum_correlation_slope(
    distances_km,
    best_shift_minutes,
    maximum_correlation,
    minimum_correlation=0.5,
    distance_min_km=None,
    distance_max_km=None,
):
    """
    Fit:

        shift_minutes = slope * distance_km + intercept
    """
    distances_km = np.asarray(
        distances_km,
        dtype=float,
    )

    best_shift_minutes = np.asarray(
        best_shift_minutes,
        dtype=float,
    )

    maximum_correlation = np.asarray(
        maximum_correlation,
        dtype=float,
    )

    fit_mask = (
        np.isfinite(distances_km)
        & np.isfinite(best_shift_minutes)
        & np.isfinite(maximum_correlation)
        & (
            maximum_correlation
            >= minimum_correlation
        )
    )

    if distance_min_km is not None:
        fit_mask &= (
            distances_km
            >= distance_min_km
        )

    if distance_max_km is not None:
        fit_mask &= (
            distances_km
            <= distance_max_km
        )

    number_of_points = np.count_nonzero(
        fit_mask
    )

    if number_of_points < 2:
        return {
            "slope_min_per_km": np.nan,
            "intercept_minutes": np.nan,
            "speed_km_per_min": np.nan,
            "speed_m_per_s": np.nan,
            "r_squared": np.nan,
            "number_of_fit_points": number_of_points,
            "fit_mask": fit_mask,
        }

    fit_distances = distances_km[
        fit_mask
    ]

    fit_shifts = best_shift_minutes[
        fit_mask
    ]

    slope_min_per_km, intercept_minutes = np.polyfit(
        fit_distances,
        fit_shifts,
        deg=1,
    )

    predicted_shifts = (
        slope_min_per_km
        * fit_distances
        + intercept_minutes
    )

    residual_sum_squares = np.sum(
        (
            fit_shifts
            - predicted_shifts
        ) ** 2
    )

    total_sum_squares = np.sum(
        (
            fit_shifts
            - np.mean(fit_shifts)
        ) ** 2
    )

    if total_sum_squares > 0:
        r_squared = (
            1.0
            - residual_sum_squares
            / total_sum_squares
        )
    else:
        r_squared = np.nan

    if np.isclose(
        slope_min_per_km,
        0.0,
    ):
        speed_km_per_min = np.nan
        speed_m_per_s = np.nan
    else:
        speed_km_per_min = (
            1.0
            / slope_min_per_km
        )

        speed_m_per_s = (
            speed_km_per_min
            * 1000.0
            / 60.0
        )

    return {
        "slope_min_per_km": slope_min_per_km,
        "intercept_minutes": intercept_minutes,
        "speed_km_per_min": speed_km_per_min,
        "speed_m_per_s": speed_m_per_s,
        "r_squared": r_squared,
        "number_of_fit_points": number_of_points,
        "fit_mask": fit_mask,
    }


# ==================================================
# Keogram color limit
# ==================================================
def determine_keogram_abs_limit(
    full_keogram,
    limited_keogram,
):
    """
    Return one symmetric absolute color limit shared by both
    keogram panels.
    """
    if KEOGRAM_ABS_LIMIT is not None:
        abs_limit = float(
            KEOGRAM_ABS_LIMIT
        )

        if (
            not np.isfinite(abs_limit)
            or abs_limit <= 0
        ):
            raise ValueError(
                "KEOGRAM_ABS_LIMIT must be positive and finite."
            )

        return abs_limit

    finite_values = []

    for data in [
        full_keogram,
        limited_keogram,
    ]:
        data = np.asarray(
            data,
            dtype=float,
        )

        finite = np.abs(
            data[np.isfinite(data)]
        )

        if finite.size > 0:
            finite_values.append(
                finite
            )

    if len(finite_values) == 0:
        return 1.0

    all_finite = np.concatenate(
        finite_values
    )

    abs_limit = np.nanpercentile(
        all_finite,
        KEOGRAM_AUTO_PERCENTILE,
    )

    if (
        not np.isfinite(abs_limit)
        or abs_limit <= 0
    ):
        abs_limit = np.nanmax(
            all_finite
        )

    if (
        not np.isfinite(abs_limit)
        or abs_limit <= 0
    ):
        abs_limit = 1.0

    return float(
        abs_limit
    )


# ==================================================
# Plot keograms, correlation map, ridge, and profile
# ==================================================
def plot_correlation_map(
    angle_deg,
    full_keogram,
    full_times_min,
    full_distance_km,
    limited_keogram,
    limited_times_min,
    limited_distance_km,
    correlation_map,
    shifts_minutes,
    selected_distances,
    best_shift_minutes,
    maximum_correlation,
    slope_results,
    zero_km_correlation_profile,
    zero_km_distance,
    zero_km_local_maxima,
    dataset_name,
    output_file,
):
    """
    Plot:
      1. full-range original keogram;
      2. limited-range smoothed/corrected keogram;
      3. correlation versus time shift and distance;
      4. correlation versus shift for the 0 km row.
    """
    fig, (
        ax_full,
        ax_limited,
        ax_map,
        ax_profile,
    ) = plt.subplots(
        nrows=4,
        ncols=1,
        figsize=(11, 16),
        gridspec_kw={
            "height_ratios": [
                1.55,
                1.55,
                2.5,
                1.15,
            ],
        },
    )

    keogram_abs_limit = determine_keogram_abs_limit(
        full_keogram=full_keogram,
        limited_keogram=limited_keogram,
    )

    # ==================================================
    # Full-range original keogram
    # ==================================================
    full_mesh = ax_full.pcolormesh(
        full_times_min,
        full_distance_km,
        full_keogram.T,
        shading="auto",
        cmap=KEOGRAM_CMAP,
        vmin=-keogram_abs_limit,
        vmax=keogram_abs_limit,
    )

    full_cbar = fig.colorbar(
        full_mesh,
        ax=ax_full,
        fraction=0.025,
        pad=0.02,
    )

    full_cbar.set_label(
        "Keogram value"
    )

    ax_full.axhline(
        0.0,
        color="black",
        linestyle="--",
        linewidth=1.0,
    )

    ax_full.set_xlim(
        np.nanmin(full_times_min),
        np.nanmax(full_times_min),
    )

    ax_full.set_ylim(
        np.nanmin(full_distance_km),
        np.nanmax(full_distance_km),
    )

    ax_full.set_title(
        f"{angle_deg}° Full-Range Original Keogram"
    )

    ax_full.set_xlabel(
        "Time (min)"
    )

    ax_full.set_ylabel(
        "Distance along slice (km)"
    )

    # ==================================================
    # Limited-range processed keogram
    # ==================================================
    limited_mesh = ax_limited.pcolormesh(
        limited_times_min,
        limited_distance_km,
        limited_keogram.T,
        shading="auto",
        cmap=KEOGRAM_CMAP,
        #vmin=-keogram_abs_limit,
        #vmax=keogram_abs_limit,
    )

    limited_cbar = fig.colorbar(
        limited_mesh,
        ax=ax_limited,
        fraction=0.025,
        pad=0.02,
    )

    limited_cbar.set_label(
        "Keogram value"
    )

    ax_limited.axhline(
        0.0,
        color="black",
        linestyle="--",
        linewidth=1.0,
    )

    ax_limited.set_xlim(
        CORRELATION_TIME_MIN,
        CORRELATION_TIME_MAX,
    )

    ax_limited.set_ylim(
        CORRELATION_DISTANCE_MIN_KM,
        CORRELATION_DISTANCE_MAX_KM,
    )

    ax_limited.set_title(
        "Limited-Range Processed Keogram\n"
        f"{dataset_name}"
    )

    ax_limited.set_xlabel(
        "Time (min)"
    )

    ax_limited.set_ylabel(
        "Distance along slice (km)"
    )

    # ==================================================
    # Correlation map
    # ==================================================
    if len(shifts_minutes) > 1:
        shift_spacing = np.median(
            np.diff(shifts_minutes)
        )
    else:
        shift_spacing = 1.0

    if len(selected_distances) > 1:
        distance_spacing = np.median(
            np.diff(selected_distances)
        )
    else:
        distance_spacing = 1.0

    im = ax_map.imshow(
        correlation_map,
        origin="lower",
        aspect="auto",
        interpolation="nearest",
        cmap=CORRELATION_CMAP,
        vmin=CORRELATION_VMIN,
        vmax=CORRELATION_VMAX,
        extent=[
            shifts_minutes[0]
            - shift_spacing / 2.0,
            shifts_minutes[-1]
            + shift_spacing / 2.0,
            selected_distances[0]
            - distance_spacing / 2.0,
            selected_distances[-1]
            + distance_spacing / 2.0,
        ],
    )

    cbar = fig.colorbar(
        im,
        ax=ax_map,
        fraction=0.025,
        pad=0.02,
    )

    cbar.set_label(
        "Pearson correlation with 0 km row"
    )

    ax_map.axvline(
        0.0,
        color="black",
        linestyle="--",
        linewidth=1.2,
        label="Zero shift",
    )

    ax_map.axhline(
        zero_km_distance,
        color="white",
        linestyle=":",
        linewidth=1.5,
        label=f"Profile row ({zero_km_distance:g} km)",
    )

    ridge_mask = (
        np.isfinite(best_shift_minutes)
        & np.isfinite(maximum_correlation)
        & (
            maximum_correlation
            >= MIN_CORRELATION_FOR_SLOPE
        )
    )

    if np.any(ridge_mask):
        ax_map.plot(
            best_shift_minutes[ridge_mask],
            selected_distances[ridge_mask],
            color="black",
            linewidth=1.2,
            label="Maximum correlation",
        )

        ax_map.scatter(
            best_shift_minutes[ridge_mask],
            selected_distances[ridge_mask],
            color="black",
            s=8,
        )

    slope = slope_results[
        "slope_min_per_km"
    ]

    intercept = slope_results[
        "intercept_minutes"
    ]

    r_squared = slope_results[
        "r_squared"
    ]

    speed_m_per_s = slope_results[
        "speed_m_per_s"
    ]

    if np.isfinite(slope):
        fit_distances = np.linspace(
            SLOPE_FIT_DISTANCE_MIN_KM,
            SLOPE_FIT_DISTANCE_MAX_KM,
            300,
        )

        fit_shifts = (
            slope
            * fit_distances
            + intercept
        )

        ax_map.plot(
            fit_shifts,
            fit_distances,
            color="lime",
            linestyle="--",
            linewidth=2.0,
            label="Linear ridge fit",
        )

        fit_text = (
            f"Shift slope = {slope:.5f} min/km\n"
            f"Equivalent speed = {speed_m_per_s:.2f} m/s\n"
            f"$R^2$ = {r_squared:.3f}"
        )

        ax_map.text(
            0.02,
            0.03,
            fit_text,
            transform=ax_map.transAxes,
            ha="left",
            va="bottom",
            fontsize=9,
            bbox={
                "facecolor": "white",
                "edgecolor": "black",
                "alpha": 0.85,
            },
        )

    ax_map.set_xlim(
        SHIFT_MIN_MINUTES,
        SHIFT_MAX_MINUTES,
    )

    ax_map.set_ylim(
        CORRELATION_DISTANCE_MIN_KM,
        CORRELATION_DISTANCE_MAX_KM,
    )

    ax_map.set_title(
        f"{angle_deg}° Shifted Row Correlation\n"
        f"{dataset_name}"
    )

    ax_map.set_xlabel(
        "Shift applied to row (min)"
    )

    ax_map.set_ylabel(
        "Distance along slice (km)"
    )

    ax_map.legend(
        loc="upper right",
        fontsize=8,
    )

    # ==================================================
    # 0 km correlation-shift line
    # ==================================================
    ax_profile.plot(
        shifts_minutes,
        zero_km_correlation_profile,
        color="black",
        linewidth=1.4,
        label=f"{zero_km_distance:g} km correlation profile",
    )

    ax_profile.axvline(
        0.0,
        color="gray",
        linestyle="--",
        linewidth=1.0,
    )

    left_peak = zero_km_local_maxima[
        "left"
    ]

    right_peak = zero_km_local_maxima[
        "right"
    ]

    if np.isfinite(
        left_peak["shift_minutes"]
    ):
        ax_profile.scatter(
            left_peak["shift_minutes"],
            left_peak["correlation"],
            s=55,
            marker="o",
            zorder=5,
            label=(
                "Left local max: "
                f"{left_peak['shift_minutes']:.2f} min"
            ),
        )

        ax_profile.axvline(
            left_peak["shift_minutes"],
            linestyle=":",
            linewidth=1.2,
        )

        ax_profile.annotate(
            (
                f"{left_peak['shift_minutes']:.2f} min\n"
                f"r={left_peak['correlation']:.3f}"
            ),
            xy=(
                left_peak["shift_minutes"],
                left_peak["correlation"],
            ),
            xytext=(8, -25),
            textcoords="offset points",
            ha="left",
            va="bottom",
            fontsize=8,
            arrowprops={
                "arrowstyle": "->",
                "linewidth": 0.8,
            },
        )

    if np.isfinite(
        right_peak["shift_minutes"]
    ):
        ax_profile.scatter(
            right_peak["shift_minutes"],
            right_peak["correlation"],
            s=55,
            marker="s",
            zorder=5,
            label=(
                "Right local max: "
                f"{right_peak['shift_minutes']:.2f} min"
            ),
        )

        ax_profile.axvline(
            right_peak["shift_minutes"],
            linestyle=":",
            linewidth=1.2,
        )

        ax_profile.annotate(
            (
                f"{right_peak['shift_minutes']:.2f} min\n"
                f"r={right_peak['correlation']:.3f}"
            ),
            xy=(
                right_peak["shift_minutes"],
                right_peak["correlation"],
            ),
            xytext=(-8, -25),
            textcoords="offset points",
            ha="right",
            va="bottom",
            fontsize=8,
            arrowprops={
                "arrowstyle": "->",
                "linewidth": 0.8,
            },
        )

    ax_profile.set_xlim(
        SHIFT_MIN_MINUTES,
        SHIFT_MAX_MINUTES,
    )

    ax_profile.set_xlabel(
        "Shift applied to row (min)"
    )

    ax_profile.set_ylabel(
        "Correlation"
    )

    ax_profile.set_title(
        f"Correlation-Shift Line at {zero_km_distance:g} km"
    )

    ax_profile.grid(
        True,
        alpha=0.3,
    )

    ax_profile.legend(
        loc="best",
        fontsize=8,
    )

    plt.tight_layout()

    fig.savefig(
        output_file,
        dpi=DPI,
        bbox_inches="tight",
    )

    plt.close(fig)


# ==================================================
# Print slope results
# ==================================================
def print_slope_results(
    angle_deg,
    label,
    results,
):
    print(
        f"\n{angle_deg}° — {label}:"
    )

    print(
        f"  Shift slope: "
        f"{results['slope_min_per_km']:.8f} min/km"
    )

    print(
        f"  Intercept: "
        f"{results['intercept_minutes']:.6f} min"
    )

    print(
        f"  Equivalent speed: "
        f"{results['speed_m_per_s']:.4f} m/s"
    )

    print(
        f"  R²: "
        f"{results['r_squared']:.6f}"
    )

    print(
        f"  Points used: "
        f"{results['number_of_fit_points']}"
    )


# ==================================================
# Output paths
# ==================================================
sigma_t_label = safe_number_label(
    SMOOTH_SIGMA_TIME_PIXELS
)

sigma_d_label = safe_number_label(
    SMOOTH_SIGMA_DISTANCE_PIXELS
)

distance_folder = (
    f"r_"
    f"{int(CORRELATION_DISTANCE_MIN_KM)}_"
    f"{int(CORRELATION_DISTANCE_MAX_KM)}"
)

shift_folder = (
    f"shift_min_"
    f"{safe_number_label(SHIFT_MIN_MINUTES)}_"
    f"{safe_number_label(SHIFT_MAX_MINUTES)}"
)

if APPLY_2D_SMOOTHING:
    smoothing_folder = (
        f"sig_"
        f"{sigma_t_label}_"
        f"{sigma_d_label}"
    )
else:
    smoothing_folder = "raw"

OUT_DIR = (
    BASE_OUT_DIR
    / distance_folder
    / smoothing_folder
    / "pchip_bias"
    / "row_correlation_minutes_all_angles"
    / shift_folder
)

UNCORRECTED_OUT_DIR = (
    OUT_DIR
    / "uncorrected"
)

CORRECTED_OUT_DIR = (
    OUT_DIR
    / "corrected"
)

SLOPE_REPORT_OUT_DIR = (
    OUT_DIR
    / "slope_reports"
)

for directory in [
    OUT_DIR,
    UNCORRECTED_OUT_DIR,
    CORRECTED_OUT_DIR,
    SLOPE_REPORT_OUT_DIR,
]:
    directory.mkdir(
        parents=True,
        exist_ok=True,
    )


# ==================================================
# Read common time array
# ==================================================
TIMES_FILE = (
    BASE_OUT_DIR
    / "times_s.npy"
)

times_s = np.load(
    TIMES_FILE
).astype(float)

times_min = times_s / 60.0

if len(times_min) < 2:
    raise ValueError(
        "At least two time samples are required."
    )


# ==================================================
# Determine minutes per time pixel
# ==================================================
time_differences_min = np.diff(
    times_min
)

finite_time_differences = time_differences_min[
    np.isfinite(time_differences_min)
]

if len(finite_time_differences) == 0:
    raise ValueError(
        "Could not determine the time spacing."
    )

MINUTES_PER_PIXEL = np.median(
    finite_time_differences
)

if (
    not np.isfinite(MINUTES_PER_PIXEL)
    or MINUTES_PER_PIXEL <= 0
):
    raise ValueError(
        "The calculated minutes-per-pixel value is invalid."
    )

maximum_time_spacing_deviation = np.max(
    np.abs(
        finite_time_differences
        - MINUTES_PER_PIXEL
    )
)

print(
    "Angles to process:"
)

print(
    KEOGRAM_ANGLES_DEG
)

print(
    f"Number of angles: "
    f"{len(KEOGRAM_ANGLES_DEG)}"
)

print(
    "\nTime sampling:"
)

print(
    f"One time pixel = "
    f"{MINUTES_PER_PIXEL:.6f} minutes"
)

print(
    f"One time pixel = "
    f"{MINUTES_PER_PIXEL * 60.0:.3f} seconds"
)

print(
    f"Maximum spacing deviation = "
    f"{maximum_time_spacing_deviation:.6e} minutes"
)


# ==================================================
# Convert minute shifts to integer pixel shifts
# ==================================================
shift_min_pixels = int(
    np.ceil(
        SHIFT_MIN_MINUTES
        / MINUTES_PER_PIXEL
    )
)

shift_max_pixels = int(
    np.floor(
        SHIFT_MAX_MINUTES
        / MINUTES_PER_PIXEL
    )
)

shifts_pixels = np.arange(
    shift_min_pixels,
    shift_max_pixels + 1,
    dtype=int,
)

shifts_minutes = (
    shifts_pixels
    * MINUTES_PER_PIXEL
)

print(
    "\nShift range:"
)

print(
    f"Requested: "
    f"{SHIFT_MIN_MINUTES:.2f} to "
    f"{SHIFT_MAX_MINUTES:.2f} min"
)

print(
    f"Actual: "
    f"{shifts_minutes[0]:.2f} to "
    f"{shifts_minutes[-1]:.2f} min"
)

print(
    f"Pixel shifts: "
    f"{shifts_pixels[0]} to "
    f"{shifts_pixels[-1]}"
)


# ==================================================
# Select common time range
# ==================================================
time_mask = (
    (times_min >= CORRELATION_TIME_MIN)
    & (times_min <= CORRELATION_TIME_MAX)
)

if np.count_nonzero(time_mask) < MIN_VALID_CORRELATION_POINTS:
    raise ValueError(
        "The selected time interval has too few samples."
    )

times_correlation_min = times_min[
    time_mask
]


# ==================================================
# Save common arrays
# ==================================================
np.save(
    OUT_DIR / "angles_deg.npy",
    KEOGRAM_ANGLES_DEG,
)

np.save(
    OUT_DIR / "shifts_minutes.npy",
    shifts_minutes,
)

np.save(
    OUT_DIR / "shifts_pixels.npy",
    shifts_pixels,
)

np.save(
    OUT_DIR / "correlation_times_min.npy",
    times_correlation_min,
)


# ==================================================
# Storage for combined angle results
# ==================================================
combined_results = []
combined_zero_km_peak_results = []

successful_angles = []
failed_angles = []


# ==================================================
# Process every angle
# ==================================================
for angle_number, angle_deg in enumerate(
    KEOGRAM_ANGLES_DEG,
    start=1,
):
    print(
        "\n"
        + "=" * 70
    )

    print(
        f"Processing angle "
        f"{angle_deg}° "
        f"({angle_number}/{len(KEOGRAM_ANGLES_DEG)})"
    )

    print(
        "=" * 70
    )

    KEOGRAM_FILE = (
        BASE_OUT_DIR
        / f"BLO_keogram_angle_{angle_deg:03d}.npy"
    )

    if not KEOGRAM_FILE.exists():
        print(
            f"WARNING: File not found: "
            f"{KEOGRAM_FILE}"
        )

        failed_angles.append(
            (
                int(angle_deg),
                "file not found",
            )
        )

        continue

    try:
        # ==================================================
        # Read keogram
        # ==================================================
        keogram_original = np.load(
            KEOGRAM_FILE
        ).astype(float)

        if keogram_original.ndim != 2:
            raise ValueError(
                f"Expected a 2D keogram, found "
                f"{keogram_original.shape}."
            )

        nt, nspace = keogram_original.shape

        if len(times_min) != nt:
            raise ValueError(
                f"Keogram has {nt} time samples, but "
                f"times_s.npy has {len(times_min)}."
            )

        # ==================================================
        # Construct distance axis
        # ==================================================
        distance_km = np.arange(
            -(nspace // 2),
            nspace // 2 + 1,
            dtype=float,
        )

        if len(distance_km) != nspace:
            distance_km = (
                np.arange(
                    nspace,
                    dtype=float,
                )
                - nspace // 2
            )

        available_distance_min = np.nanmin(
            distance_km
        )

        available_distance_max = np.nanmax(
            distance_km
        )

        if (
            CORRELATION_DISTANCE_MIN_KM
            < available_distance_min
            or CORRELATION_DISTANCE_MAX_KM
            > available_distance_max
        ):
            raise ValueError(
                f"Requested distance range "
                f"{CORRELATION_DISTANCE_MIN_KM} to "
                f"{CORRELATION_DISTANCE_MAX_KM} km "
                f"is outside available range "
                f"{available_distance_min} to "
                f"{available_distance_max} km."
            )

        # ==================================================
        # Select time interval
        # ==================================================
        keogram_time_selected = keogram_original[
            time_mask,
            :
        ]

        # ==================================================
        # Optional smoothing
        # ==================================================
        if APPLY_2D_SMOOTHING:
            keogram_uncorrected = smooth_keogram_2d(
                data=keogram_time_selected,
                sigma_time=SMOOTH_SIGMA_TIME_PIXELS,
                sigma_distance=SMOOTH_SIGMA_DISTANCE_PIXELS,
                mode=SMOOTH_MODE,
                weight_tolerance=SMOOTH_WEIGHT_TOLERANCE,
            )

            uncorrected_dataset_name = (
                f"Uncorrected, smoothed "
                f"σt={SMOOTH_SIGMA_TIME_PIXELS:g}, "
                f"σd={SMOOTH_SIGMA_DISTANCE_PIXELS:g}"
            )
        else:
            keogram_uncorrected = (
                keogram_time_selected.copy()
            )

            uncorrected_dataset_name = (
                "Uncorrected, unsmoothed"
            )

        # ==================================================
        # Create bias correction
        # ==================================================
        (
            bias_correction_keogram,
            bias_vs_time,
            bias_distance_factor,
        ) = create_bias_correction_keogram(
            times_min=times_correlation_min,
            distance_km=distance_km,
            bias_time_points_min=BIAS_TIME_POINTS_MIN,
            bias_at_max_distance=BIAS_AT_MAX_DISTANCE,
            bias_distance_min_km=BIAS_DISTANCE_MIN_KM,
            bias_distance_max_km=BIAS_DISTANCE_MAX_KM,
            distance_power=BIAS_DISTANCE_POWER,
            zero_outside_distance_range=(
                BIAS_ZERO_OUTSIDE_DISTANCE_RANGE
            ),
            zero_outside_time_range=(
                BIAS_ZERO_OUTSIDE_TIME_RANGE
            ),
            clip_negative_time_correction=(
                CLIP_NEGATIVE_TIME_CORRECTION
            ),
        )

        # ==================================================
        # Apply bias correction
        # ==================================================
        if APPLY_BIAS_CORRECTION:
            keogram_corrected = (
                keogram_uncorrected
                + bias_correction_keogram
            )

            corrected_dataset_name = (
                "Bias corrected: PCHIP time dependence, "
                f"distance power={BIAS_DISTANCE_POWER:g}"
            )
        else:
            keogram_corrected = (
                keogram_uncorrected.copy()
            )

            corrected_dataset_name = (
                "Bias correction disabled"
            )

        keogram_corrected[
            ~np.isfinite(keogram_uncorrected)
        ] = np.nan

        # ==================================================
        # Select distance rows
        # ==================================================
        selected_distance_mask = (
            (
                distance_km
                >= CORRELATION_DISTANCE_MIN_KM
            )
            & (
                distance_km
                <= CORRELATION_DISTANCE_MAX_KM
            )
        )

        selected_distances = distance_km[
            selected_distance_mask
        ]

        if len(selected_distances) == 0:
            raise ValueError(
                "No rows lie inside the selected distance range."
            )

        # ==================================================
        # Per-angle folders
        # ==================================================
        angle_uncorrected_dir = (
            UNCORRECTED_OUT_DIR
            / f"angle_{angle_deg:03d}"
        )

        angle_corrected_dir = (
            CORRECTED_OUT_DIR
            / f"angle_{angle_deg:03d}"
        )

        angle_uncorrected_dir.mkdir(
            parents=True,
            exist_ok=True,
        )

        angle_corrected_dir.mkdir(
            parents=True,
            exist_ok=True,
        )

        # ==================================================
        # Uncorrected map
        # ==================================================
        (
            correlation_map_uncorrected,
            selected_distances_uncorrected,
            reference_index_uncorrected,
        ) = calculate_correlation_map(
            keogram=keogram_uncorrected,
            distance_km=distance_km,
            reference_distance_km=0.0,
            selected_distance_mask=selected_distance_mask,
            shifts_pixels=shifts_pixels,
            minimum_points=MIN_VALID_CORRELATION_POINTS,
        )

        (
            best_shift_uncorrected_minutes,
            maximum_correlation_uncorrected,
        ) = find_best_shifts(
            correlation_map=correlation_map_uncorrected,
            shifts_minutes=shifts_minutes,
        )

        uncorrected_slope_results = (
            fit_maximum_correlation_slope(
                distances_km=(
                    selected_distances_uncorrected
                ),
                best_shift_minutes=(
                    best_shift_uncorrected_minutes
                ),
                maximum_correlation=(
                    maximum_correlation_uncorrected
                ),
                minimum_correlation=(
                    MIN_CORRELATION_FOR_SLOPE
                ),
                distance_min_km=(
                    SLOPE_FIT_DISTANCE_MIN_KM
                ),
                distance_max_km=(
                    SLOPE_FIT_DISTANCE_MAX_KM
                ),
            )
        )

        # ==================================================
        # Corrected map
        # ==================================================
        (
            correlation_map_corrected,
            selected_distances_corrected,
            reference_index_corrected,
        ) = calculate_correlation_map(
            keogram=keogram_corrected,
            distance_km=distance_km,
            reference_distance_km=0.0,
            selected_distance_mask=selected_distance_mask,
            shifts_pixels=shifts_pixels,
            minimum_points=MIN_VALID_CORRELATION_POINTS,
        )

        (
            best_shift_corrected_minutes,
            maximum_correlation_corrected,
        ) = find_best_shifts(
            correlation_map=correlation_map_corrected,
            shifts_minutes=shifts_minutes,
        )

        corrected_slope_results = (
            fit_maximum_correlation_slope(
                distances_km=(
                    selected_distances_corrected
                ),
                best_shift_minutes=(
                    best_shift_corrected_minutes
                ),
                maximum_correlation=(
                    maximum_correlation_corrected
                ),
                minimum_correlation=(
                    MIN_CORRELATION_FOR_SLOPE
                ),
                distance_min_km=(
                    SLOPE_FIT_DISTANCE_MIN_KM
                ),
                distance_max_km=(
                    SLOPE_FIT_DISTANCE_MAX_KM
                ),
            )
        )

        # ==================================================
        # Extract 0 km correlation-shift profiles
        # ==================================================
        (
            zero_km_profile_uncorrected,
            zero_km_row_uncorrected,
            zero_km_distance_uncorrected,
        ) = extract_zero_km_correlation_profile(
            correlation_map=correlation_map_uncorrected,
            selected_distances=selected_distances_uncorrected,
        )

        (
            zero_km_profile_corrected,
            zero_km_row_corrected,
            zero_km_distance_corrected,
        ) = extract_zero_km_correlation_profile(
            correlation_map=correlation_map_corrected,
            selected_distances=selected_distances_corrected,
        )

        zero_km_peaks_uncorrected = (
            find_left_and_right_local_maxima(
                correlation_profile=zero_km_profile_uncorrected,
                shifts_minutes=shifts_minutes,
                minimum_prominence=(
                    LOCAL_MAXIMUM_MIN_PROMINENCE
                ),
                minimum_distance_pixels=(
                    LOCAL_MAXIMUM_MIN_DISTANCE_PIXELS
                ),
            )
        )

        zero_km_peaks_corrected = (
            find_left_and_right_local_maxima(
                correlation_profile=zero_km_profile_corrected,
                shifts_minutes=shifts_minutes,
                minimum_prominence=(
                    LOCAL_MAXIMUM_MIN_PROMINENCE
                ),
                minimum_distance_pixels=(
                    LOCAL_MAXIMUM_MIN_DISTANCE_PIXELS
                ),
            )
        )

        # ==================================================
        # Plot maps
        # ==================================================
        uncorrected_plot_file = (
            angle_uncorrected_dir
            / (
                f"correlation_minutes_"
                f"{angle_deg:03d}_uncorrected.png"
            )
        )

        corrected_plot_file = (
            angle_corrected_dir
            / (
                f"correlation_minutes_"
                f"{angle_deg:03d}_corrected.png"
            )
        )

        plot_correlation_map(
            angle_deg=angle_deg,
            full_keogram=keogram_original,
            full_times_min=times_min,
            full_distance_km=distance_km,
            limited_keogram=keogram_uncorrected[
                :,
                selected_distance_mask,
            ],
            limited_times_min=times_correlation_min,
            limited_distance_km=selected_distances_uncorrected,
            correlation_map=(
                correlation_map_uncorrected
            ),
            shifts_minutes=shifts_minutes,
            selected_distances=(
                selected_distances_uncorrected
            ),
            best_shift_minutes=(
                best_shift_uncorrected_minutes
            ),
            maximum_correlation=(
                maximum_correlation_uncorrected
            ),
            slope_results=(
                uncorrected_slope_results
            ),
            zero_km_correlation_profile=(
                zero_km_profile_uncorrected
            ),
            zero_km_distance=(
                zero_km_distance_uncorrected
            ),
            zero_km_local_maxima=(
                zero_km_peaks_uncorrected
            ),
            dataset_name=uncorrected_dataset_name,
            output_file=uncorrected_plot_file,
        )

        plot_correlation_map(
            angle_deg=angle_deg,
            full_keogram=keogram_original,
            full_times_min=times_min,
            full_distance_km=distance_km,
            limited_keogram=keogram_corrected[
                :,
                selected_distance_mask,
            ],
            limited_times_min=times_correlation_min,
            limited_distance_km=selected_distances_corrected,
            correlation_map=(
                correlation_map_corrected
            ),
            shifts_minutes=shifts_minutes,
            selected_distances=(
                selected_distances_corrected
            ),
            best_shift_minutes=(
                best_shift_corrected_minutes
            ),
            maximum_correlation=(
                maximum_correlation_corrected
            ),
            slope_results=(
                corrected_slope_results
            ),
            zero_km_correlation_profile=(
                zero_km_profile_corrected
            ),
            zero_km_distance=(
                zero_km_distance_corrected
            ),
            zero_km_local_maxima=(
                zero_km_peaks_corrected
            ),
            dataset_name=corrected_dataset_name,
            output_file=corrected_plot_file,
        )

        # ==================================================
        # Save arrays
        # ==================================================
        np.save(
            angle_uncorrected_dir
            / (
                f"correlation_"
                f"{angle_deg:03d}_uncorrected.npy"
            ),
            correlation_map_uncorrected,
        )

        np.save(
            angle_uncorrected_dir
            / (
                f"best_shift_minutes_"
                f"{angle_deg:03d}_uncorrected.npy"
            ),
            best_shift_uncorrected_minutes,
        )

        np.save(
            angle_uncorrected_dir
            / (
                f"maximum_correlation_"
                f"{angle_deg:03d}_uncorrected.npy"
            ),
            maximum_correlation_uncorrected,
        )

        np.save(
            angle_corrected_dir
            / (
                f"correlation_"
                f"{angle_deg:03d}_corrected.npy"
            ),
            correlation_map_corrected,
        )

        np.save(
            angle_corrected_dir
            / (
                f"best_shift_minutes_"
                f"{angle_deg:03d}_corrected.npy"
            ),
            best_shift_corrected_minutes,
        )

        np.save(
            angle_corrected_dir
            / (
                f"maximum_correlation_"
                f"{angle_deg:03d}_corrected.npy"
            ),
            maximum_correlation_corrected,
        )

        np.save(
            angle_uncorrected_dir
            / "distances_km.npy",
            selected_distances_uncorrected,
        )

        np.save(
            angle_corrected_dir
            / "distances_km.npy",
            selected_distances_corrected,
        )

        np.save(
            angle_uncorrected_dir
            / "zero_km_correlation_profile.npy",
            zero_km_profile_uncorrected,
        )

        np.save(
            angle_corrected_dir
            / "zero_km_correlation_profile.npy",
            zero_km_profile_corrected,
        )

        zero_km_profile_uncorrected_csv = np.column_stack(
            [
                shifts_minutes,
                zero_km_profile_uncorrected,
            ]
        )

        zero_km_profile_corrected_csv = np.column_stack(
            [
                shifts_minutes,
                zero_km_profile_corrected,
            ]
        )

        np.savetxt(
            angle_uncorrected_dir
            / "zero_km_correlation_profile.csv",
            zero_km_profile_uncorrected_csv,
            delimiter=",",
            header="shift_minutes,correlation",
            comments="",
        )

        np.savetxt(
            angle_corrected_dir
            / "zero_km_correlation_profile.csv",
            zero_km_profile_corrected_csv,
            delimiter=",",
            header="shift_minutes,correlation",
            comments="",
        )

        # ==================================================
        # Save ridge CSV files
        # ==================================================
        uncorrected_summary = np.column_stack(
            [
                selected_distances_uncorrected,
                best_shift_uncorrected_minutes,
                maximum_correlation_uncorrected,
            ]
        )

        corrected_summary = np.column_stack(
            [
                selected_distances_corrected,
                best_shift_corrected_minutes,
                maximum_correlation_corrected,
            ]
        )

        uncorrected_csv = (
            angle_uncorrected_dir
            / (
                f"ridge_"
                f"{angle_deg:03d}_uncorrected.csv"
            )
        )

        corrected_csv = (
            angle_corrected_dir
            / (
                f"ridge_"
                f"{angle_deg:03d}_corrected.csv"
            )
        )

        np.savetxt(
            uncorrected_csv,
            uncorrected_summary,
            delimiter=",",
            header=(
                "distance_km,"
                "best_shift_minutes,"
                "maximum_correlation"
            ),
            comments="",
        )

        np.savetxt(
            corrected_csv,
            corrected_summary,
            delimiter=",",
            header=(
                "distance_km,"
                "best_shift_minutes,"
                "maximum_correlation"
            ),
            comments="",
        )

        # ==================================================
        # Save individual slope report
        # ==================================================
        slope_output_file = (
            SLOPE_REPORT_OUT_DIR
            / (
                f"correlation_slope_"
                f"{angle_deg:03d}.txt"
            )
        )

        with open(
            slope_output_file,
            "w",
            encoding="utf-8",
        ) as file:
            file.write(
                f"Angle: {angle_deg} degrees\n"
            )

            file.write(
                f"Minutes per pixel: "
                f"{MINUTES_PER_PIXEL:.10f}\n\n"
            )

            file.write(
                "UNCORRECTED\n"
            )

            for key, value in (
                uncorrected_slope_results.items()
            ):
                if key != "fit_mask":
                    file.write(
                        f"{key}: {value}\n"
                    )

            file.write(
                "\nCORRECTED\n"
            )

            for key, value in (
                corrected_slope_results.items()
            ):
                if key != "fit_mask":
                    file.write(
                        f"{key}: {value}\n"
                    )

            file.write(
                "\n0 KM CORRELATION-PROFILE LOCAL MAXIMA\n"
            )

            file.write(
                "Uncorrected left shift (min): "
                f"{zero_km_peaks_uncorrected['left']['shift_minutes']}\n"
            )
            file.write(
                "Uncorrected left correlation: "
                f"{zero_km_peaks_uncorrected['left']['correlation']}\n"
            )
            file.write(
                "Uncorrected right shift (min): "
                f"{zero_km_peaks_uncorrected['right']['shift_minutes']}\n"
            )
            file.write(
                "Uncorrected right correlation: "
                f"{zero_km_peaks_uncorrected['right']['correlation']}\n"
            )
            file.write(
                "Corrected left shift (min): "
                f"{zero_km_peaks_corrected['left']['shift_minutes']}\n"
            )
            file.write(
                "Corrected left correlation: "
                f"{zero_km_peaks_corrected['left']['correlation']}\n"
            )
            file.write(
                "Corrected right shift (min): "
                f"{zero_km_peaks_corrected['right']['shift_minutes']}\n"
            )
            file.write(
                "Corrected right correlation: "
                f"{zero_km_peaks_corrected['right']['correlation']}\n"
            )

        # ==================================================
        # Add to combined summary
        # ==================================================
        combined_results.append(
            [
                float(angle_deg),

                uncorrected_slope_results[
                    "slope_min_per_km"
                ],
                uncorrected_slope_results[
                    "intercept_minutes"
                ],
                uncorrected_slope_results[
                    "speed_km_per_min"
                ],
                uncorrected_slope_results[
                    "speed_m_per_s"
                ],
                uncorrected_slope_results[
                    "r_squared"
                ],
                uncorrected_slope_results[
                    "number_of_fit_points"
                ],

                corrected_slope_results[
                    "slope_min_per_km"
                ],
                corrected_slope_results[
                    "intercept_minutes"
                ],
                corrected_slope_results[
                    "speed_km_per_min"
                ],
                corrected_slope_results[
                    "speed_m_per_s"
                ],
                corrected_slope_results[
                    "r_squared"
                ],
                corrected_slope_results[
                    "number_of_fit_points"
                ],
            ]
        )

        combined_zero_km_peak_results.append(
            [
                float(angle_deg),
                zero_km_peaks_uncorrected["left"][
                    "shift_minutes"
                ],
                zero_km_peaks_uncorrected["left"][
                    "correlation"
                ],
                zero_km_peaks_uncorrected["right"][
                    "shift_minutes"
                ],
                zero_km_peaks_uncorrected["right"][
                    "correlation"
                ],
                zero_km_peaks_corrected["left"][
                    "shift_minutes"
                ],
                zero_km_peaks_corrected["left"][
                    "correlation"
                ],
                zero_km_peaks_corrected["right"][
                    "shift_minutes"
                ],
                zero_km_peaks_corrected["right"][
                    "correlation"
                ],
            ]
        )

        successful_angles.append(
            int(angle_deg)
        )

        print_slope_results(
            angle_deg=angle_deg,
            label=(
                "Uncorrected maximum-correlation ridge"
            ),
            results=uncorrected_slope_results,
        )

        print_slope_results(
            angle_deg=angle_deg,
            label=(
                "Corrected maximum-correlation ridge"
            ),
            results=corrected_slope_results,
        )

        print(
            "\n0 km correlation-profile local maxima:"
        )
        print(
            "  Uncorrected left: "
            f"{zero_km_peaks_uncorrected['left']['shift_minutes']:.3f} min, "
            f"r={zero_km_peaks_uncorrected['left']['correlation']:.6f}"
        )
        print(
            "  Uncorrected right: "
            f"{zero_km_peaks_uncorrected['right']['shift_minutes']:.3f} min, "
            f"r={zero_km_peaks_uncorrected['right']['correlation']:.6f}"
        )
        print(
            "  Corrected left: "
            f"{zero_km_peaks_corrected['left']['shift_minutes']:.3f} min, "
            f"r={zero_km_peaks_corrected['left']['correlation']:.6f}"
        )
        print(
            "  Corrected right: "
            f"{zero_km_peaks_corrected['right']['shift_minutes']:.3f} min, "
            f"r={zero_km_peaks_corrected['right']['correlation']:.6f}"
        )

    except Exception as error:
        print(
            f"ERROR processing {angle_deg}°:"
        )

        print(
            error
        )

        failed_angles.append(
            (
                int(angle_deg),
                str(error),
            )
        )


# ==================================================
# Save combined 0 km local-maximum summary
# ==================================================
combined_zero_km_peak_results = np.asarray(
    combined_zero_km_peak_results,
    dtype=float,
)

combined_zero_km_peak_csv = (
    OUT_DIR
    / "all_angle_zero_km_local_maxima.csv"
)

combined_zero_km_peak_npy = (
    OUT_DIR
    / "all_angle_zero_km_local_maxima.npy"
)

combined_zero_km_peak_header = (
    "angle_deg,"
    "uncorrected_left_shift_minutes,"
    "uncorrected_left_correlation,"
    "uncorrected_right_shift_minutes,"
    "uncorrected_right_correlation,"
    "corrected_left_shift_minutes,"
    "corrected_left_correlation,"
    "corrected_right_shift_minutes,"
    "corrected_right_correlation"
)

if combined_zero_km_peak_results.size > 0:
    np.savetxt(
        combined_zero_km_peak_csv,
        combined_zero_km_peak_results,
        delimiter=",",
        header=combined_zero_km_peak_header,
        comments="",
    )

    np.save(
        combined_zero_km_peak_npy,
        combined_zero_km_peak_results,
    )


# ==================================================
# Save combined angle summary
# ==================================================
combined_results = np.asarray(
    combined_results,
    dtype=float,
)

combined_csv = (
    OUT_DIR
    / "all_angle_slope_summary.csv"
)

combined_npy = (
    OUT_DIR
    / "all_angle_slope_summary.npy"
)

combined_header = (
    "angle_deg,"
    "uncorrected_slope_min_per_km,"
    "uncorrected_intercept_minutes,"
    "uncorrected_speed_km_per_min,"
    "uncorrected_speed_m_per_s,"
    "uncorrected_r_squared,"
    "uncorrected_number_of_fit_points,"
    "corrected_slope_min_per_km,"
    "corrected_intercept_minutes,"
    "corrected_speed_km_per_min,"
    "corrected_speed_m_per_s,"
    "corrected_r_squared,"
    "corrected_number_of_fit_points"
)

if combined_results.size > 0:
    np.savetxt(
        combined_csv,
        combined_results,
        delimiter=",",
        header=combined_header,
        comments="",
    )

    np.save(
        combined_npy,
        combined_results,
    )


# ==================================================
# Plot speed versus angle
# ==================================================
if combined_results.size > 0:
    angles_processed = combined_results[
        :,
        0,
    ]

    uncorrected_speed_m_per_s = combined_results[
        :,
        4,
    ]

    corrected_speed_m_per_s = combined_results[
        :,
        10,
    ]

    fig_speed, ax_speed = plt.subplots(
        figsize=(10, 5)
    )

    ax_speed.plot(
        angles_processed,
        uncorrected_speed_m_per_s,
        marker="o",
        linewidth=1.5,
        label="Uncorrected",
    )

    ax_speed.plot(
        angles_processed,
        corrected_speed_m_per_s,
        marker="o",
        linewidth=1.5,
        label="Corrected",
    )

    ax_speed.axhline(
        0.0,
        color="black",
        linestyle="--",
        linewidth=1.0,
    )

    ax_speed.set_xlim(
        KEOGRAM_ANGLES_DEG.min(),
        KEOGRAM_ANGLES_DEG.max(),
    )

    ax_speed.set_xlabel(
        "Keogram angle (degrees)"
    )

    ax_speed.set_ylabel(
        "Equivalent signed speed (m/s)"
    )

    ax_speed.set_title(
        "Maximum-Correlation Ridge Speed versus Angle"
    )

    ax_speed.grid(
        True,
        alpha=0.3,
    )

    ax_speed.legend(
        loc="best",
    )

    plt.tight_layout()

    speed_angle_plot = (
        OUT_DIR
        / "speed_vs_angle.png"
    )

    fig_speed.savefig(
        speed_angle_plot,
        dpi=DPI,
        bbox_inches="tight",
    )

    plt.close(
        fig_speed
    )


# ==================================================
# Plot R-squared versus angle
# ==================================================
if combined_results.size > 0:
    uncorrected_r_squared = combined_results[
        :,
        5,
    ]

    corrected_r_squared = combined_results[
        :,
        11,
    ]

    fig_r2, ax_r2 = plt.subplots(
        figsize=(10, 5)
    )

    ax_r2.plot(
        angles_processed,
        uncorrected_r_squared,
        marker="o",
        linewidth=1.5,
        label="Uncorrected",
    )

    ax_r2.plot(
        angles_processed,
        corrected_r_squared,
        marker="o",
        linewidth=1.5,
        label="Corrected",
    )

    ax_r2.set_xlim(
        KEOGRAM_ANGLES_DEG.min(),
        KEOGRAM_ANGLES_DEG.max(),
    )

    ax_r2.set_ylim(
        0.0,
        1.0,
    )

    ax_r2.set_xlabel(
        "Keogram angle (degrees)"
    )

    ax_r2.set_ylabel(
        "R²"
    )

    ax_r2.set_title(
        "Maximum-Correlation Ridge Fit R² versus Angle"
    )

    ax_r2.grid(
        True,
        alpha=0.3,
    )

    ax_r2.legend(
        loc="best",
    )

    plt.tight_layout()

    r2_angle_plot = (
        OUT_DIR
        / "r_squared_vs_angle.png"
    )

    fig_r2.savefig(
        r2_angle_plot,
        dpi=DPI,
        bbox_inches="tight",
    )

    plt.close(
        fig_r2
    )


# ==================================================
# Save processing log
# ==================================================
processing_log_file = (
    OUT_DIR
    / "processing_log.txt"
)

with open(
    processing_log_file,
    "w",
    encoding="utf-8",
) as file:
    file.write(
        "Requested angles:\n"
    )

    file.write(
        ", ".join(
            str(angle)
            for angle in KEOGRAM_ANGLES_DEG
        )
        + "\n\n"
    )

    file.write(
        f"Number requested: "
        f"{len(KEOGRAM_ANGLES_DEG)}\n"
    )

    file.write(
        f"Number successful: "
        f"{len(successful_angles)}\n"
    )

    file.write(
        f"Number failed: "
        f"{len(failed_angles)}\n\n"
    )

    file.write(
        "Successful angles:\n"
    )

    file.write(
        ", ".join(
            str(angle)
            for angle in successful_angles
        )
        + "\n\n"
    )

    file.write(
        "Failed angles:\n"
    )

    if failed_angles:
        for angle_deg, reason in failed_angles:
            file.write(
                f"{angle_deg}: {reason}\n"
            )
    else:
        file.write(
            "None\n"
        )


# ==================================================
# Print final summary
# ==================================================
print(
    "\n"
    + "=" * 70
)

print(
    "ALL-ANGLE PROCESSING COMPLETE"
)

print(
    "=" * 70
)

print(
    "Angles requested:"
)

print(
    KEOGRAM_ANGLES_DEG
)

print(
    f"\nNumber requested: "
    f"{len(KEOGRAM_ANGLES_DEG)}"
)

print(
    f"Number successful: "
    f"{len(successful_angles)}"
)

print(
    f"Number failed: "
    f"{len(failed_angles)}"
)

print(
    "\nSuccessful angles:"
)

print(
    successful_angles
)

if failed_angles:
    print(
        "\nFailed angles:"
    )

    for angle_deg, reason in failed_angles:
        print(
            f"  {angle_deg}°: {reason}"
        )

print(
    "\nCombined CSV:"
)

print(
    combined_csv
)

print(
    "\nOutput folder:"
)

print(
    OUT_DIR.resolve()
)
```
