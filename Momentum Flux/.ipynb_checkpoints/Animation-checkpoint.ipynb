## Animation of the frames + corresponding relative amplitude. 

```python
from pathlib import Path
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import imageio.v2 as imageio

# =========================
# User settings
# =========================
IMG_DIR = Path(
    r"C:\Users\domin\juwavelet\Momentum Flux\AWE\01770\remap85_from_l7a\slice_003\cluster_004_global_00007\awe_radiance_pre_interpolation_frames\overlays_crests_phase\ROI_22_120x120"
)

CSV_PATH = IMG_DIR / "crest_relative_amplitude_evolution.csv"
OUT_GIF = IMG_DIR / "crest_relative_amplitude_evolution.gif"

FPS = 3
N_TOTAL_FRAMES = 25
FIG_DPI = 130


def natural_key(path):
    return [
        int(t) if t.isdigit() else t.lower()
        for t in re.split(r"(\d+)", path.name)
    ]


images = sorted(
    [
        p for p in IMG_DIR.glob("*.png")
        if "amplitude_evolution" not in p.name
    ],
    key=natural_key
)

df = pd.read_csv(CSV_PATH)

pt_cols = [c for c in df.columns if re.fullmatch(r"pt\d+", c)]
pt_cols = sorted(pt_cols, key=lambda s: int(s[2:]))

if len(pt_cols) == 0:
    raise ValueError("No pt1, pt2, ... columns found in CSV.")

gif_frames = []

colors = plt.cm.tab10(np.arange(len(pt_cols)))

for i, img_path in enumerate(images):
    if i >= len(df):
        break

    img = imageio.imread(img_path)
    df_now = df.iloc[: i + 1]

    x = (
        df_now["image_index"].to_numpy()
        if "image_index" in df_now.columns
        else np.arange(len(df_now))
    )

    fig, axs = plt.subplots(
        1, 2,
        figsize=(14, 4),
        dpi=FIG_DPI,
        gridspec_kw={"width_ratios": [1.0, 0.8]}
    )

    # =========================
    # Left: original image
    # =========================
    axs[0].imshow(img)
    axs[0].axis("off")
    axs[0].set_title(img_path.name, fontsize=9)

    # =========================
    # Right: amplitude evolution
    # =========================
    for j, col in enumerate(pt_cols):
        y_now = df_now[col].to_numpy()

        # Previous/current points only; no connecting lines
        axs[1].scatter(
            x,
            y_now,
            s=22,
            color=colors[j],
            alpha=0.65,
            label=col
        )

        # Highlight current frame
        axs[1].scatter(
            x[-1],
            y_now[-1],
            s=95,
            color=colors[j],
            edgecolors="black",
            linewidths=0.8
        )

    # fixed 25-frame placeholder axis
    axs[1].set_xlim(-0.5, N_TOTAL_FRAMES - 0.5)
    axs[1].set_ylim(-1.05, 1.05)
    axs[1].set_xticks(np.arange(0, N_TOTAL_FRAMES, 2))

    axs[1].axhline(0, linewidth=0.8)
    axs[1].grid(True, alpha=0.25)

    axs[1].set_title("Relative amplitude evolution", fontsize=11)
    axs[1].set_xlabel("Frame")
    axs[1].set_ylabel("Relative amplitude")

    axs[1].legend(
        loc="upper right",
        fontsize=7,
        ncol=1,
        frameon=True
    )

    fig.tight_layout()

    fig.canvas.draw()
    frame = np.asarray(fig.canvas.buffer_rgba())[:, :, :3].copy()
    gif_frames.append(frame)

    plt.close(fig)

imageio.mimsave(OUT_GIF, gif_frames, loop=1, fps=FPS)

print("Saved GIF:")
print(OUT_GIF)
```
