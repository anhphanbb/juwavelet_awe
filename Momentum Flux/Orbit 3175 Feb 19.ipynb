```python
# %%
from __future__ import annotations

from pathlib import Path
from typing import Tuple, Dict, Any, List

import copy
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
from netCDF4 import Dataset

from juwavelet import transform, utils, plot_utils

try:
    from scipy import stats
except Exception:
    stats = None

# --------------------------
# User params
# --------------------------
FILE_PATH = r"C:\Users\domin\juwavelet\examples\CWT Whole Orbit\l3\OneFile\awe_l3a_tmp_2025153T1853_08658_v01.nc"  # can be L5C / L3C / L3A
FRAME_IDX = 0

# Slicing on the UNBINNED flattened field
X_CHUNK = 600
Y_SLICE = slice(None, None)
SLICE_PAD = 50  # extend each slice by this many UNBINNED pixels on each side

# Choose ONE slicing offset (no comparisons)
X_OFFSET = 0  # set to 0 or 300 or whatever you want

# Optional preprocessing
REMOVE_TREND = True
STANDARDIZE = False
GAUSS_BLUR_SIGMA = 0.0

# Optional 2x2 binning BEFORE processing
APPLY_BINNING = True
BIN_FACTOR = 2

# Edge taper on remaining field (before CWT)
APPLY_TAPER = True
TAPER_EDGE_PIXELS = 10  # in BINNED pixels if APPLY_BINNING else in original pixels

# Wavelet params
S0 = 16
DJ = 1 / 8
JS = 24
JT = 18
ASPECT = 1

# Pixel spacing (km per pixel)
DX_KM_BASE = 2.0
DY_KM_BASE = 2.0

# How many clusters to show (largest by POWER)
N_SHOW = 6

# Per cluster maps
AMP_MIN_FRACTION_OF_CLUSTER_MAX = 0.10
QUIVER_STEP_X = 20
QUIVER_STEP_Y = 20
QUIVER_SCALE = 1

# Red noise style filtering (robust WPS thresholding per scale)
APPLY_RED_NOISE_FILTER = True

# Optional diagnostics output from the red noise filter
SAVE_REDNOISE_DIAGNOSTICS = True

# Save settings
OUTPUT_ROOT = Path("outputs_decompose2d_slices_single_noise_filtering_AUTO")
SAVE_DPI = 200

# --------------------------
# Auto-detect product type
# --------------------------
file_name = Path(FILE_PATH).name.lower()
if "_l3a_" in file_name:
    PRODUCT_TYPE = "l3a"
elif "_l3c_" in file_name:
    PRODUCT_TYPE = "l3c"
elif "_l5c_" in file_name:
    PRODUCT_TYPE = "l5c"
else:
    raise ValueError(f"Cannot detect product type from file name: {FILE_PATH}")

print("Detected product type:", PRODUCT_TYPE)

# --------------------------
# Parameter runs (AUTO)
# --------------------------
if PRODUCT_TYPE == "l3a":
    RUNS: List[Dict[str, Any]] = [
        dict(
            name="base",
            MIN_AMP=1.25,
            THR=0.6,
            WHITE_NOISE_THRESHOLD=0.1,
            WPS_NOISE_THRESHOLD=8.0,
        ),
    ]
else:
    # l5c and l3c (Radiance) use the same settings
    RUNS = [
        dict(
            name="base",
            MIN_AMP=0.2,
            THR=0.1,
            WHITE_NOISE_THRESHOLD=0.02,
            WPS_NOISE_THRESHOLD=8.0,
        ),
    ]

print("Using RUNS:", RUNS)

# --------------------------
# Plot style
# --------------------------
matplotlib.rcParams.update(
    {
        "axes.labelsize": 16,
        "font.size": 16,
        "xtick.labelsize": 14,
        "ytick.labelsize": 14,
        "legend.fontsize": 14,
        "figure.figsize": [2 * 6.94, 2 * 4.29],
    }
)

# --------------------------
# Small helpers
# --------------------------
def save_fig(fig: plt.Figure, outpath: Path) -> None:
    outpath.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(outpath, dpi=SAVE_DPI, bbox_inches="tight")
    plt.close(fig)


def _fmt(v: float) -> str:
    if not np.isfinite(v):
        return "∞"
    v = float(v)
    if abs(v) >= 1000 or abs(v) < 0.1:
        return f"{v:.2e}"
    if abs(v) >= 100:
        return f"{v:.0f}"
    return f"{v:.1f}"


def diverging_limits(*arrays: np.ndarray, p_lo: float = 2, p_hi: float = 98) -> Tuple[float, float]:
    vals = []
    for A in arrays:
        if A is None:
            continue
        a = np.asarray(A)
        good = np.isfinite(a)
        if np.any(good):
            vals.append(a[good].ravel())
    if not vals:
        return (-1.0, 1.0)
    stacked = np.concatenate(vals)
    lo, hi = np.nanpercentile(stacked, [p_lo, p_hi])
    vmax = max(abs(lo), abs(hi))
    if not np.isfinite(vmax) or vmax == 0:
        vmax = 1.0
    return (-vmax, vmax)


# --------------------------
# Data helpers
# --------------------------
def read_frame_auto(path: str, frame_idx: int) -> np.ndarray:
    with Dataset(path, "r") as nc:
        if PRODUCT_TYPE == "l3a":
            varname = "Temperature"
        else:
            varname = "Radiance"

        data = nc.variables[varname][:]  # (t, y, x)

    if data.ndim != 3:
        raise ValueError(f"{varname} expected 3D (t,y,x); got {data.shape}")

    if not (0 <= frame_idx < data.shape[0]):
        raise IndexError(f"FRAME_IDX={frame_idx} out of range 0..{data.shape[0]-1}")

    return np.asarray(data[frame_idx, :, :], dtype=float)


def flatten_swath(Z: np.ndarray, target_ny: int = 300) -> np.ndarray:
    ny, nx = Z.shape
    out = np.full_like(Z, np.nan)

    bottoms = np.full(nx, np.nan, dtype=float)
    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size:
            bottoms[j] = valid_idx.max()

    valid_bottoms = bottoms[np.isfinite(bottoms)]
    if valid_bottoms.size == 0:
        med = float(np.nanmedian(Z)) if np.isfinite(np.nanmedian(Z)) else 0.0
        out = np.nan_to_num(Z, nan=med, posinf=med, neginf=med)
        if target_ny is not None and ny > target_ny:
            out = out[-target_ny:, :]
        return out

    max_bottom = int(valid_bottoms.max())

    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size == 0:
            continue
        bottom = int(valid_idx.max())
        shift = max_bottom - bottom
        if shift == 0:
            out[:, j] = col
        elif shift > 0:
            out[shift:, j] = col[: ny - shift]
        else:
            shift_up = -shift
            out[: ny - shift_up, j] = col[shift_up:]

    med = float(np.nanmedian(out)) if np.isfinite(np.nanmedian(out)) else 0.0
    out = np.nan_to_num(out, nan=med, posinf=med, neginf=med)

    if target_ny is not None and ny > target_ny:
        start = max_bottom + 1 - target_ny
        start = max(start, 0)
        end = start + target_ny
        out = out[start:end, :]

    return out


def bin2d_block_mean(Z: np.ndarray, factor: int) -> np.ndarray:
    if factor is None or int(factor) <= 1:
        return Z
    f = int(factor)
    ny, nx = Z.shape
    ny2 = (ny // f) * f
    nx2 = (nx // f) * f
    if ny2 == 0 or nx2 == 0:
        raise ValueError(f"Binning factor {f} too large for array {Z.shape}")
    Z2 = Z[:ny2, :nx2]
    return Z2.reshape(ny2 // f, f, nx2 // f, f).mean(axis=(1, 3))


def unbin_repeat_xy(A: np.ndarray, factor: int) -> np.ndarray:
    """
    Nearest-neighbor upsample (repeat) from (y,x) to (y*factor, x*factor).
    Preserves NaNs by repetition.
    """
    if factor is None or int(factor) <= 1:
        return A
    f = int(factor)
    return np.repeat(np.repeat(A, f, axis=0), f, axis=1)


def fit_plane(Z: np.ndarray) -> np.ndarray:
    ny, nx = Z.shape
    yy, xx = np.mgrid[0:ny, 0:nx]
    G = np.stack([xx.ravel(), yy.ravel(), np.ones(nx * ny)], axis=1)
    coeff, *_ = np.linalg.lstsq(G, Z.ravel(), rcond=None)
    return coeff[0] * xx + coeff[1] * yy + coeff[2]


def extract_first_fourier_bg(Z: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
    ny, nx = Z.shape
    F = np.fft.fft2(Z)
    F_bg = np.zeros_like(F, dtype=complex)

    if nx > 1:
        F_bg[0, 1] = F[0, 1]
        F_bg[0, -1] = F[0, -1]
    if ny > 1:
        F_bg[1, 0] = F[1, 0]
        F_bg[-1, 0] = F[-1, 0]

    bg = np.fft.ifft2(F_bg).real
    return bg, Z - bg


def maybe_blur(Z: np.ndarray, sigma: float) -> np.ndarray:
    if sigma <= 0:
        return Z
    try:
        from scipy.ndimage import gaussian_filter
    except Exception:
        return Z
    return gaussian_filter(Z, sigma=float(sigma))


def _hanning_edge_1d(n: int, edge: int) -> np.ndarray:
    w = np.ones(n, dtype=float)
    if edge <= 0:
        return w
    edge = min(edge, n // 2)
    ramp = 0.5 * (1.0 - np.cos(np.linspace(0, np.pi, edge, endpoint=True)))
    w[:edge] = ramp
    w[-edge:] = ramp[::-1]
    return w


def apply_edge_hanning_taper(Z: np.ndarray, edge: int) -> np.ndarray:
    if edge <= 0:
        return Z
    ny, nx = Z.shape
    wx = _hanning_edge_1d(nx, edge)
    wy = _hanning_edge_1d(ny, edge)
    W = np.outer(wy, wx)
    return Z * W


def effective_lambda(cwt: dict, s_idx: int, t_idx: int) -> Tuple[float, float, float]:
    lamx = float(cwt["wavelength_x"][s_idx, t_idx])
    lamy = float(cwt["wavelength_y"][s_idx, t_idx])
    inv2 = 0.0
    if np.isfinite(lamx) and lamx != 0:
        inv2 += (1.0 / lamx) ** 2
    if np.isfinite(lamy) and lamy != 0:
        inv2 += (1.0 / lamy) ** 2
    lam = np.inf if inv2 == 0 else 1.0 / np.sqrt(inv2)
    return lamx, lamy, lam


def compute_cluster_power_table(cwt: dict, iwave: np.ndarray) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    dec = cwt["decomposition"]
    power4d = np.abs(dec) ** 2

    cluster_ids = np.unique(iwave[iwave >= 0]).astype(int)
    if cluster_ids.size == 0:
        return cluster_ids, np.array([]), np.array([])

    P_abs = np.zeros(cluster_ids.size, dtype=float)
    for i, kcl in enumerate(cluster_ids):
        P_abs[i] = float(np.sum(power4d[iwave == kcl]))

    total = float(np.sum(P_abs))
    if total > 0 and np.isfinite(total):
        P_pct = 100.0 * P_abs / total
    else:
        P_pct = np.full_like(P_abs, np.nan)

    return cluster_ids, P_abs, P_pct


def cluster_amp_theta_maps(
    cwt: dict,
    iwave: np.ndarray,
    kcl: int,
    amp_min_fraction_of_cluster_max: float = 0.2,
) -> Tuple[np.ndarray, np.ndarray]:
    dec = cwt["decomposition"]  # (ns, nt, nx, ny) complex
    power = np.abs(dec) ** 2

    mask4d = (iwave == kcl)
    if not np.any(mask4d):
        ny = dec.shape[3]
        nx = dec.shape[2]
        return np.full((ny, nx), np.nan), np.full((ny, nx), np.nan)

    ns, nt, nx, ny = power.shape
    power2 = power.reshape(ns * nt, nx * ny)
    mask2 = mask4d.reshape(ns * nt, nx * ny)

    p = np.where(mask2, power2, -np.inf)
    pmax = np.max(p, axis=0)

    valid = np.isfinite(pmax) & (pmax > 0)

    A = np.full(nx * ny, np.nan, dtype=float)
    theta = np.full(nx * ny, np.nan, dtype=float)

    if np.any(valid):
        arg = np.argmax(p[:, valid], axis=0)
        t_idx = arg % nt

        A_valid = np.sqrt(pmax[valid])
        theta_valid = np.asarray(cwt["theta"], dtype=float)[t_idx]

        A[valid] = A_valid
        theta[valid] = theta_valid

    A_xy = A.reshape(nx, ny).T
    theta_xy = theta.reshape(nx, ny).T

    if np.isfinite(np.nanmax(A_xy)):
        thr = float(amp_min_fraction_of_cluster_max) * float(np.nanmax(A_xy))
        low = A_xy < thr
        A_xy[low] = np.nan
        theta_xy[low] = np.nan

    return A_xy, theta_xy


def theta_to_uv(theta_xy: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
    return np.cos(theta_xy), np.sin(theta_xy)


# --------------------------
# Red noise style filtering helper
# --------------------------
def apply_red_noise_filter_cwt(
    cwt: dict,
    white_noise_threshold: float = 0.1,
    wps_noise_threshold: float = 8.0,
) -> Tuple[dict, Dict[str, Any]]:
    """
    Robust per-scale WPS filtering.
    Returns filtered_cwt (deep copy) and a small diagnostics dict.
    """
    diag: Dict[str, Any] = {}
    if stats is None:
        diag["skipped"] = True
        diag["reason"] = "scipy not available"
        return cwt, diag

    cwt_copy = copy.deepcopy(cwt)
    dec = cwt_copy["decomposition"]  # (ns, nt, nx, ny) complex

    WPS0 = np.abs(dec) ** 2
    diag["n_coeff_total"] = int(WPS0.size)

    mask_1 = WPS0 < (float(white_noise_threshold) ** 2)
    dec[mask_1] = 0

    WPS1 = np.abs(dec) ** 2
    diag["n_zero_after_white"] = int(np.count_nonzero(WPS1 == 0))

    median_WPS = np.median(WPS1, axis=(1, 2, 3))  # (ns,)
    mad = stats.median_abs_deviation(WPS1, axis=(1, 2, 3), scale=1.0)  # (ns,)
    sMAD_WPS = 1.4826 * mad

    sMAD_safe = np.where(sMAD_WPS > 0, sMAD_WPS, np.nan)
    WPS_scaled = (WPS1 - median_WPS[:, None, None, None]) / sMAD_safe[:, None, None, None]

    mask = WPS_scaled < float(wps_noise_threshold)
    dec[mask] = 0

    WPS2 = np.abs(dec) ** 2
    diag["n_zero_after_scaled"] = int(np.count_nonzero(WPS2 == 0))

    finite_scaled = np.isfinite(WPS_scaled)

    diag["sMAD_min"] = float(np.nanmin(sMAD_WPS))
    diag["sMAD_median"] = float(np.nanmedian(sMAD_WPS))

    if np.any(finite_scaled):
        diag["WPS_scaled_min"] = float(np.nanmin(WPS_scaled[finite_scaled]))
        diag["WPS_scaled_median"] = float(np.nanmedian(WPS_scaled[finite_scaled]))
        diag["WPS_scaled_max"] = float(np.nanmax(WPS_scaled[finite_scaled]))
    else:
        diag["WPS_scaled_min"] = np.nan
        diag["WPS_scaled_median"] = np.nan
        diag["WPS_scaled_max"] = np.nan

    return cwt_copy, diag


# --------------------------
# Quicklook plots
# --------------------------
def quicklook_save(x1d: np.ndarray, y1d: np.ndarray, Z: np.ndarray, title: str, outpath: Path) -> None:
    fig, ax = plt.subplots()
    im = ax.pcolormesh(x1d, y1d, Z, shading="auto", cmap="gray")
    fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04).set_label("arb")
    ax.set_xlabel("x (km)")
    ax.set_ylabel("y (km)")
    ax.set_title(title)
    ax.set_aspect("equal", adjustable="box")
    fig.tight_layout()
    save_fig(fig, outpath)


def quicklook_background_decomp_4panel_save(
    x1d_km: np.ndarray,
    y1d_km: np.ndarray,
    original: np.ndarray,
    background: np.ndarray,
    remaining: np.ndarray,
    remaining_tapered: np.ndarray,
    taper_edge_px: int,
    outpath: Path,
    wavy_after_noise_xy: np.ndarray | None = None,
) -> None:
    med = np.nanmedian(original)
    orig_c = original - med
    bg_c = background - med
    rem_c = remaining
    rem_tapered_c = remaining_tapered

    arrays = [orig_c, bg_c, rem_c, rem_tapered_c]
    if wavy_after_noise_xy is not None:
        arrays.append(wavy_after_noise_xy)

    vmin, vmax = diverging_limits(*arrays, p_lo=1, p_hi=99)

    fig, axs = plt.subplots(2, 3, figsize=(16, 10), constrained_layout=True)
    axs = axs.flatten()

    panels = [
        (orig_c, "Original (median sub)"),
        (bg_c, "Background (plane + first Fourier)"),
        (rem_c, "Remaining = original - background"),
        (rem_tapered_c, f"Remaining with edge taper (width={taper_edge_px})"),
    ]

    if wavy_after_noise_xy is not None:
        panels.append((wavy_after_noise_xy, "Wavy stuff (wavelet reconstruction after noise filtering)"))

    im0 = None
    for ax, (Z, title) in zip(axs, panels):
        im0 = ax.pcolormesh(x1d_km, y1d_km, Z, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
        ax.set_title(title)
        ax.set_xlabel("x (km)")
        ax.set_ylabel("y (km)")
        ax.set_aspect("equal", adjustable="box")

    for ax in axs[len(panels) :]:
        ax.axis("off")

    cbar = fig.colorbar(im0, ax=axs[: len(panels)], orientation="vertical", fraction=0.04, pad=0.02)
    cbar.set_label("arb")
    save_fig(fig, outpath)


def plot_cluster_individual_save(
    cwt: dict,
    iwave: np.ndarray,
    kcl: int,
    x1d_km: np.ndarray,
    y1d_km: np.ndarray,
    orig_decomp: np.ndarray,
    bg_xy: np.ndarray,
    dx_km: float,
    dy_km: float,
    outpath: Path,
) -> None:
    decomposition = cwt["decomposition"]
    decomposition[:] = orig_decomp
    decomposition[iwave != kcl] = 0

    rec = transform.reconstruct2d(cwt)  # (nx, ny)
    rec_xy = rec.T - np.median(rec)

    A_xy, theta_xy = cluster_amp_theta_maps(
        cwt, iwave, kcl, amp_min_fraction_of_cluster_max=AMP_MIN_FRACTION_OF_CLUSTER_MAX
    )

    dec0 = orig_decomp
    power0 = np.abs(dec0) ** 2
    mask4d = (iwave == kcl)
    ns, nt, nx, ny = power0.shape

    power2 = power0.reshape(ns * nt, nx * ny)
    mask2 = mask4d.reshape(ns * nt, nx * ny)
    p = np.where(mask2, power2, -np.inf)
    pmax = np.max(p, axis=0)

    lam_xy = np.full(nx * ny, np.nan, dtype=float)
    valid = np.isfinite(pmax) & (pmax > 0)
    if np.any(valid):
        arg = np.argmax(p[:, valid], axis=0)
        s_idx = arg // nt
        t_idx = arg % nt

        lamx = np.asarray(cwt["wavelength_x"], dtype=float)[s_idx, t_idx]
        lamy = np.asarray(cwt["wavelength_y"], dtype=float)[s_idx, t_idx]

        inv2 = np.zeros_like(lamx, dtype=float)
        goodx = np.isfinite(lamx) & (lamx != 0)
        goody = np.isfinite(lamy) & (lamy != 0)
        inv2[goodx] += (1.0 / lamx[goodx]) ** 2
        inv2[goody] += (1.0 / lamy[goody]) ** 2

        lam = np.full_like(inv2, np.nan, dtype=float)
        ok = inv2 > 0
        lam[ok] = 1.0 / np.sqrt(inv2[ok])
        lam_xy[valid] = lam

    lam_xy = lam_xy.reshape(nx, ny).T

    if np.isfinite(np.nanmax(A_xy)):
        thr = float(AMP_MIN_FRACTION_OF_CLUSTER_MAX) * float(np.nanmax(A_xy))
        lam_xy[A_xy < thr] = np.nan

    bg_vmin, bg_vmax = diverging_limits(bg_xy, p_lo=2, p_hi=98)
    cl_vmin, cl_vmax = diverging_limits(rec_xy, p_lo=2, p_hi=98)

    theta_vals = np.asarray(cwt["theta"], dtype=float)
    theta_min = float(np.nanmin(theta_vals))
    theta_max = float(np.nanmax(theta_vals))

    if np.any(np.isfinite(lam_xy)):
        lam_vmin, lam_vmax = np.nanpercentile(lam_xy, [5, 95])
        if (not np.isfinite(lam_vmin)) or (not np.isfinite(lam_vmax)) or (lam_vmin == lam_vmax):
            lam_vmin = float(np.nanmin(lam_xy))
            lam_vmax = float(np.nanmax(lam_xy))
    else:
        lam_vmin, lam_vmax = 0.0, 1.0

    fig, axs = plt.subplots(2, 2, figsize=(16, 11), constrained_layout=True)
    ax00, ax01, ax10, ax11 = axs[0, 0], axs[0, 1], axs[1, 0], axs[1, 1]

    ax00.pcolormesh(x1d_km, y1d_km, bg_xy, shading="auto", cmap="gray", vmin=bg_vmin, vmax=bg_vmax)
    im00 = ax00.pcolormesh(x1d_km, y1d_km, rec_xy, shading="auto", cmap="RdBu_r", vmin=cl_vmin, vmax=cl_vmax, alpha=0.6)
    ax00.set_title(f"Cluster {kcl} overlay")
    ax00.set_xlabel("x (km)")
    ax00.set_ylabel("y (km)")
    ax00.set_aspect("equal", adjustable="box")
    fig.colorbar(im00, ax=ax00, fraction=0.046, pad=0.02)

    im01 = ax01.pcolormesh(x1d_km, y1d_km, A_xy, shading="auto", cmap="plasma")
    n_pix = int(np.count_nonzero(np.isfinite(A_xy)))
    area_km2 = n_pix * float(dx_km) * float(dy_km)
    ax01.set_title("A(x,y) = max |coeff| in cluster\n" + rf"Area = {area_km2:.0f} km$^2$ ({n_pix} px)")
    ax01.set_xlabel("x (km)")
    ax01.set_ylabel("y (km)")
    ax01.set_aspect("equal", adjustable="box")
    fig.colorbar(im01, ax=ax01, fraction=0.046, pad=0.02)

    im10 = ax10.pcolormesh(x1d_km, y1d_km, lam_xy, shading="auto", cmap="viridis", vmin=lam_vmin, vmax=lam_vmax)
    if np.any(np.isfinite(lam_xy)):
        lam_med = float(np.nanmedian(lam_xy))
        lam_p10, lam_p90 = np.nanpercentile(lam_xy, [10, 90])
        ax10.set_title(
            r"Dominant $\lambda(x,y)$ in cluster" + "\n"
            + rf"median={_fmt(lam_med)} km (10–90%: {_fmt(lam_p10)}–{_fmt(lam_p90)} km)"
        )
    else:
        ax10.set_title(r"Dominant $\lambda(x,y)$ in cluster")
    ax10.set_xlabel("x (km)")
    ax10.set_ylabel("y (km)")
    ax10.set_aspect("equal", adjustable="box")
    fig.colorbar(im10, ax=ax10, fraction=0.046, pad=0.02).set_label("km")

    im11 = ax11.pcolormesh(x1d_km, y1d_km, theta_xy, shading="auto", cmap="PiYG", vmin=theta_min, vmax=theta_max)
    ax11.set_title("θ(x,y) + quiver")
    ax11.set_xlabel("x (km)")
    ax11.set_ylabel("y (km)")
    ax11.set_aspect("equal", adjustable="box")
    fig.colorbar(im11, ax=ax11, fraction=0.046, pad=0.02)

    if np.any(np.isfinite(theta_xy)):
        u, v = theta_to_uv(theta_xy)
        arrow_len_km = 40.0
        u = u * arrow_len_km
        v = v * arrow_len_km

        Xq, Yq = np.meshgrid(x1d_km, y1d_km)
        Xqs = Xq[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        Yqs = Yq[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        uqs = u[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        vqs = v[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        good = np.isfinite(uqs) & np.isfinite(vqs)
        ax11.quiver(
            Xqs[good], Yqs[good], uqs[good], vqs[good],
            angles="xy", scale_units="xy", scale=QUIVER_SCALE, width=0.0025
        )

    save_fig(fig, outpath)


# --------------------------
# Cluster products (for L6C/L6A outputs)
# --------------------------
def cluster_dominant_lambda_map(
    cwt: dict,
    iwave: np.ndarray,
    kcl: int,
    amp_mask_xy: np.ndarray | None = None,
) -> np.ndarray:
    """
    Per-pixel dominant wavelength map λ(x,y) for ONE cluster.
    Returns shape (ny, nx) in XY.

    If amp_mask_xy is provided (A_xy with NaNs), apply the same NaN mask to λ.
    """
    dec = cwt["decomposition"]  # (ns, nt, nx, ny) complex
    power0 = np.abs(dec) ** 2
    mask4d = (iwave == kcl)
    if not np.any(mask4d):
        ny = dec.shape[3]
        nx = dec.shape[2]
        return np.full((ny, nx), np.nan)

    ns, nt, nx, ny = power0.shape
    power2 = power0.reshape(ns * nt, nx * ny)
    mask2 = mask4d.reshape(ns * nt, nx * ny)

    p = np.where(mask2, power2, -np.inf)
    pmax = np.max(p, axis=0)
    valid = np.isfinite(pmax) & (pmax > 0)

    lam_xy = np.full(nx * ny, np.nan, dtype=float)
    if np.any(valid):
        arg = np.argmax(p[:, valid], axis=0)
        s_idx = arg // nt
        t_idx = arg % nt

        lamx = np.asarray(cwt["wavelength_x"], dtype=float)[s_idx, t_idx]
        lamy = np.asarray(cwt["wavelength_y"], dtype=float)[s_idx, t_idx]

        inv2 = np.zeros_like(lamx, dtype=float)
        goodx = np.isfinite(lamx) & (lamx != 0)
        goody = np.isfinite(lamy) & (lamy != 0)
        inv2[goodx] += (1.0 / lamx[goodx]) ** 2
        inv2[goody] += (1.0 / lamy[goody]) ** 2

        lam = np.full_like(inv2, np.nan, dtype=float)
        ok = inv2 > 0
        lam[ok] = 1.0 / np.sqrt(inv2[ok])

        lam_xy[valid] = lam

    lam_xy = lam_xy.reshape(nx, ny).T  # XY

    if amp_mask_xy is not None:
        lam_xy[~np.isfinite(amp_mask_xy)] = np.nan

    return lam_xy


def copy_lxc_to_l6x_with_clusters(
    lxc_path: str | Path,
    l6x_path: str | Path,
    clusters: List[Dict[str, Any]],
    y_dim: int,
    x_dim: int,
    unbin_factor: int = 1,
) -> None:
    """
    Copy everything from input file, then append cluster products.

    Adds:
      - dimension: cluster
      - dimension: y_cluster (=y_dim*unbin_factor), x_cluster (=x_dim*unbin_factor)
      - SlicesNo(cluster)
      - Area(cluster)
      - ClusterPowerAbs(cluster)
      - ClusterPowerPct(cluster)
      - Amplitude(cluster,y_cluster,x_cluster)
      - DominantWavelength(cluster,y_cluster,x_cluster)
      - Angle(cluster,y_cluster,x_cluster)
      - ClusterReconstruction(cluster,y_cluster,x_cluster)   (optional)

    If unbin_factor>1, upsample products by repetition.
    """
    def _unbin_repeat_xy(A: np.ndarray, factor: int) -> np.ndarray:
        if factor is None or int(factor) <= 1:
            return A
        f = int(factor)
        return np.repeat(np.repeat(A, f, axis=0), f, axis=1)

    lxc_path = Path(lxc_path)
    l6x_path = Path(l6x_path)
    l6x_path.parent.mkdir(parents=True, exist_ok=True)

    ncl = int(len(clusters))
    has_rec = any((c.get("rec_xy", None) is not None) for c in clusters)

    fup = int(unbin_factor) if (unbin_factor is not None and int(unbin_factor) > 1) else 1
    y_dim_w = int(y_dim) * fup
    x_dim_w = int(x_dim) * fup

    with Dataset(lxc_path, "r") as src, Dataset(l6x_path, "w", format="NETCDF4") as dst:
        dst.setncatts({k: src.getncattr(k) for k in src.ncattrs()})

        for dname, dim in src.dimensions.items():
            dst.createDimension(dname, (len(dim) if not dim.isunlimited() else None))

        for vname, var in src.variables.items():
            out = dst.createVariable(
                vname,
                var.datatype,
                var.dimensions,
                zlib=True,
                complevel=4,
                fill_value=getattr(var, "_FillValue", None),
            )
            out.setncatts({k: var.getncattr(k) for k in var.ncattrs()})
            out[:] = var[:]

        dst.createDimension("cluster", ncl)
        dst.createDimension("y_cluster", y_dim_w)
        dst.createDimension("x_cluster", x_dim_w)

        v_slice = dst.createVariable("SlicesNo", "i4", ("cluster",), zlib=True, complevel=4)
        v_area = dst.createVariable("Area", "f4", ("cluster",), zlib=True, complevel=4)
        v_pabs = dst.createVariable("ClusterPowerAbs", "f8", ("cluster",), zlib=True, complevel=4)
        v_ppct = dst.createVariable("ClusterPowerPct", "f4", ("cluster",), zlib=True, complevel=4)

        v_amp = dst.createVariable(
            "Amplitude", "f4", ("cluster", "y_cluster", "x_cluster"),
            zlib=True, complevel=4, fill_value=np.float32(np.nan)
        )
        v_lam = dst.createVariable(
            "DominantWavelength", "f4", ("cluster", "y_cluster", "x_cluster"),
            zlib=True, complevel=4, fill_value=np.float32(np.nan)
        )
        v_ang = dst.createVariable(
            "Angle", "f4", ("cluster", "y_cluster", "x_cluster"),
            zlib=True, complevel=4, fill_value=np.float32(np.nan)
        )

        v_rec = None
        if has_rec:
            v_rec = dst.createVariable(
                "ClusterReconstruction", "f4", ("cluster", "y_cluster", "x_cluster"),
                zlib=True, complevel=4, fill_value=np.float32(np.nan)
            )
            v_rec.units = "arb"
            v_rec.long_name = "Per-cluster wavelet reconstruction map (relative; median-subtracted)"

        v_area.units = "km2"
        v_lam.units = "km"
        v_ang.units = "rad"
        v_slice.long_name = "Slice index (starts at 0) for each cluster"
        v_pabs.units = "arb"
        v_ppct.units = "percent"

        if PRODUCT_TYPE == "l3a":
            v_amp.units = "K"
        else:
            v_amp.units = "kR"

        if ncl == 0:
            return

        v_slice[:] = np.asarray([c["slice_no"] for c in clusters], dtype=np.int32)
        v_area[:] = np.asarray([c["area_km2"] for c in clusters], dtype=np.float32)
        v_pabs[:] = np.asarray([c.get("power_abs", np.nan) for c in clusters], dtype=np.float64)
        v_ppct[:] = np.asarray([c.get("power_pct", np.nan) for c in clusters], dtype=np.float32)

        amp_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32)
        lam_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32)
        ang_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32)
        rec_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32) if has_rec else None

        for i, c in enumerate(clusters):
            A = _unbin_repeat_xy(c["A_xy"], fup)
            L = _unbin_repeat_xy(c["lam_xy"], fup)
            T = _unbin_repeat_xy(c["theta_xy"], fup)

            def _force_shape(M: np.ndarray) -> np.ndarray:
                if M.shape == (y_dim_w, x_dim_w):
                    return M.astype(np.float32)
                tmp = np.full((y_dim_w, x_dim_w), np.nan, dtype=np.float32)
                yy = min(y_dim_w, M.shape[0])
                xx = min(x_dim_w, M.shape[1])
                tmp[:yy, :xx] = M[:yy, :xx]
                return tmp

            amp_stack[i] = _force_shape(A)
            lam_stack[i] = _force_shape(L)
            ang_stack[i] = _force_shape(T)

            if has_rec and rec_stack is not None and v_rec is not None:
                R = c.get("rec_xy", None)
                if R is not None:
                    R = _force_shape(_unbin_repeat_xy(R, fup))
                    rec_stack[i] = R

        v_amp[:] = amp_stack
        v_lam[:] = lam_stack
        v_ang[:] = ang_stack
        if has_rec and v_rec is not None and rec_stack is not None:
            v_rec[:] = rec_stack


def make_output_cluster_path(input_path: str | Path) -> Path:
    p = Path(input_path)
    name = p.name
    if "_l5c_" in name:
        name = name.replace("_l5c_", "_l6c_")
    elif "_l3c_" in name:
        name = name.replace("_l3c_", "_l6c_")
    elif "_l3a_" in name:
        name = name.replace("_l3a_", "_l6a_")
    else:
        name = p.stem + "_clusters.nc"
    return p.with_name(name)


# --------------------------
# Slice pipeline
# --------------------------
def run_pipeline_for_slice(
    rad2d_flat: np.ndarray,
    x_slice: slice,
    outdir: Path,
    x_start_unbinned: int,
    core_start_unbinned: int,
    core_end_unbinned: int,
    cfg: Dict[str, Any],
    slice_no: int,
) -> Dict[str, Any]:
    """
    Runs processing on an extended slice (x_slice), but returns only CORE tiles for stitching.
    Figure 1 includes extra panel: wavelet reconstruction after noise filtering.
    Figure 9 has 3 panels:
      Remaining (input to CWT)
      Wavy reconstruction after noise filtering
      Reconstruction after clustering (all clusters)
    """
    outdir.mkdir(parents=True, exist_ok=True)

    min_amp = float(cfg.get("MIN_AMP"))
    thr = float(cfg.get("THR"))
    white_noise_threshold = float(cfg.get("WHITE_NOISE_THRESHOLD"))
    wps_noise_threshold = float(cfg.get("WPS_NOISE_THRESHOLD"))
    run_name = str(cfg.get("name", "run"))

    STORE_CLUSTER_RECON = True  # save per-cluster recon maps to output NetCDF

    if APPLY_BINNING and int(BIN_FACTOR) > 1:
        f = int(BIN_FACTOR)
        dx_km = DX_KM_BASE * f
        dy_km = DY_KM_BASE * f
    else:
        f = 1
        dx_km = DX_KM_BASE
        dy_km = DY_KM_BASE

    core_width_unbinned = int(core_end_unbinned - core_start_unbinned)
    core_w_binned = core_width_unbinned // f
    rel0_unbinned = int(core_start_unbinned - x_start_unbinned)
    rel0_binned = rel0_unbinned // f
    rel1_binned = rel0_binned + core_w_binned

    def crop_pad_core(A: np.ndarray) -> np.ndarray:
        nyA, nxA = A.shape
        left = max(0, rel0_binned)
        right = min(nxA, rel1_binned)
        core = A[:, left:right]
        need = core_w_binned - core.shape[1]
        if need > 0:
            core = np.concatenate([core, np.full((nyA, need), np.nan)], axis=1)
        elif need < 0:
            core = core[:, :core_w_binned]
        return core

    raw_win = np.asarray(rad2d_flat[Y_SLICE, x_slice], dtype=float)

    if APPLY_BINNING and int(BIN_FACTOR) > 1:
        raw_win = bin2d_block_mean(raw_win, BIN_FACTOR)

    ny, nx = raw_win.shape
    x1d_km = np.arange(nx, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    bin_tag = f" (binned {BIN_FACTOR}x{BIN_FACTOR})" if (APPLY_BINNING and int(BIN_FACTOR) > 1) else ""

    if PRODUCT_TYPE == "l3a":
        label0 = "Temperature"
    else:
        label0 = "Radiance"

    quicklook_save(
        x1d_km, y1d_km, raw_win,
        f"{label0} (window, flattened){bin_tag}\n{run_name} MIN_AMP={min_amp} THR={thr}",
        outdir / "00_window.png",
    )

    proc0 = raw_win.astype(float, copy=True)

    if REMOVE_TREND:
        plane = fit_plane(proc0)
        detrended = proc0 - plane
    else:
        plane = np.zeros_like(proc0)
        detrended = proc0

    fourier_bg, _ = extract_first_fourier_bg(detrended)
    background = plane + fourier_bg
    remaining = proc0 - background

    if APPLY_TAPER:
        remaining_tapered = apply_edge_hanning_taper(remaining, TAPER_EDGE_PIXELS)
    else:
        remaining_tapered = remaining.copy()

    proc = remaining_tapered.copy()
    if STANDARDIZE:
        m = float(np.median(proc))
        s = float(np.std(proc))
        proc = (proc - m) / s if s > 0 else (proc - m)
    proc = maybe_blur(proc, GAUSS_BLUR_SIGMA)

    wave_xy = proc.T  # (nx, ny)

    cwt_raw = transform.decompose2d(
        wave_xy, dx=dx_km, dy=dy_km, s0=S0, dj=DJ, js=JS, jt=JT, aspect=ASPECT
    )

    if APPLY_RED_NOISE_FILTER:
        cwt, diag = apply_red_noise_filter_cwt(
            cwt_raw,
            white_noise_threshold=white_noise_threshold,
            wps_noise_threshold=wps_noise_threshold,
        )
        if SAVE_REDNOISE_DIAGNOSTICS:
            lines = []
            lines.append("Red noise filter diagnostics\n\n")
            lines.append(f"name                  = {run_name}\n")
            lines.append(f"WHITE_NOISE_THRESHOLD = {white_noise_threshold}\n")
            lines.append(f"WPS_NOISE_THRESHOLD   = {wps_noise_threshold}\n\n")
            if diag.get("skipped", False):
                lines.append(f"SKIPPED: {diag.get('reason','unknown')}\n")
            else:
                lines.append(f"n_coeff_total        = {diag.get('n_coeff_total')}\n")
                lines.append(f"n_zero_after_white   = {diag.get('n_zero_after_white')}\n")
                lines.append(f"n_zero_after_scaled  = {diag.get('n_zero_after_scaled')}\n")
                lines.append(f"sMAD_WPS min         = {diag.get('sMAD_min')}\n")
                lines.append(f"sMAD_WPS median      = {diag.get('sMAD_median')}\n")
                lines.append(f"WPS_scaled min       = {diag.get('WPS_scaled_min')}\n")
                lines.append(f"WPS_scaled median    = {diag.get('WPS_scaled_median')}\n")
                lines.append(f"WPS_scaled max       = {diag.get('WPS_scaled_max')}\n")
            (outdir / "02a_rednoise_diagnostics.txt").write_text("".join(lines))
    else:
        cwt = cwt_raw

    rec_wavy = transform.reconstruct2d(cwt)  # (nx, ny)
    wavy_xy = rec_wavy.T - np.median(rec_wavy)

    quicklook_background_decomp_4panel_save(
        x1d_km, y1d_km,
        proc0, background, remaining, remaining_tapered, TAPER_EDGE_PIXELS,
        outdir / "01_background_decomp_5panel.png",
        wavy_after_noise_xy=wavy_xy,
    )

    amps, idxs, iwave = utils.identify_cluster2d(cwt, min_amp=min_amp, thr=thr)
    decomposition = cwt["decomposition"]
    orig_decomp = decomposition.copy()

    cmap_cwt = matplotlib.cm.turbo
    norm_cwt = matplotlib.colors.BoundaryNorm(np.exp(np.linspace(np.log(0.05), np.log(1.0), 10)), cmap_cwt.N)
    fig_cwt, _ = utils.plot_decomposition2d(cwt, redux_s=2, redux_t=2, cmap=cmap_cwt, norm=norm_cwt)
    save_fig(fig_cwt, outdir / "02_decomposition_overview.png")

    plot_utils.plot_fov_wavelet_spectrum(cwt, stat="max")
    fig_spec = plt.gcf()
    save_fig(fig_spec, outdir / "03_fov_wavelet_spectrum.png")

    bg = remaining_tapered - np.median(remaining_tapered)

    cluster_ids, P_abs, P_pct = compute_cluster_power_table(cwt, iwave)
    if cluster_ids.size == 0:
        (outdir / "NO_CLUSTERS.txt").write_text(
            "No clusters found.\n\n"
            f"Run name: {run_name}\n"
            f"Used MIN_AMP={min_amp}, THR={thr}\n"
            f"Used WHITE_NOISE_THRESHOLD={white_noise_threshold}, WPS_NOISE_THRESHOLD={wps_noise_threshold}\n\n"
            "Try lowering MIN_AMP or THR.\n"
            "If red noise filter is on, try lowering WPS_NOISE_THRESHOLD or WHITE_NOISE_THRESHOLD.\n"
        )
        remaining_core_xy = crop_pad_core(proc)
        nan_core = np.full_like(remaining_core_xy, np.nan)
        return {
            "has_clusters": False,
            "x_start_unbinned": core_start_unbinned,
            "dx_km": dx_km,
            "dy_km": dy_km,
            "remaining_xy": remaining_core_xy,
            "wavy_xy": nan_core,
            "rec_all_xy": nan_core,
            "cluster_records": [],
        }

    power_abs_map = {int(k): float(p) for k, p in zip(cluster_ids, P_abs)}
    power_pct_map = {int(k): float(p) for k, p in zip(cluster_ids, P_pct)}

    cluster_records: List[Dict[str, Any]] = []
    for kcl in cluster_ids.astype(int):
        kcl = int(kcl)

        A_xy, theta_xy = cluster_amp_theta_maps(
            cwt, iwave, kcl,
            amp_min_fraction_of_cluster_max=AMP_MIN_FRACTION_OF_CLUSTER_MAX,
        )
        lam_xy = cluster_dominant_lambda_map(cwt, iwave, kcl, amp_mask_xy=A_xy)

        rec_core = None
        if STORE_CLUSTER_RECON:
            decomposition[:] = orig_decomp
            decomposition[iwave != kcl] = 0
            rec = transform.reconstruct2d(cwt)  # (nx, ny)
            rec_xy = rec.T
            rec_xy = rec_xy - np.nanmedian(rec_xy)
            rec_core = crop_pad_core(rec_xy)
            decomposition[:] = orig_decomp

        A_core = crop_pad_core(A_xy)
        th_core = crop_pad_core(theta_xy)
        lam_core = crop_pad_core(lam_xy)

        n_pix = int(np.count_nonzero(np.isfinite(A_core)))
        area_km2 = float(n_pix) * float(dx_km) * float(dy_km)

        cluster_records.append(
            dict(
                slice_no=int(slice_no),
                kcl=kcl,
                area_km2=area_km2,
                power_abs=power_abs_map.get(kcl, np.nan),
                power_pct=power_pct_map.get(kcl, np.nan),
                A_xy=A_core,
                lam_xy=lam_core,
                theta_xy=th_core,
                rec_xy=rec_core,
            )
        )

    order_power = np.argsort(P_abs)[::-1]
    cluster_ids_sorted = cluster_ids[order_power]
    P_abs_sorted = P_abs[order_power]
    P_pct_sorted = P_pct[order_power]

    lines = []
    lines.append("Cluster power ranking (clustered only)\n\n")
    lines.append(f"name = {run_name}\n")
    lines.append(f"MIN_AMP = {min_amp}\n")
    lines.append(f"THR = {thr}\n")
    lines.append(f"WHITE_NOISE_THRESHOLD = {white_noise_threshold}\n")
    lines.append(f"WPS_NOISE_THRESHOLD = {wps_noise_threshold}\n\n")
    lines.append("rank  kcl      P_abs           P_pct\n")
    for r, (kcl, Pk, pk) in enumerate(zip(cluster_ids_sorted, P_abs_sorted, P_pct_sorted), start=1):
        lines.append(f"{r:4d}  {int(kcl):3d}  {Pk: .6e}   {pk:6.2f}%\n")
    (outdir / "04_cluster_power_ranking.txt").write_text("".join(lines))

    n_show = int(min(N_SHOW, cluster_ids_sorted.size))
    biggest = cluster_ids_sorted[:n_show]

    nrow = int(np.ceil(np.sqrt(n_show)))
    ncol = int(np.ceil(n_show / nrow))
    theta_vals = np.asarray(cwt["theta"], dtype=float)

    def plot_cluster_grid(kind: str, outname: str) -> None:
        fig, axs = plt.subplots(nrow, ncol, figsize=(12, 10))
        axs = np.atleast_1d(axs).reshape(-1)

        bg_vmin, bg_vmax = diverging_limits(bg, p_lo=2, p_hi=98)
        number_box = {"boxstyle": "square", "lw": 0.67, "facecolor": "white", "edgecolor": "black"}

        for ax, kcl in zip(axs, biggest):
            decomposition[:] = orig_decomp
            decomposition[iwave != kcl] = 0
            rec = transform.reconstruct2d(cwt)
            rec_centered = rec.T - np.median(rec)

            s_idx = int(idxs[kcl][0])
            t_idx = int(idxs[kcl][1])
            _, _, lam = effective_lambda(cwt, s_idx, t_idx)

            theta_dom_rad = float(theta_vals[t_idx])
            theta_dom_deg = (np.degrees(theta_dom_rad) + 180.0) % 180.0

            ii = int(np.where(cluster_ids_sorted == kcl)[0][0])
            Pk = float(P_abs_sorted[ii])
            pk = float(P_pct_sorted[ii])

            if kind == "standalone":
                ax.pcolormesh(x1d_km, y1d_km, rec_centered, shading="auto", cmap="RdBu_r", vmin=-1.0, vmax=1.0)
            elif kind == "overlay":
                ax.pcolormesh(x1d_km, y1d_km, bg, shading="auto", cmap="gray", vmin=bg_vmin, vmax=bg_vmax)
                ax.pcolormesh(x1d_km, y1d_km, rec_centered, shading="auto", cmap="RdBu_r", vmin=-1.0, vmax=1.0, alpha=0.6)
            else:
                raise ValueError(kind)

            ax.set_title(rf"$\lambda$={_fmt(lam)} km, $\theta$={theta_dom_deg:.0f}$^\circ$" + "\n" + rf"P={Pk:.2e} ({pk:.1f}%)")
            ax.set_aspect("equal", adjustable="box")

        for ax in axs[n_show:]:
            ax.axis("off")
        for ax in axs[:n_show]:
            ax.set_xlabel("x (km)")
            ax.set_ylabel("y (km)")

        for i, ax in enumerate(axs, start=1):
            if ax.has_data():
                ax.text(0.12, 0.15, str(i), transform=ax.transAxes, bbox=number_box, va="top", ha="right")

        fig.tight_layout()
        save_fig(fig, outdir / outname)

    plot_cluster_grid("standalone", "05_clusters_standalone.png")
    plot_cluster_grid("overlay", "06_clusters_overlay.png")

    for kcl in biggest:
        plot_cluster_individual_save(
            cwt=cwt,
            iwave=iwave,
            kcl=int(kcl),
            x1d_km=x1d_km,
            y1d_km=y1d_km,
            orig_decomp=orig_decomp,
            bg_xy=bg,
            dx_km=dx_km,
            dy_km=dy_km,
            outpath=outdir / f"07_cluster_{int(kcl):03d}_maps.png",
        )

    decomposition[:] = orig_decomp
    decomposition[iwave < 0] = 0
    rec_all = transform.reconstruct2d(cwt)  # (nx, ny)
    rec_all_xy = rec_all.T - np.median(rec_all)
    decomposition[:] = orig_decomp

    remaining_after_preproc_xy = proc
    vmin, vmax = diverging_limits(remaining_after_preproc_xy, wavy_xy, rec_all_xy, p_lo=2, p_hi=98)

    figR, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(18, 8), constrained_layout=True, sharex=True, sharey=True)

    im1 = ax1.pcolormesh(x1d_km, y1d_km, remaining_after_preproc_xy, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    ax1.set_title("Remaining (after preprocessing, input to CWT)")
    ax1.set_xlabel("x (km)")
    ax1.set_ylabel("y (km)")
    ax1.set_aspect("equal", adjustable="box")

    im2 = ax2.pcolormesh(x1d_km, y1d_km, wavy_xy, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    ax2.set_title("Wavy stuff (wavelet reconstruction after noise filtering)")
    ax2.set_xlabel("x (km)")
    ax2.set_ylabel("y (km)")
    ax2.set_aspect("equal", adjustable="box")

    im3 = ax3.pcolormesh(x1d_km, y1d_km, rec_all_xy, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    ax3.set_title("Reconstruction after clustering (all clusters combined)")
    ax3.set_xlabel("x (km)")
    ax3.set_ylabel("y (km)")
    ax3.set_aspect("equal", adjustable="box")

    cbar = figR.colorbar(im1, ax=[ax1, ax2, ax3], orientation="vertical", fraction=0.015, pad=0.01)
    cbar.set_label("arb")
    save_fig(figR, outdir / "09_remaining_vs_wavy_vs_allclusters.png")

    remaining_core_xy = crop_pad_core(remaining_after_preproc_xy)
    wavy_core_xy = crop_pad_core(wavy_xy)
    rec_all_core_xy = crop_pad_core(rec_all_xy)

    return {
        "has_clusters": True,
        "x_start_unbinned": core_start_unbinned,
        "dx_km": dx_km,
        "dy_km": dy_km,
        "remaining_xy": remaining_core_xy,
        "wavy_xy": wavy_core_xy,
        "rec_all_xy": rec_all_core_xy,
        "cluster_records": cluster_records,
    }


def stitch_and_save_final_09(slice_results: List[Dict[str, Any]], outroot: Path, tag: str) -> None:
    if not slice_results:
        return

    slice_results = sorted(slice_results, key=lambda d: int(d.get("x_start_unbinned", 0)))

    dx_km = float(slice_results[0]["dx_km"])
    dy_km = float(slice_results[0]["dy_km"])
    x0_unbinned = int(slice_results[0].get("x_start_unbinned", 0))
    x0_km = x0_unbinned * DX_KM_BASE

    remaining_stitched = np.concatenate([d["remaining_xy"] for d in slice_results], axis=1)
    wavy_stitched = np.concatenate([d["wavy_xy"] for d in slice_results], axis=1)
    rec_all_stitched = np.concatenate([d["rec_all_xy"] for d in slice_results], axis=1)

    ny, nx_tot = remaining_stitched.shape
    x1d_km = x0_km + np.arange(nx_tot, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    vmin, vmax = diverging_limits(remaining_stitched, wavy_stitched, rec_all_stitched, p_lo=2, p_hi=98)

    figR, (axT, axM, axB) = plt.subplots(3, 1, figsize=(18, 6), constrained_layout=True, sharex=True, sharey=True)

    imT = axT.pcolormesh(x1d_km, y1d_km, remaining_stitched, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    axT.set_title(f"STITCHED (core tiles) {tag}: Remaining (after preprocessing, input to CWT)")
    axT.set_xlabel("x (km)")
    axT.set_ylabel("y (km)")
    axT.set_aspect("equal", adjustable="box")

    imM = axM.pcolormesh(x1d_km, y1d_km, wavy_stitched, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    axM.set_title(f"STITCHED (core tiles) {tag}: Wavy stuff (wavelet reconstruction after noise filtering)")
    axM.set_xlabel("x (km)")
    axM.set_ylabel("y (km)")
    axM.set_aspect("equal", adjustable="box")

    imB = axB.pcolormesh(x1d_km, y1d_km, rec_all_stitched, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    axB.set_title(f"STITCHED (core tiles) {tag}: Reconstruction after clustering (all clusters combined)")
    axB.set_xlabel("x (km)")
    axB.set_ylabel("y (km)")
    axB.set_aspect("equal", adjustable="box")

    axT.set_xlim(600, 10000)

    cbar = figR.colorbar(imT, ax=[axT, axM, axB], orientation="vertical", fraction=0.012, pad=0.01)
    if PRODUCT_TYPE == "l3a":
        cbar.set_label("K (Relative)")
    else:
        cbar.set_label("kRayleigh (Relative)")
    save_fig(figR, outroot / f"FINAL_09_remaining_vs_wavy_vs_allclusters_STITCHED_{tag}.png")


def run_all_slices_single_offset(cfg: Dict[str, Any]) -> None:
    OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)

    rad2d = read_frame_auto(FILE_PATH, FRAME_IDX)
    rad2d_flat = flatten_swath(rad2d, target_ny=300)
    nx_total = rad2d_flat.shape[1]

    x_offset = int(X_OFFSET)
    if x_offset >= nx_total:
        raise ValueError(f"X_OFFSET={x_offset} too large for nx_total={nx_total}")

    run_name = str(cfg.get("name", "run"))
    outroot = OUTPUT_ROOT / f"{run_name}_offset_{x_offset:04d}"
    outroot.mkdir(parents=True, exist_ok=True)

    (outroot / "RUN_CONFIG.txt").write_text("\n".join([f"{k} = {v}" for k, v in cfg.items()]) + "\n")

    core_starts = list(range(x_offset, nx_total, X_CHUNK))
    stitch_results: List[Dict[str, Any]] = []
    all_cluster_records: List[Dict[str, Any]] = []

    f = int(BIN_FACTOR) if (APPLY_BINNING and int(BIN_FACTOR) > 1) else 1
    y_dim = int(rad2d_flat[Y_SLICE, :].shape[0] // f)
    x_dim = int(X_CHUNK // f)

    for slice_no, core_start in enumerate(core_starts):
        core_end = min(core_start + X_CHUNK, nx_total)

        ext_start = max(0, core_start - SLICE_PAD)
        ext_end = min(nx_total, core_end + SLICE_PAD)
        x_slice = slice(ext_start, ext_end)

        outdir = outroot / f"xcore_{core_start:05d}_{core_end:05d}_xext_{ext_start:05d}_{ext_end:05d}"
        print(f"[{run_name} | offset {x_offset}] core={core_start}:{core_end} ext={ext_start}:{ext_end} saved to {outdir}")

        res = run_pipeline_for_slice(
            rad2d_flat=rad2d_flat,
            x_slice=x_slice,
            outdir=outdir,
            x_start_unbinned=ext_start,
            core_start_unbinned=core_start,
            core_end_unbinned=core_end,
            cfg=cfg,
            slice_no=slice_no,
        )
        stitch_results.append(res)
        all_cluster_records.extend(res.get("cluster_records", []))

    stitch_and_save_final_09(stitch_results, outroot, tag=f"{run_name}_offset_{x_offset:04d}")
    print("Done. Outputs in", outroot.resolve())

    out_path = make_output_cluster_path(FILE_PATH)
    out_path = outroot / out_path.name

    copy_lxc_to_l6x_with_clusters(
        lxc_path=FILE_PATH,
        l6x_path=out_path,
        clusters=all_cluster_records,
        y_dim=y_dim,
        x_dim=x_dim,
        unbin_factor=(BIN_FACTOR if (APPLY_BINNING and int(BIN_FACTOR) > 1) else 1),
    )

    print("Saved output with clusters:", out_path.resolve())


for cfg in RUNS:
    run_all_slices_single_offset(cfg)
```

## Clean up memory, optional saving plots

```python
# %%
from __future__ import annotations

from pathlib import Path
from typing import Tuple, Dict, Any, List

import copy
import gc

import numpy as np
import matplotlib
import matplotlib.pyplot as plt
from netCDF4 import Dataset

from juwavelet import transform, utils, plot_utils

try:
    from scipy import stats
except Exception:
    stats = None


# --------------------------
# User params
# --------------------------
FILE_PATH = r"Z:\socfiles\l3a\2024\03_v23\2x2\awe_l3a_tmp_2024075T0538_01770_v23.nc" # can be L5C / L3C / L3A
FRAME_IDX = 0

# Slicing on the UNBINNED flattened field
X_CHUNK = 600
Y_SLICE = slice(None, None)
SLICE_PAD = 50  # extend each slice by this many UNBINNED pixels on each side

# Choose ONE slicing offset (no comparisons)
X_OFFSET = 0  # set to 0 or 300 or whatever you want

# Optional preprocessing
REMOVE_TREND = True
STANDARDIZE = False
GAUSS_BLUR_SIGMA = 0.0

# Optional 2x2 binning BEFORE processing
APPLY_BINNING = True
BIN_FACTOR = 2

# Edge taper on remaining field (before CWT)
APPLY_TAPER = True
TAPER_EDGE_PIXELS = 10  # in BINNED pixels if APPLY_BINNING else in original pixels

# Wavelet params
S0 = 16
DJ = 1 / 8
JS = 24
JT = 18
ASPECT = 1

# Pixel spacing (km per pixel)
DX_KM_BASE = 2.0
DY_KM_BASE = 2.0

# How many clusters to show (largest by POWER)
N_SHOW = 6

# Per cluster maps
AMP_MIN_FRACTION_OF_CLUSTER_MAX = 0.10
QUIVER_STEP_X = 20
QUIVER_STEP_Y = 20
QUIVER_SCALE = 1

# Red noise style filtering (robust WPS thresholding per scale)
APPLY_RED_NOISE_FILTER = True

# Optional diagnostics output from the red noise filter
SAVE_REDNOISE_DIAGNOSTICS = True

# Save settings
OUTPUT_ROOT = Path("outputs_decompose2d_slices_single_noise_filtering_AUTO/01770_l3a")
SAVE_DPI = 200

# NEW: plot saving switches
SAVE_SLICE_PLOTS = True        # saves everything inside each slice outdir
SAVE_STITCHED_PLOT = True      # saves the stitched plot at the run root


# --------------------------
# Auto detect product type
# --------------------------
file_name = Path(FILE_PATH).name.lower()
if "_l3a_" in file_name:
    PRODUCT_TYPE = "l3a"
elif "_l3c_" in file_name:
    PRODUCT_TYPE = "l3c"
elif "_l5c_" in file_name:
    PRODUCT_TYPE = "l5c"
else:
    raise ValueError(f"Cannot detect product type from file name: {FILE_PATH}")

print("Detected product type:", PRODUCT_TYPE)


# --------------------------
# Parameter runs (AUTO)
# --------------------------
if PRODUCT_TYPE == "l3a":
    RUNS: List[Dict[str, Any]] = [
        dict(
            name="base",
            MIN_AMP=1.25,
            THR=0.6,
            WHITE_NOISE_THRESHOLD=0.1,
            WPS_NOISE_THRESHOLD=8.0,
        ),
    ]
else:
    RUNS = [
        dict(
            name="base",
            MIN_AMP=0.2,
            THR=0.1,
            WHITE_NOISE_THRESHOLD=0.02,
            WPS_NOISE_THRESHOLD=8.0,
        ),
    ]

print("Using RUNS:", RUNS)


# --------------------------
# Plot style
# --------------------------
matplotlib.rcParams.update(
    {
        "axes.labelsize": 16,
        "font.size": 16,
        "xtick.labelsize": 14,
        "ytick.labelsize": 14,
        "legend.fontsize": 14,
        "figure.figsize": [2 * 6.94, 2 * 4.29],
    }
)


# --------------------------
# Small helpers
# --------------------------
def save_fig(fig: plt.Figure, outpath: Path, *, enable: bool = True) -> None:
    """
    Save and close a figure if enable=True.
    Always closes the figure to avoid matplotlib memory buildup.
    """
    try:
        if enable:
            outpath.parent.mkdir(parents=True, exist_ok=True)
            fig.savefig(outpath, dpi=SAVE_DPI, bbox_inches="tight")
    finally:
        plt.close(fig)


def cleanup_memory() -> None:
    """
    Best effort cleanup after each slice to keep RAM stable.
    """
    try:
        plt.close("all")
    except Exception:
        pass
    gc.collect()


def _fmt(v: float) -> str:
    if not np.isfinite(v):
        return "∞"
    v = float(v)
    if abs(v) >= 1000 or abs(v) < 0.1:
        return f"{v:.2e}"
    if abs(v) >= 100:
        return f"{v:.0f}"
    return f"{v:.1f}"


def diverging_limits(*arrays: np.ndarray, p_lo: float = 2, p_hi: float = 98) -> Tuple[float, float]:
    vals = []
    for A in arrays:
        if A is None:
            continue
        a = np.asarray(A)
        good = np.isfinite(a)
        if np.any(good):
            vals.append(a[good].ravel())
    if not vals:
        return (-1.0, 1.0)
    stacked = np.concatenate(vals)
    lo, hi = np.nanpercentile(stacked, [p_lo, p_hi])
    vmax = max(abs(lo), abs(hi))
    if not np.isfinite(vmax) or vmax == 0:
        vmax = 1.0
    return (-vmax, vmax)


# --------------------------
# Data helpers
# --------------------------
def read_frame_auto(path: str, frame_idx: int) -> np.ndarray:
    with Dataset(path, "r") as nc:
        if PRODUCT_TYPE == "l3a":
            varname = "Temperature"
        else:
            varname = "Radiance"
        data = nc.variables[varname][:]  # (t, y, x)

    if data.ndim != 3:
        raise ValueError(f"{varname} expected 3D (t,y,x); got {data.shape}")

    if not (0 <= frame_idx < data.shape[0]):
        raise IndexError(f"FRAME_IDX={frame_idx} out of range 0..{data.shape[0]-1}")

    return np.asarray(data[frame_idx, :, :], dtype=float)


def flatten_swath(Z: np.ndarray, target_ny: int = 300) -> np.ndarray:
    ny, nx = Z.shape
    out = np.full_like(Z, np.nan)

    bottoms = np.full(nx, np.nan, dtype=float)
    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size:
            bottoms[j] = valid_idx.max()

    valid_bottoms = bottoms[np.isfinite(bottoms)]
    if valid_bottoms.size == 0:
        med = float(np.nanmedian(Z)) if np.isfinite(np.nanmedian(Z)) else 0.0
        out = np.nan_to_num(Z, nan=med, posinf=med, neginf=med)
        if target_ny is not None and ny > target_ny:
            out = out[-target_ny:, :]
        return out

    max_bottom = int(valid_bottoms.max())

    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size == 0:
            continue
        bottom = int(valid_idx.max())
        shift = max_bottom - bottom
        if shift == 0:
            out[:, j] = col
        elif shift > 0:
            out[shift:, j] = col[: ny - shift]
        else:
            shift_up = -shift
            out[: ny - shift_up, j] = col[shift_up:]

    med = float(np.nanmedian(out)) if np.isfinite(np.nanmedian(out)) else 0.0
    out = np.nan_to_num(out, nan=med, posinf=med, neginf=med)

    if target_ny is not None and ny > target_ny:
        start = max_bottom + 1 - target_ny
        start = max(start, 0)
        end = start + target_ny
        out = out[start:end, :]

    return out


def bin2d_block_mean(Z: np.ndarray, factor: int) -> np.ndarray:
    if factor is None or int(factor) <= 1:
        return Z
    f = int(factor)
    ny, nx = Z.shape
    ny2 = (ny // f) * f
    nx2 = (nx // f) * f
    if ny2 == 0 or nx2 == 0:
        raise ValueError(f"Binning factor {f} too large for array {Z.shape}")
    Z2 = Z[:ny2, :nx2]
    return Z2.reshape(ny2 // f, f, nx2 // f, f).mean(axis=(1, 3))


def unbin_repeat_xy(A: np.ndarray, factor: int) -> np.ndarray:
    """
    Nearest neighbor upsample (repeat) from (y,x) to (y*factor, x*factor).
    Preserves NaNs by repetition.
    """
    if factor is None or int(factor) <= 1:
        return A
    f = int(factor)
    return np.repeat(np.repeat(A, f, axis=0), f, axis=1)


def fit_plane(Z: np.ndarray) -> np.ndarray:
    ny, nx = Z.shape
    yy, xx = np.mgrid[0:ny, 0:nx]
    G = np.stack([xx.ravel(), yy.ravel(), np.ones(nx * ny)], axis=1)
    coeff, *_ = np.linalg.lstsq(G, Z.ravel(), rcond=None)
    return coeff[0] * xx + coeff[1] * yy + coeff[2]


def extract_first_fourier_bg(Z: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
    ny, nx = Z.shape
    F = np.fft.fft2(Z)
    F_bg = np.zeros_like(F, dtype=complex)

    if nx > 1:
        F_bg[0, 1] = F[0, 1]
        F_bg[0, -1] = F[0, -1]
    if ny > 1:
        F_bg[1, 0] = F[1, 0]
        F_bg[-1, 0] = F[-1, 0]

    bg = np.fft.ifft2(F_bg).real
    return bg, Z - bg


def maybe_blur(Z: np.ndarray, sigma: float) -> np.ndarray:
    if sigma <= 0:
        return Z
    try:
        from scipy.ndimage import gaussian_filter
    except Exception:
        return Z
    return gaussian_filter(Z, sigma=float(sigma))


def _hanning_edge_1d(n: int, edge: int) -> np.ndarray:
    w = np.ones(n, dtype=float)
    if edge <= 0:
        return w
    edge = min(edge, n // 2)
    ramp = 0.5 * (1.0 - np.cos(np.linspace(0, np.pi, edge, endpoint=True)))
    w[:edge] = ramp
    w[-edge:] = ramp[::-1]
    return w


def apply_edge_hanning_taper(Z: np.ndarray, edge: int) -> np.ndarray:
    if edge <= 0:
        return Z
    ny, nx = Z.shape
    wx = _hanning_edge_1d(nx, edge)
    wy = _hanning_edge_1d(ny, edge)
    W = np.outer(wy, wx)
    return Z * W


def effective_lambda(cwt: dict, s_idx: int, t_idx: int) -> Tuple[float, float, float]:
    lamx = float(cwt["wavelength_x"][s_idx, t_idx])
    lamy = float(cwt["wavelength_y"][s_idx, t_idx])
    inv2 = 0.0
    if np.isfinite(lamx) and lamx != 0:
        inv2 += (1.0 / lamx) ** 2
    if np.isfinite(lamy) and lamy != 0:
        inv2 += (1.0 / lamy) ** 2
    lam = np.inf if inv2 == 0 else 1.0 / np.sqrt(inv2)
    return lamx, lamy, lam


def compute_cluster_power_table(cwt: dict, iwave: np.ndarray) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    dec = cwt["decomposition"]
    power4d = np.abs(dec) ** 2

    cluster_ids = np.unique(iwave[iwave >= 0]).astype(int)
    if cluster_ids.size == 0:
        return cluster_ids, np.array([]), np.array([])

    P_abs = np.zeros(cluster_ids.size, dtype=float)
    for i, kcl in enumerate(cluster_ids):
        P_abs[i] = float(np.sum(power4d[iwave == kcl]))

    total = float(np.sum(P_abs))
    if total > 0 and np.isfinite(total):
        P_pct = 100.0 * P_abs / total
    else:
        P_pct = np.full_like(P_abs, np.nan)

    return cluster_ids, P_abs, P_pct


def cluster_amp_theta_maps(
    cwt: dict,
    iwave: np.ndarray,
    kcl: int,
    amp_min_fraction_of_cluster_max: float = 0.2,
) -> Tuple[np.ndarray, np.ndarray]:
    dec = cwt["decomposition"]  # (ns, nt, nx, ny) complex
    power = np.abs(dec) ** 2

    mask4d = (iwave == kcl)
    if not np.any(mask4d):
        ny = dec.shape[3]
        nx = dec.shape[2]
        return np.full((ny, nx), np.nan), np.full((ny, nx), np.nan)

    ns, nt, nx, ny = power.shape
    power2 = power.reshape(ns * nt, nx * ny)
    mask2 = mask4d.reshape(ns * nt, nx * ny)

    p = np.where(mask2, power2, -np.inf)
    pmax = np.max(p, axis=0)

    valid = np.isfinite(pmax) & (pmax > 0)

    A = np.full(nx * ny, np.nan, dtype=float)
    theta = np.full(nx * ny, np.nan, dtype=float)

    if np.any(valid):
        arg = np.argmax(p[:, valid], axis=0)
        t_idx = arg % nt

        A_valid = np.sqrt(pmax[valid])
        theta_valid = np.asarray(cwt["theta"], dtype=float)[t_idx]

        A[valid] = A_valid
        theta[valid] = theta_valid

    A_xy = A.reshape(nx, ny).T
    theta_xy = theta.reshape(nx, ny).T

    if np.isfinite(np.nanmax(A_xy)):
        thr = float(amp_min_fraction_of_cluster_max) * float(np.nanmax(A_xy))
        low = A_xy < thr
        A_xy[low] = np.nan
        theta_xy[low] = np.nan

    return A_xy, theta_xy


def theta_to_uv(theta_xy: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
    return np.cos(theta_xy), np.sin(theta_xy)


# --------------------------
# Red noise style filtering helper
# --------------------------
def apply_red_noise_filter_cwt(
    cwt: dict,
    white_noise_threshold: float = 0.1,
    wps_noise_threshold: float = 8.0,
) -> Tuple[dict, Dict[str, Any]]:
    """
    Robust per scale WPS filtering.
    Returns filtered_cwt (deep copy) and a small diagnostics dict.
    """
    diag: Dict[str, Any] = {}
    if stats is None:
        diag["skipped"] = True
        diag["reason"] = "scipy not available"
        return cwt, diag

    cwt_copy = copy.deepcopy(cwt)
    dec = cwt_copy["decomposition"]  # (ns, nt, nx, ny) complex

    WPS0 = np.abs(dec) ** 2
    diag["n_coeff_total"] = int(WPS0.size)

    mask_1 = WPS0 < (float(white_noise_threshold) ** 2)
    dec[mask_1] = 0

    WPS1 = np.abs(dec) ** 2
    diag["n_zero_after_white"] = int(np.count_nonzero(WPS1 == 0))

    median_WPS = np.median(WPS1, axis=(1, 2, 3))  # (ns,)
    mad = stats.median_abs_deviation(WPS1, axis=(1, 2, 3), scale=1.0)  # (ns,)
    sMAD_WPS = 1.4826 * mad

    sMAD_safe = np.where(sMAD_WPS > 0, sMAD_WPS, np.nan)
    WPS_scaled = (WPS1 - median_WPS[:, None, None, None]) / sMAD_safe[:, None, None, None]

    mask = WPS_scaled < float(wps_noise_threshold)
    dec[mask] = 0

    WPS2 = np.abs(dec) ** 2
    diag["n_zero_after_scaled"] = int(np.count_nonzero(WPS2 == 0))

    finite_scaled = np.isfinite(WPS_scaled)

    diag["sMAD_min"] = float(np.nanmin(sMAD_WPS))
    diag["sMAD_median"] = float(np.nanmedian(sMAD_WPS))

    if np.any(finite_scaled):
        diag["WPS_scaled_min"] = float(np.nanmin(WPS_scaled[finite_scaled]))
        diag["WPS_scaled_median"] = float(np.nanmedian(WPS_scaled[finite_scaled]))
        diag["WPS_scaled_max"] = float(np.nanmax(WPS_scaled[finite_scaled]))
    else:
        diag["WPS_scaled_min"] = np.nan
        diag["WPS_scaled_median"] = np.nan
        diag["WPS_scaled_max"] = np.nan

    return cwt_copy, diag


# --------------------------
# Quicklook plots
# --------------------------
def quicklook_save(
    x1d: np.ndarray,
    y1d: np.ndarray,
    Z: np.ndarray,
    title: str,
    outpath: Path,
    *,
    save_plots: bool,
) -> None:
    fig, ax = plt.subplots()
    im = ax.pcolormesh(x1d, y1d, Z, shading="auto", cmap="gray")
    fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04).set_label("arb")
    ax.set_xlabel("x (km)")
    ax.set_ylabel("y (km)")
    ax.set_title(title)
    ax.set_aspect("equal", adjustable="box")
    fig.tight_layout()
    save_fig(fig, outpath, enable=save_plots)


def quicklook_background_decomp_4panel_save(
    x1d_km: np.ndarray,
    y1d_km: np.ndarray,
    original: np.ndarray,
    background: np.ndarray,
    remaining: np.ndarray,
    remaining_tapered: np.ndarray,
    taper_edge_px: int,
    outpath: Path,
    *,
    wavy_after_noise_xy: np.ndarray | None = None,
    save_plots: bool,
) -> None:
    med = np.nanmedian(original)
    orig_c = original - med
    bg_c = background - med
    rem_c = remaining
    rem_tapered_c = remaining_tapered

    arrays = [orig_c, bg_c, rem_c, rem_tapered_c]
    if wavy_after_noise_xy is not None:
        arrays.append(wavy_after_noise_xy)

    vmin, vmax = diverging_limits(*arrays, p_lo=1, p_hi=99)

    fig, axs = plt.subplots(2, 3, figsize=(16, 10), constrained_layout=True)
    axs = axs.flatten()

    panels = [
        (orig_c, "Original (median sub)"),
        (bg_c, "Background (plane + first Fourier)"),
        (rem_c, "Remaining = original - background"),
        (rem_tapered_c, f"Remaining with edge taper (width={taper_edge_px})"),
    ]

    if wavy_after_noise_xy is not None:
        panels.append((wavy_after_noise_xy, "Wavy stuff (wavelet reconstruction after noise filtering)"))

    im0 = None
    for ax, (Zp, title) in zip(axs, panels):
        im0 = ax.pcolormesh(x1d_km, y1d_km, Zp, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
        ax.set_title(title)
        ax.set_xlabel("x (km)")
        ax.set_ylabel("y (km)")
        ax.set_aspect("equal", adjustable="box")

    for ax in axs[len(panels) :]:
        ax.axis("off")

    cbar = fig.colorbar(im0, ax=axs[: len(panels)], orientation="vertical", fraction=0.04, pad=0.02)
    cbar.set_label("arb")
    save_fig(fig, outpath, enable=save_plots)


def plot_cluster_individual_save(
    cwt: dict,
    iwave: np.ndarray,
    kcl: int,
    x1d_km: np.ndarray,
    y1d_km: np.ndarray,
    orig_decomp: np.ndarray,
    bg_xy: np.ndarray,
    dx_km: float,
    dy_km: float,
    outpath: Path,
    *,
    save_plots: bool,
) -> None:
    decomposition = cwt["decomposition"]
    decomposition[:] = orig_decomp
    decomposition[iwave != kcl] = 0

    rec = transform.reconstruct2d(cwt)  # (nx, ny)
    rec_xy = rec.T - np.median(rec)

    A_xy, theta_xy = cluster_amp_theta_maps(
        cwt, iwave, kcl, amp_min_fraction_of_cluster_max=AMP_MIN_FRACTION_OF_CLUSTER_MAX
    )

    dec0 = orig_decomp
    power0 = np.abs(dec0) ** 2
    mask4d = (iwave == kcl)
    ns, nt, nx, ny = power0.shape

    power2 = power0.reshape(ns * nt, nx * ny)
    mask2 = mask4d.reshape(ns * nt, nx * ny)
    p = np.where(mask2, power2, -np.inf)
    pmax = np.max(p, axis=0)

    lam_xy = np.full(nx * ny, np.nan, dtype=float)
    valid = np.isfinite(pmax) & (pmax > 0)
    if np.any(valid):
        arg = np.argmax(p[:, valid], axis=0)
        s_idx = arg // nt
        t_idx = arg % nt

        lamx = np.asarray(cwt["wavelength_x"], dtype=float)[s_idx, t_idx]
        lamy = np.asarray(cwt["wavelength_y"], dtype=float)[s_idx, t_idx]

        inv2 = np.zeros_like(lamx, dtype=float)
        goodx = np.isfinite(lamx) & (lamx != 0)
        goody = np.isfinite(lamy) & (lamy != 0)
        inv2[goodx] += (1.0 / lamx[goodx]) ** 2
        inv2[goody] += (1.0 / lamy[goody]) ** 2

        lam = np.full_like(inv2, np.nan, dtype=float)
        ok = inv2 > 0
        lam[ok] = 1.0 / np.sqrt(inv2[ok])
        lam_xy[valid] = lam

    lam_xy = lam_xy.reshape(nx, ny).T

    if np.isfinite(np.nanmax(A_xy)):
        thr = float(AMP_MIN_FRACTION_OF_CLUSTER_MAX) * float(np.nanmax(A_xy))
        lam_xy[A_xy < thr] = np.nan

    bg_vmin, bg_vmax = diverging_limits(bg_xy, p_lo=2, p_hi=98)
    cl_vmin, cl_vmax = diverging_limits(rec_xy, p_lo=2, p_hi=98)

    theta_vals = np.asarray(cwt["theta"], dtype=float)
    theta_min = float(np.nanmin(theta_vals))
    theta_max = float(np.nanmax(theta_vals))

    if np.any(np.isfinite(lam_xy)):
        lam_vmin, lam_vmax = np.nanpercentile(lam_xy, [5, 95])
        if (not np.isfinite(lam_vmin)) or (not np.isfinite(lam_vmax)) or (lam_vmin == lam_vmax):
            lam_vmin = float(np.nanmin(lam_xy))
            lam_vmax = float(np.nanmax(lam_xy))
    else:
        lam_vmin, lam_vmax = 0.0, 1.0

    fig, axs = plt.subplots(2, 2, figsize=(16, 11), constrained_layout=True)
    ax00, ax01, ax10, ax11 = axs[0, 0], axs[0, 1], axs[1, 0], axs[1, 1]

    ax00.pcolormesh(x1d_km, y1d_km, bg_xy, shading="auto", cmap="gray", vmin=bg_vmin, vmax=bg_vmax)
    im00 = ax00.pcolormesh(
        x1d_km, y1d_km, rec_xy, shading="auto", cmap="RdBu_r", vmin=cl_vmin, vmax=cl_vmax, alpha=0.6
    )
    ax00.set_title(f"Cluster {kcl} overlay")
    ax00.set_xlabel("x (km)")
    ax00.set_ylabel("y (km)")
    ax00.set_aspect("equal", adjustable="box")
    fig.colorbar(im00, ax=ax00, fraction=0.046, pad=0.02)

    im01 = ax01.pcolormesh(x1d_km, y1d_km, A_xy, shading="auto", cmap="plasma")
    n_pix = int(np.count_nonzero(np.isfinite(A_xy)))
    area_km2 = n_pix * float(dx_km) * float(dy_km)
    ax01.set_title("A(x,y) = max |coeff| in cluster\n" + rf"Area = {area_km2:.0f} km$^2$ ({n_pix} px)")
    ax01.set_xlabel("x (km)")
    ax01.set_ylabel("y (km)")
    ax01.set_aspect("equal", adjustable="box")
    fig.colorbar(im01, ax=ax01, fraction=0.046, pad=0.02)

    im10 = ax10.pcolormesh(x1d_km, y1d_km, lam_xy, shading="auto", cmap="viridis", vmin=lam_vmin, vmax=lam_vmax)
    if np.any(np.isfinite(lam_xy)):
        lam_med = float(np.nanmedian(lam_xy))
        lam_p10, lam_p90 = np.nanpercentile(lam_xy, [10, 90])
        ax10.set_title(
            r"Dominant $\lambda(x,y)$ in cluster" + "\n"
            + rf"median={_fmt(lam_med)} km (10–90%: {_fmt(lam_p10)}–{_fmt(lam_p90)} km)"
        )
    else:
        ax10.set_title(r"Dominant $\lambda(x,y)$ in cluster")
    ax10.set_xlabel("x (km)")
    ax10.set_ylabel("y (km)")
    ax10.set_aspect("equal", adjustable="box")
    fig.colorbar(im10, ax=ax10, fraction=0.046, pad=0.02).set_label("km")

    im11 = ax11.pcolormesh(x1d_km, y1d_km, theta_xy, shading="auto", cmap="PiYG", vmin=theta_min, vmax=theta_max)
    ax11.set_title("θ(x,y) + quiver")
    ax11.set_xlabel("x (km)")
    ax11.set_ylabel("y (km)")
    ax11.set_aspect("equal", adjustable="box")
    fig.colorbar(im11, ax=ax11, fraction=0.046, pad=0.02)

    if np.any(np.isfinite(theta_xy)):
        u, v = theta_to_uv(theta_xy)
        arrow_len_km = 40.0
        u = u * arrow_len_km
        v = v * arrow_len_km

        Xq, Yq = np.meshgrid(x1d_km, y1d_km)
        Xqs = Xq[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        Yqs = Yq[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        uqs = u[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        vqs = v[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        good = np.isfinite(uqs) & np.isfinite(vqs)
        ax11.quiver(
            Xqs[good],
            Yqs[good],
            uqs[good],
            vqs[good],
            angles="xy",
            scale_units="xy",
            scale=QUIVER_SCALE,
            width=0.0025,
        )

    save_fig(fig, outpath, enable=save_plots)


# --------------------------
# Cluster products (for L6C/L6A outputs)
# --------------------------
def cluster_dominant_lambda_map(
    cwt: dict,
    iwave: np.ndarray,
    kcl: int,
    amp_mask_xy: np.ndarray | None = None,
) -> np.ndarray:
    """
    Per pixel dominant wavelength map λ(x,y) for one cluster.
    Returns shape (ny, nx) in XY.

    If amp_mask_xy is provided (A_xy with NaNs), apply the same NaN mask to λ.
    """
    dec = cwt["decomposition"]  # (ns, nt, nx, ny) complex
    power0 = np.abs(dec) ** 2
    mask4d = (iwave == kcl)
    if not np.any(mask4d):
        ny = dec.shape[3]
        nx = dec.shape[2]
        return np.full((ny, nx), np.nan)

    ns, nt, nx, ny = power0.shape
    power2 = power0.reshape(ns * nt, nx * ny)
    mask2 = mask4d.reshape(ns * nt, nx * ny)

    p = np.where(mask2, power2, -np.inf)
    pmax = np.max(p, axis=0)
    valid = np.isfinite(pmax) & (pmax > 0)

    lam_xy = np.full(nx * ny, np.nan, dtype=float)
    if np.any(valid):
        arg = np.argmax(p[:, valid], axis=0)
        s_idx = arg // nt
        t_idx = arg % nt

        lamx = np.asarray(cwt["wavelength_x"], dtype=float)[s_idx, t_idx]
        lamy = np.asarray(cwt["wavelength_y"], dtype=float)[s_idx, t_idx]

        inv2 = np.zeros_like(lamx, dtype=float)
        goodx = np.isfinite(lamx) & (lamx != 0)
        goody = np.isfinite(lamy) & (lamy != 0)
        inv2[goodx] += (1.0 / lamx[goodx]) ** 2
        inv2[goody] += (1.0 / lamy[goody]) ** 2

        lam = np.full_like(inv2, np.nan, dtype=float)
        ok = inv2 > 0
        lam[ok] = 1.0 / np.sqrt(inv2[ok])

        lam_xy[valid] = lam

    lam_xy = lam_xy.reshape(nx, ny).T  # XY

    if amp_mask_xy is not None:
        lam_xy[~np.isfinite(amp_mask_xy)] = np.nan

    return lam_xy


def copy_lxc_to_l6x_with_clusters(
    lxc_path: str | Path,
    l6x_path: str | Path,
    clusters: List[Dict[str, Any]],
    y_dim: int,
    x_dim: int,
    unbin_factor: int = 1,
) -> None:
    """
    Copy everything from input file, then append cluster products.

    Adds:
      dimension: cluster
      dimension: y_cluster (=y_dim*unbin_factor), x_cluster (=x_dim*unbin_factor)
      SlicesNo(cluster)
      Area(cluster)
      ClusterPowerAbs(cluster)
      ClusterPowerPct(cluster)
      Amplitude(cluster,y_cluster,x_cluster)
      DominantWavelength(cluster,y_cluster,x_cluster)
      Angle(cluster,y_cluster,x_cluster)
      ClusterReconstruction(cluster,y_cluster,x_cluster) optional

    If unbin_factor>1, upsample products by repetition.
    """
    def _unbin_repeat_xy_local(A: np.ndarray, factor: int) -> np.ndarray:
        if factor is None or int(factor) <= 1:
            return A
        f = int(factor)
        return np.repeat(np.repeat(A, f, axis=0), f, axis=1)

    lxc_path = Path(lxc_path)
    l6x_path = Path(l6x_path)
    l6x_path.parent.mkdir(parents=True, exist_ok=True)

    ncl = int(len(clusters))
    has_rec = any((c.get("rec_xy", None) is not None) for c in clusters)

    fup = int(unbin_factor) if (unbin_factor is not None and int(unbin_factor) > 1) else 1
    y_dim_w = int(y_dim) * fup
    x_dim_w = int(x_dim) * fup

    with Dataset(lxc_path, "r") as src, Dataset(l6x_path, "w", format="NETCDF4") as dst:
        dst.setncatts({k: src.getncattr(k) for k in src.ncattrs()})

        for dname, dim in src.dimensions.items():
            dst.createDimension(dname, (len(dim) if not dim.isunlimited() else None))

        for vname, var in src.variables.items():
            out = dst.createVariable(
                vname,
                var.datatype,
                var.dimensions,
                zlib=True,
                complevel=4,
                fill_value=getattr(var, "_FillValue", None),
            )
            out.setncatts({k: var.getncattr(k) for k in var.ncattrs()})
            out[:] = var[:]

        dst.createDimension("cluster", ncl)
        dst.createDimension("y_cluster", y_dim_w)
        dst.createDimension("x_cluster", x_dim_w)

        v_slice = dst.createVariable("SlicesNo", "i4", ("cluster",), zlib=True, complevel=4)
        v_area = dst.createVariable("Area", "f4", ("cluster",), zlib=True, complevel=4)
        v_pabs = dst.createVariable("ClusterPowerAbs", "f8", ("cluster",), zlib=True, complevel=4)
        v_ppct = dst.createVariable("ClusterPowerPct", "f4", ("cluster",), zlib=True, complevel=4)

        v_amp = dst.createVariable(
            "Amplitude",
            "f4",
            ("cluster", "y_cluster", "x_cluster"),
            zlib=True,
            complevel=4,
            fill_value=np.float32(np.nan),
        )
        v_lam = dst.createVariable(
            "DominantWavelength",
            "f4",
            ("cluster", "y_cluster", "x_cluster"),
            zlib=True,
            complevel=4,
            fill_value=np.float32(np.nan),
        )
        v_ang = dst.createVariable(
            "Angle",
            "f4",
            ("cluster", "y_cluster", "x_cluster"),
            zlib=True,
            complevel=4,
            fill_value=np.float32(np.nan),
        )

        v_rec = None
        if has_rec:
            v_rec = dst.createVariable(
                "ClusterReconstruction",
                "f4",
                ("cluster", "y_cluster", "x_cluster"),
                zlib=True,
                complevel=4,
                fill_value=np.float32(np.nan),
            )
            v_rec.units = "arb"
            v_rec.long_name = "Per cluster wavelet reconstruction map (relative; median subtracted)"

        v_area.units = "km2"
        v_lam.units = "km"
        v_ang.units = "rad"
        v_slice.long_name = "Slice index (starts at 0) for each cluster"
        v_pabs.units = "arb"
        v_ppct.units = "percent"

        if PRODUCT_TYPE == "l3a":
            v_amp.units = "K"
        else:
            v_amp.units = "kR"

        if ncl == 0:
            return

        v_slice[:] = np.asarray([c["slice_no"] for c in clusters], dtype=np.int32)
        v_area[:] = np.asarray([c["area_km2"] for c in clusters], dtype=np.float32)
        v_pabs[:] = np.asarray([c.get("power_abs", np.nan) for c in clusters], dtype=np.float64)
        v_ppct[:] = np.asarray([c.get("power_pct", np.nan) for c in clusters], dtype=np.float32)

        amp_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32)
        lam_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32)
        ang_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32)
        rec_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32) if has_rec else None

        def _force_shape(M: np.ndarray) -> np.ndarray:
            if M.shape == (y_dim_w, x_dim_w):
                return M.astype(np.float32)
            tmp = np.full((y_dim_w, x_dim_w), np.nan, dtype=np.float32)
            yy = min(y_dim_w, M.shape[0])
            xx = min(x_dim_w, M.shape[1])
            tmp[:yy, :xx] = M[:yy, :xx]
            return tmp

        for i, c in enumerate(clusters):
            A = _unbin_repeat_xy_local(c["A_xy"], fup)
            L = _unbin_repeat_xy_local(c["lam_xy"], fup)
            T = _unbin_repeat_xy_local(c["theta_xy"], fup)

            amp_stack[i] = _force_shape(A)
            lam_stack[i] = _force_shape(L)
            ang_stack[i] = _force_shape(T)

            if has_rec and rec_stack is not None and v_rec is not None:
                R = c.get("rec_xy", None)
                if R is not None:
                    R = _force_shape(_unbin_repeat_xy_local(R, fup))
                    rec_stack[i] = R

        v_amp[:] = amp_stack
        v_lam[:] = lam_stack
        v_ang[:] = ang_stack
        if has_rec and v_rec is not None and rec_stack is not None:
            v_rec[:] = rec_stack


def make_output_cluster_path(input_path: str | Path) -> Path:
    p = Path(input_path)
    name = p.name
    if "_l5c_" in name:
        name = name.replace("_l5c_", "_l6c_")
    elif "_l3c_" in name:
        name = name.replace("_l3c_", "_l6c_")
    elif "_l3a_" in name:
        name = name.replace("_l3a_", "_l6a_")
    else:
        name = p.stem + "_clusters.nc"
    return p.with_name(name)


# --------------------------
# Slice pipeline
# --------------------------
def run_pipeline_for_slice(
    rad2d_flat: np.ndarray,
    x_slice: slice,
    outdir: Path,
    x_start_unbinned: int,
    core_start_unbinned: int,
    core_end_unbinned: int,
    cfg: Dict[str, Any],
    slice_no: int,
) -> Dict[str, Any]:
    """
    Runs processing on an extended slice (x_slice), but returns only core tiles for stitching.

    Per slice plots are controlled by SAVE_SLICE_PLOTS.
    """
    outdir.mkdir(parents=True, exist_ok=True)
    save_plots = bool(SAVE_SLICE_PLOTS)

    min_amp = float(cfg.get("MIN_AMP"))
    thr = float(cfg.get("THR"))
    white_noise_threshold = float(cfg.get("WHITE_NOISE_THRESHOLD"))
    wps_noise_threshold = float(cfg.get("WPS_NOISE_THRESHOLD"))
    run_name = str(cfg.get("name", "run"))

    STORE_CLUSTER_RECON = True  # per cluster recon maps stored in output NetCDF

    if APPLY_BINNING and int(BIN_FACTOR) > 1:
        f = int(BIN_FACTOR)
        dx_km = DX_KM_BASE * f
        dy_km = DY_KM_BASE * f
    else:
        f = 1
        dx_km = DX_KM_BASE
        dy_km = DY_KM_BASE

    core_width_unbinned = int(core_end_unbinned - core_start_unbinned)
    core_w_binned = core_width_unbinned // f
    rel0_unbinned = int(core_start_unbinned - x_start_unbinned)
    rel0_binned = rel0_unbinned // f
    rel1_binned = rel0_binned + core_w_binned

    def crop_pad_core(A: np.ndarray) -> np.ndarray:
        nyA, nxA = A.shape
        left = max(0, rel0_binned)
        right = min(nxA, rel1_binned)
        core = A[:, left:right]
        need = core_w_binned - core.shape[1]
        if need > 0:
            core = np.concatenate([core, np.full((nyA, need), np.nan)], axis=1)
        elif need < 0:
            core = core[:, :core_w_binned]
        return core

    raw_win = np.asarray(rad2d_flat[Y_SLICE, x_slice], dtype=float)

    if APPLY_BINNING and int(BIN_FACTOR) > 1:
        raw_win = bin2d_block_mean(raw_win, BIN_FACTOR)

    ny, nx = raw_win.shape
    x1d_km = np.arange(nx, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    bin_tag = f" (binned {BIN_FACTOR}x{BIN_FACTOR})" if (APPLY_BINNING and int(BIN_FACTOR) > 1) else ""

    label0 = "Temperature" if PRODUCT_TYPE == "l3a" else "Radiance"

    quicklook_save(
        x1d_km,
        y1d_km,
        raw_win,
        f"{label0} (window, flattened){bin_tag}\n{run_name} MIN_AMP={min_amp} THR={thr}",
        outdir / "00_window.png",
        save_plots=save_plots,
    )

    proc0 = raw_win.astype(float, copy=True)

    if REMOVE_TREND:
        plane = fit_plane(proc0)
        detrended = proc0 - plane
    else:
        plane = np.zeros_like(proc0)
        detrended = proc0

    fourier_bg, _ = extract_first_fourier_bg(detrended)
    background = plane + fourier_bg
    remaining = proc0 - background

    if APPLY_TAPER:
        remaining_tapered = apply_edge_hanning_taper(remaining, TAPER_EDGE_PIXELS)
    else:
        remaining_tapered = remaining.copy()

    proc = remaining_tapered.copy()
    if STANDARDIZE:
        m = float(np.median(proc))
        s = float(np.std(proc))
        proc = (proc - m) / s if s > 0 else (proc - m)
    proc = maybe_blur(proc, GAUSS_BLUR_SIGMA)

    wave_xy = proc.T  # (nx, ny)

    cwt_raw = transform.decompose2d(
        wave_xy, dx=dx_km, dy=dy_km, s0=S0, dj=DJ, js=JS, jt=JT, aspect=ASPECT
    )

    if APPLY_RED_NOISE_FILTER:
        cwt, diag = apply_red_noise_filter_cwt(
            cwt_raw,
            white_noise_threshold=white_noise_threshold,
            wps_noise_threshold=wps_noise_threshold,
        )
        if SAVE_REDNOISE_DIAGNOSTICS:
            lines = []
            lines.append("Red noise filter diagnostics\n\n")
            lines.append(f"name                  = {run_name}\n")
            lines.append(f"WHITE_NOISE_THRESHOLD = {white_noise_threshold}\n")
            lines.append(f"WPS_NOISE_THRESHOLD   = {wps_noise_threshold}\n\n")
            if diag.get("skipped", False):
                lines.append(f"SKIPPED: {diag.get('reason','unknown')}\n")
            else:
                lines.append(f"n_coeff_total        = {diag.get('n_coeff_total')}\n")
                lines.append(f"n_zero_after_white   = {diag.get('n_zero_after_white')}\n")
                lines.append(f"n_zero_after_scaled  = {diag.get('n_zero_after_scaled')}\n")
                lines.append(f"sMAD_WPS min         = {diag.get('sMAD_min')}\n")
                lines.append(f"sMAD_WPS median      = {diag.get('sMAD_median')}\n")
                lines.append(f"WPS_scaled min       = {diag.get('WPS_scaled_min')}\n")
                lines.append(f"WPS_scaled median    = {diag.get('WPS_scaled_median')}\n")
                lines.append(f"WPS_scaled max       = {diag.get('WPS_scaled_max')}\n")
            (outdir / "02a_rednoise_diagnostics.txt").write_text("".join(lines))
    else:
        cwt = cwt_raw

    rec_wavy = transform.reconstruct2d(cwt)  # (nx, ny)
    wavy_xy = rec_wavy.T - np.median(rec_wavy)

    quicklook_background_decomp_4panel_save(
        x1d_km,
        y1d_km,
        proc0,
        background,
        remaining,
        remaining_tapered,
        TAPER_EDGE_PIXELS,
        outdir / "01_background_decomp_5panel.png",
        wavy_after_noise_xy=wavy_xy,
        save_plots=save_plots,
    )

    amps, idxs, iwave = utils.identify_cluster2d(cwt, min_amp=min_amp, thr=thr)
    decomposition = cwt["decomposition"]
    orig_decomp = decomposition.copy()

    cmap_cwt = matplotlib.cm.turbo
    norm_cwt = matplotlib.colors.BoundaryNorm(np.exp(np.linspace(np.log(0.05), np.log(1.0), 10)), cmap_cwt.N)
    fig_cwt, _ = utils.plot_decomposition2d(cwt, redux_s=2, redux_t=2, cmap=cmap_cwt, norm=norm_cwt)
    save_fig(fig_cwt, outdir / "02_decomposition_overview.png", enable=save_plots)

    plot_utils.plot_fov_wavelet_spectrum(cwt, stat="max")
    fig_spec = plt.gcf()
    save_fig(fig_spec, outdir / "03_fov_wavelet_spectrum.png", enable=save_plots)

    bg = remaining_tapered - np.median(remaining_tapered)

    cluster_ids, P_abs, P_pct = compute_cluster_power_table(cwt, iwave)
    if cluster_ids.size == 0:
        (outdir / "NO_CLUSTERS.txt").write_text(
            "No clusters found.\n\n"
            f"Run name: {run_name}\n"
            f"Used MIN_AMP={min_amp}, THR={thr}\n"
            f"Used WHITE_NOISE_THRESHOLD={white_noise_threshold}, WPS_NOISE_THRESHOLD={wps_noise_threshold}\n\n"
            "Try lowering MIN_AMP or THR.\n"
            "If red noise filter is on, try lowering WPS_NOISE_THRESHOLD or WHITE_NOISE_THRESHOLD.\n"
        )
        remaining_core_xy = crop_pad_core(proc)
        nan_core = np.full_like(remaining_core_xy, np.nan)

        cleanup_memory()
        return {
            "has_clusters": False,
            "x_start_unbinned": core_start_unbinned,
            "dx_km": dx_km,
            "dy_km": dy_km,
            "remaining_xy": remaining_core_xy,
            "wavy_xy": nan_core,
            "rec_all_xy": nan_core,
            "cluster_records": [],
        }

    power_abs_map = {int(k): float(p) for k, p in zip(cluster_ids, P_abs)}
    power_pct_map = {int(k): float(p) for k, p in zip(cluster_ids, P_pct)}

    cluster_records: List[Dict[str, Any]] = []
    for kcl in cluster_ids.astype(int):
        kcl = int(kcl)

        A_xy, theta_xy = cluster_amp_theta_maps(
            cwt,
            iwave,
            kcl,
            amp_min_fraction_of_cluster_max=AMP_MIN_FRACTION_OF_CLUSTER_MAX,
        )
        lam_xy = cluster_dominant_lambda_map(cwt, iwave, kcl, amp_mask_xy=A_xy)

        rec_core = None
        if STORE_CLUSTER_RECON:
            decomposition[:] = orig_decomp
            decomposition[iwave != kcl] = 0
            rec = transform.reconstruct2d(cwt)  # (nx, ny)
            rec_xy = rec.T
            rec_xy = rec_xy - np.nanmedian(rec_xy)
            rec_core = crop_pad_core(rec_xy)
            decomposition[:] = orig_decomp

        A_core = crop_pad_core(A_xy)
        th_core = crop_pad_core(theta_xy)
        lam_core = crop_pad_core(lam_xy)

        n_pix = int(np.count_nonzero(np.isfinite(A_core)))
        area_km2 = float(n_pix) * float(dx_km) * float(dy_km)

        cluster_records.append(
            dict(
                slice_no=int(slice_no),
                kcl=kcl,
                area_km2=area_km2,
                power_abs=power_abs_map.get(kcl, np.nan),
                power_pct=power_pct_map.get(kcl, np.nan),
                A_xy=A_core,
                lam_xy=lam_core,
                theta_xy=th_core,
                rec_xy=rec_core,
            )
        )

    order_power = np.argsort(P_abs)[::-1]
    cluster_ids_sorted = cluster_ids[order_power]
    P_abs_sorted = P_abs[order_power]
    P_pct_sorted = P_pct[order_power]

    lines = []
    lines.append("Cluster power ranking (clustered only)\n\n")
    lines.append(f"name = {run_name}\n")
    lines.append(f"MIN_AMP = {min_amp}\n")
    lines.append(f"THR = {thr}\n")
    lines.append(f"WHITE_NOISE_THRESHOLD = {white_noise_threshold}\n")
    lines.append(f"WPS_NOISE_THRESHOLD = {wps_noise_threshold}\n\n")
    lines.append("rank  kcl      P_abs           P_pct\n")
    for r, (kcl, Pk, pk) in enumerate(zip(cluster_ids_sorted, P_abs_sorted, P_pct_sorted), start=1):
        lines.append(f"{r:4d}  {int(kcl):3d}  {Pk: .6e}   {pk:6.2f}%\n")
    (outdir / "04_cluster_power_ranking.txt").write_text("".join(lines))

    n_show = int(min(N_SHOW, cluster_ids_sorted.size))
    biggest = cluster_ids_sorted[:n_show]

    nrow = int(np.ceil(np.sqrt(n_show)))
    ncol = int(np.ceil(n_show / nrow))
    theta_vals = np.asarray(cwt["theta"], dtype=float)

    def plot_cluster_grid(kind: str, outname: str) -> None:
        fig, axs = plt.subplots(nrow, ncol, figsize=(12, 10))
        axs = np.atleast_1d(axs).reshape(-1)

        bg_vmin, bg_vmax = diverging_limits(bg, p_lo=2, p_hi=98)
        number_box = {"boxstyle": "square", "lw": 0.67, "facecolor": "white", "edgecolor": "black"}

        for ax, kcl in zip(axs, biggest):
            decomposition[:] = orig_decomp
            decomposition[iwave != kcl] = 0
            rec = transform.reconstruct2d(cwt)
            rec_centered = rec.T - np.median(rec)

            s_idx = int(idxs[kcl][0])
            t_idx = int(idxs[kcl][1])
            _, _, lam = effective_lambda(cwt, s_idx, t_idx)

            theta_dom_rad = float(theta_vals[t_idx])
            theta_dom_deg = (np.degrees(theta_dom_rad) + 180.0) % 180.0

            ii = int(np.where(cluster_ids_sorted == kcl)[0][0])
            Pk = float(P_abs_sorted[ii])
            pk = float(P_pct_sorted[ii])

            if kind == "standalone":
                ax.pcolormesh(x1d_km, y1d_km, rec_centered, shading="auto", cmap="RdBu_r", vmin=-1.0, vmax=1.0)
            elif kind == "overlay":
                ax.pcolormesh(x1d_km, y1d_km, bg, shading="auto", cmap="gray", vmin=bg_vmin, vmax=bg_vmax)
                ax.pcolormesh(x1d_km, y1d_km, rec_centered, shading="auto", cmap="RdBu_r", vmin=-1.0, vmax=1.0, alpha=0.6)
            else:
                raise ValueError(kind)

            ax.set_title(
                rf"$\lambda$={_fmt(lam)} km, $\theta$={theta_dom_deg:.0f}$^\circ$"
                + "\n"
                + rf"P={Pk:.2e} ({pk:.1f}%)"
            )
            ax.set_aspect("equal", adjustable="box")

        for ax in axs[n_show:]:
            ax.axis("off")
        for ax in axs[:n_show]:
            ax.set_xlabel("x (km)")
            ax.set_ylabel("y (km)")

        for i, ax in enumerate(axs, start=1):
            if ax.has_data():
                ax.text(0.12, 0.15, str(i), transform=ax.transAxes, bbox=number_box, va="top", ha="right")

        fig.tight_layout()
        save_fig(fig, outdir / outname, enable=save_plots)

    plot_cluster_grid("standalone", "05_clusters_standalone.png")
    plot_cluster_grid("overlay", "06_clusters_overlay.png")

    for kcl in biggest:
        plot_cluster_individual_save(
            cwt=cwt,
            iwave=iwave,
            kcl=int(kcl),
            x1d_km=x1d_km,
            y1d_km=y1d_km,
            orig_decomp=orig_decomp,
            bg_xy=bg,
            dx_km=dx_km,
            dy_km=dy_km,
            outpath=outdir / f"07_cluster_{int(kcl):03d}_maps.png",
            save_plots=save_plots,
        )

    decomposition[:] = orig_decomp
    decomposition[iwave < 0] = 0
    rec_all = transform.reconstruct2d(cwt)  # (nx, ny)
    rec_all_xy = rec_all.T - np.median(rec_all)
    decomposition[:] = orig_decomp

    remaining_after_preproc_xy = proc
    vmin, vmax = diverging_limits(remaining_after_preproc_xy, wavy_xy, rec_all_xy, p_lo=2, p_hi=98)

    figR, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(18, 8), constrained_layout=True, sharex=True, sharey=True)

    im1 = ax1.pcolormesh(x1d_km, y1d_km, remaining_after_preproc_xy, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    ax1.set_title("Remaining (after preprocessing, input to CWT)")
    ax1.set_xlabel("x (km)")
    ax1.set_ylabel("y (km)")
    ax1.set_aspect("equal", adjustable="box")

    im2 = ax2.pcolormesh(x1d_km, y1d_km, wavy_xy, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    ax2.set_title("Wavy stuff (wavelet reconstruction after noise filtering)")
    ax2.set_xlabel("x (km)")
    ax2.set_ylabel("y (km)")
    ax2.set_aspect("equal", adjustable="box")

    im3 = ax3.pcolormesh(x1d_km, y1d_km, rec_all_xy, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    ax3.set_title("Reconstruction after clustering (all clusters combined)")
    ax3.set_xlabel("x (km)")
    ax3.set_ylabel("y (km)")
    ax3.set_aspect("equal", adjustable="box")

    cbar = figR.colorbar(im1, ax=[ax1, ax2, ax3], orientation="vertical", fraction=0.015, pad=0.01)
    cbar.set_label("arb")
    save_fig(figR, outdir / "09_remaining_vs_wavy_vs_allclusters.png", enable=save_plots)

    remaining_core_xy = crop_pad_core(remaining_after_preproc_xy)
    wavy_core_xy = crop_pad_core(wavy_xy)
    rec_all_core_xy = crop_pad_core(rec_all_xy)

    # Free big intermediates before returning
    try:
        del raw_win, proc0, plane, detrended, fourier_bg, background, remaining, remaining_tapered, proc, wave_xy
        del cwt_raw, cwt, amps, idxs, iwave, decomposition, orig_decomp, rec_wavy, rec_all
        del bg, cluster_ids, P_abs, P_pct, cluster_ids_sorted, P_abs_sorted, P_pct_sorted
    except Exception:
        pass
    cleanup_memory()

    return {
        "has_clusters": True,
        "x_start_unbinned": core_start_unbinned,
        "dx_km": dx_km,
        "dy_km": dy_km,
        "remaining_xy": remaining_core_xy,
        "wavy_xy": wavy_core_xy,
        "rec_all_xy": rec_all_core_xy,
        "cluster_records": cluster_records,
    }


def stitch_and_save_final_09(slice_results: List[Dict[str, Any]], outroot: Path, tag: str) -> None:
    if not slice_results:
        return

    slice_results = sorted(slice_results, key=lambda d: int(d.get("x_start_unbinned", 0)))

    dx_km = float(slice_results[0]["dx_km"])
    dy_km = float(slice_results[0]["dy_km"])
    x0_unbinned = int(slice_results[0].get("x_start_unbinned", 0))
    x0_km = x0_unbinned * DX_KM_BASE

    remaining_stitched = np.concatenate([d["remaining_xy"] for d in slice_results], axis=1)
    wavy_stitched = np.concatenate([d["wavy_xy"] for d in slice_results], axis=1)
    rec_all_stitched = np.concatenate([d["rec_all_xy"] for d in slice_results], axis=1)

    ny, nx_tot = remaining_stitched.shape
    x1d_km = x0_km + np.arange(nx_tot, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    vmin, vmax = diverging_limits(remaining_stitched, wavy_stitched, rec_all_stitched, p_lo=2, p_hi=98)

    figR, (axT, axM, axB) = plt.subplots(3, 1, figsize=(18, 6), constrained_layout=True, sharex=True, sharey=True)

    imT = axT.pcolormesh(x1d_km, y1d_km, remaining_stitched, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    axT.set_title(f"STITCHED (core tiles) {tag}: Remaining (after preprocessing, input to CWT)")
    axT.set_xlabel("x (km)")
    axT.set_ylabel("y (km)")
    axT.set_aspect("equal", adjustable="box")

    imM = axM.pcolormesh(x1d_km, y1d_km, wavy_stitched, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    axM.set_title(f"STITCHED (core tiles) {tag}: Wavy stuff (wavelet reconstruction after noise filtering)")
    axM.set_xlabel("x (km)")
    axM.set_ylabel("y (km)")
    axM.set_aspect("equal", adjustable="box")

    imB = axB.pcolormesh(x1d_km, y1d_km, rec_all_stitched, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    axB.set_title(f"STITCHED (core tiles) {tag}: Reconstruction after clustering (all clusters combined)")
    axB.set_xlabel("x (km)")
    axB.set_ylabel("y (km)")
    axB.set_aspect("equal", adjustable="box")

    # keep your xlim choice
    axT.set_xlim(600, 10000)

    cbar = figR.colorbar(imT, ax=[axT, axM, axB], orientation="vertical", fraction=0.012, pad=0.01)
    if PRODUCT_TYPE == "l3a":
        cbar.set_label("K (Relative)")
    else:
        cbar.set_label("kRayleigh (Relative)")

    save_fig(figR, outroot / f"FINAL_09_remaining_vs_wavy_vs_allclusters_STITCHED_{tag}.png", enable=bool(SAVE_STITCHED_PLOT))

    # stitched arrays can be large too
    try:
        del remaining_stitched, wavy_stitched, rec_all_stitched
    except Exception:
        pass
    cleanup_memory()


def run_all_slices_single_offset(cfg: Dict[str, Any]) -> None:
    OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)

    rad2d = read_frame_auto(FILE_PATH, FRAME_IDX)
    rad2d_flat = flatten_swath(rad2d, target_ny=300)
    nx_total = rad2d_flat.shape[1]

    x_offset = int(X_OFFSET)
    if x_offset >= nx_total:
        raise ValueError(f"X_OFFSET={x_offset} too large for nx_total={nx_total}")

    run_name = str(cfg.get("name", "run"))
    outroot = OUTPUT_ROOT / f"{run_name}_offset_{x_offset:04d}"
    outroot.mkdir(parents=True, exist_ok=True)

    (outroot / "RUN_CONFIG.txt").write_text("\n".join([f"{k} = {v}" for k, v in cfg.items()]) + "\n")

    core_starts = list(range(x_offset, nx_total, X_CHUNK))
    stitch_results: List[Dict[str, Any]] = []
    all_cluster_records: List[Dict[str, Any]] = []

    f = int(BIN_FACTOR) if (APPLY_BINNING and int(BIN_FACTOR) > 1) else 1
    y_dim = int(rad2d_flat[Y_SLICE, :].shape[0] // f)
    x_dim = int(X_CHUNK // f)

    for slice_no, core_start in enumerate(core_starts):
        core_end = min(core_start + X_CHUNK, nx_total)

        ext_start = max(0, core_start - SLICE_PAD)
        ext_end = min(nx_total, core_end + SLICE_PAD)
        x_slice = slice(ext_start, ext_end)

        outdir = outroot / f"xcore_{core_start:05d}_{core_end:05d}_xext_{ext_start:05d}_{ext_end:05d}"
        print(f"[{run_name} | offset {x_offset}] core={core_start}:{core_end} ext={ext_start}:{ext_end} saved to {outdir}")

        res = run_pipeline_for_slice(
            rad2d_flat=rad2d_flat,
            x_slice=x_slice,
            outdir=outdir,
            x_start_unbinned=ext_start,
            core_start_unbinned=core_start,
            core_end_unbinned=core_end,
            cfg=cfg,
            slice_no=slice_no,
        )

        stitch_results.append(res)
        all_cluster_records.extend(res.get("cluster_records", []))

        # slice level cleanup
        cleanup_memory()

    stitch_and_save_final_09(stitch_results, outroot, tag=f"{run_name}_offset_{x_offset:04d}")
    print("Done. Outputs in", outroot.resolve())

    out_path = make_output_cluster_path(FILE_PATH)
    out_path = outroot / out_path.name

    copy_lxc_to_l6x_with_clusters(
        lxc_path=FILE_PATH,
        l6x_path=out_path,
        clusters=all_cluster_records,
        y_dim=y_dim,
        x_dim=x_dim,
        unbin_factor=(BIN_FACTOR if (APPLY_BINNING and int(BIN_FACTOR) > 1) else 1),
    )

    print("Saved output with clusters:", out_path.resolve())

    # final cleanup
    try:
        del rad2d, rad2d_flat, stitch_results, all_cluster_records
    except Exception:
        pass
    cleanup_memory()


for cfg in RUNS:
    run_all_slices_single_offset(cfg)
```

## Batch processing

```python
# %%
from __future__ import annotations

from pathlib import Path
from typing import Tuple, Dict, Any, List

import gc
import copy

import numpy as np
import matplotlib
import matplotlib.pyplot as plt
from netCDF4 import Dataset

from juwavelet import transform, utils, plot_utils

try:
    from scipy import stats
except Exception:
    stats = None


# ==========================
# CHANGES (summary)
# ==========================
# 1) True plotting disable
#    All plotting functions now return immediately when saving is off
#    This skips creating figures entirely (big speedup)
#
# 2) Stitched plot saving can be fully disabled
#    stitch_and_save_final_09 returns immediately if SAVE_STITCHED_PLOT is False
#
# 3) Red noise diagnostics writing can be disabled
#    No per slice text output when SAVE_REDNOISE_DIAGNOSTICS is False
#
# 4) Red noise filter no longer deep copies the entire cwt dict
#    It shallow copies the dict and copies only decomposition (much faster, less memory)
#
# 5) Fixed batch loop bug
#    continuelo -> continue


# --------------------------
# User params
# --------------------------
INPUT_FOLDER = Path("l3/OneFile")
FILE_GLOB = "*.nc"

OUTPUT_ROOT = Path("outputs_decompose2d_slices_single_noise_filtering_AUTO_BATCH")
SAVE_DPI = 200

OUTPUT_L6_FOLDER = Path("l6/OneFile")

FRAME_IDX = 0

X_CHUNK = 600
Y_SLICE = slice(None, None)
SLICE_PAD = 50

X_OFFSET = 0

REMOVE_TREND = True
STANDARDIZE = False
GAUSS_BLUR_SIGMA = 0.0

APPLY_BINNING = True
BIN_FACTOR = 2

APPLY_TAPER = True
TAPER_EDGE_PIXELS = 10

S0 = 16
DJ = 1 / 8
JS = 24
JT = 18
ASPECT = 1

DX_KM_BASE = 2.0
DY_KM_BASE = 2.0

N_SHOW = 6

AMP_MIN_FRACTION_OF_CLUSTER_MAX = 0.10
QUIVER_STEP_X = 20
QUIVER_STEP_Y = 20
QUIVER_SCALE = 1

APPLY_RED_NOISE_FILTER = True

# Turn these on only if you actually want files written
SAVE_REDNOISE_DIAGNOSTICS = False
SAVE_SLICE_PLOTS = False
SAVE_STITCHED_PLOT = True


# --------------------------
# Plot style
# --------------------------
matplotlib.rcParams.update(
    {
        "axes.labelsize": 16,
        "font.size": 16,
        "xtick.labelsize": 14,
        "ytick.labelsize": 14,
        "legend.fontsize": 14,
        "figure.figsize": [2 * 6.94, 2 * 4.29],
    }
)


# --------------------------
# Small helpers
# --------------------------
def save_fig(fig: plt.Figure, outpath: Path, *, enable: bool = True) -> None:
    """
    Save and close a figure if enable=True.
    Always closes the figure to avoid matplotlib memory buildup.
    """
    try:
        if enable:
            outpath.parent.mkdir(parents=True, exist_ok=True)
            fig.savefig(outpath, dpi=SAVE_DPI, bbox_inches="tight")
    finally:
        plt.close(fig)


def cleanup_memory() -> None:
    try:
        plt.close("all")
    except Exception:
        pass
    gc.collect()


def _fmt(v: float) -> str:
    if not np.isfinite(v):
        return "∞"
    v = float(v)
    if abs(v) >= 1000 or abs(v) < 0.1:
        return f"{v:.2e}"
    if abs(v) >= 100:
        return f"{v:.0f}"
    return f"{v:.1f}"


def diverging_limits(*arrays: np.ndarray, p_lo: float = 2, p_hi: float = 98) -> Tuple[float, float]:
    vals = []
    for A in arrays:
        if A is None:
            continue
        a = np.asarray(A)
        good = np.isfinite(a)
        if np.any(good):
            vals.append(a[good].ravel())
    if not vals:
        return (-1.0, 1.0)
    stacked = np.concatenate(vals)
    lo, hi = np.nanpercentile(stacked, [p_lo, p_hi])
    vmax = max(abs(lo), abs(hi))
    if not np.isfinite(vmax) or vmax == 0:
        vmax = 1.0
    return (-vmax, vmax)


# --------------------------
# Product detection + RUNS
# --------------------------
def detect_product_type(file_path: Path) -> str:
    name = file_path.name.lower()
    if "_l3a_" in name:
        return "l3a"
    if "_l3c_" in name:
        return "l3c"
    if "_l5c_" in name:
        return "l5c"
    raise ValueError(f"Cannot detect product type from file name: {file_path.name}")


def get_runs_for_product(product_type: str) -> List[Dict[str, Any]]:
    if product_type == "l3a":
        return [
            dict(
                name="base",
                MIN_AMP=1.25,
                THR=0.6,
                WHITE_NOISE_THRESHOLD=0.1,
                WPS_NOISE_THRESHOLD=8.0,
            ),
        ]
    return [
        dict(
            name="base",
            MIN_AMP=0.2,
            THR=0.15,
            WHITE_NOISE_THRESHOLD=0.02,
            WPS_NOISE_THRESHOLD=8.0,
        ),
    ]


# --------------------------
# Data helpers
# --------------------------
def read_frame(path: str | Path, frame_idx: int, product_type: str) -> np.ndarray:
    with Dataset(str(path), "r") as nc:
        varname = "Temperature" if product_type == "l3a" else "Radiance"
        data = nc.variables[varname][:]  # (t, y, x)

    if data.ndim != 3:
        raise ValueError(f"{varname} expected 3D (t,y,x); got {data.shape}")

    if not (0 <= frame_idx < data.shape[0]):
        raise IndexError(f"FRAME_IDX={frame_idx} out of range 0..{data.shape[0]-1}")

    return np.asarray(data[frame_idx, :, :], dtype=float)


def flatten_swath(Z: np.ndarray, target_ny: int = 300) -> np.ndarray:
    ny, nx = Z.shape
    out = np.full_like(Z, np.nan)

    bottoms = np.full(nx, np.nan, dtype=float)
    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size:
            bottoms[j] = valid_idx.max()

    valid_bottoms = bottoms[np.isfinite(bottoms)]
    if valid_bottoms.size == 0:
        med = float(np.nanmedian(Z)) if np.isfinite(np.nanmedian(Z)) else 0.0
        out = np.nan_to_num(Z, nan=med, posinf=med, neginf=med)
        if target_ny is not None and ny > target_ny:
            out = out[-target_ny:, :]
        return out

    max_bottom = int(valid_bottoms.max())

    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size == 0:
            continue
        bottom = int(valid_idx.max())
        shift = max_bottom - bottom
        if shift == 0:
            out[:, j] = col
        elif shift > 0:
            out[shift:, j] = col[: ny - shift]
        else:
            shift_up = -shift
            out[: ny - shift_up, j] = col[shift_up:]

    med = float(np.nanmedian(out)) if np.isfinite(np.nanmedian(out)) else 0.0
    out = np.nan_to_num(out, nan=med, posinf=med, neginf=med)

    if target_ny is not None and ny > target_ny:
        start = max_bottom + 1 - target_ny
        start = max(start, 0)
        end = start + target_ny
        out = out[start:end, :]

    return out


def bin2d_block_mean(Z: np.ndarray, factor: int) -> np.ndarray:
    if factor is None or int(factor) <= 1:
        return Z
    f = int(factor)
    ny, nx = Z.shape
    ny2 = (ny // f) * f
    nx2 = (nx // f) * f
    if ny2 == 0 or nx2 == 0:
        raise ValueError(f"Binning factor {f} too large for array {Z.shape}")
    Z2 = Z[:ny2, :nx2]
    return Z2.reshape(ny2 // f, f, nx2 // f, f).mean(axis=(1, 3))


def unbin_repeat_xy(A: np.ndarray, factor: int) -> np.ndarray:
    if factor is None or int(factor) <= 1:
        return A
    f = int(factor)
    return np.repeat(np.repeat(A, f, axis=0), f, axis=1)


def fit_plane(Z: np.ndarray) -> np.ndarray:
    ny, nx = Z.shape
    yy, xx = np.mgrid[0:ny, 0:nx]
    G = np.stack([xx.ravel(), yy.ravel(), np.ones(nx * ny)], axis=1)
    coeff, *_ = np.linalg.lstsq(G, Z.ravel(), rcond=None)
    return coeff[0] * xx + coeff[1] * yy + coeff[2]


def extract_first_fourier_bg(Z: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
    ny, nx = Z.shape
    F = np.fft.fft2(Z)
    F_bg = np.zeros_like(F, dtype=complex)

    if nx > 1:
        F_bg[0, 1] = F[0, 1]
        F_bg[0, -1] = F[0, -1]
    if ny > 1:
        F_bg[1, 0] = F[1, 0]
        F_bg[-1, 0] = F[-1, 0]

    bg = np.fft.ifft2(F_bg).real
    return bg, Z - bg


def maybe_blur(Z: np.ndarray, sigma: float) -> np.ndarray:
    if sigma <= 0:
        return Z
    try:
        from scipy.ndimage import gaussian_filter
    except Exception:
        return Z
    return gaussian_filter(Z, sigma=float(sigma))


def _hanning_edge_1d(n: int, edge: int) -> np.ndarray:
    w = np.ones(n, dtype=float)
    if edge <= 0:
        return w
    edge = min(edge, n // 2)
    ramp = 0.5 * (1.0 - np.cos(np.linspace(0, np.pi, edge, endpoint=True)))
    w[:edge] = ramp
    w[-edge:] = ramp[::-1]
    return w


def apply_edge_hanning_taper(Z: np.ndarray, edge: int) -> np.ndarray:
    if edge <= 0:
        return Z
    ny, nx = Z.shape
    wx = _hanning_edge_1d(nx, edge)
    wy = _hanning_edge_1d(ny, edge)
    W = np.outer(wy, wx)
    return Z * W


def effective_lambda(cwt: dict, s_idx: int, t_idx: int) -> Tuple[float, float, float]:
    lamx = float(cwt["wavelength_x"][s_idx, t_idx])
    lamy = float(cwt["wavelength_y"][s_idx, t_idx])
    inv2 = 0.0
    if np.isfinite(lamx) and lamx != 0:
        inv2 += (1.0 / lamx) ** 2
    if np.isfinite(lamy) and lamy != 0:
        inv2 += (1.0 / lamy) ** 2
    lam = np.inf if inv2 == 0 else 1.0 / np.sqrt(inv2)
    return lamx, lamy, lam


def compute_cluster_power_table(cwt: dict, iwave: np.ndarray) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    dec = cwt["decomposition"]
    power4d = np.abs(dec) ** 2

    cluster_ids = np.unique(iwave[iwave >= 0]).astype(int)
    if cluster_ids.size == 0:
        return cluster_ids, np.array([]), np.array([])

    P_abs = np.zeros(cluster_ids.size, dtype=float)
    for i, kcl in enumerate(cluster_ids):
        P_abs[i] = float(np.sum(power4d[iwave == kcl]))

    total = float(np.sum(P_abs))
    if total > 0 and np.isfinite(total):
        P_pct = 100.0 * P_abs / total
    else:
        P_pct = np.full_like(P_abs, np.nan)

    return cluster_ids, P_abs, P_pct


def cluster_amp_theta_maps(
    cwt: dict,
    iwave: np.ndarray,
    kcl: int,
    amp_min_fraction_of_cluster_max: float = 0.2,
) -> Tuple[np.ndarray, np.ndarray]:
    dec = cwt["decomposition"]  # (ns, nt, nx, ny) complex
    power = np.abs(dec) ** 2

    mask4d = (iwave == kcl)
    if not np.any(mask4d):
        ny = dec.shape[3]
        nx = dec.shape[2]
        return np.full((ny, nx), np.nan), np.full((ny, nx), np.nan)

    ns, nt, nx, ny = power.shape
    power2 = power.reshape(ns * nt, nx * ny)
    mask2 = mask4d.reshape(ns * nt, nx * ny)

    p = np.where(mask2, power2, -np.inf)
    pmax = np.max(p, axis=0)

    valid = np.isfinite(pmax) & (pmax > 0)

    A = np.full(nx * ny, np.nan, dtype=float)
    theta = np.full(nx * ny, np.nan, dtype=float)

    if np.any(valid):
        arg = np.argmax(p[:, valid], axis=0)
        t_idx = arg % nt

        A_valid = np.sqrt(pmax[valid])
        theta_valid = np.asarray(cwt["theta"], dtype=float)[t_idx]

        A[valid] = A_valid
        theta[valid] = theta_valid

    A_xy = A.reshape(nx, ny).T
    theta_xy = theta.reshape(nx, ny).T

    if np.isfinite(np.nanmax(A_xy)):
        thr = float(amp_min_fraction_of_cluster_max) * float(np.nanmax(A_xy))
        low = A_xy < thr
        A_xy[low] = np.nan
        theta_xy[low] = np.nan

    return A_xy, theta_xy


def theta_to_uv(theta_xy: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
    return np.cos(theta_xy), np.sin(theta_xy)


# --------------------------
# Red noise style filtering helper
# --------------------------
def apply_red_noise_filter_cwt(
    cwt: dict,
    white_noise_threshold: float = 0.1,
    wps_noise_threshold: float = 8.0,
) -> Tuple[dict, Dict[str, Any]]:
    """
    Robust per scale WPS filtering.
    Returns filtered_cwt and diagnostics.

    Change: no deepcopy of entire cwt dict.
    Only decomposition is copied since that is what we mutate.
    """
    diag: Dict[str, Any] = {}
    if stats is None:
        diag["skipped"] = True
        diag["reason"] = "scipy not available"
        return cwt, diag

    # Shallow copy of dict and copy only decomposition
    cwt_copy = dict(cwt)
    dec0 = cwt["decomposition"]
    dec = dec0.copy()
    cwt_copy["decomposition"] = dec

    WPS0 = np.abs(dec) ** 2
    diag["n_coeff_total"] = int(WPS0.size)

    mask_1 = WPS0 < (float(white_noise_threshold) ** 2)
    dec[mask_1] = 0

    WPS1 = np.abs(dec) ** 2
    diag["n_zero_after_white"] = int(np.count_nonzero(WPS1 == 0))

    median_WPS = np.median(WPS1, axis=(1, 2, 3))  # (ns,)
    mad = stats.median_abs_deviation(WPS1, axis=(1, 2, 3), scale=1.0)  # (ns,)
    sMAD_WPS = 1.4826 * mad

    sMAD_safe = np.where(sMAD_WPS > 0, sMAD_WPS, np.nan)
    WPS_scaled = (WPS1 - median_WPS[:, None, None, None]) / sMAD_safe[:, None, None, None]

    mask = WPS_scaled < float(wps_noise_threshold)
    dec[mask] = 0

    WPS2 = np.abs(dec) ** 2
    diag["n_zero_after_scaled"] = int(np.count_nonzero(WPS2 == 0))

    finite_scaled = np.isfinite(WPS_scaled)

    diag["sMAD_min"] = float(np.nanmin(sMAD_WPS))
    diag["sMAD_median"] = float(np.nanmedian(sMAD_WPS))

    if np.any(finite_scaled):
        diag["WPS_scaled_min"] = float(np.nanmin(WPS_scaled[finite_scaled]))
        diag["WPS_scaled_median"] = float(np.nanmedian(WPS_scaled[finite_scaled]))
        diag["WPS_scaled_max"] = float(np.nanmax(WPS_scaled[finite_scaled]))
    else:
        diag["WPS_scaled_min"] = np.nan
        diag["WPS_scaled_median"] = np.nan
        diag["WPS_scaled_max"] = np.nan

    return cwt_copy, diag


# --------------------------
# Quicklook plots (now true no-op if not saving)
# --------------------------
def quicklook_save(
    x1d: np.ndarray,
    y1d: np.ndarray,
    Z: np.ndarray,
    title: str,
    outpath: Path,
    *,
    save_plots: bool,
) -> None:
    if not save_plots:
        return
    fig, ax = plt.subplots()
    im = ax.pcolormesh(x1d, y1d, Z, shading="auto", cmap="gray")
    fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04).set_label("arb")
    ax.set_xlabel("x (km)")
    ax.set_ylabel("y (km)")
    ax.set_title(title)
    ax.set_aspect("equal", adjustable="box")
    fig.tight_layout()
    save_fig(fig, outpath, enable=True)


def quicklook_background_decomp_4panel_save(
    x1d_km: np.ndarray,
    y1d_km: np.ndarray,
    original: np.ndarray,
    background: np.ndarray,
    remaining: np.ndarray,
    remaining_tapered: np.ndarray,
    taper_edge_px: int,
    outpath: Path,
    *,
    wavy_after_noise_xy: np.ndarray | None = None,
    save_plots: bool,
) -> None:
    if not save_plots:
        return

    med = np.nanmedian(original)
    orig_c = original - med
    bg_c = background - med
    rem_c = remaining
    rem_tapered_c = remaining_tapered

    arrays = [orig_c, bg_c, rem_c, rem_tapered_c]
    if wavy_after_noise_xy is not None:
        arrays.append(wavy_after_noise_xy)

    vmin, vmax = diverging_limits(*arrays, p_lo=1, p_hi=99)

    fig, axs = plt.subplots(2, 3, figsize=(16, 10), constrained_layout=True)
    axs = axs.flatten()

    panels = [
        (orig_c, "Original (median sub)"),
        (bg_c, "Background (plane + first Fourier)"),
        (rem_c, "Remaining = original - background"),
        (rem_tapered_c, f"Remaining with edge taper (width={taper_edge_px})"),
    ]

    if wavy_after_noise_xy is not None:
        panels.append((wavy_after_noise_xy, "Wavy stuff (wavelet reconstruction after noise filtering)"))

    im0 = None
    for ax, (Zp, title) in zip(axs, panels):
        im0 = ax.pcolormesh(x1d_km, y1d_km, Zp, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
        ax.set_title(title)
        ax.set_xlabel("x (km)")
        ax.set_ylabel("y (km)")
        ax.set_aspect("equal", adjustable="box")

    for ax in axs[len(panels) :]:
        ax.axis("off")

    cbar = fig.colorbar(im0, ax=axs[: len(panels)], orientation="vertical", fraction=0.04, pad=0.02)
    cbar.set_label("arb")
    save_fig(fig, outpath, enable=True)


def plot_cluster_individual_save(
    cwt: dict,
    iwave: np.ndarray,
    kcl: int,
    x1d_km: np.ndarray,
    y1d_km: np.ndarray,
    orig_decomp: np.ndarray,
    bg_xy: np.ndarray,
    dx_km: float,
    dy_km: float,
    outpath: Path,
    *,
    save_plots: bool,
) -> None:
    if not save_plots:
        return

    decomposition = cwt["decomposition"]
    decomposition[:] = orig_decomp
    decomposition[iwave != kcl] = 0

    rec = transform.reconstruct2d(cwt)  # (nx, ny)
    rec_xy = rec.T - np.median(rec)

    A_xy, theta_xy = cluster_amp_theta_maps(
        cwt, iwave, kcl, amp_min_fraction_of_cluster_max=AMP_MIN_FRACTION_OF_CLUSTER_MAX
    )

    dec0 = orig_decomp
    power0 = np.abs(dec0) ** 2
    mask4d = (iwave == kcl)
    ns, nt, nx, ny = power0.shape

    power2 = power0.reshape(ns * nt, nx * ny)
    mask2 = mask4d.reshape(ns * nt, nx * ny)
    p = np.where(mask2, power2, -np.inf)
    pmax = np.max(p, axis=0)

    lam_xy = np.full(nx * ny, np.nan, dtype=float)
    valid = np.isfinite(pmax) & (pmax > 0)
    if np.any(valid):
        arg = np.argmax(p[:, valid], axis=0)
        s_idx = arg // nt
        t_idx = arg % nt

        lamx = np.asarray(cwt["wavelength_x"], dtype=float)[s_idx, t_idx]
        lamy = np.asarray(cwt["wavelength_y"], dtype=float)[s_idx, t_idx]

        inv2 = np.zeros_like(lamx, dtype=float)
        goodx = np.isfinite(lamx) & (lamx != 0)
        goody = np.isfinite(lamy) & (lamy != 0)
        inv2[goodx] += (1.0 / lamx[goodx]) ** 2
        inv2[goody] += (1.0 / lamy[goody]) ** 2

        lam = np.full_like(inv2, np.nan, dtype=float)
        ok = inv2 > 0
        lam[ok] = 1.0 / np.sqrt(inv2[ok])
        lam_xy[valid] = lam

    lam_xy = lam_xy.reshape(nx, ny).T

    if np.isfinite(np.nanmax(A_xy)):
        thr = float(AMP_MIN_FRACTION_OF_CLUSTER_MAX) * float(np.nanmax(A_xy))
        lam_xy[A_xy < thr] = np.nan

    bg_vmin, bg_vmax = diverging_limits(bg_xy, p_lo=2, p_hi=98)
    cl_vmin, cl_vmax = diverging_limits(rec_xy, p_lo=2, p_hi=98)

    theta_vals = np.asarray(cwt["theta"], dtype=float)
    theta_min = float(np.nanmin(theta_vals))
    theta_max = float(np.nanmax(theta_vals))

    if np.any(np.isfinite(lam_xy)):
        lam_vmin, lam_vmax = np.nanpercentile(lam_xy, [5, 95])
        if (not np.isfinite(lam_vmin)) or (not np.isfinite(lam_vmax)) or (lam_vmin == lam_vmax):
            lam_vmin = float(np.nanmin(lam_xy))
            lam_vmax = float(np.nanmax(lam_xy))
    else:
        lam_vmin, lam_vmax = 0.0, 1.0

    fig, axs = plt.subplots(2, 2, figsize=(16, 11), constrained_layout=True)
    ax00, ax01, ax10, ax11 = axs[0, 0], axs[0, 1], axs[1, 0], axs[1, 1]

    ax00.pcolormesh(x1d_km, y1d_km, bg_xy, shading="auto", cmap="gray", vmin=bg_vmin, vmax=bg_vmax)
    im00 = ax00.pcolormesh(
        x1d_km, y1d_km, rec_xy, shading="auto", cmap="RdBu_r", vmin=cl_vmin, vmax=cl_vmax, alpha=0.6
    )
    ax00.set_title(f"Cluster {kcl} overlay")
    ax00.set_xlabel("x (km)")
    ax00.set_ylabel("y (km)")
    ax00.set_aspect("equal", adjustable="box")
    fig.colorbar(im00, ax=ax00, fraction=0.046, pad=0.02)

    im01 = ax01.pcolormesh(x1d_km, y1d_km, A_xy, shading="auto", cmap="plasma")
    n_pix = int(np.count_nonzero(np.isfinite(A_xy)))
    area_km2 = n_pix * float(dx_km) * float(dy_km)
    ax01.set_title("A(x,y) = max |coeff| in cluster\n" + rf"Area = {area_km2:.0f} km$^2$ ({n_pix} px)")
    ax01.set_xlabel("x (km)")
    ax01.set_ylabel("y (km)")
    ax01.set_aspect("equal", adjustable="box")
    fig.colorbar(im01, ax=ax01, fraction=0.046, pad=0.02)

    im10 = ax10.pcolormesh(x1d_km, y1d_km, lam_xy, shading="auto", cmap="viridis", vmin=lam_vmin, vmax=lam_vmax)
    if np.any(np.isfinite(lam_xy)):
        lam_med = float(np.nanmedian(lam_xy))
        lam_p10, lam_p90 = np.nanpercentile(lam_xy, [10, 90])
        ax10.set_title(
            r"Dominant $\lambda(x,y)$ in cluster" + "\n"
            + rf"median={_fmt(lam_med)} km (10–90%: {_fmt(lam_p10)}–{_fmt(lam_p90)} km)"
        )
    else:
        ax10.set_title(r"Dominant $\lambda(x,y)$ in cluster")
    ax10.set_xlabel("x (km)")
    ax10.set_ylabel("y (km)")
    ax10.set_aspect("equal", adjustable="box")
    fig.colorbar(im10, ax=ax10, fraction=0.046, pad=0.02).set_label("km")

    im11 = ax11.pcolormesh(x1d_km, y1d_km, theta_xy, shading="auto", cmap="PiYG", vmin=theta_min, vmax=theta_max)
    ax11.set_title("θ(x,y) + quiver")
    ax11.set_xlabel("x (km)")
    ax11.set_ylabel("y (km)")
    ax11.set_aspect("equal", adjustable="box")
    fig.colorbar(im11, ax=ax11, fraction=0.046, pad=0.02)

    if np.any(np.isfinite(theta_xy)):
        u, v = theta_to_uv(theta_xy)
        arrow_len_km = 40.0
        u = u * arrow_len_km
        v = v * arrow_len_km

        Xq, Yq = np.meshgrid(x1d_km, y1d_km)
        Xqs = Xq[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        Yqs = Yq[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        uqs = u[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        vqs = v[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        good = np.isfinite(uqs) & np.isfinite(vqs)
        ax11.quiver(
            Xqs[good],
            Yqs[good],
            uqs[good],
            vqs[good],
            angles="xy",
            scale_units="xy",
            scale=QUIVER_SCALE,
            width=0.0025,
        )

    save_fig(fig, outpath, enable=True)


# --------------------------
# Cluster products (for L6C/L6A outputs)
# --------------------------
def cluster_dominant_lambda_map(
    cwt: dict,
    iwave: np.ndarray,
    kcl: int,
    amp_mask_xy: np.ndarray | None = None,
) -> np.ndarray:
    dec = cwt["decomposition"]  # (ns, nt, nx, ny) complex
    power0 = np.abs(dec) ** 2
    mask4d = (iwave == kcl)
    if not np.any(mask4d):
        ny = dec.shape[3]
        nx = dec.shape[2]
        return np.full((ny, nx), np.nan)

    ns, nt, nx, ny = power0.shape
    power2 = power0.reshape(ns * nt, nx * ny)
    mask2 = mask4d.reshape(ns * nt, nx * ny)

    p = np.where(mask2, power2, -np.inf)
    pmax = np.max(p, axis=0)
    valid = np.isfinite(pmax) & (pmax > 0)

    lam_xy = np.full(nx * ny, np.nan, dtype=float)
    if np.any(valid):
        arg = np.argmax(p[:, valid], axis=0)
        s_idx = arg // nt
        t_idx = arg % nt

        lamx = np.asarray(cwt["wavelength_x"], dtype=float)[s_idx, t_idx]
        lamy = np.asarray(cwt["wavelength_y"], dtype=float)[s_idx, t_idx]

        inv2 = np.zeros_like(lamx, dtype=float)
        goodx = np.isfinite(lamx) & (lamx != 0)
        goody = np.isfinite(lamy) & (lamy != 0)
        inv2[goodx] += (1.0 / lamx[goodx]) ** 2
        inv2[goody] += (1.0 / lamy[goody]) ** 2

        lam = np.full_like(inv2, np.nan, dtype=float)
        ok = inv2 > 0
        lam[ok] = 1.0 / np.sqrt(inv2[ok])

        lam_xy[valid] = lam

    lam_xy = lam_xy.reshape(nx, ny).T

    if amp_mask_xy is not None:
        lam_xy[~np.isfinite(amp_mask_xy)] = np.nan

    return lam_xy


def copy_lxc_to_l6x_with_clusters(
    lxc_path: str | Path,
    l6x_path: str | Path,
    clusters: List[Dict[str, Any]],
    y_dim: int,
    x_dim: int,
    unbin_factor: int = 1,
    product_type: str = "l5c",
) -> None:
    def _unbin_repeat_xy_local(A: np.ndarray, factor: int) -> np.ndarray:
        if factor is None or int(factor) <= 1:
            return A
        f = int(factor)
        return np.repeat(np.repeat(A, f, axis=0), f, axis=1)

    lxc_path = Path(lxc_path)
    l6x_path = Path(l6x_path)
    l6x_path.parent.mkdir(parents=True, exist_ok=True)

    ncl = int(len(clusters))
    has_rec = any((c.get("rec_xy", None) is not None) for c in clusters)

    fup = int(unbin_factor) if (unbin_factor is not None and int(unbin_factor) > 1) else 1
    y_dim_w = int(y_dim) * fup
    x_dim_w = int(x_dim) * fup

    with Dataset(lxc_path, "r") as src, Dataset(l6x_path, "w", format="NETCDF4") as dst:
        dst.setncatts({k: src.getncattr(k) for k in src.ncattrs()})

        for dname, dim in src.dimensions.items():
            dst.createDimension(dname, (len(dim) if not dim.isunlimited() else None))

        for vname, var in src.variables.items():
            out = dst.createVariable(
                vname,
                var.datatype,
                var.dimensions,
                zlib=True,
                complevel=4,
                fill_value=getattr(var, "_FillValue", None),
            )
            out.setncatts({k: var.getncattr(k) for k in var.ncattrs()})
            out[:] = var[:]

        dst.createDimension("cluster", ncl)
        dst.createDimension("y_cluster", y_dim_w)
        dst.createDimension("x_cluster", x_dim_w)

        v_slice = dst.createVariable("SlicesNo", "i4", ("cluster",), zlib=True, complevel=4)
        v_area = dst.createVariable("Area", "f4", ("cluster",), zlib=True, complevel=4)
        v_pabs = dst.createVariable("ClusterPowerAbs", "f8", ("cluster",), zlib=True, complevel=4)
        v_ppct = dst.createVariable("ClusterPowerPct", "f4", ("cluster",), zlib=True, complevel=4)

        v_amp = dst.createVariable(
            "Amplitude",
            "f4",
            ("cluster", "y_cluster", "x_cluster"),
            zlib=True,
            complevel=4,
            fill_value=np.float32(np.nan),
        )
        v_lam = dst.createVariable(
            "DominantWavelength",
            "f4",
            ("cluster", "y_cluster", "x_cluster"),
            zlib=True,
            complevel=4,
            fill_value=np.float32(np.nan),
        )
        v_ang = dst.createVariable(
            "Angle",
            "f4",
            ("cluster", "y_cluster", "x_cluster"),
            zlib=True,
            complevel=4,
            fill_value=np.float32(np.nan),
        )

        v_rec = None
        if has_rec:
            v_rec = dst.createVariable(
                "ClusterReconstruction",
                "f4",
                ("cluster", "y_cluster", "x_cluster"),
                zlib=True,
                complevel=4,
                fill_value=np.float32(np.nan),
            )
            v_rec.units = "arb"
            v_rec.long_name = "Per cluster wavelet reconstruction map (relative; median subtracted)"

        v_area.units = "km2"
        v_lam.units = "km"
        v_ang.units = "rad"
        v_slice.long_name = "Slice index (starts at 0) for each cluster"
        v_pabs.units = "arb"
        v_ppct.units = "percent"

        v_amp.units = "K" if product_type == "l3a" else "kR"

        if ncl == 0:
            return

        v_slice[:] = np.asarray([c["slice_no"] for c in clusters], dtype=np.int32)
        v_area[:] = np.asarray([c["area_km2"] for c in clusters], dtype=np.float32)
        v_pabs[:] = np.asarray([c.get("power_abs", np.nan) for c in clusters], dtype=np.float64)
        v_ppct[:] = np.asarray([c.get("power_pct", np.nan) for c in clusters], dtype=np.float32)

        amp_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32)
        lam_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32)
        ang_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32)
        rec_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32) if has_rec else None

        def _force_shape(M: np.ndarray) -> np.ndarray:
            if M.shape == (y_dim_w, x_dim_w):
                return M.astype(np.float32)
            tmp = np.full((y_dim_w, x_dim_w), np.nan, dtype=np.float32)
            yy = min(y_dim_w, M.shape[0])
            xx = min(x_dim_w, M.shape[1])
            tmp[:yy, :xx] = M[:yy, :xx]
            return tmp

        for i, c in enumerate(clusters):
            A = _unbin_repeat_xy_local(c["A_xy"], fup)
            L = _unbin_repeat_xy_local(c["lam_xy"], fup)
            T = _unbin_repeat_xy_local(c["theta_xy"], fup)

            amp_stack[i] = _force_shape(A)
            lam_stack[i] = _force_shape(L)
            ang_stack[i] = _force_shape(T)

            if has_rec and rec_stack is not None and v_rec is not None:
                R = c.get("rec_xy", None)
                if R is not None:
                    R = _force_shape(_unbin_repeat_xy_local(R, fup))
                    rec_stack[i] = R

        v_amp[:] = amp_stack
        v_lam[:] = lam_stack
        v_ang[:] = ang_stack
        if has_rec and v_rec is not None and rec_stack is not None:
            v_rec[:] = rec_stack


def make_output_cluster_path(input_path: str | Path) -> Path:
    p = Path(input_path)
    name = p.name
    if "_l5c_" in name:
        name = name.replace("_l5c_", "_l6c_")
    elif "_l3c_" in name:
        name = name.replace("_l3c_", "_l6c_")
    elif "_l3a_" in name:
        name = name.replace("_l3a_", "_l6a_")
    else:
        name = p.stem + "_clusters.nc"
    return p.with_name(name)


# --------------------------
# Slice pipeline
# --------------------------
def run_pipeline_for_slice(
    rad2d_flat: np.ndarray,
    x_slice: slice,
    outdir: Path,
    x_start_unbinned: int,
    core_start_unbinned: int,
    core_end_unbinned: int,
    cfg: Dict[str, Any],
    slice_no: int,
    product_type: str,
) -> Dict[str, Any]:
    outdir.mkdir(parents=True, exist_ok=True)
    save_plots = bool(SAVE_SLICE_PLOTS)

    min_amp = float(cfg.get("MIN_AMP"))
    thr = float(cfg.get("THR"))
    white_noise_threshold = float(cfg.get("WHITE_NOISE_THRESHOLD"))
    wps_noise_threshold = float(cfg.get("WPS_NOISE_THRESHOLD"))
    run_name = str(cfg.get("name", "run"))

    STORE_CLUSTER_RECON = True

    if APPLY_BINNING and int(BIN_FACTOR) > 1:
        f = int(BIN_FACTOR)
        dx_km = DX_KM_BASE * f
        dy_km = DY_KM_BASE * f
    else:
        f = 1
        dx_km = DX_KM_BASE
        dy_km = DY_KM_BASE

    core_width_unbinned = int(core_end_unbinned - core_start_unbinned)
    core_w_binned = core_width_unbinned // f
    rel0_unbinned = int(core_start_unbinned - x_start_unbinned)
    rel0_binned = rel0_unbinned // f
    rel1_binned = rel0_binned + core_w_binned

    def crop_pad_core(A: np.ndarray) -> np.ndarray:
        nyA, nxA = A.shape
        left = max(0, rel0_binned)
        right = min(nxA, rel1_binned)
        core = A[:, left:right]
        need = core_w_binned - core.shape[1]
        if need > 0:
            core = np.concatenate([core, np.full((nyA, need), np.nan)], axis=1)
        elif need < 0:
            core = core[:, :core_w_binned]
        return core

    raw_win = np.asarray(rad2d_flat[Y_SLICE, x_slice], dtype=float)

    if APPLY_BINNING and int(BIN_FACTOR) > 1:
        raw_win = bin2d_block_mean(raw_win, BIN_FACTOR)

    ny, nx = raw_win.shape
    x1d_km = np.arange(nx, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    bin_tag = f" (binned {BIN_FACTOR}x{BIN_FACTOR})" if (APPLY_BINNING and int(BIN_FACTOR) > 1) else ""
    label0 = "Temperature" if product_type == "l3a" else "Radiance"

    quicklook_save(
        x1d_km,
        y1d_km,
        raw_win,
        f"{label0} (window, flattened){bin_tag}\n{run_name} MIN_AMP={min_amp} THR={thr}",
        outdir / "00_window.png",
        save_plots=save_plots,
    )

    proc0 = raw_win.astype(float, copy=True)

    if REMOVE_TREND:
        plane = fit_plane(proc0)
        detrended = proc0 - plane
    else:
        plane = np.zeros_like(proc0)
        detrended = proc0

    fourier_bg, _ = extract_first_fourier_bg(detrended)
    background = plane + fourier_bg
    remaining = proc0 - background

    if APPLY_TAPER:
        remaining_tapered = apply_edge_hanning_taper(remaining, TAPER_EDGE_PIXELS)
    else:
        remaining_tapered = remaining.copy()

    proc = remaining_tapered.copy()
    if STANDARDIZE:
        m = float(np.median(proc))
        s = float(np.std(proc))
        proc = (proc - m) / s if s > 0 else (proc - m)
    proc = maybe_blur(proc, GAUSS_BLUR_SIGMA)

    wave_xy = proc.T  # (nx, ny)

    cwt_raw = transform.decompose2d(
        wave_xy, dx=dx_km, dy=dy_km, s0=S0, dj=DJ, js=JS, jt=JT, aspect=ASPECT
    )

    if APPLY_RED_NOISE_FILTER:
        cwt, diag = apply_red_noise_filter_cwt(
            cwt_raw,
            white_noise_threshold=white_noise_threshold,
            wps_noise_threshold=wps_noise_threshold,
        )

        if SAVE_REDNOISE_DIAGNOSTICS:
            lines = []
            lines.append("Red noise filter diagnostics\n\n")
            lines.append(f"name                  = {run_name}\n")
            lines.append(f"WHITE_NOISE_THRESHOLD = {white_noise_threshold}\n")
            lines.append(f"WPS_NOISE_THRESHOLD   = {wps_noise_threshold}\n\n")
            if diag.get("skipped", False):
                lines.append(f"SKIPPED: {diag.get('reason','unknown')}\n")
            else:
                lines.append(f"n_coeff_total        = {diag.get('n_coeff_total')}\n")
                lines.append(f"n_zero_after_white   = {diag.get('n_zero_after_white')}\n")
                lines.append(f"n_zero_after_scaled  = {diag.get('n_zero_after_scaled')}\n")
                lines.append(f"sMAD_WPS min         = {diag.get('sMAD_min')}\n")
                lines.append(f"sMAD_WPS median      = {diag.get('sMAD_median')}\n")
                lines.append(f"WPS_scaled min       = {diag.get('WPS_scaled_min')}\n")
                lines.append(f"WPS_scaled median    = {diag.get('WPS_scaled_median')}\n")
                lines.append(f"WPS_scaled max       = {diag.get('WPS_scaled_max')}\n")
            (outdir / "02a_rednoise_diagnostics.txt").write_text("".join(lines))
    else:
        cwt = cwt_raw

    rec_wavy = transform.reconstruct2d(cwt)  # (nx, ny)
    wavy_xy = rec_wavy.T - np.median(rec_wavy)

    quicklook_background_decomp_4panel_save(
        x1d_km,
        y1d_km,
        proc0,
        background,
        remaining,
        remaining_tapered,
        TAPER_EDGE_PIXELS,
        outdir / "01_background_decomp_5panel.png",
        wavy_after_noise_xy=wavy_xy,
        save_plots=save_plots,
    )

    amps, idxs, iwave = utils.identify_cluster2d(cwt, min_amp=min_amp, thr=thr)
    decomposition = cwt["decomposition"]
    orig_decomp = decomposition.copy()

    if save_plots:
        cmap_cwt = matplotlib.cm.turbo
        norm_cwt = matplotlib.colors.BoundaryNorm(np.exp(np.linspace(np.log(0.05), np.log(1.0), 10)), cmap_cwt.N)
        fig_cwt, _ = utils.plot_decomposition2d(cwt, redux_s=2, redux_t=2, cmap=cmap_cwt, norm=norm_cwt)
        save_fig(fig_cwt, outdir / "02_decomposition_overview.png", enable=True)

        plot_utils.plot_fov_wavelet_spectrum(cwt, stat="max")
        fig_spec = plt.gcf()
        save_fig(fig_spec, outdir / "03_fov_wavelet_spectrum.png", enable=True)

    bg = remaining_tapered - np.median(remaining_tapered)

    cluster_ids, P_abs, P_pct = compute_cluster_power_table(cwt, iwave)
    if cluster_ids.size == 0:
        (outdir / "NO_CLUSTERS.txt").write_text(
            "No clusters found.\n\n"
            f"Run name: {run_name}\n"
            f"Used MIN_AMP={min_amp}, THR={thr}\n"
            f"Used WHITE_NOISE_THRESHOLD={white_noise_threshold}, WPS_NOISE_THRESHOLD={wps_noise_threshold}\n\n"
            "Try lowering MIN_AMP or THR.\n"
            "If red noise filter is on, try lowering WPS_NOISE_THRESHOLD or WHITE_NOISE_THRESHOLD.\n"
        )
        remaining_core_xy = crop_pad_core(proc)
        nan_core = np.full_like(remaining_core_xy, np.nan)

        cleanup_memory()
        return {
            "has_clusters": False,
            "x_start_unbinned": core_start_unbinned,
            "dx_km": dx_km,
            "dy_km": dy_km,
            "remaining_xy": remaining_core_xy,
            "wavy_xy": nan_core,
            "rec_all_xy": nan_core,
            "cluster_records": [],
        }

    power_abs_map = {int(k): float(p) for k, p in zip(cluster_ids, P_abs)}
    power_pct_map = {int(k): float(p) for k, p in zip(cluster_ids, P_pct)}

    cluster_records: List[Dict[str, Any]] = []
    for kcl in cluster_ids.astype(int):
        kcl = int(kcl)

        A_xy, theta_xy = cluster_amp_theta_maps(
            cwt,
            iwave,
            kcl,
            amp_min_fraction_of_cluster_max=AMP_MIN_FRACTION_OF_CLUSTER_MAX,
        )
        lam_xy = cluster_dominant_lambda_map(cwt, iwave, kcl, amp_mask_xy=A_xy)

        rec_core = None
        if STORE_CLUSTER_RECON:
            decomposition[:] = orig_decomp
            decomposition[iwave != kcl] = 0
            rec = transform.reconstruct2d(cwt)  # (nx, ny)
            rec_xy = rec.T
            rec_xy = rec_xy - np.nanmedian(rec_xy)
            rec_core = crop_pad_core(rec_xy)
            decomposition[:] = orig_decomp

        A_core = crop_pad_core(A_xy)
        th_core = crop_pad_core(theta_xy)
        lam_core = crop_pad_core(lam_xy)

        n_pix = int(np.count_nonzero(np.isfinite(A_core)))
        area_km2 = float(n_pix) * float(dx_km) * float(dy_km)

        cluster_records.append(
            dict(
                slice_no=int(slice_no),
                kcl=kcl,
                area_km2=area_km2,
                power_abs=power_abs_map.get(kcl, np.nan),
                power_pct=power_pct_map.get(kcl, np.nan),
                A_xy=A_core,
                lam_xy=lam_core,
                theta_xy=th_core,
                rec_xy=rec_core,
            )
        )

    order_power = np.argsort(P_abs)[::-1]
    cluster_ids_sorted = cluster_ids[order_power]
    P_abs_sorted = P_abs[order_power]
    P_pct_sorted = P_pct[order_power]

    lines = []
    lines.append("Cluster power ranking (clustered only)\n\n")
    lines.append(f"name = {run_name}\n")
    lines.append(f"MIN_AMP = {min_amp}\n")
    lines.append(f"THR = {thr}\n")
    lines.append(f"WHITE_NOISE_THRESHOLD = {white_noise_threshold}\n")
    lines.append(f"WPS_NOISE_THRESHOLD = {wps_noise_threshold}\n\n")
    lines.append("rank  kcl      P_abs           P_pct\n")
    for r, (kcl, Pk, pk) in enumerate(zip(cluster_ids_sorted, P_abs_sorted, P_pct_sorted), start=1):
        lines.append(f"{r:4d}  {int(kcl):3d}  {Pk: .6e}   {pk:6.2f}%\n")
    (outdir / "04_cluster_power_ranking.txt").write_text("".join(lines))

    n_show = int(min(N_SHOW, cluster_ids_sorted.size))
    biggest = cluster_ids_sorted[:n_show]
    theta_vals = np.asarray(cwt["theta"], dtype=float)

    if save_plots:
        nrow = int(np.ceil(np.sqrt(n_show)))
        ncol = int(np.ceil(n_show / nrow))

        def plot_cluster_grid(kind: str, outname: str) -> None:
            fig, axs = plt.subplots(nrow, ncol, figsize=(12, 10))
            axs = np.atleast_1d(axs).reshape(-1)

            bg_vmin, bg_vmax = diverging_limits(bg, p_lo=2, p_hi=98)
            number_box = {"boxstyle": "square", "lw": 0.67, "facecolor": "white", "edgecolor": "black"}

            for ax, kcl in zip(axs, biggest):
                decomposition[:] = orig_decomp
                decomposition[iwave != kcl] = 0
                rec = transform.reconstruct2d(cwt)
                rec_centered = rec.T - np.median(rec)

                s_idx = int(idxs[kcl][0])
                t_idx = int(idxs[kcl][1])
                _, _, lam = effective_lambda(cwt, s_idx, t_idx)

                theta_dom_rad = float(theta_vals[t_idx])
                theta_dom_deg = (np.degrees(theta_dom_rad) + 180.0) % 180.0

                ii = int(np.where(cluster_ids_sorted == kcl)[0][0])
                Pk = float(P_abs_sorted[ii])
                pk = float(P_pct_sorted[ii])

                if kind == "standalone":
                    ax.pcolormesh(x1d_km, y1d_km, rec_centered, shading="auto", cmap="RdBu_r", vmin=-1.0, vmax=1.0)
                elif kind == "overlay":
                    ax.pcolormesh(x1d_km, y1d_km, bg, shading="auto", cmap="gray", vmin=bg_vmin, vmax=bg_vmax)
                    ax.pcolormesh(
                        x1d_km, y1d_km, rec_centered, shading="auto", cmap="RdBu_r", vmin=-1.0, vmax=1.0, alpha=0.6
                    )
                else:
                    raise ValueError(kind)

                ax.set_title(
                    rf"$\lambda$={_fmt(lam)} km, $\theta$={theta_dom_deg:.0f}$^\circ$"
                    + "\n"
                    + rf"P={Pk:.2e} ({pk:.1f}%)"
                )
                ax.set_aspect("equal", adjustable="box")

            for ax in axs[n_show:]:
                ax.axis("off")
            for ax in axs[:n_show]:
                ax.set_xlabel("x (km)")
                ax.set_ylabel("y (km)")

            for i, ax in enumerate(axs, start=1):
                if ax.has_data():
                    ax.text(0.12, 0.15, str(i), transform=ax.transAxes, bbox=number_box, va="top", ha="right")

            fig.tight_layout()
            save_fig(fig, outdir / outname, enable=True)

        plot_cluster_grid("standalone", "05_clusters_standalone.png")
        plot_cluster_grid("overlay", "06_clusters_overlay.png")

        for kcl in biggest:
            plot_cluster_individual_save(
                cwt=cwt,
                iwave=iwave,
                kcl=int(kcl),
                x1d_km=x1d_km,
                y1d_km=y1d_km,
                orig_decomp=orig_decomp,
                bg_xy=bg,
                dx_km=dx_km,
                dy_km=dy_km,
                outpath=outdir / f"07_cluster_{int(kcl):03d}_maps.png",
                save_plots=True,
            )

    decomposition[:] = orig_decomp
    decomposition[iwave < 0] = 0
    rec_all = transform.reconstruct2d(cwt)  # (nx, ny)
    rec_all_xy = rec_all.T - np.median(rec_all)
    decomposition[:] = orig_decomp

    remaining_after_preproc_xy = proc

    if save_plots:
        vmin, vmax = diverging_limits(remaining_after_preproc_xy, wavy_xy, rec_all_xy, p_lo=2, p_hi=98)

        figR, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(18, 8), constrained_layout=True, sharex=True, sharey=True)

        im1 = ax1.pcolormesh(
            x1d_km, y1d_km, remaining_after_preproc_xy, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax
        )
        ax1.set_title("Remaining (after preprocessing, input to CWT)")
        ax1.set_xlabel("x (km)")
        ax1.set_ylabel("y (km)")
        ax1.set_aspect("equal", adjustable="box")

        im2 = ax2.pcolormesh(x1d_km, y1d_km, wavy_xy, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
        ax2.set_title("Wavy stuff (wavelet reconstruction after noise filtering)")
        ax2.set_xlabel("x (km)")
        ax2.set_ylabel("y (km)")
        ax2.set_aspect("equal", adjustable="box")

        im3 = ax3.pcolormesh(x1d_km, y1d_km, rec_all_xy, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
        ax3.set_title("Reconstruction after clustering (all clusters combined)")
        ax3.set_xlabel("x (km)")
        ax3.set_ylabel("y (km)")
        ax3.set_aspect("equal", adjustable="box")

        cbar = figR.colorbar(im1, ax=[ax1, ax2, ax3], orientation="vertical", fraction=0.015, pad=0.01)
        cbar.set_label("arb")
        save_fig(figR, outdir / "09_remaining_vs_wavy_vs_allclusters.png", enable=True)

    remaining_core_xy = crop_pad_core(remaining_after_preproc_xy)
    wavy_core_xy = crop_pad_core(wavy_xy)
    rec_all_core_xy = crop_pad_core(rec_all_xy)

    try:
        del raw_win, proc0, plane, detrended, fourier_bg, background, remaining, remaining_tapered, proc, wave_xy
        del cwt_raw, cwt, amps, idxs, iwave, decomposition, orig_decomp, rec_wavy, rec_all
        del bg, cluster_ids, P_abs, P_pct, cluster_ids_sorted, P_abs_sorted, P_pct_sorted
    except Exception:
        pass
    cleanup_memory()

    return {
        "has_clusters": True,
        "x_start_unbinned": core_start_unbinned,
        "dx_km": dx_km,
        "dy_km": dy_km,
        "remaining_xy": remaining_core_xy,
        "wavy_xy": wavy_core_xy,
        "rec_all_xy": rec_all_core_xy,
        "cluster_records": cluster_records,
    }


def stitch_and_save_final_09(slice_results: List[Dict[str, Any]], outroot: Path, tag: str, product_type: str) -> None:
    if not SAVE_STITCHED_PLOT:
        return
    if not slice_results:
        return

    slice_results = sorted(slice_results, key=lambda d: int(d.get("x_start_unbinned", 0)))

    dx_km = float(slice_results[0]["dx_km"])
    dy_km = float(slice_results[0]["dy_km"])
    x0_unbinned = int(slice_results[0].get("x_start_unbinned", 0))
    x0_km = x0_unbinned * DX_KM_BASE

    remaining_stitched = np.concatenate([d["remaining_xy"] for d in slice_results], axis=1)
    wavy_stitched = np.concatenate([d["wavy_xy"] for d in slice_results], axis=1)
    rec_all_stitched = np.concatenate([d["rec_all_xy"] for d in slice_results], axis=1)

    ny, nx_tot = remaining_stitched.shape
    x1d_km = x0_km + np.arange(nx_tot, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    vmin, vmax = diverging_limits(remaining_stitched, wavy_stitched, rec_all_stitched, p_lo=2, p_hi=98)

    figR, (axT, axM, axB) = plt.subplots(3, 1, figsize=(18, 6), constrained_layout=True, sharex=True, sharey=True)

    imT = axT.pcolormesh(x1d_km, y1d_km, remaining_stitched, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    axT.set_title(f"STITCHED (core tiles) {tag}: Remaining (after preprocessing, input to CWT)")
    axT.set_xlabel("x (km)")
    axT.set_ylabel("y (km)")
    axT.set_aspect("equal", adjustable="box")

    imM = axM.pcolormesh(x1d_km, y1d_km, wavy_stitched, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    axM.set_title(f"STITCHED (core tiles) {tag}: Wavy stuff (wavelet reconstruction after noise filtering)")
    axM.set_xlabel("x (km)")
    axM.set_ylabel("y (km)")
    axM.set_aspect("equal", adjustable="box")

    imB = axB.pcolormesh(x1d_km, y1d_km, rec_all_stitched, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    axB.set_title(f"STITCHED (core tiles) {tag}: Reconstruction after clustering (all clusters combined)")
    axB.set_xlabel("x (km)")
    axB.set_ylabel("y (km)")
    axB.set_aspect("equal", adjustable="box")

    cbar = figR.colorbar(imT, ax=[axT, axM, axB], orientation="vertical", fraction=0.012, pad=0.01)
    cbar.set_label("K (Relative)" if product_type == "l3a" else "kRayleigh (Relative)")

    save_fig(
        figR,
        outroot / f"FINAL_09_remaining_vs_wavy_vs_allclusters_STITCHED_{tag}.png",
        enable=True,
    )

    try:
        del remaining_stitched, wavy_stitched, rec_all_stitched
    except Exception:
        pass
    cleanup_memory()


# --------------------------
# Per file runner
# --------------------------
def run_one_file(file_path: Path) -> None:
    product_type = detect_product_type(file_path)
    runs = get_runs_for_product(product_type)

    print("\n========================================")
    print("File:", file_path.name)
    print("Detected product:", product_type)
    print("Runs:", runs)

    rad2d = read_frame(file_path, FRAME_IDX, product_type=product_type)
    rad2d_flat = flatten_swath(rad2d, target_ny=300)
    nx_total = rad2d_flat.shape[1]

    x_offset = int(X_OFFSET)
    if x_offset >= nx_total:
        raise ValueError(f"X_OFFSET={x_offset} too large for nx_total={nx_total} in file {file_path.name}")

    for cfg in runs:
        run_name = str(cfg.get("name", "run"))

        outroot = OUTPUT_ROOT / file_path.stem / f"{run_name}_offset_{x_offset:04d}"
        outroot.mkdir(parents=True, exist_ok=True)

        (outroot / "RUN_CONFIG.txt").write_text(
            "\n".join([f"{k} = {v}" for k, v in cfg.items()])
            + f"\nproduct_type = {product_type}\nfile = {file_path.name}\n"
        )

        core_starts = list(range(x_offset, nx_total, X_CHUNK))
        stitch_results: List[Dict[str, Any]] = []
        all_cluster_records: List[Dict[str, Any]] = []

        f = int(BIN_FACTOR) if (APPLY_BINNING and int(BIN_FACTOR) > 1) else 1
        y_dim = int(rad2d_flat[Y_SLICE, :].shape[0] // f)
        x_dim = int(X_CHUNK // f)

        for slice_no, core_start in enumerate(core_starts):
            core_end = min(core_start + X_CHUNK, nx_total)

            ext_start = max(0, core_start - SLICE_PAD)
            ext_end = min(nx_total, core_end + SLICE_PAD)
            x_slice = slice(ext_start, ext_end)

            outdir = outroot / f"xcore_{core_start:05d}_{core_end:05d}_xext_{ext_start:05d}_{ext_end:05d}"
            print(f"[{file_path.name} | {run_name}] core={core_start}:{core_end} ext={ext_start}:{ext_end}")

            res = run_pipeline_for_slice(
                rad2d_flat=rad2d_flat,
                x_slice=x_slice,
                outdir=outdir,
                x_start_unbinned=ext_start,
                core_start_unbinned=core_start,
                core_end_unbinned=core_end,
                cfg=cfg,
                slice_no=slice_no,
                product_type=product_type,
            )

            stitch_results.append(res)
            all_cluster_records.extend(res.get("cluster_records", []))

            cleanup_memory()

        stitch_and_save_final_09(
            stitch_results,
            outroot,
            tag=f"{run_name}_offset_{x_offset:04d}",
            product_type=product_type,
        )
        print("Done per slice processing. Outputs in", outroot.resolve())

        OUTPUT_L6_FOLDER.mkdir(parents=True, exist_ok=True)
        out_path = OUTPUT_L6_FOLDER / make_output_cluster_path(file_path).name

        copy_lxc_to_l6x_with_clusters(
            lxc_path=file_path,
            l6x_path=out_path,
            clusters=all_cluster_records,
            y_dim=y_dim,
            x_dim=x_dim,
            unbin_factor=(BIN_FACTOR if (APPLY_BINNING and int(BIN_FACTOR) > 1) else 1),
            product_type=product_type,
        )

        print("Saved output with clusters:", out_path.resolve())

    try:
        del rad2d, rad2d_flat
    except Exception:
        pass
    cleanup_memory()


# --------------------------
# Batch loop over folder
# --------------------------
OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)

files = sorted(INPUT_FOLDER.glob(FILE_GLOB))
if not files:
    raise FileNotFoundError(f"No files matched {FILE_GLOB} in {INPUT_FOLDER.resolve()}")

print("Found", len(files), "files in", INPUT_FOLDER.resolve())
print("Will write L6* cluster netCDF to:", OUTPUT_L6_FOLDER.resolve())
print("Will write plot outputs to:", OUTPUT_ROOT.resolve())
print("SAVE_SLICE_PLOTS =", SAVE_SLICE_PLOTS)
print("SAVE_STITCHED_PLOT =", SAVE_STITCHED_PLOT)
print("SAVE_REDNOISE_DIAGNOSTICS =", SAVE_REDNOISE_DIAGNOSTICS)

for fpath in files:
    try:
        run_one_file(fpath)
    except Exception as e:
        print("FAILED:", fpath.name)
        print("Reason:", repr(e))
        cleanup_memory()
        continue

print("\nAll done.")
```

```python
# %%
from __future__ import annotations

from pathlib import Path
from typing import Tuple, Dict, Any, List

import gc
import copy

import numpy as np
import matplotlib
import matplotlib.pyplot as plt
from netCDF4 import Dataset

from juwavelet import transform, utils, plot_utils

try:
    from scipy import stats
except Exception:
    stats = None


# ==========================
# CHANGES (summary)
# ==========================
# 1) Trim leading and trailing columns that are initially all NaN (or non finite) in the RAW frame
#    This happens BEFORE flatten_swath fills NaNs, so you do not carry fake empty edge columns
#    Helps first and last slice not include empty padding
#
# 2) X_OFFSET is interpreted in ORIGINAL x coordinates (global)
#    After trimming, it is converted to trimmed local coordinates and clamped
#
# 3) Stitching uses global x start for correct x axis after trimming
#    slice results carry x_start_unbinned_global
#
# 4) True plotting disable
#    Plotting functions return immediately when saving is off
#
# 5) Stitched plot saving can be fully disabled
#    stitch_and_save_final_09 returns immediately if SAVE_STITCHED_PLOT is False
#
# 6) Red noise diagnostics writing can be disabled
#    No per slice text output when SAVE_REDNOISE_DIAGNOSTICS is False
#
# 7) Red noise filter no longer deep copies the entire cwt dict
#    It shallow copies the dict and copies only decomposition
#
# 8) Fixed batch loop bug
#    continuelo -> continue


# --------------------------
# User params
# --------------------------
INPUT_FOLDER = Path("l3")
FILE_GLOB = "*.nc"

OUTPUT_ROOT = Path("outputs_decompose2d_slices_single_noise_filtering_AUTO_BATCH")
SAVE_DPI = 200

OUTPUT_L6_FOLDER = INPUT_FOLDER.parent / "l6"

FRAME_IDX = 0

X_CHUNK = 600
Y_SLICE = slice(None, None)
SLICE_PAD = 50

X_OFFSET = 0

REMOVE_TREND = True
STANDARDIZE = False
GAUSS_BLUR_SIGMA = 0.0

APPLY_BINNING = True
BIN_FACTOR = 2

APPLY_TAPER = True
TAPER_EDGE_PIXELS = 10

S0 = 16
DJ = 1 / 8
JS = 24
JT = 18
ASPECT = 1

DX_KM_BASE = 2.0
DY_KM_BASE = 2.0

N_SHOW = 6

AMP_MIN_FRACTION_OF_CLUSTER_MAX = 0.10
QUIVER_STEP_X = 20
QUIVER_STEP_Y = 20
QUIVER_SCALE = 1

APPLY_RED_NOISE_FILTER = True

# Turn these on only if you actually want files written
SAVE_REDNOISE_DIAGNOSTICS = False
SAVE_SLICE_PLOTS = False
SAVE_STITCHED_PLOT = True


# --------------------------
# Plot style
# --------------------------
matplotlib.rcParams.update(
    {
        "axes.labelsize": 16,
        "font.size": 16,
        "xtick.labelsize": 14,
        "ytick.labelsize": 14,
        "legend.fontsize": 14,
        "figure.figsize": [2 * 6.94, 2 * 4.29],
    }
)


# --------------------------
# Small helpers
# --------------------------
def save_fig(fig: plt.Figure, outpath: Path, *, enable: bool = True) -> None:
    """
    Save and close a figure if enable=True.
    Always closes the figure to avoid matplotlib memory buildup.
    """
    try:
        if enable:
            outpath.parent.mkdir(parents=True, exist_ok=True)
            fig.savefig(outpath, dpi=SAVE_DPI, bbox_inches="tight")
    finally:
        plt.close(fig)


def cleanup_memory() -> None:
    try:
        plt.close("all")
    except Exception:
        pass
    gc.collect()


def _fmt(v: float) -> str:
    if not np.isfinite(v):
        return "∞"
    v = float(v)
    if abs(v) >= 1000 or abs(v) < 0.1:
        return f"{v:.2e}"
    if abs(v) >= 100:
        return f"{v:.0f}"
    return f"{v:.1f}"


def diverging_limits(*arrays: np.ndarray, p_lo: float = 2, p_hi: float = 98) -> Tuple[float, float]:
    vals = []
    for A in arrays:
        if A is None:
            continue
        a = np.asarray(A)
        good = np.isfinite(a)
        if np.any(good):
            vals.append(a[good].ravel())
    if not vals:
        return (-1.0, 1.0)
    stacked = np.concatenate(vals)
    lo, hi = np.nanpercentile(stacked, [p_lo, p_hi])
    vmax = max(abs(lo), abs(hi))
    if not np.isfinite(vmax) or vmax == 0:
        vmax = 1.0
    return (-vmax, vmax)


def trim_all_nan_edge_columns(Z: np.ndarray) -> Tuple[np.ndarray, int, int]:
    """
    Trim leading and trailing columns that are entirely non finite (all NaN or inf) in the raw 2D frame.
    Returns Z_trimmed, x0, x1_exclusive where x0 is the first kept column index in the original Z.
    """
    if Z.ndim != 2:
        raise ValueError(f"Expected 2D array, got {Z.shape}")

    col_has_data = np.any(np.isfinite(Z), axis=0)
    if not np.any(col_has_data):
        return Z, 0, Z.shape[1]

    x0 = int(np.argmax(col_has_data))
    x1 = int(len(col_has_data) - np.argmax(col_has_data[::-1]))
    return Z[:, x0:x1], x0, x1


# --------------------------
# Product detection + RUNS
# --------------------------
def detect_product_type(file_path: Path) -> str:
    name = file_path.name.lower()
    if "_l3a_" in name:
        return "l3a"
    if "_l3c_" in name:
        return "l3c"
    if "_l5c_" in name:
        return "l5c"
    raise ValueError(f"Cannot detect product type from file name: {file_path.name}")


def get_runs_for_product(product_type: str) -> List[Dict[str, Any]]:
    if product_type == "l3a":
        return [
            dict(
                name="base",
                MIN_AMP=1.25,
                THR=0.6,
                WHITE_NOISE_THRESHOLD=0.1,
                WPS_NOISE_THRESHOLD=8.0,
            ),
        ]
    return [
        dict(
            name="base",
            MIN_AMP=0.2,
            THR=0.15,
            WHITE_NOISE_THRESHOLD=0.02,
            WPS_NOISE_THRESHOLD=8.0,
        ),
    ]

# --------------------------
# Data helpers
# --------------------------
def read_frame(path: str | Path, frame_idx: int, product_type: str) -> np.ndarray:
    with Dataset(str(path), "r") as nc:
        varname = "Temperature" if product_type == "l3a" else "Radiance"
        data = nc.variables[varname][:]  # (t, y, x)

    if data.ndim != 3:
        raise ValueError(f"{varname} expected 3D (t,y,x); got {data.shape}")

    if not (0 <= frame_idx < data.shape[0]):
        raise IndexError(f"FRAME_IDX={frame_idx} out of range 0..{data.shape[0]-1}")

    return np.asarray(data[frame_idx, :, :], dtype=float)


def flatten_swath(Z: np.ndarray, target_ny: int = 300) -> np.ndarray:
    ny, nx = Z.shape
    out = np.full_like(Z, np.nan)

    bottoms = np.full(nx, np.nan, dtype=float)
    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size:
            bottoms[j] = valid_idx.max()

    valid_bottoms = bottoms[np.isfinite(bottoms)]
    if valid_bottoms.size == 0:
        med = float(np.nanmedian(Z)) if np.isfinite(np.nanmedian(Z)) else 0.0
        out = np.nan_to_num(Z, nan=med, posinf=med, neginf=med)
        if target_ny is not None and ny > target_ny:
            out = out[-target_ny:, :]
        return out

    max_bottom = int(valid_bottoms.max())

    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size == 0:
            continue
        bottom = int(valid_idx.max())
        shift = max_bottom - bottom
        if shift == 0:
            out[:, j] = col
        elif shift > 0:
            out[shift:, j] = col[: ny - shift]
        else:
            shift_up = -shift
            out[: ny - shift_up, j] = col[shift_up:]

    med = float(np.nanmedian(out)) if np.isfinite(np.nanmedian(out)) else 0.0
    out = np.nan_to_num(out, nan=med, posinf=med, neginf=med)

    if target_ny is not None and ny > target_ny:
        start = max_bottom + 1 - target_ny
        start = max(start, 0)
        end = start + target_ny
        out = out[start:end, :]

    return out


def bin2d_block_mean(Z: np.ndarray, factor: int) -> np.ndarray:
    if factor is None or int(factor) <= 1:
        return Z
    f = int(factor)
    ny, nx = Z.shape
    ny2 = (ny // f) * f
    nx2 = (nx // f) * f
    if ny2 == 0 or nx2 == 0:
        raise ValueError(f"Binning factor {f} too large for array {Z.shape}")
    Z2 = Z[:ny2, :nx2]
    return Z2.reshape(ny2 // f, f, nx2 // f, f).mean(axis=(1, 3))


def unbin_repeat_xy(A: np.ndarray, factor: int) -> np.ndarray:
    if factor is None or int(factor) <= 1:
        return A
    f = int(factor)
    return np.repeat(np.repeat(A, f, axis=0), f, axis=1)


def fit_plane(Z: np.ndarray) -> np.ndarray:
    ny, nx = Z.shape
    yy, xx = np.mgrid[0:ny, 0:nx]
    G = np.stack([xx.ravel(), yy.ravel(), np.ones(nx * ny)], axis=1)
    coeff, *_ = np.linalg.lstsq(G, Z.ravel(), rcond=None)
    return coeff[0] * xx + coeff[1] * yy + coeff[2]


def extract_first_fourier_bg(Z: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
    ny, nx = Z.shape
    F = np.fft.fft2(Z)
    F_bg = np.zeros_like(F, dtype=complex)

    if nx > 1:
        F_bg[0, 1] = F[0, 1]
        F_bg[0, -1] = F[0, -1]
    if ny > 1:
        F_bg[1, 0] = F[1, 0]
        F_bg[-1, 0] = F[-1, 0]

    bg = np.fft.ifft2(F_bg).real
    return bg, Z - bg


def maybe_blur(Z: np.ndarray, sigma: float) -> np.ndarray:
    if sigma <= 0:
        return Z
    try:
        from scipy.ndimage import gaussian_filter
    except Exception:
        return Z
    return gaussian_filter(Z, sigma=float(sigma))


def _hanning_edge_1d(n: int, edge: int) -> np.ndarray:
    w = np.ones(n, dtype=float)
    if edge <= 0:
        return w
    edge = min(edge, n // 2)
    ramp = 0.5 * (1.0 - np.cos(np.linspace(0, np.pi, edge, endpoint=True)))
    w[:edge] = ramp
    w[-edge:] = ramp[::-1]
    return w


def apply_edge_hanning_taper(Z: np.ndarray, edge: int) -> np.ndarray:
    if edge <= 0:
        return Z
    ny, nx = Z.shape
    wx = _hanning_edge_1d(nx, edge)
    wy = _hanning_edge_1d(ny, edge)
    W = np.outer(wy, wx)
    return Z * W


def effective_lambda(cwt: dict, s_idx: int, t_idx: int) -> Tuple[float, float, float]:
    lamx = float(cwt["wavelength_x"][s_idx, t_idx])
    lamy = float(cwt["wavelength_y"][s_idx, t_idx])
    inv2 = 0.0
    if np.isfinite(lamx) and lamx != 0:
        inv2 += (1.0 / lamx) ** 2
    if np.isfinite(lamy) and lamy != 0:
        inv2 += (1.0 / lamy) ** 2
    lam = np.inf if inv2 == 0 else 1.0 / np.sqrt(inv2)
    return lamx, lamy, lam


def compute_cluster_power_table(cwt: dict, iwave: np.ndarray) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    dec = cwt["decomposition"]
    power4d = np.abs(dec) ** 2

    cluster_ids = np.unique(iwave[iwave >= 0]).astype(int)
    if cluster_ids.size == 0:
        return cluster_ids, np.array([]), np.array([])

    P_abs = np.zeros(cluster_ids.size, dtype=float)
    for i, kcl in enumerate(cluster_ids):
        P_abs[i] = float(np.sum(power4d[iwave == kcl]))

    total = float(np.sum(P_abs))
    if total > 0 and np.isfinite(total):
        P_pct = 100.0 * P_abs / total
    else:
        P_pct = np.full_like(P_abs, np.nan)

    return cluster_ids, P_abs, P_pct


def cluster_amp_theta_maps(
    cwt: dict,
    iwave: np.ndarray,
    kcl: int,
    amp_min_fraction_of_cluster_max: float = 0.2,
) -> Tuple[np.ndarray, np.ndarray]:
    dec = cwt["decomposition"]  # (ns, nt, nx, ny) complex
    power = np.abs(dec) ** 2

    mask4d = (iwave == kcl)
    if not np.any(mask4d):
        ny = dec.shape[3]
        nx = dec.shape[2]
        return np.full((ny, nx), np.nan), np.full((ny, nx), np.nan)

    ns, nt, nx, ny = power.shape
    power2 = power.reshape(ns * nt, nx * ny)
    mask2 = mask4d.reshape(ns * nt, nx * ny)

    p = np.where(mask2, power2, -np.inf)
    pmax = np.max(p, axis=0)

    valid = np.isfinite(pmax) & (pmax > 0)

    A = np.full(nx * ny, np.nan, dtype=float)
    theta = np.full(nx * ny, np.nan, dtype=float)

    if np.any(valid):
        arg = np.argmax(p[:, valid], axis=0)
        t_idx = arg % nt

        A_valid = np.sqrt(pmax[valid])
        theta_valid = np.asarray(cwt["theta"], dtype=float)[t_idx]

        A[valid] = A_valid
        theta[valid] = theta_valid

    A_xy = A.reshape(nx, ny).T
    theta_xy = theta.reshape(nx, ny).T

    if np.isfinite(np.nanmax(A_xy)):
        thr = float(amp_min_fraction_of_cluster_max) * float(np.nanmax(A_xy))
        low = A_xy < thr
        A_xy[low] = np.nan
        theta_xy[low] = np.nan

    return A_xy, theta_xy


def theta_to_uv(theta_xy: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
    return np.cos(theta_xy), np.sin(theta_xy)


# --------------------------
# Red noise style filtering helper
# --------------------------
def apply_red_noise_filter_cwt(
    cwt: dict,
    white_noise_threshold: float = 0.1,
    wps_noise_threshold: float = 8.0,
) -> Tuple[dict, Dict[str, Any]]:
    """
    Robust per scale WPS filtering.
    Returns filtered_cwt and diagnostics.

    Change: no deepcopy of entire cwt dict.
    Only decomposition is copied since that is what we mutate.
    """
    diag: Dict[str, Any] = {}
    if stats is None:
        diag["skipped"] = True
        diag["reason"] = "scipy not available"
        return cwt, diag

    cwt_copy = dict(cwt)
    dec0 = cwt["decomposition"]
    dec = dec0.copy()
    cwt_copy["decomposition"] = dec

    WPS0 = np.abs(dec) ** 2
    diag["n_coeff_total"] = int(WPS0.size)

    mask_1 = WPS0 < (float(white_noise_threshold) ** 2)
    dec[mask_1] = 0

    WPS1 = np.abs(dec) ** 2
    diag["n_zero_after_white"] = int(np.count_nonzero(WPS1 == 0))

    median_WPS = np.median(WPS1, axis=(1, 2, 3))  # (ns,)
    mad = stats.median_abs_deviation(WPS1, axis=(1, 2, 3), scale=1.0)  # (ns,)
    sMAD_WPS = 1.4826 * mad

    sMAD_safe = np.where(sMAD_WPS > 0, sMAD_WPS, np.nan)
    WPS_scaled = (WPS1 - median_WPS[:, None, None, None]) / sMAD_safe[:, None, None, None]

    mask = WPS_scaled < float(wps_noise_threshold)
    dec[mask] = 0

    WPS2 = np.abs(dec) ** 2
    diag["n_zero_after_scaled"] = int(np.count_nonzero(WPS2 == 0))

    finite_scaled = np.isfinite(WPS_scaled)

    diag["sMAD_min"] = float(np.nanmin(sMAD_WPS))
    diag["sMAD_median"] = float(np.nanmedian(sMAD_WPS))

    if np.any(finite_scaled):
        diag["WPS_scaled_min"] = float(np.nanmin(WPS_scaled[finite_scaled]))
        diag["WPS_scaled_median"] = float(np.nanmedian(WPS_scaled[finite_scaled]))
        diag["WPS_scaled_max"] = float(np.nanmax(WPS_scaled[finite_scaled]))
    else:
        diag["WPS_scaled_min"] = np.nan
        diag["WPS_scaled_median"] = np.nan
        diag["WPS_scaled_max"] = np.nan

    return cwt_copy, diag


# --------------------------
# Quicklook plots (true no op if not saving)
# --------------------------
def quicklook_save(
    x1d: np.ndarray,
    y1d: np.ndarray,
    Z: np.ndarray,
    title: str,
    outpath: Path,
    *,
    save_plots: bool,
) -> None:
    if not save_plots:
        return
    fig, ax = plt.subplots()
    im = ax.pcolormesh(x1d, y1d, Z, shading="auto", cmap="gray")
    fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04).set_label("arb")
    ax.set_xlabel("x (km)")
    ax.set_ylabel("y (km)")
    ax.set_title(title)
    ax.set_aspect("equal", adjustable="box")
    fig.tight_layout()
    save_fig(fig, outpath, enable=True)


def quicklook_background_decomp_4panel_save(
    x1d_km: np.ndarray,
    y1d_km: np.ndarray,
    original: np.ndarray,
    background: np.ndarray,
    remaining: np.ndarray,
    remaining_tapered: np.ndarray,
    taper_edge_px: int,
    outpath: Path,
    *,
    wavy_after_noise_xy: np.ndarray | None = None,
    save_plots: bool,
) -> None:
    if not save_plots:
        return

    med = np.nanmedian(original)
    orig_c = original - med
    bg_c = background - med
    rem_c = remaining
    rem_tapered_c = remaining_tapered

    arrays = [orig_c, bg_c, rem_c, rem_tapered_c]
    if wavy_after_noise_xy is not None:
        arrays.append(wavy_after_noise_xy)

    vmin, vmax = diverging_limits(*arrays, p_lo=1, p_hi=99)

    fig, axs = plt.subplots(2, 3, figsize=(16, 10), constrained_layout=True)
    axs = axs.flatten()

    panels = [
        (orig_c, "Original (median sub)"),
        (bg_c, "Background (plane + first Fourier)"),
        (rem_c, "Remaining = original - background"),
        (rem_tapered_c, f"Remaining with edge taper (width={taper_edge_px})"),
    ]

    if wavy_after_noise_xy is not None:
        panels.append((wavy_after_noise_xy, "Wavy stuff (wavelet reconstruction after noise filtering)"))

    im0 = None
    for ax, (Zp, title) in zip(axs, panels):
        im0 = ax.pcolormesh(x1d_km, y1d_km, Zp, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
        ax.set_title(title)
        ax.set_xlabel("x (km)")
        ax.set_ylabel("y (km)")
        ax.set_aspect("equal", adjustable="box")

    for ax in axs[len(panels) :]:
        ax.axis("off")

    cbar = fig.colorbar(im0, ax=axs[: len(panels)], orientation="vertical", fraction=0.04, pad=0.02)
    cbar.set_label("arb")
    save_fig(fig, outpath, enable=True)


def plot_cluster_individual_save(
    cwt: dict,
    iwave: np.ndarray,
    kcl: int,
    x1d_km: np.ndarray,
    y1d_km: np.ndarray,
    orig_decomp: np.ndarray,
    bg_xy: np.ndarray,
    dx_km: float,
    dy_km: float,
    outpath: Path,
    *,
    save_plots: bool,
) -> None:
    if not save_plots:
        return

    decomposition = cwt["decomposition"]
    decomposition[:] = orig_decomp
    decomposition[iwave != kcl] = 0

    rec = transform.reconstruct2d(cwt)  # (nx, ny)
    rec_xy = rec.T - np.median(rec)

    A_xy, theta_xy = cluster_amp_theta_maps(
        cwt, iwave, kcl, amp_min_fraction_of_cluster_max=AMP_MIN_FRACTION_OF_CLUSTER_MAX
    )

    dec0 = orig_decomp
    power0 = np.abs(dec0) ** 2
    mask4d = (iwave == kcl)
    ns, nt, nx, ny = power0.shape

    power2 = power0.reshape(ns * nt, nx * ny)
    mask2 = mask4d.reshape(ns * nt, nx * ny)
    p = np.where(mask2, power2, -np.inf)
    pmax = np.max(p, axis=0)

    lam_xy = np.full(nx * ny, np.nan, dtype=float)
    valid = np.isfinite(pmax) & (pmax > 0)
    if np.any(valid):
        arg = np.argmax(p[:, valid], axis=0)
        s_idx = arg // nt
        t_idx = arg % nt

        lamx = np.asarray(cwt["wavelength_x"], dtype=float)[s_idx, t_idx]
        lamy = np.asarray(cwt["wavelength_y"], dtype=float)[s_idx, t_idx]

        inv2 = np.zeros_like(lamx, dtype=float)
        goodx = np.isfinite(lamx) & (lamx != 0)
        goody = np.isfinite(lamy) & (lamy != 0)
        inv2[goodx] += (1.0 / lamx[goodx]) ** 2
        inv2[goody] += (1.0 / lamy[goody]) ** 2

        lam = np.full_like(inv2, np.nan, dtype=float)
        ok = inv2 > 0
        lam[ok] = 1.0 / np.sqrt(inv2[ok])
        lam_xy[valid] = lam

    lam_xy = lam_xy.reshape(nx, ny).T

    if np.isfinite(np.nanmax(A_xy)):
        thr = float(AMP_MIN_FRACTION_OF_CLUSTER_MAX) * float(np.nanmax(A_xy))
        lam_xy[A_xy < thr] = np.nan

    bg_vmin, bg_vmax = diverging_limits(bg_xy, p_lo=2, p_hi=98)
    cl_vmin, cl_vmax = diverging_limits(rec_xy, p_lo=2, p_hi=98)

    theta_vals = np.asarray(cwt["theta"], dtype=float)
    theta_min = float(np.nanmin(theta_vals))
    theta_max = float(np.nanmax(theta_vals))

    if np.any(np.isfinite(lam_xy)):
        lam_vmin, lam_vmax = np.nanpercentile(lam_xy, [5, 95])
        if (not np.isfinite(lam_vmin)) or (not np.isfinite(lam_vmax)) or (lam_vmin == lam_vmax):
            lam_vmin = float(np.nanmin(lam_xy))
            lam_vmax = float(np.nanmax(lam_xy))
    else:
        lam_vmin, lam_vmax = 0.0, 1.0

    fig, axs = plt.subplots(2, 2, figsize=(16, 11), constrained_layout=True)
    ax00, ax01, ax10, ax11 = axs[0, 0], axs[0, 1], axs[1, 0], axs[1, 1]

    ax00.pcolormesh(x1d_km, y1d_km, bg_xy, shading="auto", cmap="gray", vmin=bg_vmin, vmax=bg_vmax)
    im00 = ax00.pcolormesh(
        x1d_km, y1d_km, rec_xy, shading="auto", cmap="RdBu_r", vmin=cl_vmin, vmax=cl_vmax, alpha=0.6
    )
    ax00.set_title(f"Cluster {kcl} overlay")
    ax00.set_xlabel("x (km)")
    ax00.set_ylabel("y (km)")
    ax00.set_aspect("equal", adjustable="box")
    fig.colorbar(im00, ax=ax00, fraction=0.046, pad=0.02)

    im01 = ax01.pcolormesh(x1d_km, y1d_km, A_xy, shading="auto", cmap="plasma")
    n_pix = int(np.count_nonzero(np.isfinite(A_xy)))
    area_km2 = n_pix * float(dx_km) * float(dy_km)
    ax01.set_title("A(x,y) = max |coeff| in cluster\n" + rf"Area = {area_km2:.0f} km$^2$ ({n_pix} px)")
    ax01.set_xlabel("x (km)")
    ax01.set_ylabel("y (km)")
    ax01.set_aspect("equal", adjustable="box")
    fig.colorbar(im01, ax=ax01, fraction=0.046, pad=0.02)

    im10 = ax10.pcolormesh(x1d_km, y1d_km, lam_xy, shading="auto", cmap="viridis", vmin=lam_vmin, vmax=lam_vmax)
    if np.any(np.isfinite(lam_xy)):
        lam_med = float(np.nanmedian(lam_xy))
        lam_p10, lam_p90 = np.nanpercentile(lam_xy, [10, 90])
        ax10.set_title(
            r"Dominant $\lambda(x,y)$ in cluster" + "\n"
            + rf"median={_fmt(lam_med)} km (10–90%: {_fmt(lam_p10)}–{_fmt(lam_p90)} km)"
        )
    else:
        ax10.set_title(r"Dominant $\lambda(x,y)$ in cluster")
    ax10.set_xlabel("x (km)")
    ax10.set_ylabel("y (km)")
    ax10.set_aspect("equal", adjustable="box")
    fig.colorbar(im10, ax=ax10, fraction=0.046, pad=0.02).set_label("km")

    im11 = ax11.pcolormesh(x1d_km, y1d_km, theta_xy, shading="auto", cmap="PiYG", vmin=theta_min, vmax=theta_max)
    ax11.set_title("θ(x,y) + quiver")
    ax11.set_xlabel("x (km)")
    ax11.set_ylabel("y (km)")
    ax11.set_aspect("equal", adjustable="box")
    fig.colorbar(im11, ax=ax11, fraction=0.046, pad=0.02)

    if np.any(np.isfinite(theta_xy)):
        u, v = theta_to_uv(theta_xy)
        arrow_len_km = 40.0
        u = u * arrow_len_km
        v = v * arrow_len_km

        Xq, Yq = np.meshgrid(x1d_km, y1d_km)
        Xqs = Xq[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        Yqs = Yq[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        uqs = u[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        vqs = v[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        good = np.isfinite(uqs) & np.isfinite(vqs)
        ax11.quiver(
            Xqs[good],
            Yqs[good],
            uqs[good],
            vqs[good],
            angles="xy",
            scale_units="xy",
            scale=QUIVER_SCALE,
            width=0.0025,
        )

    save_fig(fig, outpath, enable=True)


# --------------------------
# Cluster products (for L6C/L6A outputs)
# --------------------------
def cluster_dominant_lambda_map(
    cwt: dict,
    iwave: np.ndarray,
    kcl: int,
    amp_mask_xy: np.ndarray | None = None,
) -> np.ndarray:
    dec = cwt["decomposition"]  # (ns, nt, nx, ny) complex
    power0 = np.abs(dec) ** 2
    mask4d = (iwave == kcl)
    if not np.any(mask4d):
        ny = dec.shape[3]
        nx = dec.shape[2]
        return np.full((ny, nx), np.nan)

    ns, nt, nx, ny = power0.shape
    power2 = power0.reshape(ns * nt, nx * ny)
    mask2 = mask4d.reshape(ns * nt, nx * ny)

    p = np.where(mask2, power2, -np.inf)
    pmax = np.max(p, axis=0)
    valid = np.isfinite(pmax) & (pmax > 0)

    lam_xy = np.full(nx * ny, np.nan, dtype=float)
    if np.any(valid):
        arg = np.argmax(p[:, valid], axis=0)
        s_idx = arg // nt
        t_idx = arg % nt

        lamx = np.asarray(cwt["wavelength_x"], dtype=float)[s_idx, t_idx]
        lamy = np.asarray(cwt["wavelength_y"], dtype=float)[s_idx, t_idx]

        inv2 = np.zeros_like(lamx, dtype=float)
        goodx = np.isfinite(lamx) & (lamx != 0)
        goody = np.isfinite(lamy) & (lamy != 0)
        inv2[goodx] += (1.0 / lamx[goodx]) ** 2
        inv2[goody] += (1.0 / lamy[goody]) ** 2

        lam = np.full_like(inv2, np.nan, dtype=float)
        ok = inv2 > 0
        lam[ok] = 1.0 / np.sqrt(inv2[ok])

        lam_xy[valid] = lam

    lam_xy = lam_xy.reshape(nx, ny).T

    if amp_mask_xy is not None:
        lam_xy[~np.isfinite(amp_mask_xy)] = np.nan

    return lam_xy


def copy_lxc_to_l6x_with_clusters(
    lxc_path: str | Path,
    l6x_path: str | Path,
    clusters: List[Dict[str, Any]],
    y_dim: int,
    x_dim: int,
    unbin_factor: int = 1,
    product_type: str = "l5c",
) -> None:
    def _unbin_repeat_xy_local(A: np.ndarray, factor: int) -> np.ndarray:
        if factor is None or int(factor) <= 1:
            return A
        f = int(factor)
        return np.repeat(np.repeat(A, f, axis=0), f, axis=1)

    lxc_path = Path(lxc_path)
    l6x_path = Path(l6x_path)
    l6x_path.parent.mkdir(parents=True, exist_ok=True)

    ncl = int(len(clusters))
    has_rec = any((c.get("rec_xy", None) is not None) for c in clusters)

    fup = int(unbin_factor) if (unbin_factor is not None and int(unbin_factor) > 1) else 1
    y_dim_w = int(y_dim) * fup
    x_dim_w = int(x_dim) * fup

    with Dataset(lxc_path, "r") as src, Dataset(l6x_path, "w", format="NETCDF4") as dst:
        dst.setncatts({k: src.getncattr(k) for k in src.ncattrs()})

        for dname, dim in src.dimensions.items():
            dst.createDimension(dname, (len(dim) if not dim.isunlimited() else None))

        for vname, var in src.variables.items():
            out = dst.createVariable(
                vname,
                var.datatype,
                var.dimensions,
                zlib=True,
                complevel=4,
                fill_value=getattr(var, "_FillValue", None),
            )
            out.setncatts({k: var.getncattr(k) for k in var.ncattrs()})
            out[:] = var[:]

        dst.createDimension("cluster", ncl)
        dst.createDimension("y_cluster", y_dim_w)
        dst.createDimension("x_cluster", x_dim_w)

        v_slice = dst.createVariable("SlicesNo", "i4", ("cluster",), zlib=True, complevel=4)
        v_area = dst.createVariable("Area", "f4", ("cluster",), zlib=True, complevel=4)
        v_pabs = dst.createVariable("ClusterPowerAbs", "f8", ("cluster",), zlib=True, complevel=4)
        v_ppct = dst.createVariable("ClusterPowerPct", "f4", ("cluster",), zlib=True, complevel=4)

        v_amp = dst.createVariable(
            "Amplitude",
            "f4",
            ("cluster", "y_cluster", "x_cluster"),
            zlib=True,
            complevel=4,
            fill_value=np.float32(np.nan),
        )
        v_lam = dst.createVariable(
            "DominantWavelength",
            "f4",
            ("cluster", "y_cluster", "x_cluster"),
            zlib=True,
            complevel=4,
            fill_value=np.float32(np.nan),
        )
        v_ang = dst.createVariable(
            "Angle",
            "f4",
            ("cluster", "y_cluster", "x_cluster"),
            zlib=True,
            complevel=4,
            fill_value=np.float32(np.nan),
        )

        v_rec = None
        if has_rec:
            v_rec = dst.createVariable(
                "ClusterReconstruction",
                "f4",
                ("cluster", "y_cluster", "x_cluster"),
                zlib=True,
                complevel=4,
                fill_value=np.float32(np.nan),
            )
            v_rec.units = "arb"
            v_rec.long_name = "Per cluster wavelet reconstruction map (relative; median subtracted)"

        v_area.units = "km2"
        v_lam.units = "km"
        v_ang.units = "rad"
        v_slice.long_name = "Slice index (starts at 0) for each cluster"
        v_pabs.units = "arb"
        v_ppct.units = "percent"

        v_amp.units = "K" if product_type == "l3a" else "kR"

        if ncl == 0:
            return

        v_slice[:] = np.asarray([c["slice_no"] for c in clusters], dtype=np.int32)
        v_area[:] = np.asarray([c["area_km2"] for c in clusters], dtype=np.float32)
        v_pabs[:] = np.asarray([c.get("power_abs", np.nan) for c in clusters], dtype=np.float64)
        v_ppct[:] = np.asarray([c.get("power_pct", np.nan) for c in clusters], dtype=np.float32)

        amp_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32)
        lam_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32)
        ang_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32)
        rec_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32) if has_rec else None

        def _force_shape(M: np.ndarray) -> np.ndarray:
            if M.shape == (y_dim_w, x_dim_w):
                return M.astype(np.float32)
            tmp = np.full((y_dim_w, x_dim_w), np.nan, dtype=np.float32)
            yy = min(y_dim_w, M.shape[0])
            xx = min(x_dim_w, M.shape[1])
            tmp[:yy, :xx] = M[:yy, :xx]
            return tmp

        for i, c in enumerate(clusters):
            A = _unbin_repeat_xy_local(c["A_xy"], fup)
            L = _unbin_repeat_xy_local(c["lam_xy"], fup)
            T = _unbin_repeat_xy_local(c["theta_xy"], fup)

            amp_stack[i] = _force_shape(A)
            lam_stack[i] = _force_shape(L)
            ang_stack[i] = _force_shape(T)

            if has_rec and rec_stack is not None and v_rec is not None:
                R = c.get("rec_xy", None)
                if R is not None:
                    R = _force_shape(_unbin_repeat_xy_local(R, fup))
                    rec_stack[i] = R

        v_amp[:] = amp_stack
        v_lam[:] = lam_stack
        v_ang[:] = ang_stack
        if has_rec and v_rec is not None and rec_stack is not None:
            v_rec[:] = rec_stack


def make_output_cluster_path(input_path: str | Path) -> Path:
    p = Path(input_path)
    name = p.name
    if "_l5c_" in name:
        name = name.replace("_l5c_", "_l6c_")
    elif "_l3c_" in name:
        name = name.replace("_l3c_", "_l6c_")
    elif "_l3a_" in name:
        name = name.replace("_l3a_", "_l6a_")
    else:
        name = p.stem + "_clusters.nc"
    return p.with_name(name)


# --------------------------
# Slice pipeline
# --------------------------
def run_pipeline_for_slice(
    rad2d_flat: np.ndarray,
    x_slice: slice,
    outdir: Path,
    x_start_unbinned: int,
    core_start_unbinned: int,
    core_end_unbinned: int,
    x_base_unbinned: int,
    cfg: Dict[str, Any],
    slice_no: int,
    product_type: str,
) -> Dict[str, Any]:
    outdir.mkdir(parents=True, exist_ok=True)
    save_plots = bool(SAVE_SLICE_PLOTS)

    min_amp = float(cfg.get("MIN_AMP"))
    thr = float(cfg.get("THR"))
    white_noise_threshold = float(cfg.get("WHITE_NOISE_THRESHOLD"))
    wps_noise_threshold = float(cfg.get("WPS_NOISE_THRESHOLD"))
    run_name = str(cfg.get("name", "run"))

    STORE_CLUSTER_RECON = True

    if APPLY_BINNING and int(BIN_FACTOR) > 1:
        f = int(BIN_FACTOR)
        dx_km = DX_KM_BASE * f
        dy_km = DY_KM_BASE * f
    else:
        f = 1
        dx_km = DX_KM_BASE
        dy_km = DY_KM_BASE

    core_width_unbinned = int(core_end_unbinned - core_start_unbinned)
    core_w_binned = core_width_unbinned // f
    rel0_unbinned = int(core_start_unbinned - x_start_unbinned)
    rel0_binned = rel0_unbinned // f
    rel1_binned = rel0_binned + core_w_binned

    def crop_pad_core(A: np.ndarray) -> np.ndarray:
        nyA, nxA = A.shape
        left = max(0, rel0_binned)
        right = min(nxA, rel1_binned)
        core = A[:, left:right]
        need = core_w_binned - core.shape[1]
        if need > 0:
            core = np.concatenate([core, np.full((nyA, need), np.nan)], axis=1)
        elif need < 0:
            core = core[:, :core_w_binned]
        return core

    raw_win = np.asarray(rad2d_flat[Y_SLICE, x_slice], dtype=float)

    if APPLY_BINNING and int(BIN_FACTOR) > 1:
        raw_win = bin2d_block_mean(raw_win, BIN_FACTOR)

    ny, nx = raw_win.shape
    x1d_km = np.arange(nx, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    bin_tag = f" (binned {BIN_FACTOR}x{BIN_FACTOR})" if (APPLY_BINNING and int(BIN_FACTOR) > 1) else ""
    label0 = "Temperature" if product_type == "l3a" else "Radiance"

    quicklook_save(
        x1d_km,
        y1d_km,
        raw_win,
        f"{label0} (window, flattened){bin_tag}\n{run_name} MIN_AMP={min_amp} THR={thr}",
        outdir / "00_window.png",
        save_plots=save_plots,
    )

    proc0 = raw_win.astype(float, copy=True)

    if REMOVE_TREND:
        plane = fit_plane(proc0)
        detrended = proc0 - plane
    else:
        plane = np.zeros_like(proc0)
        detrended = proc0

    fourier_bg, _ = extract_first_fourier_bg(detrended)
    background = plane + fourier_bg
    remaining = proc0 - background

    if APPLY_TAPER:
        remaining_tapered = apply_edge_hanning_taper(remaining, TAPER_EDGE_PIXELS)
    else:
        remaining_tapered = remaining.copy()

    proc = remaining_tapered.copy()
    if STANDARDIZE:
        m = float(np.median(proc))
        s = float(np.std(proc))
        proc = (proc - m) / s if s > 0 else (proc - m)
    proc = maybe_blur(proc, GAUSS_BLUR_SIGMA)

    wave_xy = proc.T  # (nx, ny)

    cwt_raw = transform.decompose2d(
        wave_xy, dx=dx_km, dy=dy_km, s0=S0, dj=DJ, js=JS, jt=JT, aspect=ASPECT
    )

    if APPLY_RED_NOISE_FILTER:
        cwt, diag = apply_red_noise_filter_cwt(
            cwt_raw,
            white_noise_threshold=white_noise_threshold,
            wps_noise_threshold=wps_noise_threshold,
        )

        if SAVE_REDNOISE_DIAGNOSTICS:
            lines = []
            lines.append("Red noise filter diagnostics\n\n")
            lines.append(f"name                  = {run_name}\n")
            lines.append(f"WHITE_NOISE_THRESHOLD = {white_noise_threshold}\n")
            lines.append(f"WPS_NOISE_THRESHOLD   = {wps_noise_threshold}\n\n")
            if diag.get("skipped", False):
                lines.append(f"SKIPPED: {diag.get('reason','unknown')}\n")
            else:
                lines.append(f"n_coeff_total        = {diag.get('n_coeff_total')}\n")
                lines.append(f"n_zero_after_white   = {diag.get('n_zero_after_white')}\n")
                lines.append(f"n_zero_after_scaled  = {diag.get('n_zero_after_scaled')}\n")
                lines.append(f"sMAD_WPS min         = {diag.get('sMAD_min')}\n")
                lines.append(f"sMAD_WPS median      = {diag.get('sMAD_median')}\n")
                lines.append(f"WPS_scaled min       = {diag.get('WPS_scaled_min')}\n")
                lines.append(f"WPS_scaled median    = {diag.get('WPS_scaled_median')}\n")
                lines.append(f"WPS_scaled max       = {diag.get('WPS_scaled_max')}\n")
            (outdir / "02a_rednoise_diagnostics.txt").write_text("".join(lines))
    else:
        cwt = cwt_raw

    rec_wavy = transform.reconstruct2d(cwt)  # (nx, ny)
    wavy_xy = rec_wavy.T - np.median(rec_wavy)

    quicklook_background_decomp_4panel_save(
        x1d_km,
        y1d_km,
        proc0,
        background,
        remaining,
        remaining_tapered,
        TAPER_EDGE_PIXELS,
        outdir / "01_background_decomp_5panel.png",
        wavy_after_noise_xy=wavy_xy,
        save_plots=save_plots,
    )

    amps, idxs, iwave = utils.identify_cluster2d(cwt, min_amp=min_amp, thr=thr)
    decomposition = cwt["decomposition"]
    orig_decomp = decomposition.copy()

    if save_plots:
        cmap_cwt = matplotlib.cm.turbo
        norm_cwt = matplotlib.colors.BoundaryNorm(np.exp(np.linspace(np.log(0.05), np.log(1.0), 10)), cmap_cwt.N)
        fig_cwt, _ = utils.plot_decomposition2d(cwt, redux_s=2, redux_t=2, cmap=cmap_cwt, norm=norm_cwt)
        save_fig(fig_cwt, outdir / "02_decomposition_overview.png", enable=True)

        plot_utils.plot_fov_wavelet_spectrum(cwt, stat="max")
        fig_spec = plt.gcf()
        save_fig(fig_spec, outdir / "03_fov_wavelet_spectrum.png", enable=True)

    bg = remaining_tapered - np.median(remaining_tapered)

    cluster_ids, P_abs, P_pct = compute_cluster_power_table(cwt, iwave)
    if cluster_ids.size == 0:
        (outdir / "NO_CLUSTERS.txt").write_text(
            "No clusters found.\n\n"
            f"Run name: {run_name}\n"
            f"Used MIN_AMP={min_amp}, THR={thr}\n"
            f"Used WHITE_NOISE_THRESHOLD={white_noise_threshold}, WPS_NOISE_THRESHOLD={wps_noise_threshold}\n\n"
            "Try lowering MIN_AMP or THR.\n"
            "If red noise filter is on, try lowering WPS_NOISE_THRESHOLD or WHITE_NOISE_THRESHOLD.\n"
        )
        remaining_core_xy = crop_pad_core(proc)
        nan_core = np.full_like(remaining_core_xy, np.nan)

        cleanup_memory()
        return {
            "has_clusters": False,
            "x_start_unbinned": int(core_start_unbinned),
            "x_start_unbinned_global": int(core_start_unbinned + x_base_unbinned),
            "dx_km": dx_km,
            "dy_km": dy_km,
            "remaining_xy": remaining_core_xy,
            "wavy_xy": nan_core,
            "rec_all_xy": nan_core,
            "cluster_records": [],
        }

    power_abs_map = {int(k): float(p) for k, p in zip(cluster_ids, P_abs)}
    power_pct_map = {int(k): float(p) for k, p in zip(cluster_ids, P_pct)}

    cluster_records: List[Dict[str, Any]] = []
    for kcl in cluster_ids.astype(int):
        kcl = int(kcl)

        A_xy, theta_xy = cluster_amp_theta_maps(
            cwt,
            iwave,
            kcl,
            amp_min_fraction_of_cluster_max=AMP_MIN_FRACTION_OF_CLUSTER_MAX,
        )
        lam_xy = cluster_dominant_lambda_map(cwt, iwave, kcl, amp_mask_xy=A_xy)

        rec_core = None
        if STORE_CLUSTER_RECON:
            decomposition[:] = orig_decomp
            decomposition[iwave != kcl] = 0
            rec = transform.reconstruct2d(cwt)  # (nx, ny)
            rec_xy = rec.T
            rec_xy = rec_xy - np.nanmedian(rec_xy)
            rec_core = crop_pad_core(rec_xy)
            decomposition[:] = orig_decomp

        A_core = crop_pad_core(A_xy)
        th_core = crop_pad_core(theta_xy)
        lam_core = crop_pad_core(lam_xy)

        n_pix = int(np.count_nonzero(np.isfinite(A_core)))
        area_km2 = float(n_pix) * float(dx_km) * float(dy_km)

        cluster_records.append(
            dict(
                slice_no=int(slice_no),
                kcl=kcl,
                area_km2=area_km2,
                power_abs=power_abs_map.get(kcl, np.nan),
                power_pct=power_pct_map.get(kcl, np.nan),
                A_xy=A_core,
                lam_xy=lam_core,
                theta_xy=th_core,
                rec_xy=rec_core,
            )
        )

    order_power = np.argsort(P_abs)[::-1]
    cluster_ids_sorted = cluster_ids[order_power]
    P_abs_sorted = P_abs[order_power]
    P_pct_sorted = P_pct[order_power]

    lines = []
    lines.append("Cluster power ranking (clustered only)\n\n")
    lines.append(f"name = {run_name}\n")
    lines.append(f"MIN_AMP = {min_amp}\n")
    lines.append(f"THR = {thr}\n")
    lines.append(f"WHITE_NOISE_THRESHOLD = {white_noise_threshold}\n")
    lines.append(f"WPS_NOISE_THRESHOLD = {wps_noise_threshold}\n\n")
    lines.append("rank  kcl      P_abs           P_pct\n")
    for r, (kcl, Pk, pk) in enumerate(zip(cluster_ids_sorted, P_abs_sorted, P_pct_sorted), start=1):
        lines.append(f"{r:4d}  {int(kcl):3d}  {Pk: .6e}   {pk:6.2f}%\n")
    (outdir / "04_cluster_power_ranking.txt").write_text("".join(lines))

    n_show = int(min(N_SHOW, cluster_ids_sorted.size))
    biggest = cluster_ids_sorted[:n_show]
    theta_vals = np.asarray(cwt["theta"], dtype=float)

    if save_plots:
        nrow = int(np.ceil(np.sqrt(n_show)))
        ncol = int(np.ceil(n_show / nrow))

        def plot_cluster_grid(kind: str, outname: str) -> None:
            fig, axs = plt.subplots(nrow, ncol, figsize=(12, 10))
            axs = np.atleast_1d(axs).reshape(-1)

            bg_vmin, bg_vmax = diverging_limits(bg, p_lo=2, p_hi=98)
            number_box = {"boxstyle": "square", "lw": 0.67, "facecolor": "white", "edgecolor": "black"}

            for ax, kcl in zip(axs, biggest):
                decomposition[:] = orig_decomp
                decomposition[iwave != kcl] = 0
                rec = transform.reconstruct2d(cwt)
                rec_centered = rec.T - np.median(rec)

                s_idx = int(idxs[kcl][0])
                t_idx = int(idxs[kcl][1])
                _, _, lam = effective_lambda(cwt, s_idx, t_idx)

                theta_dom_rad = float(theta_vals[t_idx])
                theta_dom_deg = (np.degrees(theta_dom_rad) + 180.0) % 180.0

                ii = int(np.where(cluster_ids_sorted == kcl)[0][0])
                Pk = float(P_abs_sorted[ii])
                pk = float(P_pct_sorted[ii])

                if kind == "standalone":
                    ax.pcolormesh(x1d_km, y1d_km, rec_centered, shading="auto", cmap="RdBu_r", vmin=-1.0, vmax=1.0)
                elif kind == "overlay":
                    ax.pcolormesh(x1d_km, y1d_km, bg, shading="auto", cmap="gray", vmin=bg_vmin, vmax=bg_vmax)
                    ax.pcolormesh(
                        x1d_km, y1d_km, rec_centered, shading="auto", cmap="RdBu_r", vmin=-1.0, vmax=1.0, alpha=0.6
                    )
                else:
                    raise ValueError(kind)

                ax.set_title(
                    rf"$\lambda$={_fmt(lam)} km, $\theta$={theta_dom_deg:.0f}$^\circ$"
                    + "\n"
                    + rf"P={Pk:.2e} ({pk:.1f}%)"
                )
                ax.set_aspect("equal", adjustable="box")

            for ax in axs[n_show:]:
                ax.axis("off")
            for ax in axs[:n_show]:
                ax.set_xlabel("x (km)")
                ax.set_ylabel("y (km)")

            for i, ax in enumerate(axs, start=1):
                if ax.has_data():
                    ax.text(0.12, 0.15, str(i), transform=ax.transAxes, bbox=number_box, va="top", ha="right")

            fig.tight_layout()
            save_fig(fig, outdir / outname, enable=True)

        plot_cluster_grid("standalone", "05_clusters_standalone.png")
        plot_cluster_grid("overlay", "06_clusters_overlay.png")

        for kcl in biggest:
            plot_cluster_individual_save(
                cwt=cwt,
                iwave=iwave,
                kcl=int(kcl),
                x1d_km=x1d_km,
                y1d_km=y1d_km,
                orig_decomp=orig_decomp,
                bg_xy=bg,
                dx_km=dx_km,
                dy_km=dy_km,
                outpath=outdir / f"07_cluster_{int(kcl):03d}_maps.png",
                save_plots=True,
            )

    decomposition[:] = orig_decomp
    decomposition[iwave < 0] = 0
    rec_all = transform.reconstruct2d(cwt)  # (nx, ny)
    rec_all_xy = rec_all.T - np.median(rec_all)
    decomposition[:] = orig_decomp

    remaining_after_preproc_xy = proc

    if save_plots:
        vmin, vmax = diverging_limits(remaining_after_preproc_xy, wavy_xy, rec_all_xy, p_lo=2, p_hi=98)

        figR, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(18, 8), constrained_layout=True, sharex=True, sharey=True)

        im1 = ax1.pcolormesh(
            x1d_km, y1d_km, remaining_after_preproc_xy, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax
        )
        ax1.set_title("Remaining (after preprocessing, input to CWT)")
        ax1.set_xlabel("x (km)")
        ax1.set_ylabel("y (km)")
        ax1.set_aspect("equal", adjustable="box")

        im2 = ax2.pcolormesh(x1d_km, y1d_km, wavy_xy, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
        ax2.set_title("Wavy stuff (wavelet reconstruction after noise filtering)")
        ax2.set_xlabel("x (km)")
        ax2.set_ylabel("y (km)")
        ax2.set_aspect("equal", adjustable="box")

        im3 = ax3.pcolormesh(x1d_km, y1d_km, rec_all_xy, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
        ax3.set_title("Reconstruction after clustering (all clusters combined)")
        ax3.set_xlabel("x (km)")
        ax3.set_ylabel("y (km)")
        ax3.set_aspect("equal", adjustable="box")

        cbar = figR.colorbar(im1, ax=[ax1, ax2, ax3], orientation="vertical", fraction=0.015, pad=0.01)
        cbar.set_label("arb")
        save_fig(figR, outdir / "09_remaining_vs_wavy_vs_allclusters.png", enable=True)

    remaining_core_xy = crop_pad_core(remaining_after_preproc_xy)
    wavy_core_xy = crop_pad_core(wavy_xy)
    rec_all_core_xy = crop_pad_core(rec_all_xy)

    try:
        del raw_win, proc0, plane, detrended, fourier_bg, background, remaining, remaining_tapered, proc, wave_xy
        del cwt_raw, cwt, amps, idxs, iwave, decomposition, orig_decomp, rec_wavy, rec_all
        del bg, cluster_ids, P_abs, P_pct, cluster_ids_sorted, P_abs_sorted, P_pct_sorted
    except Exception:
        pass
    cleanup_memory()

    return {
        "has_clusters": True,
        "x_start_unbinned": int(core_start_unbinned),
        "x_start_unbinned_global": int(core_start_unbinned + x_base_unbinned),
        "dx_km": dx_km,
        "dy_km": dy_km,
        "remaining_xy": remaining_core_xy,
        "wavy_xy": wavy_core_xy,
        "rec_all_xy": rec_all_core_xy,
        "cluster_records": cluster_records,
    }


def stitch_and_save_final_09(slice_results: List[Dict[str, Any]], outroot: Path, tag: str, product_type: str) -> None:
    if not SAVE_STITCHED_PLOT:
        return
    if not slice_results:
        return

    slice_results = sorted(slice_results, key=lambda d: int(d.get("x_start_unbinned_global", 0)))

    dx_km = float(slice_results[0]["dx_km"])
    dy_km = float(slice_results[0]["dy_km"])
    x0_unbinned_global = int(slice_results[0].get("x_start_unbinned_global", 0))
    x0_km = x0_unbinned_global * DX_KM_BASE

    remaining_stitched = np.concatenate([d["remaining_xy"] for d in slice_results], axis=1)
    wavy_stitched = np.concatenate([d["wavy_xy"] for d in slice_results], axis=1)
    rec_all_stitched = np.concatenate([d["rec_all_xy"] for d in slice_results], axis=1)

    ny, nx_tot = remaining_stitched.shape
    x1d_km = x0_km + np.arange(nx_tot, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    vmin, vmax = diverging_limits(remaining_stitched, wavy_stitched, rec_all_stitched, p_lo=2, p_hi=98)

    figR, (axT, axM, axB) = plt.subplots(3, 1, figsize=(18, 6), constrained_layout=True, sharex=True, sharey=True)

    imT = axT.pcolormesh(x1d_km, y1d_km, remaining_stitched, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    axT.set_title(f"STITCHED (core tiles) {tag}: Remaining (after preprocessing, input to CWT)")
    axT.set_xlabel("x (km)")
    axT.set_ylabel("y (km)")
    axT.set_aspect("equal", adjustable="box")

    imM = axM.pcolormesh(x1d_km, y1d_km, wavy_stitched, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    axM.set_title(f"STITCHED (core tiles) {tag}: Wavy stuff (wavelet reconstruction after noise filtering)")
    axM.set_xlabel("x (km)")
    axM.set_ylabel("y (km)")
    axM.set_aspect("equal", adjustable="box")

    imB = axB.pcolormesh(x1d_km, y1d_km, rec_all_stitched, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    axB.set_title(f"STITCHED (core tiles) {tag}: Reconstruction after clustering (all clusters combined)")
    axB.set_xlabel("x (km)")
    axB.set_ylabel("y (km)")
    axB.set_aspect("equal", adjustable="box")

    cbar = figR.colorbar(imT, ax=[axT, axM, axB], orientation="vertical", fraction=0.012, pad=0.01)
    cbar.set_label("K (Relative)" if product_type == "l3a" else "kRayleigh (Relative)")

    save_fig(
        figR,
        outroot / f"FINAL_09_remaining_vs_wavy_vs_allclusters_STITCHED_{tag}.png",
        enable=True,
    )

    try:
        del remaining_stitched, wavy_stitched, rec_all_stitched
    except Exception:
        pass
    cleanup_memory()


# --------------------------
# Per file runner
# --------------------------
def run_one_file(file_path: Path) -> None:
    product_type = detect_product_type(file_path)
    runs = get_runs_for_product(product_type)

    print("\n========================================")
    print("File:", file_path.name)
    print("Detected product:", product_type)
    print("Runs:", runs)

    rad2d = read_frame(file_path, FRAME_IDX, product_type=product_type)

    # NEW: trim edge columns that are initially all NaN before flatten_swath fills them
    rad2d_trim, x0_trim, x1_trim = trim_all_nan_edge_columns(rad2d)
    x_base_unbinned = int(x0_trim)

    if rad2d_trim.shape[1] != rad2d.shape[1]:
        print(f"Trimmed raw frame columns: kept [{x0_trim}:{x1_trim}] (nx {rad2d.shape[1]} -> {rad2d_trim.shape[1]})")
    else:
        print("No raw edge columns trimmed")

    rad2d_flat = flatten_swath(rad2d_trim, target_ny=300)
    nx_total = rad2d_flat.shape[1]

    # X_OFFSET is in original global coordinates
    x_offset_global = int(X_OFFSET)
    x_offset = x_offset_global - x_base_unbinned
    if x_offset < 0:
        x_offset = 0

    if x_offset >= nx_total:
        raise ValueError(
            f"X_OFFSET={x_offset_global} (global) maps to {x_offset} (trimmed local) "
            f"but nx_total after trimming is {nx_total} in file {file_path.name}"
        )

    for cfg in runs:
        run_name = str(cfg.get("name", "run"))

        outroot = OUTPUT_ROOT / file_path.stem / f"{run_name}_offset_{x_offset_global:04d}"
        outroot.mkdir(parents=True, exist_ok=True)

        (outroot / "RUN_CONFIG.txt").write_text(
            "\n".join([f"{k} = {v}" for k, v in cfg.items()])
            + f"\nproduct_type = {product_type}\nfile = {file_path.name}\n"
            + f"x_trim_kept = [{x0_trim}:{x1_trim}]\n"
            + f"x_base_unbinned = {x_base_unbinned}\n"
            + f"X_OFFSET_global = {x_offset_global}\n"
            + f"X_OFFSET_trimmed_local = {x_offset}\n"
        )

        core_starts = list(range(x_offset, nx_total, X_CHUNK))
        stitch_results: List[Dict[str, Any]] = []
        all_cluster_records: List[Dict[str, Any]] = []

        f = int(BIN_FACTOR) if (APPLY_BINNING and int(BIN_FACTOR) > 1) else 1
        y_dim = int(rad2d_flat[Y_SLICE, :].shape[0] // f)
        x_dim = int(X_CHUNK // f)

        for slice_no, core_start in enumerate(core_starts):
            core_end = min(core_start + X_CHUNK, nx_total)

            ext_start = max(0, core_start - SLICE_PAD)
            ext_end = min(nx_total, core_end + SLICE_PAD)
            x_slice = slice(ext_start, ext_end)

            outdir = outroot / f"xcore_{core_start:05d}_{core_end:05d}_xext_{ext_start:05d}_{ext_end:05d}"
            print(f"[{file_path.name} | {run_name}] core={core_start}:{core_end} ext={ext_start}:{ext_end}")

            res = run_pipeline_for_slice(
                rad2d_flat=rad2d_flat,
                x_slice=x_slice,
                outdir=outdir,
                x_start_unbinned=ext_start,
                core_start_unbinned=core_start,
                core_end_unbinned=core_end,
                x_base_unbinned=x_base_unbinned,
                cfg=cfg,
                slice_no=slice_no,
                product_type=product_type,
            )

            stitch_results.append(res)
            all_cluster_records.extend(res.get("cluster_records", []))

            cleanup_memory()

        stitch_and_save_final_09(
            stitch_results,
            outroot,
            tag=f"{run_name}_offset_{x_offset_global:04d}",
            product_type=product_type,
        )
        print("Done per slice processing. Outputs in", outroot.resolve())

        OUTPUT_L6_FOLDER.mkdir(parents=True, exist_ok=True)
        out_path = OUTPUT_L6_FOLDER / make_output_cluster_path(file_path).name

        copy_lxc_to_l6x_with_clusters(
            lxc_path=file_path,
            l6x_path=out_path,
            clusters=all_cluster_records,
            y_dim=y_dim,
            x_dim=x_dim,
            unbin_factor=(BIN_FACTOR if (APPLY_BINNING and int(BIN_FACTOR) > 1) else 1),
            product_type=product_type,
        )

        print("Saved output with clusters:", out_path.resolve())

    try:
        del rad2d, rad2d_trim, rad2d_flat
    except Exception:
        pass
    cleanup_memory()


# --------------------------
# Batch loop over folder
# --------------------------
OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)

files = sorted(INPUT_FOLDER.glob(FILE_GLOB))
if not files:
    raise FileNotFoundError(f"No files matched {FILE_GLOB} in {INPUT_FOLDER.resolve()}")

print("Found", len(files), "files in", INPUT_FOLDER.resolve())
print("Will write L6* cluster netCDF to:", OUTPUT_L6_FOLDER.resolve())
print("Will write plot outputs to:", OUTPUT_ROOT.resolve())
print("SAVE_SLICE_PLOTS =", SAVE_SLICE_PLOTS)
print("SAVE_STITCHED_PLOT =", SAVE_STITCHED_PLOT)
print("SAVE_REDNOISE_DIAGNOSTICS =", SAVE_REDNOISE_DIAGNOSTICS)

for fpath in files:
    try:
        run_one_file(fpath)
    except Exception as e:
        print("FAILED:", fpath.name)
        print("Reason:", repr(e))
        cleanup_memory()
        continue

print("\nAll done.")
```

## .py file run_batch_parallel***

## Calculating MF using both l6a and l6c

```python
# %%
"""
ALL-SLICES matching + per-slice combined L6A recon + FULL-ORBIT stitched recon
+ slice Temperature + momentum flux for PASSED clusters only
+ sum MF per slice + stitch MF across orbit
+ output NetCDF contains original L6A global attrs + variables, but ONLY PASSED clusters
  are kept for cluster-dimension variables.

Momentum flux (per cluster pixel, masked to cluster area):
MF = 1/2 * g^2 / N^2 * (lambda_z / lambda_x) * (Amplitude / Temperature)^2 * (1/c^2)

Constants:
g = 9.52 m/s^2
lambda_z = 10 km
c = 1
N^2 = 0.0004 everywhere (synthetic)
u=v=0 everywhere (synthetic)

Key rules:
- SliceNo already exists in L6A, keep it (but subset to passed clusters)
- Area, Amplitude, DominantWavelength, Angle already exist (subset to passed clusters)
- Also slice Temperature into Temp_cluster (passed clusters only)
- Add geodesic Bearing (0 = East, counter-clockwise) as Bearing + Bearing_cluster
- Only save passed clusters (output file cluster vars are subset)
- Add MF_cluster, MF_slice_sum, MF_toprows
"""

from __future__ import annotations

from pathlib import Path
from typing import Dict, Any, List, Tuple

import numpy as np
import matplotlib.pyplot as plt
from netCDF4 import Dataset

# --------------------------
# User params
# --------------------------
L6C_PATH = Path("awe_l6c_q20_2024165T2120_03175_v01.nc")  # Radiance-based clusters
L6A_PATH = Path("awe_l6a_tmp_2024165T2120_03175_v01.nc")  # Temperature-based clusters

OUTDIR = Path("outputs_ALLSLICES_l6a_to_l6c_matching_ratio_STITCH_MF")
OUTDIR.mkdir(parents=True, exist_ok=True)
SAVE_DPI = 200

DX_KM = 2.0
DY_KM = 2.0

# ---- Synthetic u,v,N^2 settings ----
TEMP_VAR_NAME = "Temperature"  # variable name in L6A
TIME_INDEX = 0                 # which t index to use when slicing
N2_VALUE = 4e-4                # N=0.02 -> N^2=0.0004
U_VALUE = 0.0
V_VALUE = 0.0

# ---- Momentum flux constants ----
G_MS2 = 9.52
LAMBDA_Z_KM = 10.0
C_CANCEL = 1.0
CP_JKGK = 1005.0

# Output NetCDF
OUT_L6A_PASSED = OUTDIR / f"{L6A_PATH.stem}_PASSED_with_u_v_NS2_Bearing_TempCluster_MF.nc"

# --------------------------
# Matching thresholds (tune)
# --------------------------
MIN_IOU = 0.05
MIN_INTERSECTION_PIX = 50
MAX_LAMBDA_MED_RATIO = 1.25
MAX_ANGLE_DIFF_DEG = 15.0

# Save best match plot even if FAIL (does NOT affect passing)
ASSIGN_BEST_EVEN_IF_FAILS = True


# --------------------------
# Helpers
# --------------------------
def save_fig(fig: plt.Figure, outpath: Path) -> None:
    outpath.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(outpath, dpi=SAVE_DPI, bbox_inches="tight")
    plt.close(fig)


def robust_limits(A: np.ndarray, p_lo: float = 2, p_hi: float = 98) -> Tuple[float, float]:
    a = np.asarray(A)
    good = np.isfinite(a)
    if not np.any(good):
        return (-1.0, 1.0)
    lo, hi = np.nanpercentile(a[good], [p_lo, p_hi])
    if (not np.isfinite(lo)) or (not np.isfinite(hi)) or lo == hi:
        lo = float(np.nanmin(a[good]))
        hi = float(np.nanmax(a[good]))
    return (float(lo), float(hi))


def nanpercentile_safe(a: np.ndarray, q: float) -> float:
    good = np.isfinite(a)
    if not np.any(good):
        return np.nan
    return float(np.nanpercentile(a[good], q))


def circ_diff_deg_180(a_deg: float, b_deg: float) -> float:
    if (not np.isfinite(a_deg)) or (not np.isfinite(b_deg)):
        return np.nan
    d = abs(a_deg - b_deg) % 180.0
    return min(d, 180.0 - d)


def circ_median_deg_180(theta_rad: np.ndarray, mask: np.ndarray) -> float:
    th = np.asarray(theta_rad, dtype=float)
    m = mask & np.isfinite(th)
    if not np.any(m):
        return np.nan
    ang = th[m]
    z = np.exp(1j * 2.0 * ang)
    zmean = np.mean(z)
    if (not np.isfinite(zmean.real)) or (not np.isfinite(zmean.imag)) or abs(zmean) == 0:
        return np.nan
    th_mean = 0.5 * np.angle(zmean)
    deg = (np.degrees(th_mean) + 180.0) % 180.0
    return float(deg)


# -------------------------
# Geodesic bearing: 0 = East, counter-clockwise (CCW)
# -------------------------
def initial_bearing_deg_north_cw(lat1_deg, lon1_deg, lat2_deg, lon2_deg):
    """
    Vectorized initial bearing (forward azimuth) from (lat1, lon1) to (lat2, lon2).
    This returns azimuth degrees in [0, 360) measured from North, clockwise.
    """
    lat1 = np.deg2rad(lat1_deg)
    lon1 = np.deg2rad(lon1_deg)
    lat2 = np.deg2rad(lat2_deg)
    lon2 = np.deg2rad(lon2_deg)

    dlon = lon2 - lon1
    x = np.sin(dlon) * np.cos(lat2)
    y = np.cos(lat1) * np.sin(lat2) - np.sin(lat1) * np.cos(lat2) * np.cos(dlon)

    az = np.arctan2(x, y)  # radians, azimuth from North
    az_deg = (np.rad2deg(az) + 360.0) % 360.0
    return az_deg


def bearing_deg_east_ccw_from_points(lat1_deg, lon1_deg, lat2_deg, lon2_deg):
    """
    Convert azimuth (North=0, CW) into your convention:
    East=0 and increasing counter-clockwise (CCW).
    """
    az_north_cw = initial_bearing_deg_north_cw(lat1_deg, lon1_deg, lat2_deg, lon2_deg)
    # North-CW -> East-CCW
    # East-CCW angle = (90 - az_north_cw) mod 360
    return (90.0 - az_north_cw) % 360.0


def compute_bearing_to_x_neighbor_east_ccw(lat2d: np.ndarray, lon2d: np.ndarray) -> np.ndarray:
    """
    Bearing at each pixel to its +x neighbor (last column uses -x neighbor),
    in degrees: 0=East, CCW positive.
    """
    lat = np.asarray(lat2d, dtype=float)
    lon = np.asarray(lon2d, dtype=float)
    ny, nx = lat.shape

    lat2 = np.empty_like(lat)
    lon2 = np.empty_like(lon)

    lat2[:, :-1] = lat[:, 1:]
    lon2[:, :-1] = lon[:, 1:]
    lat2[:, -1] = lat[:, -2]
    lon2[:, -1] = lon[:, -2]

    bearing = bearing_deg_east_ccw_from_points(lat, lon, lat2, lon2)

    good = np.isfinite(lat) & np.isfinite(lon) & np.isfinite(lat2) & np.isfinite(lon2)
    return np.where(good, bearing, np.nan)


def load_l6_for_matching(path: Path) -> Dict[str, Any]:
    with Dataset(path, "r") as nc:
        out: Dict[str, Any] = {
            "SlicesNo": np.asarray(nc.variables["SlicesNo"][:], dtype=int),
            "Amplitude": np.asarray(nc.variables["Amplitude"][:], dtype=float),
            "DominantWavelength": np.asarray(nc.variables["DominantWavelength"][:], dtype=float),
            "Angle": np.asarray(nc.variables["Angle"][:], dtype=float),
        }
        if "ClusterReconstruction" in nc.variables:
            out["ClusterReconstruction"] = np.asarray(nc.variables["ClusterReconstruction"][:], dtype=float)
        else:
            out["ClusterReconstruction"] = None
    return out


def get_cluster_maps(data: Dict[str, Any], idx: int) -> Dict[str, np.ndarray]:
    A = np.asarray(data["Amplitude"][idx], dtype=float)
    L = np.asarray(data["DominantWavelength"][idx], dtype=float)
    T = np.asarray(data["Angle"][idx], dtype=float)
    if data.get("ClusterReconstruction", None) is None:
        R = np.full_like(A, np.nan)
    else:
        R = np.asarray(data["ClusterReconstruction"][idx], dtype=float)
    return dict(R=R, A=A, L=L, T=T)


def compute_match_metrics(
    A_src: np.ndarray, L_src: np.ndarray, T_src: np.ndarray,
    A_cand: np.ndarray, L_cand: np.ndarray, T_cand: np.ndarray,
) -> Dict[str, Any]:
    mask_src = np.isfinite(A_src)
    mask_cand = np.isfinite(A_cand)
    inter = mask_src & mask_cand
    union = mask_src | mask_cand

    n_inter = int(np.count_nonzero(inter))
    n_union = int(np.count_nonzero(union))
    iou = (n_inter / n_union) if n_union > 0 else 0.0

    lam_src = L_src[inter]
    lam_cand = L_cand[inter]
    lam_src_med = nanpercentile_safe(lam_src, 50)
    lam_cand_med = nanpercentile_safe(lam_cand, 50)

    if np.isfinite(lam_src_med) and np.isfinite(lam_cand_med) and lam_src_med > 0 and lam_cand_med > 0:
        lam_ratio = max(lam_src_med, lam_cand_med) / min(lam_src_med, lam_cand_med)
    else:
        lam_ratio = np.nan

    ang_src_deg = circ_median_deg_180(T_src, inter)
    ang_cand_deg = circ_median_deg_180(T_cand, inter)
    ang_diff_deg = circ_diff_deg_180(ang_src_deg, ang_cand_deg)

    return dict(
        n_intersection=n_inter,
        iou=float(iou),
        lam_src_med=float(lam_src_med),
        lam_cand_med=float(lam_cand_med),
        lam_ratio=float(lam_ratio),
        ang_src_deg=float(ang_src_deg),
        ang_cand_deg=float(ang_cand_deg),
        ang_diff_deg=float(ang_diff_deg),
    )


def passes_thresholds(m: Dict[str, Any]) -> bool:
    if m["iou"] < MIN_IOU:
        return False
    if m["n_intersection"] < MIN_INTERSECTION_PIX:
        return False
    if (not np.isfinite(m["lam_ratio"])) or (m["lam_ratio"] > MAX_LAMBDA_MED_RATIO):
        return False
    if (not np.isfinite(m["ang_diff_deg"])) or (m["ang_diff_deg"] > MAX_ANGLE_DIFF_DEG):
        return False
    return True


def score(m: Dict[str, Any]) -> float:
    ang = m["ang_diff_deg"]
    if not np.isfinite(ang):
        ang = 999.0
    r = m["lam_ratio"]
    if (not np.isfinite(r)) or r <= 0:
        return 0.0
    return float(m["iou"]) * (1.0 / float(r)) / (1.0 + float(ang))


def plot_side_by_side_4rows(
    left: Dict[str, np.ndarray],
    right: Dict[str, np.ndarray],
    title_left: str,
    title_right: str,
    outpath: Path,
    dx_km: float,
    dy_km: float,
) -> None:
    ny, nx = left["A"].shape
    x1d_km = np.arange(nx, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    Rl_vmin, Rl_vmax = robust_limits(left["R"])
    Rr_vmin, Rr_vmax = robust_limits(right["R"])
    Al_vmin, Al_vmax = robust_limits(left["A"])
    Ar_vmin, Ar_vmax = robust_limits(right["A"])

    L_vmin, L_vmax = robust_limits(np.concatenate([left["L"].ravel(), right["L"].ravel()]), 5, 95)
    T_vmin, T_vmax = robust_limits(np.concatenate([left["T"].ravel(), right["T"].ravel()]))

    fig, axs = plt.subplots(4, 2, figsize=(16, 14), constrained_layout=True, sharex=True, sharey=True)

    def _panel(ax, Z, vmin, vmax, label, cmap=None):
        im = ax.pcolormesh(x1d_km, y1d_km, Z, shading="auto", vmin=vmin, vmax=vmax, cmap=cmap)
        cb = fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01)
        cb.set_label(label)
        ax.set_aspect("equal", adjustable="box")

    axs[0, 0].set_title(title_left)
    axs[0, 1].set_title(title_right)

    _panel(axs[0, 0], left["R"], Rl_vmin, Rl_vmax, "arb", cmap="RdBu_r")
    _panel(axs[0, 1], right["R"], Rr_vmin, Rr_vmax, "arb", cmap="RdBu_r")
    axs[0, 0].set_ylabel("Recon\ny (km)")

    _panel(axs[1, 0], left["A"], Al_vmin, Al_vmax, "K")
    _panel(axs[1, 1], right["A"], Ar_vmin, Ar_vmax, "kR")
    axs[1, 0].set_ylabel("Amp\ny (km)")

    _panel(axs[2, 0], left["L"], L_vmin, L_vmax, "km")
    _panel(axs[2, 1], right["L"], L_vmin, L_vmax, "km")
    axs[2, 0].set_ylabel("Lambda\ny (km)")

    _panel(axs[3, 0], left["T"], T_vmin, T_vmax, "rad")
    _panel(axs[3, 1], right["T"], T_vmin, T_vmax, "rad")
    axs[3, 0].set_ylabel("Angle\ny (km)")

    for r in range(4):
        for c in range(2):
            axs[r, c].set_xlabel("x (km)")

    save_fig(fig, outpath)


# --------------------------
# Flatten swath BEFORE slicing
# --------------------------
def flatten_swath(Z: np.ndarray, target_ny: int | None = None) -> np.ndarray:
    ny, nx = Z.shape
    out = np.full_like(Z, np.nan)

    bottoms = np.full(nx, np.nan, dtype=float)
    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size:
            bottoms[j] = valid_idx.max()

    valid_bottoms = bottoms[np.isfinite(bottoms)]
    if valid_bottoms.size == 0:
        med = float(np.nanmedian(Z)) if np.isfinite(np.nanmedian(Z)) else 0.0
        out = np.nan_to_num(Z, nan=med, posinf=med, neginf=med)
        if target_ny is not None and ny > target_ny:
            out = out[-target_ny:, :]
        return out

    max_bottom = int(valid_bottoms.max())

    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size == 0:
            continue
        bottom = int(valid_idx.max())
        shift = max_bottom - bottom
        if shift == 0:
            out[:, j] = col
        elif shift > 0:
            out[shift:, j] = col[: ny - shift]
        else:
            shift_up = -shift
            out[: ny - shift_up, j] = col[shift_up:]

    med = float(np.nanmedian(out)) if np.isfinite(np.nanmedian(out)) else 0.0
    out = np.nan_to_num(out, nan=med, posinf=med, neginf=med)

    if target_ny is not None and ny > target_ny:
        start = max_bottom + 1 - target_ny
        start = max(start, 0)
        end = start + target_ny
        out = out[start:end, :]

    return out


# --------------------------
# L6A cloning with PASSED-only cluster variables + new vars (u,v,N^2,Bearing, MF)
# --------------------------
def create_passed_only_l6a_with_new_vars(
    src_l6a: Path,
    dst_l6a: Path,
    passed_cluster_indices: np.ndarray,
    time_index: int,
) -> Tuple[str, str, str]:
    """
    Creates a passed only copy of L6A and adds

    Full field vars
    Bearing  degrees, 0 East, CCW, written only for top y_tile rows at time_index

    Cluster masked vars
    Temp_cluster
    u_cluster
    v_cluster
    NSquared_cluster
    Bearing_cluster
    MF_cluster
    MF_zonal_cluster        MF cos^2(phi)
    MF_meridional_cluster   MF sin^2(phi)

    Per slice sums on tile grid
    MF_slice_sum
    MF_zonal_slice_sum
    MF_meridional_slice_sum

    Full field stitched top rows at time_index
    MF_toprows
    MF_zonal_toprows
    MF_meridional_toprows
    """
    passed_cluster_indices = np.asarray(passed_cluster_indices, dtype=int)
    if passed_cluster_indices.size == 0:
        raise ValueError("No passed clusters. Output would be empty.")

    dst_l6a.parent.mkdir(parents=True, exist_ok=True)

    with Dataset(src_l6a, "r") as src:
        amp_var = src.variables["Amplitude"]
        cluster_dim, ytile_dim, xtile_dim = amp_var.dimensions[:3]

        temp_var = src.variables[TEMP_VAR_NAME]
        temp_dims = temp_var.dimensions
        temp_shape = temp_var.shape
        temp_raw_2d = np.asarray(temp_var[time_index], dtype=float)

        amp_all = np.asarray(src.variables["Amplitude"][:], dtype=float)
        _, y_tile, x_tile = amp_all.shape

        # full field u v N2
        u_raw_2d = (
            np.asarray(src.variables["u"][time_index], dtype=float)
            if "u" in src.variables
            else np.full_like(temp_raw_2d, U_VALUE, dtype=float)
        )
        v_raw_2d = (
            np.asarray(src.variables["v"][time_index], dtype=float)
            if "v" in src.variables
            else np.full_like(temp_raw_2d, V_VALUE, dtype=float)
        )
        n2_raw_2d = (
            np.asarray(src.variables["NSquared"][time_index], dtype=float)
            if "NSquared" in src.variables
            else np.full_like(temp_raw_2d, N2_VALUE, dtype=float)
        )

        # bearing from lat lon if present
        if ("Latitude" in src.variables) and ("Longitude" in src.variables):
            lat_raw_2d = np.asarray(src.variables["Latitude"][time_index], dtype=float)
            lon_raw_2d = np.asarray(src.variables["Longitude"][time_index], dtype=float)
            bearing_raw_2d = compute_bearing_to_x_neighbor_east_ccw(lat_raw_2d, lon_raw_2d)
        else:
            bearing_raw_2d = np.full_like(temp_raw_2d, np.nan, dtype=float)

        # flatten to match tile y dimension
        temp_full_2d = flatten_swath(temp_raw_2d, target_ny=y_tile)
        u_full_2d = flatten_swath(u_raw_2d, target_ny=y_tile)
        v_full_2d = flatten_swath(v_raw_2d, target_ny=y_tile)
        n2_full_2d = flatten_swath(n2_raw_2d, target_ny=y_tile)
        bearing_full_2d = flatten_swath(bearing_raw_2d, target_ny=y_tile)

        n_passed = int(passed_cluster_indices.size)

        with Dataset(dst_l6a, "w") as dst:
            # global attrs
            for aname in src.ncattrs():
                dst.setncattr(aname, src.getncattr(aname))

            # dimensions, subset cluster dim
            for dname, dim in src.dimensions.items():
                if dname == cluster_dim:
                    dst.createDimension(dname, n_passed)
                else:
                    dst.createDimension(dname, (None if dim.isunlimited() else len(dim)))

            # variables copied, subset cluster dim vars
            for vname, vsrc in src.variables.items():
                fill_value = vsrc.getncattr("_FillValue") if "_FillValue" in vsrc.ncattrs() else None
                vdst = dst.createVariable(
                    vname,
                    vsrc.dtype,
                    vsrc.dimensions,
                    zlib=True,
                    complevel=4,
                    fill_value=fill_value,
                )
                for aname in vsrc.ncattrs():
                    if aname != "_FillValue":
                        vdst.setncattr(aname, vsrc.getncattr(aname))

                if cluster_dim in vsrc.dimensions:
                    ax = vsrc.dimensions.index(cluster_dim)
                    data_sub = np.take(vsrc[:], passed_cluster_indices, axis=ax)
                    vdst[:] = data_sub
                else:
                    vdst[:] = vsrc[:]

            # full field Bearing
            bearing_full = np.zeros(temp_shape, dtype=np.float32)
            bearing_full[time_index, 0:y_tile, :] = bearing_full_2d.astype(np.float32)

            if "Bearing" in dst.variables:
                dst.variables["Bearing"][:] = bearing_full
            else:
                vB = dst.createVariable("Bearing", "f4", temp_dims, zlib=True, complevel=4)
                vB.setncattr("long_name", "Geodesic bearing to +x neighbor (0=East, CCW)")
                vB.setncattr("units", "degree")
                vB[:] = bearing_full

            # allocate cluster masked arrays (NaN outside cluster mask)
            temp_cluster = np.full((n_passed, y_tile, x_tile), np.nan, dtype=np.float32)
            u_cluster = np.full_like(temp_cluster, np.nan)
            v_cluster = np.full_like(temp_cluster, np.nan)
            n2_cluster = np.full_like(temp_cluster, np.nan)
            bearing_cluster = np.full_like(temp_cluster, np.nan)
            
            mf_cluster = np.full_like(temp_cluster, np.nan)
            mf_zonal_cluster = np.full_like(temp_cluster, np.nan)
            mf_meridional_cluster = np.full_like(temp_cluster, np.nan)

            amp_passed = np.asarray(dst.variables["Amplitude"][:], dtype=float)
            lam_passed = np.asarray(dst.variables["DominantWavelength"][:], dtype=float)
            slice_passed = np.asarray(dst.variables["SlicesNo"][:], dtype=int)
            ang_rad_passed = np.asarray(dst.variables["Angle"][:], dtype=float)  # radians, propagation direction in tile

            for j in range(n_passed):
                s = int(slice_passed[j])
                x0 = s * x_tile
                x1 = x0 + x_tile

                mask = np.isfinite(amp_passed[j])

                Ttile = temp_full_2d[:, x0:x1]
                Utile = u_full_2d[:, x0:x1]
                Vtile = v_full_2d[:, x0:x1]
                N2tile = n2_full_2d[:, x0:x1]
                Btile = bearing_full_2d[:, x0:x1]

                temp_cluster[j][mask] = Ttile[mask]
                u_cluster[j][mask] = Utile[mask]
                v_cluster[j][mask] = Vtile[mask]
                n2_cluster[j][mask] = N2tile[mask]
                bearing_cluster[j][mask] = Btile[mask]

                A = amp_passed[j]
                Lx = lam_passed[j]

                ang_deg_tile = (np.rad2deg(ang_rad_passed[j]) % 360.0)

                good = mask
                good = good & np.isfinite(Ttile) & np.isfinite(N2tile) & np.isfinite(Lx) & np.isfinite(Btile)
                good = good & (Ttile != 0.0) & (N2tile > 0.0) & (Lx > 0.0)

                if np.any(good):
                    mf_val = (
                        0.5
                        * (G_MS2**2)
                        / N2tile[good]
                        * (LAMBDA_Z_KM / Lx[good])
                        * (A[good] / Ttile[good]) ** 2
                        * (1.0 / (C_CANCEL**2))
                    )
                    mf_cluster[j][good] = mf_val

                    # phi is geographic propagation angle, degrees, 0 East, CCW
                    phi_deg = (Btile[good] + ang_deg_tile[good]) % 360.0
                    phi_rad = np.deg2rad(phi_deg)

                    mf_zonal_cluster[j][good] = mf_val * (np.cos(phi_rad) ** 2)
                    mf_meridional_cluster[j][good] = mf_val * (np.sin(phi_rad) ** 2)

            # write cluster vars
            if "Temp_cluster" in dst.variables:
                dst.variables["Temp_cluster"][:] = temp_cluster
            else:
                dst.createVariable("Temp_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim))[:] = temp_cluster

            if "u_cluster" in dst.variables:
                dst.variables["u_cluster"][:] = u_cluster
            else:
                dst.createVariable("u_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim))[:] = u_cluster

            if "v_cluster" in dst.variables:
                dst.variables["v_cluster"][:] = v_cluster
            else:
                dst.createVariable("v_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim))[:] = v_cluster

            if "NSquared_cluster" in dst.variables:
                dst.variables["NSquared_cluster"][:] = n2_cluster
            else:
                dst.createVariable("NSquared_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim))[:] = n2_cluster

            if "Bearing_cluster" in dst.variables:
                dst.variables["Bearing_cluster"][:] = bearing_cluster
            else:
                vBC = dst.createVariable("Bearing_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim))
                vBC.setncattr("long_name", "Geodesic bearing to +x neighbor (0=East, CCW), cluster masked")
                vBC.setncattr("units", "degree")
                vBC[:] = bearing_cluster

            if "MF_cluster" in dst.variables:
                dst.variables["MF_cluster"][:] = mf_cluster
            else:
                vMF = dst.createVariable("MF_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim))
                vMF.setncattr("long_name", "Momentum flux proxy per pixel, cluster masked")
                vMF.setncattr("units", "m^2 s^-2")
                vMF[:] = mf_cluster

            if "MF_zonal_cluster" in dst.variables:
                dst.variables["MF_zonal_cluster"][:] = mf_zonal_cluster
            else:
                vMFZ = dst.createVariable("MF_zonal_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim))
                vMFZ.setncattr("long_name", "Zonal aligned MF magnitude share, MF cos^2(phi)")
                vMFZ.setncattr("units", "m^2 s^-2")
                vMFZ[:] = mf_zonal_cluster

            if "MF_meridional_cluster" in dst.variables:
                dst.variables["MF_meridional_cluster"][:] = mf_meridional_cluster
            else:
                vMFM = dst.createVariable("MF_meridional_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim))
                vMFM.setncattr("long_name", "Meridional aligned MF magnitude share, MF sin^2(phi)")
                vMFM.setncattr("units", "m^2 s^-2")
                vMFM[:] = mf_meridional_cluster

            # slice sums and stitched top rows
            smin, smax = int(np.min(slice_passed)), int(np.max(slice_passed))
            full_slices = np.arange(smin, smax + 1)

            if "slice" not in dst.dimensions:
                dst.createDimension("slice", len(full_slices))

            if "SliceNo_passed_slices" in dst.variables:
                dst.variables["SliceNo_passed_slices"][:] = full_slices
            else:
                dst.createVariable("SliceNo_passed_slices", "i4", ("slice",))[:] = full_slices

            mf_slice_sum = np.zeros((len(full_slices), y_tile, x_tile), dtype=np.float32)
            mfz_slice_sum = np.zeros_like(mf_slice_sum)
            mfm_slice_sum = np.zeros_like(mf_slice_sum)

            for k, s in enumerate(full_slices):
                idx = np.where(slice_passed == s)[0]
                if idx.size > 0:
                    mf_slice_sum[k]  = np.nansum(mf_cluster[idx], axis=0)
                    mfz_slice_sum[k] = np.nansum(mf_zonal_cluster[idx], axis=0)
                    mfm_slice_sum[k] = np.nansum(mf_meridional_cluster[idx], axis=0)

                # else keep zeros

            if "MF_slice_sum" in dst.variables:
                dst.variables["MF_slice_sum"][:] = mf_slice_sum
            else:
                dst.createVariable("MF_slice_sum", "f4", ("slice", ytile_dim, xtile_dim))[:] = mf_slice_sum

            if "MF_zonal_slice_sum" in dst.variables:
                dst.variables["MF_zonal_slice_sum"][:] = mfz_slice_sum
            else:
                v = dst.createVariable("MF_zonal_slice_sum", "f4", ("slice", ytile_dim, xtile_dim))
                v.setncattr("long_name", "Per slice zonal MF magnitude share sum")
                v.setncattr("units", "m^2 s^-2")
                v[:] = mfz_slice_sum

            if "MF_meridional_slice_sum" in dst.variables:
                dst.variables["MF_meridional_slice_sum"][:] = mfm_slice_sum
            else:
                v = dst.createVariable("MF_meridional_slice_sum", "f4", ("slice", ytile_dim, xtile_dim))
                v.setncattr("long_name", "Per slice meridional MF magnitude share sum")
                v.setncattr("units", "m^2 s^-2")
                v[:] = mfm_slice_sum

            MF_toprows = np.zeros(temp_shape, dtype=np.float32)
            MFZ_toprows = np.zeros(temp_shape, dtype=np.float32)
            MFM_toprows = np.zeros(temp_shape, dtype=np.float32)

            for k, s in enumerate(full_slices):
                x0 = s * x_tile
                x1 = x0 + x_tile
                MF_toprows[time_index, 0:y_tile, x0:x1] = mf_slice_sum[k]
                MFZ_toprows[time_index, 0:y_tile, x0:x1] = mfz_slice_sum[k]
                MFM_toprows[time_index, 0:y_tile, x0:x1] = mfm_slice_sum[k]

            if "MF_toprows" in dst.variables:
                dst.variables["MF_toprows"][:] = MF_toprows
            else:
                dst.createVariable("MF_toprows", "f4", temp_dims)[:] = MF_toprows

            if "MF_zonal_toprows" in dst.variables:
                dst.variables["MF_zonal_toprows"][:] = MFZ_toprows
            else:
                v = dst.createVariable("MF_zonal_toprows", "f4", temp_dims)
                v.setncattr("long_name", "Stitched zonal MF magnitude share, top tile rows only")
                v.setncattr("units", "m^2 s^-2")
                v[:] = MFZ_toprows

            if "MF_meridional_toprows" in dst.variables:
                dst.variables["MF_meridional_toprows"][:] = MFM_toprows
            else:
                v = dst.createVariable("MF_meridional_toprows", "f4", temp_dims)
                v.setncattr("long_name", "Stitched meridional MF magnitude share, top tile rows only")
                v.setncattr("units", "m^2 s^-2")
                v[:] = MFM_toprows

    return cluster_dim, ytile_dim, xtile_dim

# --------------------------
# Main: matching -> passed clusters -> output passed-only netCDF -> stitched MF plot
# --------------------------
l6c = load_l6_for_matching(L6C_PATH)
l6a_match = load_l6_for_matching(L6A_PATH)

all_slices_a = np.unique(l6a_match["SlicesNo"]).astype(int)
all_slices_a = all_slices_a[all_slices_a >= 0]
all_slices_a = np.sort(all_slices_a)

print("Slices in L6A:", all_slices_a.tolist())

summary_lines: List[str] = []
summary_lines.append(f"L6A: {L6A_PATH}\n")
summary_lines.append(f"L6C: {L6C_PATH}\n\n")
summary_lines.append("slice  nA  nC  nBestSaved  nPassed  note\n")

passed_global_indices: List[int] = []

for slice_no in all_slices_a:
    idx_a = np.where(l6a_match["SlicesNo"] == int(slice_no))[0]
    idx_c = np.where(l6c["SlicesNo"] == int(slice_no))[0]

    slice_dir = OUTDIR / f"slice_{slice_no:03d}"
    pairs_dir = slice_dir / "pairs"
    pairs_dir.mkdir(parents=True, exist_ok=True)

    if idx_a.size == 0:
        summary_lines.append(f"{slice_no:5d}  {0:2d}  {idx_c.size:2d}  {0:10d}  {0:7d}  no L6A clusters\n")
        continue
    if idx_c.size == 0:
        (slice_dir / "NO_L6C_CLUSTERS.txt").write_text(f"No L6C clusters for slice {slice_no}\n")
        summary_lines.append(f"{slice_no:5d}  {idx_a.size:2d}  {0:2d}  {0:10d}  {0:7d}  no L6C clusters\n")
        continue

    all_candidate_rows: List[Dict[str, Any]] = []
    best_rows: List[Dict[str, Any]] = []

    for a_idx in idx_a:
        a_idx = int(a_idx)
        A_a = l6a_match["Amplitude"][a_idx]
        L_a = l6a_match["DominantWavelength"][a_idx]
        T_a = l6a_match["Angle"][a_idx]

        candidates: List[Dict[str, Any]] = []
        for c_idx in idx_c:
            c_idx = int(c_idx)
            A_c = l6c["Amplitude"][c_idx]
            L_c = l6c["DominantWavelength"][c_idx]
            T_c = l6c["Angle"][c_idx]

            m = compute_match_metrics(A_a, L_a, T_a, A_c, L_c, T_c)
            m["cluster_a"] = a_idx
            m["cluster_c"] = c_idx
            m["passes"] = bool(passes_thresholds(m))
            m["score"] = float(score(m))
            candidates.append(m)
            all_candidate_rows.append(m)

        passed = [c for c in candidates if c["passes"]]
        if passed:
            best = max(passed, key=lambda d: d["score"])
        else:
            best = max(candidates, key=lambda d: d["score"])

        if best["passes"] or ASSIGN_BEST_EVEN_IF_FAILS:
            best_rows.append(best)

            left_maps = get_cluster_maps(l6a_match, best["cluster_a"])
            right_maps = get_cluster_maps(l6c, best["cluster_c"])

            pass_tag = "PASS" if best["passes"] else "FAIL"
            outpng = pairs_dir / f"slice_{slice_no:03d}_A_{best['cluster_a']:04d}_to_C_{best['cluster_c']:04d}_{pass_tag}.png"

            title_left = (
                f"L6A Temp | slice {slice_no} | idx {best['cluster_a']} | "
                f"n={np.count_nonzero(np.isfinite(left_maps['A']))}"
            )
            title_right = (
                f"L6C Rad | slice {slice_no} | idx {best['cluster_c']} | "
                f"IoU={best['iou']:.3f} inter={best['n_intersection']} "
                f"ratio={best['lam_ratio']:.3f} Δθ={best['ang_diff_deg']:.1f}°"
            )

            plot_side_by_side_4rows(
                left=left_maps,
                right=right_maps,
                title_left=title_left,
                title_right=title_right,
                outpath=outpng,
                dx_km=DX_KM,
                dy_km=DY_KM,
            )

    cand_path = slice_dir / f"slice_{slice_no:03d}_ALL_candidates_ratio.txt"
    lines = []
    lines.append(f"L6A file: {L6A_PATH}\n")
    lines.append(f"L6C file: {L6C_PATH}\n")
    lines.append(f"Slice: {slice_no}\n\n")
    lines.append(
        "cluster_a  cluster_c  passes  score    iou    inter  "
        "lam_ratio  ang_diff_deg  thetaA_deg  thetaC_deg  lamA_med  lamC_med\n"
    )
    for m in sorted(all_candidate_rows, key=lambda d: (d["cluster_a"], -d["score"])):
        lines.append(
            f"{m['cluster_a']:9d}  {m['cluster_c']:9d}  {int(m['passes']):6d}  {m['score']:6.3f}  "
            f"{m['iou']:.3f}  {m['n_intersection']:5d}  "
            f"{m['lam_ratio']:8.3f}  {m['ang_diff_deg']:12.1f}  "
            f"{m['ang_src_deg']:10.1f}  {m['ang_cand_deg']:10.1f}  "
            f"{m['lam_src_med']:8.1f}  {m['lam_cand_med']:8.1f}\n"
        )
    cand_path.write_text("".join(lines))

    best_path = slice_dir / f"slice_{slice_no:03d}_BEST_matches_A_to_C_ratio.txt"
    lines = []
    lines.append(f"L6A file: {L6A_PATH}\n")
    lines.append(f"L6C file: {L6C_PATH}\n")
    lines.append(f"Slice: {slice_no}\n\n")
    lines.append(
        "cluster_a  best_cluster_c  passes  score    iou    inter  "
        "lam_ratio  ang_diff_deg  thetaA_deg  thetaC_deg  lamA_med  lamC_med\n"
    )
    for m in sorted(best_rows, key=lambda d: d["cluster_a"]):
        lines.append(
            f"{m['cluster_a']:9d}  {m['cluster_c']:14d}  {int(m['passes']):6d}  {m['score']:6.3f}  "
            f"{m['iou']:.3f}  {m['n_intersection']:5d}  "
            f"{m['lam_ratio']:8.3f}  {m['ang_diff_deg']:12.1f}  "
            f"{m['ang_src_deg']:10.1f}  {m['ang_cand_deg']:10.1f}  "
            f"{m['lam_src_med']:8.1f}  {m['lam_cand_med']:8.1f}\n"
        )
    best_path.write_text("".join(lines))

    passed_here = [int(m["cluster_a"]) for m in best_rows if bool(m["passes"])]
    passed_global_indices.extend(passed_here)

    summary_lines.append(
        f"{slice_no:5d}  {idx_a.size:2d}  {idx_c.size:2d}  {len(best_rows):10d}  {len(passed_here):7d}  ok\n"
    )

summary_path = OUTDIR / "SUMMARY_all_slices.txt"
summary_path.write_text("".join(summary_lines))

passed_global_indices = sorted(set(passed_global_indices))
passed_global_indices_np = np.asarray(passed_global_indices, dtype=int)
print("Total passed clusters:", passed_global_indices_np.size)

if passed_global_indices_np.size == 0:
    raise RuntimeError("No passed clusters found. Nothing to save/stitch.")

cluster_dim_name, ytile_dim_name, xtile_dim_name = create_passed_only_l6a_with_new_vars(
    src_l6a=L6A_PATH,
    dst_l6a=OUT_L6A_PASSED,
    passed_cluster_indices=passed_global_indices_np,
    time_index=TIME_INDEX,
)

# ---- Plot MF_toprows stitched (2D at TIME_INDEX) ----
with Dataset(OUT_L6A_PASSED, "r") as nc:
    MF2 = np.asarray(nc.variables["MF_toprows"][TIME_INDEX], dtype=float)  # (y,x)
    ny, nx = MF2.shape
    x1d_km = np.arange(nx, dtype=float) * DX_KM
    y1d_km = np.arange(ny, dtype=float) * DY_KM

    vmin, vmax = robust_limits(MF2, 2, 99.9)

    fig, ax = plt.subplots(figsize=(22, 3), constrained_layout=True)
    im = ax.pcolormesh(x1d_km, y1d_km, MF2, shading="auto", vmin=vmin, vmax=vmax)
    ax.set_title("Momentum Flux (passed clusters summed)")
    ax.set_xlabel("x (km)")
    ax.set_ylabel("y (km)")
    ax.set_ylim(0, 600)
    ax.set_aspect("equal", adjustable="box")
    fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01).set_label("MF (m²/s²)")
    save_fig(fig, OUTDIR / "MF_stitched.png")

print("\nDone.")
print("Outputs in:", OUTDIR.resolve())
print("Summary:", summary_path.resolve())
print("PASSED-only augmented L6A:", OUT_L6A_PASSED.resolve())
print("New vars in output: Bearing, Bearing_cluster, Temp_cluster, u_cluster, v_cluster, NSquared_cluster, MF_cluster, MF_slice_sum, MF_toprows")


# --------------------------
# Plot per-passed-cluster panels:
# Recon, Amp, Lambda, Angle, Bearing, N^2, Temperature, MF
# --------------------------
def plot_passed_clusters_panels(nc_path: Path, outdir: Path) -> None:
    """
    Per passed cluster panels, 5 rows x 2 columns

    Row 1  Recon | Amplitude
    Row 2  Lambda | Temperature
    Row 3  Angle deg | Bearing deg
    Row 4  N2 | MF
    Row 5  MF zonal share | MF meridional share
    """
    outdir.mkdir(parents=True, exist_ok=True)

    with Dataset(nc_path, "r") as nc:
        amp = np.asarray(nc.variables["Amplitude"][:], dtype=float)
        lam = np.asarray(nc.variables["DominantWavelength"][:], dtype=float)

        ang_rad = np.asarray(nc.variables["Angle"][:], dtype=float)
        ang_deg = (np.rad2deg(ang_rad) % 360.0)

        brg = (
            np.asarray(nc.variables["Bearing_cluster"][:], dtype=float)
            if "Bearing_cluster" in nc.variables
            else np.full_like(amp, np.nan)
        )

        n2 = np.asarray(nc.variables["NSquared_cluster"][:], dtype=float)
        tmp = np.asarray(nc.variables["Temp_cluster"][:], dtype=float)
        mf = np.asarray(nc.variables["MF_cluster"][:], dtype=float)

        mfz = (
            np.asarray(nc.variables["MF_zonal_cluster"][:], dtype=float)
            if "MF_zonal_cluster" in nc.variables
            else np.full_like(mf, np.nan)
        )
        mfm = (
            np.asarray(nc.variables["MF_meridional_cluster"][:], dtype=float)
            if "MF_meridional_cluster" in nc.variables
            else np.full_like(mf, np.nan)
        )

        slc = np.asarray(nc.variables["SlicesNo"][:], dtype=int)

        if "ClusterReconstruction" in nc.variables:
            recon = np.asarray(nc.variables["ClusterReconstruction"][:], dtype=float)
        else:
            recon = np.full_like(amp, np.nan)

        n_cluster, ny, nx = amp.shape
        x1d_km = np.arange(nx, dtype=float) * DX_KM
        y1d_km = np.arange(ny, dtype=float) * DY_KM

        ang_vmin, ang_vmax = (0.0, 360.0)
        brg_vmin, brg_vmax = (0.0, 360.0)

        n2_center = float(N2_VALUE)
        n2_vmin, n2_vmax = (0.5 * n2_center, 1.5 * n2_center)

        for i in range(n_cluster):
            s = int(slc[i])

            r_vmin, r_vmax = robust_limits(recon[i], 2, 98)
            a_vmin, a_vmax = robust_limits(amp[i], 2, 98)
            l_vmin, l_vmax = robust_limits(lam[i], 5, 95)
            T_vmin, T_vmax = robust_limits(tmp[i], 2, 98)

            mf_vmin, mf_vmax = robust_limits(mf[i], 2, 98)
            mfz_vmin, mfz_vmax = robust_limits(mfz[i], 2, 98)
            mfm_vmin, mfm_vmax = robust_limits(mfm[i], 2, 98)

            fig, axs = plt.subplots(
                5, 2,
                figsize=(16, 22),
                constrained_layout=True,
                sharex=True,
                sharey=True,
            )

            def _panel(ax, Z, vmin, vmax, label, cmap=None):
                im = ax.pcolormesh(
                    x1d_km, y1d_km, Z,
                    shading="auto",
                    vmin=vmin,
                    vmax=vmax,
                    cmap=cmap,
                )
                cb = fig.colorbar(im, ax=ax, fraction=0.04, pad=0.02)
                cb.set_label(label)
                ax.set_aspect("equal", adjustable="box")

            axs[0, 0].set_title(f"Passed cluster {i:04d}  SliceNo {s}")

            _panel(axs[0, 0], recon[i], r_vmin, r_vmax, "Recon", cmap="RdBu_r")
            _panel(axs[0, 1], amp[i], a_vmin, a_vmax, "Amplitude K")

            _panel(axs[1, 0], lam[i], l_vmin, l_vmax, "DominantWavelength km")
            _panel(axs[1, 1], tmp[i], T_vmin, T_vmax, "Temperature K")

            _panel(axs[2, 0], ang_deg[i], ang_vmin, ang_vmax, "Angle deg propagation")
            _panel(axs[2, 1], brg[i], brg_vmin, brg_vmax, "Bearing deg 0 East CCW")

            _panel(axs[3, 0], n2[i], n2_vmin, n2_vmax, "N2 1 per s2")
            _panel(axs[3, 1], mf[i], mf_vmin, mf_vmax, "MF m2 per s2")

            _panel(axs[4, 0], mfz[i], mfz_vmin, mfz_vmax, "MF zonal share cos2")
            _panel(axs[4, 1], mfm[i], mfm_vmin, mfm_vmax, "MF meridional share sin2")

            for r in range(5):
                axs[r, 0].set_ylabel("y km")
                axs[r, 1].set_ylabel("y km")

            axs[4, 0].set_xlabel("x km")
            axs[4, 1].set_xlabel("x km")

            outpng = outdir / (
                f"passed_cluster_{i:04d}_slice_{s:03d}"
                "_Recon_Amp_Lam_Temp_Ang_Bearing_N2_MF_MFz_MFm.png"
            )
            save_fig(fig, outpng)

    print("Per cluster plots saved to:", outdir.resolve())

def plot_stitched_mf_components(nc_path: Path, outpng: Path, time_index: int) -> None:
    """
    Makes a stitched 3 panel figure

    Panel 1 MF_toprows
    Panel 2 MF_zonal_toprows
    Panel 3 MF_meridional_toprows
    """
    outpng.parent.mkdir(parents=True, exist_ok=True)

    with Dataset(nc_path, "r") as nc:
        MF = np.asarray(nc.variables["MF_toprows"][time_index], dtype=float)
        MFZ = np.asarray(nc.variables["MF_zonal_toprows"][time_index], dtype=float)
        MFM = np.asarray(nc.variables["MF_meridional_toprows"][time_index], dtype=float)

    ny, nx = MF.shape
    x1d_km = np.arange(nx, dtype=float) * DX_KM
    y1d_km = np.arange(ny, dtype=float) * DY_KM

    stacked = np.concatenate([MF.ravel(), MFZ.ravel(), MFM.ravel()])
    vmin, vmax = robust_limits(stacked, 2, 99.9)

    fig, axs = plt.subplots(3, 1, figsize=(22, 8), constrained_layout=True, sharex=True, sharey=True)

    im0 = axs[0].pcolormesh(x1d_km, y1d_km, MF, shading="auto", vmin=vmin, vmax=vmax)
    axs[0].set_title("Stitched MF passed clusters summed")
    axs[0].set_ylabel("y km")
    axs[0].set_aspect("equal", adjustable="box")

    im1 = axs[1].pcolormesh(x1d_km, y1d_km, MFZ, shading="auto", vmin=vmin, vmax=vmax)
    axs[1].set_title("Stitched MF zonal share cos2")
    axs[1].set_ylabel("y km")
    axs[1].set_aspect("equal", adjustable="box")

    im2 = axs[2].pcolormesh(x1d_km, y1d_km, MFM, shading="auto", vmin=vmin, vmax=vmax)
    axs[2].set_title("Stitched MF meridional share sin2")
    axs[2].set_xlabel("x km")
    axs[2].set_ylabel("y km")
    axs[2].set_aspect("equal", adjustable="box")

    for ax in axs:
        ax.set_ylim(0, 600)

    cbar = fig.colorbar(im2, ax=axs, fraction=0.02, pad=0.01)
    cbar.set_label("MF m2 per s2")

    save_fig(fig, outpng)


plot_passed_clusters_panels(
    nc_path=OUT_L6A_PASSED,
    outdir=OUTDIR / "passed_cluster_plots_panels",
)

plot_stitched_mf_components(
    nc_path=OUT_L6A_PASSED,
    outpng=OUTDIR / "MF_stitched_components.png",
    time_index=TIME_INDEX,
)
```

## Clean up memory

Brief list of changes

Matching load no longer reads ClusterReconstruction[:] into memory. It is lazy loaded only for the specific best pair plot.

Candidate logging is streamed as strings (cand_lines) instead of storing a huge list of dicts for all A by C pairs.

In create_passed_only_l6a_with_new_vars, the tile shape comes from amp_var.shape (metadata) instead of Amplitude[:].

Per cluster plotting reads one cluster at a time (var[i]) instead of var[:] for all clusters.

Added aggressive cleanup (del, gc.collect(), and plt.close) after heavy steps so memory can actually drop instead of only going up.

```python
# %%
"""
ALL-SLICES matching + per-slice combined L6A recon + FULL-ORBIT stitched recon
+ slice Temperature + momentum flux for PASSED clusters only
+ sum MF per slice + stitch MF across orbit
+ output NetCDF contains original L6A global attrs + variables, but ONLY PASSED clusters
  are kept for cluster-dimension variables.

Momentum flux (per cluster pixel, masked to cluster area):
MF = 1/2 * g^2 / N^2 * (lambda_z / lambda_x) * (Amplitude / Temperature)^2 * (1/c^2)

Constants:
g = 9.52 m/s^2
lambda_z = 10 km
c = 1
N^2 = 0.0004 everywhere (synthetic)
u=v=0 everywhere (synthetic)

Key rules:
- SliceNo already exists in L6A, keep it (but subset to passed clusters)
- Area, Amplitude, DominantWavelength, Angle already exist (subset to passed clusters)
- Also slice Temperature into Temp_cluster (passed clusters only)
- Add geodesic Bearing (0 = East, counter-clockwise) as Bearing + Bearing_cluster
- Only save passed clusters (output file cluster vars are subset)
- Add MF_cluster, MF_slice_sum, MF_toprows
"""

from __future__ import annotations

from pathlib import Path
from typing import Dict, Any, List, Tuple

import gc
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from netCDF4 import Dataset

# --------------------------
# User params
# --------------------------
L6C_PATH = Path("awe_l6c_q20_2024165T2120_03175_v01.nc")  # Radiance-based clusters
L6A_PATH = Path("awe_l6a_tmp_2024165T2120_03175_v01.nc")  # Temperature-based clusters

OUTDIR = Path("outputs_ALLSLICES_l6a_to_l6c_matching_ratio_STITCH_MF")
OUTDIR.mkdir(parents=True, exist_ok=True)
SAVE_DPI = 200

DX_KM = 2.0
DY_KM = 2.0

# ---- Synthetic u,v,N^2 settings ----
TEMP_VAR_NAME = "Temperature"  # variable name in L6A
TIME_INDEX = 0                 # which t index to use when slicing
N2_VALUE = 4e-4                # N=0.02 -> N^2=0.0004
U_VALUE = 0.0
V_VALUE = 0.0

# ---- Momentum flux constants ----
G_MS2 = 9.52
LAMBDA_Z_KM = 10.0
C_CANCEL = 1.0
CP_JKGK = 1005.0

# Output NetCDF
OUT_L6A_PASSED = OUTDIR / f"{L6A_PATH.stem}_PASSED_with_u_v_NS2_Bearing_TempCluster_MF.nc"

# --------------------------
# Matching thresholds (tune)
# --------------------------
MIN_IOU = 0.05
MIN_INTERSECTION_PIX = 50
MAX_LAMBDA_MED_RATIO = 1.25
MAX_ANGLE_DIFF_DEG = 15.0

# Save best match plot even if FAIL (does NOT affect passing)
ASSIGN_BEST_EVEN_IF_FAILS = True

# --------------------------
# Helpers
# --------------------------
def save_fig(fig: plt.Figure, outpath: Path) -> None:
    outpath.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(outpath, dpi=SAVE_DPI, bbox_inches="tight")
    plt.close(fig)


def robust_limits(A: np.ndarray, p_lo: float = 2, p_hi: float = 98) -> Tuple[float, float]:
    a = np.asarray(A)
    good = np.isfinite(a)
    if not np.any(good):
        return (-1.0, 1.0)
    lo, hi = np.nanpercentile(a[good], [p_lo, p_hi])
    if (not np.isfinite(lo)) or (not np.isfinite(hi)) or lo == hi:
        lo = float(np.nanmin(a[good]))
        hi = float(np.nanmax(a[good]))
    return (float(lo), float(hi))


def nanpercentile_safe(a: np.ndarray, q: float) -> float:
    good = np.isfinite(a)
    if not np.any(good):
        return np.nan
    return float(np.nanpercentile(a[good], q))


def circ_diff_deg_180(a_deg: float, b_deg: float) -> float:
    if (not np.isfinite(a_deg)) or (not np.isfinite(b_deg)):
        return np.nan
    d = abs(a_deg - b_deg) % 180.0
    return min(d, 180.0 - d)


def circ_median_deg_180(theta_rad: np.ndarray, mask: np.ndarray) -> float:
    th = np.asarray(theta_rad, dtype=float)
    m = mask & np.isfinite(th)
    if not np.any(m):
        return np.nan
    ang = th[m]
    z = np.exp(1j * 2.0 * ang)
    zmean = np.mean(z)
    if (not np.isfinite(zmean.real)) or (not np.isfinite(zmean.imag)) or abs(zmean) == 0:
        return np.nan
    th_mean = 0.5 * np.angle(zmean)
    deg = (np.degrees(th_mean) + 180.0) % 180.0
    return float(deg)


# -------------------------
# Geodesic bearing: 0 = East, counter-clockwise (CCW)
# -------------------------
def initial_bearing_deg_north_cw(lat1_deg, lon1_deg, lat2_deg, lon2_deg):
    """
    Vectorized initial bearing (forward azimuth) from (lat1, lon1) to (lat2, lon2).
    Returns azimuth degrees in [0, 360) measured from North, clockwise.
    """
    lat1 = np.deg2rad(lat1_deg)
    lon1 = np.deg2rad(lon1_deg)
    lat2 = np.deg2rad(lat2_deg)
    lon2 = np.deg2rad(lon2_deg)

    dlon = lon2 - lon1
    x = np.sin(dlon) * np.cos(lat2)
    y = np.cos(lat1) * np.sin(lat2) - np.sin(lat1) * np.cos(lat2) * np.cos(dlon)

    az = np.arctan2(x, y)
    az_deg = (np.rad2deg(az) + 360.0) % 360.0
    return az_deg


def bearing_deg_east_ccw_from_points(lat1_deg, lon1_deg, lat2_deg, lon2_deg):
    """
    Convert azimuth (North=0, CW) into your convention:
    East=0 and increasing counter-clockwise (CCW).
    East-CCW angle = (90 - az_north_cw) mod 360
    """
    az_north_cw = initial_bearing_deg_north_cw(lat1_deg, lon1_deg, lat2_deg, lon2_deg)
    return (90.0 - az_north_cw) % 360.0


def compute_bearing_to_x_neighbor_east_ccw(lat2d: np.ndarray, lon2d: np.ndarray) -> np.ndarray:
    """
    Bearing at each pixel to its +x neighbor (last column uses -x neighbor),
    in degrees: 0=East, CCW positive.
    """
    lat = np.asarray(lat2d, dtype=float)
    lon = np.asarray(lon2d, dtype=float)
    ny, nx = lat.shape

    lat2 = np.empty_like(lat)
    lon2 = np.empty_like(lon)

    lat2[:, :-1] = lat[:, 1:]
    lon2[:, :-1] = lon[:, 1:]
    lat2[:, -1] = lat[:, -2]
    lon2[:, -1] = lon[:, -2]

    bearing = bearing_deg_east_ccw_from_points(lat, lon, lat2, lon2)

    good = np.isfinite(lat) & np.isfinite(lon) & np.isfinite(lat2) & np.isfinite(lon2)
    return np.where(good, bearing, np.nan)


# --------------------------
# IO: load only the variables needed for matching (no ClusterReconstruction here)
# --------------------------
def load_l6_for_matching(path: Path) -> Dict[str, Any]:
    with Dataset(path, "r") as nc:
        return {
            "path": str(path),
            "SlicesNo": np.asarray(nc.variables["SlicesNo"][:], dtype=int),
            "Amplitude": np.asarray(nc.variables["Amplitude"][:], dtype=float),
            "DominantWavelength": np.asarray(nc.variables["DominantWavelength"][:], dtype=float),
            "Angle": np.asarray(nc.variables["Angle"][:], dtype=float),
        }


def get_cluster_maps(data: Dict[str, Any], idx: int) -> Dict[str, np.ndarray]:
    """
    Lazy load ClusterReconstruction only for the requested idx.
    This keeps memory much flatter than reading ClusterReconstruction[:] upfront.
    """
    A = np.asarray(data["Amplitude"][idx], dtype=float)
    L = np.asarray(data["DominantWavelength"][idx], dtype=float)
    T = np.asarray(data["Angle"][idx], dtype=float)

    R = np.full_like(A, np.nan)
    with Dataset(data["path"], "r") as nc:
        if "ClusterReconstruction" in nc.variables:
            R = np.asarray(nc.variables["ClusterReconstruction"][idx], dtype=float)

    return dict(R=R, A=A, L=L, T=T)


def compute_match_metrics(
    A_src: np.ndarray, L_src: np.ndarray, T_src: np.ndarray,
    A_cand: np.ndarray, L_cand: np.ndarray, T_cand: np.ndarray,
) -> Dict[str, Any]:
    mask_src = np.isfinite(A_src)
    mask_cand = np.isfinite(A_cand)
    inter = mask_src & mask_cand
    union = mask_src | mask_cand

    n_inter = int(np.count_nonzero(inter))
    n_union = int(np.count_nonzero(union))
    iou = (n_inter / n_union) if n_union > 0 else 0.0

    lam_src = L_src[inter]
    lam_cand = L_cand[inter]
    lam_src_med = nanpercentile_safe(lam_src, 50)
    lam_cand_med = nanpercentile_safe(lam_cand, 50)

    if np.isfinite(lam_src_med) and np.isfinite(lam_cand_med) and lam_src_med > 0 and lam_cand_med > 0:
        lam_ratio = max(lam_src_med, lam_cand_med) / min(lam_src_med, lam_cand_med)
    else:
        lam_ratio = np.nan

    ang_src_deg = circ_median_deg_180(T_src, inter)
    ang_cand_deg = circ_median_deg_180(T_cand, inter)
    ang_diff_deg = circ_diff_deg_180(ang_src_deg, ang_cand_deg)

    return dict(
        n_intersection=n_inter,
        iou=float(iou),
        lam_src_med=float(lam_src_med),
        lam_cand_med=float(lam_cand_med),
        lam_ratio=float(lam_ratio),
        ang_src_deg=float(ang_src_deg),
        ang_cand_deg=float(ang_cand_deg),
        ang_diff_deg=float(ang_diff_deg),
    )


def passes_thresholds(m: Dict[str, Any]) -> bool:
    if m["iou"] < MIN_IOU:
        return False
    if m["n_intersection"] < MIN_INTERSECTION_PIX:
        return False
    if (not np.isfinite(m["lam_ratio"])) or (m["lam_ratio"] > MAX_LAMBDA_MED_RATIO):
        return False
    if (not np.isfinite(m["ang_diff_deg"])) or (m["ang_diff_deg"] > MAX_ANGLE_DIFF_DEG):
        return False
    return True


def score(m: Dict[str, Any]) -> float:
    ang = m["ang_diff_deg"]
    if not np.isfinite(ang):
        ang = 999.0
    r = m["lam_ratio"]
    if (not np.isfinite(r)) or r <= 0:
        return 0.0
    return float(m["iou"]) * (1.0 / float(r)) / (1.0 + float(ang))


def plot_side_by_side_4rows(
    left: Dict[str, np.ndarray],
    right: Dict[str, np.ndarray],
    title_left: str,
    title_right: str,
    outpath: Path,
    dx_km: float,
    dy_km: float,
) -> None:
    ny, nx = left["A"].shape
    x1d_km = np.arange(nx, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    Rl_vmin, Rl_vmax = robust_limits(left["R"])
    Rr_vmin, Rr_vmax = robust_limits(right["R"])
    Al_vmin, Al_vmax = robust_limits(left["A"])
    Ar_vmin, Ar_vmax = robust_limits(right["A"])

    L_vmin, L_vmax = robust_limits(np.concatenate([left["L"].ravel(), right["L"].ravel()]), 5, 95)
    T_vmin, T_vmax = robust_limits(np.concatenate([left["T"].ravel(), right["T"].ravel()]))

    fig, axs = plt.subplots(4, 2, figsize=(16, 14), constrained_layout=True, sharex=True, sharey=True)

    def _panel(ax, Z, vmin, vmax, label, cmap=None):
        im = ax.pcolormesh(x1d_km, y1d_km, Z, shading="auto", vmin=vmin, vmax=vmax, cmap=cmap)
        cb = fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01)
        cb.set_label(label)
        ax.set_aspect("equal", adjustable="box")

    axs[0, 0].set_title(title_left)
    axs[0, 1].set_title(title_right)

    _panel(axs[0, 0], left["R"], Rl_vmin, Rl_vmax, "arb", cmap="RdBu_r")
    _panel(axs[0, 1], right["R"], Rr_vmin, Rr_vmax, "arb", cmap="RdBu_r")
    axs[0, 0].set_ylabel("Recon\ny (km)")

    _panel(axs[1, 0], left["A"], Al_vmin, Al_vmax, "K")
    _panel(axs[1, 1], right["A"], Ar_vmin, Ar_vmax, "kR")
    axs[1, 0].set_ylabel("Amp\ny (km)")

    _panel(axs[2, 0], left["L"], L_vmin, L_vmax, "km")
    _panel(axs[2, 1], right["L"], L_vmin, L_vmax, "km")
    axs[2, 0].set_ylabel("Lambda\ny (km)")

    _panel(axs[3, 0], left["T"], T_vmin, T_vmax, "rad")
    _panel(axs[3, 1], right["T"], T_vmin, T_vmax, "rad")
    axs[3, 0].set_ylabel("Angle\ny (km)")

    for r in range(4):
        for c in range(2):
            axs[r, c].set_xlabel("x (km)")

    save_fig(fig, outpath)


# --------------------------
# Flatten swath BEFORE slicing
# --------------------------
def flatten_swath(Z: np.ndarray, target_ny: int | None = None) -> np.ndarray:
    ny, nx = Z.shape
    out = np.full_like(Z, np.nan)

    bottoms = np.full(nx, np.nan, dtype=float)
    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size:
            bottoms[j] = valid_idx.max()

    valid_bottoms = bottoms[np.isfinite(bottoms)]
    if valid_bottoms.size == 0:
        med = float(np.nanmedian(Z)) if np.isfinite(np.nanmedian(Z)) else 0.0
        out = np.nan_to_num(Z, nan=med, posinf=med, neginf=med)
        if target_ny is not None and ny > target_ny:
            out = out[-target_ny:, :]
        return out

    max_bottom = int(valid_bottoms.max())

    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size == 0:
            continue
        bottom = int(valid_idx.max())
        shift = max_bottom - bottom
        if shift == 0:
            out[:, j] = col
        elif shift > 0:
            out[shift:, j] = col[: ny - shift]
        else:
            shift_up = -shift
            out[: ny - shift_up, j] = col[shift_up:]

    med = float(np.nanmedian(out)) if np.isfinite(np.nanmedian(out)) else 0.0
    out = np.nan_to_num(out, nan=med, posinf=med, neginf=med)

    if target_ny is not None and ny > target_ny:
        start = max_bottom + 1 - target_ny
        start = max(start, 0)
        end = start + target_ny
        out = out[start:end, :]

    return out


# --------------------------
# L6A cloning with PASSED-only cluster variables + new vars (u,v,N^2,Bearing, MF)
# --------------------------
def create_passed_only_l6a_with_new_vars(
    src_l6a: Path,
    dst_l6a: Path,
    passed_cluster_indices: np.ndarray,
    time_index: int,
) -> Tuple[str, str, str]:
    """
    Creates a passed only copy of L6A and adds

    Full field vars
    Bearing degrees, 0 East, CCW, written only for top y_tile rows at time_index

    Cluster masked vars
    Temp_cluster
    u_cluster
    v_cluster
    NSquared_cluster
    Bearing_cluster
    MF_cluster
    MF_zonal_cluster        MF cos^2(phi)
    MF_meridional_cluster   MF sin^2(phi)

    Per slice sums on tile grid
    MF_slice_sum
    MF_zonal_slice_sum
    MF_meridional_slice_sum

    Full field stitched top rows at time_index
    MF_toprows
    MF_zonal_toprows
    MF_meridional_toprows
    """
    passed_cluster_indices = np.asarray(passed_cluster_indices, dtype=int)
    if passed_cluster_indices.size == 0:
        raise ValueError("No passed clusters. Output would be empty.")

    dst_l6a.parent.mkdir(parents=True, exist_ok=True)

    with Dataset(src_l6a, "r") as src:
        amp_var = src.variables["Amplitude"]
        cluster_dim, ytile_dim, xtile_dim = amp_var.dimensions[:3]

        # IMPORTANT: do not read amp_all[:] just to get shapes
        _, y_tile, x_tile = amp_var.shape

        temp_var = src.variables[TEMP_VAR_NAME]
        temp_dims = temp_var.dimensions
        temp_shape = temp_var.shape
        temp_raw_2d = np.asarray(temp_var[time_index], dtype=float)

        # full field u v N2
        u_raw_2d = (
            np.asarray(src.variables["u"][time_index], dtype=float)
            if "u" in src.variables
            else np.full_like(temp_raw_2d, U_VALUE, dtype=float)
        )
        v_raw_2d = (
            np.asarray(src.variables["v"][time_index], dtype=float)
            if "v" in src.variables
            else np.full_like(temp_raw_2d, V_VALUE, dtype=float)
        )
        n2_raw_2d = (
            np.asarray(src.variables["NSquared"][time_index], dtype=float)
            if "NSquared" in src.variables
            else np.full_like(temp_raw_2d, N2_VALUE, dtype=float)
        )

        # bearing from lat lon if present
        if ("Latitude" in src.variables) and ("Longitude" in src.variables):
            lat_raw_2d = np.asarray(src.variables["Latitude"][time_index], dtype=float)
            lon_raw_2d = np.asarray(src.variables["Longitude"][time_index], dtype=float)
            bearing_raw_2d = compute_bearing_to_x_neighbor_east_ccw(lat_raw_2d, lon_raw_2d)
        else:
            bearing_raw_2d = np.full_like(temp_raw_2d, np.nan, dtype=float)

        # flatten to match tile y dimension
        temp_full_2d = flatten_swath(temp_raw_2d, target_ny=y_tile)
        u_full_2d = flatten_swath(u_raw_2d, target_ny=y_tile)
        v_full_2d = flatten_swath(v_raw_2d, target_ny=y_tile)
        n2_full_2d = flatten_swath(n2_raw_2d, target_ny=y_tile)
        bearing_full_2d = flatten_swath(bearing_raw_2d, target_ny=y_tile)

        n_passed = int(passed_cluster_indices.size)

        with Dataset(dst_l6a, "w") as dst:
            # global attrs
            for aname in src.ncattrs():
                dst.setncattr(aname, src.getncattr(aname))

            # dimensions, subset cluster dim
            for dname, dim in src.dimensions.items():
                if dname == cluster_dim:
                    dst.createDimension(dname, n_passed)
                else:
                    dst.createDimension(dname, (None if dim.isunlimited() else len(dim)))

            # variables copied, subset cluster dim vars
            for vname, vsrc in src.variables.items():
                fill_value = vsrc.getncattr("_FillValue") if "_FillValue" in vsrc.ncattrs() else None
                vdst = dst.createVariable(
                    vname,
                    vsrc.dtype,
                    vsrc.dimensions,
                    zlib=True,
                    complevel=4,
                    fill_value=fill_value,
                )
                for aname in vsrc.ncattrs():
                    if aname != "_FillValue":
                        vdst.setncattr(aname, vsrc.getncattr(aname))

                if cluster_dim in vsrc.dimensions:
                    ax = vsrc.dimensions.index(cluster_dim)
                    data_sub = np.take(vsrc[:], passed_cluster_indices, axis=ax)
                    vdst[:] = data_sub
                else:
                    vdst[:] = vsrc[:]

            # full field Bearing
            bearing_full = np.zeros(temp_shape, dtype=np.float32)
            bearing_full[time_index, 0:y_tile, :] = bearing_full_2d.astype(np.float32)

            if "Bearing" in dst.variables:
                dst.variables["Bearing"][:] = bearing_full
            else:
                vB = dst.createVariable("Bearing", "f4", temp_dims, zlib=True, complevel=4)
                vB.setncattr("long_name", "Geodesic bearing to +x neighbor (0=East, CCW)")
                vB.setncattr("units", "degree")
                vB[:] = bearing_full

            # allocate cluster masked arrays (NaN outside cluster mask)
            temp_cluster = np.full((n_passed, y_tile, x_tile), np.nan, dtype=np.float32)
            u_cluster = np.full_like(temp_cluster, np.nan)
            v_cluster = np.full_like(temp_cluster, np.nan)
            n2_cluster = np.full_like(temp_cluster, np.nan)
            bearing_cluster = np.full_like(temp_cluster, np.nan)

            mf_cluster = np.full_like(temp_cluster, np.nan)
            mf_zonal_cluster = np.full_like(temp_cluster, np.nan)
            mf_meridional_cluster = np.full_like(temp_cluster, np.nan)

            amp_passed = np.asarray(dst.variables["Amplitude"][:], dtype=float)
            lam_passed = np.asarray(dst.variables["DominantWavelength"][:], dtype=float)
            slice_passed = np.asarray(dst.variables["SlicesNo"][:], dtype=int)
            ang_rad_passed = np.asarray(dst.variables["Angle"][:], dtype=float)

            for j in range(n_passed):
                s = int(slice_passed[j])
                x0 = s * x_tile
                x1 = x0 + x_tile

                mask = np.isfinite(amp_passed[j])

                Ttile = temp_full_2d[:, x0:x1]
                Utile = u_full_2d[:, x0:x1]
                Vtile = v_full_2d[:, x0:x1]
                N2tile = n2_full_2d[:, x0:x1]
                Btile = bearing_full_2d[:, x0:x1]

                temp_cluster[j][mask] = Ttile[mask]
                u_cluster[j][mask] = Utile[mask]
                v_cluster[j][mask] = Vtile[mask]
                n2_cluster[j][mask] = N2tile[mask]
                bearing_cluster[j][mask] = Btile[mask]

                A = amp_passed[j]
                Lx = lam_passed[j]
                ang_deg_tile = (np.rad2deg(ang_rad_passed[j]) % 360.0)

                good = mask
                good = good & np.isfinite(Ttile) & np.isfinite(N2tile) & np.isfinite(Lx) & np.isfinite(Btile)
                good = good & (Ttile != 0.0) & (N2tile > 0.0) & (Lx > 0.0)

                if np.any(good):
                    mf_val = (
                        0.5
                        * (G_MS2**2)
                        / N2tile[good]
                        * (LAMBDA_Z_KM / Lx[good])
                        * (A[good] / Ttile[good]) ** 2
                        * (1.0 / (C_CANCEL**2))
                    )
                    mf_cluster[j][good] = mf_val

                    # phi is geographic propagation angle, degrees, 0 East, CCW
                    phi_deg = (Btile[good] + ang_deg_tile[good]) % 360.0
                    phi_rad = np.deg2rad(phi_deg)

                    mf_zonal_cluster[j][good] = mf_val * (np.cos(phi_rad) ** 2)
                    mf_meridional_cluster[j][good] = mf_val * (np.sin(phi_rad) ** 2)

            # write cluster vars
            if "Temp_cluster" in dst.variables:
                dst.variables["Temp_cluster"][:] = temp_cluster
            else:
                v = dst.createVariable("Temp_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim))
                v.setncattr("long_name", "Temperature, cluster masked")
                v.setncattr("units", "K")
                v[:] = temp_cluster

            if "u_cluster" in dst.variables:
                dst.variables["u_cluster"][:] = u_cluster
            else:
                v = dst.createVariable("u_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim))
                v.setncattr("long_name", "u, cluster masked")
                v.setncattr("units", "m s-1")
                v[:] = u_cluster

            if "v_cluster" in dst.variables:
                dst.variables["v_cluster"][:] = v_cluster
            else:
                v = dst.createVariable("v_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim))
                v.setncattr("long_name", "v, cluster masked")
                v.setncattr("units", "m s-1")
                v[:] = v_cluster

            if "NSquared_cluster" in dst.variables:
                dst.variables["NSquared_cluster"][:] = n2_cluster
            else:
                v = dst.createVariable("NSquared_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim))
                v.setncattr("long_name", "N^2, cluster masked")
                v.setncattr("units", "s-2")
                v[:] = n2_cluster

            if "Bearing_cluster" in dst.variables:
                dst.variables["Bearing_cluster"][:] = bearing_cluster
            else:
                vBC = dst.createVariable("Bearing_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim))
                vBC.setncattr("long_name", "Geodesic bearing to +x neighbor (0=East, CCW), cluster masked")
                vBC.setncattr("units", "degree")
                vBC[:] = bearing_cluster

            if "MF_cluster" in dst.variables:
                dst.variables["MF_cluster"][:] = mf_cluster
            else:
                vMF = dst.createVariable("MF_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim))
                vMF.setncattr("long_name", "Momentum flux proxy per pixel, cluster masked")
                vMF.setncattr("units", "m2 s-2")
                vMF[:] = mf_cluster

            if "MF_zonal_cluster" in dst.variables:
                dst.variables["MF_zonal_cluster"][:] = mf_zonal_cluster
            else:
                vMFZ = dst.createVariable("MF_zonal_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim))
                vMFZ.setncattr("long_name", "Zonal aligned MF magnitude share, MF cos^2(phi)")
                vMFZ.setncattr("units", "m2 s-2")
                vMFZ[:] = mf_zonal_cluster

            if "MF_meridional_cluster" in dst.variables:
                dst.variables["MF_meridional_cluster"][:] = mf_meridional_cluster
            else:
                vMFM = dst.createVariable("MF_meridional_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim))
                vMFM.setncattr("long_name", "Meridional aligned MF magnitude share, MF sin^2(phi)")
                vMFM.setncattr("units", "m2 s-2")
                vMFM[:] = mf_meridional_cluster

            # slice sums and stitched top rows
            smin, smax = int(np.min(slice_passed)), int(np.max(slice_passed))
            full_slices = np.arange(smin, smax + 1)

            if "slice" not in dst.dimensions:
                dst.createDimension("slice", len(full_slices))

            if "SliceNo_passed_slices" in dst.variables:
                dst.variables["SliceNo_passed_slices"][:] = full_slices
            else:
                dst.createVariable("SliceNo_passed_slices", "i4", ("slice",))[:] = full_slices

            mf_slice_sum = np.zeros((len(full_slices), y_tile, x_tile), dtype=np.float32)
            mfz_slice_sum = np.zeros_like(mf_slice_sum)
            mfm_slice_sum = np.zeros_like(mf_slice_sum)

            for k, s in enumerate(full_slices):
                idx = np.where(slice_passed == s)[0]
                if idx.size > 0:
                    mf_slice_sum[k] = np.nansum(mf_cluster[idx], axis=0)
                    mfz_slice_sum[k] = np.nansum(mf_zonal_cluster[idx], axis=0)
                    mfm_slice_sum[k] = np.nansum(mf_meridional_cluster[idx], axis=0)

            if "MF_slice_sum" in dst.variables:
                dst.variables["MF_slice_sum"][:] = mf_slice_sum
            else:
                v = dst.createVariable("MF_slice_sum", "f4", ("slice", ytile_dim, xtile_dim))
                v.setncattr("long_name", "Per slice MF sum (passed clusters), tile grid")
                v.setncattr("units", "m2 s-2")
                v[:] = mf_slice_sum

            if "MF_zonal_slice_sum" in dst.variables:
                dst.variables["MF_zonal_slice_sum"][:] = mfz_slice_sum
            else:
                v = dst.createVariable("MF_zonal_slice_sum", "f4", ("slice", ytile_dim, xtile_dim))
                v.setncattr("long_name", "Per slice zonal MF magnitude share sum")
                v.setncattr("units", "m2 s-2")
                v[:] = mfz_slice_sum

            if "MF_meridional_slice_sum" in dst.variables:
                dst.variables["MF_meridional_slice_sum"][:] = mfm_slice_sum
            else:
                v = dst.createVariable("MF_meridional_slice_sum", "f4", ("slice", ytile_dim, xtile_dim))
                v.setncattr("long_name", "Per slice meridional MF magnitude share sum")
                v.setncattr("units", "m2 s-2")
                v[:] = mfm_slice_sum

            MF_toprows = np.zeros(temp_shape, dtype=np.float32)
            MFZ_toprows = np.zeros(temp_shape, dtype=np.float32)
            MFM_toprows = np.zeros(temp_shape, dtype=np.float32)

            for k, s in enumerate(full_slices):
                x0 = s * x_tile
                x1 = x0 + x_tile
                MF_toprows[time_index, 0:y_tile, x0:x1] = mf_slice_sum[k]
                MFZ_toprows[time_index, 0:y_tile, x0:x1] = mfz_slice_sum[k]
                MFM_toprows[time_index, 0:y_tile, x0:x1] = mfm_slice_sum[k]

            if "MF_toprows" in dst.variables:
                dst.variables["MF_toprows"][:] = MF_toprows
            else:
                v = dst.createVariable("MF_toprows", "f4", temp_dims)
                v.setncattr("long_name", "Stitched MF sum (passed clusters) on top tile rows only")
                v.setncattr("units", "m2 s-2")
                v[:] = MF_toprows

            if "MF_zonal_toprows" in dst.variables:
                dst.variables["MF_zonal_toprows"][:] = MFZ_toprows
            else:
                v = dst.createVariable("MF_zonal_toprows", "f4", temp_dims)
                v.setncattr("long_name", "Stitched zonal MF magnitude share, top tile rows only")
                v.setncattr("units", "m2 s-2")
                v[:] = MFZ_toprows

            if "MF_meridional_toprows" in dst.variables:
                dst.variables["MF_meridional_toprows"][:] = MFM_toprows
            else:
                v = dst.createVariable("MF_meridional_toprows", "f4", temp_dims)
                v.setncattr("long_name", "Stitched meridional MF magnitude share, top tile rows only")
                v.setncattr("units", "m2 s-2")
                v[:] = MFM_toprows

        # help Python actually let go of big arrays once the file is written
        del temp_raw_2d, u_raw_2d, v_raw_2d, n2_raw_2d, bearing_raw_2d
        del temp_full_2d, u_full_2d, v_full_2d, n2_full_2d, bearing_full_2d
        del temp_cluster, u_cluster, v_cluster, n2_cluster, bearing_cluster
        del mf_cluster, mf_zonal_cluster, mf_meridional_cluster
        gc.collect()

    return cluster_dim, ytile_dim, xtile_dim


# --------------------------
# Plot per passed cluster panels, memory safe (read one cluster at a time)
# --------------------------
def plot_passed_clusters_panels(nc_path: Path, outdir: Path) -> None:
    """
    Per passed cluster panels, 5 rows x 2 columns

    Row 1  Recon | Amplitude
    Row 2  Lambda | Temperature
    Row 3  Angle deg | Bearing deg
    Row 4  N2 | MF
    Row 5  MF zonal share | MF meridional share
    """
    outdir.mkdir(parents=True, exist_ok=True)

    with Dataset(nc_path, "r") as nc:
        # assume cluster dimension name exists on Amplitude
        cluster_dim = nc.variables["Amplitude"].dimensions[0]
        ytile_dim = nc.variables["Amplitude"].dimensions[1]
        xtile_dim = nc.variables["Amplitude"].dimensions[2]
        n_cluster = nc.dimensions[cluster_dim].size

        slc = np.asarray(nc.variables["SlicesNo"][:], dtype=int)

        ny = nc.dimensions[ytile_dim].size
        nx = nc.dimensions[xtile_dim].size
        x1d_km = np.arange(nx, dtype=float) * DX_KM
        y1d_km = np.arange(ny, dtype=float) * DY_KM

        ang_vmin, ang_vmax = (0.0, 360.0)
        brg_vmin, brg_vmax = (0.0, 360.0)

        n2_center = float(N2_VALUE)
        n2_vmin, n2_vmax = (0.5 * n2_center, 1.5 * n2_center)

        for i in range(n_cluster):
            s = int(slc[i])

            amp = np.asarray(nc.variables["Amplitude"][i], dtype=float)
            lam = np.asarray(nc.variables["DominantWavelength"][i], dtype=float)
            ang_rad = np.asarray(nc.variables["Angle"][i], dtype=float)
            ang_deg = (np.rad2deg(ang_rad) % 360.0)

            brg = (
                np.asarray(nc.variables["Bearing_cluster"][i], dtype=float)
                if "Bearing_cluster" in nc.variables
                else np.full_like(amp, np.nan)
            )
            n2 = np.asarray(nc.variables["NSquared_cluster"][i], dtype=float)
            tmp = np.asarray(nc.variables["Temp_cluster"][i], dtype=float)
            mf = np.asarray(nc.variables["MF_cluster"][i], dtype=float)

            mfz = (
                np.asarray(nc.variables["MF_zonal_cluster"][i], dtype=float)
                if "MF_zonal_cluster" in nc.variables
                else np.full_like(mf, np.nan)
            )
            mfm = (
                np.asarray(nc.variables["MF_meridional_cluster"][i], dtype=float)
                if "MF_meridional_cluster" in nc.variables
                else np.full_like(mf, np.nan)
            )

            if "ClusterReconstruction" in nc.variables:
                recon = np.asarray(nc.variables["ClusterReconstruction"][i], dtype=float)
            else:
                recon = np.full_like(amp, np.nan)

            r_vmin, r_vmax = robust_limits(recon, 2, 98)
            a_vmin, a_vmax = robust_limits(amp, 2, 98)
            l_vmin, l_vmax = robust_limits(lam, 5, 95)
            t_vmin, t_vmax = robust_limits(tmp, 2, 98)

            mf_vmin, mf_vmax = robust_limits(mf, 2, 98)
            mfz_vmin, mfz_vmax = robust_limits(mfz, 2, 98)
            mfm_vmin, mfm_vmax = robust_limits(mfm, 2, 98)

            fig, axs = plt.subplots(
                5, 2,
                figsize=(16, 22),
                constrained_layout=True,
                sharex=True,
                sharey=True,
            )

            def _panel(ax, Z, vmin, vmax, label, cmap=None):
                im = ax.pcolormesh(
                    x1d_km, y1d_km, Z,
                    shading="auto",
                    vmin=vmin,
                    vmax=vmax,
                    cmap=cmap,
                )
                cb = fig.colorbar(im, ax=ax, fraction=0.04, pad=0.02)
                cb.set_label(label)
                ax.set_aspect("equal", adjustable="box")

            axs[0, 0].set_title(f"Passed cluster {i:04d}  SliceNo {s}")

            _panel(axs[0, 0], recon, r_vmin, r_vmax, "Recon", cmap="RdBu_r")
            _panel(axs[0, 1], amp, a_vmin, a_vmax, "Amplitude K")

            _panel(axs[1, 0], lam, l_vmin, l_vmax, "DominantWavelength km")
            _panel(axs[1, 1], tmp, t_vmin, t_vmax, "Temperature K")

            _panel(axs[2, 0], ang_deg, ang_vmin, ang_vmax, "Angle deg propagation")
            _panel(axs[2, 1], brg, brg_vmin, brg_vmax, "Bearing deg 0 East CCW")

            _panel(axs[3, 0], n2, n2_vmin, n2_vmax, "N2 s-2")
            _panel(axs[3, 1], mf, mf_vmin, mf_vmax, "MF m2 s-2")

            _panel(axs[4, 0], mfz, mfz_vmin, mfz_vmax, "MF zonal share cos2")
            _panel(axs[4, 1], mfm, mfm_vmin, mfm_vmax, "MF meridional share sin2")

            for r in range(5):
                axs[r, 0].set_ylabel("y km")
                axs[r, 1].set_ylabel("y km")
            axs[4, 0].set_xlabel("x km")
            axs[4, 1].set_xlabel("x km")

            outpng = outdir / (
                f"passed_cluster_{i:04d}_slice_{s:03d}"
                "_Recon_Amp_Lam_Temp_Ang_Bearing_N2_MF_MFz_MFm.png"
            )
            save_fig(fig, outpng)

            # free per cluster arrays aggressively
            del amp, lam, ang_rad, ang_deg, brg, n2, tmp, mf, mfz, mfm, recon, fig, axs
            gc.collect()

    mpl.pyplot.close("all")
    gc.collect()
    print("Per cluster plots saved to:", outdir.resolve())


def plot_stitched_mf_components(nc_path: Path, outpng: Path, time_index: int) -> None:
    """
    Makes a stitched 3 panel figure

    Panel 1 MF_toprows
    Panel 2 MF_zonal_toprows
    Panel 3 MF_meridional_toprows
    """
    outpng.parent.mkdir(parents=True, exist_ok=True)

    with Dataset(nc_path, "r") as nc:
        MF = np.asarray(nc.variables["MF_toprows"][time_index], dtype=float)
        MFZ = np.asarray(nc.variables["MF_zonal_toprows"][time_index], dtype=float)
        MFM = np.asarray(nc.variables["MF_meridional_toprows"][time_index], dtype=float)

    ny, nx = MF.shape
    x1d_km = np.arange(nx, dtype=float) * DX_KM
    y1d_km = np.arange(ny, dtype=float) * DY_KM

    stacked = np.concatenate([MF.ravel(), MFZ.ravel(), MFM.ravel()])
    vmin, vmax = robust_limits(stacked, 2, 99.9)

    fig, axs = plt.subplots(3, 1, figsize=(22, 8), constrained_layout=True, sharex=True, sharey=True)

    im0 = axs[0].pcolormesh(x1d_km, y1d_km, MF, shading="auto", vmin=vmin, vmax=vmax)
    axs[0].set_title("Stitched MF passed clusters summed")
    axs[0].set_ylabel("y km")
    axs[0].set_aspect("equal", adjustable="box")

    im1 = axs[1].pcolormesh(x1d_km, y1d_km, MFZ, shading="auto", vmin=vmin, vmax=vmax)
    axs[1].set_title("Stitched MF zonal share cos2")
    axs[1].set_ylabel("y km")
    axs[1].set_aspect("equal", adjustable="box")

    im2 = axs[2].pcolormesh(x1d_km, y1d_km, MFM, shading="auto", vmin=vmin, vmax=vmax)
    axs[2].set_title("Stitched MF meridional share sin2")
    axs[2].set_xlabel("x km")
    axs[2].set_ylabel("y km")
    axs[2].set_aspect("equal", adjustable="box")

    for ax in axs:
        ax.set_ylim(0, 600)

    cbar = fig.colorbar(im2, ax=axs, fraction=0.02, pad=0.01)
    cbar.set_label("MF m2 s-2")

    save_fig(fig, outpng)

    del MF, MFZ, MFM, stacked, fig, axs
    gc.collect()


# --------------------------
# Main: matching -> passed clusters -> output passed-only netCDF -> stitched MF plot
# --------------------------
l6c = load_l6_for_matching(L6C_PATH)
l6a_match = load_l6_for_matching(L6A_PATH)

all_slices_a = np.unique(l6a_match["SlicesNo"]).astype(int)
all_slices_a = all_slices_a[all_slices_a >= 0]
all_slices_a = np.sort(all_slices_a)

print("Slices in L6A:", all_slices_a.tolist())

summary_lines: List[str] = []
summary_lines.append(f"L6A: {L6A_PATH}\n")
summary_lines.append(f"L6C: {L6C_PATH}\n\n")
summary_lines.append("slice  nA  nC  nBestSaved  nPassed  note\n")

passed_global_indices: List[int] = []

for slice_no in all_slices_a:
    idx_a = np.where(l6a_match["SlicesNo"] == int(slice_no))[0]
    idx_c = np.where(l6c["SlicesNo"] == int(slice_no))[0]

    slice_dir = OUTDIR / f"slice_{slice_no:03d}"
    pairs_dir = slice_dir / "pairs"
    pairs_dir.mkdir(parents=True, exist_ok=True)

    if idx_a.size == 0:
        summary_lines.append(f"{slice_no:5d}  {0:2d}  {idx_c.size:2d}  {0:10d}  {0:7d}  no L6A clusters\n")
        continue
    if idx_c.size == 0:
        (slice_dir / "NO_L6C_CLUSTERS.txt").write_text(f"No L6C clusters for slice {slice_no}\n")
        summary_lines.append(f"{slice_no:5d}  {idx_a.size:2d}  {0:2d}  {0:10d}  {0:7d}  no L6C clusters\n")
        continue

    best_rows: List[Dict[str, Any]] = []

    # Stream candidate lines instead of storing giant lists of dicts
    cand_lines: List[str] = []
    cand_header = []
    cand_header.append(f"L6A file: {L6A_PATH}\n")
    cand_header.append(f"L6C file: {L6C_PATH}\n")
    cand_header.append(f"Slice: {slice_no}\n\n")
    cand_header.append(
        "cluster_a  cluster_c  passes  score    iou    inter  "
        "lam_ratio  ang_diff_deg  thetaA_deg  thetaC_deg  lamA_med  lamC_med\n"
    )

    for a_idx in idx_a:
        a_idx = int(a_idx)
        A_a = l6a_match["Amplitude"][a_idx]
        L_a = l6a_match["DominantWavelength"][a_idx]
        T_a = l6a_match["Angle"][a_idx]

        candidates: List[Dict[str, Any]] = []
        for c_idx in idx_c:
            c_idx = int(c_idx)
            A_c = l6c["Amplitude"][c_idx]
            L_c = l6c["DominantWavelength"][c_idx]
            T_c = l6c["Angle"][c_idx]

            m = compute_match_metrics(A_a, L_a, T_a, A_c, L_c, T_c)
            m["cluster_a"] = a_idx
            m["cluster_c"] = c_idx
            m["passes"] = bool(passes_thresholds(m))
            m["score"] = float(score(m))

            candidates.append(m)

            cand_lines.append(
                f"{m['cluster_a']:9d}  {m['cluster_c']:9d}  {int(m['passes']):6d}  {m['score']:6.3f}  "
                f"{m['iou']:.3f}  {m['n_intersection']:5d}  "
                f"{m['lam_ratio']:8.3f}  {m['ang_diff_deg']:12.1f}  "
                f"{m['ang_src_deg']:10.1f}  {m['ang_cand_deg']:10.1f}  "
                f"{m['lam_src_med']:8.1f}  {m['lam_cand_med']:8.1f}\n"
            )

        passed = [c for c in candidates if c["passes"]]
        if passed:
            best = max(passed, key=lambda d: d["score"])
        else:
            best = max(candidates, key=lambda d: d["score"])

        if best["passes"] or ASSIGN_BEST_EVEN_IF_FAILS:
            best_rows.append(best)

            # Lazy map loads include ClusterReconstruction only for these two indices
            left_maps = get_cluster_maps(l6a_match, best["cluster_a"])
            right_maps = get_cluster_maps(l6c, best["cluster_c"])

            pass_tag = "PASS" if best["passes"] else "FAIL"
            outpng = pairs_dir / f"slice_{slice_no:03d}_A_{best['cluster_a']:04d}_to_C_{best['cluster_c']:04d}_{pass_tag}.png"

            title_left = (
                f"L6A Temp | slice {slice_no} | idx {best['cluster_a']} | "
                f"n={np.count_nonzero(np.isfinite(left_maps['A']))}"
            )
            title_right = (
                f"L6C Rad | slice {slice_no} | idx {best['cluster_c']} | "
                f"IoU={best['iou']:.3f} inter={best['n_intersection']} "
                f"ratio={best['lam_ratio']:.3f} Δθ={best['ang_diff_deg']:.1f}°"
            )

            plot_side_by_side_4rows(
                left=left_maps,
                right=right_maps,
                title_left=title_left,
                title_right=title_right,
                outpath=outpng,
                dx_km=DX_KM,
                dy_km=DY_KM,
            )

            # free per plot
            del left_maps, right_maps
            gc.collect()

        del candidates
        gc.collect()

    cand_path = slice_dir / f"slice_{slice_no:03d}_ALL_candidates_ratio.txt"
    cand_path.write_text("".join(cand_header) + "".join(cand_lines))
    del cand_lines
    gc.collect()

    best_path = slice_dir / f"slice_{slice_no:03d}_BEST_matches_A_to_C_ratio.txt"
    best_lines = []
    best_lines.append(f"L6A file: {L6A_PATH}\n")
    best_lines.append(f"L6C file: {L6C_PATH}\n")
    best_lines.append(f"Slice: {slice_no}\n\n")
    best_lines.append(
        "cluster_a  best_cluster_c  passes  score    iou    inter  "
        "lam_ratio  ang_diff_deg  thetaA_deg  thetaC_deg  lamA_med  lamC_med\n"
    )
    for m in sorted(best_rows, key=lambda d: d["cluster_a"]):
        best_lines.append(
            f"{m['cluster_a']:9d}  {m['cluster_c']:14d}  {int(m['passes']):6d}  {m['score']:6.3f}  "
            f"{m['iou']:.3f}  {m['n_intersection']:5d}  "
            f"{m['lam_ratio']:8.3f}  {m['ang_diff_deg']:12.1f}  "
            f"{m['ang_src_deg']:10.1f}  {m['ang_cand_deg']:10.1f}  "
            f"{m['lam_src_med']:8.1f}  {m['lam_cand_med']:8.1f}\n"
        )
    best_path.write_text("".join(best_lines))

    passed_here = [int(m["cluster_a"]) for m in best_rows if bool(m["passes"])]
    passed_global_indices.extend(passed_here)

    summary_lines.append(
        f"{slice_no:5d}  {idx_a.size:2d}  {idx_c.size:2d}  {len(best_rows):10d}  {len(passed_here):7d}  ok\n"
    )

    del best_rows, best_lines
    gc.collect()

summary_path = OUTDIR / "SUMMARY_all_slices.txt"
summary_path.write_text("".join(summary_lines))

passed_global_indices = sorted(set(passed_global_indices))
passed_global_indices_np = np.asarray(passed_global_indices, dtype=int)
print("Total passed clusters:", passed_global_indices_np.size)

if passed_global_indices_np.size == 0:
    raise RuntimeError("No passed clusters found. Nothing to save or stitch.")

cluster_dim_name, ytile_dim_name, xtile_dim_name = create_passed_only_l6a_with_new_vars(
    src_l6a=L6A_PATH,
    dst_l6a=OUT_L6A_PASSED,
    passed_cluster_indices=passed_global_indices_np,
    time_index=TIME_INDEX,
)

# plot stitched MF quicklook (single panel)
with Dataset(OUT_L6A_PASSED, "r") as nc:
    MF2 = np.asarray(nc.variables["MF_toprows"][TIME_INDEX], dtype=float)
    ny, nx = MF2.shape
    x1d_km = np.arange(nx, dtype=float) * DX_KM
    y1d_km = np.arange(ny, dtype=float) * DY_KM

    vmin, vmax = robust_limits(MF2, 2, 99.9)

    fig, ax = plt.subplots(figsize=(22, 3), constrained_layout=True)
    im = ax.pcolormesh(x1d_km, y1d_km, MF2, shading="auto", vmin=vmin, vmax=vmax)
    ax.set_title("Momentum Flux (passed clusters summed)")
    ax.set_xlabel("x (km)")
    ax.set_ylabel("y (km)")
    ax.set_ylim(0, 600)
    ax.set_aspect("equal", adjustable="box")
    fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01).set_label("MF (m2 s-2)")
    save_fig(fig, OUTDIR / "MF_stitched.png")

del MF2, fig, ax
gc.collect()

# heavy plotting
plot_passed_clusters_panels(
    nc_path=OUT_L6A_PASSED,
    outdir=OUTDIR / "passed_cluster_plots_panels",
)

plot_stitched_mf_components(
    nc_path=OUT_L6A_PASSED,
    outpng=OUTDIR / "MF_stitched_components.png",
    time_index=TIME_INDEX,
)

# final cleanup
del l6c, l6a_match, passed_global_indices_np
mpl.pyplot.close("all")
gc.collect()

print("\nDone.")
print("Outputs in:", OUTDIR.resolve())
print("Summary:", summary_path.resolve())
print("PASSED-only augmented L6A:", OUT_L6A_PASSED.resolve())
print("New vars in output: Bearing, Bearing_cluster, Temp_cluster, u_cluster, v_cluster, NSquared_cluster, MF_cluster, MF_slice_sum, MF_toprows")
```

```python
# %%
"""
ALL-SLICES matching + per-slice combined L6A recon + FULL-ORBIT stitched recon
+ slice Temperature + momentum flux for PASSED clusters only
+ sum MF per slice + stitch MF across orbit
+ output NetCDF contains original L6A global attrs + variables, but ONLY PASSED clusters
  are kept for cluster-dimension variables.

Momentum flux (per cluster pixel, masked to cluster area):
MF = 1/2 * g^2 / N^2 * (lambda_z / lambda_x) * (Amplitude / Temperature)^2 * (1/c^2)

Key rules:
- SliceNo already exists in L6A, keep it (but subset to passed clusters)
- Area, Amplitude, DominantWavelength, Angle already exist (subset to passed clusters)
- Also slice Temperature into Temp_cluster (passed clusters only)
- Add geodesic Bearing (0 = East, counter-clockwise) as Bearing + Bearing_cluster
- Only save passed clusters (output file cluster vars are subset)
- Add MF_cluster, MF_slice_sum, MF_toprows
"""

from __future__ import annotations

from pathlib import Path
from typing import Dict, Any, List, Tuple

import gc
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from netCDF4 import Dataset

# --------------------------
# User params
# --------------------------
L6C_PATH = Path("awe_l6c_q20_2024165T2120_03175_v01.nc")  # Radiance-based clusters
L6A_PATH = Path("awe_l6a_tmp_2024165T2120_03175_v01.nc")  # Temperature-based clusters

OUTDIR = Path("outputs_ALLSLICES_l6a_to_l6c_matching_ratio_STITCH_MF")
OUTDIR.mkdir(parents=True, exist_ok=True)
SAVE_DPI = 200

DX_KM = 2.0
DY_KM = 2.0

# ---- Synthetic u,v,N^2 settings ----
TEMP_VAR_NAME = "Temperature"  # variable name in L6A
TIME_INDEX = 0                 # which t index to use when slicing
N2_VALUE = 4e-4                # N=0.02 -> N^2=0.0004
U_VALUE = 0.0
V_VALUE = 0.0

# ---- Momentum flux constants ----
G_MS2 = 9.52
LAMBDA_Z_KM = 10.0
C_CANCEL = 1.0

# Output NetCDF
OUT_L6A_PASSED = OUTDIR / f"{L6A_PATH.stem}_PASSED_with_u_v_NS2_Bearing_TempCluster_MF.nc"

# --------------------------
# Matching thresholds (tune)
# --------------------------
MIN_IOU = 0.05
MIN_INTERSECTION_PIX = 50
MAX_LAMBDA_MED_RATIO = 1.25
MAX_ANGLE_DIFF_DEG = 15.0

# Save best match selection even if FAIL (does NOT affect passing)
ASSIGN_BEST_EVEN_IF_FAILS = True

# --------------------------
# Plot saving switches
# --------------------------
SAVE_PAIR_PLOTS = False          # side-by-side A vs C match PNGs per slice
SAVE_PER_CLUSTER_PANELS = False  # big 5x2 panels for every passed cluster
SAVE_STITCHED_PLOTS = True       # MF_stitched.png + MF_stitched_components.png

# --------------------------
# Helpers
# --------------------------
def save_fig(fig: plt.Figure, outpath: Path) -> None:
    outpath.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(outpath, dpi=SAVE_DPI, bbox_inches="tight")
    plt.close(fig)


def robust_limits(A: np.ndarray, p_lo: float = 2, p_hi: float = 98) -> Tuple[float, float]:
    a = np.asarray(A)
    good = np.isfinite(a)
    if not np.any(good):
        return (-1.0, 1.0)
    lo, hi = np.nanpercentile(a[good], [p_lo, p_hi])
    if (not np.isfinite(lo)) or (not np.isfinite(hi)) or lo == hi:
        lo = float(np.nanmin(a[good]))
        hi = float(np.nanmax(a[good]))
    return (float(lo), float(hi))


def nanpercentile_safe(a: np.ndarray, q: float) -> float:
    good = np.isfinite(a)
    if not np.any(good):
        return np.nan
    return float(np.nanpercentile(a[good], q))


def circ_diff_deg_180(a_deg: float, b_deg: float) -> float:
    if (not np.isfinite(a_deg)) or (not np.isfinite(b_deg)):
        return np.nan
    d = abs(a_deg - b_deg) % 180.0
    return min(d, 180.0 - d)


def circ_median_deg_180(theta_rad: np.ndarray, mask: np.ndarray) -> float:
    th = np.asarray(theta_rad, dtype=float)
    m = mask & np.isfinite(th)
    if not np.any(m):
        return np.nan
    ang = th[m]
    z = np.exp(1j * 2.0 * ang)
    zmean = np.mean(z)
    if (not np.isfinite(zmean.real)) or (not np.isfinite(zmean.imag)) or abs(zmean) == 0:
        return np.nan
    th_mean = 0.5 * np.angle(zmean)
    deg = (np.degrees(th_mean) + 180.0) % 180.0
    return float(deg)


# -------------------------
# Geodesic bearing: 0 = East, counter-clockwise (CCW)
# -------------------------
def initial_bearing_deg_north_cw(lat1_deg, lon1_deg, lat2_deg, lon2_deg):
    """
    Vectorized initial bearing (forward azimuth) from (lat1, lon1) to (lat2, lon2).
    Returns azimuth degrees in [0, 360) measured from North, clockwise.
    """
    lat1 = np.deg2rad(lat1_deg)
    lon1 = np.deg2rad(lon1_deg)
    lat2 = np.deg2rad(lat2_deg)
    lon2 = np.deg2rad(lon2_deg)

    dlon = lon2 - lon1
    x = np.sin(dlon) * np.cos(lat2)
    y = np.cos(lat1) * np.sin(lat2) - np.sin(lat1) * np.cos(lat2) * np.cos(dlon)

    az = np.arctan2(x, y)
    az_deg = (np.rad2deg(az) + 360.0) % 360.0
    return az_deg


def bearing_deg_east_ccw_from_points(lat1_deg, lon1_deg, lat2_deg, lon2_deg):
    """
    Convert azimuth (North=0, CW) into your convention:
    East=0 and increasing counter-clockwise (CCW).
    East-CCW angle = (90 - az_north_cw) mod 360
    """
    az_north_cw = initial_bearing_deg_north_cw(lat1_deg, lon1_deg, lat2_deg, lon2_deg)
    return (90.0 - az_north_cw) % 360.0


def compute_bearing_to_x_neighbor_east_ccw(lat2d: np.ndarray, lon2d: np.ndarray) -> np.ndarray:
    """
    Bearing at each pixel to its +x neighbor (last column uses -x neighbor),
    in degrees: 0=East, CCW positive.
    """
    lat = np.asarray(lat2d, dtype=float)
    lon = np.asarray(lon2d, dtype=float)
    ny, nx = lat.shape

    lat2 = np.empty_like(lat)
    lon2 = np.empty_like(lon)

    lat2[:, :-1] = lat[:, 1:]
    lon2[:, :-1] = lon[:, 1:]
    lat2[:, -1] = lat[:, -2]
    lon2[:, -1] = lon[:, -2]

    bearing = bearing_deg_east_ccw_from_points(lat, lon, lat2, lon2)

    good = np.isfinite(lat) & np.isfinite(lon) & np.isfinite(lat2) & np.isfinite(lon2)
    return np.where(good, bearing, np.nan)


# --------------------------
# IO: load only variables needed for matching (no ClusterReconstruction here)
# --------------------------
def load_l6_for_matching(path: Path) -> Dict[str, Any]:
    with Dataset(path, "r") as nc:
        return {
            "path": str(path),
            "SlicesNo": np.asarray(nc.variables["SlicesNo"][:], dtype=int),
            "Amplitude": np.asarray(nc.variables["Amplitude"][:], dtype=float),
            "DominantWavelength": np.asarray(nc.variables["DominantWavelength"][:], dtype=float),
            "Angle": np.asarray(nc.variables["Angle"][:], dtype=float),
        }


def get_cluster_maps(data: Dict[str, Any], idx: int) -> Dict[str, np.ndarray]:
    """
    Lazy load ClusterReconstruction only for the requested idx.
    """
    A = np.asarray(data["Amplitude"][idx], dtype=float)
    L = np.asarray(data["DominantWavelength"][idx], dtype=float)
    T = np.asarray(data["Angle"][idx], dtype=float)

    R = np.full_like(A, np.nan)
    with Dataset(data["path"], "r") as nc:
        if "ClusterReconstruction" in nc.variables:
            R = np.asarray(nc.variables["ClusterReconstruction"][idx], dtype=float)

    return dict(R=R, A=A, L=L, T=T)


def compute_match_metrics(
    A_src: np.ndarray, L_src: np.ndarray, T_src: np.ndarray,
    A_cand: np.ndarray, L_cand: np.ndarray, T_cand: np.ndarray,
) -> Dict[str, Any]:
    mask_src = np.isfinite(A_src)
    mask_cand = np.isfinite(A_cand)
    inter = mask_src & mask_cand
    union = mask_src | mask_cand

    n_inter = int(np.count_nonzero(inter))
    n_union = int(np.count_nonzero(union))
    iou = (n_inter / n_union) if n_union > 0 else 0.0

    lam_src = L_src[inter]
    lam_cand = L_cand[inter]
    lam_src_med = nanpercentile_safe(lam_src, 50)
    lam_cand_med = nanpercentile_safe(lam_cand, 50)

    if np.isfinite(lam_src_med) and np.isfinite(lam_cand_med) and lam_src_med > 0 and lam_cand_med > 0:
        lam_ratio = max(lam_src_med, lam_cand_med) / min(lam_src_med, lam_cand_med)
    else:
        lam_ratio = np.nan

    ang_src_deg = circ_median_deg_180(T_src, inter)
    ang_cand_deg = circ_median_deg_180(T_cand, inter)
    ang_diff_deg = circ_diff_deg_180(ang_src_deg, ang_cand_deg)

    return dict(
        n_intersection=n_inter,
        iou=float(iou),
        lam_src_med=float(lam_src_med),
        lam_cand_med=float(lam_cand_med),
        lam_ratio=float(lam_ratio),
        ang_src_deg=float(ang_src_deg),
        ang_cand_deg=float(ang_cand_deg),
        ang_diff_deg=float(ang_diff_deg),
    )


def passes_thresholds(m: Dict[str, Any]) -> bool:
    if m["iou"] < MIN_IOU:
        return False
    if m["n_intersection"] < MIN_INTERSECTION_PIX:
        return False
    if (not np.isfinite(m["lam_ratio"])) or (m["lam_ratio"] > MAX_LAMBDA_MED_RATIO):
        return False
    if (not np.isfinite(m["ang_diff_deg"])) or (m["ang_diff_deg"] > MAX_ANGLE_DIFF_DEG):
        return False
    return True


def score(m: Dict[str, Any]) -> float:
    ang = m["ang_diff_deg"]
    if not np.isfinite(ang):
        ang = 999.0
    r = m["lam_ratio"]
    if (not np.isfinite(r)) or r <= 0:
        return 0.0
    return float(m["iou"]) * (1.0 / float(r)) / (1.0 + float(ang))


def plot_side_by_side_4rows(
    left: Dict[str, np.ndarray],
    right: Dict[str, np.ndarray],
    title_left: str,
    title_right: str,
    outpath: Path,
    dx_km: float,
    dy_km: float,
) -> None:
    ny, nx = left["A"].shape
    x1d_km = np.arange(nx, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    Rl_vmin, Rl_vmax = robust_limits(left["R"])
    Rr_vmin, Rr_vmax = robust_limits(right["R"])
    Al_vmin, Al_vmax = robust_limits(left["A"])
    Ar_vmin, Ar_vmax = robust_limits(right["A"])

    L_vmin, L_vmax = robust_limits(np.concatenate([left["L"].ravel(), right["L"].ravel()]), 5, 95)
    T_vmin, T_vmax = robust_limits(np.concatenate([left["T"].ravel(), right["T"].ravel()]))

    fig, axs = plt.subplots(4, 2, figsize=(16, 14), constrained_layout=True, sharex=True, sharey=True)

    def _panel(ax, Z, vmin, vmax, label, cmap=None):
        im = ax.pcolormesh(x1d_km, y1d_km, Z, shading="auto", vmin=vmin, vmax=vmax, cmap=cmap)
        cb = fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01)
        cb.set_label(label)
        ax.set_aspect("equal", adjustable="box")

    axs[0, 0].set_title(title_left)
    axs[0, 1].set_title(title_right)

    _panel(axs[0, 0], left["R"], Rl_vmin, Rl_vmax, "arb", cmap="RdBu_r")
    _panel(axs[0, 1], right["R"], Rr_vmin, Rr_vmax, "arb", cmap="RdBu_r")
    axs[0, 0].set_ylabel("Recon\ny (km)")

    _panel(axs[1, 0], left["A"], Al_vmin, Al_vmax, "K")
    _panel(axs[1, 1], right["A"], Ar_vmin, Ar_vmax, "kR")
    axs[1, 0].set_ylabel("Amp\ny (km)")

    _panel(axs[2, 0], left["L"], L_vmin, L_vmax, "km")
    _panel(axs[2, 1], right["L"], L_vmin, L_vmax, "km")
    axs[2, 0].set_ylabel("Lambda\ny (km)")

    _panel(axs[3, 0], left["T"], T_vmin, T_vmax, "rad")
    _panel(axs[3, 1], right["T"], T_vmin, T_vmax, "rad")
    axs[3, 0].set_ylabel("Angle\ny (km)")

    for r in range(4):
        for c in range(2):
            axs[r, c].set_xlabel("x (km)")

    save_fig(fig, outpath)


# --------------------------
# Flatten swath BEFORE slicing
# --------------------------
def flatten_swath(Z: np.ndarray, target_ny: int | None = None) -> np.ndarray:
    ny, nx = Z.shape
    out = np.full_like(Z, np.nan)

    bottoms = np.full(nx, np.nan, dtype=float)
    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size:
            bottoms[j] = valid_idx.max()

    valid_bottoms = bottoms[np.isfinite(bottoms)]
    if valid_bottoms.size == 0:
        med = float(np.nanmedian(Z)) if np.isfinite(np.nanmedian(Z)) else 0.0
        out = np.nan_to_num(Z, nan=med, posinf=med, neginf=med)
        if target_ny is not None and ny > target_ny:
            out = out[-target_ny:, :]
        return out

    max_bottom = int(valid_bottoms.max())

    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size == 0:
            continue
        bottom = int(valid_idx.max())
        shift = max_bottom - bottom
        if shift == 0:
            out[:, j] = col
        elif shift > 0:
            out[shift:, j] = col[: ny - shift]
        else:
            shift_up = -shift
            out[: ny - shift_up, j] = col[shift_up:]

    med = float(np.nanmedian(out)) if np.isfinite(np.nanmedian(out)) else 0.0
    out = np.nan_to_num(out, nan=med, posinf=med, neginf=med)

    if target_ny is not None and ny > target_ny:
        start = max_bottom + 1 - target_ny
        start = max(start, 0)
        end = start + target_ny
        out = out[start:end, :]

    return out


# --------------------------
# L6A cloning with PASSED-only cluster variables + new vars (u,v,N^2,Bearing, MF)
# --------------------------
def create_passed_only_l6a_with_new_vars(
    src_l6a: Path,
    dst_l6a: Path,
    passed_cluster_indices: np.ndarray,
    time_index: int,
) -> Tuple[str, str, str]:
    passed_cluster_indices = np.asarray(passed_cluster_indices, dtype=int)
    if passed_cluster_indices.size == 0:
        raise ValueError("No passed clusters. Output would be empty.")

    dst_l6a.parent.mkdir(parents=True, exist_ok=True)

    with Dataset(src_l6a, "r") as src:
        amp_var = src.variables["Amplitude"]
        cluster_dim, ytile_dim, xtile_dim = amp_var.dimensions[:3]
        _, y_tile, x_tile = amp_var.shape

        temp_var = src.variables[TEMP_VAR_NAME]
        temp_dims = temp_var.dimensions
        temp_shape = temp_var.shape
        temp_raw_2d = np.asarray(temp_var[time_index], dtype=float)

        u_raw_2d = (
            np.asarray(src.variables["u"][time_index], dtype=float)
            if "u" in src.variables
            else np.full_like(temp_raw_2d, U_VALUE, dtype=float)
        )
        v_raw_2d = (
            np.asarray(src.variables["v"][time_index], dtype=float)
            if "v" in src.variables
            else np.full_like(temp_raw_2d, V_VALUE, dtype=float)
        )
        n2_raw_2d = (
            np.asarray(src.variables["NSquared"][time_index], dtype=float)
            if "NSquared" in src.variables
            else np.full_like(temp_raw_2d, N2_VALUE, dtype=float)
        )

        if ("Latitude" in src.variables) and ("Longitude" in src.variables):
            lat_raw_2d = np.asarray(src.variables["Latitude"][time_index], dtype=float)
            lon_raw_2d = np.asarray(src.variables["Longitude"][time_index], dtype=float)
            bearing_raw_2d = compute_bearing_to_x_neighbor_east_ccw(lat_raw_2d, lon_raw_2d)
        else:
            bearing_raw_2d = np.full_like(temp_raw_2d, np.nan, dtype=float)

        temp_full_2d = flatten_swath(temp_raw_2d, target_ny=y_tile)
        u_full_2d = flatten_swath(u_raw_2d, target_ny=y_tile)
        v_full_2d = flatten_swath(v_raw_2d, target_ny=y_tile)
        n2_full_2d = flatten_swath(n2_raw_2d, target_ny=y_tile)
        bearing_full_2d = flatten_swath(bearing_raw_2d, target_ny=y_tile)

        n_passed = int(passed_cluster_indices.size)

        with Dataset(dst_l6a, "w") as dst:
            for aname in src.ncattrs():
                dst.setncattr(aname, src.getncattr(aname))

            for dname, dim in src.dimensions.items():
                if dname == cluster_dim:
                    dst.createDimension(dname, n_passed)
                else:
                    dst.createDimension(dname, (None if dim.isunlimited() else len(dim)))

            for vname, vsrc in src.variables.items():
                fill_value = vsrc.getncattr("_FillValue") if "_FillValue" in vsrc.ncattrs() else None
                vdst = dst.createVariable(
                    vname,
                    vsrc.dtype,
                    vsrc.dimensions,
                    zlib=True,
                    complevel=4,
                    fill_value=fill_value,
                )
                for aname in vsrc.ncattrs():
                    if aname != "_FillValue":
                        vdst.setncattr(aname, vsrc.getncattr(aname))

                if cluster_dim in vsrc.dimensions:
                    ax = vsrc.dimensions.index(cluster_dim)
                    data_sub = np.take(vsrc[:], passed_cluster_indices, axis=ax)
                    vdst[:] = data_sub
                else:
                    vdst[:] = vsrc[:]

            bearing_full = np.zeros(temp_shape, dtype=np.float32)
            bearing_full[time_index, 0:y_tile, :] = bearing_full_2d.astype(np.float32)

            if "Bearing" in dst.variables:
                dst.variables["Bearing"][:] = bearing_full
            else:
                vB = dst.createVariable("Bearing", "f4", temp_dims, zlib=True, complevel=4)
                vB.setncattr("long_name", "Geodesic bearing to +x neighbor (0=East, CCW)")
                vB.setncattr("units", "degree")
                vB[:] = bearing_full

            temp_cluster = np.full((n_passed, y_tile, x_tile), np.nan, dtype=np.float32)
            u_cluster = np.full_like(temp_cluster, np.nan)
            v_cluster = np.full_like(temp_cluster, np.nan)
            n2_cluster = np.full_like(temp_cluster, np.nan)
            bearing_cluster = np.full_like(temp_cluster, np.nan)

            mf_cluster = np.full_like(temp_cluster, np.nan)
            mf_zonal_cluster = np.full_like(temp_cluster, np.nan)
            mf_meridional_cluster = np.full_like(temp_cluster, np.nan)

            amp_passed = np.asarray(dst.variables["Amplitude"][:], dtype=float)
            lam_passed = np.asarray(dst.variables["DominantWavelength"][:], dtype=float)
            slice_passed = np.asarray(dst.variables["SlicesNo"][:], dtype=int)
            ang_rad_passed = np.asarray(dst.variables["Angle"][:], dtype=float)

            for j in range(n_passed):
                s = int(slice_passed[j])
                x0 = s * x_tile
                x1 = x0 + x_tile

                mask = np.isfinite(amp_passed[j])

                Ttile = temp_full_2d[:, x0:x1]
                Utile = u_full_2d[:, x0:x1]
                Vtile = v_full_2d[:, x0:x1]
                N2tile = n2_full_2d[:, x0:x1]
                Btile = bearing_full_2d[:, x0:x1]

                temp_cluster[j][mask] = Ttile[mask]
                u_cluster[j][mask] = Utile[mask]
                v_cluster[j][mask] = Vtile[mask]
                n2_cluster[j][mask] = N2tile[mask]
                bearing_cluster[j][mask] = Btile[mask]

                A = amp_passed[j]
                Lx = lam_passed[j]
                ang_deg_tile = (np.rad2deg(ang_rad_passed[j]) % 360.0)

                good = mask
                good = good & np.isfinite(Ttile) & np.isfinite(N2tile) & np.isfinite(Lx) & np.isfinite(Btile)
                good = good & (Ttile != 0.0) & (N2tile > 0.0) & (Lx > 0.0)

                if np.any(good):
                    mf_val = (
                        0.5
                        * (G_MS2**2)
                        / N2tile[good]
                        * (LAMBDA_Z_KM / Lx[good])
                        * (A[good] / Ttile[good]) ** 2
                        * (1.0 / (C_CANCEL**2))
                    )
                    mf_cluster[j][good] = mf_val

                    phi_deg = (Btile[good] + ang_deg_tile[good]) % 360.0
                    phi_rad = np.deg2rad(phi_deg)

                    mf_zonal_cluster[j][good] = mf_val * (np.cos(phi_rad) ** 2)
                    mf_meridional_cluster[j][good] = mf_val * (np.sin(phi_rad) ** 2)

            def _write(name: str, arr: np.ndarray, dims: Tuple[str, ...], long_name: str, units: str):
                if name in dst.variables:
                    dst.variables[name][:] = arr
                else:
                    v = dst.createVariable(name, "f4", dims)
                    v.setncattr("long_name", long_name)
                    v.setncattr("units", units)
                    v[:] = arr

            _write("Temp_cluster", temp_cluster, (cluster_dim, ytile_dim, xtile_dim), "Temperature, cluster masked", "K")
            _write("u_cluster", u_cluster, (cluster_dim, ytile_dim, xtile_dim), "u, cluster masked", "m s-1")
            _write("v_cluster", v_cluster, (cluster_dim, ytile_dim, xtile_dim), "v, cluster masked", "m s-1")
            _write("NSquared_cluster", n2_cluster, (cluster_dim, ytile_dim, xtile_dim), "N^2, cluster masked", "s-2")
            _write("Bearing_cluster", bearing_cluster, (cluster_dim, ytile_dim, xtile_dim), "Bearing (0=East, CCW), cluster masked", "degree")
            _write("MF_cluster", mf_cluster, (cluster_dim, ytile_dim, xtile_dim), "MF proxy per pixel, cluster masked", "m2 s-2")
            _write("MF_zonal_cluster", mf_zonal_cluster, (cluster_dim, ytile_dim, xtile_dim), "MF zonal share, cos^2(phi)", "m2 s-2")
            _write("MF_meridional_cluster", mf_meridional_cluster, (cluster_dim, ytile_dim, xtile_dim), "MF meridional share, sin^2(phi)", "m2 s-2")

            smin, smax = int(np.min(slice_passed)), int(np.max(slice_passed))
            full_slices = np.arange(smin, smax + 1)

            if "slice" not in dst.dimensions:
                dst.createDimension("slice", len(full_slices))
            if "SliceNo_passed_slices" not in dst.variables:
                dst.createVariable("SliceNo_passed_slices", "i4", ("slice",))[:] = full_slices
            else:
                dst.variables["SliceNo_passed_slices"][:] = full_slices

            mf_slice_sum = np.zeros((len(full_slices), y_tile, x_tile), dtype=np.float32)
            mfz_slice_sum = np.zeros_like(mf_slice_sum)
            mfm_slice_sum = np.zeros_like(mf_slice_sum)

            for k, s in enumerate(full_slices):
                idx = np.where(slice_passed == s)[0]
                if idx.size > 0:
                    mf_slice_sum[k] = np.nansum(mf_cluster[idx], axis=0)
                    mfz_slice_sum[k] = np.nansum(mf_zonal_cluster[idx], axis=0)
                    mfm_slice_sum[k] = np.nansum(mf_meridional_cluster[idx], axis=0)

            _write("MF_slice_sum", mf_slice_sum, ("slice", ytile_dim, xtile_dim), "Per slice MF sum (passed clusters), tile grid", "m2 s-2")
            _write("MF_zonal_slice_sum", mfz_slice_sum, ("slice", ytile_dim, xtile_dim), "Per slice zonal MF sum (passed clusters)", "m2 s-2")
            _write("MF_meridional_slice_sum", mfm_slice_sum, ("slice", ytile_dim, xtile_dim), "Per slice meridional MF sum (passed clusters)", "m2 s-2")

            MF_toprows = np.zeros(temp_shape, dtype=np.float32)
            MFZ_toprows = np.zeros(temp_shape, dtype=np.float32)
            MFM_toprows = np.zeros(temp_shape, dtype=np.float32)

            for k, s in enumerate(full_slices):
                x0 = s * x_tile
                x1 = x0 + x_tile
                MF_toprows[time_index, 0:y_tile, x0:x1] = mf_slice_sum[k]
                MFZ_toprows[time_index, 0:y_tile, x0:x1] = mfz_slice_sum[k]
                MFM_toprows[time_index, 0:y_tile, x0:x1] = mfm_slice_sum[k]

            _write("MF_toprows", MF_toprows, temp_dims, "Stitched MF sum (passed clusters) on top tile rows only", "m2 s-2")
            _write("MF_zonal_toprows", MFZ_toprows, temp_dims, "Stitched zonal MF share on top tile rows only", "m2 s-2")
            _write("MF_meridional_toprows", MFM_toprows, temp_dims, "Stitched meridional MF share on top tile rows only", "m2 s-2")

        del temp_raw_2d, u_raw_2d, v_raw_2d, n2_raw_2d, bearing_raw_2d
        del temp_full_2d, u_full_2d, v_full_2d, n2_full_2d, bearing_full_2d
        del temp_cluster, u_cluster, v_cluster, n2_cluster, bearing_cluster
        del mf_cluster, mf_zonal_cluster, mf_meridional_cluster
        gc.collect()

    return cluster_dim, ytile_dim, xtile_dim


def plot_passed_clusters_panels(nc_path: Path, outdir: Path) -> None:
    """
    Memory safe: read one cluster at a time.
    """
    outdir.mkdir(parents=True, exist_ok=True)

    with Dataset(nc_path, "r") as nc:
        cluster_dim = nc.variables["Amplitude"].dimensions[0]
        ytile_dim = nc.variables["Amplitude"].dimensions[1]
        xtile_dim = nc.variables["Amplitude"].dimensions[2]
        n_cluster = nc.dimensions[cluster_dim].size

        slc = np.asarray(nc.variables["SlicesNo"][:], dtype=int)

        ny = nc.dimensions[ytile_dim].size
        nx = nc.dimensions[xtile_dim].size
        x1d_km = np.arange(nx, dtype=float) * DX_KM
        y1d_km = np.arange(ny, dtype=float) * DY_KM

        ang_vmin, ang_vmax = (0.0, 360.0)
        brg_vmin, brg_vmax = (0.0, 360.0)
        n2_center = float(N2_VALUE)
        n2_vmin, n2_vmax = (0.5 * n2_center, 1.5 * n2_center)

        for i in range(n_cluster):
            s = int(slc[i])

            amp = np.asarray(nc.variables["Amplitude"][i], dtype=float)
            lam = np.asarray(nc.variables["DominantWavelength"][i], dtype=float)
            ang_rad = np.asarray(nc.variables["Angle"][i], dtype=float)
            ang_deg = (np.rad2deg(ang_rad) % 360.0)

            brg = np.asarray(nc.variables["Bearing_cluster"][i], dtype=float)
            n2 = np.asarray(nc.variables["NSquared_cluster"][i], dtype=float)
            tmp = np.asarray(nc.variables["Temp_cluster"][i], dtype=float)
            mf = np.asarray(nc.variables["MF_cluster"][i], dtype=float)
            mfz = np.asarray(nc.variables["MF_zonal_cluster"][i], dtype=float)
            mfm = np.asarray(nc.variables["MF_meridional_cluster"][i], dtype=float)

            if "ClusterReconstruction" in nc.variables:
                recon = np.asarray(nc.variables["ClusterReconstruction"][i], dtype=float)
            else:
                recon = np.full_like(amp, np.nan)

            r_vmin, r_vmax = robust_limits(recon, 2, 98)
            a_vmin, a_vmax = robust_limits(amp, 2, 98)
            l_vmin, l_vmax = robust_limits(lam, 5, 95)
            t_vmin, t_vmax = robust_limits(tmp, 2, 98)
            mf_vmin, mf_vmax = robust_limits(mf, 2, 98)
            mfz_vmin, mfz_vmax = robust_limits(mfz, 2, 98)
            mfm_vmin, mfm_vmax = robust_limits(mfm, 2, 98)

            fig, axs = plt.subplots(5, 2, figsize=(16, 22), constrained_layout=True, sharex=True, sharey=True)

            def _panel(ax, Z, vmin, vmax, label, cmap=None):
                im = ax.pcolormesh(x1d_km, y1d_km, Z, shading="auto", vmin=vmin, vmax=vmax, cmap=cmap)
                cb = fig.colorbar(im, ax=ax, fraction=0.04, pad=0.02)
                cb.set_label(label)
                ax.set_aspect("equal", adjustable="box")

            axs[0, 0].set_title(f"Passed cluster {i:04d}  SliceNo {s}")

            _panel(axs[0, 0], recon, r_vmin, r_vmax, "Recon", cmap="RdBu_r")
            _panel(axs[0, 1], amp, a_vmin, a_vmax, "Amplitude K")
            _panel(axs[1, 0], lam, l_vmin, l_vmax, "DominantWavelength km")
            _panel(axs[1, 1], tmp, t_vmin, t_vmax, "Temperature K")
            _panel(axs[2, 0], ang_deg, ang_vmin, ang_vmax, "Angle deg propagation")
            _panel(axs[2, 1], brg, brg_vmin, brg_vmax, "Bearing deg 0 East CCW")
            _panel(axs[3, 0], n2, n2_vmin, n2_vmax, "N2 s-2")
            _panel(axs[3, 1], mf, mf_vmin, mf_vmax, "MF m2 s-2")
            _panel(axs[4, 0], mfz, mfz_vmin, mfz_vmax, "MF zonal share cos2")
            _panel(axs[4, 1], mfm, mfm_vmin, mfm_vmax, "MF meridional share sin2")

            axs[4, 0].set_xlabel("x km")
            axs[4, 1].set_xlabel("x km")

            outpng = outdir / (
                f"passed_cluster_{i:04d}_slice_{s:03d}"
                "_Recon_Amp_Lam_Temp_Ang_Bearing_N2_MF_MFz_MFm.png"
            )
            save_fig(fig, outpng)

            del amp, lam, ang_rad, ang_deg, brg, n2, tmp, mf, mfz, mfm, recon, fig, axs
            gc.collect()

    mpl.pyplot.close("all")
    gc.collect()
    print("Per cluster plots saved to:", outdir.resolve())


def plot_stitched_mf_components(nc_path: Path, outpng: Path, time_index: int) -> None:
    outpng.parent.mkdir(parents=True, exist_ok=True)

    with Dataset(nc_path, "r") as nc:
        MF = np.asarray(nc.variables["MF_toprows"][time_index], dtype=float)
        MFZ = np.asarray(nc.variables["MF_zonal_toprows"][time_index], dtype=float)
        MFM = np.asarray(nc.variables["MF_meridional_toprows"][time_index], dtype=float)

    ny, nx = MF.shape
    x1d_km = np.arange(nx, dtype=float) * DX_KM
    y1d_km = np.arange(ny, dtype=float) * DY_KM

    stacked = np.concatenate([MF.ravel(), MFZ.ravel(), MFM.ravel()])
    vmin, vmax = robust_limits(stacked, 2, 99.9)

    fig, axs = plt.subplots(3, 1, figsize=(22, 8), constrained_layout=True, sharex=True, sharey=True)

    im0 = axs[0].pcolormesh(x1d_km, y1d_km, MF, shading="auto", vmin=vmin, vmax=vmax)
    axs[0].set_title("Stitched MF passed clusters summed")
    axs[0].set_ylabel("y km")
    axs[0].set_aspect("equal", adjustable="box")

    im1 = axs[1].pcolormesh(x1d_km, y1d_km, MFZ, shading="auto", vmin=vmin, vmax=vmax)
    axs[1].set_title("Stitched MF zonal share cos2")
    axs[1].set_ylabel("y km")
    axs[1].set_aspect("equal", adjustable="box")

    im2 = axs[2].pcolormesh(x1d_km, y1d_km, MFM, shading="auto", vmin=vmin, vmax=vmax)
    axs[2].set_title("Stitched MF meridional share sin2")
    axs[2].set_xlabel("x km")
    axs[2].set_ylabel("y km")
    axs[2].set_aspect("equal", adjustable="box")

    for ax in axs:
        ax.set_ylim(0, 600)

    cbar = fig.colorbar(im2, ax=axs, fraction=0.02, pad=0.01)
    cbar.set_label("MF m2 s-2")

    save_fig(fig, outpng)

    del MF, MFZ, MFM, stacked, fig, axs
    gc.collect()


# --------------------------
# Main: matching -> passed clusters -> output passed-only netCDF -> stitched plots
# --------------------------
l6c = load_l6_for_matching(L6C_PATH)
l6a_match = load_l6_for_matching(L6A_PATH)

all_slices_a = np.unique(l6a_match["SlicesNo"]).astype(int)
all_slices_a = all_slices_a[all_slices_a >= 0]
all_slices_a = np.sort(all_slices_a)

print("Slices in L6A:", all_slices_a.tolist())

summary_lines: List[str] = []
summary_lines.append(f"L6A: {L6A_PATH}\n")
summary_lines.append(f"L6C: {L6C_PATH}\n\n")
summary_lines.append("slice  nA  nC  nBestSaved  nPassed  note\n")

passed_global_indices: List[int] = []

for slice_no in all_slices_a:
    idx_a = np.where(l6a_match["SlicesNo"] == int(slice_no))[0]
    idx_c = np.where(l6c["SlicesNo"] == int(slice_no))[0]

    slice_dir = OUTDIR / f"slice_{slice_no:03d}"
    pairs_dir = slice_dir / "pairs"
    pairs_dir.mkdir(parents=True, exist_ok=True)

    if idx_a.size == 0:
        summary_lines.append(f"{slice_no:5d}  {0:2d}  {idx_c.size:2d}  {0:10d}  {0:7d}  no L6A clusters\n")
        continue
    if idx_c.size == 0:
        (slice_dir / "NO_L6C_CLUSTERS.txt").write_text(f"No L6C clusters for slice {slice_no}\n")
        summary_lines.append(f"{slice_no:5d}  {idx_a.size:2d}  {0:2d}  {0:10d}  {0:7d}  no L6C clusters\n")
        continue

    best_rows: List[Dict[str, Any]] = []

    cand_lines: List[str] = []
    cand_header = []
    cand_header.append(f"L6A file: {L6A_PATH}\n")
    cand_header.append(f"L6C file: {L6C_PATH}\n")
    cand_header.append(f"Slice: {slice_no}\n\n")
    cand_header.append(
        "cluster_a  cluster_c  passes  score    iou    inter  "
        "lam_ratio  ang_diff_deg  thetaA_deg  thetaC_deg  lamA_med  lamC_med\n"
    )

    for a_idx in idx_a:
        a_idx = int(a_idx)
        A_a = l6a_match["Amplitude"][a_idx]
        L_a = l6a_match["DominantWavelength"][a_idx]
        T_a = l6a_match["Angle"][a_idx]

        candidates: List[Dict[str, Any]] = []
        for c_idx in idx_c:
            c_idx = int(c_idx)
            A_c = l6c["Amplitude"][c_idx]
            L_c = l6c["DominantWavelength"][c_idx]
            T_c = l6c["Angle"][c_idx]

            m = compute_match_metrics(A_a, L_a, T_a, A_c, L_c, T_c)
            m["cluster_a"] = a_idx
            m["cluster_c"] = c_idx
            m["passes"] = bool(passes_thresholds(m))
            m["score"] = float(score(m))
            candidates.append(m)

            cand_lines.append(
                f"{m['cluster_a']:9d}  {m['cluster_c']:9d}  {int(m['passes']):6d}  {m['score']:6.3f}  "
                f"{m['iou']:.3f}  {m['n_intersection']:5d}  "
                f"{m['lam_ratio']:8.3f}  {m['ang_diff_deg']:12.1f}  "
                f"{m['ang_src_deg']:10.1f}  {m['ang_cand_deg']:10.1f}  "
                f"{m['lam_src_med']:8.1f}  {m['lam_cand_med']:8.1f}\n"
            )

        passed = [c for c in candidates if c["passes"]]
        if passed:
            best = max(passed, key=lambda d: d["score"])
        else:
            best = max(candidates, key=lambda d: d["score"])

        if best["passes"] or ASSIGN_BEST_EVEN_IF_FAILS:
            best_rows.append(best)

            if SAVE_PAIR_PLOTS:
                left_maps = get_cluster_maps(l6a_match, best["cluster_a"])
                right_maps = get_cluster_maps(l6c, best["cluster_c"])

                pass_tag = "PASS" if best["passes"] else "FAIL"
                outpng = pairs_dir / f"slice_{slice_no:03d}_A_{best['cluster_a']:04d}_to_C_{best['cluster_c']:04d}_{pass_tag}.png"

                title_left = (
                    f"L6A Temp | slice {slice_no} | idx {best['cluster_a']} | "
                    f"n={np.count_nonzero(np.isfinite(left_maps['A']))}"
                )
                title_right = (
                    f"L6C Rad | slice {slice_no} | idx {best['cluster_c']} | "
                    f"IoU={best['iou']:.3f} inter={best['n_intersection']} "
                    f"ratio={best['lam_ratio']:.3f} Δθ={best['ang_diff_deg']:.1f}°"
                )

                plot_side_by_side_4rows(
                    left=left_maps,
                    right=right_maps,
                    title_left=title_left,
                    title_right=title_right,
                    outpath=outpng,
                    dx_km=DX_KM,
                    dy_km=DY_KM,
                )

                del left_maps, right_maps
                gc.collect()

        del candidates
        gc.collect()

    cand_path = slice_dir / f"slice_{slice_no:03d}_ALL_candidates_ratio.txt"
    cand_path.write_text("".join(cand_header) + "".join(cand_lines))
    del cand_lines
    gc.collect()

    best_path = slice_dir / f"slice_{slice_no:03d}_BEST_matches_A_to_C_ratio.txt"
    best_lines = []
    best_lines.append(f"L6A file: {L6A_PATH}\n")
    best_lines.append(f"L6C file: {L6C_PATH}\n")
    best_lines.append(f"Slice: {slice_no}\n\n")
    best_lines.append(
        "cluster_a  best_cluster_c  passes  score    iou    inter  "
        "lam_ratio  ang_diff_deg  thetaA_deg  thetaC_deg  lamA_med  lamC_med\n"
    )
    for m in sorted(best_rows, key=lambda d: d["cluster_a"]):
        best_lines.append(
            f"{m['cluster_a']:9d}  {m['cluster_c']:14d}  {int(m['passes']):6d}  {m['score']:6.3f}  "
            f"{m['iou']:.3f}  {m['n_intersection']:5d}  "
            f"{m['lam_ratio']:8.3f}  {m['ang_diff_deg']:12.1f}  "
            f"{m['ang_src_deg']:10.1f}  {m['ang_cand_deg']:10.1f}  "
            f"{m['lam_src_med']:8.1f}  {m['lam_cand_med']:8.1f}\n"
        )
    best_path.write_text("".join(best_lines))

    passed_here = [int(m["cluster_a"]) for m in best_rows if bool(m["passes"])]
    passed_global_indices.extend(passed_here)

    summary_lines.append(
        f"{slice_no:5d}  {idx_a.size:2d}  {idx_c.size:2d}  {len(best_rows):10d}  {len(passed_here):7d}  ok\n"
    )

    del best_rows, best_lines
    gc.collect()

summary_path = OUTDIR / "SUMMARY_all_slices.txt"
summary_path.write_text("".join(summary_lines))

passed_global_indices = sorted(set(passed_global_indices))
passed_global_indices_np = np.asarray(passed_global_indices, dtype=int)
print("Total passed clusters:", passed_global_indices_np.size)

if passed_global_indices_np.size == 0:
    raise RuntimeError("No passed clusters found. Nothing to save or stitch.")

create_passed_only_l6a_with_new_vars(
    src_l6a=L6A_PATH,
    dst_l6a=OUT_L6A_PASSED,
    passed_cluster_indices=passed_global_indices_np,
    time_index=TIME_INDEX,
)

if SAVE_STITCHED_PLOTS:
    with Dataset(OUT_L6A_PASSED, "r") as nc:
        MF2 = np.asarray(nc.variables["MF_toprows"][TIME_INDEX], dtype=float)
        ny, nx = MF2.shape
        x1d_km = np.arange(nx, dtype=float) * DX_KM
        y1d_km = np.arange(ny, dtype=float) * DY_KM

        vmin, vmax = robust_limits(MF2, 2, 99.9)

        fig, ax = plt.subplots(figsize=(22, 3), constrained_layout=True)
        im = ax.pcolormesh(x1d_km, y1d_km, MF2, shading="auto", vmin=vmin, vmax=vmax)
        ax.set_title("Momentum Flux (passed clusters summed)")
        ax.set_xlabel("x (km)")
        ax.set_ylabel("y (km)")
        ax.set_ylim(0, 600)
        ax.set_aspect("equal", adjustable="box")
        fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01).set_label("MF (m2 s-2)")
        save_fig(fig, OUTDIR / "MF_stitched.png")

    plot_stitched_mf_components(
        nc_path=OUT_L6A_PASSED,
        outpng=OUTDIR / "MF_stitched_components.png",
        time_index=TIME_INDEX,
    )

if SAVE_PER_CLUSTER_PANELS:
    plot_passed_clusters_panels(
        nc_path=OUT_L6A_PASSED,
        outdir=OUTDIR / "passed_cluster_plots_panels",
    )

del l6c, l6a_match, passed_global_indices_np
mpl.pyplot.close("all")
gc.collect()

print("\nDone.")
print("Outputs in:", OUTDIR.resolve())
print("Summary:", summary_path.resolve())
print("PASSED-only augmented L6A:", OUT_L6A_PASSED.resolve())
print("New vars in output: Bearing, Bearing_cluster, Temp_cluster, u_cluster, v_cluster, NSquared_cluster, MF_cluster, MF_slice_sum, MF_toprows")
```

## Batch Calculating

```python
# %%
"""
BATCH: ALL-SLICES matching + per-slice combined L6A recon + FULL-ORBIT stitched recon
+ slice Temperature + momentum flux for PASSED clusters only
+ sum MF per slice + stitch MF across orbit
+ output NetCDF contains original L6A global attrs + variables, but ONLY PASSED clusters
  are kept for cluster-dimension variables.

Momentum flux (per cluster pixel, masked to cluster area):
MF = 1/2 * g^2 / N^2 * (lambda_z / lambda_x) * (Amplitude / Temperature)^2 * (1/c^2)

Key rules:
- SliceNo already exists in L6A, keep it (but subset to passed clusters)
- Area, Amplitude, DominantWavelength, Angle already exist (subset to passed clusters)
- Also slice Temperature into Temp_cluster (passed clusters only)
- Add geodesic Bearing (0 = East, counter-clockwise) as Bearing + Bearing_cluster
- Only save passed clusters (output file cluster vars are subset)
- Add MF_cluster, MF_slice_sum, MF_toprows

BATCH behavior:
- Scan L6_DIR for L6A and L6C files
- Pair by common suffix key like: 2024165T2120_03175_v01
- For each pair, write the new file into L7_DIR and rename l6a -> l7a
- Save plots + text into outputs folder per key
"""

from __future__ import annotations

from pathlib import Path
from typing import Dict, Any, List, Tuple
import traceback
import gc

import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from netCDF4 import Dataset

# ============================================================
# Batch params
# ============================================================
L6_DIR = Path("l6")  # contains many L6A/L6C files
L7_DIR = Path("l7")  # where new NetCDFs go
L7_DIR.mkdir(parents=True, exist_ok=True)

BATCH_OUTROOT = Path("outputs_BATCH_l6a_to_l6c_matching_ratio_STITCH_MF")
BATCH_OUTROOT.mkdir(parents=True, exist_ok=True)

SAVE_DPI = 200
DX_KM = 2.0
DY_KM = 2.0

# ---- Synthetic u,v,N^2 settings ----
TEMP_VAR_NAME = "Temperature"
TIME_INDEX = 0
N2_VALUE = 4e-4
U_VALUE = 0.0
V_VALUE = 0.0

# ---- Momentum flux constants ----
G_MS2 = 9.52
LAMBDA_Z_KM = 10.0
C_CANCEL = 1.0

# --------------------------
# Matching thresholds (tune)
# --------------------------
MIN_IOU = 0.05
MIN_INTERSECTION_PIX = 50
MAX_LAMBDA_MED_RATIO = 1.25
MAX_ANGLE_DIFF_DEG = 15.0

# Save best match even if FAIL (does NOT affect passing)
ASSIGN_BEST_EVEN_IF_FAILS = True

# --------------------------
# Plot saving switches (batch friendly)
# --------------------------
SAVE_PAIR_PLOTS = False          # side-by-side A vs C match PNGs per slice (heavy)
SAVE_PER_CLUSTER_PANELS = False  # big 5x2 panels per passed cluster (heavy)
SAVE_STITCHED_PLOTS = True       # MF_stitched.png + MF_stitched_components.png


# ============================================================
# Helpers
# ============================================================
def save_fig(fig: plt.Figure, outpath: Path) -> None:
    outpath.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(outpath, dpi=SAVE_DPI, bbox_inches="tight")
    plt.close(fig)


def robust_limits(A: np.ndarray, p_lo: float = 2, p_hi: float = 98) -> Tuple[float, float]:
    a = np.asarray(A)
    good = np.isfinite(a)
    if not np.any(good):
        return (-1.0, 1.0)
    lo, hi = np.nanpercentile(a[good], [p_lo, p_hi])
    if (not np.isfinite(lo)) or (not np.isfinite(hi)) or lo == hi:
        lo = float(np.nanmin(a[good]))
        hi = float(np.nanmax(a[good]))
    return (float(lo), float(hi))


def nanpercentile_safe(a: np.ndarray, q: float) -> float:
    good = np.isfinite(a)
    if not np.any(good):
        return np.nan
    return float(np.nanpercentile(a[good], q))


def circ_diff_deg_180(a_deg: float, b_deg: float) -> float:
    if (not np.isfinite(a_deg)) or (not np.isfinite(b_deg)):
        return np.nan
    d = abs(a_deg - b_deg) % 180.0
    return min(d, 180.0 - d)


def circ_median_deg_180(theta_rad: np.ndarray, mask: np.ndarray) -> float:
    th = np.asarray(theta_rad, dtype=float)
    m = mask & np.isfinite(th)
    if not np.any(m):
        return np.nan
    ang = th[m]
    z = np.exp(1j * 2.0 * ang)
    zmean = np.mean(z)
    if (not np.isfinite(zmean.real)) or (not np.isfinite(zmean.imag)) or abs(zmean) == 0:
        return np.nan
    th_mean = 0.5 * np.angle(zmean)
    deg = (np.degrees(th_mean) + 180.0) % 180.0
    return float(deg)


# -------------------------
# Geodesic bearing: 0 = East, counter-clockwise (CCW)
# -------------------------
def initial_bearing_deg_north_cw(lat1_deg, lon1_deg, lat2_deg, lon2_deg):
    lat1 = np.deg2rad(lat1_deg)
    lon1 = np.deg2rad(lon1_deg)
    lat2 = np.deg2rad(lat2_deg)
    lon2 = np.deg2rad(lon2_deg)

    dlon = lon2 - lon1
    x = np.sin(dlon) * np.cos(lat2)
    y = np.cos(lat1) * np.sin(lat2) - np.sin(lat1) * np.cos(lat2) * np.cos(dlon)

    az = np.arctan2(x, y)
    return (np.rad2deg(az) + 360.0) % 360.0


def bearing_deg_east_ccw_from_points(lat1_deg, lon1_deg, lat2_deg, lon2_deg):
    az_north_cw = initial_bearing_deg_north_cw(lat1_deg, lon1_deg, lat2_deg, lon2_deg)
    return (90.0 - az_north_cw) % 360.0


def compute_bearing_to_x_neighbor_east_ccw(lat2d: np.ndarray, lon2d: np.ndarray) -> np.ndarray:
    lat = np.asarray(lat2d, dtype=float)
    lon = np.asarray(lon2d, dtype=float)

    lat2 = np.empty_like(lat)
    lon2 = np.empty_like(lon)

    lat2[:, :-1] = lat[:, 1:]
    lon2[:, :-1] = lon[:, 1:]
    lat2[:, -1] = lat[:, -2]
    lon2[:, -1] = lon[:, -2]

    bearing = bearing_deg_east_ccw_from_points(lat, lon, lat2, lon2)
    good = np.isfinite(lat) & np.isfinite(lon) & np.isfinite(lat2) & np.isfinite(lon2)
    return np.where(good, bearing, np.nan)


# ============================================================
# Matching IO (memory friendly): do NOT load ClusterReconstruction upfront
# ============================================================
def load_l6_for_matching(path: Path) -> Dict[str, Any]:
    with Dataset(path, "r") as nc:
        return {
            "path": str(path),
            "SlicesNo": np.asarray(nc.variables["SlicesNo"][:], dtype=int),
            "Amplitude": np.asarray(nc.variables["Amplitude"][:], dtype=float),
            "DominantWavelength": np.asarray(nc.variables["DominantWavelength"][:], dtype=float),
            "Angle": np.asarray(nc.variables["Angle"][:], dtype=float),
        }


def get_cluster_maps(data: Dict[str, Any], idx: int) -> Dict[str, np.ndarray]:
    """
    Lazy load ClusterReconstruction only for requested idx (only used if SAVE_PAIR_PLOTS=True).
    """
    A = np.asarray(data["Amplitude"][idx], dtype=float)
    L = np.asarray(data["DominantWavelength"][idx], dtype=float)
    T = np.asarray(data["Angle"][idx], dtype=float)

    R = np.full_like(A, np.nan)
    with Dataset(data["path"], "r") as nc:
        if "ClusterReconstruction" in nc.variables:
            R = np.asarray(nc.variables["ClusterReconstruction"][idx], dtype=float)

    return dict(R=R, A=A, L=L, T=T)


def compute_match_metrics(
    A_src: np.ndarray, L_src: np.ndarray, T_src: np.ndarray,
    A_cand: np.ndarray, L_cand: np.ndarray, T_cand: np.ndarray,
) -> Dict[str, Any]:
    mask_src = np.isfinite(A_src)
    mask_cand = np.isfinite(A_cand)
    inter = mask_src & mask_cand
    union = mask_src | mask_cand

    n_inter = int(np.count_nonzero(inter))
    n_union = int(np.count_nonzero(union))
    iou = (n_inter / n_union) if n_union > 0 else 0.0

    lam_src = L_src[inter]
    lam_cand = L_cand[inter]
    lam_src_med = nanpercentile_safe(lam_src, 50)
    lam_cand_med = nanpercentile_safe(lam_cand, 50)

    if np.isfinite(lam_src_med) and np.isfinite(lam_cand_med) and lam_src_med > 0 and lam_cand_med > 0:
        lam_ratio = max(lam_src_med, lam_cand_med) / min(lam_src_med, lam_cand_med)
    else:
        lam_ratio = np.nan

    ang_src_deg = circ_median_deg_180(T_src, inter)
    ang_cand_deg = circ_median_deg_180(T_cand, inter)
    ang_diff_deg = circ_diff_deg_180(ang_src_deg, ang_cand_deg)

    return dict(
        n_intersection=n_inter,
        iou=float(iou),
        lam_src_med=float(lam_src_med),
        lam_cand_med=float(lam_cand_med),
        lam_ratio=float(lam_ratio),
        ang_src_deg=float(ang_src_deg),
        ang_cand_deg=float(ang_cand_deg),
        ang_diff_deg=float(ang_diff_deg),
    )


def passes_thresholds(m: Dict[str, Any]) -> bool:
    if m["iou"] < MIN_IOU:
        return False
    if m["n_intersection"] < MIN_INTERSECTION_PIX:
        return False
    if (not np.isfinite(m["lam_ratio"])) or (m["lam_ratio"] > MAX_LAMBDA_MED_RATIO):
        return False
    if (not np.isfinite(m["ang_diff_deg"])) or (m["ang_diff_deg"] > MAX_ANGLE_DIFF_DEG):
        return False
    return True


def score(m: Dict[str, Any]) -> float:
    ang = m["ang_diff_deg"]
    if not np.isfinite(ang):
        ang = 999.0
    r = m["lam_ratio"]
    if (not np.isfinite(r)) or r <= 0:
        return 0.0
    return float(m["iou"]) * (1.0 / float(r)) / (1.0 + float(ang))


def plot_side_by_side_4rows(
    left: Dict[str, np.ndarray],
    right: Dict[str, np.ndarray],
    title_left: str,
    title_right: str,
    outpath: Path,
    dx_km: float,
    dy_km: float,
) -> None:
    ny, nx = left["A"].shape
    x1d_km = np.arange(nx, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    Rl_vmin, Rl_vmax = robust_limits(left["R"])
    Rr_vmin, Rr_vmax = robust_limits(right["R"])
    Al_vmin, Al_vmax = robust_limits(left["A"])
    Ar_vmin, Ar_vmax = robust_limits(right["A"])

    L_vmin, L_vmax = robust_limits(np.concatenate([left["L"].ravel(), right["L"].ravel()]), 5, 95)
    T_vmin, T_vmax = robust_limits(np.concatenate([left["T"].ravel(), right["T"].ravel()]))

    fig, axs = plt.subplots(4, 2, figsize=(16, 14), constrained_layout=True, sharex=True, sharey=True)

    def _panel(ax, Z, vmin, vmax, label, cmap=None):
        im = ax.pcolormesh(x1d_km, y1d_km, Z, shading="auto", vmin=vmin, vmax=vmax, cmap=cmap)
        cb = fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01)
        cb.set_label(label)
        ax.set_aspect("equal", adjustable="box")

    axs[0, 0].set_title(title_left)
    axs[0, 1].set_title(title_right)

    _panel(axs[0, 0], left["R"], Rl_vmin, Rl_vmax, "arb", cmap="RdBu_r")
    _panel(axs[0, 1], right["R"], Rr_vmin, Rr_vmax, "arb", cmap="RdBu_r")
    axs[0, 0].set_ylabel("Recon\ny (km)")

    _panel(axs[1, 0], left["A"], Al_vmin, Al_vmax, "K")
    _panel(axs[1, 1], right["A"], Ar_vmin, Ar_vmax, "kR")
    axs[1, 0].set_ylabel("Amp\ny (km)")

    _panel(axs[2, 0], left["L"], L_vmin, L_vmax, "km")
    _panel(axs[2, 1], right["L"], L_vmin, L_vmax, "km")
    axs[2, 0].set_ylabel("Lambda\ny (km)")

    _panel(axs[3, 0], left["T"], T_vmin, T_vmax, "rad")
    _panel(axs[3, 1], right["T"], T_vmin, T_vmax, "rad")
    axs[3, 0].set_ylabel("Angle\ny (km)")

    for r in range(4):
        for c in range(2):
            axs[r, c].set_xlabel("x (km)")

    save_fig(fig, outpath)


# --------------------------
# Flatten swath BEFORE slicing
# --------------------------
def flatten_swath(Z: np.ndarray, target_ny: int | None = None) -> np.ndarray:
    ny, nx = Z.shape
    out = np.full_like(Z, np.nan)

    bottoms = np.full(nx, np.nan, dtype=float)
    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size:
            bottoms[j] = valid_idx.max()

    valid_bottoms = bottoms[np.isfinite(bottoms)]
    if valid_bottoms.size == 0:
        med = float(np.nanmedian(Z)) if np.isfinite(np.nanmedian(Z)) else 0.0
        out = np.nan_to_num(Z, nan=med, posinf=med, neginf=med)
        if target_ny is not None and ny > target_ny:
            out = out[-target_ny:, :]
        return out

    max_bottom = int(valid_bottoms.max())

    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size == 0:
            continue
        bottom = int(valid_idx.max())
        shift = max_bottom - bottom
        if shift == 0:
            out[:, j] = col
        elif shift > 0:
            out[shift:, j] = col[: ny - shift]
        else:
            shift_up = -shift
            out[: ny - shift_up, j] = col[shift_up:]

    med = float(np.nanmedian(out)) if np.isfinite(np.nanmedian(out)) else 0.0
    out = np.nan_to_num(out, nan=med, posinf=med, neginf=med)

    if target_ny is not None and ny > target_ny:
        start = max_bottom + 1 - target_ny
        start = max(start, 0)
        end = start + target_ny
        out = out[start:end, :]

    return out


# ============================================================
# L6A cloning with PASSED-only cluster variables + new vars
# ============================================================
def create_passed_only_l6a_with_new_vars(
    src_l6a: Path,
    dst_l7a: Path,
    passed_cluster_indices: np.ndarray,
    time_index: int,
) -> Tuple[str, str, str]:
    passed_cluster_indices = np.asarray(passed_cluster_indices, dtype=int)
    if passed_cluster_indices.size == 0:
        raise ValueError("No passed clusters. Output would be empty.")

    dst_l7a.parent.mkdir(parents=True, exist_ok=True)

    with Dataset(src_l6a, "r") as src:
        amp_var = src.variables["Amplitude"]
        cluster_dim, ytile_dim, xtile_dim = amp_var.dimensions[:3]
        _, y_tile, x_tile = amp_var.shape

        temp_var = src.variables[TEMP_VAR_NAME]
        temp_dims = temp_var.dimensions
        temp_shape = temp_var.shape
        temp_raw_2d = np.asarray(temp_var[time_index], dtype=float)

        u_raw_2d = (
            np.asarray(src.variables["u"][time_index], dtype=float)
            if "u" in src.variables
            else np.full_like(temp_raw_2d, U_VALUE, dtype=float)
        )
        v_raw_2d = (
            np.asarray(src.variables["v"][time_index], dtype=float)
            if "v" in src.variables
            else np.full_like(temp_raw_2d, V_VALUE, dtype=float)
        )
        n2_raw_2d = (
            np.asarray(src.variables["NSquared"][time_index], dtype=float)
            if "NSquared" in src.variables
            else np.full_like(temp_raw_2d, N2_VALUE, dtype=float)
        )

        if ("Latitude" in src.variables) and ("Longitude" in src.variables):
            lat_raw_2d = np.asarray(src.variables["Latitude"][time_index], dtype=float)
            lon_raw_2d = np.asarray(src.variables["Longitude"][time_index], dtype=float)
            bearing_raw_2d = compute_bearing_to_x_neighbor_east_ccw(lat_raw_2d, lon_raw_2d)
        else:
            bearing_raw_2d = np.full_like(temp_raw_2d, np.nan, dtype=float)

        temp_full_2d = flatten_swath(temp_raw_2d, target_ny=y_tile)
        u_full_2d = flatten_swath(u_raw_2d, target_ny=y_tile)
        v_full_2d = flatten_swath(v_raw_2d, target_ny=y_tile)
        n2_full_2d = flatten_swath(n2_raw_2d, target_ny=y_tile)
        bearing_full_2d = flatten_swath(bearing_raw_2d, target_ny=y_tile)

        n_passed = int(passed_cluster_indices.size)

        with Dataset(dst_l7a, "w") as dst:
            for aname in src.ncattrs():
                dst.setncattr(aname, src.getncattr(aname))

            for dname, dim in src.dimensions.items():
                if dname == cluster_dim:
                    dst.createDimension(dname, n_passed)
                else:
                    dst.createDimension(dname, (None if dim.isunlimited() else len(dim)))

            for vname, vsrc in src.variables.items():
                fill_value = vsrc.getncattr("_FillValue") if "_FillValue" in vsrc.ncattrs() else None
                vdst = dst.createVariable(
                    vname,
                    vsrc.dtype,
                    vsrc.dimensions,
                    zlib=True,
                    complevel=4,
                    fill_value=fill_value,
                )
                for aname in vsrc.ncattrs():
                    if aname != "_FillValue":
                        vdst.setncattr(aname, vsrc.getncattr(aname))

                if cluster_dim in vsrc.dimensions:
                    ax = vsrc.dimensions.index(cluster_dim)
                    vdst[:] = np.take(vsrc[:], passed_cluster_indices, axis=ax)
                else:
                    vdst[:] = vsrc[:]

            bearing_full = np.zeros(temp_shape, dtype=np.float32)
            bearing_full[time_index, 0:y_tile, :] = bearing_full_2d.astype(np.float32)

            if "Bearing" in dst.variables:
                dst.variables["Bearing"][:] = bearing_full
            else:
                vB = dst.createVariable("Bearing", "f4", temp_dims, zlib=True, complevel=4)
                vB.setncattr("long_name", "Geodesic bearing to +x neighbor (0=East, CCW)")
                vB.setncattr("units", "degree")
                vB[:] = bearing_full

            temp_cluster = np.full((n_passed, y_tile, x_tile), np.nan, dtype=np.float32)
            u_cluster = np.full_like(temp_cluster, np.nan)
            v_cluster = np.full_like(temp_cluster, np.nan)
            n2_cluster = np.full_like(temp_cluster, np.nan)
            bearing_cluster = np.full_like(temp_cluster, np.nan)

            mf_cluster = np.full_like(temp_cluster, np.nan)
            mf_zonal_cluster = np.full_like(temp_cluster, np.nan)
            mf_meridional_cluster = np.full_like(temp_cluster, np.nan)

            amp_passed = np.asarray(dst.variables["Amplitude"][:], dtype=float)
            lam_passed = np.asarray(dst.variables["DominantWavelength"][:], dtype=float)
            slice_passed = np.asarray(dst.variables["SlicesNo"][:], dtype=int)
            ang_rad_passed = np.asarray(dst.variables["Angle"][:], dtype=float)

            for j in range(n_passed):
                s = int(slice_passed[j])
                x0 = s * x_tile
                x1 = x0 + x_tile

                mask = np.isfinite(amp_passed[j])

                Ttile = temp_full_2d[:, x0:x1]
                Utile = u_full_2d[:, x0:x1]
                Vtile = v_full_2d[:, x0:x1]
                N2tile = n2_full_2d[:, x0:x1]
                Btile = bearing_full_2d[:, x0:x1]

                temp_cluster[j][mask] = Ttile[mask]
                u_cluster[j][mask] = Utile[mask]
                v_cluster[j][mask] = Vtile[mask]
                n2_cluster[j][mask] = N2tile[mask]
                bearing_cluster[j][mask] = Btile[mask]

                A = amp_passed[j]
                Lx = lam_passed[j]
                ang_deg_tile = (np.rad2deg(ang_rad_passed[j]) % 360.0)

                good = mask
                good = good & np.isfinite(Ttile) & np.isfinite(N2tile) & np.isfinite(Lx) & np.isfinite(Btile)
                good = good & (Ttile != 0.0) & (N2tile > 0.0) & (Lx > 0.0)

                if np.any(good):
                    mf_val = (
                        0.5
                        * (G_MS2**2)
                        / N2tile[good]
                        * (LAMBDA_Z_KM / Lx[good])
                        * (A[good] / Ttile[good]) ** 2
                        * (1.0 / (C_CANCEL**2))
                    )
                    mf_cluster[j][good] = mf_val

                    phi_deg = (Btile[good] + ang_deg_tile[good]) % 360.0
                    phi_rad = np.deg2rad(phi_deg)

                    mf_zonal_cluster[j][good] = mf_val * (np.cos(phi_rad) ** 2)
                    mf_meridional_cluster[j][good] = mf_val * (np.sin(phi_rad) ** 2)

            def _write(name: str, arr: np.ndarray, dims: Tuple[str, ...], long_name: str, units: str):
                if name in dst.variables:
                    dst.variables[name][:] = arr
                else:
                    v = dst.createVariable(name, "f4", dims, zlib=True, complevel=4)
                    v.setncattr("long_name", long_name)
                    v.setncattr("units", units)
                    v[:] = arr

            _write("Temp_cluster", temp_cluster, (cluster_dim, ytile_dim, xtile_dim), "Temperature, cluster masked", "K")
            _write("u_cluster", u_cluster, (cluster_dim, ytile_dim, xtile_dim), "u wind, cluster masked", "m/s")
            _write("v_cluster", v_cluster, (cluster_dim, ytile_dim, xtile_dim), "v wind, cluster masked", "m/s")
            _write("NSquared_cluster", n2_cluster, (cluster_dim, ytile_dim, xtile_dim), "N^2, cluster masked", "s^-2")
            _write("Bearing_cluster", bearing_cluster, (cluster_dim, ytile_dim, xtile_dim), "Bearing (0=East, CCW), cluster masked", "degree")

            _write("MF_cluster", mf_cluster, (cluster_dim, ytile_dim, xtile_dim), "MF proxy per pixel, cluster masked", "m2 s-2")
            _write("MF_zonal_cluster", mf_zonal_cluster, (cluster_dim, ytile_dim, xtile_dim), "MF cos^2(phi)", "m2 s-2")
            _write("MF_meridional_cluster", mf_meridional_cluster, (cluster_dim, ytile_dim, xtile_dim), "MF sin^2(phi)", "m2 s-2")

            smin, smax = int(np.min(slice_passed)), int(np.max(slice_passed))
            full_slices = np.arange(smin, smax + 1)

            if "slice" not in dst.dimensions:
                dst.createDimension("slice", len(full_slices))
            if "SliceNo_passed_slices" not in dst.variables:
                dst.createVariable("SliceNo_passed_slices", "i4", ("slice",))[:] = full_slices
            else:
                dst.variables["SliceNo_passed_slices"][:] = full_slices

            mf_slice_sum = np.zeros((len(full_slices), y_tile, x_tile), dtype=np.float32)
            mfz_slice_sum = np.zeros_like(mf_slice_sum)
            mfm_slice_sum = np.zeros_like(mf_slice_sum)

            for k, s in enumerate(full_slices):
                idx = np.where(slice_passed == s)[0]
                if idx.size > 0:
                    mf_slice_sum[k] = np.nansum(mf_cluster[idx], axis=0)
                    mfz_slice_sum[k] = np.nansum(mf_zonal_cluster[idx], axis=0)
                    mfm_slice_sum[k] = np.nansum(mf_meridional_cluster[idx], axis=0)

            _write("MF_slice_sum", mf_slice_sum, ("slice", ytile_dim, xtile_dim), "Per slice MF sum (passed clusters), tile grid", "m2 s-2")
            _write("MF_zonal_slice_sum", mfz_slice_sum, ("slice", ytile_dim, xtile_dim), "Per slice zonal MF sum (passed clusters), tile grid", "m2 s-2")
            _write("MF_meridional_slice_sum", mfm_slice_sum, ("slice", ytile_dim, xtile_dim), "Per slice meridional MF sum (passed clusters), tile grid", "m2 s-2")

            MF_toprows = np.zeros(temp_shape, dtype=np.float32)
            MFZ_toprows = np.zeros(temp_shape, dtype=np.float32)
            MFM_toprows = np.zeros(temp_shape, dtype=np.float32)

            for k, s in enumerate(full_slices):
                x0 = s * x_tile
                x1 = x0 + x_tile
                MF_toprows[time_index, 0:y_tile, x0:x1] = mf_slice_sum[k]
                MFZ_toprows[time_index, 0:y_tile, x0:x1] = mfz_slice_sum[k]
                MFM_toprows[time_index, 0:y_tile, x0:x1] = mfm_slice_sum[k]

            _write("MF_toprows", MF_toprows, temp_dims, "Stitched MF sum (passed clusters) on top tile rows only", "m2 s-2")
            _write("MF_zonal_toprows", MFZ_toprows, temp_dims, "Stitched zonal MF share on top tile rows only", "m2 s-2")
            _write("MF_meridional_toprows", MFM_toprows, temp_dims, "Stitched meridional MF share on top tile rows only", "m2 s-2")

        del temp_raw_2d, u_raw_2d, v_raw_2d, n2_raw_2d, bearing_raw_2d
        del temp_full_2d, u_full_2d, v_full_2d, n2_full_2d, bearing_full_2d
        del temp_cluster, u_cluster, v_cluster, n2_cluster, bearing_cluster
        del mf_cluster, mf_zonal_cluster, mf_meridional_cluster
        gc.collect()

    return cluster_dim, ytile_dim, xtile_dim


# ============================================================
# Plotting
# ============================================================
def plot_stitched_mf_components(nc_path: Path, outpng: Path, time_index: int) -> None:
    outpng.parent.mkdir(parents=True, exist_ok=True)

    with Dataset(nc_path, "r") as nc:
        MF = np.asarray(nc.variables["MF_toprows"][time_index], dtype=float)
        MFZ = np.asarray(nc.variables["MF_zonal_toprows"][time_index], dtype=float)
        MFM = np.asarray(nc.variables["MF_meridional_toprows"][time_index], dtype=float)

    ny, nx = MF.shape
    x1d_km = np.arange(nx, dtype=float) * DX_KM
    y1d_km = np.arange(ny, dtype=float) * DY_KM

    stacked = np.concatenate([MF.ravel(), MFZ.ravel(), MFM.ravel()])
    vmin, vmax = robust_limits(stacked, 2, 99.9)

    fig, axs = plt.subplots(3, 1, figsize=(22, 8), constrained_layout=True, sharex=True, sharey=True)

    axs[0].pcolormesh(x1d_km, y1d_km, MF, shading="auto", vmin=vmin, vmax=vmax)
    axs[0].set_title("Stitched MF passed clusters summed")
    axs[0].set_ylabel("y km")
    axs[0].set_aspect("equal", adjustable="box")

    axs[1].pcolormesh(x1d_km, y1d_km, MFZ, shading="auto", vmin=vmin, vmax=vmax)
    axs[1].set_title("Stitched MF zonal share cos2")
    axs[1].set_ylabel("y km")
    axs[1].set_aspect("equal", adjustable="box")

    im2 = axs[2].pcolormesh(x1d_km, y1d_km, MFM, shading="auto", vmin=vmin, vmax=vmax)
    axs[2].set_title("Stitched MF meridional share sin2")
    axs[2].set_xlabel("x km")
    axs[2].set_ylabel("y km")
    axs[2].set_aspect("equal", adjustable="box")

    for ax in axs:
        ax.set_ylim(0, 600)

    cbar = fig.colorbar(im2, ax=axs, fraction=0.02, pad=0.01)
    cbar.set_label("MF m2 s-2")
    save_fig(fig, outpng)

    del MF, MFZ, MFM, stacked, fig, axs
    gc.collect()


# ============================================================
# Pairing utilities
# ============================================================
def _key_from_stem(stem: str) -> str:
    """
    Turn:
      awe_l6c_q20_2024165T2120_03175_v01
      awe_l6a_tmp_2024165T2120_03175_v01
    into the common key:
      2024165T2120_03175_v01
    """
    parts = stem.split("_")
    if len(parts) < 5:
        return stem
    kind = parts[1]  # l6a or l6c
    if kind in {"l6c", "l6a"}:
        return "_".join(parts[3:])
    return "_".join(parts[2:])


def find_l6_pairs(l6_dir: Path) -> List[Tuple[Path, Path, str]]:
    """
    Returns list of (l6c_path, l6a_path, key)
    """
    l6_dir = Path(l6_dir)
    l6c_files = sorted(l6_dir.glob("*l6c*.nc"))
    l6a_files = sorted(l6_dir.glob("*l6a*.nc"))

    l6c_by_key: Dict[str, Path] = {_key_from_stem(p.stem): p for p in l6c_files}
    l6a_by_key: Dict[str, Path] = {_key_from_stem(p.stem): p for p in l6a_files}

    keys = sorted(set(l6c_by_key) & set(l6a_by_key))
    return [(l6c_by_key[k], l6a_by_key[k], k) for k in keys]


def make_l7a_name_from_l6a(l6a_path: Path) -> str:
    """
    Rename l6a -> l7a in the filename, keep everything else.
    """
    name = l6a_path.name
    name = name.replace("_l6a_", "_l7a_")
    name = name.replace("l6a", "l7a")
    return name


# ============================================================
# One pair runner
# ============================================================
def run_one_pair(l6c_path: Path, l6a_path: Path, outdir: Path) -> Dict[str, Any]:
    outdir.mkdir(parents=True, exist_ok=True)

    out_l7a = L7_DIR / make_l7a_name_from_l6a(l6a_path)

    l6c = load_l6_for_matching(l6c_path)
    l6a = load_l6_for_matching(l6a_path)

    all_slices_a = np.unique(l6a["SlicesNo"]).astype(int)
    all_slices_a = all_slices_a[all_slices_a >= 0]
    all_slices_a = np.sort(all_slices_a)

    summary_lines: List[str] = []
    summary_lines.append(f"L6A: {l6a_path}\n")
    summary_lines.append(f"L6C: {l6c_path}\n\n")
    summary_lines.append("slice  nA  nC  nBestSaved  nPassed  note\n")

    passed_global_indices: List[int] = []

    for slice_no in all_slices_a:
        idx_a = np.where(l6a["SlicesNo"] == int(slice_no))[0]
        idx_c = np.where(l6c["SlicesNo"] == int(slice_no))[0]

        slice_dir = outdir / f"slice_{slice_no:03d}"
        pairs_dir = slice_dir / "pairs"
        pairs_dir.mkdir(parents=True, exist_ok=True)

        if idx_a.size == 0:
            summary_lines.append(f"{slice_no:5d}  {0:2d}  {idx_c.size:2d}  {0:10d}  {0:7d}  no L6A clusters\n")
            continue
        if idx_c.size == 0:
            (slice_dir / "NO_L6C_CLUSTERS.txt").write_text(f"No L6C clusters for slice {slice_no}\n")
            summary_lines.append(f"{slice_no:5d}  {idx_a.size:2d}  {0:2d}  {0:10d}  {0:7d}  no L6C clusters\n")
            continue

        all_candidate_rows: List[Dict[str, Any]] = []
        best_rows: List[Dict[str, Any]] = []

        for a_idx in idx_a:
            a_idx = int(a_idx)
            A_a = l6a["Amplitude"][a_idx]
            L_a = l6a["DominantWavelength"][a_idx]
            T_a = l6a["Angle"][a_idx]

            candidates: List[Dict[str, Any]] = []
            for c_idx in idx_c:
                c_idx = int(c_idx)
                A_c = l6c["Amplitude"][c_idx]
                L_c = l6c["DominantWavelength"][c_idx]
                T_c = l6c["Angle"][c_idx]

                m = compute_match_metrics(A_a, L_a, T_a, A_c, L_c, T_c)
                m["cluster_a"] = a_idx
                m["cluster_c"] = c_idx
                m["passes"] = bool(passes_thresholds(m))
                m["score"] = float(score(m))
                candidates.append(m)
                all_candidate_rows.append(m)

            passed = [c for c in candidates if c["passes"]]
            best = max(passed, key=lambda d: d["score"]) if passed else max(candidates, key=lambda d: d["score"])

            if best["passes"] or ASSIGN_BEST_EVEN_IF_FAILS:
                best_rows.append(best)

                if SAVE_PAIR_PLOTS:
                    left_maps = get_cluster_maps(l6a, best["cluster_a"])
                    right_maps = get_cluster_maps(l6c, best["cluster_c"])

                    pass_tag = "PASS" if best["passes"] else "FAIL"
                    outpng = pairs_dir / f"slice_{slice_no:03d}_A_{best['cluster_a']:04d}_to_C_{best['cluster_c']:04d}_{pass_tag}.png"

                    title_left = (
                        f"L6A Temp | slice {slice_no} | idx {best['cluster_a']} | "
                        f"n={np.count_nonzero(np.isfinite(left_maps['A']))}"
                    )
                    title_right = (
                        f"L6C Rad | slice {slice_no} | idx {best['cluster_c']} | "
                        f"IoU={best['iou']:.3f} inter={best['n_intersection']} "
                        f"ratio={best['lam_ratio']:.3f} Δθ={best['ang_diff_deg']:.1f}°"
                    )

                    plot_side_by_side_4rows(
                        left=left_maps,
                        right=right_maps,
                        title_left=title_left,
                        title_right=title_right,
                        outpath=outpng,
                        dx_km=DX_KM,
                        dy_km=DY_KM,
                    )

                    del left_maps, right_maps
                    gc.collect()

        # Write candidate lists for this slice
        (slice_dir / f"slice_{slice_no:03d}_ALL_candidates_ratio.txt").write_text(
            "cluster_a  cluster_c  passes  score    iou    inter  lam_ratio  ang_diff_deg  thetaA_deg  thetaC_deg  lamA_med  lamC_med\n"
            + "".join(
                f"{m['cluster_a']:9d}  {m['cluster_c']:9d}  {int(m['passes']):6d}  {m['score']:6.3f}  "
                f"{m['iou']:.3f}  {m['n_intersection']:5d}  "
                f"{m['lam_ratio']:8.3f}  {m['ang_diff_deg']:12.1f}  "
                f"{m['ang_src_deg']:10.1f}  {m['ang_cand_deg']:10.1f}  "
                f"{m['lam_src_med']:8.1f}  {m['lam_cand_med']:8.1f}\n"
                for m in sorted(all_candidate_rows, key=lambda d: (d["cluster_a"], -d["score"]))
            )
        )

        (slice_dir / f"slice_{slice_no:03d}_BEST_matches_A_to_C_ratio.txt").write_text(
            "cluster_a  best_cluster_c  passes  score    iou    inter  lam_ratio  ang_diff_deg  thetaA_deg  thetaC_deg  lamA_med  lamC_med\n"
            + "".join(
                f"{m['cluster_a']:9d}  {m['cluster_c']:14d}  {int(m['passes']):6d}  {m['score']:6.3f}  "
                f"{m['iou']:.3f}  {m['n_intersection']:5d}  "
                f"{m['lam_ratio']:8.3f}  {m['ang_diff_deg']:12.1f}  "
                f"{m['ang_src_deg']:10.1f}  {m['ang_cand_deg']:10.1f}  "
                f"{m['lam_src_med']:8.1f}  {m['lam_cand_med']:8.1f}\n"
                for m in sorted(best_rows, key=lambda d: d["cluster_a"])
            )
        )

        passed_here = [int(m["cluster_a"]) for m in best_rows if bool(m["passes"])]
        passed_global_indices.extend(passed_here)

        summary_lines.append(
            f"{slice_no:5d}  {idx_a.size:2d}  {idx_c.size:2d}  {len(best_rows):10d}  {len(passed_here):7d}  ok\n"
        )

        del all_candidate_rows, best_rows
        gc.collect()

    summary_path = outdir / "SUMMARY_all_slices.txt"
    summary_path.write_text("".join(summary_lines))

    passed_global_indices = sorted(set(passed_global_indices))
    passed_idx = np.asarray(passed_global_indices, dtype=int)

    if passed_idx.size == 0:
        return dict(ok=False, l6c=str(l6c_path), l6a=str(l6a_path), outdir=str(outdir), l7a=str(out_l7a), n_passed=0,
                    summary=str(summary_path), note="No passed clusters")

    create_passed_only_l6a_with_new_vars(
        src_l6a=l6a_path,
        dst_l7a=out_l7a,
        passed_cluster_indices=passed_idx,
        time_index=TIME_INDEX,
    )

    if SAVE_STITCHED_PLOTS:
        with Dataset(out_l7a, "r") as nc:
            MF2 = np.asarray(nc.variables["MF_toprows"][TIME_INDEX], dtype=float)
            ny, nx = MF2.shape
            x1d_km = np.arange(nx, dtype=float) * DX_KM
            y1d_km = np.arange(ny, dtype=float) * DY_KM
            vmin, vmax = robust_limits(MF2, 2, 99.9)

            fig, ax = plt.subplots(figsize=(22, 3), constrained_layout=True)
            im = ax.pcolormesh(x1d_km, y1d_km, MF2, shading="auto", vmin=vmin, vmax=vmax)
            ax.set_title("Momentum Flux (passed clusters summed)")
            ax.set_xlabel("x (km)")
            ax.set_ylabel("y (km)")
            ax.set_ylim(0, 600)
            ax.set_aspect("equal", adjustable="box")
            fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01).set_label("MF (m2 s-2)")
            save_fig(fig, outdir / "MF_stitched.png")

        plot_stitched_mf_components(out_l7a, outdir / "MF_stitched_components.png", TIME_INDEX)

    # Per-cluster panels are optional and heavy
    # (kept as a hook, but default is False)
    if SAVE_PER_CLUSTER_PANELS:
        # You can plug your per-cluster plotting function here if you want it in batch.
        pass

    del l6c, l6a, passed_idx
    mpl.pyplot.close("all")
    gc.collect()

    return dict(ok=True, l6c=str(l6c_path), l6a=str(l6a_path), outdir=str(outdir), l7a=str(out_l7a),
                n_passed=int(passed_global_indices.__len__()), summary=str(summary_path), note="ok")


# ============================================================
# Main batch
# ============================================================
pairs = find_l6_pairs(L6_DIR)
print(f"Found {len(pairs)} L6C/L6A pairs in: {L6_DIR.resolve()}")

results: List[Dict[str, Any]] = []
errors: List[Dict[str, Any]] = []

for l6c_path, l6a_path, key in pairs:
    pair_outdir = BATCH_OUTROOT / key
    print(f"\n--- Running pair key={key} ---")
    print("L6C:", l6c_path.name)
    print("L6A:", l6a_path.name)
    try:
        res = run_one_pair(l6c_path, l6a_path, pair_outdir)
        results.append(res)
        print(f"OK={res['ok']}  n_passed={res.get('n_passed', 'NA')}")
        print("L7A:", res.get("l7a"))
        print("Outdir:", pair_outdir)
    except Exception as e:
        err = dict(
            ok=False,
            l6c=str(l6c_path),
            l6a=str(l6a_path),
            key=key,
            outdir=str(pair_outdir),
            error=str(e),
            traceback=traceback.format_exc(),
        )
        errors.append(err)
        print("FAILED:", e)

# batch summary
summary_txt = BATCH_OUTROOT / "BATCH_SUMMARY.txt"
lines: List[str] = []
lines.append(f"Input dir: {L6_DIR.resolve()}\n")
lines.append(f"Output nc dir (l7): {L7_DIR.resolve()}\n")
lines.append(f"Pairs found: {len(pairs)}\n")
lines.append(f"Success: {sum(1 for r in results if r.get('ok'))}\n")
lines.append(f"Fail: {len(errors)}\n\n")

lines.append("Per pair:\n")
for r in results:
    key = Path(r["outdir"]).name
    lines.append(f"{key}  ok={r['ok']}  n_passed={r.get('n_passed', 0)}  l7a={r.get('l7a')}\n")

if errors:
    lines.append("\nErrors:\n")
    for e in errors:
        lines.append(f"{e['key']}  {e['error']}\n")

summary_txt.write_text("".join(lines))

mpl.pyplot.close("all")
gc.collect()

print("\nBatch done.")
print("Batch outroot:", BATCH_OUTROOT.resolve())
print("L7 output folder:", L7_DIR.resolve())
print("Batch summary:", summary_txt.resolve())
```

## Flat Lat, Lon, T

```python
# %%
"""
BATCH: ALL-SLICES matching + per-slice combined L6A recon + FULL-ORBIT stitched recon
+ slice Temperature + momentum flux for PASSED clusters only
+ sum MF per slice + stitch MF across orbit
+ output NetCDF contains original L6A global attrs + variables, but ONLY PASSED clusters
  are kept for cluster-dimension variables.

Momentum flux (per cluster pixel, masked to cluster area):
MF = 1/2 * g^2 / N^2 * (lambda_z / lambda_x) * (Amplitude / Temperature)^2 * (1/c^2)

Key rules:
- SliceNo already exists in L6A, keep it (but subset to passed clusters)
- Area, Amplitude, DominantWavelength, Angle already exist (subset to passed clusters)
- Also slice Temperature into Temp_cluster (passed clusters only)
- Add geodesic Bearing (0 = East, counter-clockwise) as Bearing + Bearing_cluster
- Only save passed clusters (output file cluster vars are subset)
- Add MF_cluster, MF_slice_sum, MF_toprows
- Also write flattened toprows Temperature, Latitude, Longitude aligned with MF_toprows

BATCH behavior:
- Scan L6_DIR for L6A and L6C files
- Pair by common suffix key like: 2024165T2120_03175_v01
- For each pair, write the new file into L7_DIR and rename l6a -> l7a
- Save plots + text into outputs folder per key
"""

from __future__ import annotations

from pathlib import Path
from typing import Dict, Any, List, Tuple
import traceback
import gc

import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from netCDF4 import Dataset

# ============================================================
# Batch params
# ============================================================
L6_DIR = Path("l6")  # contains many L6A/L6C files
L7_DIR = Path("l7")  # where new NetCDFs go
L7_DIR.mkdir(parents=True, exist_ok=True)

BATCH_OUTROOT = Path("outputs_BATCH_l6a_to_l6c_matching_ratio_STITCH_MF")
BATCH_OUTROOT.mkdir(parents=True, exist_ok=True)

SAVE_DPI = 200
DX_KM = 2.0
DY_KM = 2.0

# ---- Synthetic u,v,N^2 settings ----
TEMP_VAR_NAME = "Temperature"
TIME_INDEX = 0
N2_VALUE = 4e-4
U_VALUE = 0.0
V_VALUE = 0.0

# ---- Momentum flux constants ----
G_MS2 = 9.52
LAMBDA_Z_KM = 10.0
C_CANCEL = 1.0

# --------------------------
# Matching thresholds (tune)
# --------------------------
MIN_IOU = 0.05
MIN_INTERSECTION_PIX = 50
MAX_LAMBDA_MED_RATIO = 1.25
MAX_ANGLE_DIFF_DEG = 15.0

# Save best match even if FAIL (does NOT affect passing)
ASSIGN_BEST_EVEN_IF_FAILS = True

# --------------------------
# Plot saving switches (batch friendly)
# --------------------------
SAVE_PAIR_PLOTS = False          # side-by-side A vs C match PNGs per slice (heavy)
SAVE_PER_CLUSTER_PANELS = False  # big panels per passed cluster (heavy)
SAVE_STITCHED_PLOTS = True       # MF_stitched.png + MF_stitched_components.png


# ============================================================
# Helpers
# ============================================================
def save_fig(fig: plt.Figure, outpath: Path) -> None:
    outpath.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(outpath, dpi=SAVE_DPI, bbox_inches="tight")
    plt.close(fig)


def robust_limits(A: np.ndarray, p_lo: float = 2, p_hi: float = 98) -> Tuple[float, float]:
    a = np.asarray(A)
    good = np.isfinite(a)
    if not np.any(good):
        return (-1.0, 1.0)
    lo, hi = np.nanpercentile(a[good], [p_lo, p_hi])
    if (not np.isfinite(lo)) or (not np.isfinite(hi)) or lo == hi:
        lo = float(np.nanmin(a[good]))
        hi = float(np.nanmax(a[good]))
    return (float(lo), float(hi))


def nanpercentile_safe(a: np.ndarray, q: float) -> float:
    good = np.isfinite(a)
    if not np.any(good):
        return np.nan
    return float(np.nanpercentile(a[good], q))


def circ_diff_deg_180(a_deg: float, b_deg: float) -> float:
    if (not np.isfinite(a_deg)) or (not np.isfinite(b_deg)):
        return np.nan
    d = abs(a_deg - b_deg) % 180.0
    return min(d, 180.0 - d)


def circ_median_deg_180(theta_rad: np.ndarray, mask: np.ndarray) -> float:
    th = np.asarray(theta_rad, dtype=float)
    m = mask & np.isfinite(th)
    if not np.any(m):
        return np.nan
    ang = th[m]
    z = np.exp(1j * 2.0 * ang)
    zmean = np.mean(z)
    if (not np.isfinite(zmean.real)) or (not np.isfinite(zmean.imag)) or abs(zmean) == 0:
        return np.nan
    th_mean = 0.5 * np.angle(zmean)
    deg = (np.degrees(th_mean) + 180.0) % 180.0
    return float(deg)


# -------------------------
# Geodesic bearing: 0 = East, CCW
# -------------------------
def initial_bearing_deg_north_cw(lat1_deg, lon1_deg, lat2_deg, lon2_deg):
    lat1 = np.deg2rad(lat1_deg)
    lon1 = np.deg2rad(lon1_deg)
    lat2 = np.deg2rad(lat2_deg)
    lon2 = np.deg2rad(lon2_deg)

    dlon = lon2 - lon1
    x = np.sin(dlon) * np.cos(lat2)
    y = np.cos(lat1) * np.sin(lat2) - np.sin(lat1) * np.cos(lat2) * np.cos(dlon)

    az = np.arctan2(x, y)
    return (np.rad2deg(az) + 360.0) % 360.0


def bearing_deg_east_ccw_from_points(lat1_deg, lon1_deg, lat2_deg, lon2_deg):
    az_north_cw = initial_bearing_deg_north_cw(lat1_deg, lon1_deg, lat2_deg, lon2_deg)
    return (90.0 - az_north_cw) % 360.0


def compute_bearing_to_x_neighbor_east_ccw(lat2d: np.ndarray, lon2d: np.ndarray) -> np.ndarray:
    lat = np.asarray(lat2d, dtype=float)
    lon = np.asarray(lon2d, dtype=float)

    lat2 = np.empty_like(lat)
    lon2 = np.empty_like(lon)

    lat2[:, :-1] = lat[:, 1:]
    lon2[:, :-1] = lon[:, 1:]
    lat2[:, -1] = lat[:, -2]
    lon2[:, -1] = lon[:, -2]

    bearing = bearing_deg_east_ccw_from_points(lat, lon, lat2, lon2)
    good = np.isfinite(lat) & np.isfinite(lon) & np.isfinite(lat2) & np.isfinite(lon2)
    return np.where(good, bearing, np.nan)


# ============================================================
# Matching IO (memory friendly): do NOT load ClusterReconstruction upfront
# ============================================================
def load_l6_for_matching(path: Path) -> Dict[str, Any]:
    with Dataset(path, "r") as nc:
        return {
            "path": str(path),
            "SlicesNo": np.asarray(nc.variables["SlicesNo"][:], dtype=int),
            "Amplitude": np.asarray(nc.variables["Amplitude"][:], dtype=float),
            "DominantWavelength": np.asarray(nc.variables["DominantWavelength"][:], dtype=float),
            "Angle": np.asarray(nc.variables["Angle"][:], dtype=float),
        }


def get_cluster_maps(data: Dict[str, Any], idx: int) -> Dict[str, np.ndarray]:
    """
    Lazy load ClusterReconstruction only for requested idx (only used if SAVE_PAIR_PLOTS=True).
    """
    A = np.asarray(data["Amplitude"][idx], dtype=float)
    L = np.asarray(data["DominantWavelength"][idx], dtype=float)
    T = np.asarray(data["Angle"][idx], dtype=float)

    R = np.full_like(A, np.nan)
    with Dataset(data["path"], "r") as nc:
        if "ClusterReconstruction" in nc.variables:
            R = np.asarray(nc.variables["ClusterReconstruction"][idx], dtype=float)

    return dict(R=R, A=A, L=L, T=T)


def compute_match_metrics(
    A_src: np.ndarray, L_src: np.ndarray, T_src: np.ndarray,
    A_cand: np.ndarray, L_cand: np.ndarray, T_cand: np.ndarray,
) -> Dict[str, Any]:
    mask_src = np.isfinite(A_src)
    mask_cand = np.isfinite(A_cand)
    inter = mask_src & mask_cand
    union = mask_src | mask_cand

    n_inter = int(np.count_nonzero(inter))
    n_union = int(np.count_nonzero(union))
    iou = (n_inter / n_union) if n_union > 0 else 0.0

    lam_src = L_src[inter]
    lam_cand = L_cand[inter]
    lam_src_med = nanpercentile_safe(lam_src, 50)
    lam_cand_med = nanpercentile_safe(lam_cand, 50)

    if np.isfinite(lam_src_med) and np.isfinite(lam_cand_med) and lam_src_med > 0 and lam_cand_med > 0:
        lam_ratio = max(lam_src_med, lam_cand_med) / min(lam_src_med, lam_cand_med)
    else:
        lam_ratio = np.nan

    ang_src_deg = circ_median_deg_180(T_src, inter)
    ang_cand_deg = circ_median_deg_180(T_cand, inter)
    ang_diff_deg = circ_diff_deg_180(ang_src_deg, ang_cand_deg)

    return dict(
        n_intersection=n_inter,
        iou=float(iou),
        lam_src_med=float(lam_src_med),
        lam_cand_med=float(lam_cand_med),
        lam_ratio=float(lam_ratio),
        ang_src_deg=float(ang_src_deg),
        ang_cand_deg=float(ang_cand_deg),
        ang_diff_deg=float(ang_diff_deg),
    )


def passes_thresholds(m: Dict[str, Any]) -> bool:
    if m["iou"] < MIN_IOU:
        return False
    if m["n_intersection"] < MIN_INTERSECTION_PIX:
        return False
    if (not np.isfinite(m["lam_ratio"])) or (m["lam_ratio"] > MAX_LAMBDA_MED_RATIO):
        return False
    if (not np.isfinite(m["ang_diff_deg"])) or (m["ang_diff_deg"] > MAX_ANGLE_DIFF_DEG):
        return False
    return True


def score(m: Dict[str, Any]) -> float:
    ang = m["ang_diff_deg"]
    if not np.isfinite(ang):
        ang = 999.0
    r = m["lam_ratio"]
    if (not np.isfinite(r)) or r <= 0:
        return 0.0
    return float(m["iou"]) * (1.0 / float(r)) / (1.0 + float(ang))


def plot_side_by_side_4rows(
    left: Dict[str, np.ndarray],
    right: Dict[str, np.ndarray],
    title_left: str,
    title_right: str,
    outpath: Path,
    dx_km: float,
    dy_km: float,
) -> None:
    ny, nx = left["A"].shape
    x1d_km = np.arange(nx, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    Rl_vmin, Rl_vmax = robust_limits(left["R"])
    Rr_vmin, Rr_vmax = robust_limits(right["R"])
    Al_vmin, Al_vmax = robust_limits(left["A"])
    Ar_vmin, Ar_vmax = robust_limits(right["A"])

    L_vmin, L_vmax = robust_limits(np.concatenate([left["L"].ravel(), right["L"].ravel()]), 5, 95)
    T_vmin, T_vmax = robust_limits(np.concatenate([left["T"].ravel(), right["T"].ravel()]))

    fig, axs = plt.subplots(4, 2, figsize=(16, 14), constrained_layout=True, sharex=True, sharey=True)

    def _panel(ax, Z, vmin, vmax, label, cmap=None):
        im = ax.pcolormesh(x1d_km, y1d_km, Z, shading="auto", vmin=vmin, vmax=vmax, cmap=cmap)
        cb = fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01)
        cb.set_label(label)
        ax.set_aspect("equal", adjustable="box")

    axs[0, 0].set_title(title_left)
    axs[0, 1].set_title(title_right)

    _panel(axs[0, 0], left["R"], Rl_vmin, Rl_vmax, "arb", cmap="RdBu_r")
    _panel(axs[0, 1], right["R"], Rr_vmin, Rr_vmax, "arb", cmap="RdBu_r")
    axs[0, 0].set_ylabel("Recon\ny (km)")

    _panel(axs[1, 0], left["A"], Al_vmin, Al_vmax, "K")
    _panel(axs[1, 1], right["A"], Ar_vmin, Ar_vmax, "kR")
    axs[1, 0].set_ylabel("Amp\ny (km)")

    _panel(axs[2, 0], left["L"], L_vmin, L_vmax, "km")
    _panel(axs[2, 1], right["L"], L_vmin, L_vmax, "km")
    axs[2, 0].set_ylabel("Lambda\ny (km)")

    _panel(axs[3, 0], left["T"], T_vmin, T_vmax, "rad")
    _panel(axs[3, 1], right["T"], T_vmin, T_vmax, "rad")
    axs[3, 0].set_ylabel("Angle\ny (km)")

    for r in range(4):
        for c in range(2):
            axs[r, c].set_xlabel("x (km)")

    save_fig(fig, outpath)


# --------------------------
# Flatten swath BEFORE slicing
# --------------------------
def flatten_swath(Z: np.ndarray, target_ny: int | None = None) -> np.ndarray:
    ny, nx = Z.shape
    out = np.full_like(Z, np.nan)

    bottoms = np.full(nx, np.nan, dtype=float)
    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size:
            bottoms[j] = valid_idx.max()

    valid_bottoms = bottoms[np.isfinite(bottoms)]
    if valid_bottoms.size == 0:
        med = float(np.nanmedian(Z)) if np.isfinite(np.nanmedian(Z)) else 0.0
        out = np.nan_to_num(Z, nan=med, posinf=med, neginf=med)
        if target_ny is not None and ny > target_ny:
            out = out[-target_ny:, :]
        return out

    max_bottom = int(valid_bottoms.max())

    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size == 0:
            continue
        bottom = int(valid_idx.max())
        shift = max_bottom - bottom
        if shift == 0:
            out[:, j] = col
        elif shift > 0:
            out[shift:, j] = col[: ny - shift]
        else:
            shift_up = -shift
            out[: ny - shift_up, j] = col[shift_up:]

    med = float(np.nanmedian(out)) if np.isfinite(np.nanmedian(out)) else 0.0
    out = np.nan_to_num(out, nan=med, posinf=med, neginf=med)

    if target_ny is not None and ny > target_ny:
        start = max_bottom + 1 - target_ny
        start = max(start, 0)
        end = start + target_ny
        out = out[start:end, :]

    return out


# ============================================================
# L6A cloning with PASSED-only cluster variables + new vars
# ============================================================
def create_passed_only_l6a_with_new_vars(
    src_l6a: Path,
    dst_l7a: Path,
    passed_cluster_indices: np.ndarray,
    time_index: int,
) -> Tuple[str, str, str]:
    passed_cluster_indices = np.asarray(passed_cluster_indices, dtype=int)
    if passed_cluster_indices.size == 0:
        raise ValueError("No passed clusters. Output would be empty.")

    dst_l7a.parent.mkdir(parents=True, exist_ok=True)

    with Dataset(src_l6a, "r") as src:
        amp_var = src.variables["Amplitude"]
        cluster_dim, ytile_dim, xtile_dim = amp_var.dimensions[:3]
        _, y_tile, x_tile = amp_var.shape

        temp_var = src.variables[TEMP_VAR_NAME]
        temp_dims = temp_var.dimensions
        temp_shape = temp_var.shape
        temp_raw_2d = np.asarray(temp_var[time_index], dtype=float)

        u_raw_2d = (
            np.asarray(src.variables["u"][time_index], dtype=float)
            if "u" in src.variables
            else np.full_like(temp_raw_2d, U_VALUE, dtype=float)
        )
        v_raw_2d = (
            np.asarray(src.variables["v"][time_index], dtype=float)
            if "v" in src.variables
            else np.full_like(temp_raw_2d, V_VALUE, dtype=float)
        )
        n2_raw_2d = (
            np.asarray(src.variables["NSquared"][time_index], dtype=float)
            if "NSquared" in src.variables
            else np.full_like(temp_raw_2d, N2_VALUE, dtype=float)
        )

        # ---- Latitude/Longitude (optional) + bearing uses RAW geo ----
        if ("Latitude" in src.variables) and ("Longitude" in src.variables):
            lat_raw_2d = np.asarray(src.variables["Latitude"][time_index], dtype=float)
            lon_raw_2d = np.asarray(src.variables["Longitude"][time_index], dtype=float)

            bearing_raw_2d = compute_bearing_to_x_neighbor_east_ccw(lat_raw_2d, lon_raw_2d)

            lat_full_2d = flatten_swath(lat_raw_2d, target_ny=y_tile)
            lon_full_2d = flatten_swath(lon_raw_2d, target_ny=y_tile)
        else:
            lat_raw_2d = None
            lon_raw_2d = None
            bearing_raw_2d = np.full_like(temp_raw_2d, np.nan, dtype=float)
            lat_full_2d = np.full((y_tile, temp_raw_2d.shape[1]), np.nan, dtype=float)
            lon_full_2d = np.full((y_tile, temp_raw_2d.shape[1]), np.nan, dtype=float)

        temp_full_2d = flatten_swath(temp_raw_2d, target_ny=y_tile)
        u_full_2d = flatten_swath(u_raw_2d, target_ny=y_tile)
        v_full_2d = flatten_swath(v_raw_2d, target_ny=y_tile)
        n2_full_2d = flatten_swath(n2_raw_2d, target_ny=y_tile)
        bearing_full_2d = flatten_swath(bearing_raw_2d, target_ny=y_tile)

        n_passed = int(passed_cluster_indices.size)

        with Dataset(dst_l7a, "w") as dst:
            for aname in src.ncattrs():
                dst.setncattr(aname, src.getncattr(aname))

            for dname, dim in src.dimensions.items():
                if dname == cluster_dim:
                    dst.createDimension(dname, n_passed)
                else:
                    dst.createDimension(dname, (None if dim.isunlimited() else len(dim)))

            # Copy all variables, but subset cluster_dim variables to PASSED ONLY
            for vname, vsrc in src.variables.items():
                fill_value = vsrc.getncattr("_FillValue") if "_FillValue" in vsrc.ncattrs() else None
                vdst = dst.createVariable(
                    vname,
                    vsrc.dtype,
                    vsrc.dimensions,
                    zlib=True,
                    complevel=4,
                    fill_value=fill_value,
                )
                for aname in vsrc.ncattrs():
                    if aname != "_FillValue":
                        vdst.setncattr(aname, vsrc.getncattr(aname))

                if cluster_dim in vsrc.dimensions:
                    ax = vsrc.dimensions.index(cluster_dim)
                    vdst[:] = np.take(vsrc[:], passed_cluster_indices, axis=ax)
                else:
                    vdst[:] = vsrc[:]

            # Bearing full field (same dims as Temperature)
            bearing_full = np.zeros(temp_shape, dtype=np.float32)
            bearing_full[time_index, 0:y_tile, :] = bearing_full_2d.astype(np.float32)

            if "Bearing" in dst.variables:
                dst.variables["Bearing"][:] = bearing_full
            else:
                vB = dst.createVariable("Bearing", "f4", temp_dims, zlib=True, complevel=4)
                vB.setncattr("long_name", "Geodesic bearing to +x neighbor (0=East, CCW)")
                vB.setncattr("units", "degree")
                vB[:] = bearing_full

            # Allocate per cluster masked tiles
            temp_cluster = np.full((n_passed, y_tile, x_tile), np.nan, dtype=np.float32)
            u_cluster = np.full_like(temp_cluster, np.nan)
            v_cluster = np.full_like(temp_cluster, np.nan)
            n2_cluster = np.full_like(temp_cluster, np.nan)
            bearing_cluster = np.full_like(temp_cluster, np.nan)

            mf_cluster = np.full_like(temp_cluster, np.nan)
            mfz_cluster = np.full_like(temp_cluster, np.nan)
            mfm_cluster = np.full_like(temp_cluster, np.nan)

            amp_passed = np.asarray(dst.variables["Amplitude"][:], dtype=float)
            lam_passed = np.asarray(dst.variables["DominantWavelength"][:], dtype=float)
            slice_passed = np.asarray(dst.variables["SlicesNo"][:], dtype=int)
            ang_rad_passed = np.asarray(dst.variables["Angle"][:], dtype=float)

            for j in range(n_passed):
                s = int(slice_passed[j])
                x0 = s * x_tile
                x1 = x0 + x_tile

                mask = np.isfinite(amp_passed[j])

                Ttile = temp_full_2d[:, x0:x1]
                Utile = u_full_2d[:, x0:x1]
                Vtile = v_full_2d[:, x0:x1]
                N2tile = n2_full_2d[:, x0:x1]
                Btile = bearing_full_2d[:, x0:x1]

                temp_cluster[j][mask] = Ttile[mask]
                u_cluster[j][mask] = Utile[mask]
                v_cluster[j][mask] = Vtile[mask]
                n2_cluster[j][mask] = N2tile[mask]
                bearing_cluster[j][mask] = Btile[mask]

                A = amp_passed[j]
                Lx = lam_passed[j]
                ang_deg_tile = (np.rad2deg(ang_rad_passed[j]) % 360.0)

                good = mask
                good = good & np.isfinite(Ttile) & np.isfinite(N2tile) & np.isfinite(Lx) & np.isfinite(Btile)
                good = good & (Ttile != 0.0) & (N2tile > 0.0) & (Lx > 0.0)

                if np.any(good):
                    mf_val = (
                        0.5
                        * (G_MS2**2)
                        / N2tile[good]
                        * (LAMBDA_Z_KM / Lx[good])
                        * (A[good] / Ttile[good]) ** 2
                        * (1.0 / (C_CANCEL**2))
                    )
                    mf_cluster[j][good] = mf_val

                    phi_deg = (Btile[good] + ang_deg_tile[good]) % 360.0
                    phi_rad = np.deg2rad(phi_deg)

                    mfz_cluster[j][good] = mf_val * (np.cos(phi_rad) ** 2)
                    mfm_cluster[j][good] = mf_val * (np.sin(phi_rad) ** 2)

            def _write(name: str, arr: np.ndarray, dims: Tuple[str, ...], long_name: str, units: str):
                if name in dst.variables:
                    dst.variables[name][:] = arr
                else:
                    v = dst.createVariable(name, "f4", dims, zlib=True, complevel=4)
                    v.setncattr("long_name", long_name)
                    v.setncattr("units", units)
                    v[:] = arr

            _write("Temp_cluster", temp_cluster, (cluster_dim, ytile_dim, xtile_dim), "Temperature, cluster masked", "K")
            _write("u_cluster", u_cluster, (cluster_dim, ytile_dim, xtile_dim), "u wind, cluster masked", "m/s")
            _write("v_cluster", v_cluster, (cluster_dim, ytile_dim, xtile_dim), "v wind, cluster masked", "m/s")
            _write("NSquared_cluster", n2_cluster, (cluster_dim, ytile_dim, xtile_dim), "N^2, cluster masked", "s^-2")
            _write("Bearing_cluster", bearing_cluster, (cluster_dim, ytile_dim, xtile_dim), "Bearing (0=East, CCW), cluster masked", "degree")

            _write("MF_cluster", mf_cluster, (cluster_dim, ytile_dim, xtile_dim), "MF proxy per pixel, cluster masked", "m2 s-2")
            _write("MFz_cluster", mfz_cluster, (cluster_dim, ytile_dim, xtile_dim), "MF cos^2(phi)", "m2 s-2")
            _write("MFm_cluster", mfm_cluster, (cluster_dim, ytile_dim, xtile_dim), "MF sin^2(phi)", "m2 s-2")

            # Slice sums on tile grid
            smin, smax = int(np.min(slice_passed)), int(np.max(slice_passed))
            full_slices = np.arange(smin, smax + 1)

            if "slice" not in dst.dimensions:
                dst.createDimension("slice", len(full_slices))
            if "SliceNo_passed_slices" not in dst.variables:
                dst.createVariable("SliceNo_passed_slices", "i4", ("slice",))[:] = full_slices
            else:
                dst.variables["SliceNo_passed_slices"][:] = full_slices

            mf_slice = np.zeros((len(full_slices), y_tile, x_tile), dtype=np.float32)
            mfz_slice = np.zeros_like(mf_slice)
            mfm_slice = np.zeros_like(mf_slice)

            for k, s in enumerate(full_slices):
                idx = np.where(slice_passed == s)[0]
                if idx.size > 0:
                    mf_slice[k] = np.nansum(mf_cluster[idx], axis=0)
                    mfz_slice[k] = np.nansum(mfz_cluster[idx], axis=0)
                    mfm_slice[k] = np.nansum(mfm_cluster[idx], axis=0)

            _write("MF_slice", mf_slice, ("slice", ytile_dim, xtile_dim), "Per slice MF sum (passed clusters), tile grid", "m2 s-2")
            _write("MFz_slice", mfz_slice, ("slice", ytile_dim, xtile_dim), "Per slice zonal MF sum (passed clusters), tile grid", "m2 s-2")
            _write("MFm_slice", mfm_slice, ("slice", ytile_dim, xtile_dim), "Per slice meridional MF sum (passed clusters), tile grid", "m2 s-2")

            # Stitch onto full x (top tile rows only)
            MF_top = np.full(temp_shape, np.nan, dtype=np.float32)
            MFz_top = np.full(temp_shape, np.nan, dtype=np.float32)
            MFm_top = np.full(temp_shape, np.nan, dtype=np.float32)

            for k, s in enumerate(full_slices):
                x0 = s * x_tile
                x1 = x0 + x_tile
                MF_top[time_index, 0:y_tile, x0:x1] = mf_slice[k]
                MFz_top[time_index, 0:y_tile, x0:x1] = mfz_slice[k]
                MFm_top[time_index, 0:y_tile, x0:x1] = mfm_slice[k]

            _write("MF_top", MF_top, temp_dims, "Stitched MF sum (passed clusters) on top tile rows only", "m2 s-2")
            _write("MFz_top", MFz_top, temp_dims, "Stitched zonal MF share on top tile rows only", "m2 s-2")
            _write("MFm_top", MFm_top, temp_dims, "Stitched meridional MF share on top tile rows only", "m2 s-2")

            # Flattened toprows Temperature, Latitude, Longitude aligned with MF_top
            T_top = np.full(temp_shape, np.nan, dtype=np.float32)
            Lat_top = np.full(temp_shape, np.nan, dtype=np.float32)
            Lon_top = np.full(temp_shape, np.nan, dtype=np.float32)

            T_top[time_index, 0:y_tile, :] = temp_full_2d.astype(np.float32)
            Lat_top[time_index, 0:y_tile, :] = lat_full_2d.astype(np.float32)
            Lon_top[time_index, 0:y_tile, :] = lon_full_2d.astype(np.float32)

            _write("T_top", T_top, temp_dims, "Flattened Temperature on MF_top grid (top tile rows only)", "K")
            _write("Lat_top", Lat_top, temp_dims, "Flattened Latitude on MF_top grid (top tile rows only)", "degree_north")
            _write("Lon_top", Lon_top, temp_dims, "Flattened Longitude on MF_top grid (top tile rows only)", "degree_east")

        del temp_raw_2d, u_raw_2d, v_raw_2d, n2_raw_2d
        del temp_full_2d, u_full_2d, v_full_2d, n2_full_2d
        del bearing_raw_2d, bearing_full_2d
        del temp_cluster, u_cluster, v_cluster, n2_cluster, bearing_cluster
        del mf_cluster, mfz_cluster, mfm_cluster
        gc.collect()

    return cluster_dim, ytile_dim, xtile_dim


# ============================================================
# Plotting
# ============================================================
def plot_stitched_mf_components(nc_path: Path, outpng: Path, time_index: int) -> None:
    outpng.parent.mkdir(parents=True, exist_ok=True)

    with Dataset(nc_path, "r") as nc:
        MF = np.asarray(nc.variables["MF_top"][time_index], dtype=float)
        MFz = np.asarray(nc.variables["MFz_top"][time_index], dtype=float)
        MFm = np.asarray(nc.variables["MFm_top"][time_index], dtype=float)

    ny, nx = MF.shape
    x1d_km = np.arange(nx, dtype=float) * DX_KM
    y1d_km = np.arange(ny, dtype=float) * DY_KM

    stacked = np.concatenate([MF.ravel(), MFz.ravel(), MFm.ravel()])
    vmin, vmax = robust_limits(stacked, 2, 99.9)

    fig, axs = plt.subplots(3, 1, figsize=(22, 8), constrained_layout=True, sharex=True, sharey=True)

    axs[0].pcolormesh(x1d_km, y1d_km, MF, shading="auto", vmin=vmin, vmax=vmax)
    axs[0].set_title("Stitched MF (passed clusters summed)")
    axs[0].set_ylabel("y km")
    axs[0].set_aspect("equal", adjustable="box")

    axs[1].pcolormesh(x1d_km, y1d_km, MFz, shading="auto", vmin=vmin, vmax=vmax)
    axs[1].set_title("Stitched MF zonal share cos2")
    axs[1].set_ylabel("y km")
    axs[1].set_aspect("equal", adjustable="box")

    im2 = axs[2].pcolormesh(x1d_km, y1d_km, MFm, shading="auto", vmin=vmin, vmax=vmax)
    axs[2].set_title("Stitched MF meridional share sin2")
    axs[2].set_xlabel("x km")
    axs[2].set_ylabel("y km")
    axs[2].set_aspect("equal", adjustable="box")

    for ax in axs:
        ax.set_ylim(0, 600)

    cbar = fig.colorbar(im2, ax=axs, fraction=0.02, pad=0.01)
    cbar.set_label("MF m2 s-2")
    save_fig(fig, outpng)

    del MF, MFz, MFm, stacked, fig, axs
    gc.collect()


# ============================================================
# Pairing utilities
# ============================================================
def _key_from_stem(stem: str) -> str:
    """
    Turn:
      awe_l6c_q20_2024165T2120_03175_v01
      awe_l6a_tmp_2024165T2120_03175_v01
    into the common key:
      2024165T2120_03175_v01
    """
    parts = stem.split("_")
    if len(parts) < 5:
        return stem
    kind = parts[1]  # l6a or l6c
    if kind in {"l6c", "l6a"}:
        return "_".join(parts[3:])
    return "_".join(parts[2:])


def find_l6_pairs(l6_dir: Path) -> List[Tuple[Path, Path, str]]:
    """
    Returns list of (l6c_path, l6a_path, key)
    """
    l6_dir = Path(l6_dir)
    l6c_files = sorted(l6_dir.glob("*l6c*.nc"))
    l6a_files = sorted(l6_dir.glob("*l6a*.nc"))

    l6c_by_key: Dict[str, Path] = {_key_from_stem(p.stem): p for p in l6c_files}
    l6a_by_key: Dict[str, Path] = {_key_from_stem(p.stem): p for p in l6a_files}

    keys = sorted(set(l6c_by_key) & set(l6a_by_key))
    return [(l6c_by_key[k], l6a_by_key[k], k) for k in keys]


def make_l7a_name_from_l6a(l6a_path: Path) -> str:
    """
    Rename l6a -> l7a in the filename, keep everything else.
    """
    name = l6a_path.name
    name = name.replace("_l6a_", "_l7a_")
    name = name.replace("l6a", "l7a")
    return name


# ============================================================
# One pair runner
# ============================================================
def run_one_pair(l6c_path: Path, l6a_path: Path, outdir: Path) -> Dict[str, Any]:
    outdir.mkdir(parents=True, exist_ok=True)

    out_l7a = L7_DIR / make_l7a_name_from_l6a(l6a_path)

    l6c = load_l6_for_matching(l6c_path)
    l6a = load_l6_for_matching(l6a_path)

    all_slices_a = np.unique(l6a["SlicesNo"]).astype(int)
    all_slices_a = all_slices_a[all_slices_a >= 0]
    all_slices_a = np.sort(all_slices_a)

    summary_lines: List[str] = []
    summary_lines.append(f"L6A: {l6a_path}\n")
    summary_lines.append(f"L6C: {l6c_path}\n\n")
    summary_lines.append("slice  nA  nC  nBestSaved  nPassed  note\n")

    passed_global_indices: List[int] = []

    for slice_no in all_slices_a:
        idx_a = np.where(l6a["SlicesNo"] == int(slice_no))[0]
        idx_c = np.where(l6c["SlicesNo"] == int(slice_no))[0]

        slice_dir = outdir / f"slice_{slice_no:03d}"
        pairs_dir = slice_dir / "pairs"
        pairs_dir.mkdir(parents=True, exist_ok=True)

        if idx_a.size == 0:
            summary_lines.append(f"{slice_no:5d}  {0:2d}  {idx_c.size:2d}  {0:10d}  {0:7d}  no L6A clusters\n")
            continue
        if idx_c.size == 0:
            (slice_dir / "NO_L6C_CLUSTERS.txt").write_text(f"No L6C clusters for slice {slice_no}\n")
            summary_lines.append(f"{slice_no:5d}  {idx_a.size:2d}  {0:2d}  {0:10d}  {0:7d}  no L6C clusters\n")
            continue

        all_candidate_rows: List[Dict[str, Any]] = []
        best_rows: List[Dict[str, Any]] = []

        for a_idx in idx_a:
            a_idx = int(a_idx)
            A_a = l6a["Amplitude"][a_idx]
            L_a = l6a["DominantWavelength"][a_idx]
            T_a = l6a["Angle"][a_idx]

            candidates: List[Dict[str, Any]] = []
            for c_idx in idx_c:
                c_idx = int(c_idx)
                A_c = l6c["Amplitude"][c_idx]
                L_c = l6c["DominantWavelength"][c_idx]
                T_c = l6c["Angle"][c_idx]

                m = compute_match_metrics(A_a, L_a, T_a, A_c, L_c, T_c)
                m["cluster_a"] = a_idx
                m["cluster_c"] = c_idx
                m["passes"] = bool(passes_thresholds(m))
                m["score"] = float(score(m))
                candidates.append(m)
                all_candidate_rows.append(m)

            passed = [c for c in candidates if c["passes"]]
            best = max(passed, key=lambda d: d["score"]) if passed else max(candidates, key=lambda d: d["score"])

            if best["passes"] or ASSIGN_BEST_EVEN_IF_FAILS:
                best_rows.append(best)

                if SAVE_PAIR_PLOTS:
                    left_maps = get_cluster_maps(l6a, best["cluster_a"])
                    right_maps = get_cluster_maps(l6c, best["cluster_c"])

                    pass_tag = "PASS" if best["passes"] else "FAIL"
                    outpng = pairs_dir / f"slice_{slice_no:03d}_A_{best['cluster_a']:04d}_to_C_{best['cluster_c']:04d}_{pass_tag}.png"

                    title_left = (
                        f"L6A Temp | slice {slice_no} | idx {best['cluster_a']} | "
                        f"n={np.count_nonzero(np.isfinite(left_maps['A']))}"
                    )
                    title_right = (
                        f"L6C Rad | slice {slice_no} | idx {best['cluster_c']} | "
                        f"IoU={best['iou']:.3f} inter={best['n_intersection']} "
                        f"ratio={best['lam_ratio']:.3f} dTheta={best['ang_diff_deg']:.1f}deg"
                    )

                    plot_side_by_side_4rows(
                        left=left_maps,
                        right=right_maps,
                        title_left=title_left,
                        title_right=title_right,
                        outpath=outpng,
                        dx_km=DX_KM,
                        dy_km=DY_KM,
                    )

                    del left_maps, right_maps
                    gc.collect()

        (slice_dir / f"slice_{slice_no:03d}_ALL_candidates_ratio.txt").write_text(
            "cluster_a  cluster_c  passes  score    iou    inter  lam_ratio  ang_diff_deg  thetaA_deg  thetaC_deg  lamA_med  lamC_med\n"
            + "".join(
                f"{m['cluster_a']:9d}  {m['cluster_c']:9d}  {int(m['passes']):6d}  {m['score']:6.3f}  "
                f"{m['iou']:.3f}  {m['n_intersection']:5d}  "
                f"{m['lam_ratio']:8.3f}  {m['ang_diff_deg']:12.1f}  "
                f"{m['ang_src_deg']:10.1f}  {m['ang_cand_deg']:10.1f}  "
                f"{m['lam_src_med']:8.1f}  {m['lam_cand_med']:8.1f}\n"
                for m in sorted(all_candidate_rows, key=lambda d: (d["cluster_a"], -d["score"]))
            )
        )

        (slice_dir / f"slice_{slice_no:03d}_BEST_matches_A_to_C_ratio.txt").write_text(
            "cluster_a  best_cluster_c  passes  score    iou    inter  lam_ratio  ang_diff_deg  thetaA_deg  thetaC_deg  lamA_med  lamC_med\n"
            + "".join(
                f"{m['cluster_a']:9d}  {m['cluster_c']:14d}  {int(m['passes']):6d}  {m['score']:6.3f}  "
                f"{m['iou']:.3f}  {m['n_intersection']:5d}  "
                f"{m['lam_ratio']:8.3f}  {m['ang_diff_deg']:12.1f}  "
                f"{m['ang_src_deg']:10.1f}  {m['ang_cand_deg']:10.1f}  "
                f"{m['lam_src_med']:8.1f}  {m['lam_cand_med']:8.1f}\n"
                for m in sorted(best_rows, key=lambda d: d["cluster_a"])
            )
        )

        passed_here = [int(m["cluster_a"]) for m in best_rows if bool(m["passes"])]
        passed_global_indices.extend(passed_here)

        summary_lines.append(
            f"{slice_no:5d}  {idx_a.size:2d}  {idx_c.size:2d}  {len(best_rows):10d}  {len(passed_here):7d}  ok\n"
        )

        del all_candidate_rows, best_rows
        gc.collect()

    summary_path = outdir / "SUMMARY_all_slices.txt"
    summary_path.write_text("".join(summary_lines))

    passed_global_indices = sorted(set(passed_global_indices))
    passed_idx = np.asarray(passed_global_indices, dtype=int)

    if passed_idx.size == 0:
        return dict(
            ok=False,
            l6c=str(l6c_path),
            l6a=str(l6a_path),
            outdir=str(outdir),
            l7a=str(out_l7a),
            n_passed=0,
            summary=str(summary_path),
            note="No passed clusters",
        )

    create_passed_only_l6a_with_new_vars(
        src_l6a=l6a_path,
        dst_l7a=out_l7a,
        passed_cluster_indices=passed_idx,
        time_index=TIME_INDEX,
    )

    if SAVE_STITCHED_PLOTS:
        with Dataset(out_l7a, "r") as nc:
            MF2 = np.asarray(nc.variables["MF_top"][TIME_INDEX], dtype=float)
            ny, nx = MF2.shape
            x1d_km = np.arange(nx, dtype=float) * DX_KM
            y1d_km = np.arange(ny, dtype=float) * DY_KM
            vmin, vmax = robust_limits(MF2, 2, 99.9)

            fig, ax = plt.subplots(figsize=(22, 3), constrained_layout=True)
            im = ax.pcolormesh(x1d_km, y1d_km, MF2, shading="auto", vmin=vmin, vmax=vmax)
            ax.set_title("Momentum Flux (passed clusters summed)")
            ax.set_xlabel("x (km)")
            ax.set_ylabel("y (km)")
            ax.set_ylim(0, 600)
            ax.set_aspect("equal", adjustable="box")
            fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01).set_label("MF (m2 s-2)")
            save_fig(fig, outdir / "MF_stitched.png")

        plot_stitched_mf_components(out_l7a, outdir / "MF_stitched_components.png", TIME_INDEX)

    if SAVE_PER_CLUSTER_PANELS:
        pass

    del l6c, l6a, passed_idx
    mpl.pyplot.close("all")
    gc.collect()

    return dict(
        ok=True,
        l6c=str(l6c_path),
        l6a=str(l6a_path),
        outdir=str(outdir),
        l7a=str(out_l7a),
        n_passed=int(len(passed_global_indices)),
        summary=str(summary_path),
        note="ok",
    )


# ============================================================
# Main batch
# ============================================================
pairs = find_l6_pairs(L6_DIR)
print(f"Found {len(pairs)} L6C/L6A pairs in: {L6_DIR.resolve()}")

results: List[Dict[str, Any]] = []
errors: List[Dict[str, Any]] = []

for l6c_path, l6a_path, key in pairs:
    pair_outdir = BATCH_OUTROOT / key
    print(f"\nRunning pair key={key}")
    print("L6C:", l6c_path.name)
    print("L6A:", l6a_path.name)
    try:
        res = run_one_pair(l6c_path, l6a_path, pair_outdir)
        results.append(res)
        print(f"OK={res['ok']}  n_passed={res.get('n_passed', 'NA')}")
        print("L7A:", res.get("l7a"))
        print("Outdir:", pair_outdir)
    except Exception as e:
        err = dict(
            ok=False,
            l6c=str(l6c_path),
            l6a=str(l6a_path),
            key=key,
            outdir=str(pair_outdir),
            error=str(e),
            traceback=traceback.format_exc(),
        )
        errors.append(err)
        print("FAILED:", e)

summary_txt = BATCH_OUTROOT / "BATCH_SUMMARY.txt"
lines: List[str] = []
lines.append(f"Input dir: {L6_DIR.resolve()}\n")
lines.append(f"Output nc dir (l7): {L7_DIR.resolve()}\n")
lines.append(f"Pairs found: {len(pairs)}\n")
lines.append(f"Success: {sum(1 for r in results if r.get('ok'))}\n")
lines.append(f"Fail: {len(errors)}\n\n")

lines.append("Per pair:\n")
for r in results:
    key = Path(r["outdir"]).name
    lines.append(f"{key}  ok={r['ok']}  n_passed={r.get('n_passed', 0)}  l7a={r.get('l7a')}\n")

if errors:
    lines.append("\nErrors:\n")
    for e in errors:
        lines.append(f"{e['key']}  {e['error']}\n")

summary_txt.write_text("".join(lines))

mpl.pyplot.close("all")
gc.collect()

print("\nBatch done.")
print("Batch outroot:", BATCH_OUTROOT.resolve())
print("L7 output folder:", L7_DIR.resolve())
print("Batch summary:", summary_txt.resolve())
```

## Overall plot. Use mean value binning. 

```python
# %%
"""
Global stacked Cartopy maps for MFz and MFm from L7A outputs
for ALL months in a year.

What this does:
1. Loops over all 12 months in YEAR
2. Reads files from: BASE_L7_DIR / YEAR / MM
3. Makes one stacked figure per month
4. Saves all monthly figures into the SAME folder
5. Uses month/year in titles and filenames
"""

from __future__ import annotations

from pathlib import Path
from typing import List, Tuple, Dict
import calendar
import gc

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
from netCDF4 import Dataset

# -------------------------
# User params
# -------------------------
YEAR = 2025
MONTHS = range(11, 13)   # 1..12

#BASE_L7_DIR = Path("/Volumes/aweread/socfiles/public/l7")
BASE_L7_DIR = Path("Z:\socfiles\public\l7")
FILE_GLOB = "*l7a*.nc"
TIME_INDEX = 0

LAT_NAME = "Lat_top"
LON_NAME = "Lon_top"

STACK_PAIR = ("MFz_top", "MFm_top")

DISPLAY_NAMES = {
    "MF_top": "Momentum Flux",
    "MFz_top": "Zonal Momentum Flux",
    "MFm_top": "Meridional Momentum Flux",
}

# All monthly stacked figures go here
OUTDIR = Path(f"outputs_MF_cartopy/{YEAR}/stacked_all_months")
OUTDIR.mkdir(parents=True, exist_ok=True)

MAX_POINTS_TOTAL = 4_500_000
GRID_DEG = 1.0

CLIP_PLO = 2.0
CLIP_PHI = 99.0

# -------------------------
# Latitude limit
# -------------------------
LAT_MIN = -60.0
LAT_MAX = 60.0

# -------------------------
# LOG scaling
# -------------------------
USE_LOG_SCALE = False
ZERO_MODE = "floor"   # "drop" or "floor"
LOG_FLOOR = 1e-4

# -------------------------
# Cartopy
# -------------------------
try:
    import cartopy.crs as ccrs
    import cartopy.feature as cfeature
except Exception as e:
    raise ImportError(
        "Cartopy is not available in this environment. "
        "If you're in conda: conda install -c conda-forge cartopy"
    ) from e


# -------------------------
# Helpers
# -------------------------
def wrap_lon180(lon: np.ndarray) -> np.ndarray:
    return ((lon + 180.0) % 360.0) - 180.0


def apply_zero_mode(v: np.ndarray) -> np.ndarray:
    out = v.copy()
    bad = np.isfinite(out) & (out <= 0)
    if ZERO_MODE == "drop":
        out[bad] = np.nan
    elif ZERO_MODE == "floor":
        out[bad] = LOG_FLOOR
    else:
        raise ValueError("ZERO_MODE must be 'drop' or 'floor'")
    return out


def robust_clip_linear(v: np.ndarray) -> Tuple[float, float]:
    good = np.isfinite(v)
    if not np.any(good):
        return 0.0, 1.0
    lo, hi = np.nanpercentile(v[good], [CLIP_PLO, CLIP_PHI])
    if (not np.isfinite(lo)) or (not np.isfinite(hi)) or (lo == hi):
        lo = float(np.nanmin(v[good]))
        hi = float(np.nanmax(v[good]))
    return float(lo), float(hi)


def robust_clip_log(v: np.ndarray) -> Tuple[float, float]:
    good = np.isfinite(v) & (v > 0)
    if not np.any(good):
        return LOG_FLOOR, LOG_FLOOR * 10.0
    lo, hi = np.nanpercentile(v[good], [CLIP_PLO, CLIP_PHI])
    lo = max(float(lo), LOG_FLOOR)
    hi = max(float(hi), lo * 10.0)
    return lo, hi


def binned_mean_2d(
    lat: np.ndarray, lon: np.ndarray, mf: np.ndarray, ddeg: float
) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    lon_edges = np.arange(-180.0, 180.0 + ddeg, ddeg)
    lat_edges = np.arange(LAT_MIN, LAT_MAX + ddeg, ddeg)

    nlon = len(lon_edges) - 1
    nlat = len(lat_edges) - 1

    grid_sum = np.zeros((nlat, nlon), dtype=float)
    grid_cnt = np.zeros((nlat, nlon), dtype=np.int64)

    j = np.digitize(lon, lon_edges) - 1
    i = np.digitize(lat, lat_edges) - 1

    good = (
        (i >= 0) & (i < nlat) &
        (j >= 0) & (j < nlon) &
        np.isfinite(mf)
    )

    if not np.any(good):
        return lon_edges, lat_edges, np.full((nlat, nlon), np.nan, dtype=float)

    np.add.at(grid_sum, (i[good], j[good]), mf[good])
    np.add.at(grid_cnt, (i[good], j[good]), 1)

    grid_mean = np.full((nlat, nlon), np.nan, dtype=float)
    mask = grid_cnt > 0
    grid_mean[mask] = grid_sum[mask] / grid_cnt[mask]

    return lon_edges, lat_edges, grid_mean


def collect_component_data(
    l7_dir: Path,
    file_glob: str,
    components: Tuple[str, str],
) -> Dict[str, Dict[str, List[np.ndarray]]]:
    files = sorted(l7_dir.glob(file_glob))
    print(f"Found {len(files)} files in {l7_dir.resolve()} matching {file_glob}")

    data = {
        comp: {"lat": [], "lon": [], "val": []}
        for comp in components
    }

    if len(files) == 0:
        return data

    per_file_cap = max(50_000, MAX_POINTS_TOTAL // max(1, len(files)))

    for p in files:
        with Dataset(p, "r") as nc:
            if LAT_NAME not in nc.variables or LON_NAME not in nc.variables:
                print(f"Skip {p.name}: missing {LAT_NAME} or {LON_NAME}")
                continue

            lat_var = nc.variables[LAT_NAME]
            lon_var = nc.variables[LON_NAME]

            if lat_var.ndim == 3:
                if TIME_INDEX >= lat_var.shape[0] or TIME_INDEX >= lon_var.shape[0]:
                    print(f"Skip {p.name}: TIME_INDEX out of range")
                    continue
                lat = np.asarray(lat_var[TIME_INDEX], dtype=float)
                lon = np.asarray(lon_var[TIME_INDEX], dtype=float)
            elif lat_var.ndim == 2:
                lat = np.asarray(lat_var[:], dtype=float)
                lon = np.asarray(lon_var[:], dtype=float)
            else:
                print(f"Skip {p.name}: unsupported dims for {LAT_NAME}/{LON_NAME}")
                continue

            lon = wrap_lon180(lon)

            base = (
                np.isfinite(lat) &
                np.isfinite(lon) &
                (lat >= LAT_MIN) &
                (lat <= LAT_MAX)
            )

            idx = np.where(base.ravel())[0]
            if idx.size == 0:
                print(f"Skip {p.name}: no valid points in latitude range")
                continue

            if idx.size > per_file_cap:
                rng = np.random.default_rng(0)
                idx = rng.choice(idx, size=per_file_cap, replace=False)

            lat_flat = lat.ravel()[idx]
            lon_flat = lon.ravel()[idx]

            counts_this_file = {}

            for comp in components:
                if comp not in nc.variables:
                    counts_this_file[comp] = 0
                    continue

                var = nc.variables[comp]
                if var.ndim == 3:
                    if TIME_INDEX >= var.shape[0]:
                        counts_this_file[comp] = 0
                        continue
                    val = np.asarray(var[TIME_INDEX], dtype=float).ravel()[idx]
                elif var.ndim == 2:
                    val = np.asarray(var[:], dtype=float).ravel()[idx]
                else:
                    counts_this_file[comp] = 0
                    continue

                good = np.isfinite(val)
                n_good = int(np.count_nonzero(good))
                counts_this_file[comp] = n_good

                if n_good == 0:
                    continue

                data[comp]["lat"].append(lat_flat[good])
                data[comp]["lon"].append(lon_flat[good])
                data[comp]["val"].append(val[good])

            print(f"{p.name}: added points per component: {counts_this_file}")

    return data


def make_monthly_stacked_plot(year: int, month: int):
    month_str = f"{month:02d}"
    month_name = calendar.month_name[month]
    l7_dir = BASE_L7_DIR / str(year) / month_str

    if not l7_dir.exists():
        print(f"Skip {year}-{month_str}: folder does not exist -> {l7_dir}")
        return

    data = collect_component_data(l7_dir, FILE_GLOB, STACK_PAIR)

    mf_top, mf_bot = STACK_PAIR
    if not data[mf_top]["val"] or not data[mf_bot]["val"]:
        print(f"Skip {year}-{month_str}: missing samples for {STACK_PAIR}")
        return

    LAT1 = np.concatenate(data[mf_top]["lat"])
    LON1 = np.concatenate(data[mf_top]["lon"])
    V1 = np.concatenate(data[mf_top]["val"])

    LAT2 = np.concatenate(data[mf_bot]["lat"])
    LON2 = np.concatenate(data[mf_bot]["lon"])
    V2 = np.concatenate(data[mf_bot]["val"])

    if V1.size > MAX_POINTS_TOTAL:
        rng = np.random.default_rng(2)
        sel = rng.choice(np.arange(V1.size), size=MAX_POINTS_TOTAL, replace=False)
        LAT1, LON1, V1 = LAT1[sel], LON1[sel], V1[sel]

    if V2.size > MAX_POINTS_TOTAL:
        rng = np.random.default_rng(3)
        sel = rng.choice(np.arange(V2.size), size=MAX_POINTS_TOTAL, replace=False)
        LAT2, LON2, V2 = LAT2[sel], LON2[sel], V2[sel]

    pretty_top = DISPLAY_NAMES.get(mf_top, mf_top)
    pretty_bot = DISPLAY_NAMES.get(mf_bot, mf_bot)

    if USE_LOG_SCALE:
        V1d = apply_zero_mode(V1)
        V2d = apply_zero_mode(V2)
        vall = np.concatenate([V1d, V2d])
        lo, hi = robust_clip_log(vall)
        norm_shared = LogNorm(vmin=lo, vmax=hi)
        vmin = None
        vmax = None
        grid1 = apply_zero_mode(binned_mean_2d(LAT1, LON1, V1d, GRID_DEG)[2])
        grid2 = apply_zero_mode(binned_mean_2d(LAT2, LON2, V2d, GRID_DEG)[2])
        cbar_label = r"Momentum Flux (m$^2$ s$^{-2}$, log)"
        suptitle = f"{month_name} {year}: {pretty_top} and {pretty_bot} (shared LOG scale)"
    else:
        vall = np.concatenate([V1, V2])
        lo, hi = robust_clip_linear(vall)
        norm_shared = None
        #vmin = lo
        #vmax = hi
        vmin = 0
        vmax = 3
        grid1 = np.clip(binned_mean_2d(LAT1, LON1, V1, GRID_DEG)[2], lo, hi)
        grid2 = np.clip(binned_mean_2d(LAT2, LON2, V2, GRID_DEG)[2], lo, hi)
        cbar_label = r"Momentum Flux (m$^2$ s$^{-2}$)"
        suptitle = f"{month_name} {year}: {pretty_top} and {pretty_bot} (shared linear scale)"

    lon_e1, lat_e1, _ = binned_mean_2d(LAT1, LON1, V1 if not USE_LOG_SCALE else V1d, GRID_DEG)
    lon_e2, lat_e2, _ = binned_mean_2d(LAT2, LON2, V2 if not USE_LOG_SCALE else V2d, GRID_DEG)

    outpath = OUTDIR / f"{year}_{month_str}_MFz_MFm_stacked_global_maps.png"

    proj = ccrs.PlateCarree()
    fig = plt.figure(figsize=(14, 10), constrained_layout=True)
    gs = fig.add_gridspec(2, 1)

    ax1 = fig.add_subplot(gs[0, 0], projection=proj)
    ax1.set_extent([-180, 180, LAT_MIN, LAT_MAX], crs=proj)
    ax1.coastlines(linewidth=0.6)
    ax1.add_feature(cfeature.BORDERS, linewidth=0.3)
    ax1.gridlines(draw_labels=False, linewidth=0.2, alpha=0.5)
    im1 = ax1.pcolormesh(
        lon_e1, lat_e1, grid1,
        transform=proj,
        shading="auto",
        norm=norm_shared,
        vmin=vmin,
        vmax=vmax,
    )
    ax1.set_title(pretty_top)

    ax2 = fig.add_subplot(gs[1, 0], projection=proj)
    ax2.set_extent([-180, 180, LAT_MIN, LAT_MAX], crs=proj)
    ax2.coastlines(linewidth=0.6)
    ax2.add_feature(cfeature.BORDERS, linewidth=0.3)
    ax2.gridlines(draw_labels=False, linewidth=0.2, alpha=0.5)
    im2 = ax2.pcolormesh(
        lon_e2, lat_e2, grid2,
        transform=proj,
        shading="auto",
        norm=norm_shared,
        vmin=vmin,
        vmax=vmax,
    )
    ax2.set_title(pretty_bot)

    cb = fig.colorbar(im2, ax=[ax1, ax2], fraction=0.03, pad=0.02)
    cb.set_label(cbar_label)

    fig.suptitle(suptitle, fontsize=14)
    fig.savefig(outpath, dpi=200)
    plt.close(fig)

    print(f"Saved: {outpath.resolve()}")
    gc.collect()


# -------------------------
# Run all months
# -------------------------
for month in MONTHS:
    print("\n" + "=" * 80)
    print(f"Processing {YEAR}-{month:02d}")
    print("=" * 80)
    make_monthly_stacked_plot(YEAR, month)

print("\nDone.")
print(f"All monthly stacked figures are in: {OUTDIR.resolve()}")
```

## DOY vs Lat plots for MF

```python
# %%
"""
DOY vs latitude plots for MFz_top and MFm_top from L7A outputs
for all days in 2024.

Memory strategy:
1. Find all L7A files in all months of YEAR
2. Group files by DOY
3. Process one DOY at a time
4. For each file, copy it to a temporary local folder before opening
5. For that DOY, accumulate latitude-bin sums/counts
6. Write the DOY result into the final grid
7. Release that day's arrays before moving on
"""

from __future__ import annotations

from pathlib import Path
from collections import defaultdict
import re
import gc
import shutil
import tempfile

import numpy as np
import matplotlib.pyplot as plt
from netCDF4 import Dataset


# -------------------------
# User params
# -------------------------
YEAR = 2024
BASE_L7_DIR = Path("/Volumes/aweread/socfiles/public/l7")
FILE_GLOB = "*l7a*.nc"
TIME_INDEX = 0

LAT_NAME = "Lat_top"
COMPONENTS = ("MFz_top", "MFm_top")

DISPLAY_NAMES = {
    "MFz_top": "Zonal Momentum Flux",
    "MFm_top": "Meridional Momentum Flux",
}

OUTDIR = Path(f"outputs_MF_doy_latitude/{YEAR}")
OUTDIR.mkdir(parents=True, exist_ok=True)

# Latitude binning
LAT_MIN = -60.0
LAT_MAX = 60.0
LAT_BIN_DEG = 1.0

# Plot scaling
USE_SYMMETRIC_PERCENTILE = True
PCTL = 99.0   # use symmetric percentile for color limits

# Filename DOY pattern like: awe_l7a_tmp_2024005T0917_00688_v01.nc
DOY_RE = re.compile(rf"{YEAR}(\d{{3}})T\d{{4}}")


# -------------------------
# Helpers
# -------------------------
def extract_doy_from_name(path: Path) -> int | None:
    m = DOY_RE.search(path.name)
    if m is None:
        return None
    return int(m.group(1))


def get_var_2d(nc: Dataset, var_name: str, time_index: int = 0) -> np.ndarray | None:
    if var_name not in nc.variables:
        return None

    var = nc.variables[var_name]

    try:
        var.set_auto_mask(False)
    except Exception:
        pass

    if var.ndim == 3:
        if time_index >= var.shape[0]:
            return None
        arr = np.array(var[time_index, :, :], dtype=np.float64, copy=True)
    elif var.ndim == 2:
        arr = np.array(var[:, :], dtype=np.float64, copy=True)
    else:
        return None

    return arr


def robust_symmetric_limit(arr: np.ndarray, percentile: float = 99.0) -> float:
    good = np.isfinite(arr)
    if not np.any(good):
        return 1.0
    vmax = np.nanpercentile(np.abs(arr[good]), percentile)
    if not np.isfinite(vmax) or vmax == 0:
        vmax = np.nanmax(np.abs(arr[good]))
    if not np.isfinite(vmax) or vmax == 0:
        vmax = 1.0
    return float(vmax)


def copy_to_temp(src_path: Path, temp_dir: Path) -> Path:
    local_path = temp_dir / src_path.name
    shutil.copy2(src_path, local_path)
    return local_path


# -------------------------
# Build file list grouped by DOY
# -------------------------
files_by_doy = defaultdict(list)

for month in range(1, 13):
    month_dir = BASE_L7_DIR / str(YEAR) / f"{month:02d}"
    if not month_dir.exists():
        print(f"Skip missing folder: {month_dir}")
        continue

    for p in sorted(month_dir.glob(FILE_GLOB)):
        doy = extract_doy_from_name(p)
        if doy is None:
            print(f"Could not parse DOY from filename, skip: {p.name}")
            continue
        files_by_doy[doy].append(p)

all_doys = sorted(files_by_doy.keys())

if len(all_doys) == 0:
    raise FileNotFoundError(f"No L7A files found for year {YEAR} in {BASE_L7_DIR}")

print(f"Found {sum(len(v) for v in files_by_doy.values())} files across {len(all_doys)} DOYs")


# -------------------------
# Final grids: DOY x latitude
# -------------------------
lat_edges = np.arange(LAT_MIN, LAT_MAX + LAT_BIN_DEG, LAT_BIN_DEG)
lat_centers = 0.5 * (lat_edges[:-1] + lat_edges[1:])
nlat = len(lat_centers)

# Full year axis
ndoy = 366 if ((YEAR % 4 == 0 and YEAR % 100 != 0) or (YEAR % 400 == 0)) else 365
doy_axis = np.arange(1, ndoy + 1)

final_grids = {
    comp: np.full((ndoy, nlat), np.nan, dtype=np.float32)
    for comp in COMPONENTS
}


# -------------------------
# Process one DOY at a time
# -------------------------
for i_doy, doy in enumerate(all_doys, start=1):
    doy_files = files_by_doy[doy]
    print("\n" + "=" * 80)
    print(f"Processing DOY {doy:03d}  ({i_doy}/{len(all_doys)})  with {len(doy_files)} files")
    print("=" * 80)

    # Per-DOY accumulators
    doy_sum = {comp: np.zeros(nlat, dtype=np.float64) for comp in COMPONENTS}
    doy_cnt = {comp: np.zeros(nlat, dtype=np.int64) for comp in COMPONENTS}

    with tempfile.TemporaryDirectory(prefix=f"l7a_{YEAR}_{doy:03d}_") as tmpdir:
        tmpdir = Path(tmpdir)

        for p in doy_files:
            local_p = None
            lat = None
            arr = None

            try:
                print(f"  Copying {p.name} to temp...")
                local_p = copy_to_temp(p, tmpdir)

                with Dataset(local_p, "r") as nc:
                    try:
                        nc.set_auto_mask(False)
                    except Exception:
                        pass
                    try:
                        nc.set_auto_maskandscale(False)
                    except Exception:
                        pass

                    lat = get_var_2d(nc, LAT_NAME, TIME_INDEX)
                    if lat is None:
                        print(f"  Skip {p.name}: missing or invalid {LAT_NAME}")
                        continue

                    lat_flat = lat.ravel()
                    base = np.isfinite(lat_flat) & (lat_flat >= LAT_MIN) & (lat_flat <= LAT_MAX)

                    if not np.any(base):
                        print(f"  Skip {p.name}: no valid latitude points")
                        continue

                    lat_use = lat_flat[base]
                    lat_idx = np.digitize(lat_use, lat_edges) - 1
                    good_lat = (lat_idx >= 0) & (lat_idx < nlat)

                    if not np.any(good_lat):
                        print(f"  Skip {p.name}: no valid latitude bins")
                        continue

                    lat_idx = lat_idx[good_lat]

                    for comp in COMPONENTS:
                        arr = get_var_2d(nc, comp, TIME_INDEX)
                        if arr is None:
                            print(f"  {p.name}: missing {comp}")
                            continue

                        val = arr.ravel()[base][good_lat]
                        good = np.isfinite(val)

                        if not np.any(good):
                            continue

                        np.add.at(doy_sum[comp], lat_idx[good], val[good])
                        np.add.at(doy_cnt[comp], lat_idx[good], 1)

                print(f"  Finished {p.name}")

            except Exception as e:
                print(f"  Error reading {p.name}: {e}")
                continue

            finally:
                del lat, arr
                gc.collect()

    # Write this DOY into final grid
    doy_row = doy - 1
    for comp in COMPONENTS:
        mask = doy_cnt[comp] > 0
        if np.any(mask):
            final_grids[comp][doy_row, mask] = (doy_sum[comp][mask] / doy_cnt[comp][mask]).astype(np.float32)

    # Release memory for this DOY
    del doy_sum, doy_cnt, doy_files
    gc.collect()


# -------------------------
# Plot helper
# -------------------------
def make_doy_lat_plot(grid: np.ndarray, comp: str, outname: str):
    fig, ax = plt.subplots(figsize=(16, 6))

    if USE_SYMMETRIC_PERCENTILE:
        vmax = robust_symmetric_limit(grid, PCTL)
        vmin = -vmax
    else:
        good = np.isfinite(grid)
        if np.any(good):
            vmin = float(np.nanmin(grid))
            vmax = float(np.nanmax(grid))
            if vmin == vmax:
                vmin, vmax = -1.0, 1.0
        else:
            vmin, vmax = -1.0, 1.0

    # transpose so y = latitude, x = DOY
    im = ax.pcolormesh(
        doy_axis,
        lat_edges,
        grid.T,
        shading="auto",
        cmap="RdBu_r",
        vmin=vmin,
        vmax=vmax,
    )

    ax.set_title(f"{DISPLAY_NAMES.get(comp, comp)} in {YEAR}")
    ax.set_xlabel("Day of Year")
    ax.set_ylabel("Latitude (deg)")
    ax.set_xlim(1, ndoy)
    ax.set_ylim(LAT_MIN, LAT_MAX)

    cbar = fig.colorbar(im, ax=ax, pad=0.02)
    cbar.set_label(r"Momentum Flux (m$^2$ s$^{-2}$)")

    outpath = OUTDIR / outname
    fig.tight_layout()
    fig.savefig(outpath, dpi=200)
    plt.close(fig)

    print(f"Saved: {outpath.resolve()}")


# -------------------------
# Make plots
# -------------------------
make_doy_lat_plot(
    final_grids["MFz_top"],
    "MFz_top",
    f"{YEAR}_MFz_top_DOY_vs_latitude.png",
)

make_doy_lat_plot(
    final_grids["MFm_top"],
    "MFm_top",
    f"{YEAR}_MFm_top_DOY_vs_latitude.png",
)


# -------------------------
# Optional: save arrays too
# -------------------------
np.savez_compressed(
    OUTDIR / f"{YEAR}_MFz_MFm_DOY_latitude_grids.npz",
    doy_axis=doy_axis,
    lat_edges=lat_edges,
    lat_centers=lat_centers,
    MFz_top=final_grids["MFz_top"],
    MFm_top=final_grids["MFm_top"],
)

print("\nDone.")
print(f"Outputs saved in: {OUTDIR.resolve()}")
```

## Test day 5

```python
# %%
from pathlib import Path
import gc
import numpy as np
from netCDF4 import Dataset

p = Path("/Volumes/aweread/socfiles/public/l7/2024/01/awe_l7a_tmp_2024005T0917_00688_v01.nc")
TIME_INDEX = 0

LAT_NAME = "Lat_top"
STACK_PAIR = ("MFz_top", "MFm_top")

LAT_MIN = -60.0
LAT_MAX = 60.0
LAT_BIN_DEG = 1.0

USE_SUBSAMPLING = True
MAX_POINTS_PER_ROW = 2000
SUBSAMPLE_SEED = 0

lat_edges = np.arange(LAT_MIN, LAT_MAX + LAT_BIN_DEG, LAT_BIN_DEG)
nlat = len(lat_edges) - 1


def read_row(var, row_idx, time_index):
    if var.ndim == 3:
        return np.asarray(var[time_index, row_idx, :], dtype=float)
    elif var.ndim == 2:
        return np.asarray(var[row_idx, :], dtype=float)
    return None


def maybe_subsample_row(lat_row, val_row, max_points, seed):
    n = lat_row.size
    if n <= max_points:
        return lat_row, val_row
    rng = np.random.default_rng(seed)
    idx = rng.choice(n, size=max_points, replace=False)
    return lat_row[idx], val_row[idx]


def accumulate_row_into_profile(lat_row, val_row, lat_edges, prof_sum, prof_cnt):
    if lat_row.size == 0:
        return 0
    bin_idx = np.digitize(lat_row, lat_edges) - 1
    good = (
        np.isfinite(lat_row)
        & np.isfinite(val_row)
        & (bin_idx >= 0)
        & (bin_idx < len(prof_sum))
    )
    if not np.any(good):
        return 0
    bin_idx = bin_idx[good]
    val_good = val_row[good]
    np.add.at(prof_sum, bin_idx, val_good)
    np.add.at(prof_cnt, bin_idx, 1)
    return int(val_good.size)


print("Testing file:", p)

with Dataset(p, "r") as nc:
    lat_var = nc.variables[LAT_NAME]
    ny = lat_var.shape[1] if lat_var.ndim == 3 else lat_var.shape[0]
    print("ny =", ny)

    for icomp, comp in enumerate(STACK_PAIR):
        print(f"\n=== Component: {comp} ===")
        val_var = nc.variables[comp]

        prof_sum = np.zeros(nlat, dtype=np.float64)
        prof_cnt = np.zeros(nlat, dtype=np.int64)

        for iy in range(ny):
            if iy % 25 == 0:
                print(f"  row {iy}/{ny-1}")

            lat_row = read_row(lat_var, iy, TIME_INDEX)
            val_row = read_row(val_var, iy, TIME_INDEX)

            good = (
                np.isfinite(lat_row)
                & np.isfinite(val_row)
                & (lat_row >= LAT_MIN)
                & (lat_row <= LAT_MAX)
            )

            if np.any(good):
                lat_good = lat_row[good]
                val_good = val_row[good]

                if USE_SUBSAMPLING and lat_good.size > MAX_POINTS_PER_ROW:
                    seed = SUBSAMPLE_SEED + iy * 10 + icomp
                    lat_good, val_good = maybe_subsample_row(
                        lat_good, val_good, MAX_POINTS_PER_ROW, seed
                    )

                accumulate_row_into_profile(
                    lat_good, val_good, lat_edges, prof_sum, prof_cnt
                )

            del lat_row, val_row, good
            gc.collect()

        print(f"Finished component {comp}")
```

## Andes Region

```python
#!/usr/bin/env python3
"""
June 2024 Andes zoom maps for AWE momentum flux components.

Reads L7A outputs from:
  l7_062024/

Plots:
  Meridional Momentum Flux
  Zonal Momentum Flux

Main output:
  outputs_L7A_MF_cartopy_andes/side_by_side/
  Meridional_Zonal_Momentum_Flux_June_2024_andes_maps.png

Colorbar:
  Diverging RdBu_r
  Symmetric around zero
  Shared between panels
"""

from __future__ import annotations

from pathlib import Path
from typing import List, Tuple, Dict, Optional
import gc

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
from netCDF4 import Dataset


# ==================================================
# User params
# ==================================================
L7_DIR = Path("l7_062024")
FILE_GLOB = "*l7a*.nc"
TIME_INDEX = 0

LAT_NAME: Optional[str] = None
LON_NAME: Optional[str] = None

LAT_CANDIDATES = [
    "Lat_top", "Latitude_top", "lat_top",
    "Lat", "Latitude", "lat", "LAT",
    "Lat_flat", "Latitude_flat", "lat_flat",
]

LON_CANDIDATES = [
    "Lon_top", "Longitude_top", "lon_top",
    "Lon", "Longitude", "lon", "LON",
    "Lon_flat", "Longitude_flat", "lon_flat",
]

# Only plot these two components
MF_COMPONENTS = ["MFm_top", "MFz_top"]

PLOT_NAMES = {
    "MFm_top": "Meridional Momentum Flux",
    "MFz_top": "Zonal Momentum Flux",
}

# Side-by-side order
STACK_PAIR = ("MFm_top", "MFz_top")

OUTDIR = Path("outputs_L7A_MF_cartopy_andes")
OUTDIR.mkdir(parents=True, exist_ok=True)

MAX_POINTS_TOTAL = 4_500_000
GRID_DEG = 0.5

CLIP_PLO = 2.0
CLIP_PHI = 98.0


# ==================================================
# Andes zoom region
# ==================================================
ANDES_LON_MIN = -92.0
ANDES_LON_MAX = -30.0
ANDES_LAT_MIN = -60.0
ANDES_LAT_MAX = 0.0


# ==================================================
# Colorbar / colormap settings
# ==================================================
USE_LOG_SCALE = False

ZERO_MODE = "floor"
LOG_FLOOR = 1e-4

# Match MF component plots: diverging, centered around zero
CMAP = "RdBu_r"


# ==================================================
# Cartopy import
# ==================================================
try:
    import cartopy.crs as ccrs
    import cartopy.feature as cfeature
except Exception as e:
    raise ImportError(
        "Cartopy is not available. "
        "If using conda, install with: conda install -c conda-forge cartopy"
    ) from e


# ==================================================
# Helpers
# ==================================================
def wrap_lon180(lon_deg: np.ndarray) -> np.ndarray:
    return ((lon_deg + 180.0) % 360.0) - 180.0


def apply_zero_mode(v: np.ndarray, zero_mode: str, floor: float) -> np.ndarray:
    out = v.copy()
    bad = np.isfinite(out) & (out <= 0)

    if zero_mode == "drop":
        out[bad] = np.nan
    elif zero_mode == "floor":
        out[bad] = floor
    else:
        raise ValueError(f"Unknown ZERO_MODE={zero_mode!r}. Use 'drop' or 'floor'.")

    return out


def robust_clip_linear(v: np.ndarray, plo: float, phi: float) -> Tuple[float, float]:
    good = np.isfinite(v)
    if not np.any(good):
        return 0.0, 1.0

    lo, hi = np.nanpercentile(v[good], [plo, phi])

    if (not np.isfinite(lo)) or (not np.isfinite(hi)) or lo == hi:
        lo = float(np.nanmin(v[good]))
        hi = float(np.nanmax(v[good]))

    return float(lo), float(hi)


def robust_symmetric_clip(v: np.ndarray, phi: float) -> Tuple[float, float]:
    good = np.isfinite(v)

    if not np.any(good):
        return -1.0, 1.0

    p = np.nanpercentile(np.abs(v[good]), phi)

    if (not np.isfinite(p)) or p == 0:
        p = float(np.nanmax(np.abs(v[good])))

    if (not np.isfinite(p)) or p == 0:
        p = 1.0

    return -float(p), float(p)


def robust_clip_log(v: np.ndarray, plo: float, phi: float, floor: float) -> Tuple[float, float]:
    good = np.isfinite(v) & (v > 0)

    if not np.any(good):
        return floor, floor * 10.0

    lo, hi = np.nanpercentile(v[good], [plo, phi])
    lo = max(float(lo), float(floor))
    hi = max(float(hi), lo * 10.0)

    return lo, hi


def pick_varname(
    nc: Dataset,
    preferred: Optional[str],
    candidates: List[str],
) -> Optional[str]:
    if preferred is not None:
        return preferred if preferred in nc.variables else None

    for name in candidates:
        if name in nc.variables:
            return name

    return None


def binned_mean_2d_region(
    lat: np.ndarray,
    lon: np.ndarray,
    mf: np.ndarray,
    ddeg: float,
) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    lon0 = np.floor(ANDES_LON_MIN / ddeg) * ddeg
    lon1 = np.ceil(ANDES_LON_MAX / ddeg) * ddeg
    lat0 = np.floor(ANDES_LAT_MIN / ddeg) * ddeg
    lat1 = np.ceil(ANDES_LAT_MAX / ddeg) * ddeg

    lon_edges = np.arange(lon0, lon1 + ddeg, ddeg)
    lat_edges = np.arange(lat0, lat1 + ddeg, ddeg)

    nlon = lon_edges.size - 1
    nlat = lat_edges.size - 1

    grid_sum = np.zeros((nlat, nlon), dtype=np.float64)
    grid_cnt = np.zeros((nlat, nlon), dtype=np.int64)

    in_region = (
        np.isfinite(lat)
        & np.isfinite(lon)
        & np.isfinite(mf)
        & (lon >= ANDES_LON_MIN)
        & (lon <= ANDES_LON_MAX)
        & (lat >= ANDES_LAT_MIN)
        & (lat <= ANDES_LAT_MAX)
    )

    if not np.any(in_region):
        return lon_edges, lat_edges, np.full((nlat, nlon), np.nan, dtype=float)

    lat_r = lat[in_region]
    lon_r = lon[in_region]
    mf_r = mf[in_region].astype(np.float64, copy=False)

    j = np.digitize(lon_r, lon_edges) - 1
    i = np.digitize(lat_r, lat_edges) - 1

    good = (i >= 0) & (i < nlat) & (j >= 0) & (j < nlon)

    if not np.any(good):
        return lon_edges, lat_edges, np.full((nlat, nlon), np.nan, dtype=float)

    i = i[good]
    j = j[good]
    v = mf_r[good]

    np.add.at(grid_sum, (i, j), v)
    np.add.at(grid_cnt, (i, j), 1)

    grid_mean = np.full((nlat, nlon), np.nan, dtype=np.float64)
    mask = grid_cnt > 0
    grid_mean[mask] = grid_sum[mask] / grid_cnt[mask]

    return lon_edges, lat_edges, grid_mean.astype(float)


def plot_individual_andes_map(
    lon_edges: np.ndarray,
    lat_edges: np.ndarray,
    grid_disp: np.ndarray,
    title: str,
    cbar_label: str,
    outpath: Path,
    use_log: bool,
    norm: LogNorm | None,
    vmin: float | None,
    vmax: float | None,
) -> None:
    proj = ccrs.PlateCarree()

    fig = plt.figure(figsize=(12, 8), constrained_layout=True)
    ax = plt.axes(projection=proj)

    ax.set_extent(
        [ANDES_LON_MIN, ANDES_LON_MAX, ANDES_LAT_MIN, ANDES_LAT_MAX],
        crs=ccrs.PlateCarree(),
    )

    ax.coastlines(linewidth=0.9, color="black")
    ax.add_feature(cfeature.BORDERS, linewidth=0.5, edgecolor="black")

    gl = ax.gridlines(draw_labels=True, linewidth=0.25, alpha=0.6)
    gl.top_labels = False
    gl.right_labels = False

    im = ax.pcolormesh(
        lon_edges,
        lat_edges,
        grid_disp,
        transform=ccrs.PlateCarree(),
        shading="auto",
        cmap=CMAP,
        norm=norm,
        vmin=None if use_log else vmin,
        vmax=None if use_log else vmax,
    )

    cb = plt.colorbar(im, ax=ax, fraction=0.035, pad=0.02)
    cb.set_label(cbar_label)

    ax.set_title(title)

    fig.savefig(outpath, dpi=200)
    plt.close(fig)


# ==================================================
# Checks
# ==================================================
if USE_LOG_SCALE:
    if ZERO_MODE not in ("drop", "floor"):
        raise ValueError("ZERO_MODE must be 'drop' or 'floor'")

    if ZERO_MODE == "floor" and (not np.isfinite(LOG_FLOOR) or LOG_FLOOR <= 0):
        raise ValueError("LOG_FLOOR must be finite and positive when ZERO_MODE='floor'")


# ==================================================
# Load and sample data
# ==================================================
files = sorted(L7_DIR.glob(FILE_GLOB))
print(f"Found {len(files)} files in {L7_DIR.resolve()} matching {FILE_GLOB}")

if len(files) == 0:
    raise RuntimeError("No L7A files found.")

per_file_cap = max(50_000, MAX_POINTS_TOTAL // max(1, len(files)))

lat_all: Dict[str, List[np.ndarray]] = {name: [] for name in MF_COMPONENTS}
lon_all: Dict[str, List[np.ndarray]] = {name: [] for name in MF_COMPONENTS}
mf_all: Dict[str, List[np.ndarray]] = {name: [] for name in MF_COMPONENTS}

picked_latlon_once = False

for p in files:
    with Dataset(p, "r") as nc:
        lat_name = pick_varname(nc, LAT_NAME, LAT_CANDIDATES)
        lon_name = pick_varname(nc, LON_NAME, LON_CANDIDATES)

        if lat_name is None or lon_name is None:
            print(
                f"Skip {p.name}: could not find lat/lon variables. "
                f"Available vars example: {list(nc.variables.keys())[:25]}"
            )
            continue

        if not picked_latlon_once:
            print(f"Using lat var: {lat_name} | lon var: {lon_name}")
            picked_latlon_once = True

        lat = np.asarray(nc.variables[lat_name][TIME_INDEX], dtype=float)
        lon = np.asarray(nc.variables[lon_name][TIME_INDEX], dtype=float)
        lon = wrap_lon180(lon)

        base_good = (
            np.isfinite(lat)
            & np.isfinite(lon)
            & (lon >= ANDES_LON_MIN)
            & (lon <= ANDES_LON_MAX)
            & (lat >= ANDES_LAT_MIN)
            & (lat <= ANDES_LAT_MAX)
        )

        base_idx = np.where(base_good.ravel())[0]

        if base_idx.size == 0:
            print(f"Skip {p.name}: no valid lat/lon points inside Andes box")
            continue

        if base_idx.size > per_file_cap:
            rng = np.random.default_rng(0)
            base_idx = rng.choice(base_idx, size=per_file_cap, replace=False)

        la = lat.ravel()[base_idx]
        lo = lon.ravel()[base_idx]

        any_used = False
        counts_this_file = {}

        for mf_name in MF_COMPONENTS:
            if mf_name not in nc.variables:
                counts_this_file[mf_name] = 0
                continue

            mf_raw = np.asarray(
                nc.variables[mf_name][TIME_INDEX],
                dtype=float,
            ).ravel()[base_idx]

            good = np.isfinite(mf_raw)
            n = int(np.count_nonzero(good))
            counts_this_file[mf_name] = n

            if n == 0:
                continue

            lat_all[mf_name].append(la[good])
            lon_all[mf_name].append(lo[good])
            mf_all[mf_name].append(mf_raw[good])

            any_used = True

        if any_used:
            print(f"{p.name}: added points per component: {counts_this_file}")
        else:
            print(f"{p.name}: no valid MF points for requested components")


# ==================================================
# Individual plots
# ==================================================
for mf_name in MF_COMPONENTS:
    if not mf_all[mf_name]:
        print(f"\nNo samples for {mf_name}. Skipping plot.")
        continue

    LAT = np.concatenate(lat_all[mf_name])
    LON = np.concatenate(lon_all[mf_name])
    MF_raw = np.concatenate(mf_all[mf_name])

    if MF_raw.size > MAX_POINTS_TOTAL:
        rng = np.random.default_rng(1)
        idx = rng.choice(np.arange(MF_raw.size), size=MAX_POINTS_TOTAL, replace=False)
        LAT, LON, MF_raw = LAT[idx], LON[idx], MF_raw[idx]

    if USE_LOG_SCALE:
        MF = apply_zero_mode(MF_raw, ZERO_MODE, LOG_FLOOR)
        mf_lo, mf_hi = robust_clip_log(MF, CLIP_PLO, CLIP_PHI, LOG_FLOOR)
        norm = LogNorm(vmin=mf_lo, vmax=mf_hi)
        vmin = vmax = None
        cbar_label = f"{PLOT_NAMES[mf_name]} (m² s⁻², log)"
    else:
        MF = MF_raw
        mf_lo, mf_hi = robust_symmetric_clip(MF, CLIP_PHI)
        norm = None
        vmin, vmax = mf_lo, mf_hi
        cbar_label = "(m²/s²)"

    lon_edges, lat_edges, grid_mean = binned_mean_2d_region(LAT, LON, MF, GRID_DEG)

    if USE_LOG_SCALE:
        grid_disp = apply_zero_mode(grid_mean, ZERO_MODE, LOG_FLOOR)
    else:
        grid_disp = np.clip(grid_mean, mf_lo, mf_hi)

    out_comp = OUTDIR / mf_name
    out_comp.mkdir(parents=True, exist_ok=True)

    title = (
        f"{PLOT_NAMES[mf_name]}\n"
        f"Andes Region, June 2024 Monthly Mean"
    )

    plot_individual_andes_map(
        lon_edges=lon_edges,
        lat_edges=lat_edges,
        grid_disp=grid_disp,
        title=title,
        cbar_label=cbar_label,
        outpath=out_comp / f"{mf_name}_June_2024_andes_map.png",
        use_log=USE_LOG_SCALE,
        norm=norm,
        vmin=vmin,
        vmax=vmax,
    )


# ==================================================
# Side-by-side Meridional + Zonal plot
# ==================================================
mf_left, mf_right = STACK_PAIR

if mf_all.get(mf_left) and mf_all.get(mf_right):
    LAT1 = np.concatenate(lat_all[mf_left])
    LON1 = np.concatenate(lon_all[mf_left])
    V1 = np.concatenate(mf_all[mf_left])

    LAT2 = np.concatenate(lat_all[mf_right])
    LON2 = np.concatenate(lon_all[mf_right])
    V2 = np.concatenate(mf_all[mf_right])

    if V1.size > MAX_POINTS_TOTAL:
        rng = np.random.default_rng(2)
        idx = rng.choice(np.arange(V1.size), size=MAX_POINTS_TOTAL, replace=False)
        LAT1, LON1, V1 = LAT1[idx], LON1[idx], V1[idx]

    if V2.size > MAX_POINTS_TOTAL:
        rng = np.random.default_rng(3)
        idx = rng.choice(np.arange(V2.size), size=MAX_POINTS_TOTAL, replace=False)
        LAT2, LON2, V2 = LAT2[idx], LON2[idx], V2[idx]

    if USE_LOG_SCALE:
        V1d = apply_zero_mode(V1, ZERO_MODE, LOG_FLOOR)
        V2d = apply_zero_mode(V2, ZERO_MODE, LOG_FLOOR)

        Vall = np.concatenate([V1d, V2d])
        v_lo, v_hi = robust_clip_log(Vall, CLIP_PLO, CLIP_PHI, LOG_FLOOR)

        norm_shared = LogNorm(vmin=v_lo, vmax=v_hi)
        vmin = vmax = None
        cbar_label = "Momentum Flux (m² s⁻², log)"

        grid1_vals = V1d
        grid2_vals = V2d

    else:
        Vall = np.concatenate([V1, V2])
        v_lo, v_hi = robust_symmetric_clip(Vall, CLIP_PHI)

        norm_shared = None
        vmin, vmax = v_lo, v_hi
        cbar_label = "m²/s²"

        grid1_vals = V1
        grid2_vals = V2

    lon_edges1, lat_edges1, grid1 = binned_mean_2d_region(
        LAT1, LON1, grid1_vals, GRID_DEG
    )

    lon_edges2, lat_edges2, grid2 = binned_mean_2d_region(
        LAT2, LON2, grid2_vals, GRID_DEG
    )

    if USE_LOG_SCALE:
        grid1_disp = apply_zero_mode(grid1, ZERO_MODE, LOG_FLOOR)
        grid2_disp = apply_zero_mode(grid2, ZERO_MODE, LOG_FLOOR)
    else:
        grid1_disp = np.clip(grid1, v_lo, v_hi)
        grid2_disp = np.clip(grid2, v_lo, v_hi)

    side_dir = OUTDIR / "side_by_side"
    side_dir.mkdir(parents=True, exist_ok=True)

    outpath = side_dir / "Meridional_Zonal_Momentum_Flux_June_2024_andes_maps.png"

    proj = ccrs.PlateCarree()

    fig = plt.figure(figsize=(12, 6.5), constrained_layout=True)
    gs = fig.add_gridspec(1, 2)

    ax1 = fig.add_subplot(gs[0, 0], projection=proj)
    ax1.set_extent(
        [ANDES_LON_MIN, ANDES_LON_MAX, ANDES_LAT_MIN, ANDES_LAT_MAX],
        crs=ccrs.PlateCarree(),
    )
    ax1.coastlines(linewidth=0.9, color="black")
    ax1.add_feature(cfeature.BORDERS, linewidth=0.5, edgecolor="black")

    gl1 = ax1.gridlines(draw_labels=True, linewidth=0.25, alpha=0.6)
    gl1.top_labels = False
    gl1.right_labels = False

    im1 = ax1.pcolormesh(
        lon_edges1,
        lat_edges1,
        grid1_disp,
        transform=ccrs.PlateCarree(),
        shading="auto",
        cmap=CMAP,
        norm=norm_shared,
        vmin=None if USE_LOG_SCALE else vmin,
        vmax=None if USE_LOG_SCALE else vmax,
    )

    ax1.set_title(PLOT_NAMES[mf_left])

    ax2 = fig.add_subplot(gs[0, 1], projection=proj)
    ax2.set_extent(
        [ANDES_LON_MIN, ANDES_LON_MAX, ANDES_LAT_MIN, ANDES_LAT_MAX],
        crs=ccrs.PlateCarree(),
    )
    ax2.coastlines(linewidth=0.9, color="black")
    ax2.add_feature(cfeature.BORDERS, linewidth=0.5, edgecolor="black")

    gl2 = ax2.gridlines(draw_labels=True, linewidth=0.25, alpha=0.6)
    gl2.top_labels = False
    gl2.right_labels = False
    gl2.left_labels = False

    im2 = ax2.pcolormesh(
        lon_edges2,
        lat_edges2,
        grid2_disp,
        transform=ccrs.PlateCarree(),
        shading="auto",
        cmap=CMAP,
        norm=norm_shared,
        vmin=None if USE_LOG_SCALE else vmin,
        vmax=None if USE_LOG_SCALE else vmax,
    )

    ax2.set_title(PLOT_NAMES[mf_right])

    cb = fig.colorbar(
        im2,
        ax=[ax1, ax2],
        fraction=0.022,
        pad=0.015,
        aspect=35,
    )
    cb.set_label(cbar_label)

    fig.suptitle(
        "Gravity Wave Momentum Flux Components Over the Andes\n"
        "June 2024 Monthly Mean",
        fontsize=14,
    )

    fig.savefig(outpath, dpi=200)
    plt.close(fig)

    print("\nSaved side-by-side Andes plot to:", outpath.resolve())

else:
    print(f"\nSkipping side-by-side plot: missing samples for {STACK_PAIR}")


gc.collect()
print("\nSaved Andes plots to:", OUTDIR.resolve())
```

## Top-bottom

```python
#!/usr/bin/env python3
"""
June 2024 Andes zoom maps for AWE momentum flux components.

Plots:
  Top:    Meridional Momentum Flux
  Bottom: Zonal Momentum Flux

Colorbar:
  RdBu_r
  Symmetric around zero
  Shared between panels
"""

from __future__ import annotations

from pathlib import Path
from typing import List, Tuple, Dict, Optional
import gc

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
from netCDF4 import Dataset


# ==================================================
# User params
# ==================================================
L7_DIR = Path("l7_062024")
FILE_GLOB = "*l7a*.nc"
TIME_INDEX = 0

LAT_NAME: Optional[str] = None
LON_NAME: Optional[str] = None

LAT_CANDIDATES = [
    "Lat_top", "Latitude_top", "lat_top",
    "Lat", "Latitude", "lat", "LAT",
    "Lat_flat", "Latitude_flat", "lat_flat",
]

LON_CANDIDATES = [
    "Lon_top", "Longitude_top", "lon_top",
    "Lon", "Longitude", "lon", "LON",
    "Lon_flat", "Longitude_flat", "lon_flat",
]

MF_COMPONENTS = ["MFm_top", "MFz_top"]

PLOT_NAMES = {
    "MFm_top": "Meridional Momentum Flux",
    "MFz_top": "Zonal Momentum Flux",
}

# Top-bottom order
STACK_PAIR = ("MFm_top", "MFz_top")

OUTDIR = Path("outputs_L7A_MF_cartopy_andes")
OUTDIR.mkdir(parents=True, exist_ok=True)

MAX_POINTS_TOTAL = 4_500_000
GRID_DEG = 0.5

CLIP_PLO = 2.0
CLIP_PHI = 98.0


# ==================================================
# Andes zoom region
# ==================================================
ANDES_LON_MIN = -92.0
ANDES_LON_MAX = -30.0
ANDES_LAT_MIN = -57.0
ANDES_LAT_MAX = -12


# ==================================================
# Colorbar / colormap settings
# ==================================================
USE_LOG_SCALE = False

ZERO_MODE = "floor"
LOG_FLOOR = 1e-4

CMAP = "RdBu_r"


# ==================================================
# Cartopy import
# ==================================================
try:
    import cartopy.crs as ccrs
    import cartopy.feature as cfeature
except Exception as e:
    raise ImportError(
        "Cartopy is not available. "
        "If using conda, install with: conda install -c conda-forge cartopy"
    ) from e


# ==================================================
# Helpers
# ==================================================
def wrap_lon180(lon_deg: np.ndarray) -> np.ndarray:
    return ((lon_deg + 180.0) % 360.0) - 180.0


def apply_zero_mode(v: np.ndarray, zero_mode: str, floor: float) -> np.ndarray:
    out = v.copy()
    bad = np.isfinite(out) & (out <= 0)

    if zero_mode == "drop":
        out[bad] = np.nan
    elif zero_mode == "floor":
        out[bad] = floor
    else:
        raise ValueError(f"Unknown ZERO_MODE={zero_mode!r}. Use 'drop' or 'floor'.")

    return out


def robust_symmetric_clip(v: np.ndarray, phi: float) -> Tuple[float, float]:
    good = np.isfinite(v)

    if not np.any(good):
        return -1.0, 1.0

    p = np.nanpercentile(np.abs(v[good]), phi)

    if (not np.isfinite(p)) or p == 0:
        p = float(np.nanmax(np.abs(v[good])))

    if (not np.isfinite(p)) or p == 0:
        p = 1.0

    return -float(p), float(p)


def robust_clip_log(v: np.ndarray, plo: float, phi: float, floor: float) -> Tuple[float, float]:
    good = np.isfinite(v) & (v > 0)

    if not np.any(good):
        return floor, floor * 10.0

    lo, hi = np.nanpercentile(v[good], [plo, phi])
    lo = max(float(lo), float(floor))
    hi = max(float(hi), lo * 10.0)

    return lo, hi


def pick_varname(
    nc: Dataset,
    preferred: Optional[str],
    candidates: List[str],
) -> Optional[str]:
    if preferred is not None:
        return preferred if preferred in nc.variables else None

    for name in candidates:
        if name in nc.variables:
            return name

    return None


def binned_mean_2d_region(
    lat: np.ndarray,
    lon: np.ndarray,
    mf: np.ndarray,
    ddeg: float,
) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    lon0 = np.floor(ANDES_LON_MIN / ddeg) * ddeg
    lon1 = np.ceil(ANDES_LON_MAX / ddeg) * ddeg
    lat0 = np.floor(ANDES_LAT_MIN / ddeg) * ddeg
    lat1 = np.ceil(ANDES_LAT_MAX / ddeg) * ddeg

    lon_edges = np.arange(lon0, lon1 + ddeg, ddeg)
    lat_edges = np.arange(lat0, lat1 + ddeg, ddeg)

    nlon = lon_edges.size - 1
    nlat = lat_edges.size - 1

    grid_sum = np.zeros((nlat, nlon), dtype=np.float64)
    grid_cnt = np.zeros((nlat, nlon), dtype=np.int64)

    in_region = (
        np.isfinite(lat)
        & np.isfinite(lon)
        & np.isfinite(mf)
        & (lon >= ANDES_LON_MIN)
        & (lon <= ANDES_LON_MAX)
        & (lat >= ANDES_LAT_MIN)
        & (lat <= ANDES_LAT_MAX)
    )

    if not np.any(in_region):
        return lon_edges, lat_edges, np.full((nlat, nlon), np.nan, dtype=float)

    lat_r = lat[in_region]
    lon_r = lon[in_region]
    mf_r = mf[in_region].astype(np.float64, copy=False)

    j = np.digitize(lon_r, lon_edges) - 1
    i = np.digitize(lat_r, lat_edges) - 1

    good = (i >= 0) & (i < nlat) & (j >= 0) & (j < nlon)

    if not np.any(good):
        return lon_edges, lat_edges, np.full((nlat, nlon), np.nan, dtype=float)

    i = i[good]
    j = j[good]
    v = mf_r[good]

    np.add.at(grid_sum, (i, j), v)
    np.add.at(grid_cnt, (i, j), 1)

    grid_mean = np.full((nlat, nlon), np.nan, dtype=np.float64)
    mask = grid_cnt > 0
    grid_mean[mask] = grid_sum[mask] / grid_cnt[mask]

    return lon_edges, lat_edges, grid_mean.astype(float)


def plot_individual_andes_map(
    lon_edges: np.ndarray,
    lat_edges: np.ndarray,
    grid_disp: np.ndarray,
    title: str,
    cbar_label: str,
    outpath: Path,
    use_log: bool,
    norm: LogNorm | None,
    vmin: float | None,
    vmax: float | None,
) -> None:
    proj = ccrs.PlateCarree()

    fig = plt.figure(figsize=(12, 8), constrained_layout=True)
    ax = plt.axes(projection=proj)

    ax.set_extent(
        [ANDES_LON_MIN, ANDES_LON_MAX, ANDES_LAT_MIN, ANDES_LAT_MAX],
        crs=ccrs.PlateCarree(),
    )

    ax.coastlines(linewidth=0.9, color="black")
    ax.add_feature(cfeature.BORDERS, linewidth=0.5, edgecolor="black")

    gl = ax.gridlines(draw_labels=True, linewidth=0.25, alpha=0.6)
    gl.top_labels = False
    gl.right_labels = False

    im = ax.pcolormesh(
        lon_edges,
        lat_edges,
        grid_disp,
        transform=ccrs.PlateCarree(),
        shading="auto",
        cmap=CMAP,
        norm=norm,
        vmin=None if use_log else vmin,
        vmax=None if use_log else vmax,
    )

    cb = plt.colorbar(im, ax=ax, fraction=0.035, pad=0.02)
    cb.set_label(cbar_label)

    ax.set_title(title)

    fig.savefig(outpath, dpi=200)
    plt.close(fig)


# ==================================================
# Checks
# ==================================================
if USE_LOG_SCALE:
    if ZERO_MODE not in ("drop", "floor"):
        raise ValueError("ZERO_MODE must be 'drop' or 'floor'")

    if ZERO_MODE == "floor" and (not np.isfinite(LOG_FLOOR) or LOG_FLOOR <= 0):
        raise ValueError("LOG_FLOOR must be finite and positive when ZERO_MODE='floor'")


# ==================================================
# Load and sample data
# ==================================================
files = sorted(L7_DIR.glob(FILE_GLOB))
print(f"Found {len(files)} files in {L7_DIR.resolve()} matching {FILE_GLOB}")

if len(files) == 0:
    raise RuntimeError("No L7A files found.")

per_file_cap = max(50_000, MAX_POINTS_TOTAL // max(1, len(files)))

lat_all: Dict[str, List[np.ndarray]] = {name: [] for name in MF_COMPONENTS}
lon_all: Dict[str, List[np.ndarray]] = {name: [] for name in MF_COMPONENTS}
mf_all: Dict[str, List[np.ndarray]] = {name: [] for name in MF_COMPONENTS}

picked_latlon_once = False

for p in files:
    with Dataset(p, "r") as nc:
        lat_name = pick_varname(nc, LAT_NAME, LAT_CANDIDATES)
        lon_name = pick_varname(nc, LON_NAME, LON_CANDIDATES)

        if lat_name is None or lon_name is None:
            print(
                f"Skip {p.name}: could not find lat/lon variables. "
                f"Available vars example: {list(nc.variables.keys())[:25]}"
            )
            continue

        if not picked_latlon_once:
            print(f"Using lat var: {lat_name} | lon var: {lon_name}")
            picked_latlon_once = True

        lat = np.asarray(nc.variables[lat_name][TIME_INDEX], dtype=float)
        lon = np.asarray(nc.variables[lon_name][TIME_INDEX], dtype=float)
        lon = wrap_lon180(lon)

        base_good = (
            np.isfinite(lat)
            & np.isfinite(lon)
            & (lon >= ANDES_LON_MIN)
            & (lon <= ANDES_LON_MAX)
            & (lat >= ANDES_LAT_MIN)
            & (lat <= ANDES_LAT_MAX)
        )

        base_idx = np.where(base_good.ravel())[0]

        if base_idx.size == 0:
            print(f"Skip {p.name}: no valid lat/lon points inside Andes box")
            continue

        if base_idx.size > per_file_cap:
            rng = np.random.default_rng(0)
            base_idx = rng.choice(base_idx, size=per_file_cap, replace=False)

        la = lat.ravel()[base_idx]
        lo = lon.ravel()[base_idx]

        any_used = False
        counts_this_file = {}

        for mf_name in MF_COMPONENTS:
            if mf_name not in nc.variables:
                counts_this_file[mf_name] = 0
                continue

            mf_raw = np.asarray(
                nc.variables[mf_name][TIME_INDEX],
                dtype=float,
            ).ravel()[base_idx]

            good = np.isfinite(mf_raw)
            n = int(np.count_nonzero(good))
            counts_this_file[mf_name] = n

            if n == 0:
                continue

            lat_all[mf_name].append(la[good])
            lon_all[mf_name].append(lo[good])
            mf_all[mf_name].append(mf_raw[good])

            any_used = True

        if any_used:
            print(f"{p.name}: added points per component: {counts_this_file}")
        else:
            print(f"{p.name}: no valid MF points for requested components")


# ==================================================
# Individual plots
# ==================================================
for mf_name in MF_COMPONENTS:
    if not mf_all[mf_name]:
        print(f"\nNo samples for {mf_name}. Skipping plot.")
        continue

    LAT = np.concatenate(lat_all[mf_name])
    LON = np.concatenate(lon_all[mf_name])
    MF_raw = np.concatenate(mf_all[mf_name])

    if MF_raw.size > MAX_POINTS_TOTAL:
        rng = np.random.default_rng(1)
        idx = rng.choice(np.arange(MF_raw.size), size=MAX_POINTS_TOTAL, replace=False)
        LAT, LON, MF_raw = LAT[idx], LON[idx], MF_raw[idx]

    if USE_LOG_SCALE:
        MF = apply_zero_mode(MF_raw, ZERO_MODE, LOG_FLOOR)
        mf_lo, mf_hi = robust_clip_log(MF, CLIP_PLO, CLIP_PHI, LOG_FLOOR)
        norm = LogNorm(vmin=mf_lo, vmax=mf_hi)
        vmin = vmax = None
        cbar_label = f"{PLOT_NAMES[mf_name]} (m² s⁻², log)"
    else:
        MF = MF_raw
        mf_lo, mf_hi = robust_symmetric_clip(MF, CLIP_PHI)
        norm = None
        vmin, vmax = mf_lo, mf_hi
        cbar_label = "m²/s²"

    lon_edges, lat_edges, grid_mean = binned_mean_2d_region(LAT, LON, MF, GRID_DEG)

    if USE_LOG_SCALE:
        grid_disp = apply_zero_mode(grid_mean, ZERO_MODE, LOG_FLOOR)
    else:
        grid_disp = np.clip(grid_mean, mf_lo, mf_hi)

    out_comp = OUTDIR / mf_name
    out_comp.mkdir(parents=True, exist_ok=True)

    title = (
        f"{PLOT_NAMES[mf_name]}\n"
        f"Andes Region, June 2024 Monthly Mean"
    )

    plot_individual_andes_map(
        lon_edges=lon_edges,
        lat_edges=lat_edges,
        grid_disp=grid_disp,
        title=title,
        cbar_label=cbar_label,
        outpath=out_comp / f"{mf_name}_June_2024_andes_map.png",
        use_log=USE_LOG_SCALE,
        norm=norm,
        vmin=vmin,
        vmax=vmax,
    )


# ==================================================
# Top-bottom Meridional + Zonal plot
# ==================================================
mf_top, mf_bottom = STACK_PAIR

if mf_all.get(mf_top) and mf_all.get(mf_bottom):
    LAT1 = np.concatenate(lat_all[mf_top])
    LON1 = np.concatenate(lon_all[mf_top])
    V1 = np.concatenate(mf_all[mf_top])

    LAT2 = np.concatenate(lat_all[mf_bottom])
    LON2 = np.concatenate(lon_all[mf_bottom])
    V2 = np.concatenate(mf_all[mf_bottom])

    if V1.size > MAX_POINTS_TOTAL:
        rng = np.random.default_rng(2)
        idx = rng.choice(np.arange(V1.size), size=MAX_POINTS_TOTAL, replace=False)
        LAT1, LON1, V1 = LAT1[idx], LON1[idx], V1[idx]

    if V2.size > MAX_POINTS_TOTAL:
        rng = np.random.default_rng(3)
        idx = rng.choice(np.arange(V2.size), size=MAX_POINTS_TOTAL, replace=False)
        LAT2, LON2, V2 = LAT2[idx], LON2[idx], V2[idx]

    if USE_LOG_SCALE:
        V1d = apply_zero_mode(V1, ZERO_MODE, LOG_FLOOR)
        V2d = apply_zero_mode(V2, ZERO_MODE, LOG_FLOOR)

        Vall = np.concatenate([V1d, V2d])
        v_lo, v_hi = robust_clip_log(Vall, CLIP_PLO, CLIP_PHI, LOG_FLOOR)

        norm_shared = LogNorm(vmin=v_lo, vmax=v_hi)
        vmin = vmax = None
        cbar_label = "Momentum Flux (m² s⁻², log)"

        grid1_vals = V1d
        grid2_vals = V2d

    else:
        Vall = np.concatenate([V1, V2])
        v_lo, v_hi = robust_symmetric_clip(Vall, CLIP_PHI)

        norm_shared = None
        vmin, vmax = v_lo, v_hi
        cbar_label = "m²/s²"

        grid1_vals = V1
        grid2_vals = V2

    lon_edges1, lat_edges1, grid1 = binned_mean_2d_region(
        LAT1, LON1, grid1_vals, GRID_DEG
    )

    lon_edges2, lat_edges2, grid2 = binned_mean_2d_region(
        LAT2, LON2, grid2_vals, GRID_DEG
    )

    if USE_LOG_SCALE:
        grid1_disp = apply_zero_mode(grid1, ZERO_MODE, LOG_FLOOR)
        grid2_disp = apply_zero_mode(grid2, ZERO_MODE, LOG_FLOOR)
    else:
        grid1_disp = np.clip(grid1, v_lo, v_hi)
        grid2_disp = np.clip(grid2, v_lo, v_hi)

    stacked_dir = OUTDIR / "stacked"
    stacked_dir.mkdir(parents=True, exist_ok=True)

    outpath = stacked_dir / "Meridional_Zonal_Momentum_Flux_June_2024_andes_maps_stacked.png"

    proj = ccrs.PlateCarree()

    fig = plt.figure(figsize=(7, 8.5), constrained_layout=True)
    gs = fig.add_gridspec(2, 1)

    ax1 = fig.add_subplot(gs[0, 0], projection=proj)
    ax1.set_extent(
        [ANDES_LON_MIN, ANDES_LON_MAX, ANDES_LAT_MIN, ANDES_LAT_MAX],
        crs=ccrs.PlateCarree(),
    )
    ax1.coastlines(linewidth=0.9, color="black")
    ax1.add_feature(cfeature.BORDERS, linewidth=0.5, edgecolor="black")

    gl1 = ax1.gridlines(draw_labels=True, linewidth=0.25, alpha=0.6)
    gl1.top_labels = False
    gl1.right_labels = False
    gl1.bottom_labels = False

    im1 = ax1.pcolormesh(
        lon_edges1,
        lat_edges1,
        grid1_disp,
        transform=ccrs.PlateCarree(),
        shading="auto",
        cmap=CMAP,
        norm=norm_shared,
        vmin=None if USE_LOG_SCALE else vmin,
        vmax=None if USE_LOG_SCALE else vmax,
    )

    ax1.set_title(PLOT_NAMES[mf_top])

    ax2 = fig.add_subplot(gs[1, 0], projection=proj)
    ax2.set_extent(
        [ANDES_LON_MIN, ANDES_LON_MAX, ANDES_LAT_MIN, ANDES_LAT_MAX],
        crs=ccrs.PlateCarree(),
    )
    ax2.coastlines(linewidth=0.9, color="black")
    ax2.add_feature(cfeature.BORDERS, linewidth=0.5, edgecolor="black")

    gl2 = ax2.gridlines(draw_labels=True, linewidth=0.25, alpha=0.6)
    gl2.top_labels = False
    gl2.right_labels = False

    im2 = ax2.pcolormesh(
        lon_edges2,
        lat_edges2,
        grid2_disp,
        transform=ccrs.PlateCarree(),
        shading="auto",
        cmap=CMAP,
        norm=norm_shared,
        vmin=None if USE_LOG_SCALE else vmin,
        vmax=None if USE_LOG_SCALE else vmax,
    )

    ax2.set_title(PLOT_NAMES[mf_bottom])

    cb = fig.colorbar(
        im2,
        ax=[ax1, ax2],
        fraction=0.022,
        pad=0.015,
        aspect=35,
    )
    cb.set_label(cbar_label)

    fig.suptitle(
        "Gravity Wave Momentum Flux Components Over the Andes\n"
        "June 2024 Monthly Mean",
        fontsize=14,
    )

    fig.savefig(outpath, dpi=200)
    plt.close(fig)

    print("\nSaved stacked Andes plot to:", outpath.resolve())

else:
    print(f"\nSkipping stacked plot: missing samples for {STACK_PAIR}")


gc.collect()
print("\nSaved Andes plots to:", OUTDIR.resolve())
```
