## One way of slicing, applying noise filtering.

```python
# %%
from __future__ import annotations

from pathlib import Path
from typing import Tuple, Dict, Any, List

import copy
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
from netCDF4 import Dataset

from juwavelet import transform, utils, plot_utils

try:
    from scipy import stats
except Exception:
    stats = None

# --------------------------
# User params
# --------------------------
FILE_PATH = "awe_l5c_q20_2024165T2120_03175_v01.nc"
FRAME_IDX = 0

# Slicing on the UNBINNED flattened field
X_CHUNK = 600
Y_SLICE = slice(None, None)
SLICE_PAD = 50  # extend each slice by this many UNBINNED pixels on each side

# Choose ONE slicing offset (no comparisons)
X_OFFSET = 0  # set to 0 or 300 or whatever you want

# Optional preprocessing
REMOVE_TREND = True
STANDARDIZE = False
GAUSS_BLUR_SIGMA = 0.0

# Optional 2x2 binning BEFORE processing
APPLY_BINNING = True
BIN_FACTOR = 2

# Edge taper on remaining field (before CWT)
APPLY_TAPER = True
TAPER_EDGE_PIXELS = 10  # in BINNED pixels if APPLY_BINNING else in original pixels

# Wavelet params
S0 = 16
DJ = 1 / 8
JS = 24
JT = 18
ASPECT = 1

# Pixel spacing (km per pixel)
DX_KM_BASE = 2.0
DY_KM_BASE = 2.0

# How many clusters to show (largest by POWER)
N_SHOW = 6

# Per cluster maps
AMP_MIN_FRACTION_OF_CLUSTER_MAX = 0.10
QUIVER_STEP_X = 20
QUIVER_STEP_Y = 20
QUIVER_SCALE = 1

# Red noise style filtering (robust WPS thresholding per scale)
APPLY_RED_NOISE_FILTER = True

# Optional diagnostics output from the red noise filter
SAVE_REDNOISE_DIAGNOSTICS = True

# Save settings
OUTPUT_ROOT = Path("outputs_decompose2d_slices_single_noise_filtering_q20_save")
SAVE_DPI = 200

# --------------------------
# Parameter runs
# --------------------------
RUNS: List[Dict[str, Any]] = [
    dict(
        name="base",
        MIN_AMP=0.2,
        THR=0.1,
        WHITE_NOISE_THRESHOLD=0.02, # Decreased from .03
        WPS_NOISE_THRESHOLD=8.0,
    ),
]

# --------------------------
# Plot style
# --------------------------
matplotlib.rcParams.update(
    {
        "axes.labelsize": 16,
        "font.size": 16,
        "xtick.labelsize": 14,
        "ytick.labelsize": 14,
        "legend.fontsize": 14,
        "figure.figsize": [2 * 6.94, 2 * 4.29],
    }
)

# --------------------------
# Small helpers
# --------------------------
def save_fig(fig: plt.Figure, outpath: Path) -> None:
    outpath.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(outpath, dpi=SAVE_DPI, bbox_inches="tight")
    plt.close(fig)


def _fmt(v: float) -> str:
    if not np.isfinite(v):
        return "∞"
    v = float(v)
    if abs(v) >= 1000 or abs(v) < 0.1:
        return f"{v:.2e}"
    if abs(v) >= 100:
        return f"{v:.0f}"
    return f"{v:.1f}"


def diverging_limits(*arrays: np.ndarray, p_lo: float = 2, p_hi: float = 98) -> Tuple[float, float]:
    vals = []
    for A in arrays:
        if A is None:
            continue
        a = np.asarray(A)
        good = np.isfinite(a)
        if np.any(good):
            vals.append(a[good].ravel())
    if not vals:
        return (-1.0, 1.0)
    stacked = np.concatenate(vals)
    lo, hi = np.nanpercentile(stacked, [p_lo, p_hi])
    vmax = max(abs(lo), abs(hi))
    if not np.isfinite(vmax) or vmax == 0:
        vmax = 1.0
    return (-vmax, vmax)


# --------------------------
# Data helpers
# --------------------------
def read_radiance_frame(path: str, frame_idx: int) -> np.ndarray:
    with Dataset(path, "r") as nc:
        rad = nc.variables["Radiance"][:]  # (t, y, x)
    if rad.ndim != 3:
        raise ValueError(f"Temperature expected 3D (t,y,x); got {rad.shape}")
    if not (0 <= frame_idx < rad.shape[0]):
        raise IndexError(f"FRAME_IDX={frame_idx} out of range 0..{rad.shape[0]-1}")
    return np.asarray(rad[frame_idx, :, :], dtype=float)


def flatten_swath(Z: np.ndarray, target_ny: int = 300) -> np.ndarray:
    ny, nx = Z.shape
    out = np.full_like(Z, np.nan)

    bottoms = np.full(nx, np.nan, dtype=float)
    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size:
            bottoms[j] = valid_idx.max()

    valid_bottoms = bottoms[np.isfinite(bottoms)]
    if valid_bottoms.size == 0:
        med = float(np.nanmedian(Z)) if np.isfinite(np.nanmedian(Z)) else 0.0
        out = np.nan_to_num(Z, nan=med, posinf=med, neginf=med)
        if target_ny is not None and ny > target_ny:
            out = out[-target_ny:, :]
        return out

    max_bottom = int(valid_bottoms.max())

    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size == 0:
            continue
        bottom = int(valid_idx.max())
        shift = max_bottom - bottom
        if shift == 0:
            out[:, j] = col
        elif shift > 0:
            out[shift:, j] = col[: ny - shift]
        else:
            shift_up = -shift
            out[: ny - shift_up, j] = col[shift_up:]

    med = float(np.nanmedian(out)) if np.isfinite(np.nanmedian(out)) else 0.0
    out = np.nan_to_num(out, nan=med, posinf=med, neginf=med)

    if target_ny is not None and ny > target_ny:
        start = max_bottom + 1 - target_ny
        start = max(start, 0)
        end = start + target_ny
        out = out[start:end, :]

    return out


def bin2d_block_mean(Z: np.ndarray, factor: int) -> np.ndarray:
    if factor is None or int(factor) <= 1:
        return Z
    f = int(factor)
    ny, nx = Z.shape
    ny2 = (ny // f) * f
    nx2 = (nx // f) * f
    if ny2 == 0 or nx2 == 0:
        raise ValueError(f"Binning factor {f} too large for array {Z.shape}")
    Z2 = Z[:ny2, :nx2]
    return Z2.reshape(ny2 // f, f, nx2 // f, f).mean(axis=(1, 3))

def unbin_repeat_xy(A: np.ndarray, factor: int) -> np.ndarray:
    """
    Nearest-neighbor upsample (repeat) from (y,x) to (y*factor, x*factor).
    Preserves NaNs by repetition.
    """
    if factor is None or int(factor) <= 1:
        return A
    f = int(factor)
    return np.repeat(np.repeat(A, f, axis=0), f, axis=1)

def fit_plane(Z: np.ndarray) -> np.ndarray:
    ny, nx = Z.shape
    yy, xx = np.mgrid[0:ny, 0:nx]
    G = np.stack([xx.ravel(), yy.ravel(), np.ones(nx * ny)], axis=1)
    coeff, *_ = np.linalg.lstsq(G, Z.ravel(), rcond=None)
    return coeff[0] * xx + coeff[1] * yy + coeff[2]


def extract_first_fourier_bg(Z: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
    ny, nx = Z.shape
    F = np.fft.fft2(Z)
    F_bg = np.zeros_like(F, dtype=complex)

    if nx > 1:
        F_bg[0, 1] = F[0, 1]
        F_bg[0, -1] = F[0, -1]
    if ny > 1:
        F_bg[1, 0] = F[1, 0]
        F_bg[-1, 0] = F[-1, 0]

    bg = np.fft.ifft2(F_bg).real
    return bg, Z - bg


def maybe_blur(Z: np.ndarray, sigma: float) -> np.ndarray:
    if sigma <= 0:
        return Z
    try:
        from scipy.ndimage import gaussian_filter
    except Exception:
        return Z
    return gaussian_filter(Z, sigma=float(sigma))


def _hanning_edge_1d(n: int, edge: int) -> np.ndarray:
    w = np.ones(n, dtype=float)
    if edge <= 0:
        return w
    edge = min(edge, n // 2)
    ramp = 0.5 * (1.0 - np.cos(np.linspace(0, np.pi, edge, endpoint=True)))
    w[:edge] = ramp
    w[-edge:] = ramp[::-1]
    return w


def apply_edge_hanning_taper(Z: np.ndarray, edge: int) -> np.ndarray:
    if edge <= 0:
        return Z
    ny, nx = Z.shape
    wx = _hanning_edge_1d(nx, edge)
    wy = _hanning_edge_1d(ny, edge)
    W = np.outer(wy, wx)
    return Z * W


def effective_lambda(cwt: dict, s_idx: int, t_idx: int) -> Tuple[float, float, float]:
    lamx = float(cwt["wavelength_x"][s_idx, t_idx])
    lamy = float(cwt["wavelength_y"][s_idx, t_idx])
    inv2 = 0.0
    if np.isfinite(lamx) and lamx != 0:
        inv2 += (1.0 / lamx) ** 2
    if np.isfinite(lamy) and lamy != 0:
        inv2 += (1.0 / lamy) ** 2
    lam = np.inf if inv2 == 0 else 1.0 / np.sqrt(inv2)
    return lamx, lamy, lam


def compute_cluster_power_table(cwt: dict, iwave: np.ndarray) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
    dec = cwt["decomposition"]
    power4d = np.abs(dec) ** 2

    cluster_ids = np.unique(iwave[iwave >= 0]).astype(int)
    if cluster_ids.size == 0:
        return cluster_ids, np.array([]), np.array([])

    P_abs = np.zeros(cluster_ids.size, dtype=float)
    for i, kcl in enumerate(cluster_ids):
        P_abs[i] = float(np.sum(power4d[iwave == kcl]))

    total = float(np.sum(P_abs))
    if total > 0 and np.isfinite(total):
        P_pct = 100.0 * P_abs / total
    else:
        P_pct = np.full_like(P_abs, np.nan)

    return cluster_ids, P_abs, P_pct


def cluster_amp_theta_maps(
    cwt: dict,
    iwave: np.ndarray,
    kcl: int,
    amp_min_fraction_of_cluster_max: float = 0.2,
) -> Tuple[np.ndarray, np.ndarray]:
    dec = cwt["decomposition"]  # (ns, nt, nx, ny) complex
    power = np.abs(dec) ** 2

    mask4d = (iwave == kcl)
    if not np.any(mask4d):
        ny = dec.shape[3]
        nx = dec.shape[2]
        return np.full((ny, nx), np.nan), np.full((ny, nx), np.nan)

    ns, nt, nx, ny = power.shape
    power2 = power.reshape(ns * nt, nx * ny)
    mask2 = mask4d.reshape(ns * nt, nx * ny)

    p = np.where(mask2, power2, -np.inf)
    pmax = np.max(p, axis=0)

    valid = np.isfinite(pmax) & (pmax > 0)

    A = np.full(nx * ny, np.nan, dtype=float)
    theta = np.full(nx * ny, np.nan, dtype=float)

    if np.any(valid):
        arg = np.argmax(p[:, valid], axis=0)
        t_idx = arg % nt

        A_valid = np.sqrt(pmax[valid])
        theta_valid = np.asarray(cwt["theta"], dtype=float)[t_idx]

        A[valid] = A_valid
        theta[valid] = theta_valid

    A_xy = A.reshape(nx, ny).T
    theta_xy = theta.reshape(nx, ny).T

    if np.isfinite(np.nanmax(A_xy)):
        thr = float(amp_min_fraction_of_cluster_max) * float(np.nanmax(A_xy))
        low = A_xy < thr
        A_xy[low] = np.nan
        theta_xy[low] = np.nan

    return A_xy, theta_xy


def theta_to_uv(theta_xy: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
    return np.cos(theta_xy), np.sin(theta_xy)


# --------------------------
# Red noise style filtering helper
# --------------------------
def apply_red_noise_filter_cwt(
    cwt: dict,
    white_noise_threshold: float = 0.1,
    wps_noise_threshold: float = 8.0,
) -> Tuple[dict, Dict[str, Any]]:
    """
    Robust per scale WPS filtering.

    Returns filtered_cwt (deep copy) and a small diagnostics dict.
    """
    diag: Dict[str, Any] = {}
    if stats is None:
        diag["skipped"] = True
        diag["reason"] = "scipy not available"
        return cwt, diag

    cwt_copy = copy.deepcopy(cwt)
    dec = cwt_copy["decomposition"]  # (ns, nt, nx, ny) complex

    WPS0 = np.abs(dec) ** 2
    diag["n_coeff_total"] = int(WPS0.size)

    mask_1 = WPS0 < (float(white_noise_threshold) ** 2)
    dec[mask_1] = 0

    WPS1 = np.abs(dec) ** 2
    diag["n_zero_after_white"] = int(np.count_nonzero(WPS1 == 0))

    median_WPS = np.median(WPS1, axis=(1, 2, 3))  # (ns,)
    mad = stats.median_abs_deviation(WPS1, axis=(1, 2, 3), scale=1.0)  # (ns,)
    sMAD_WPS = 1.4826 * mad

    sMAD_safe = np.where(sMAD_WPS > 0, sMAD_WPS, np.nan)
    WPS_scaled = (WPS1 - median_WPS[:, None, None, None]) / sMAD_safe[:, None, None, None]

    mask = WPS_scaled < float(wps_noise_threshold)
    dec[mask] = 0

    WPS2 = np.abs(dec) ** 2
    diag["n_zero_after_scaled"] = int(np.count_nonzero(WPS2 == 0))

    finite_scaled = np.isfinite(WPS_scaled)

    diag["sMAD_min"] = float(np.nanmin(sMAD_WPS))
    diag["sMAD_median"] = float(np.nanmedian(sMAD_WPS))

    if np.any(finite_scaled):
        diag["WPS_scaled_min"] = float(np.nanmin(WPS_scaled[finite_scaled]))
        diag["WPS_scaled_median"] = float(np.nanmedian(WPS_scaled[finite_scaled]))
        diag["WPS_scaled_max"] = float(np.nanmax(WPS_scaled[finite_scaled]))
    else:
        diag["WPS_scaled_min"] = np.nan
        diag["WPS_scaled_median"] = np.nan
        diag["WPS_scaled_max"] = np.nan

    return cwt_copy, diag


# --------------------------
# Quicklook plots
# --------------------------
def quicklook_save(x1d: np.ndarray, y1d: np.ndarray, Z: np.ndarray, title: str, outpath: Path) -> None:
    fig, ax = plt.subplots()
    im = ax.pcolormesh(x1d, y1d, Z, shading="auto", cmap="gray")
    fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04).set_label("Radiance (arb)")
    ax.set_xlabel("x (km)")
    ax.set_ylabel("y (km)")
    ax.set_title(title)
    ax.set_aspect("equal", adjustable="box")
    fig.tight_layout()
    save_fig(fig, outpath)


def quicklook_background_decomp_4panel_save(
    x1d_km: np.ndarray,
    y1d_km: np.ndarray,
    original: np.ndarray,
    background: np.ndarray,
    remaining: np.ndarray,
    remaining_tapered: np.ndarray,
    taper_edge_px: int,
    outpath: Path,
    wavy_after_noise_xy: np.ndarray | None = None,
) -> None:
    med = np.nanmedian(original)
    orig_c = original - med
    bg_c = background - med
    rem_c = remaining
    rem_tapered_c = remaining_tapered

    arrays = [orig_c, bg_c, rem_c, rem_tapered_c]
    if wavy_after_noise_xy is not None:
        arrays.append(wavy_after_noise_xy)

    vmin, vmax = diverging_limits(*arrays, p_lo=1, p_hi=99)

    fig, axs = plt.subplots(2, 3, figsize=(16, 10), constrained_layout=True)
    axs = axs.flatten()

    panels = [
        (orig_c, "Original (median sub)"),
        (bg_c, "Background (plane + first Fourier)"),
        (rem_c, "Remaining = original - background"),
        (rem_tapered_c, f"Remaining with edge taper (width={taper_edge_px})"),
    ]

    if wavy_after_noise_xy is not None:
        panels.append((wavy_after_noise_xy, "Wavy stuff (wavelet reconstruction after noise filtering)"))

    im0 = None
    for ax, (Z, title) in zip(axs, panels):
        im0 = ax.pcolormesh(
            x1d_km, y1d_km, Z,
            shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax
        )
        ax.set_title(title)
        ax.set_xlabel("x (km)")
        ax.set_ylabel("y (km)")
        ax.set_aspect("equal", adjustable="box")

    for ax in axs[len(panels):]:
        ax.axis("off")

    cbar = fig.colorbar(im0, ax=axs[:len(panels)], orientation="vertical", fraction=0.04, pad=0.02)
    cbar.set_label("Radiance (arb)")
    save_fig(fig, outpath)


def plot_cluster_individual_save(
    cwt: dict,
    iwave: np.ndarray,
    kcl: int,
    x1d_km: np.ndarray,
    y1d_km: np.ndarray,
    orig_decomp: np.ndarray,
    bg_xy: np.ndarray,
    dx_km: float,
    dy_km: float,
    outpath: Path,
) -> None:
    decomposition = cwt["decomposition"]
    decomposition[:] = orig_decomp
    decomposition[iwave != kcl] = 0

    rec = transform.reconstruct2d(cwt)  # (nx, ny)
    rec_xy = rec.T - np.median(rec)

    A_xy, theta_xy = cluster_amp_theta_maps(
        cwt, iwave, kcl, amp_min_fraction_of_cluster_max=AMP_MIN_FRACTION_OF_CLUSTER_MAX
    )

    dec0 = orig_decomp
    power0 = np.abs(dec0) ** 2
    mask4d = (iwave == kcl)
    ns, nt, nx, ny = power0.shape

    power2 = power0.reshape(ns * nt, nx * ny)
    mask2 = mask4d.reshape(ns * nt, nx * ny)
    p = np.where(mask2, power2, -np.inf)
    pmax = np.max(p, axis=0)

    lam_xy = np.full(nx * ny, np.nan, dtype=float)
    valid = np.isfinite(pmax) & (pmax > 0)
    if np.any(valid):
        arg = np.argmax(p[:, valid], axis=0)
        s_idx = arg // nt
        t_idx = arg % nt

        lamx = np.asarray(cwt["wavelength_x"], dtype=float)[s_idx, t_idx]
        lamy = np.asarray(cwt["wavelength_y"], dtype=float)[s_idx, t_idx]

        inv2 = np.zeros_like(lamx, dtype=float)
        goodx = np.isfinite(lamx) & (lamx != 0)
        goody = np.isfinite(lamy) & (lamy != 0)
        inv2[goodx] += (1.0 / lamx[goodx]) ** 2
        inv2[goody] += (1.0 / lamy[goody]) ** 2

        lam = np.full_like(inv2, np.nan, dtype=float)
        ok = inv2 > 0
        lam[ok] = 1.0 / np.sqrt(inv2[ok])
        lam_xy[valid] = lam

    lam_xy = lam_xy.reshape(nx, ny).T

    if np.isfinite(np.nanmax(A_xy)):
        thr = float(AMP_MIN_FRACTION_OF_CLUSTER_MAX) * float(np.nanmax(A_xy))
        lam_xy[A_xy < thr] = np.nan

    bg_vmin, bg_vmax = diverging_limits(bg_xy, p_lo=2, p_hi=98)
    cl_vmin, cl_vmax = diverging_limits(rec_xy, p_lo=2, p_hi=98)

    theta_vals = np.asarray(cwt["theta"], dtype=float)
    theta_min = float(np.nanmin(theta_vals))
    theta_max = float(np.nanmax(theta_vals))

    if np.any(np.isfinite(lam_xy)):
        lam_vmin, lam_vmax = np.nanpercentile(lam_xy, [5, 95])
        if (not np.isfinite(lam_vmin)) or (not np.isfinite(lam_vmax)) or (lam_vmin == lam_vmax):
            lam_vmin = float(np.nanmin(lam_xy))
            lam_vmax = float(np.nanmax(lam_xy))
    else:
        lam_vmin, lam_vmax = 0.0, 1.0

    fig, axs = plt.subplots(2, 2, figsize=(16, 11), constrained_layout=True)
    ax00, ax01, ax10, ax11 = axs[0, 0], axs[0, 1], axs[1, 0], axs[1, 1]

    ax00.pcolormesh(x1d_km, y1d_km, bg_xy, shading="auto", cmap="gray", vmin=bg_vmin, vmax=bg_vmax)
    im00 = ax00.pcolormesh(
        x1d_km, y1d_km, rec_xy, shading="auto", cmap="RdBu_r", vmin=cl_vmin, vmax=cl_vmax, alpha=0.6
    )
    ax00.set_title(f"Cluster {kcl} overlay")
    ax00.set_xlabel("x (km)")
    ax00.set_ylabel("y (km)")
    ax00.set_aspect("equal", adjustable="box")
    fig.colorbar(im00, ax=ax00, fraction=0.046, pad=0.02)

    im01 = ax01.pcolormesh(x1d_km, y1d_km, A_xy, shading="auto", cmap="plasma")
    n_pix = int(np.count_nonzero(np.isfinite(A_xy)))
    area_km2 = n_pix * float(dx_km) * float(dy_km)
    ax01.set_title("A(x,y) = max |coeff| in cluster\n" + rf"Area = {area_km2:.0f} km$^2$ ({n_pix} px)")
    ax01.set_xlabel("x (km)")
    ax01.set_ylabel("y (km)")
    ax01.set_aspect("equal", adjustable="box")
    fig.colorbar(im01, ax=ax01, fraction=0.046, pad=0.02)

    im10 = ax10.pcolormesh(x1d_km, y1d_km, lam_xy, shading="auto", cmap="viridis", vmin=lam_vmin, vmax=lam_vmax)
    if np.any(np.isfinite(lam_xy)):
        lam_med = float(np.nanmedian(lam_xy))
        lam_p10, lam_p90 = np.nanpercentile(lam_xy, [10, 90])
        ax10.set_title(
            r"Dominant $\lambda(x,y)$ in cluster" + "\n"
            + rf"median={_fmt(lam_med)} km (10–90%: {_fmt(lam_p10)}–{_fmt(lam_p90)} km)"
        )
    else:
        ax10.set_title(r"Dominant $\lambda(x,y)$ in cluster")
    ax10.set_xlabel("x (km)")
    ax10.set_ylabel("y (km)")
    ax10.set_aspect("equal", adjustable="box")
    fig.colorbar(im10, ax=ax10, fraction=0.046, pad=0.02).set_label("km")

    im11 = ax11.pcolormesh(x1d_km, y1d_km, theta_xy, shading="auto", cmap="PiYG", vmin=theta_min, vmax=theta_max)
    ax11.set_title("θ(x,y) + quiver")
    ax11.set_xlabel("x (km)")
    ax11.set_ylabel("y (km)")
    ax11.set_aspect("equal", adjustable="box")
    fig.colorbar(im11, ax=ax11, fraction=0.046, pad=0.02)

    if np.any(np.isfinite(theta_xy)):
        u, v = theta_to_uv(theta_xy)
        arrow_len_km = 40.0
        u = u * arrow_len_km
        v = v * arrow_len_km

        Xq, Yq = np.meshgrid(x1d_km, y1d_km)
        Xqs = Xq[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        Yqs = Yq[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        uqs = u[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        vqs = v[::QUIVER_STEP_Y, ::QUIVER_STEP_X]
        good = np.isfinite(uqs) & np.isfinite(vqs)
        ax11.quiver(
            Xqs[good], Yqs[good], uqs[good], vqs[good],
            angles="xy", scale_units="xy", scale=QUIVER_SCALE, width=0.0025
        )

    save_fig(fig, outpath)

# --------------------------
# Cluster products (for L6C)
# --------------------------
def cluster_dominant_lambda_map(
    cwt: dict,
    iwave: np.ndarray,
    kcl: int,
    amp_mask_xy: np.ndarray | None = None,
) -> np.ndarray:
    """
    Per-pixel dominant wavelength map λ(x,y) for ONE cluster.
    Returns shape (ny, nx) in XY (same as step 07 plots).

    If amp_mask_xy is provided (A_xy with NaNs), we apply the same NaN mask to λ.
    """
    dec = cwt["decomposition"]  # (ns, nt, nx, ny) complex
    power0 = np.abs(dec) ** 2
    mask4d = (iwave == kcl)
    if not np.any(mask4d):
        ny = dec.shape[3]
        nx = dec.shape[2]
        return np.full((ny, nx), np.nan)

    ns, nt, nx, ny = power0.shape
    power2 = power0.reshape(ns * nt, nx * ny)
    mask2 = mask4d.reshape(ns * nt, nx * ny)

    p = np.where(mask2, power2, -np.inf)
    pmax = np.max(p, axis=0)
    valid = np.isfinite(pmax) & (pmax > 0)

    lam_xy = np.full(nx * ny, np.nan, dtype=float)
    if np.any(valid):
        arg = np.argmax(p[:, valid], axis=0)
        s_idx = arg // nt
        t_idx = arg % nt

        lamx = np.asarray(cwt["wavelength_x"], dtype=float)[s_idx, t_idx]
        lamy = np.asarray(cwt["wavelength_y"], dtype=float)[s_idx, t_idx]

        inv2 = np.zeros_like(lamx, dtype=float)
        goodx = np.isfinite(lamx) & (lamx != 0)
        goody = np.isfinite(lamy) & (lamy != 0)
        inv2[goodx] += (1.0 / lamx[goodx]) ** 2
        inv2[goody] += (1.0 / lamy[goody]) ** 2

        lam = np.full_like(inv2, np.nan, dtype=float)
        ok = inv2 > 0
        lam[ok] = 1.0 / np.sqrt(inv2[ok])

        lam_xy[valid] = lam

    lam_xy = lam_xy.reshape(nx, ny).T  # XY

    if amp_mask_xy is not None:
        lam_xy[~np.isfinite(amp_mask_xy)] = np.nan

    return lam_xy


def copy_l5c_to_l6c_with_clusters(
    l5c_path: str | Path,
    l6c_path: str | Path,
    clusters: List[Dict[str, Any]],
    y_dim: int,
    x_dim: int,
    unbin_factor: int = 1,
) -> None:
    """
    Make an L6C file by copying everything from L5C, then appending:
      - dimension: cluster
      - dimension: y_cluster (=y_dim*unbin_factor), x_cluster (=x_dim*unbin_factor)
      - SlicesNo(cluster)
      - Area(cluster)
      - ClusterPowerAbs(cluster)
      - ClusterPowerPct(cluster)
      - Amplitude(cluster,y_cluster,x_cluster)
      - DominantWavelength(cluster,y_cluster,x_cluster)
      - Angle(cluster,y_cluster,x_cluster)
      - ClusterReconstruction(cluster,y_cluster,x_cluster)   (optional, if present in clusters)

    If unbin_factor>1, we "unbin" (upsample) the per-cluster 2D products by simple repetition
    (nearest-neighbor): (y_dim,x_dim) -> (y_dim*unbin_factor, x_dim*unbin_factor).
    """
    def _unbin_repeat_xy(A: np.ndarray, factor: int) -> np.ndarray:
        if factor is None or int(factor) <= 1:
            return A
        f = int(factor)
        # (y,x) -> (y*f, x*f)
        return np.repeat(np.repeat(A, f, axis=0), f, axis=1)

    l5c_path = Path(l5c_path)
    l6c_path = Path(l6c_path)
    l6c_path.parent.mkdir(parents=True, exist_ok=True)

    ncl = int(len(clusters))

    # Decide whether we have reconstruction maps to write
    has_rec = any((c.get("rec_xy", None) is not None) for c in clusters)

    fup = int(unbin_factor) if (unbin_factor is not None and int(unbin_factor) > 1) else 1
    y_dim_w = int(y_dim) * fup
    x_dim_w = int(x_dim) * fup

    with Dataset(l5c_path, "r") as src, Dataset(l6c_path, "w", format="NETCDF4") as dst:
        # ---- globals
        dst.setncatts({k: src.getncattr(k) for k in src.ncattrs()})

        # ---- dimensions
        for dname, dim in src.dimensions.items():
            dst.createDimension(dname, (len(dim) if not dim.isunlimited() else None))

        # ---- variables (copy)
        for vname, var in src.variables.items():
            out = dst.createVariable(
                vname,
                var.datatype,
                var.dimensions,
                zlib=True,
                complevel=4,
                fill_value=getattr(var, "_FillValue", None),
            )
            out.setncatts({k: var.getncattr(k) for k in var.ncattrs()})
            out[:] = var[:]

        # ---- new dims (WRITE dims are UNBINNED if fup>1)
        dst.createDimension("cluster", ncl)
        dst.createDimension("y_cluster", y_dim_w)
        dst.createDimension("x_cluster", x_dim_w)

        # ---- new vars
        v_slice = dst.createVariable("SlicesNo", "i4", ("cluster",), zlib=True, complevel=4)
        v_area  = dst.createVariable("Area", "f4", ("cluster",), zlib=True, complevel=4)

        v_pabs = dst.createVariable("ClusterPowerAbs", "f8", ("cluster",), zlib=True, complevel=4)
        v_ppct = dst.createVariable("ClusterPowerPct", "f4", ("cluster",), zlib=True, complevel=4)

        v_amp = dst.createVariable(
            "Amplitude", "f4", ("cluster", "y_cluster", "x_cluster"),
            zlib=True, complevel=4, fill_value=np.float32(np.nan)
        )
        v_lam = dst.createVariable(
            "DominantWavelength", "f4", ("cluster", "y_cluster", "x_cluster"),
            zlib=True, complevel=4, fill_value=np.float32(np.nan)
        )
        v_ang = dst.createVariable(
            "Angle", "f4", ("cluster", "y_cluster", "x_cluster"),
            zlib=True, complevel=4, fill_value=np.float32(np.nan)
        )

        v_rec = None
        if has_rec:
            v_rec = dst.createVariable(
                "ClusterReconstruction", "f4", ("cluster", "y_cluster", "x_cluster"),
                zlib=True, complevel=4, fill_value=np.float32(np.nan)
            )
            v_rec.units = "arb"
            v_rec.long_name = "Per-cluster wavelet reconstruction map (relative; median-subtracted)"

        # attrs
        v_area.units = "km2"
        v_amp.units = "arb"
        v_lam.units = "km"
        v_ang.units = "rad"
        v_slice.long_name = "Slice index (starts at 0) for each cluster"

        v_pabs.units = "arb"
        v_pabs.long_name = "Absolute summed wavelet power for cluster (sum |coeff|^2 over cluster mask)"
        v_ppct.units = "percent"
        v_ppct.long_name = "Cluster power as percent of total clustered power within its slice"

        # ---- write data
        if ncl == 0:
            return

        v_slice[:] = np.asarray([c["slice_no"] for c in clusters], dtype=np.int32)
        v_area[:]  = np.asarray([c["area_km2"] for c in clusters], dtype=np.float32)

        v_pabs[:] = np.asarray([c.get("power_abs", np.nan) for c in clusters], dtype=np.float64)
        v_ppct[:] = np.asarray([c.get("power_pct", np.nan) for c in clusters], dtype=np.float32)

        # stack maps (UNBINNED if fup>1)
        amp_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32)
        lam_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32)
        ang_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32)
        rec_stack = np.full((ncl, y_dim_w, x_dim_w), np.nan, dtype=np.float32) if has_rec else None

        for i, c in enumerate(clusters):
            A = c["A_xy"]
            L = c["lam_xy"]
            T = c["theta_xy"]

            # ---- "unbin" by repetition: (y_dim,x_dim) -> (y_dim_w,x_dim_w)
            A = _unbin_repeat_xy(A, fup)
            L = _unbin_repeat_xy(L, fup)
            T = _unbin_repeat_xy(T, fup)

            # enforce exact shape (y_dim_w, x_dim_w)
            if A.shape != (y_dim_w, x_dim_w):
                tmp = np.full((y_dim_w, x_dim_w), np.nan, dtype=np.float32)
                yy = min(y_dim_w, A.shape[0])
                xx = min(x_dim_w, A.shape[1])
                tmp[:yy, :xx] = A[:yy, :xx]
                A = tmp

            if L.shape != (y_dim_w, x_dim_w):
                tmp = np.full((y_dim_w, x_dim_w), np.nan, dtype=np.float32)
                yy = min(y_dim_w, L.shape[0])
                xx = min(x_dim_w, L.shape[1])
                tmp[:yy, :xx] = L[:yy, :xx]
                L = tmp

            if T.shape != (y_dim_w, x_dim_w):
                tmp = np.full((y_dim_w, x_dim_w), np.nan, dtype=np.float32)
                yy = min(y_dim_w, T.shape[0])
                xx = min(x_dim_w, T.shape[1])
                tmp[:yy, :xx] = T[:yy, :xx]
                T = tmp

            amp_stack[i] = A.astype(np.float32)
            lam_stack[i] = L.astype(np.float32)
            ang_stack[i] = T.astype(np.float32)

            if has_rec and rec_stack is not None and v_rec is not None:
                R = c.get("rec_xy", None)
                if R is not None:
                    R = _unbin_repeat_xy(R, fup)
                    if R.shape != (y_dim_w, x_dim_w):
                        tmp = np.full((y_dim_w, x_dim_w), np.nan, dtype=np.float32)
                        yy = min(y_dim_w, R.shape[0])
                        xx = min(x_dim_w, R.shape[1])
                        tmp[:yy, :xx] = R[:yy, :xx]
                        R = tmp
                    rec_stack[i] = R.astype(np.float32)

        v_amp[:] = amp_stack
        v_lam[:] = lam_stack
        v_ang[:] = ang_stack
        if has_rec and v_rec is not None and rec_stack is not None:
            v_rec[:] = rec_stack

def make_l6c_path_from_l5c(l5c_path: str | Path) -> Path:
    p = Path(l5c_path)
    name = p.name
    if "_l5c_" in name:
        name = name.replace("_l5c_", "_l6c_")
    else:
        # fallback: just append
        stem = p.stem
        name = f"{stem}_l6c.nc"
    return p.with_name(name)
    

def run_pipeline_for_slice(
    rad2d_flat: np.ndarray,
    x_slice: slice,
    outdir: Path,
    x_start_unbinned: int,
    core_start_unbinned: int,
    core_end_unbinned: int,
    cfg: Dict[str, Any],
    slice_no: int,
) -> Dict[str, Any]:
    """
    Runs processing on an extended slice (x_slice),
    but returns only CORE tiles for stitching.

    Step 08 (power) intentionally removed.

    Figure 1 includes extra panel: wavelet reconstruction after noise filtering.
    Figure 9 has 3 panels:
      Remaining (input to CWT)
      Wavy reconstruction after noise filtering
      Reconstruction after clustering (all clusters)
    """
    outdir.mkdir(parents=True, exist_ok=True)

    min_amp = float(cfg.get("MIN_AMP"))
    thr = float(cfg.get("THR"))
    white_noise_threshold = float(cfg.get("WHITE_NOISE_THRESHOLD"))
    wps_noise_threshold = float(cfg.get("WPS_NOISE_THRESHOLD"))
    run_name = str(cfg.get("name", "run"))

    # ---- controls for what we store into NetCDF later ----
    STORE_CLUSTER_RECON = True  # saves the "subplot 07 first panel" colored field (no gray bg)

    if APPLY_BINNING and int(BIN_FACTOR) > 1:
        f = int(BIN_FACTOR)
        dx_km = DX_KM_BASE * f
        dy_km = DY_KM_BASE * f
    else:
        f = 1
        dx_km = DX_KM_BASE
        dy_km = DY_KM_BASE

    core_width_unbinned = int(core_end_unbinned - core_start_unbinned)
    core_w_binned = core_width_unbinned // f
    rel0_unbinned = int(core_start_unbinned - x_start_unbinned)
    rel0_binned = rel0_unbinned // f
    rel1_binned = rel0_binned + core_w_binned

    def crop_pad_core(A: np.ndarray) -> np.ndarray:
        nyA, nxA = A.shape
        left = max(0, rel0_binned)
        right = min(nxA, rel1_binned)
        core = A[:, left:right]
        need = core_w_binned - core.shape[1]
        if need > 0:
            core = np.concatenate([core, np.full((nyA, need), np.nan)], axis=1)
        elif need < 0:
            core = core[:, :core_w_binned]
        return core

    raw_win = np.asarray(rad2d_flat[Y_SLICE, x_slice], dtype=float)

    if APPLY_BINNING and int(BIN_FACTOR) > 1:
        raw_win = bin2d_block_mean(raw_win, BIN_FACTOR)

    ny, nx = raw_win.shape
    x1d_km = np.arange(nx, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    bin_tag = f" (binned {BIN_FACTOR}x{BIN_FACTOR})" if (APPLY_BINNING and int(BIN_FACTOR) > 1) else ""
    quicklook_save(
        x1d_km, y1d_km, raw_win,
        f"Temperature (window, flattened){bin_tag}\n{run_name} MIN_AMP={min_amp} THR={thr}",
        outdir / "00_temperature_window.png",
    )

    proc0 = raw_win.astype(float, copy=True)

    if REMOVE_TREND:
        plane = fit_plane(proc0)
        detrended = proc0 - plane
    else:
        plane = np.zeros_like(proc0)
        detrended = proc0

    fourier_bg, _ = extract_first_fourier_bg(detrended)
    background = plane + fourier_bg
    remaining = proc0 - background

    if APPLY_TAPER:
        remaining_tapered = apply_edge_hanning_taper(remaining, TAPER_EDGE_PIXELS)
    else:
        remaining_tapered = remaining.copy()

    proc = remaining_tapered.copy()
    if STANDARDIZE:
        m = float(np.median(proc))
        s = float(np.std(proc))
        proc = (proc - m) / s if s > 0 else (proc - m)
    proc = maybe_blur(proc, GAUSS_BLUR_SIGMA)

    wave_xy = proc.T  # (nx, ny)

    cwt_raw = transform.decompose2d(
        wave_xy, dx=dx_km, dy=dy_km, s0=S0, dj=DJ, js=JS, jt=JT, aspect=ASPECT
    )

    if APPLY_RED_NOISE_FILTER:
        cwt, diag = apply_red_noise_filter_cwt(
            cwt_raw,
            white_noise_threshold=white_noise_threshold,
            wps_noise_threshold=wps_noise_threshold,
        )
        if SAVE_REDNOISE_DIAGNOSTICS:
            lines = []
            lines.append("Red noise filter diagnostics\n\n")
            lines.append(f"name                  = {run_name}\n")
            lines.append(f"WHITE_NOISE_THRESHOLD = {white_noise_threshold}\n")
            lines.append(f"WPS_NOISE_THRESHOLD   = {wps_noise_threshold}\n\n")
            if diag.get("skipped", False):
                lines.append(f"SKIPPED: {diag.get('reason','unknown')}\n")
            else:
                lines.append(f"n_coeff_total        = {diag.get('n_coeff_total')}\n")
                lines.append(f"n_zero_after_white   = {diag.get('n_zero_after_white')}\n")
                lines.append(f"n_zero_after_scaled  = {diag.get('n_zero_after_scaled')}\n")
                lines.append(f"sMAD_WPS min         = {diag.get('sMAD_min')}\n")
                lines.append(f"sMAD_WPS median      = {diag.get('sMAD_median')}\n")
                lines.append(f"WPS_scaled min       = {diag.get('WPS_scaled_min')}\n")
                lines.append(f"WPS_scaled median    = {diag.get('WPS_scaled_median')}\n")
                lines.append(f"WPS_scaled max       = {diag.get('WPS_scaled_max')}\n")
            (outdir / "02a_rednoise_diagnostics.txt").write_text("".join(lines))
    else:
        cwt = cwt_raw

    rec_wavy = transform.reconstruct2d(cwt)  # (nx, ny)
    wavy_xy = rec_wavy.T - np.median(rec_wavy)

    quicklook_background_decomp_4panel_save(
        x1d_km, y1d_km,
        proc0, background, remaining, remaining_tapered, TAPER_EDGE_PIXELS,
        outdir / "01_background_decomp_5panel.png",
        wavy_after_noise_xy=wavy_xy,
    )

    amps, idxs, iwave = utils.identify_cluster2d(cwt, min_amp=min_amp, thr=thr)
    decomposition = cwt["decomposition"]
    orig_decomp = decomposition.copy()

    cmap_cwt = matplotlib.cm.turbo
    norm_cwt = matplotlib.colors.BoundaryNorm(
        np.exp(np.linspace(np.log(0.05), np.log(1.0), 10)), cmap_cwt.N
    )
    fig_cwt, _ = utils.plot_decomposition2d(cwt, redux_s=2, redux_t=2, cmap=cmap_cwt, norm=norm_cwt)
    save_fig(fig_cwt, outdir / "02_decomposition_overview.png")

    plot_utils.plot_fov_wavelet_spectrum(cwt, stat="max")
    fig_spec = plt.gcf()
    save_fig(fig_spec, outdir / "03_fov_wavelet_spectrum.png")

    bg = remaining_tapered - np.median(remaining_tapered)

    # ---- power per cluster (absolute and percent) ----
    cluster_ids, P_abs, P_pct = compute_cluster_power_table(cwt, iwave)
    if cluster_ids.size == 0:
        (outdir / "NO_CLUSTERS.txt").write_text(
            "No clusters found.\n\n"
            f"Run name: {run_name}\n"
            f"Used MIN_AMP={min_amp}, THR={thr}\n"
            f"Used WHITE_NOISE_THRESHOLD={white_noise_threshold}, WPS_NOISE_THRESHOLD={wps_noise_threshold}\n\n"
            "Try lowering MIN_AMP or THR.\n"
            "If red noise filter is on, try lowering WPS_NOISE_THRESHOLD or WHITE_NOISE_THRESHOLD.\n"
        )
        remaining_core_xy = crop_pad_core(proc)
        nan_core = np.full_like(remaining_core_xy, np.nan)
        return {
            "has_clusters": False,
            "x_start_unbinned": core_start_unbinned,
            "dx_km": dx_km,
            "dy_km": dy_km,
            "remaining_xy": remaining_core_xy,
            "wavy_xy": nan_core,
            "rec_all_xy": nan_core,
            "cluster_records": [],
        }

    power_abs_map = {int(k): float(p) for k, p in zip(cluster_ids, P_abs)}
    power_pct_map = {int(k): float(p) for k, p in zip(cluster_ids, P_pct)}

    # --------------------------
    # Collect ALL clusters in this slice (core-tile maps), now including:
    #   power_abs, power_pct, and optional rec_xy (per-cluster reconstruction)
    # --------------------------
    cluster_records: List[Dict[str, Any]] = []
    for kcl in cluster_ids.astype(int):
        kcl = int(kcl)

        A_xy, theta_xy = cluster_amp_theta_maps(
            cwt, iwave, kcl,
            amp_min_fraction_of_cluster_max=AMP_MIN_FRACTION_OF_CLUSTER_MAX,
        )
        lam_xy = cluster_dominant_lambda_map(cwt, iwave, kcl, amp_mask_xy=A_xy)

        # Optional: store per-cluster reconstruction map (the colored overlay field)
        rec_core = None
        if STORE_CLUSTER_RECON:
            decomposition[:] = orig_decomp
            decomposition[iwave != kcl] = 0
            rec = transform.reconstruct2d(cwt)     # (nx, ny)
            rec_xy = rec.T
            rec_xy = rec_xy - np.nanmedian(rec_xy)  # keep it "relative" like your plots
            rec_core = crop_pad_core(rec_xy)
            decomposition[:] = orig_decomp

        A_core = crop_pad_core(A_xy)
        th_core = crop_pad_core(theta_xy)
        lam_core = crop_pad_core(lam_xy)

        n_pix = int(np.count_nonzero(np.isfinite(A_core)))
        area_km2 = float(n_pix) * float(dx_km) * float(dy_km)

        cluster_records.append(
            dict(
                slice_no=int(slice_no),
                kcl=kcl,
                area_km2=area_km2,
                power_abs=power_abs_map.get(kcl, np.nan),
                power_pct=power_pct_map.get(kcl, np.nan),
                A_xy=A_core,
                lam_xy=lam_core,
                theta_xy=th_core,
                rec_xy=rec_core,  # may be None
            )
        )

    order_power = np.argsort(P_abs)[::-1]
    cluster_ids_sorted = cluster_ids[order_power]
    P_abs_sorted = P_abs[order_power]
    P_pct_sorted = P_pct[order_power]

    lines = []
    lines.append("Cluster power ranking (clustered only)\n\n")
    lines.append(f"name = {run_name}\n")
    lines.append(f"MIN_AMP = {min_amp}\n")
    lines.append(f"THR = {thr}\n")
    lines.append(f"WHITE_NOISE_THRESHOLD = {white_noise_threshold}\n")
    lines.append(f"WPS_NOISE_THRESHOLD = {wps_noise_threshold}\n\n")
    lines.append("rank  kcl      P_abs           P_pct\n")
    for r, (kcl, Pk, pk) in enumerate(zip(cluster_ids_sorted, P_abs_sorted, P_pct_sorted), start=1):
        lines.append(f"{r:4d}  {int(kcl):3d}  {Pk: .6e}   {pk:6.2f}%\n")
    (outdir / "04_cluster_power_ranking.txt").write_text("".join(lines))

    n_show = int(min(N_SHOW, cluster_ids_sorted.size))
    biggest = cluster_ids_sorted[:n_show]

    nrow = int(np.ceil(np.sqrt(n_show)))
    ncol = int(np.ceil(n_show / nrow))
    theta_vals = np.asarray(cwt["theta"], dtype=float)

    def plot_cluster_grid(kind: str, outname: str) -> None:
        fig, axs = plt.subplots(nrow, ncol, figsize=(12, 10))
        axs = np.atleast_1d(axs).reshape(-1)

        bg_vmin, bg_vmax = diverging_limits(bg, p_lo=2, p_hi=98)
        number_box = {"boxstyle": "square", "lw": 0.67, "facecolor": "white", "edgecolor": "black"}

        for ax, kcl in zip(axs, biggest):
            decomposition[:] = orig_decomp
            decomposition[iwave != kcl] = 0
            rec = transform.reconstruct2d(cwt)
            rec_centered = rec.T - np.median(rec)

            s_idx = int(idxs[kcl][0])
            t_idx = int(idxs[kcl][1])
            _, _, lam = effective_lambda(cwt, s_idx, t_idx)

            theta_dom_rad = float(theta_vals[t_idx])
            theta_dom_deg = (np.degrees(theta_dom_rad) + 180.0) % 180.0

            ii = int(np.where(cluster_ids_sorted == kcl)[0][0])
            Pk = float(P_abs_sorted[ii])
            pk = float(P_pct_sorted[ii])

            if kind == "standalone":
                ax.pcolormesh(x1d_km, y1d_km, rec_centered, shading="auto", cmap="RdBu_r", vmin=-1.0, vmax=1.0)
            elif kind == "overlay":
                ax.pcolormesh(x1d_km, y1d_km, bg, shading="auto", cmap="gray", vmin=bg_vmin, vmax=bg_vmax)
                ax.pcolormesh(
                    x1d_km, y1d_km, rec_centered,
                    shading="auto", cmap="RdBu_r", vmin=-1.0, vmax=1.0, alpha=0.6
                )
            else:
                raise ValueError(kind)

            ax.set_title(
                rf"$\lambda$={_fmt(lam)} km, $\theta$={theta_dom_deg:.0f}$^\circ$"
                + "\n"
                + rf"P={Pk:.2e} ({pk:.1f}%)"
            )
            ax.set_aspect("equal", adjustable="box")

        for ax in axs[n_show:]:
            ax.axis("off")
        for ax in axs[:n_show]:
            ax.set_xlabel("x (km)")
            ax.set_ylabel("y (km)")

        for i, ax in enumerate(axs, start=1):
            if ax.has_data():
                ax.text(0.12, 0.15, str(i), transform=ax.transAxes, bbox=number_box, va="top", ha="right")

        fig.tight_layout()
        save_fig(fig, outdir / outname)

    plot_cluster_grid("standalone", "05_clusters_standalone.png")
    plot_cluster_grid("overlay", "06_clusters_overlay.png")

    for kcl in biggest:
        plot_cluster_individual_save(
            cwt=cwt,
            iwave=iwave,
            kcl=int(kcl),
            x1d_km=x1d_km,
            y1d_km=y1d_km,
            orig_decomp=orig_decomp,
            bg_xy=bg,
            dx_km=dx_km,
            dy_km=dy_km,
            outpath=outdir / f"07_cluster_{int(kcl):03d}_maps.png",
        )

    decomposition[:] = orig_decomp
    decomposition[iwave < 0] = 0
    rec_all = transform.reconstruct2d(cwt)  # (nx, ny)
    rec_all_xy = rec_all.T - np.median(rec_all)
    decomposition[:] = orig_decomp

    remaining_after_preproc_xy = proc
    vmin, vmax = diverging_limits(remaining_after_preproc_xy, wavy_xy, rec_all_xy, p_lo=2, p_hi=98)

    figR, (ax1, ax2, ax3) = plt.subplots(
        3, 1, figsize=(18, 8), constrained_layout=True, sharex=True, sharey=True
    )

    im1 = ax1.pcolormesh(
        x1d_km, y1d_km, remaining_after_preproc_xy,
        shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax
    )
    ax1.set_title("Remaining (after preprocessing, input to CWT)")
    ax1.set_xlabel("x (km)")
    ax1.set_ylabel("y (km)")
    ax1.set_aspect("equal", adjustable="box")

    im2 = ax2.pcolormesh(
        x1d_km, y1d_km, wavy_xy,
        shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax
    )
    ax2.set_title("Wavy stuff (wavelet reconstruction after noise filtering)")
    ax2.set_xlabel("x (km)")
    ax2.set_ylabel("y (km)")
    ax2.set_aspect("equal", adjustable="box")

    im3 = ax3.pcolormesh(
        x1d_km, y1d_km, rec_all_xy,
        shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax
    )
    ax3.set_title("Reconstruction after clustering (all clusters combined)")
    ax3.set_xlabel("x (km)")
    ax3.set_ylabel("y (km)")
    ax3.set_aspect("equal", adjustable="box")

    cbar = figR.colorbar(im1, ax=[ax1, ax2, ax3], orientation="vertical", fraction=0.015, pad=0.01)
    cbar.set_label("arb")
    save_fig(figR, outdir / "09_remaining_vs_wavy_vs_allclusters.png")

    remaining_core_xy = crop_pad_core(remaining_after_preproc_xy)
    wavy_core_xy = crop_pad_core(wavy_xy)
    rec_all_core_xy = crop_pad_core(rec_all_xy)

    return {
        "has_clusters": True,
        "x_start_unbinned": core_start_unbinned,
        "dx_km": dx_km,
        "dy_km": dy_km,
        "remaining_xy": remaining_core_xy,
        "wavy_xy": wavy_core_xy,
        "rec_all_xy": rec_all_core_xy,
        "cluster_records": cluster_records,  # now includes power + optional rec map
    }


def stitch_and_save_final_09(slice_results: List[Dict[str, Any]], outroot: Path, tag: str) -> None:
    if not slice_results:
        return

    slice_results = sorted(slice_results, key=lambda d: int(d.get("x_start_unbinned", 0)))

    dx_km = float(slice_results[0]["dx_km"])
    dy_km = float(slice_results[0]["dy_km"])
    x0_unbinned = int(slice_results[0].get("x_start_unbinned", 0))
    x0_km = x0_unbinned * DX_KM_BASE

    remaining_stitched = np.concatenate([d["remaining_xy"] for d in slice_results], axis=1)
    wavy_stitched = np.concatenate([d["wavy_xy"] for d in slice_results], axis=1)
    rec_all_stitched = np.concatenate([d["rec_all_xy"] for d in slice_results], axis=1)

    ny, nx_tot = remaining_stitched.shape
    x1d_km = x0_km + np.arange(nx_tot, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    vmin, vmax = diverging_limits(remaining_stitched, wavy_stitched, rec_all_stitched, p_lo=2, p_hi=98)

    figR, (axT, axM, axB) = plt.subplots(
        3, 1, figsize=(18, 6), constrained_layout=True, sharex=True, sharey=True
    )

    imT = axT.pcolormesh(
        x1d_km, y1d_km, remaining_stitched,
        shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax
    )
    axT.set_title(f"STITCHED (core tiles) {tag}: Remaining (after preprocessing, input to CWT)")
    axT.set_xlabel("x (km)")
    axT.set_ylabel("y (km)")
    axT.set_aspect("equal", adjustable="box")

    imM = axM.pcolormesh(
        x1d_km, y1d_km, wavy_stitched,
        shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax
    )
    axM.set_title(f"STITCHED (core tiles) {tag}: Wavy stuff (wavelet reconstruction after noise filtering)")
    axM.set_xlabel("x (km)")
    axM.set_ylabel("y (km)")
    axM.set_aspect("equal", adjustable="box")

    imB = axB.pcolormesh(
        x1d_km, y1d_km, rec_all_stitched,
        shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax
    )
    axB.set_title(f"STITCHED (core tiles) {tag}: Reconstruction after clustering (all clusters combined)")
    axB.set_xlabel("x (km)")
    axB.set_ylabel("y (km)")
    axB.set_aspect("equal", adjustable="box")

    axT.set_xlim(600, 10000)

    cbar = figR.colorbar(imT, ax=[axT, axM, axB], orientation="vertical", fraction=0.012, pad=0.01)
    cbar.set_label("kRayleigh (Relative)")
    save_fig(figR, outroot / f"FINAL_09_remaining_vs_wavy_vs_allclusters_STITCHED_{tag}.png")


def run_all_slices_single_offset(cfg: Dict[str, Any]) -> None:
    OUTPUT_ROOT.mkdir(parents=True, exist_ok=True)

    rad2d = read_radiance_frame(FILE_PATH, FRAME_IDX)
    rad2d_flat = flatten_swath(rad2d, target_ny=300)
    nx_total = rad2d_flat.shape[1]

    x_offset = int(X_OFFSET)
    if x_offset >= nx_total:
        raise ValueError(f"X_OFFSET={x_offset} too large for nx_total={nx_total}")

    run_name = str(cfg.get("name", "run"))
    outroot = OUTPUT_ROOT / f"{run_name}_offset_{x_offset:04d}"
    outroot.mkdir(parents=True, exist_ok=True)

    (outroot / "RUN_CONFIG.txt").write_text("\n".join([f"{k} = {v}" for k, v in cfg.items()]) + "\n")

    core_starts = list(range(x_offset, nx_total, X_CHUNK))
    stitch_results: List[Dict[str, Any]] = []
    all_cluster_records: List[Dict[str, Any]] = []

    # expected (y,x) of the CORE tile AFTER binning
    f = int(BIN_FACTOR) if (APPLY_BINNING and int(BIN_FACTOR) > 1) else 1
    y_dim = int(rad2d_flat[Y_SLICE, :].shape[0] // f)  # 300//2 = 150
    x_dim = int(X_CHUNK // f)                          # 600//2 = 300

    for slice_no, core_start in enumerate(core_starts):
        core_end = min(core_start + X_CHUNK, nx_total)

        ext_start = max(0, core_start - SLICE_PAD)
        ext_end = min(nx_total, core_end + SLICE_PAD)
        x_slice = slice(ext_start, ext_end)

        outdir = outroot / f"xcore_{core_start:05d}_{core_end:05d}_xext_{ext_start:05d}_{ext_end:05d}"
        print(f"[{run_name} | offset {x_offset}] core={core_start}:{core_end} ext={ext_start}:{ext_end} saved to {outdir}")

        res = run_pipeline_for_slice(
            rad2d_flat=rad2d_flat,
            x_slice=x_slice,
            outdir=outdir,
            x_start_unbinned=ext_start,
            core_start_unbinned=core_start,
            core_end_unbinned=core_end,
            cfg=cfg,
            slice_no=slice_no,
        )
        stitch_results.append(res)
        all_cluster_records.extend(res.get("cluster_records", []))

    stitch_and_save_final_09(stitch_results, outroot, tag=f"{run_name}_offset_{x_offset:04d}")
    print("Done. Outputs in", outroot.resolve())

    # ---- Write L6C with cluster products ----
    l6c_path = make_l6c_path_from_l5c(FILE_PATH)
    l6c_path = outroot / l6c_path.name  # saves next to outputs; change if you want elsewhere

    copy_l5c_to_l6c_with_clusters(
        l5c_path=FILE_PATH,
        l6c_path=l6c_path,
        clusters=all_cluster_records,
        y_dim=y_dim,
        x_dim=x_dim,
        unbin_factor=(BIN_FACTOR if (APPLY_BINNING and int(BIN_FACTOR) > 1) else 1),
    )

    print("Saved L6C with clusters:", l6c_path.resolve())


for cfg in RUNS:
    run_all_slices_single_offset(cfg)
```

## Read the saved file

```python
# %%
"""
Plot the biggest cluster in slice 5, where "biggest" = highest MAX amplitude
within that slice.

Reads an L6C file created by your pipeline (clusters saved per core tile).
Finds all clusters with SlicesNo == 5, picks the one with max(Amplitude),
then plots:
  1) Amplitude
  2) DominantWavelength
  3) Angle
and also prints basic stats.

Note: x/y here are the core-tile grid (e.g., 150 x 300 if BIN_FACTOR=2).
"""

from __future__ import annotations

from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from netCDF4 import Dataset

# --------------------------
# User params
# --------------------------
L6C_PATH = Path("awe_l6c_q20_2024165T2120_03175_v01.nc")  # <-- change
SLICE_NO = 5
OUTDIR = Path("outputs_l6c_slice6_biggest_cluster")
OUTDIR.mkdir(parents=True, exist_ok=True)

# Optional: pixel spacing on the CLUSTER grid (km)
DX_KM = 4.0  # e.g., DX_KM_BASE * BIN_FACTOR = 2*2
DY_KM = 4.0

SAVE_DPI = 200


# --------------------------
# Small helpers
# --------------------------
def save_fig(fig: plt.Figure, outpath: Path) -> None:
    outpath.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(outpath, dpi=SAVE_DPI, bbox_inches="tight")
    plt.close(fig)


def robust_limits(A: np.ndarray, p_lo: float = 2, p_hi: float = 98) -> tuple[float, float]:
    a = np.asarray(A)
    good = np.isfinite(a)
    if not np.any(good):
        return (-1.0, 1.0)
    lo, hi = np.nanpercentile(a[good], [p_lo, p_hi])
    if (not np.isfinite(lo)) or (not np.isfinite(hi)) or (lo == hi):
        lo = float(np.nanmin(a[good]))
        hi = float(np.nanmax(a[good]))
    return (float(lo), float(hi))


# --------------------------
# Load
# --------------------------
with Dataset(L6C_PATH, "r") as nc:
    slices_no = np.asarray(nc.variables["SlicesNo"][:], dtype=int)          # (cluster,)
    amp_all = np.asarray(nc.variables["Amplitude"][:], dtype=float)         # (cluster,y,x)
    lam_all = np.asarray(nc.variables["DominantWavelength"][:], dtype=float)
    ang_all = np.asarray(nc.variables["Angle"][:], dtype=float)

ncl, ny, nx = amp_all.shape
print(f"Loaded {ncl} clusters. Tile shape (y,x)=({ny},{nx})")

# --------------------------
# Select slice and biggest cluster
# --------------------------
idx = np.where(slices_no == int(SLICE_NO))[0]
if idx.size == 0:
    raise ValueError(f"No clusters found with SlicesNo == {SLICE_NO}")

# "biggest" = highest maximum amplitude
max_amp_each = np.array([np.nanmax(amp_all[i]) for i in idx], dtype=float)
best_local = int(np.nanargmax(max_amp_each))
best_cluster_index = int(idx[best_local])

A = amp_all[best_cluster_index]
L = lam_all[best_cluster_index]
T = ang_all[best_cluster_index]

print(f"Slice {SLICE_NO}: found {idx.size} clusters.")
print(f"Biggest cluster (by max amplitude) = global cluster index {best_cluster_index}")
print(f"  max(A) = {np.nanmax(A):.6g}")
print(f"  median(A) = {np.nanmedian(A):.6g}")
print(f"  n_finite(A) = {np.count_nonzero(np.isfinite(A))} pixels")

# --------------------------
# Plot
# --------------------------
x1d_km = np.arange(nx, dtype=float) * DX_KM
y1d_km = np.arange(ny, dtype=float) * DY_KM

A_vmin, A_vmax = robust_limits(A, 2, 98)

# wavelength and angle: simple robust scaling
L_vmin, L_vmax = robust_limits(L, 5, 95)
T_vmin, T_vmax = robust_limits(T, 2, 98)

fig, (ax1, ax2, ax3) = plt.subplots(
    3, 1, figsize=(14, 9), constrained_layout=True, sharex=True, sharey=True
)

im1 = ax1.pcolormesh(x1d_km, y1d_km, A, shading="auto", vmin=A_vmin, vmax=A_vmax)
ax1.set_title(f"Slice {SLICE_NO} biggest cluster (idx {best_cluster_index}): Amplitude")
ax1.set_ylabel("y (km)")
fig.colorbar(im1, ax=ax1, fraction=0.025, pad=0.01).set_label("Amplitude (K)")

im2 = ax2.pcolormesh(x1d_km, y1d_km, L, shading="auto", vmin=L_vmin, vmax=L_vmax)
ax2.set_title("DominantWavelength")
ax2.set_ylabel("y (km)")
fig.colorbar(im2, ax=ax2, fraction=0.025, pad=0.01).set_label("km")

im3 = ax3.pcolormesh(x1d_km, y1d_km, T, shading="auto", vmin=T_vmin, vmax=T_vmax)
ax3.set_title("Angle")
ax3.set_xlabel("x (km)")
ax3.set_ylabel("y (km)")
fig.colorbar(im3, ax=ax3, fraction=0.025, pad=0.01).set_label("rad")

for ax in (ax1, ax2, ax3):
    ax.set_aspect("equal", adjustable="box")

outpath = OUTDIR / f"slice_{SLICE_NO:03d}_biggest_cluster_by_maxamp.png"
save_fig(fig, outpath)
print("Saved:", outpath.resolve())
```

## Match

```python
# %%
"""
Slice matching (Temperature -> Radiance), using MEDIAN WAVELENGTH RATIO.

Updates requested:
1) In the side-by-side 4x2 plot, the FIRST TWO ROWS (Recon, Amp) do NOT need
   to share the same color scale between L6A (left) and L6C (right).
   - So: Recon and Amp use per-panel vmin/vmax (left/right independently).
   - Lambda + Angle still share scales within each row (left/right comparable).

2) Combine ALL PASSED L6A reconstructions together (sum) and save a plot.
   - Uses ClusterReconstruction if present; otherwise those clusters contribute NaNs (skipped).
"""

from __future__ import annotations

from pathlib import Path
from typing import Dict, Any, List, Tuple

import numpy as np
import matplotlib.pyplot as plt
from netCDF4 import Dataset

# --------------------------
# User params
# --------------------------
L6C_PATH = Path("awe_l6c_q20_2024165T2120_03175_v01.nc")  # Radiance-based clusters
L6A_PATH = Path("awe_l6a_tmp_2024165T2120_03175_v01.nc")  # Temperature-based clusters
SLICE_NO = 5

OUTDIR = Path("outputs_slice5_l6a_to_l6c_matching_ratio")
OUTDIR.mkdir(parents=True, exist_ok=True)
SAVE_DPI = 200

# Cluster grid spacing (km) for axes only
DX_KM = 4.0
DY_KM = 4.0

# --------------------------
# Matching thresholds (tune)
# --------------------------
MIN_IOU = 0.05
MIN_INTERSECTION_PIX = 50

# Median wavelength ratio requirement (>= 1.0). Example: 1.25 allows ±25%.
MAX_LAMBDA_MED_RATIO = 1.25

# Angle overlap requirement (degrees), modulo 180
MAX_ANGLE_DIFF_DEG = 15.0

# If True, still assign and plot the "best" match even if it fails thresholds (flag FAIL)
ASSIGN_BEST_EVEN_IF_FAILS = True


# --------------------------
# Helpers
# --------------------------
def save_fig(fig: plt.Figure, outpath: Path) -> None:
    outpath.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(outpath, dpi=SAVE_DPI, bbox_inches="tight")
    plt.close(fig)


def robust_limits(A: np.ndarray, p_lo: float = 2, p_hi: float = 98) -> Tuple[float, float]:
    a = np.asarray(A)
    good = np.isfinite(a)
    if not np.any(good):
        return (-1.0, 1.0)
    lo, hi = np.nanpercentile(a[good], [p_lo, p_hi])
    if (not np.isfinite(lo)) or (not np.isfinite(hi)) or lo == hi:
        lo = float(np.nanmin(a[good]))
        hi = float(np.nanmax(a[good]))
    return (float(lo), float(hi))


def nanpercentile_safe(a: np.ndarray, q: float) -> float:
    good = np.isfinite(a)
    if not np.any(good):
        return np.nan
    return float(np.nanpercentile(a[good], q))


def circ_diff_deg_180(a_deg: float, b_deg: float) -> float:
    if (not np.isfinite(a_deg)) or (not np.isfinite(b_deg)):
        return np.nan
    d = abs(a_deg - b_deg) % 180.0
    return min(d, 180.0 - d)


def circ_median_deg_180(theta_rad: np.ndarray, mask: np.ndarray) -> float:
    th = np.asarray(theta_rad, dtype=float)
    m = mask & np.isfinite(th)
    if not np.any(m):
        return np.nan
    ang = th[m]
    z = np.exp(1j * 2.0 * ang)  # 180-deg periodic => doubled-angle trick
    zmean = np.mean(z)
    if (not np.isfinite(zmean.real)) or (not np.isfinite(zmean.imag)) or abs(zmean) == 0:
        return np.nan
    th_mean = 0.5 * np.angle(zmean)  # radians
    deg = (np.degrees(th_mean) + 180.0) % 180.0
    return float(deg)


def load_l6(path: Path) -> Dict[str, Any]:
    with Dataset(path, "r") as nc:
        out: Dict[str, Any] = {
            "SlicesNo": np.asarray(nc.variables["SlicesNo"][:], dtype=int),
            "Amplitude": np.asarray(nc.variables["Amplitude"][:], dtype=float),
            "DominantWavelength": np.asarray(nc.variables["DominantWavelength"][:], dtype=float),
            "Angle": np.asarray(nc.variables["Angle"][:], dtype=float),
        }
        if "ClusterReconstruction" in nc.variables:
            out["ClusterReconstruction"] = np.asarray(nc.variables["ClusterReconstruction"][:], dtype=float)
        else:
            out["ClusterReconstruction"] = None
    return out


def get_cluster_maps(data: Dict[str, Any], idx: int) -> Dict[str, np.ndarray]:
    A = np.asarray(data["Amplitude"][idx], dtype=float)
    L = np.asarray(data["DominantWavelength"][idx], dtype=float)
    T = np.asarray(data["Angle"][idx], dtype=float)
    if data.get("ClusterReconstruction", None) is None:
        R = np.full_like(A, np.nan)
    else:
        R = np.asarray(data["ClusterReconstruction"][idx], dtype=float)
    return dict(R=R, A=A, L=L, T=T)


def compute_match_metrics(
    A_src: np.ndarray, L_src: np.ndarray, T_src: np.ndarray,
    A_cand: np.ndarray, L_cand: np.ndarray, T_cand: np.ndarray,
) -> Dict[str, Any]:
    mask_src = np.isfinite(A_src)
    mask_cand = np.isfinite(A_cand)

    inter = mask_src & mask_cand
    union = mask_src | mask_cand

    n_inter = int(np.count_nonzero(inter))
    n_union = int(np.count_nonzero(union))
    iou = (n_inter / n_union) if n_union > 0 else 0.0

    lam_src = L_src[inter]
    lam_cand = L_cand[inter]
    lam_src_med = nanpercentile_safe(lam_src, 50)
    lam_cand_med = nanpercentile_safe(lam_cand, 50)

    if np.isfinite(lam_src_med) and np.isfinite(lam_cand_med) and lam_src_med > 0 and lam_cand_med > 0:
        lam_ratio = max(lam_src_med, lam_cand_med) / min(lam_src_med, lam_cand_med)  # >=1
    else:
        lam_ratio = np.nan

    ang_src_deg = circ_median_deg_180(T_src, inter)
    ang_cand_deg = circ_median_deg_180(T_cand, inter)
    ang_diff_deg = circ_diff_deg_180(ang_src_deg, ang_cand_deg)

    return dict(
        n_intersection=n_inter,
        iou=float(iou),
        lam_src_med=float(lam_src_med),
        lam_cand_med=float(lam_cand_med),
        lam_ratio=float(lam_ratio),
        ang_src_deg=float(ang_src_deg),
        ang_cand_deg=float(ang_cand_deg),
        ang_diff_deg=float(ang_diff_deg),
    )


def passes_thresholds(m: Dict[str, Any]) -> bool:
    if m["iou"] < MIN_IOU:
        return False
    if m["n_intersection"] < MIN_INTERSECTION_PIX:
        return False
    if (not np.isfinite(m["lam_ratio"])) or (m["lam_ratio"] > MAX_LAMBDA_MED_RATIO):
        return False
    if (not np.isfinite(m["ang_diff_deg"])) or (m["ang_diff_deg"] > MAX_ANGLE_DIFF_DEG):
        return False
    return True


def score(m: Dict[str, Any]) -> float:
    ang = m["ang_diff_deg"]
    if not np.isfinite(ang):
        ang = 999.0

    r = m["lam_ratio"]
    if (not np.isfinite(r)) or r <= 0:
        return 0.0

    return float(m["iou"]) * (1.0 / float(r)) / (1.0 + float(ang))


def plot_side_by_side_4rows(
    left: Dict[str, np.ndarray],
    right: Dict[str, np.ndarray],
    title_left: str,
    title_right: str,
    outpath: Path,
    dx_km: float,
    dy_km: float,
) -> None:
    """
    4 rows x 2 cols:
      Row1: ClusterReconstruction  (independent scale per panel)
      Row2: Amplitude             (independent scale per panel)
      Row3: DominantWavelength    (shared scale across left/right)
      Row4: Angle                 (shared scale across left/right)
    """
    ny, nx = left["A"].shape
    x1d_km = np.arange(nx, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    # Row 1 independent
    Rl_vmin, Rl_vmax = robust_limits(left["R"])
    Rr_vmin, Rr_vmax = robust_limits(right["R"])

    # Row 2 independent
    Al_vmin, Al_vmax = robust_limits(left["A"])
    Ar_vmin, Ar_vmax = robust_limits(right["A"])

    # Row 3 shared
    L_vmin, L_vmax = robust_limits(np.concatenate([left["L"].ravel(), right["L"].ravel()]), 5, 95)

    # Row 4 shared
    T_vmin, T_vmax = robust_limits(np.concatenate([left["T"].ravel(), right["T"].ravel()]))

    fig, axs = plt.subplots(4, 2, figsize=(16, 14), constrained_layout=True, sharex=True, sharey=True)

    def _panel(ax, Z, vmin, vmax, label, cmap=None):
        im = ax.pcolormesh(x1d_km, y1d_km, Z, shading="auto", vmin=vmin, vmax=vmax, cmap=cmap)
        cb = fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01)
        cb.set_label(label)
        ax.set_aspect("equal", adjustable="box")

    axs[0, 0].set_title(title_left)
    axs[0, 1].set_title(title_right)

    # Recon
    _panel(axs[0, 0], left["R"], Rl_vmin, Rl_vmax, "arb", cmap="RdBu_r")
    _panel(axs[0, 1], right["R"], Rr_vmin, Rr_vmax, "arb", cmap="RdBu_r")
    axs[0, 0].set_ylabel("Recon\ny (km)")

    # Amp
    _panel(axs[1, 0], left["A"], Al_vmin, Al_vmax, "K")
    _panel(axs[1, 1], right["A"], Ar_vmin, Ar_vmax, "kR")
    axs[1, 0].set_ylabel("Amp\ny (km)")

    # Lambda
    _panel(axs[2, 0], left["L"], L_vmin, L_vmax, "km")
    _panel(axs[2, 1], right["L"], L_vmin, L_vmax, "km")
    axs[2, 0].set_ylabel("Lambda\ny (km)")

    # Angle
    _panel(axs[3, 0], left["T"], T_vmin, T_vmax, "rad")
    _panel(axs[3, 1], right["T"], T_vmin, T_vmax, "rad")
    axs[3, 0].set_ylabel("Angle\ny (km)")

    for r in range(4):
        for c in range(2):
            axs[r, c].set_xlabel("x (km)")

    save_fig(fig, outpath)


def combine_passed_l6a_reconstructions(
    l6a: Dict[str, Any],
    passed_a_indices: List[int],
    outdir: Path,
    slice_no: int,
) -> None:
    """
    Sum all ClusterReconstruction maps from passed L6A clusters.
    Saves:
      - combined reconstruction PNG
      - optional npy dump
    """
    if l6a.get("ClusterReconstruction", None) is None:
        (outdir / f"slice_{slice_no:03d}_COMBINED_L6A_recon_NO_RECON.txt").write_text(
            "L6A file does not contain ClusterReconstruction.\n"
            "Nothing to combine.\n"
        )
        return

    if len(passed_a_indices) == 0:
        (outdir / f"slice_{slice_no:03d}_COMBINED_L6A_recon_NONE_PASSED.txt").write_text(
            "No L6A clusters passed thresholds.\n"
            "Nothing to combine.\n"
        )
        return

    # Initialize
    first = np.asarray(l6a["ClusterReconstruction"][passed_a_indices[0]], dtype=float)
    sumR = np.zeros_like(first, dtype=float)

    n_used = 0
    for a_idx in passed_a_indices:
        R = np.asarray(l6a["ClusterReconstruction"][a_idx], dtype=float)
        if R.shape != sumR.shape:
            continue
        # sum, treating NaNs as 0
        sumR += np.nan_to_num(R, nan=0.0, posinf=0.0, neginf=0.0)
        n_used += 1

    ny, nx = sumR.shape
    x1d_km = np.arange(nx, dtype=float) * DX_KM
    y1d_km = np.arange(ny, dtype=float) * DY_KM

    vmin, vmax = robust_limits(sumR, 2, 98)

    fig, ax = plt.subplots(figsize=(14, 5), constrained_layout=True)
    im = ax.pcolormesh(x1d_km, y1d_km, sumR, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    ax.set_title(f"L6A slice {slice_no}: SUM of ClusterReconstruction over passed clusters (n={n_used})")
    ax.set_xlabel("x (km)")
    ax.set_ylabel("y (km)")
    ax.set_aspect("equal", adjustable="box")
    fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01).set_label("arb (sum)")

    outpng = outdir / f"slice_{slice_no:03d}_COMBINED_L6A_reconstruction_sum.png"
    save_fig(fig, outpng)

    # Optional: save array for later use
    np.save(outdir / f"slice_{slice_no:03d}_COMBINED_L6A_reconstruction_sum.npy", sumR)


# --------------------------
# Load files
# --------------------------
l6c = load_l6(L6C_PATH)
l6a = load_l6(L6A_PATH)

idx_c = np.where(l6c["SlicesNo"] == int(SLICE_NO))[0]
idx_a = np.where(l6a["SlicesNo"] == int(SLICE_NO))[0]

print(f"L6C slice {SLICE_NO}: {idx_c.size} clusters")
print(f"L6A slice {SLICE_NO}: {idx_a.size} clusters")

if idx_a.size == 0:
    raise ValueError(f"No L6A clusters found for slice {SLICE_NO}")
if idx_c.size == 0:
    raise ValueError(f"No L6C clusters found for slice {SLICE_NO}")

# --------------------------
# For each L6A cluster, find best matching L6C cluster
# --------------------------
all_candidate_rows: List[Dict[str, Any]] = []
best_rows: List[Dict[str, Any]] = []

pairs_dir = OUTDIR / "pairs"
pairs_dir.mkdir(parents=True, exist_ok=True)

for a_idx in idx_a:
    a_idx = int(a_idx)
    A_a = l6a["Amplitude"][a_idx]
    L_a = l6a["DominantWavelength"][a_idx]
    T_a = l6a["Angle"][a_idx]

    candidates: List[Dict[str, Any]] = []
    for c_idx in idx_c:
        c_idx = int(c_idx)
        A_c = l6c["Amplitude"][c_idx]
        L_c = l6c["DominantWavelength"][c_idx]
        T_c = l6c["Angle"][c_idx]

        m = compute_match_metrics(A_a, L_a, T_a, A_c, L_c, T_c)
        m["cluster_a"] = a_idx
        m["cluster_c"] = c_idx
        m["passes"] = bool(passes_thresholds(m))
        m["score"] = float(score(m))
        candidates.append(m)
        all_candidate_rows.append(m)

    candidates_sorted = sorted(candidates, key=lambda d: d["score"], reverse=True)
    best = candidates_sorted[0]

    if best["passes"] or ASSIGN_BEST_EVEN_IF_FAILS:
        best_rows.append(best)

        left_maps = get_cluster_maps(l6a, best["cluster_a"])
        right_maps = get_cluster_maps(l6c, best["cluster_c"])

        pass_tag = "PASS" if best["passes"] else "FAIL"
        outpng = pairs_dir / f"slice_{SLICE_NO:03d}_A_{best['cluster_a']:04d}_to_C_{best['cluster_c']:04d}_{pass_tag}.png"

        title_left = (
            f"L6A Temp | slice {SLICE_NO} | idx {best['cluster_a']} | "
            f"n={np.count_nonzero(np.isfinite(left_maps['A']))}"
        )
        title_right = (
            f"L6C Rad | slice {SLICE_NO} | idx {best['cluster_c']} | "
            f"IoU={best['iou']:.3f} inter={best['n_intersection']} "
            f"ratio={best['lam_ratio']:.3f} Δθ={best['ang_diff_deg']:.1f}°"
        )

        plot_side_by_side_4rows(
            left=left_maps,
            right=right_maps,
            title_left=title_left,
            title_right=title_right,
            outpath=outpng,
            dx_km=DX_KM,
            dy_km=DY_KM,
        )

# --------------------------
# Save matching info
# --------------------------
cand_path = OUTDIR / f"slice_{SLICE_NO:03d}_ALL_candidates_ratio.txt"
lines = []
lines.append(f"L6A file: {L6A_PATH}\n")
lines.append(f"L6C file: {L6C_PATH}\n")
lines.append(f"Slice: {SLICE_NO}\n\n")
lines.append(
    "cluster_a  cluster_c  passes  score    iou    inter  "
    "lam_ratio  ang_diff_deg  thetaA_deg  thetaC_deg  lamA_med  lamC_med\n"
)
for m in sorted(all_candidate_rows, key=lambda d: (d["cluster_a"], -d["score"])):
    lines.append(
        f"{m['cluster_a']:9d}  {m['cluster_c']:9d}  {int(m['passes']):6d}  {m['score']:6.3f}  "
        f"{m['iou']:.3f}  {m['n_intersection']:5d}  "
        f"{m['lam_ratio']:8.3f}  {m['ang_diff_deg']:12.1f}  "
        f"{m['ang_src_deg']:10.1f}  {m['ang_cand_deg']:10.1f}  "
        f"{m['lam_src_med']:8.1f}  {m['lam_cand_med']:8.1f}\n"
    )
cand_path.write_text("".join(lines))

best_path = OUTDIR / f"slice_{SLICE_NO:03d}_BEST_matches_A_to_C_ratio.txt"
lines = []
lines.append(f"L6A file: {L6A_PATH}\n")
lines.append(f"L6C file: {L6C_PATH}\n")
lines.append(f"Slice: {SLICE_NO}\n\n")
lines.append(
    "cluster_a  best_cluster_c  passes  score    iou    inter  "
    "lam_ratio  ang_diff_deg  thetaA_deg  thetaC_deg  lamA_med  lamC_med\n"
)
for m in sorted(best_rows, key=lambda d: d["cluster_a"]):
    lines.append(
        f"{m['cluster_a']:9d}  {m['cluster_c']:14d}  {int(m['passes']):6d}  {m['score']:6.3f}  "
        f"{m['iou']:.3f}  {m['n_intersection']:5d}  "
        f"{m['lam_ratio']:8.3f}  {m['ang_diff_deg']:12.1f}  "
        f"{m['ang_src_deg']:10.1f}  {m['ang_cand_deg']:10.1f}  "
        f"{m['lam_src_med']:8.1f}  {m['lam_cand_med']:8.1f}\n"
    )
best_path.write_text("".join(lines))

print("\nSaved:")
print("  candidates:", cand_path.resolve())
print("  best:", best_path.resolve())
print("  pair plots:", pairs_dir.resolve())
print(f"\nPlotted {len(best_rows)} pairs (ASSIGN_BEST_EVEN_IF_FAILS={ASSIGN_BEST_EVEN_IF_FAILS}).")

# --------------------------
# Combine PASSED L6A reconstructions
# --------------------------
passed_a_indices = [int(m["cluster_a"]) for m in best_rows if bool(m["passes"])]
combine_passed_l6a_reconstructions(l6a, passed_a_indices, OUTDIR, SLICE_NO)

print("\nCombined passed L6A reconstructions:")
print("  passed count:", len(passed_a_indices))
print("  outputs in:", OUTDIR.resolve())
```

## Match All Slices

```python
# %%
"""
ALL-SLICES matching + per-slice combined L6A recon + FULL-ORBIT stitched recon.

New (your request):
- After combining PASSED L6A ClusterReconstruction within each slice,
  stitch all slices in order to show the reconstruction of the whole orbit.

Assumptions for stitching:
- Each slice corresponds to one core tile laid left-to-right in x.
- All slices share the same (ny, nx) tile shape in ClusterReconstruction.
- Slice order is numeric SlicesNo (0,1,2,...). Missing slices are filled with NaNs.
"""

from __future__ import annotations

from pathlib import Path
from typing import Dict, Any, List, Tuple

import numpy as np
import matplotlib.pyplot as plt
from netCDF4 import Dataset

# --------------------------
# User params
# --------------------------
L6C_PATH = Path("awe_l6c_q20_2024165T2120_03175_v01.nc")  # Radiance-based clusters
L6A_PATH = Path("awe_l6a_tmp_2024165T2120_03175_v01.nc")  # Temperature-based clusters

OUTDIR = Path("outputs_ALLSLICES_l6a_to_l6c_matching_ratio_STITCH")
OUTDIR.mkdir(parents=True, exist_ok=True)
SAVE_DPI = 200

DX_KM = 2.0
DY_KM = 2.0

# --------------------------
# Matching thresholds (tune)
# --------------------------
MIN_IOU = 0.05
MIN_INTERSECTION_PIX = 50
MAX_LAMBDA_MED_RATIO = 1.25
MAX_ANGLE_DIFF_DEG = 15.0

# Save best match plot even if FAIL
ASSIGN_BEST_EVEN_IF_FAILS = True


# --------------------------
# Helpers
# --------------------------
def save_fig(fig: plt.Figure, outpath: Path) -> None:
    outpath.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(outpath, dpi=SAVE_DPI, bbox_inches="tight")
    plt.close(fig)


def robust_limits(A: np.ndarray, p_lo: float = 2, p_hi: float = 98) -> Tuple[float, float]:
    a = np.asarray(A)
    good = np.isfinite(a)
    if not np.any(good):
        return (-1.0, 1.0)
    lo, hi = np.nanpercentile(a[good], [p_lo, p_hi])
    if (not np.isfinite(lo)) or (not np.isfinite(hi)) or lo == hi:
        lo = float(np.nanmin(a[good]))
        hi = float(np.nanmax(a[good]))
    return (float(lo), float(hi))


def nanpercentile_safe(a: np.ndarray, q: float) -> float:
    good = np.isfinite(a)
    if not np.any(good):
        return np.nan
    return float(np.nanpercentile(a[good], q))


def circ_diff_deg_180(a_deg: float, b_deg: float) -> float:
    if (not np.isfinite(a_deg)) or (not np.isfinite(b_deg)):
        return np.nan
    d = abs(a_deg - b_deg) % 180.0
    return min(d, 180.0 - d)


def circ_median_deg_180(theta_rad: np.ndarray, mask: np.ndarray) -> float:
    th = np.asarray(theta_rad, dtype=float)
    m = mask & np.isfinite(th)
    if not np.any(m):
        return np.nan
    ang = th[m]
    z = np.exp(1j * 2.0 * ang)
    zmean = np.mean(z)
    if (not np.isfinite(zmean.real)) or (not np.isfinite(zmean.imag)) or abs(zmean) == 0:
        return np.nan
    th_mean = 0.5 * np.angle(zmean)
    deg = (np.degrees(th_mean) + 180.0) % 180.0
    return float(deg)


def load_l6(path: Path) -> Dict[str, Any]:
    with Dataset(path, "r") as nc:
        out: Dict[str, Any] = {
            "SlicesNo": np.asarray(nc.variables["SlicesNo"][:], dtype=int),
            "Amplitude": np.asarray(nc.variables["Amplitude"][:], dtype=float),
            "DominantWavelength": np.asarray(nc.variables["DominantWavelength"][:], dtype=float),
            "Angle": np.asarray(nc.variables["Angle"][:], dtype=float),
        }
        if "ClusterReconstruction" in nc.variables:
            out["ClusterReconstruction"] = np.asarray(nc.variables["ClusterReconstruction"][:], dtype=float)
        else:
            out["ClusterReconstruction"] = None
    return out


def get_cluster_maps(data: Dict[str, Any], idx: int) -> Dict[str, np.ndarray]:
    A = np.asarray(data["Amplitude"][idx], dtype=float)
    L = np.asarray(data["DominantWavelength"][idx], dtype=float)
    T = np.asarray(data["Angle"][idx], dtype=float)
    if data.get("ClusterReconstruction", None) is None:
        R = np.full_like(A, np.nan)
    else:
        R = np.asarray(data["ClusterReconstruction"][idx], dtype=float)
    return dict(R=R, A=A, L=L, T=T)


def compute_match_metrics(
    A_src: np.ndarray, L_src: np.ndarray, T_src: np.ndarray,
    A_cand: np.ndarray, L_cand: np.ndarray, T_cand: np.ndarray,
) -> Dict[str, Any]:
    mask_src = np.isfinite(A_src)
    mask_cand = np.isfinite(A_cand)
    inter = mask_src & mask_cand
    union = mask_src | mask_cand

    n_inter = int(np.count_nonzero(inter))
    n_union = int(np.count_nonzero(union))
    iou = (n_inter / n_union) if n_union > 0 else 0.0

    lam_src = L_src[inter]
    lam_cand = L_cand[inter]
    lam_src_med = nanpercentile_safe(lam_src, 50)
    lam_cand_med = nanpercentile_safe(lam_cand, 50)

    if np.isfinite(lam_src_med) and np.isfinite(lam_cand_med) and lam_src_med > 0 and lam_cand_med > 0:
        lam_ratio = max(lam_src_med, lam_cand_med) / min(lam_src_med, lam_cand_med)
    else:
        lam_ratio = np.nan

    ang_src_deg = circ_median_deg_180(T_src, inter)
    ang_cand_deg = circ_median_deg_180(T_cand, inter)
    ang_diff_deg = circ_diff_deg_180(ang_src_deg, ang_cand_deg)

    return dict(
        n_intersection=n_inter,
        iou=float(iou),
        lam_src_med=float(lam_src_med),
        lam_cand_med=float(lam_cand_med),
        lam_ratio=float(lam_ratio),
        ang_src_deg=float(ang_src_deg),
        ang_cand_deg=float(ang_cand_deg),
        ang_diff_deg=float(ang_diff_deg),
    )


def passes_thresholds(m: Dict[str, Any]) -> bool:
    if m["iou"] < MIN_IOU:
        return False
    if m["n_intersection"] < MIN_INTERSECTION_PIX:
        return False
    if (not np.isfinite(m["lam_ratio"])) or (m["lam_ratio"] > MAX_LAMBDA_MED_RATIO):
        return False
    if (not np.isfinite(m["ang_diff_deg"])) or (m["ang_diff_deg"] > MAX_ANGLE_DIFF_DEG):
        return False
    return True


def score(m: Dict[str, Any]) -> float:
    ang = m["ang_diff_deg"]
    if not np.isfinite(ang):
        ang = 999.0
    r = m["lam_ratio"]
    if (not np.isfinite(r)) or r <= 0:
        return 0.0
    return float(m["iou"]) * (1.0 / float(r)) / (1.0 + float(ang))


def plot_side_by_side_4rows(
    left: Dict[str, np.ndarray],
    right: Dict[str, np.ndarray],
    title_left: str,
    title_right: str,
    outpath: Path,
    dx_km: float,
    dy_km: float,
) -> None:
    """
    4 rows x 2 cols:
      Row1 Recon: independent left/right scaling
      Row2 Amp:   independent left/right scaling
      Row3 Lam:   shared scaling (left/right comparable)
      Row4 Angle: shared scaling (left/right comparable)
    """
    ny, nx = left["A"].shape
    x1d_km = np.arange(nx, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    Rl_vmin, Rl_vmax = robust_limits(left["R"])
    Rr_vmin, Rr_vmax = robust_limits(right["R"])
    Al_vmin, Al_vmax = robust_limits(left["A"])
    Ar_vmin, Ar_vmax = robust_limits(right["A"])

    L_vmin, L_vmax = robust_limits(np.concatenate([left["L"].ravel(), right["L"].ravel()]), 5, 95)
    T_vmin, T_vmax = robust_limits(np.concatenate([left["T"].ravel(), right["T"].ravel()]))

    fig, axs = plt.subplots(4, 2, figsize=(16, 14), constrained_layout=True, sharex=True, sharey=True)

    def _panel(ax, Z, vmin, vmax, label, cmap=None):
        im = ax.pcolormesh(x1d_km, y1d_km, Z, shading="auto", vmin=vmin, vmax=vmax, cmap=cmap)
        cb = fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01)
        cb.set_label(label)
        ax.set_aspect("equal", adjustable="box")

    axs[0, 0].set_title(title_left)
    axs[0, 1].set_title(title_right)

    _panel(axs[0, 0], left["R"], Rl_vmin, Rl_vmax, "arb", cmap="RdBu_r")
    _panel(axs[0, 1], right["R"], Rr_vmin, Rr_vmax, "arb", cmap="RdBu_r")
    axs[0, 0].set_ylabel("Recon\ny (km)")

    _panel(axs[1, 0], left["A"], Al_vmin, Al_vmax, "K")
    _panel(axs[1, 1], right["A"], Ar_vmin, Ar_vmax, "kR")
    axs[1, 0].set_ylabel("Amp\ny (km)")

    _panel(axs[2, 0], left["L"], L_vmin, L_vmax, "km")
    _panel(axs[2, 1], right["L"], L_vmin, L_vmax, "km")
    axs[2, 0].set_ylabel("Lambda\ny (km)")

    _panel(axs[3, 0], left["T"], T_vmin, T_vmax, "rad")
    _panel(axs[3, 1], right["T"], T_vmin, T_vmax, "rad")
    axs[3, 0].set_ylabel("Angle\ny (km)")

    for r in range(4):
        for c in range(2):
            axs[r, c].set_xlabel("x (km)")

    save_fig(fig, outpath)


def combine_passed_l6a_reconstructions_sum(
    l6a: Dict[str, Any],
    passed_a_indices: List[int],
) -> Tuple[np.ndarray | None, int]:
    """
    Returns:
      sumR (ny,nx) or None if no recon available
      n_used
    """
    if l6a.get("ClusterReconstruction", None) is None:
        return None, 0
    if len(passed_a_indices) == 0:
        return None, 0

    first = np.asarray(l6a["ClusterReconstruction"][passed_a_indices[0]], dtype=float)
    sumR = np.zeros_like(first, dtype=float)

    n_used = 0
    for a_idx in passed_a_indices:
        R = np.asarray(l6a["ClusterReconstruction"][a_idx], dtype=float)
        if R.shape != sumR.shape:
            continue
        sumR += np.nan_to_num(R, nan=0.0, posinf=0.0, neginf=0.0)
        n_used += 1

    return sumR, n_used


def save_combined_slice_recon(
    sumR: np.ndarray | None,
    n_used: int,
    outdir: Path,
    slice_no: int,
) -> None:
    if sumR is None:
        (outdir / f"slice_{slice_no:03d}_COMBINED_L6A_recon_NONE.txt").write_text(
            "No combined L6A reconstruction produced (no recon var or none passed).\n"
        )
        return

    ny, nx = sumR.shape
    x1d_km = np.arange(nx, dtype=float) * DX_KM
    y1d_km = np.arange(ny, dtype=float) * DY_KM
    vmin, vmax = robust_limits(sumR, 2, 98)

    fig, ax = plt.subplots(figsize=(14, 5), constrained_layout=True)
    im = ax.pcolormesh(x1d_km, y1d_km, sumR, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    ax.set_title(f"L6A slice {slice_no}: SUM ClusterReconstruction over passed clusters (n={n_used})")
    ax.set_xlabel("x (km)")
    ax.set_ylabel("y (km)")
    ax.set_aspect("equal", adjustable="box")
    fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01).set_label("arb (sum)")
    save_fig(fig, outdir / f"slice_{slice_no:03d}_COMBINED_L6A_reconstruction_sum.png")

    np.save(outdir / f"slice_{slice_no:03d}_COMBINED_L6A_reconstruction_sum.npy", sumR)


def stitch_slices_to_orbit(
    combined_by_slice: Dict[int, np.ndarray],
    slice_numbers: np.ndarray,
    outdir: Path,
) -> None:
    """
    Make a whole-orbit stitched reconstruction by concatenating along x.

    Missing slices => filled with NaNs.
    """
    if len(combined_by_slice) == 0:
        (outdir / "ORBIT_STITCH_NONE.txt").write_text("No per-slice combined reconstructions available.\n")
        return

    # Use first available slice for shape
    first_slice = sorted(combined_by_slice.keys())[0]
    ny, nx = combined_by_slice[first_slice].shape

    # Ensure we stitch over a contiguous slice index range (helps if missing slices)
    smin = int(np.min(slice_numbers))
    smax = int(np.max(slice_numbers))
    full_slices = list(range(smin, smax + 1))

    blocks: List[np.ndarray] = []
    n_present = 0
    for s in full_slices:
        if s in combined_by_slice and combined_by_slice[s] is not None:
            A = combined_by_slice[s]
            if A.shape != (ny, nx):
                A = np.full((ny, nx), np.nan)
            else:
                n_present += 1
            blocks.append(A)
        else:
            blocks.append(np.full((ny, nx), np.nan))

    orbit = np.concatenate(blocks, axis=1)  # (ny, nx * n_slices)

    # Axes
    x1d_km = np.arange(orbit.shape[1], dtype=float) * DX_KM
    y1d_km = np.arange(orbit.shape[0], dtype=float) * DY_KM

    vmin, vmax = robust_limits(orbit, 2, 98)

    fig, ax = plt.subplots(figsize=(20, 6), constrained_layout=True)
    im = ax.pcolormesh(x1d_km, y1d_km, orbit, shading="auto", cmap="RdBu_r", vmin=vmin, vmax=vmax)
    ax.set_title(
        f"Whole-orbit stitched L6A reconstruction (sum of passed clusters per slice)\n"
        f"slices {smin}–{smax} (present {n_present}/{len(full_slices)})"
    )
    ax.set_xlabel("x (km)")
    ax.set_ylabel("y (km)")
    ax.set_aspect("equal", adjustable="box")
    fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01).set_label("arb (stitched sum)")
    save_fig(fig, outdir / "ORBIT_STITCHED_L6A_reconstruction_sum.png")

    np.save(outdir / "ORBIT_STITCHED_L6A_reconstruction_sum.npy", orbit)


# --------------------------
# Main
# --------------------------
l6c = load_l6(L6C_PATH)
l6a = load_l6(L6A_PATH)

all_slices_a = np.unique(l6a["SlicesNo"]).astype(int)
all_slices_a = all_slices_a[all_slices_a >= 0]
all_slices_a = np.sort(all_slices_a)

print("Slices in L6A:", all_slices_a.tolist())

summary_lines: List[str] = []
summary_lines.append(f"L6A: {L6A_PATH}\n")
summary_lines.append(f"L6C: {L6C_PATH}\n\n")
summary_lines.append("slice  nA  nC  nBestSaved  nPassed  note\n")

# Store per-slice combined recon for orbit stitching
combined_by_slice: Dict[int, np.ndarray] = {}

for slice_no in all_slices_a:
    idx_a = np.where(l6a["SlicesNo"] == int(slice_no))[0]
    idx_c = np.where(l6c["SlicesNo"] == int(slice_no))[0]

    slice_dir = OUTDIR / f"slice_{slice_no:03d}"
    pairs_dir = slice_dir / "pairs"
    pairs_dir.mkdir(parents=True, exist_ok=True)

    if idx_a.size == 0:
        summary_lines.append(f"{slice_no:5d}  {0:2d}  {idx_c.size:2d}  {0:10d}  {0:7d}  no L6A clusters\n")
        continue
    if idx_c.size == 0:
        (slice_dir / "NO_L6C_CLUSTERS.txt").write_text(f"No L6C clusters for slice {slice_no}\n")
        summary_lines.append(f"{slice_no:5d}  {idx_a.size:2d}  {0:2d}  {0:10d}  {0:7d}  no L6C clusters\n")
        continue

    all_candidate_rows: List[Dict[str, Any]] = []
    best_rows: List[Dict[str, Any]] = []

    for a_idx in idx_a:
        a_idx = int(a_idx)
        A_a = l6a["Amplitude"][a_idx]
        L_a = l6a["DominantWavelength"][a_idx]
        T_a = l6a["Angle"][a_idx]

        candidates: List[Dict[str, Any]] = []
        for c_idx in idx_c:
            c_idx = int(c_idx)
            A_c = l6c["Amplitude"][c_idx]
            L_c = l6c["DominantWavelength"][c_idx]
            T_c = l6c["Angle"][c_idx]

            m = compute_match_metrics(A_a, L_a, T_a, A_c, L_c, T_c)
            m["cluster_a"] = a_idx
            m["cluster_c"] = c_idx
            m["passes"] = bool(passes_thresholds(m))
            m["score"] = float(score(m))
            candidates.append(m)
            all_candidate_rows.append(m)

        passed = [c for c in candidates if c["passes"]]

        if passed:
            best = max(passed, key=lambda d: d["score"])
        else:
            best = max(candidates, key=lambda d: d["score"])


        if best["passes"] or ASSIGN_BEST_EVEN_IF_FAILS:
            best_rows.append(best)

            left_maps = get_cluster_maps(l6a, best["cluster_a"])
            right_maps = get_cluster_maps(l6c, best["cluster_c"])

            pass_tag = "PASS" if best["passes"] else "FAIL"
            outpng = pairs_dir / f"slice_{slice_no:03d}_A_{best['cluster_a']:04d}_to_C_{best['cluster_c']:04d}_{pass_tag}.png"

            title_left = (
                f"L6A Temp | slice {slice_no} | idx {best['cluster_a']} | "
                f"n={np.count_nonzero(np.isfinite(left_maps['A']))}"
            )
            title_right = (
                f"L6C Rad | slice {slice_no} | idx {best['cluster_c']} | "
                f"IoU={best['iou']:.3f} inter={best['n_intersection']} "
                f"ratio={best['lam_ratio']:.3f} Δθ={best['ang_diff_deg']:.1f}°"
            )

            plot_side_by_side_4rows(
                left=left_maps,
                right=right_maps,
                title_left=title_left,
                title_right=title_right,
                outpath=outpng,
                dx_km=DX_KM,
                dy_km=DY_KM,
            )

    # Save tables for this slice
    cand_path = slice_dir / f"slice_{slice_no:03d}_ALL_candidates_ratio.txt"
    lines = []
    lines.append(f"L6A file: {L6A_PATH}\n")
    lines.append(f"L6C file: {L6C_PATH}\n")
    lines.append(f"Slice: {slice_no}\n\n")
    lines.append(
        "cluster_a  cluster_c  passes  score    iou    inter  "
        "lam_ratio  ang_diff_deg  thetaA_deg  thetaC_deg  lamA_med  lamC_med\n"
    )
    for m in sorted(all_candidate_rows, key=lambda d: (d["cluster_a"], -d["score"])):
        lines.append(
            f"{m['cluster_a']:9d}  {m['cluster_c']:9d}  {int(m['passes']):6d}  {m['score']:6.3f}  "
            f"{m['iou']:.3f}  {m['n_intersection']:5d}  "
            f"{m['lam_ratio']:8.3f}  {m['ang_diff_deg']:12.1f}  "
            f"{m['ang_src_deg']:10.1f}  {m['ang_cand_deg']:10.1f}  "
            f"{m['lam_src_med']:8.1f}  {m['lam_cand_med']:8.1f}\n"
        )
    cand_path.write_text("".join(lines))

    best_path = slice_dir / f"slice_{slice_no:03d}_BEST_matches_A_to_C_ratio.txt"
    lines = []
    lines.append(f"L6A file: {L6A_PATH}\n")
    lines.append(f"L6C file: {L6C_PATH}\n")
    lines.append(f"Slice: {slice_no}\n\n")
    lines.append(
        "cluster_a  best_cluster_c  passes  score    iou    inter  "
        "lam_ratio  ang_diff_deg  thetaA_deg  thetaC_deg  lamA_med  lamC_med\n"
    )
    for m in sorted(best_rows, key=lambda d: d["cluster_a"]):
        lines.append(
            f"{m['cluster_a']:9d}  {m['cluster_c']:14d}  {int(m['passes']):6d}  {m['score']:6.3f}  "
            f"{m['iou']:.3f}  {m['n_intersection']:5d}  "
            f"{m['lam_ratio']:8.3f}  {m['ang_diff_deg']:12.1f}  "
            f"{m['ang_src_deg']:10.1f}  {m['ang_cand_deg']:10.1f}  "
            f"{m['lam_src_med']:8.1f}  {m['lam_cand_med']:8.1f}\n"
        )
    best_path.write_text("".join(lines))

    passed_a_indices = [int(m["cluster_a"]) for m in best_rows if bool(m["passes"])]

    sumR, n_used = combine_passed_l6a_reconstructions_sum(l6a, passed_a_indices)
    save_combined_slice_recon(sumR, n_used, slice_dir, slice_no)

    if sumR is not None:
        combined_by_slice[int(slice_no)] = sumR

    summary_lines.append(
        f"{slice_no:5d}  {idx_a.size:2d}  {idx_c.size:2d}  {len(best_rows):10d}  {len(passed_a_indices):7d}  ok\n"
    )

summary_path = OUTDIR / "SUMMARY_all_slices.txt"
summary_path.write_text("".join(summary_lines))

# ---- Whole-orbit stitched plot ----
stitch_slices_to_orbit(combined_by_slice, all_slices_a, OUTDIR)

print("\nDone. Outputs in:", OUTDIR.resolve())
print("Summary:", summary_path.resolve())
print("Orbit stitched PNG/NPY saved in OUTDIR.")
```

## Add U, V, and NSquared. 

```python
# %%
"""
ALL-SLICES matching + per-slice combined L6A recon + FULL-ORBIT stitched recon
+ slice Temperature + momentum flux for PASSED clusters only
+ sum MF per slice + stitch MF across orbit
+ output NetCDF contains original L6A global attrs + variables, but ONLY PASSED clusters
  are kept for cluster-dimension variables.

Momentum flux (per cluster pixel, masked to cluster area):
MF = 1/2 * g^2 / N^2 * (lambda_z / lambda_x) * (Amplitude / Temperature)^2 * (1/c^2)

Constants:
g = 9.52 m/s^2
lambda_z = 10 km
c = 1
N^2 = 0.0004 everywhere (synthetic)
u=v=0 everywhere (synthetic)

Key rules:
- SliceNo already exists in L6A, keep it (but subset to passed clusters)
- Area, Amplitude, DominantWavelength, Angle already exist (subset to passed clusters)
- Also slice Temperature into Temp_cluster (passed clusters only)
- Only save passed clusters (output file cluster vars are subset)
- Add MF_cluster, MF_slice_sum, MF_toprows (renamed from MF_full)
"""

from __future__ import annotations

from pathlib import Path
from typing import Dict, Any, List, Tuple

import numpy as np
import matplotlib.pyplot as plt
from netCDF4 import Dataset

# --------------------------
# User params
# --------------------------
L6C_PATH = Path("awe_l6c_q20_2024165T2120_03175_v01.nc")  # Radiance-based clusters
L6A_PATH = Path("awe_l6a_tmp_2024165T2120_03175_v01.nc")  # Temperature-based clusters

OUTDIR = Path("outputs_ALLSLICES_l6a_to_l6c_matching_ratio_STITCH_MF")
OUTDIR.mkdir(parents=True, exist_ok=True)
SAVE_DPI = 200

DX_KM = 2.0
DY_KM = 2.0

# ---- Synthetic u,v,N^2 settings ----
TEMP_VAR_NAME = "Temperature"  # variable name in L6A
TIME_INDEX = 0                 # which t index to use when slicing
N2_VALUE = 4e-4                # N=0.02 -> N^2=0.0004
U_VALUE = 0.0
V_VALUE = 0.0

# ---- Momentum flux constants ----
G_MS2 = 9.52
LAMBDA_Z_KM = 10.0
C_CANCEL = 1.0
CP_JKGK = 1005.0


# Output NetCDF (contains original L6A attrs+vars, but cluster vars subset to PASSED only)
OUT_L6A_PASSED = OUTDIR / f"{L6A_PATH.stem}_PASSED_with_u_v_NS2_TempCluster_MF.nc"

# --------------------------
# Matching thresholds (tune)
# --------------------------
MIN_IOU = 0.05
MIN_INTERSECTION_PIX = 50
MAX_LAMBDA_MED_RATIO = 1.25
MAX_ANGLE_DIFF_DEG = 15.0

# Save best match plot even if FAIL (does NOT affect passing)
ASSIGN_BEST_EVEN_IF_FAILS = True


# --------------------------
# Helpers
# --------------------------
def save_fig(fig: plt.Figure, outpath: Path) -> None:
    outpath.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(outpath, dpi=SAVE_DPI, bbox_inches="tight")
    plt.close(fig)


def robust_limits(A: np.ndarray, p_lo: float = 2, p_hi: float = 98) -> Tuple[float, float]:
    a = np.asarray(A)
    good = np.isfinite(a)
    if not np.any(good):
        return (-1.0, 1.0)
    lo, hi = np.nanpercentile(a[good], [p_lo, p_hi])
    if (not np.isfinite(lo)) or (not np.isfinite(hi)) or lo == hi:
        lo = float(np.nanmin(a[good]))
        hi = float(np.nanmax(a[good]))
    return (float(lo), float(hi))


def nanpercentile_safe(a: np.ndarray, q: float) -> float:
    good = np.isfinite(a)
    if not np.any(good):
        return np.nan
    return float(np.nanpercentile(a[good], q))


def circ_diff_deg_180(a_deg: float, b_deg: float) -> float:
    if (not np.isfinite(a_deg)) or (not np.isfinite(b_deg)):
        return np.nan
    d = abs(a_deg - b_deg) % 180.0
    return min(d, 180.0 - d)


def circ_median_deg_180(theta_rad: np.ndarray, mask: np.ndarray) -> float:
    th = np.asarray(theta_rad, dtype=float)
    m = mask & np.isfinite(th)
    if not np.any(m):
        return np.nan
    ang = th[m]
    z = np.exp(1j * 2.0 * ang)
    zmean = np.mean(z)
    if (not np.isfinite(zmean.real)) or (not np.isfinite(zmean.imag)) or abs(zmean) == 0:
        return np.nan
    th_mean = 0.5 * np.angle(zmean)
    deg = (np.degrees(th_mean) + 180.0) % 180.0
    return float(deg)


def load_l6_for_matching(path: Path) -> Dict[str, Any]:
    with Dataset(path, "r") as nc:
        out: Dict[str, Any] = {
            "SlicesNo": np.asarray(nc.variables["SlicesNo"][:], dtype=int),
            "Amplitude": np.asarray(nc.variables["Amplitude"][:], dtype=float),
            "DominantWavelength": np.asarray(nc.variables["DominantWavelength"][:], dtype=float),
            "Angle": np.asarray(nc.variables["Angle"][:], dtype=float),
        }
        if "ClusterReconstruction" in nc.variables:
            out["ClusterReconstruction"] = np.asarray(nc.variables["ClusterReconstruction"][:], dtype=float)
        else:
            out["ClusterReconstruction"] = None
    return out


def get_cluster_maps(data: Dict[str, Any], idx: int) -> Dict[str, np.ndarray]:
    A = np.asarray(data["Amplitude"][idx], dtype=float)
    L = np.asarray(data["DominantWavelength"][idx], dtype=float)
    T = np.asarray(data["Angle"][idx], dtype=float)
    if data.get("ClusterReconstruction", None) is None:
        R = np.full_like(A, np.nan)
    else:
        R = np.asarray(data["ClusterReconstruction"][idx], dtype=float)
    return dict(R=R, A=A, L=L, T=T)


def compute_match_metrics(
    A_src: np.ndarray, L_src: np.ndarray, T_src: np.ndarray,
    A_cand: np.ndarray, L_cand: np.ndarray, T_cand: np.ndarray,
) -> Dict[str, Any]:
    mask_src = np.isfinite(A_src)
    mask_cand = np.isfinite(A_cand)
    inter = mask_src & mask_cand
    union = mask_src | mask_cand

    n_inter = int(np.count_nonzero(inter))
    n_union = int(np.count_nonzero(union))
    iou = (n_inter / n_union) if n_union > 0 else 0.0

    lam_src = L_src[inter]
    lam_cand = L_cand[inter]
    lam_src_med = nanpercentile_safe(lam_src, 50)
    lam_cand_med = nanpercentile_safe(lam_cand, 50)

    if np.isfinite(lam_src_med) and np.isfinite(lam_cand_med) and lam_src_med > 0 and lam_cand_med > 0:
        lam_ratio = max(lam_src_med, lam_cand_med) / min(lam_src_med, lam_cand_med)
    else:
        lam_ratio = np.nan

    ang_src_deg = circ_median_deg_180(T_src, inter)
    ang_cand_deg = circ_median_deg_180(T_cand, inter)
    ang_diff_deg = circ_diff_deg_180(ang_src_deg, ang_cand_deg)

    return dict(
        n_intersection=n_inter,
        iou=float(iou),
        lam_src_med=float(lam_src_med),
        lam_cand_med=float(lam_cand_med),
        lam_ratio=float(lam_ratio),
        ang_src_deg=float(ang_src_deg),
        ang_cand_deg=float(ang_cand_deg),
        ang_diff_deg=float(ang_diff_deg),
    )


def passes_thresholds(m: Dict[str, Any]) -> bool:
    if m["iou"] < MIN_IOU:
        return False
    if m["n_intersection"] < MIN_INTERSECTION_PIX:
        return False
    if (not np.isfinite(m["lam_ratio"])) or (m["lam_ratio"] > MAX_LAMBDA_MED_RATIO):
        return False
    if (not np.isfinite(m["ang_diff_deg"])) or (m["ang_diff_deg"] > MAX_ANGLE_DIFF_DEG):
        return False
    return True


def score(m: Dict[str, Any]) -> float:
    ang = m["ang_diff_deg"]
    if not np.isfinite(ang):
        ang = 999.0
    r = m["lam_ratio"]
    if (not np.isfinite(r)) or r <= 0:
        return 0.0
    return float(m["iou"]) * (1.0 / float(r)) / (1.0 + float(ang))


def plot_side_by_side_4rows(
    left: Dict[str, np.ndarray],
    right: Dict[str, np.ndarray],
    title_left: str,
    title_right: str,
    outpath: Path,
    dx_km: float,
    dy_km: float,
) -> None:
    ny, nx = left["A"].shape
    x1d_km = np.arange(nx, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    Rl_vmin, Rl_vmax = robust_limits(left["R"])
    Rr_vmin, Rr_vmax = robust_limits(right["R"])
    Al_vmin, Al_vmax = robust_limits(left["A"])
    Ar_vmin, Ar_vmax = robust_limits(right["A"])

    L_vmin, L_vmax = robust_limits(np.concatenate([left["L"].ravel(), right["L"].ravel()]), 5, 95)
    T_vmin, T_vmax = robust_limits(np.concatenate([left["T"].ravel(), right["T"].ravel()]))

    fig, axs = plt.subplots(4, 2, figsize=(16, 14), constrained_layout=True, sharex=True, sharey=True)

    def _panel(ax, Z, vmin, vmax, label, cmap=None):
        im = ax.pcolormesh(x1d_km, y1d_km, Z, shading="auto", vmin=vmin, vmax=vmax, cmap=cmap)
        cb = fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01)
        cb.set_label(label)
        ax.set_aspect("equal", adjustable="box")

    axs[0, 0].set_title(title_left)
    axs[0, 1].set_title(title_right)

    _panel(axs[0, 0], left["R"], Rl_vmin, Rl_vmax, "arb", cmap="RdBu_r")
    _panel(axs[0, 1], right["R"], Rr_vmin, Rr_vmax, "arb", cmap="RdBu_r")
    axs[0, 0].set_ylabel("Recon\ny (km)")

    _panel(axs[1, 0], left["A"], Al_vmin, Al_vmax, "K")
    _panel(axs[1, 1], right["A"], Ar_vmin, Ar_vmax, "kR")
    axs[1, 0].set_ylabel("Amp\ny (km)")

    _panel(axs[2, 0], left["L"], L_vmin, L_vmax, "km")
    _panel(axs[2, 1], right["L"], L_vmin, L_vmax, "km")
    axs[2, 0].set_ylabel("Lambda\ny (km)")

    _panel(axs[3, 0], left["T"], T_vmin, T_vmax, "rad")
    _panel(axs[3, 1], right["T"], T_vmin, T_vmax, "rad")
    axs[3, 0].set_ylabel("Angle\ny (km)")

    for r in range(4):
        for c in range(2):
            axs[r, c].set_xlabel("x (km)")

    save_fig(fig, outpath)


# --------------------------
# Flatten swath BEFORE slicing
# --------------------------
def flatten_swath(Z: np.ndarray, target_ny: int | None = None) -> np.ndarray:
    ny, nx = Z.shape
    out = np.full_like(Z, np.nan)

    bottoms = np.full(nx, np.nan, dtype=float)
    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size:
            bottoms[j] = valid_idx.max()

    valid_bottoms = bottoms[np.isfinite(bottoms)]
    if valid_bottoms.size == 0:
        med = float(np.nanmedian(Z)) if np.isfinite(np.nanmedian(Z)) else 0.0
        out = np.nan_to_num(Z, nan=med, posinf=med, neginf=med)
        if target_ny is not None and ny > target_ny:
            out = out[-target_ny:, :]
        return out

    max_bottom = int(valid_bottoms.max())

    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size == 0:
            continue
        bottom = int(valid_idx.max())
        shift = max_bottom - bottom
        if shift == 0:
            out[:, j] = col
        elif shift > 0:
            out[shift:, j] = col[: ny - shift]
        else:
            shift_up = -shift
            out[: ny - shift_up, j] = col[shift_up:]

    med = float(np.nanmedian(out)) if np.isfinite(np.nanmedian(out)) else 0.0
    out = np.nan_to_num(out, nan=med, posinf=med, neginf=med)

    if target_ny is not None and ny > target_ny:
        start = max_bottom + 1 - target_ny
        start = max(start, 0)
        end = start + target_ny
        out = out[start:end, :]

    return out


# --------------------------
# L6A cloning with PASSED-only cluster variables + new vars
# --------------------------
def create_passed_only_l6a_with_new_vars(
    src_l6a: Path,
    dst_l6a: Path,
    passed_cluster_indices: np.ndarray,
    time_index: int,
) -> Tuple[str, str, str]:
    """
    Creates dst_l6a with:
      - all src global attrs
      - all variables copied; cluster-dimension variables subset to passed clusters only
      - ensures full-field u,v,NSquared exist (t,y,x) (synthetic constants if missing)
      - adds cluster-tile vars for passed clusters:
          Temp_cluster, u_cluster, v_cluster, NSquared_cluster, MF_cluster
      - sums MF over passed clusters per slice: MF_slice_sum(slice,yTile,xTile)
      - writes MF_toprows(t,y,x) on the swath grid for convenience:
          MF_toprows[time_index, 0:yTile, x0:x1] = MF_slice_sum for slice
          missing slices remain NaN
        NOTE: MF_toprows is NOT an "unflattened" swath mapping. It's just placed in top rows.
    """
    passed_cluster_indices = np.asarray(passed_cluster_indices, dtype=int)
    if passed_cluster_indices.size == 0:
        raise ValueError("No passed clusters. Output would be empty.")

    dst_l6a.parent.mkdir(parents=True, exist_ok=True)

    with Dataset(src_l6a, "r") as src:
        if "Amplitude" not in src.variables or "SlicesNo" not in src.variables:
            raise KeyError("L6A must contain Amplitude and SlicesNo.")
        if TEMP_VAR_NAME not in src.variables:
            raise KeyError(f"L6A must contain {TEMP_VAR_NAME}.")

        amp_var = src.variables["Amplitude"]
        amp_dims = amp_var.dimensions
        if len(amp_dims) < 3:
            raise ValueError("Amplitude is expected to be (cluster, yTile, xTile).")
        cluster_dim = amp_dims[0]
        ytile_dim = amp_dims[1]
        xtile_dim = amp_dims[2]

        # Full-field temperature (t,y,x) and a 2D slice at time_index
        temp_var = src.variables[TEMP_VAR_NAME]
        temp_dims = temp_var.dimensions           # (t,y,x)
        temp_shape = temp_var.shape               # (t,y,x)

        temp_raw_2d = np.asarray(temp_var[time_index], dtype=float)  # (yFull,xFull)

        # Tile sizes from amplitude
        amp_all = np.asarray(src.variables["Amplitude"][:], dtype=float)  # (nCluster,yTile,xTile)
        n_cluster_all, y_tile, x_tile = amp_all.shape

        slices_no_all = np.asarray(src.variables["SlicesNo"][:], dtype=int)

        # Synthetic / existing full-field u/v/N^2 at time_index
        if "u" in src.variables:
            u_raw_2d = np.asarray(src.variables["u"][time_index], dtype=float)
        else:
            u_raw_2d = np.full_like(temp_raw_2d, U_VALUE, dtype=float)

        if "v" in src.variables:
            v_raw_2d = np.asarray(src.variables["v"][time_index], dtype=float)
        else:
            v_raw_2d = np.full_like(temp_raw_2d, V_VALUE, dtype=float)

        if "NSquared" in src.variables:
            n2_raw_2d = np.asarray(src.variables["NSquared"][time_index], dtype=float)
        else:
            n2_raw_2d = np.full_like(temp_raw_2d, N2_VALUE, dtype=float)

        # --------------------------
        # Flatten FIRST (and crop to exactly y_tile rows)
        # --------------------------
        temp_full_2d = flatten_swath(temp_raw_2d, target_ny=y_tile)
        u_full_2d    = flatten_swath(u_raw_2d,    target_ny=y_tile)
        v_full_2d    = flatten_swath(v_raw_2d,    target_ny=y_tile)
        n2_full_2d   = flatten_swath(n2_raw_2d,   target_ny=y_tile)

        # Sanity check mapping width for passed clusters
        smax = int(np.nanmax(slices_no_all[passed_cluster_indices]))
        x_need = (smax + 1) * x_tile
        if temp_full_2d.shape[1] < x_need:
            raise ValueError(
                "Flattened full-field x dimension too small for SliceNo mapping. "
                f"Need >= {x_need}, got {temp_full_2d.shape[1]}."
            )
        if temp_full_2d.shape[0] != y_tile:
            raise ValueError(
                f"Flattened y dimension mismatch: got {temp_full_2d.shape[0]}, expected y_tile={y_tile}."
            )

        n_passed = int(passed_cluster_indices.size)

        with Dataset(dst_l6a, "w") as dst:
            # ---- Global attrs
            for aname in src.ncattrs():
                dst.setncattr(aname, src.getncattr(aname))

            # ---- Dimensions (clone all, but override cluster length)
            for dname, dim in src.dimensions.items():
                if dname == cluster_dim:
                    dst.createDimension(dname, n_passed)
                else:
                    dst.createDimension(dname, (None if dim.isunlimited() else len(dim)))

            # ---- Copy variables (subset if they depend on cluster_dim)
            for vname, vsrc in src.variables.items():
                fill_value = vsrc.getncattr("_FillValue") if "_FillValue" in vsrc.ncattrs() else None
                vdst = dst.createVariable(
                    vname, vsrc.dtype, vsrc.dimensions,
                    zlib=True, complevel=4, fill_value=fill_value
                )

                for aname in vsrc.ncattrs():
                    if aname == "_FillValue":
                        continue
                    vdst.setncattr(aname, vsrc.getncattr(aname))

                if cluster_dim in vsrc.dimensions:
                    ax = vsrc.dimensions.index(cluster_dim)
                    data = vsrc[:]
                    data_sub = np.take(data, passed_cluster_indices, axis=ax)
                    vdst[:] = data_sub
                else:
                    vdst[:] = vsrc[:]

            # ---- Ensure full-field u,v,NSquared exist in dst (t,y,x)
            # NOTE: these are NOT flattened; they are synthetic constants on the swath grid if missing.
            def _ensure_full_field(name: str, value: float, units: str) -> None:
                if name in dst.variables:
                    return
                vv = dst.createVariable(name, "f4", temp_dims, zlib=True, complevel=4)
                vv.setncattr("long_name", name)
                vv.setncattr("units", units)
                vv[:] = np.full(temp_shape, value, dtype=np.float32)

            _ensure_full_field("u", float(U_VALUE), "m/s")
            _ensure_full_field("v", float(V_VALUE), "m/s")
            _ensure_full_field("NSquared", float(N2_VALUE), "1/s^2")

            # ---- Build per-passed-cluster vars (cluster,yTile,xTile)
            temp_cluster = np.full((n_passed, y_tile, x_tile), np.nan, dtype=np.float32)
            u_cluster = np.full_like(temp_cluster, np.nan)
            v_cluster = np.full_like(temp_cluster, np.nan)
            n2_cluster = np.full_like(temp_cluster, np.nan)
            mf_cluster = np.full_like(temp_cluster, np.nan)

            amp_passed = np.asarray(dst.variables["Amplitude"][:], dtype=float)  # (n_passed,yTile,xTile)
            lam_passed = np.asarray(dst.variables["DominantWavelength"][:], dtype=float)  # lambda_x (km)
            slice_passed = np.asarray(dst.variables["SlicesNo"][:], dtype=int)

            for j in range(n_passed):
                s = int(slice_passed[j])
                if s < 0:
                    continue
                x0 = s * x_tile
                x1 = x0 + x_tile

                # These are flattened already and exactly y_tile high
                Ttile = temp_full_2d[:, x0:x1]
                Utile = u_full_2d[:, x0:x1]
                Vtile = v_full_2d[:, x0:x1]
                N2tile = n2_full_2d[:, x0:x1]

                mask = np.isfinite(amp_passed[j])  # cluster area mask

                Tm = np.full((y_tile, x_tile), np.nan, dtype=np.float32)
                Um = np.full((y_tile, x_tile), np.nan, dtype=np.float32)
                Vm = np.full((y_tile, x_tile), np.nan, dtype=np.float32)
                N2m = np.full((y_tile, x_tile), np.nan, dtype=np.float32)

                Tm[mask] = Ttile[mask].astype(np.float32)
                Um[mask] = Utile[mask].astype(np.float32)
                Vm[mask] = Vtile[mask].astype(np.float32)
                N2m[mask] = N2tile[mask].astype(np.float32)

                temp_cluster[j] = Tm
                u_cluster[j] = Um
                v_cluster[j] = Vm
                n2_cluster[j] = N2m

                # MF per pixel (masked), lambda_x = DominantWavelength (km)
                A = amp_passed[j]
                Lx = lam_passed[j]  # km
                MF = np.full((y_tile, x_tile), np.nan, dtype=np.float32)

                good = (
                    mask
                    & np.isfinite(Tm) & (Tm != 0)
                    & np.isfinite(Lx) & (Lx > 0)
                    & np.isfinite(N2m) & (N2m > 0)
                )
                if np.any(good):
                    ratio2 = (A[good] / Tm[good]) ** 2
                    MF_val = (
                        0.5
                        * (G_MS2 ** 2)
                        / N2m[good]
                        * (LAMBDA_Z_KM / Lx[good])
                        * ratio2
                        * (1.0 / (C_CANCEL ** 2))
                    )
                    MF[good] = MF_val.astype(np.float32)

                mf_cluster[j] = MF

            # Write new cluster vars
            vT = dst.createVariable("Temp_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim), zlib=True, complevel=4)
            vT.setncattr("long_name", "Temperature sliced to cluster tile and masked to cluster area (flattened first)")
            vT.setncattr("units", getattr(src.variables[TEMP_VAR_NAME], "units", "K"))
            vT[:] = temp_cluster

            vu = dst.createVariable("u_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim), zlib=True, complevel=4)
            vu.setncattr("units", "m/s")
            vu.setncattr("note", "Sliced from flattened u field; masked to cluster area")
            vu[:] = u_cluster

            vv = dst.createVariable("v_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim), zlib=True, complevel=4)
            vv.setncattr("units", "m/s")
            vv.setncattr("note", "Sliced from flattened v field; masked to cluster area")
            vv[:] = v_cluster

            vn2 = dst.createVariable("NSquared_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim), zlib=True, complevel=4)
            vn2.setncattr("units", "1/s^2")
            vn2.setncattr("note", "Sliced from flattened N^2 field; masked to cluster area")
            vn2[:] = n2_cluster

            vmf = dst.createVariable("MF_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim), zlib=True, complevel=4)
            vmf.setncattr("long_name", "Momentum flux proxy (per pixel) for passed clusters, masked to cluster area")
            vmf.setncattr("formula", "0.5*g^2/N^2*(lambda_z/lambda_x)*(A/T)^2*(1/c^2)")
            vmf.setncattr("lambda_x_source", "DominantWavelength (km)")
            vmf.setncattr("g_ms2", float(G_MS2))
            vmf.setncattr("lambda_z_km", float(LAMBDA_Z_KM))
            vmf.setncattr("c_cancel", float(C_CANCEL))
            vmf[:] = mf_cluster

            # ---- Sum MF over passed clusters per slice
            passed_slices = np.unique(slice_passed[slice_passed >= 0]).astype(int)
            passed_slices = np.sort(passed_slices)
            n_slices = int(passed_slices.size)

            dst.createDimension("slice", n_slices)
            dst.createVariable("SliceNo_passed_slices", "i4", ("slice",))[:] = passed_slices.astype(np.int32)

            mf_slice_sum = np.full((n_slices, y_tile, x_tile), np.nan, dtype=np.float32)
            for k, s in enumerate(passed_slices):
                idx = np.where(slice_passed == int(s))[0]
                if idx.size == 0:
                    continue
                stack = mf_cluster[idx]  # (nInSlice,y,x)
                mf_slice_sum[k] = np.nansum(
                    np.nan_to_num(stack, nan=0.0, posinf=0.0, neginf=0.0),
                    axis=0
                ).astype(np.float32)

            vsum = dst.createVariable("MF_slice_sum", "f4", ("slice", ytile_dim, xtile_dim), zlib=True, complevel=4)
            vsum.setncattr("long_name", "Sum of MF over passed clusters within each slice tile")
            vsum[:] = mf_slice_sum

            # ---- Convenience: create MF_toprows(t,y,x) on swath grid (not true unflatten)
            MF_toprows = np.full(temp_shape, np.nan, dtype=np.float32)
            for k, s in enumerate(passed_slices):
                x0 = int(s) * x_tile
                x1 = x0 + x_tile
                MF_toprows[time_index, 0:y_tile, x0:x1] = mf_slice_sum[k]

            vtop = dst.createVariable("MF_toprows", "f4", temp_dims, zlib=True, complevel=4)
            vtop.setncattr(
                "long_name",
                "Stitched MF written into the first yTile rows of the swath grid (convenience view, not unflattened)"
            )
            vtop.setncattr("time_index_filled", int(time_index))
            vtop.setncattr("lambda_x_source", "DominantWavelength (km)")
            vtop.setncattr("y_placement", "MF_toprows[time_index, 0:yTile, x0:x1] = MF_slice_sum for each slice")
            vtop[:] = MF_toprows

            # Helpful globals
            dst.setncattr(
                "output_notes",
                "All original L6A attrs+vars copied. Cluster-dimension vars subset to PASSED clusters only. "
                "Temp/u/v/N^2 were flattened BEFORE slicing into cluster tiles. "
                "MF_toprows matches Temperature dims; only time_index_filled has values in 0:yTile and slice x-ranges."
            )
            dst.setncattr("cluster_area_mask", "Defined as finite(Amplitude)")
            dst.setncattr("slice_mapping", "x0 = SliceNo * xTile, x1 = x0 + xTile, y uses rows 0:yTile (top rows)")

    return cluster_dim, ytile_dim, xtile_dim


def stitch_slices_to_orbit_array(
    combined_by_slice: Dict[int, np.ndarray],
    slice_numbers: np.ndarray,
) -> np.ndarray:
    """
    Generic stitch along x for (yTile,xTile) blocks.
    Missing slices => NaNs.
    """
    if len(combined_by_slice) == 0:
        return np.empty((0, 0), dtype=float)

    first_slice = sorted(combined_by_slice.keys())[0]
    ny, nx = combined_by_slice[first_slice].shape

    smin = int(np.min(slice_numbers))
    smax = int(np.max(slice_numbers))
    full_slices = list(range(smin, smax + 1))

    blocks: List[np.ndarray] = []
    for s in full_slices:
        if s in combined_by_slice:
            A = combined_by_slice[s]
            if A.shape != (ny, nx):
                blocks.append(np.full((ny, nx), np.nan))
            else:
                blocks.append(A)
        else:
            blocks.append(np.full((ny, nx), np.nan))

    return np.concatenate(blocks, axis=1)


# --------------------------
# Main: matching -> passed clusters -> output passed-only netCDF -> stitched MF plot
# --------------------------
l6c = load_l6_for_matching(L6C_PATH)
l6a_match = load_l6_for_matching(L6A_PATH)

all_slices_a = np.unique(l6a_match["SlicesNo"]).astype(int)
all_slices_a = all_slices_a[all_slices_a >= 0]
all_slices_a = np.sort(all_slices_a)

print("Slices in L6A:", all_slices_a.tolist())

summary_lines: List[str] = []
summary_lines.append(f"L6A: {L6A_PATH}\n")
summary_lines.append(f"L6C: {L6C_PATH}\n\n")
summary_lines.append("slice  nA  nC  nBestSaved  nPassed  note\n")

passed_global_indices: List[int] = []

# Optional: combined reconstruction from PASSED clusters per slice (kept as placeholder)
combined_recon_by_slice: Dict[int, np.ndarray] = {}

for slice_no in all_slices_a:
    idx_a = np.where(l6a_match["SlicesNo"] == int(slice_no))[0]
    idx_c = np.where(l6c["SlicesNo"] == int(slice_no))[0]

    slice_dir = OUTDIR / f"slice_{slice_no:03d}"
    pairs_dir = slice_dir / "pairs"
    pairs_dir.mkdir(parents=True, exist_ok=True)

    if idx_a.size == 0:
        summary_lines.append(f"{slice_no:5d}  {0:2d}  {idx_c.size:2d}  {0:10d}  {0:7d}  no L6A clusters\n")
        continue
    if idx_c.size == 0:
        (slice_dir / "NO_L6C_CLUSTERS.txt").write_text(f"No L6C clusters for slice {slice_no}\n")
        summary_lines.append(f"{slice_no:5d}  {idx_a.size:2d}  {0:2d}  {0:10d}  {0:7d}  no L6C clusters\n")
        continue

    all_candidate_rows: List[Dict[str, Any]] = []
    best_rows: List[Dict[str, Any]] = []

    for a_idx in idx_a:
        a_idx = int(a_idx)
        A_a = l6a_match["Amplitude"][a_idx]
        L_a = l6a_match["DominantWavelength"][a_idx]
        T_a = l6a_match["Angle"][a_idx]

        candidates: List[Dict[str, Any]] = []
        for c_idx in idx_c:
            c_idx = int(c_idx)
            A_c = l6c["Amplitude"][c_idx]
            L_c = l6c["DominantWavelength"][c_idx]
            T_c = l6c["Angle"][c_idx]

            m = compute_match_metrics(A_a, L_a, T_a, A_c, L_c, T_c)
            m["cluster_a"] = a_idx
            m["cluster_c"] = c_idx
            m["passes"] = bool(passes_thresholds(m))
            m["score"] = float(score(m))
            candidates.append(m)
            all_candidate_rows.append(m)

        passed = [c for c in candidates if c["passes"]]
        if passed:
            best = max(passed, key=lambda d: d["score"])
        else:
            best = max(candidates, key=lambda d: d["score"])

        if best["passes"] or ASSIGN_BEST_EVEN_IF_FAILS:
            best_rows.append(best)

            left_maps = get_cluster_maps(l6a_match, best["cluster_a"])
            right_maps = get_cluster_maps(l6c, best["cluster_c"])

            pass_tag = "PASS" if best["passes"] else "FAIL"
            outpng = pairs_dir / f"slice_{slice_no:03d}_A_{best['cluster_a']:04d}_to_C_{best['cluster_c']:04d}_{pass_tag}.png"

            title_left = (
                f"L6A Temp | slice {slice_no} | idx {best['cluster_a']} | "
                f"n={np.count_nonzero(np.isfinite(left_maps['A']))}"
            )
            title_right = (
                f"L6C Rad | slice {slice_no} | idx {best['cluster_c']} | "
                f"IoU={best['iou']:.3f} inter={best['n_intersection']} "
                f"ratio={best['lam_ratio']:.3f} Δθ={best['ang_diff_deg']:.1f}°"
            )

            plot_side_by_side_4rows(
                left=left_maps,
                right=right_maps,
                title_left=title_left,
                title_right=title_right,
                outpath=outpng,
                dx_km=DX_KM,
                dy_km=DY_KM,
            )

    # Save tables for this slice
    cand_path = slice_dir / f"slice_{slice_no:03d}_ALL_candidates_ratio.txt"
    lines = []
    lines.append(f"L6A file: {L6A_PATH}\n")
    lines.append(f"L6C file: {L6C_PATH}\n")
    lines.append(f"Slice: {slice_no}\n\n")
    lines.append(
        "cluster_a  cluster_c  passes  score    iou    inter  "
        "lam_ratio  ang_diff_deg  thetaA_deg  thetaC_deg  lamA_med  lamC_med\n"
    )
    for m in sorted(all_candidate_rows, key=lambda d: (d["cluster_a"], -d["score"])):
        lines.append(
            f"{m['cluster_a']:9d}  {m['cluster_c']:9d}  {int(m['passes']):6d}  {m['score']:6.3f}  "
            f"{m['iou']:.3f}  {m['n_intersection']:5d}  "
            f"{m['lam_ratio']:8.3f}  {m['ang_diff_deg']:12.1f}  "
            f"{m['ang_src_deg']:10.1f}  {m['ang_cand_deg']:10.1f}  "
            f"{m['lam_src_med']:8.1f}  {m['lam_cand_med']:8.1f}\n"
        )
    cand_path.write_text("".join(lines))

    best_path = slice_dir / f"slice_{slice_no:03d}_BEST_matches_A_to_C_ratio.txt"
    lines = []
    lines.append(f"L6A file: {L6A_PATH}\n")
    lines.append(f"L6C file: {L6C_PATH}\n")
    lines.append(f"Slice: {slice_no}\n\n")
    lines.append(
        "cluster_a  best_cluster_c  passes  score    iou    inter  "
        "lam_ratio  ang_diff_deg  thetaA_deg  thetaC_deg  lamA_med  lamC_med\n"
    )
    for m in sorted(best_rows, key=lambda d: d["cluster_a"]):
        lines.append(
            f"{m['cluster_a']:9d}  {m['cluster_c']:14d}  {int(m['passes']):6d}  {m['score']:6.3f}  "
            f"{m['iou']:.3f}  {m['n_intersection']:5d}  "
            f"{m['lam_ratio']:8.3f}  {m['ang_diff_deg']:12.1f}  "
            f"{m['ang_src_deg']:10.1f}  {m['ang_cand_deg']:10.1f}  "
            f"{m['lam_src_med']:8.1f}  {m['lam_cand_med']:8.1f}\n"
        )
    best_path.write_text("".join(lines))

    passed_here = [int(m["cluster_a"]) for m in best_rows if bool(m["passes"])]
    passed_global_indices.extend(passed_here)

    summary_lines.append(
        f"{slice_no:5d}  {idx_a.size:2d}  {idx_c.size:2d}  {len(best_rows):10d}  {len(passed_here):7d}  ok\n"
    )

summary_path = OUTDIR / "SUMMARY_all_slices.txt"
summary_path.write_text("".join(summary_lines))

passed_global_indices = sorted(set(passed_global_indices))
passed_global_indices_np = np.asarray(passed_global_indices, dtype=int)
print("Total passed clusters:", passed_global_indices_np.size)

if passed_global_indices_np.size == 0:
    raise RuntimeError("No passed clusters found. Nothing to save/stitch.")

cluster_dim_name, ytile_dim_name, xtile_dim_name = create_passed_only_l6a_with_new_vars(
    src_l6a=L6A_PATH,
    dst_l6a=OUT_L6A_PASSED,
    passed_cluster_indices=passed_global_indices_np,
    time_index=TIME_INDEX,
)

# ---- Plot MF_toprows stitched (2D at TIME_INDEX) ----
with Dataset(OUT_L6A_PASSED, "r") as nc:
    MF2 = np.asarray(nc.variables["MF_toprows"][TIME_INDEX], dtype=float)  # (y,x)
    ny, nx = MF2.shape
    x1d_km = np.arange(nx, dtype=float) * DX_KM
    y1d_km = np.arange(ny, dtype=float) * DY_KM

    vmin, vmax = robust_limits(MF2, 2, 98)

    fig, ax = plt.subplots(figsize=(22, 3), constrained_layout=True)
    im = ax.pcolormesh(x1d_km, y1d_km, MF2, shading="auto", vmin=vmin, vmax=vmax)
    ax.set_title("Momentum Flux (passed clusters summed)")
    ax.set_xlabel("x (km)")
    ax.set_ylabel("y (km)")
    ax.set_ylim(0, 600)
    ax.set_aspect("equal", adjustable="box")
    fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01).set_label("MF (arb)")

    save_fig(fig, OUTDIR / "MF_stitched.png")

print("\nDone.")
print("Outputs in:", OUTDIR.resolve())
print("Summary:", summary_path.resolve())
print("PASSED-only augmented L6A:", OUT_L6A_PASSED.resolve())
print("New vars in output: u, v, NSquared, Temp_cluster, u_cluster, v_cluster, NSquared_cluster, MF_cluster, MF_slice_sum, MF_toprows")


# --------------------------
# Plot per-passed-cluster panels:
# Recon, Amp, Lambda, Angle, N^2, Temperature, MF
# --------------------------
def plot_passed_clusters_7rows(nc_path: Path, outdir: Path) -> None:
    outdir.mkdir(parents=True, exist_ok=True)

    with Dataset(nc_path, "r") as nc:
        amp = np.asarray(nc.variables["Amplitude"][:], dtype=float)
        lam = np.asarray(nc.variables["DominantWavelength"][:], dtype=float)
        ang = np.asarray(nc.variables["Angle"][:], dtype=float)
        n2  = np.asarray(nc.variables["NSquared_cluster"][:], dtype=float)
        tmp = np.asarray(nc.variables["Temp_cluster"][:], dtype=float)
        mf  = np.asarray(nc.variables["MF_cluster"][:], dtype=float)
        slc = np.asarray(nc.variables["SlicesNo"][:], dtype=int)

        if "ClusterReconstruction" in nc.variables:
            recon = np.asarray(nc.variables["ClusterReconstruction"][:], dtype=float)
        else:
            recon = np.full_like(amp, np.nan)

        n_cluster, ny, nx = amp.shape
        x1d_km = np.arange(nx, dtype=float) * DX_KM
        y1d_km = np.arange(ny, dtype=float) * DY_KM

        ang_plot = ang
        ang_label = "Angle (rad)"

        for i in range(n_cluster):
            s = int(slc[i])

            r_vmin, r_vmax = robust_limits(recon[i], 2, 98)
            a_vmin, a_vmax = robust_limits(amp[i], 2, 98)
            l_vmin, l_vmax = robust_limits(lam[i], 5, 95)
            t_vmin, t_vmax = robust_limits(ang_plot[i], 2, 98)
            T_vmin, T_vmax = robust_limits(tmp[i], 2, 98)
            m_vmin, m_vmax = robust_limits(mf[i], 2, 98)

            # N^2 is synthetic and constant → fixed limits look better
            n2_center = float(N2_VALUE)
            n2_vmin, n2_vmax = (0.5 * n2_center, 1.5 * n2_center)

            fig, axs = plt.subplots(
                7, 1, figsize=(14, 22),
                constrained_layout=True, sharex=True, sharey=True
            )

            def _panel(ax, Z, vmin, vmax, label, cmap=None):
                im = ax.pcolormesh(
                    x1d_km, y1d_km, Z,
                    shading="auto", vmin=vmin, vmax=vmax, cmap=cmap
                )
                cb = fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01)
                cb.set_label(label)
                ax.set_aspect("equal", adjustable="box")

            axs[0].set_title(f"Passed cluster {i:04d} | SliceNo={s}")

            _panel(axs[0], recon[i], r_vmin, r_vmax, "Recon (arb)", cmap="RdBu_r")
            _panel(axs[1], amp[i],   a_vmin, a_vmax, "Amplitude (K)")
            _panel(axs[2], lam[i],   l_vmin, l_vmax, "DominantWavelength (km)")
            _panel(axs[3], ang_plot[i], t_vmin, t_vmax, ang_label)
            _panel(axs[4], n2[i],    n2_vmin, n2_vmax, "N² (1/s²)")
            _panel(axs[5], tmp[i],   T_vmin, T_vmax, "Temperature (K)")
            _panel(axs[6], mf[i],    m_vmin, m_vmax, "MF (arb)")

            for ax in axs[:-1]:
                ax.set_ylabel("y (km)")
            axs[-1].set_ylabel("y (km)")
            axs[-1].set_xlabel("x (km)")

            outpng = (
                outdir
                / f"passed_cluster_{i:04d}_slice_{s:03d}"
                  "_Recon_Amp_Lam_Ang_N2_Temp_MF.png"
            )
            save_fig(fig, outpng)

    print("Per-cluster 7-row plots saved to:", outdir.resolve())


# ---- Make per-cluster plots ----
plot_passed_clusters_7rows(
    nc_path=OUT_L6A_PASSED,
    outdir=OUTDIR / "passed_cluster_plots_7rows",
)
```

## lambda_z using Dave's Equation
lambda z is negative when T < 225 K.  

```python
# %%
"""
ALL-SLICES matching + per-slice combined L6A recon + FULL-ORBIT stitched recon
+ slice Temperature + momentum flux for PASSED clusters only
+ sum MF per slice + stitch MF across orbit
+ output NetCDF contains original L6A global attrs + variables, but ONLY PASSED clusters
  are kept for cluster-dimension variables.

Momentum flux (per cluster pixel, masked to cluster area):
MF = 1/2 * g^2 / N^2 * (lambda_z / lambda_x) * (Amplitude / Temperature)^2 * (1/c^2)

NEW (your request):
- lambda_z is NO LONGER a constant.
- Compute per-pixel lambda_z (meters) then convert to km:
  lambda_z = 2*Amplitude / (N^2*T/g - g/c_p)
  c_p = 1005 J/(kg*K)
  T is Temperature (K), Amplitude is (K), N^2 is (1/s^2), g is (m/s^2)

Key rules:
- SliceNo already exists in L6A, keep it (but subset to passed clusters)
- Area, Amplitude, DominantWavelength, Angle already exist (subset to passed clusters)
- Also slice Temperature into Temp_cluster (passed clusters only)
- Only save passed clusters (output file cluster vars are subset)
- Add LambdaZ_cluster (km), MF_cluster, MF_slice_sum, MF_toprows (renamed from MF_full)
"""

from __future__ import annotations

from pathlib import Path
from typing import Dict, Any, List, Tuple

import numpy as np
import matplotlib.pyplot as plt
from netCDF4 import Dataset

# --------------------------
# User params
# --------------------------
L6C_PATH = Path("awe_l6c_q20_2024165T2120_03175_v01.nc")  # Radiance-based clusters
L6A_PATH = Path("awe_l6a_tmp_2024165T2120_03175_v01.nc")  # Temperature-based clusters

OUTDIR = Path("outputs_ALLSLICES_l6a_to_l6c_matching_ratio_STITCH_MF")
OUTDIR.mkdir(parents=True, exist_ok=True)
SAVE_DPI = 200

DX_KM = 2.0
DY_KM = 2.0

# ---- Synthetic u,v,N^2 settings ----
TEMP_VAR_NAME = "Temperature"  # variable name in L6A
TIME_INDEX = 0                 # which t index to use when slicing
N2_VALUE = 4e-4                # N=0.02 -> N^2=0.0004
U_VALUE = 0.0
V_VALUE = 0.0

# ---- Momentum flux constants ----
G_MS2 = 9.52
CP_JKGK = 1005.0          # J/(kg*K)
C_CANCEL = 1.0

# Output NetCDF (contains original L6A attrs+vars, but cluster vars subset to PASSED only)
OUT_L6A_PASSED = OUTDIR / f"{L6A_PATH.stem}_PASSED_with_u_v_NS2_TempCluster_LambdaZ_MF.nc"

# --------------------------
# Matching thresholds (tune)
# --------------------------
MIN_IOU = 0.05
MIN_INTERSECTION_PIX = 50
MAX_LAMBDA_MED_RATIO = 1.25
MAX_ANGLE_DIFF_DEG = 15.0

# Save best match plot even if FAIL (does NOT affect passing)
ASSIGN_BEST_EVEN_IF_FAILS = True


# --------------------------
# Helpers
# --------------------------
def save_fig(fig: plt.Figure, outpath: Path) -> None:
    outpath.parent.mkdir(parents=True, exist_ok=True)
    fig.savefig(outpath, dpi=SAVE_DPI, bbox_inches="tight")
    plt.close(fig)


def robust_limits(A: np.ndarray, p_lo: float = 2, p_hi: float = 98) -> Tuple[float, float]:
    a = np.asarray(A)
    good = np.isfinite(a)
    if not np.any(good):
        return (-1.0, 1.0)
    lo, hi = np.nanpercentile(a[good], [p_lo, p_hi])
    if (not np.isfinite(lo)) or (not np.isfinite(hi)) or lo == hi:
        lo = float(np.nanmin(a[good]))
        hi = float(np.nanmax(a[good]))
    return (float(lo), float(hi))


def nanpercentile_safe(a: np.ndarray, q: float) -> float:
    good = np.isfinite(a)
    if not np.any(good):
        return np.nan
    return float(np.nanpercentile(a[good], q))


def circ_diff_deg_180(a_deg: float, b_deg: float) -> float:
    if (not np.isfinite(a_deg)) or (not np.isfinite(b_deg)):
        return np.nan
    d = abs(a_deg - b_deg) % 180.0
    return min(d, 180.0 - d)


def circ_median_deg_180(theta_rad: np.ndarray, mask: np.ndarray) -> float:
    th = np.asarray(theta_rad, dtype=float)
    m = mask & np.isfinite(th)
    if not np.any(m):
        return np.nan
    ang = th[m]
    z = np.exp(1j * 2.0 * ang)
    zmean = np.mean(z)
    if (not np.isfinite(zmean.real)) or (not np.isfinite(zmean.imag)) or abs(zmean) == 0:
        return np.nan
    th_mean = 0.5 * np.angle(zmean)
    deg = (np.degrees(th_mean) + 180.0) % 180.0
    return float(deg)


def load_l6_for_matching(path: Path) -> Dict[str, Any]:
    with Dataset(path, "r") as nc:
        out: Dict[str, Any] = {
            "SlicesNo": np.asarray(nc.variables["SlicesNo"][:], dtype=int),
            "Amplitude": np.asarray(nc.variables["Amplitude"][:], dtype=float),
            "DominantWavelength": np.asarray(nc.variables["DominantWavelength"][:], dtype=float),
            "Angle": np.asarray(nc.variables["Angle"][:], dtype=float),
        }
        if "ClusterReconstruction" in nc.variables:
            out["ClusterReconstruction"] = np.asarray(nc.variables["ClusterReconstruction"][:], dtype=float)
        else:
            out["ClusterReconstruction"] = None
    return out


def get_cluster_maps(data: Dict[str, Any], idx: int) -> Dict[str, np.ndarray]:
    A = np.asarray(data["Amplitude"][idx], dtype=float)
    L = np.asarray(data["DominantWavelength"][idx], dtype=float)
    T = np.asarray(data["Angle"][idx], dtype=float)
    if data.get("ClusterReconstruction", None) is None:
        R = np.full_like(A, np.nan)
    else:
        R = np.asarray(data["ClusterReconstruction"][idx], dtype=float)
    return dict(R=R, A=A, L=L, T=T)


def compute_match_metrics(
    A_src: np.ndarray, L_src: np.ndarray, T_src: np.ndarray,
    A_cand: np.ndarray, L_cand: np.ndarray, T_cand: np.ndarray,
) -> Dict[str, Any]:
    mask_src = np.isfinite(A_src)
    mask_cand = np.isfinite(A_cand)
    inter = mask_src & mask_cand
    union = mask_src | mask_cand

    n_inter = int(np.count_nonzero(inter))
    n_union = int(np.count_nonzero(union))
    iou = (n_inter / n_union) if n_union > 0 else 0.0

    lam_src = L_src[inter]
    lam_cand = L_cand[inter]
    lam_src_med = nanpercentile_safe(lam_src, 50)
    lam_cand_med = nanpercentile_safe(lam_cand, 50)

    if np.isfinite(lam_src_med) and np.isfinite(lam_cand_med) and lam_src_med > 0 and lam_cand_med > 0:
        lam_ratio = max(lam_src_med, lam_cand_med) / min(lam_src_med, lam_cand_med)
    else:
        lam_ratio = np.nan

    ang_src_deg = circ_median_deg_180(T_src, inter)
    ang_cand_deg = circ_median_deg_180(T_cand, inter)
    ang_diff_deg = circ_diff_deg_180(ang_src_deg, ang_cand_deg)

    return dict(
        n_intersection=n_inter,
        iou=float(iou),
        lam_src_med=float(lam_src_med),
        lam_cand_med=float(lam_cand_med),
        lam_ratio=float(lam_ratio),
        ang_src_deg=float(ang_src_deg),
        ang_cand_deg=float(ang_cand_deg),
        ang_diff_deg=float(ang_diff_deg),
    )


def passes_thresholds(m: Dict[str, Any]) -> bool:
    if m["iou"] < MIN_IOU:
        return False
    if m["n_intersection"] < MIN_INTERSECTION_PIX:
        return False
    if (not np.isfinite(m["lam_ratio"])) or (m["lam_ratio"] > MAX_LAMBDA_MED_RATIO):
        return False
    if (not np.isfinite(m["ang_diff_deg"])) or (m["ang_diff_deg"] > MAX_ANGLE_DIFF_DEG):
        return False
    return True


def score(m: Dict[str, Any]) -> float:
    ang = m["ang_diff_deg"]
    if not np.isfinite(ang):
        ang = 999.0
    r = m["lam_ratio"]
    if (not np.isfinite(r)) or r <= 0:
        return 0.0
    return float(m["iou"]) * (1.0 / float(r)) / (1.0 + float(ang))


def plot_side_by_side_4rows(
    left: Dict[str, np.ndarray],
    right: Dict[str, np.ndarray],
    title_left: str,
    title_right: str,
    outpath: Path,
    dx_km: float,
    dy_km: float,
) -> None:
    ny, nx = left["A"].shape
    x1d_km = np.arange(nx, dtype=float) * dx_km
    y1d_km = np.arange(ny, dtype=float) * dy_km

    Rl_vmin, Rl_vmax = robust_limits(left["R"])
    Rr_vmin, Rr_vmax = robust_limits(right["R"])
    Al_vmin, Al_vmax = robust_limits(left["A"])
    Ar_vmin, Ar_vmax = robust_limits(right["A"])

    L_vmin, L_vmax = robust_limits(np.concatenate([left["L"].ravel(), right["L"].ravel()]), 5, 95)
    T_vmin, T_vmax = robust_limits(np.concatenate([left["T"].ravel(), right["T"].ravel()]))

    fig, axs = plt.subplots(4, 2, figsize=(16, 14), constrained_layout=True, sharex=True, sharey=True)

    def _panel(ax, Z, vmin, vmax, label, cmap=None):
        im = ax.pcolormesh(x1d_km, y1d_km, Z, shading="auto", vmin=vmin, vmax=vmax, cmap=cmap)
        cb = fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01)
        cb.set_label(label)
        ax.set_aspect("equal", adjustable="box")

    axs[0, 0].set_title(title_left)
    axs[0, 1].set_title(title_right)

    _panel(axs[0, 0], left["R"], Rl_vmin, Rl_vmax, "arb", cmap="RdBu_r")
    _panel(axs[0, 1], right["R"], Rr_vmin, Rr_vmax, "arb", cmap="RdBu_r")
    axs[0, 0].set_ylabel("Recon\ny (km)")

    _panel(axs[1, 0], left["A"], Al_vmin, Al_vmax, "K")
    _panel(axs[1, 1], right["A"], Ar_vmin, Ar_vmax, "kR")
    axs[1, 0].set_ylabel("Amp\ny (km)")

    _panel(axs[2, 0], left["L"], L_vmin, L_vmax, "km")
    _panel(axs[2, 1], right["L"], L_vmin, L_vmax, "km")
    axs[2, 0].set_ylabel("Lambda\ny (km)")

    _panel(axs[3, 0], left["T"], T_vmin, T_vmax, "rad")
    _panel(axs[3, 1], right["T"], T_vmin, T_vmax, "rad")
    axs[3, 0].set_ylabel("Angle\ny (km)")

    for r in range(4):
        for c in range(2):
            axs[r, c].set_xlabel("x (km)")

    save_fig(fig, outpath)


# --------------------------
# Flatten swath BEFORE slicing
# --------------------------
def flatten_swath(Z: np.ndarray, target_ny: int | None = None) -> np.ndarray:
    ny, nx = Z.shape
    out = np.full_like(Z, np.nan)

    bottoms = np.full(nx, np.nan, dtype=float)
    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size:
            bottoms[j] = valid_idx.max()

    valid_bottoms = bottoms[np.isfinite(bottoms)]
    if valid_bottoms.size == 0:
        med = float(np.nanmedian(Z)) if np.isfinite(np.nanmedian(Z)) else 0.0
        out = np.nan_to_num(Z, nan=med, posinf=med, neginf=med)
        if target_ny is not None and ny > target_ny:
            out = out[-target_ny:, :]
        return out

    max_bottom = int(valid_bottoms.max())

    for j in range(nx):
        col = Z[:, j]
        valid_idx = np.where(np.isfinite(col))[0]
        if valid_idx.size == 0:
            continue
        bottom = int(valid_idx.max())
        shift = max_bottom - bottom
        if shift == 0:
            out[:, j] = col
        elif shift > 0:
            out[shift:, j] = col[: ny - shift]
        else:
            shift_up = -shift
            out[: ny - shift_up, j] = col[shift_up:]

    med = float(np.nanmedian(out)) if np.isfinite(np.nanmedian(out)) else 0.0
    out = np.nan_to_num(out, nan=med, posinf=med, neginf=med)

    if target_ny is not None and ny > target_ny:
        start = max_bottom + 1 - target_ny
        start = max(start, 0)
        end = start + target_ny
        out = out[start:end, :]

    return out


# --------------------------
# L6A cloning with PASSED-only cluster variables + new vars
# --------------------------
def create_passed_only_l6a_with_new_vars(
    src_l6a: Path,
    dst_l6a: Path,
    passed_cluster_indices: np.ndarray,
    time_index: int,
) -> Tuple[str, str, str]:
    """
    Creates dst_l6a with:
      - all src global attrs
      - all variables copied; cluster-dimension variables subset to passed clusters only
      - ensures full-field u,v,NSquared exist (t,y,x) (synthetic constants if missing)
      - adds cluster-tile vars for passed clusters:
          Temp_cluster, u_cluster, v_cluster, NSquared_cluster, LambdaZ_cluster, MF_cluster
      - sums MF over passed clusters per slice: MF_slice_sum(slice,yTile,xTile)
      - writes MF_toprows(t,y,x) on the swath grid for convenience:
          MF_toprows[time_index, 0:yTile, x0:x1] = MF_slice_sum for slice
          missing slices remain NaN
        NOTE: MF_toprows is NOT an "unflattened" swath mapping. It's just placed in top rows.
    """
    passed_cluster_indices = np.asarray(passed_cluster_indices, dtype=int)
    if passed_cluster_indices.size == 0:
        raise ValueError("No passed clusters. Output would be empty.")

    dst_l6a.parent.mkdir(parents=True, exist_ok=True)

    with Dataset(src_l6a, "r") as src:
        if "Amplitude" not in src.variables or "SlicesNo" not in src.variables:
            raise KeyError("L6A must contain Amplitude and SlicesNo.")
        if TEMP_VAR_NAME not in src.variables:
            raise KeyError(f"L6A must contain {TEMP_VAR_NAME}.")

        amp_var = src.variables["Amplitude"]
        amp_dims = amp_var.dimensions
        if len(amp_dims) < 3:
            raise ValueError("Amplitude is expected to be (cluster, yTile, xTile).")
        cluster_dim = amp_dims[0]
        ytile_dim = amp_dims[1]
        xtile_dim = amp_dims[2]

        # Full-field temperature (t,y,x) and a 2D slice at time_index
        temp_var = src.variables[TEMP_VAR_NAME]
        temp_dims = temp_var.dimensions           # (t,y,x)
        temp_shape = temp_var.shape               # (t,y,x)

        temp_raw_2d = np.asarray(temp_var[time_index], dtype=float)  # (yFull,xFull)

        # Tile sizes from amplitude
        amp_all = np.asarray(src.variables["Amplitude"][:], dtype=float)  # (nCluster,yTile,xTile)
        n_cluster_all, y_tile, x_tile = amp_all.shape

        slices_no_all = np.asarray(src.variables["SlicesNo"][:], dtype=int)

        # Synthetic / existing full-field u/v/N^2 at time_index
        if "u" in src.variables:
            u_raw_2d = np.asarray(src.variables["u"][time_index], dtype=float)
        else:
            u_raw_2d = np.full_like(temp_raw_2d, U_VALUE, dtype=float)

        if "v" in src.variables:
            v_raw_2d = np.asarray(src.variables["v"][time_index], dtype=float)
        else:
            v_raw_2d = np.full_like(temp_raw_2d, V_VALUE, dtype=float)

        if "NSquared" in src.variables:
            n2_raw_2d = np.asarray(src.variables["NSquared"][time_index], dtype=float)
        else:
            n2_raw_2d = np.full_like(temp_raw_2d, N2_VALUE, dtype=float)

        # --------------------------
        # Flatten FIRST (and crop to exactly y_tile rows)
        # --------------------------
        temp_full_2d = flatten_swath(temp_raw_2d, target_ny=y_tile)
        u_full_2d    = flatten_swath(u_raw_2d,    target_ny=y_tile)
        v_full_2d    = flatten_swath(v_raw_2d,    target_ny=y_tile)
        n2_full_2d   = flatten_swath(n2_raw_2d,   target_ny=y_tile)

        # Sanity check mapping width for passed clusters
        smax = int(np.nanmax(slices_no_all[passed_cluster_indices]))
        x_need = (smax + 1) * x_tile
        if temp_full_2d.shape[1] < x_need:
            raise ValueError(
                "Flattened full-field x dimension too small for SliceNo mapping. "
                f"Need >= {x_need}, got {temp_full_2d.shape[1]}."
            )
        if temp_full_2d.shape[0] != y_tile:
            raise ValueError(
                f"Flattened y dimension mismatch: got {temp_full_2d.shape[0]}, expected y_tile={y_tile}."
            )

        n_passed = int(passed_cluster_indices.size)

        with Dataset(dst_l6a, "w") as dst:
            # ---- Global attrs
            for aname in src.ncattrs():
                dst.setncattr(aname, src.getncattr(aname))

            # ---- Dimensions (clone all, but override cluster length)
            for dname, dim in src.dimensions.items():
                if dname == cluster_dim:
                    dst.createDimension(dname, n_passed)
                else:
                    dst.createDimension(dname, (None if dim.isunlimited() else len(dim)))

            # ---- Copy variables (subset if they depend on cluster_dim)
            for vname, vsrc in src.variables.items():
                fill_value = vsrc.getncattr("_FillValue") if "_FillValue" in vsrc.ncattrs() else None
                vdst = dst.createVariable(
                    vname, vsrc.dtype, vsrc.dimensions,
                    zlib=True, complevel=4, fill_value=fill_value
                )

                for aname in vsrc.ncattrs():
                    if aname == "_FillValue":
                        continue
                    vdst.setncattr(aname, vsrc.getncattr(aname))

                if cluster_dim in vsrc.dimensions:
                    ax = vsrc.dimensions.index(cluster_dim)
                    data = vsrc[:]
                    data_sub = np.take(data, passed_cluster_indices, axis=ax)
                    vdst[:] = data_sub
                else:
                    vdst[:] = vsrc[:]

            # ---- Ensure full-field u,v,NSquared exist in dst (t,y,x)
            # NOTE: these are NOT flattened; they are synthetic constants on the swath grid if missing.
            def _ensure_full_field(name: str, value: float, units: str) -> None:
                if name in dst.variables:
                    return
                vv = dst.createVariable(name, "f4", temp_dims, zlib=True, complevel=4)
                vv.setncattr("long_name", name)
                vv.setncattr("units", units)
                vv[:] = np.full(temp_shape, value, dtype=np.float32)

            _ensure_full_field("u", float(U_VALUE), "m/s")
            _ensure_full_field("v", float(V_VALUE), "m/s")
            _ensure_full_field("NSquared", float(N2_VALUE), "1/s^2")

            # ---- Build per-passed-cluster vars (cluster,yTile,xTile)
            temp_cluster = np.full((n_passed, y_tile, x_tile), np.nan, dtype=np.float32)
            u_cluster = np.full_like(temp_cluster, np.nan)
            v_cluster = np.full_like(temp_cluster, np.nan)
            n2_cluster = np.full_like(temp_cluster, np.nan)
            lambda_z_cluster_km = np.full_like(temp_cluster, np.nan)  # (cluster,y,x) in km
            mf_cluster = np.full_like(temp_cluster, np.nan)

            amp_passed = np.asarray(dst.variables["Amplitude"][:], dtype=float)  # (n_passed,yTile,xTile)
            lam_passed = np.asarray(dst.variables["DominantWavelength"][:], dtype=float)  # lambda_x (km)
            slice_passed = np.asarray(dst.variables["SlicesNo"][:], dtype=int)

            for j in range(n_passed):
                s = int(slice_passed[j])
                if s < 0:
                    continue
                x0 = s * x_tile
                x1 = x0 + x_tile

                # These are flattened already and exactly y_tile high
                Ttile = temp_full_2d[:, x0:x1]
                Utile = u_full_2d[:, x0:x1]
                Vtile = v_full_2d[:, x0:x1]
                N2tile = n2_full_2d[:, x0:x1]

                mask = np.isfinite(amp_passed[j])  # cluster area mask

                Tm = np.full((y_tile, x_tile), np.nan, dtype=np.float32)
                Um = np.full((y_tile, x_tile), np.nan, dtype=np.float32)
                Vm = np.full((y_tile, x_tile), np.nan, dtype=np.float32)
                N2m = np.full((y_tile, x_tile), np.nan, dtype=np.float32)

                Tm[mask] = Ttile[mask].astype(np.float32)
                Um[mask] = Utile[mask].astype(np.float32)
                Vm[mask] = Vtile[mask].astype(np.float32)
                N2m[mask] = N2tile[mask].astype(np.float32)

                temp_cluster[j] = Tm
                u_cluster[j] = Um
                v_cluster[j] = Vm
                n2_cluster[j] = N2m

                # --------------------------
                # lambda_z (per pixel) then MF
                # --------------------------
                A = amp_passed[j]    # (y,x) amplitude (K)
                Lx = lam_passed[j]   # (y,x) km
                MF = np.full((y_tile, x_tile), np.nan, dtype=np.float32)
                LamZ_km = np.full((y_tile, x_tile), np.nan, dtype=np.float32)

                denom = (N2m * Tm / G_MS2) - (G_MS2 / CP_JKGK)  # K/m

                good_lz = (
                    mask
                    & np.isfinite(A)
                    & np.isfinite(Tm) & (Tm != 0)
                    & np.isfinite(N2m) & (N2m > 0)
                    & np.isfinite(denom) & (denom > 0)
                )
                if np.any(good_lz):
                    lambda_z_m = (2.0 * A[good_lz]) / denom[good_lz]   # meters
                    LamZ_km[good_lz] = (lambda_z_m / 1000.0).astype(np.float32)

                lambda_z_cluster_km[j] = LamZ_km

                good_mf = good_lz & np.isfinite(Lx) & (Lx > 0) & np.isfinite(LamZ_km) & (LamZ_km > 0)
                if np.any(good_mf):
                    ratio2 = (A[good_mf] / Tm[good_mf]) ** 2
                    MF_val = (
                        0.5
                        * (G_MS2 ** 2)
                        / N2m[good_mf]
                        * (LamZ_km[good_mf] / Lx[good_mf])   # (lambda_z/lambda_x)
                        * ratio2
                        * (1.0 / (C_CANCEL ** 2))
                    )
                    MF[good_mf] = MF_val.astype(np.float32)

                mf_cluster[j] = MF

            # Write new cluster vars
            vT = dst.createVariable("Temp_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim), zlib=True, complevel=4)
            vT.setncattr("long_name", "Temperature sliced to cluster tile and masked to cluster area (flattened first)")
            vT.setncattr("units", getattr(src.variables[TEMP_VAR_NAME], "units", "K"))
            vT[:] = temp_cluster

            vu = dst.createVariable("u_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim), zlib=True, complevel=4)
            vu.setncattr("units", "m/s")
            vu.setncattr("note", "Sliced from flattened u field; masked to cluster area")
            vu[:] = u_cluster

            vv = dst.createVariable("v_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim), zlib=True, complevel=4)
            vv.setncattr("units", "m/s")
            vv.setncattr("note", "Sliced from flattened v field; masked to cluster area")
            vv[:] = v_cluster

            vn2 = dst.createVariable("NSquared_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim), zlib=True, complevel=4)
            vn2.setncattr("units", "1/s^2")
            vn2.setncattr("note", "Sliced from flattened N^2 field; masked to cluster area")
            vn2[:] = n2_cluster

            vlz = dst.createVariable("LambdaZ_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim), zlib=True, complevel=4)
            vlz.setncattr("long_name", "Vertical wavelength proxy (km) from 2*A/(N^2*T/g - g/c_p), masked to cluster area")
            vlz.setncattr("units", "km")
            vlz.setncattr("formula", "lambda_z = 2*Amplitude /(N^2*T/g - g/c_p); stored in km")
            vlz.setncattr("cp_J_per_kgK", float(CP_JKGK))
            vlz.setncattr("g_ms2", float(G_MS2))
            vlz[:] = lambda_z_cluster_km

            vmf = dst.createVariable("MF_cluster", "f4", (cluster_dim, ytile_dim, xtile_dim), zlib=True, complevel=4)
            vmf.setncattr("long_name", "Momentum flux proxy (per pixel) for passed clusters, masked to cluster area")
            vmf.setncattr("formula", "0.5*g^2/N^2*(lambda_z/lambda_x)*(A/T)^2*(1/c^2)")
            vmf.setncattr("lambda_x_source", "DominantWavelength (km)")
            vmf.setncattr("lambda_z_source", "LambdaZ_cluster (km)")
            vmf.setncattr("g_ms2", float(G_MS2))
            vmf.setncattr("c_cancel", float(C_CANCEL))
            vmf[:] = mf_cluster

            # ---- Sum MF over passed clusters per slice
            passed_slices = np.unique(slice_passed[slice_passed >= 0]).astype(int)
            passed_slices = np.sort(passed_slices)
            n_slices = int(passed_slices.size)

            dst.createDimension("slice", n_slices)
            dst.createVariable("SliceNo_passed_slices", "i4", ("slice",))[:] = passed_slices.astype(np.int32)

            mf_slice_sum = np.full((n_slices, y_tile, x_tile), np.nan, dtype=np.float32)
            for k, s in enumerate(passed_slices):
                idx = np.where(slice_passed == int(s))[0]
                if idx.size == 0:
                    continue
                stack = mf_cluster[idx]  # (nInSlice,y,x)
                mf_slice_sum[k] = np.nansum(
                    np.nan_to_num(stack, nan=0.0, posinf=0.0, neginf=0.0),
                    axis=0
                ).astype(np.float32)

            vsum = dst.createVariable("MF_slice_sum", "f4", ("slice", ytile_dim, xtile_dim), zlib=True, complevel=4)
            vsum.setncattr("long_name", "Sum of MF over passed clusters within each slice tile")
            vsum[:] = mf_slice_sum

            # ---- Convenience: create MF_toprows(t,y,x) on swath grid (not true unflatten)
            MF_toprows = np.full(temp_shape, np.nan, dtype=np.float32)
            for k, s in enumerate(passed_slices):
                x0 = int(s) * x_tile
                x1 = x0 + x_tile
                MF_toprows[time_index, 0:y_tile, x0:x1] = mf_slice_sum[k]

            vtop = dst.createVariable("MF_toprows", "f4", temp_dims, zlib=True, complevel=4)
            vtop.setncattr(
                "long_name",
                "Stitched MF written into the first yTile rows of the swath grid (convenience view, not unflattened)"
            )
            vtop.setncattr("time_index_filled", int(time_index))
            vtop.setncattr("lambda_x_source", "DominantWavelength (km)")
            vtop.setncattr("lambda_z_source", "LambdaZ_cluster (km)")
            vtop.setncattr("y_placement", "MF_toprows[time_index, 0:yTile, x0:x1] = MF_slice_sum for each slice")
            vtop[:] = MF_toprows

            # Helpful globals
            dst.setncattr(
                "output_notes",
                "All original L6A attrs+vars copied. Cluster-dimension vars subset to PASSED clusters only. "
                "Temp/u/v/N^2 were flattened BEFORE slicing into cluster tiles. "
                "LambdaZ_cluster computed per-pixel from 2*A/(N^2*T/g - g/c_p) and stored in km. "
                "MF uses (LambdaZ_cluster / DominantWavelength). "
                "MF_toprows matches Temperature dims; only time_index_filled has values in 0:yTile and slice x-ranges."
            )
            dst.setncattr("cluster_area_mask", "Defined as finite(Amplitude)")
            dst.setncattr("slice_mapping", "x0 = SliceNo * xTile, x1 = x0 + xTile, y uses rows 0:yTile (top rows)")

    return cluster_dim, ytile_dim, xtile_dim


# --------------------------
# Main: matching -> passed clusters -> output passed-only netCDF -> stitched MF plot
# --------------------------
l6c = load_l6_for_matching(L6C_PATH)
l6a_match = load_l6_for_matching(L6A_PATH)

all_slices_a = np.unique(l6a_match["SlicesNo"]).astype(int)
all_slices_a = all_slices_a[all_slices_a >= 0]
all_slices_a = np.sort(all_slices_a)

print("Slices in L6A:", all_slices_a.tolist())

summary_lines: List[str] = []
summary_lines.append(f"L6A: {L6A_PATH}\n")
summary_lines.append(f"L6C: {L6C_PATH}\n\n")
summary_lines.append("slice  nA  nC  nBestSaved  nPassed  note\n")

passed_global_indices: List[int] = []

for slice_no in all_slices_a:
    idx_a = np.where(l6a_match["SlicesNo"] == int(slice_no))[0]
    idx_c = np.where(l6c["SlicesNo"] == int(slice_no))[0]

    slice_dir = OUTDIR / f"slice_{slice_no:03d}"
    pairs_dir = slice_dir / "pairs"
    pairs_dir.mkdir(parents=True, exist_ok=True)

    if idx_a.size == 0:
        summary_lines.append(f"{slice_no:5d}  {0:2d}  {idx_c.size:2d}  {0:10d}  {0:7d}  no L6A clusters\n")
        continue
    if idx_c.size == 0:
        (slice_dir / "NO_L6C_CLUSTERS.txt").write_text(f"No L6C clusters for slice {slice_no}\n")
        summary_lines.append(f"{slice_no:5d}  {idx_a.size:2d}  {0:2d}  {0:10d}  {0:7d}  no L6C clusters\n")
        continue

    all_candidate_rows: List[Dict[str, Any]] = []
    best_rows: List[Dict[str, Any]] = []

    for a_idx in idx_a:
        a_idx = int(a_idx)
        A_a = l6a_match["Amplitude"][a_idx]
        L_a = l6a_match["DominantWavelength"][a_idx]
        T_a = l6a_match["Angle"][a_idx]

        candidates: List[Dict[str, Any]] = []
        for c_idx in idx_c:
            c_idx = int(c_idx)
            A_c = l6c["Amplitude"][c_idx]
            L_c = l6c["DominantWavelength"][c_idx]
            T_c = l6c["Angle"][c_idx]

            m = compute_match_metrics(A_a, L_a, T_a, A_c, L_c, T_c)
            m["cluster_a"] = a_idx
            m["cluster_c"] = c_idx
            m["passes"] = bool(passes_thresholds(m))
            m["score"] = float(score(m))
            candidates.append(m)
            all_candidate_rows.append(m)

        passed = [c for c in candidates if c["passes"]]
        if passed:
            best = max(passed, key=lambda d: d["score"])
        else:
            best = max(candidates, key=lambda d: d["score"])

        if best["passes"] or ASSIGN_BEST_EVEN_IF_FAILS:
            best_rows.append(best)

            left_maps = get_cluster_maps(l6a_match, best["cluster_a"])
            right_maps = get_cluster_maps(l6c, best["cluster_c"])

            pass_tag = "PASS" if best["passes"] else "FAIL"
            outpng = pairs_dir / f"slice_{slice_no:03d}_A_{best['cluster_a']:04d}_to_C_{best['cluster_c']:04d}_{pass_tag}.png"

            title_left = (
                f"L6A Temp | slice {slice_no} | idx {best['cluster_a']} | "
                f"n={np.count_nonzero(np.isfinite(left_maps['A']))}"
            )
            title_right = (
                f"L6C Rad | slice {slice_no} | idx {best['cluster_c']} | "
                f"IoU={best['iou']:.3f} inter={best['n_intersection']} "
                f"ratio={best['lam_ratio']:.3f} Δθ={best['ang_diff_deg']:.1f}°"
            )

            plot_side_by_side_4rows(
                left=left_maps,
                right=right_maps,
                title_left=title_left,
                title_right=title_right,
                outpath=outpng,
                dx_km=DX_KM,
                dy_km=DY_KM,
            )

    # Save tables for this slice
    cand_path = slice_dir / f"slice_{slice_no:03d}_ALL_candidates_ratio.txt"
    lines = []
    lines.append(f"L6A file: {L6A_PATH}\n")
    lines.append(f"L6C file: {L6C_PATH}\n")
    lines.append(f"Slice: {slice_no}\n\n")
    lines.append(
        "cluster_a  cluster_c  passes  score    iou    inter  "
        "lam_ratio  ang_diff_deg  thetaA_deg  thetaC_deg  lamA_med  lamC_med\n"
    )
    for m in sorted(all_candidate_rows, key=lambda d: (d["cluster_a"], -d["score"])):
        lines.append(
            f"{m['cluster_a']:9d}  {m['cluster_c']:9d}  {int(m['passes']):6d}  {m['score']:6.3f}  "
            f"{m['iou']:.3f}  {m['n_intersection']:5d}  "
            f"{m['lam_ratio']:8.3f}  {m['ang_diff_deg']:12.1f}  "
            f"{m['ang_src_deg']:10.1f}  {m['ang_cand_deg']:10.1f}  "
            f"{m['lam_src_med']:8.1f}  {m['lam_cand_med']:8.1f}\n"
        )
    cand_path.write_text("".join(lines))

    best_path = slice_dir / f"slice_{slice_no:03d}_BEST_matches_A_to_C_ratio.txt"
    lines = []
    lines.append(f"L6A file: {L6A_PATH}\n")
    lines.append(f"L6C file: {L6C_PATH}\n")
    lines.append(f"Slice: {slice_no}\n\n")
    lines.append(
        "cluster_a  best_cluster_c  passes  score    iou    inter  "
        "lam_ratio  ang_diff_deg  thetaA_deg  thetaC_deg  lamA_med  lamC_med\n"
    )
    for m in sorted(best_rows, key=lambda d: d["cluster_a"]):
        lines.append(
            f"{m['cluster_a']:9d}  {m['cluster_c']:14d}  {int(m['passes']):6d}  {m['score']:6.3f}  "
            f"{m['iou']:.3f}  {m['n_intersection']:5d}  "
            f"{m['lam_ratio']:8.3f}  {m['ang_diff_deg']:12.1f}  "
            f"{m['ang_src_deg']:10.1f}  {m['ang_cand_deg']:10.1f}  "
            f"{m['lam_src_med']:8.1f}  {m['lam_cand_med']:8.1f}\n"
        )
    best_path.write_text("".join(lines))

    passed_here = [int(m["cluster_a"]) for m in best_rows if bool(m["passes"])]
    passed_global_indices.extend(passed_here)

    summary_lines.append(
        f"{slice_no:5d}  {idx_a.size:2d}  {idx_c.size:2d}  {len(best_rows):10d}  {len(passed_here):7d}  ok\n"
    )

summary_path = OUTDIR / "SUMMARY_all_slices.txt"
summary_path.write_text("".join(summary_lines))

passed_global_indices = sorted(set(passed_global_indices))
passed_global_indices_np = np.asarray(passed_global_indices, dtype=int)
print("Total passed clusters:", passed_global_indices_np.size)

if passed_global_indices_np.size == 0:
    raise RuntimeError("No passed clusters found. Nothing to save/stitch.")

cluster_dim_name, ytile_dim_name, xtile_dim_name = create_passed_only_l6a_with_new_vars(
    src_l6a=L6A_PATH,
    dst_l6a=OUT_L6A_PASSED,
    passed_cluster_indices=passed_global_indices_np,
    time_index=TIME_INDEX,
)

# ---- Plot MF_toprows stitched (2D at TIME_INDEX) ----
with Dataset(OUT_L6A_PASSED, "r") as nc:
    MF2 = np.asarray(nc.variables["MF_toprows"][TIME_INDEX], dtype=float)  # (y,x)
    ny, nx = MF2.shape
    x1d_km = np.arange(nx, dtype=float) * DX_KM
    y1d_km = np.arange(ny, dtype=float) * DY_KM

    vmin, vmax = robust_limits(MF2, 2, 98)

    fig, ax = plt.subplots(figsize=(22, 3), constrained_layout=True)
    im = ax.pcolormesh(x1d_km, y1d_km, MF2, shading="auto", vmin=vmin, vmax=vmax)
    ax.set_title("Momentum Flux (passed clusters summed)")
    ax.set_xlabel("x (km)")
    ax.set_ylabel("y (km)")
    ax.set_ylim(0, 600)
    ax.set_aspect("equal", adjustable="box")
    fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01).set_label("MF (arb)")

    save_fig(fig, OUTDIR / "MF_stitched.png")

print("\nDone (main pipeline).")
print("Outputs in:", OUTDIR.resolve())
print("Summary:", summary_path.resolve())
print("PASSED-only augmented L6A:", OUT_L6A_PASSED.resolve())
print("New vars in output: u, v, NSquared, Temp_cluster, u_cluster, v_cluster, NSquared_cluster, LambdaZ_cluster, MF_cluster, MF_slice_sum, MF_toprows")


# --------------------------
# Plot per-passed-cluster panels:
# Recon, Amp, Lambda_x, Angle, N^2, Temperature, lambda_z, MF
# --------------------------
def plot_passed_clusters_8rows(nc_path: Path, outdir: Path) -> None:
    outdir.mkdir(parents=True, exist_ok=True)

    with Dataset(nc_path, "r") as nc:
        amp = np.asarray(nc.variables["Amplitude"][:], dtype=float)
        lamx = np.asarray(nc.variables["DominantWavelength"][:], dtype=float)
        ang = np.asarray(nc.variables["Angle"][:], dtype=float)
        n2  = np.asarray(nc.variables["NSquared_cluster"][:], dtype=float)
        tmp = np.asarray(nc.variables["Temp_cluster"][:], dtype=float)
        lamz = np.asarray(nc.variables["LambdaZ_cluster"][:], dtype=float)
        mf  = np.asarray(nc.variables["MF_cluster"][:], dtype=float)
        slc = np.asarray(nc.variables["SlicesNo"][:], dtype=int)

        if "ClusterReconstruction" in nc.variables:
            recon = np.asarray(nc.variables["ClusterReconstruction"][:], dtype=float)
        else:
            recon = np.full_like(amp, np.nan)

        n_cluster, ny, nx = amp.shape
        x1d_km = np.arange(nx, dtype=float) * DX_KM
        y1d_km = np.arange(ny, dtype=float) * DY_KM

        ang_label = "Angle (rad)"

        for i in range(n_cluster):
            s = int(slc[i])

            r_vmin, r_vmax = robust_limits(recon[i], 2, 98)
            a_vmin, a_vmax = robust_limits(amp[i], 2, 98)
            lx_vmin, lx_vmax = robust_limits(lamx[i], 5, 95)
            t_vmin, t_vmax = robust_limits(ang[i], 2, 98)
            T_vmin, T_vmax = robust_limits(tmp[i], 2, 98)

            # N^2 usually constant-ish here → fixed limits look nicer
            n2_center = float(N2_VALUE)
            n2_vmin, n2_vmax = (0.5 * n2_center, 1.5 * n2_center)

            # lambda_z can have spikes → robust
            lz_vmin, lz_vmax = robust_limits(lamz[i], 2, 98)

            m_vmin, m_vmax = robust_limits(mf[i], 2, 98)

            fig, axs = plt.subplots(
                8, 1, figsize=(14, 25),
                constrained_layout=True, sharex=True, sharey=True
            )

            def _panel(ax, Z, vmin, vmax, label, cmap=None):
                im = ax.pcolormesh(
                    x1d_km, y1d_km, Z,
                    shading="auto", vmin=vmin, vmax=vmax, cmap=cmap
                )
                cb = fig.colorbar(im, ax=ax, fraction=0.03, pad=0.01)
                cb.set_label(label)
                ax.set_aspect("equal", adjustable="box")

            axs[0].set_title(f"Passed cluster {i:04d} | SliceNo={s}")

            _panel(axs[0], recon[i], r_vmin, r_vmax, "Recon (arb)", cmap="RdBu_r")
            _panel(axs[1], amp[i],   a_vmin, a_vmax, "Amplitude (K)")
            _panel(axs[2], lamx[i],  lx_vmin, lx_vmax, "DominantWavelength λx (km)")
            _panel(axs[3], ang[i],   t_vmin, t_vmax, ang_label)
            _panel(axs[4], n2[i],    n2_vmin, n2_vmax, "N² (1/s²)")
            _panel(axs[5], tmp[i],   T_vmin, T_vmax, "Temperature (K)")
            _panel(axs[6], lamz[i],  lz_vmin, lz_vmax, "λz (km)  [2A/(N²T/g - g/cp)]")
            _panel(axs[7], mf[i],    m_vmin, m_vmax, "MF (arb)")

            for ax in axs[:-1]:
                ax.set_ylabel("y (km)")
            axs[-1].set_ylabel("y (km)")
            axs[-1].set_xlabel("x (km)")

            outpng = (
                outdir
                / f"passed_cluster_{i:04d}_slice_{s:03d}"
                  "_Recon_Amp_LamX_Ang_N2_Temp_LamZ_MF.png"
            )
            save_fig(fig, outpng)

    print("Per-cluster 8-row plots saved to:", outdir.resolve())


# ---- Make per-cluster plots ----
plot_passed_clusters_8rows(
    nc_path=OUT_L6A_PASSED,
    outdir=OUTDIR / "passed_cluster_plots_8rows",
)
```
