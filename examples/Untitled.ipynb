### 0) Open Anaconda Prompt

Start menu → Anaconda Prompt

### 1) Install Git (inside Conda)

```bat
conda install -c conda-forge git
git --version
```

### 2) Create the project env

```bat
conda create -n juwavelet_env python=3.11 -y
conda activate juwavelet_env
```

### 3) Install kernel support in the env

```bat
conda install -c conda-forge ipykernel -y
```

### 4) Keep Jupyter in base

```bat
conda activate base
conda install -c conda-forge jupyterlab notebook nb_conda_kernels -y
```

### 5) Clone the repo

```bat
cd %USERPROFILE%
git clone https://jugit.fz-juelich.de/j.ungermann/juwavelet.git
cd juwavelet
```

### 6) Install Python deps for the repo

Activate your env first, then install the package

```bat
conda activate juwavelet_env
pip install -e .
```

If the repo has a `requirements.txt` instead, do:

```bat
pip install -r requirements.txt
```

### 7) Register the Jupyter kernel for this env

```bat
python -m ipykernel install --user --name juwavelet_env --display-name "Python (juwavelet)"
```

### 8) Launch Jupyter and select the kernel

```bat
conda activate base
jupyter lab
```

In Jupyter: Kernel → Change kernel → Python (juwavelet)

## Verify imports inside a notebook

In a new notebook with the “Python (juwavelet)” kernel:

```python
import juwavelet
import sys
print(sys.executable)  # should point to ...\.conda\envs\juwavelet_env\python.exe
```

## Common fixes

* `No module named ipykernel`: install it in `juwavelet_env`

  ```bat
  conda activate juwavelet_env
  conda install -c conda-forge ipykernel
  ```
* `jupyter-notebook not found`: install Notebook or JupyterLab in base

  ```bat
  conda activate base
  conda install -c conda-forge jupyterlab notebook
  ```
* Kernel not listed

  ```bat
  jupyter kernelspec list
  ```
