```python
# %%
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import cartopy.crs as ccrs
import cartopy.feature as cfeature

# -------------------------
# Input
# -------------------------
csv_path = "matched_all_2024_2025_methods.csv"
out_dir = "peak_OH_altitude_maps"
os.makedirs(out_dir, exist_ok=True)

TARGET_LAT = 41.934
TARGET_LON = -111.421
R_MAX_KM = 1000

# -------------------------
# Haversine distance
# -------------------------
def haversine(lat1, lon1, lat2, lon2):
    R = 6371.0

    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)
    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2.0) ** 2
        + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2.0) ** 2
    )

    c = 2.0 * np.arcsin(np.sqrt(a))
    return R * c

# -------------------------
# Read data
# -------------------------
df = pd.read_csv(csv_path)

lat = df["saber_lat"].to_numpy()
lon = df["saber_lon"].to_numpy()
peak_alt = df["peak_altitude_km"].to_numpy()

good = np.isfinite(lat) & np.isfinite(lon) & np.isfinite(peak_alt)
lat = lat[good]
lon = lon[good]
peak_alt = peak_alt[good]

lon = ((lon + 180) % 360) - 180

# -------------------------
# Filter within radius
# -------------------------
dist = haversine(TARGET_LAT, TARGET_LON, lat, lon)
mask = dist <= R_MAX_KM

lat_f = lat[mask]
lon_f = lon[mask]
peak_alt_f = peak_alt[mask]
dist_f = dist[mask]

# -------------------------
# Stats
# -------------------------
n = len(peak_alt_f)

if n == 0:
    raise ValueError("No points found within the selected radius.")

stats_text = (
    f"N = {n}\n"
    f"Mean = {np.mean(peak_alt_f):.2f} km\n"
    f"Std = {np.std(peak_alt_f):.2f} km\n"
    f"Median = {np.median(peak_alt_f):.2f} km\n"
    f"Min = {np.min(peak_alt_f):.2f} km\n"
    f"Max = {np.max(peak_alt_f):.2f} km\n"
    f"Range = {np.ptp(peak_alt_f):.2f} km\n"
    f"Mean distance = {np.mean(dist_f):.1f} km"
)

print("Peak OH altitude stats within radius")
print("------------------------------------")
print(stats_text)

# -------------------------
# Plot
# -------------------------
fig = plt.figure(figsize=(9, 8))
ax = plt.axes(projection=ccrs.PlateCarree())

deg_lat = R_MAX_KM / 111.0
deg_lon = R_MAX_KM / (111.0 * np.cos(np.radians(TARGET_LAT)))

ax.set_extent(
    [
        TARGET_LON - deg_lon * 1.15,
        TARGET_LON + deg_lon * 1.15,
        TARGET_LAT - deg_lat * 1.15,
        TARGET_LAT + deg_lat * 1.15,
    ],
    crs=ccrs.PlateCarree(),
)

ax.coastlines(linewidth=0.8)
ax.add_feature(cfeature.BORDERS, linewidth=0.5)
ax.add_feature(cfeature.STATES, linewidth=0.5, alpha=0.7)

gl = ax.gridlines(
    draw_labels=True,
    linewidth=0.4,
    alpha=0.5,
    linestyle="--"
)
gl.top_labels = False
gl.right_labels = False

sc = ax.scatter(
    lon_f,
    lat_f,
    c=peak_alt_f,
    s=18,
    cmap="viridis",
    transform=ccrs.PlateCarree(),
)

ax.plot(
    TARGET_LON,
    TARGET_LAT,
    "r*",
    markersize=13,
    transform=ccrs.PlateCarree(),
    label="Target",
)

# Approximate 1000 km circle
angles = np.linspace(0, 2 * np.pi, 300)
circle_lat = TARGET_LAT + deg_lat * np.sin(angles)
circle_lon = TARGET_LON + deg_lon * np.cos(angles)

ax.plot(
    circle_lon,
    circle_lat,
    "r--",
    linewidth=1.2,
    transform=ccrs.PlateCarree(),
    label=f"{R_MAX_KM} km radius",
)

cb = plt.colorbar(sc, ax=ax, pad=0.02, shrink=0.82)
cb.set_label("Peak OH Altitude (km)")

ax.text(
    0.02,
    0.02,
    stats_text,
    transform=ax.transAxes,
    fontsize=10,
    va="bottom",
    ha="left",
    bbox=dict(facecolor="white", alpha=0.8, edgecolor="gray"),
)

ax.set_title(
    f"SABER OH 16 VER Peak Altitude within {R_MAX_KM} km"
)

ax.legend(loc="upper right")

plt.tight_layout()
plt.savefig(
    f"{out_dir}/regional_peak_OH_altitude_map_with_stats.png",
    dpi=300,
    bbox_inches="tight",
)
plt.show()
```

```python
# %%
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# -------------------------
# Input
# -------------------------
csv_path = "matched_all_2024_2025_methods.csv"
out_dir = "peak_OH_altitude_DOY"
os.makedirs(out_dir, exist_ok=True)

TARGET_LAT = 41.934
TARGET_LON = -111.421
R_MAX_KM = 1000

# -------------------------
# Distance function
# -------------------------
def haversine(lat1, lon1, lat2, lon2):
    R = 6371.0

    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)
    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
    )

    return R * 2 * np.arcsin(np.sqrt(a))

# -------------------------
# Read data
# -------------------------
df = pd.read_csv(csv_path)

required_cols = ["saber_lat", "saber_lon", "saber_time", "peak_altitude_km"]
missing = [c for c in required_cols if c not in df.columns]
if missing:
    raise ValueError(f"Missing columns: {missing}")

df["saber_time"] = pd.to_datetime(df["saber_time"], errors="coerce")
df["doy"] = df["saber_time"].dt.dayofyear
df["year"] = df["saber_time"].dt.year

# clean
df = df.dropna(subset=["saber_lat", "saber_lon", "saber_time", "peak_altitude_km"])

# longitude to -180 to 180
df["saber_lon"] = ((df["saber_lon"] + 180) % 360) - 180

# -------------------------
# Filter within radius
# -------------------------
df["distance_from_target_km"] = haversine(
    TARGET_LAT,
    TARGET_LON,
    df["saber_lat"].to_numpy(),
    df["saber_lon"].to_numpy(),
)

df_f = df[df["distance_from_target_km"] <= R_MAX_KM].copy()

if len(df_f) == 0:
    raise ValueError("No points found within the selected radius.")

print(f"N within {R_MAX_KM} km = {len(df_f)}")
print(f"Mean = {df_f['peak_altitude_km'].mean():.2f} km")
print(f"Std = {df_f['peak_altitude_km'].std():.2f} km")
print(f"Median = {df_f['peak_altitude_km'].median():.2f} km")
print(
    f"Range = {df_f['peak_altitude_km'].min():.2f} "
    f"to {df_f['peak_altitude_km'].max():.2f} km"
)

# -------------------------
# Scatter: peak altitude vs DOY
# -------------------------
plt.figure(figsize=(10, 5))

for year, g in df_f.groupby("year"):
    plt.scatter(
        g["doy"],
        g["peak_altitude_km"],
        s=18,
        alpha=0.65,
        label=str(int(year)),
    )

plt.xlabel("Day of Year")
plt.ylabel("Peak OH Altitude (km)")
plt.title(
    f"SABER OH 16 VER Peak Altitude vs DOY\n"
    f"within {R_MAX_KM} km of ({TARGET_LAT:.3f}, {TARGET_LON:.3f})"
)
plt.grid(alpha=0.3)
plt.legend(title="Year")
plt.tight_layout()

plt.savefig(
    f"{out_dir}/peak_OH_altitude_vs_DOY_within_{R_MAX_KM}km.png",
    dpi=300,
    bbox_inches="tight",
)
plt.show()

# -------------------------
# Daily median and std
# -------------------------
daily = (
    df_f.groupby(["year", "doy"])["peak_altitude_km"]
    .agg(["count", "mean", "median", "std"])
    .reset_index()
)

plt.figure(figsize=(10, 5))

for year, g in daily.groupby("year"):
    plt.plot(
        g["doy"],
        g["median"],
        marker="o",
        markersize=4,
        linewidth=1.2,
        label=f"{int(year)} median",
    )

plt.xlabel("Day of Year")
plt.ylabel("Peak OH Altitude (km)")
plt.title(
    f"Daily Median SABER OH 16 VER Peak Altitude\n"
    f"within {R_MAX_KM} km of ({TARGET_LAT:.3f}, {TARGET_LON:.3f})"
)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()

plt.savefig(
    f"{out_dir}/daily_median_peak_OH_altitude_vs_DOY_within_{R_MAX_KM}km.png",
    dpi=300,
    bbox_inches="tight",
)
plt.show()

# -------------------------
# Save filtered data and daily stats
# -------------------------
df_f.to_csv(
    f"{out_dir}/filtered_points_within_{R_MAX_KM}km.csv",
    index=False,
)

daily.to_csv(
    f"{out_dir}/daily_stats_peak_OH_altitude_within_{R_MAX_KM}km.csv",
    index=False,
)

print(f"Saved plots and tables to: {out_dir}")
```
